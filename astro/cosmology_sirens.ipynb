# Standardsirenen

Das Notebook misst die Hubble-Konstante mit GW170817, der Verschmelzung zweier Neutronensterne in der Galaxie NGC 4993. Die Entfernung kommt aus den Stichproben der Parameterschätzung von LIGO und Virgo (GWTC-1), die Geschwindigkeit aus der Rotverschiebung der Galaxiengruppe samt Korrektur der Eigenbewegung. Die Bayes-Analyse folgt Abbott et al. (2017).

Dazu kommen:
- die Entartung von Entfernung und Neigung,
- der Beobachtungswinkel des Jets aus der Radio-Interferometrie (Mooley et al. 2018, Hotokezaka et al. 2019),
- der Einfluss der Eigenbewegung,
- eine Vorhersage, wie viele solche Ereignisse für 2 % nötig sind.

Kann eine einzige Gravitationswelle die Hubble-Konstante messen, ganz ohne Entfernungsleiter, und wie viele braucht man, um die Hubble-Spannung zu entscheiden?

#### **Hintergrund**

Am 17. August 2017 registrierten die beiden LIGO-Detektoren und Virgo die Gravitationswelle zweier verschmelzender Neutronensterne, GW170817. 1.7 Sekunden später sah der Satellit Fermi einen kurzen Gammablitz. Elf Stunden danach fanden optische Teleskope eine Kilonova in der Galaxie NGC 4993, rund 40 Mpc entfernt. Zum ersten Mal hatte man von derselben Quelle eine Gravitationswelle und Licht.

Bernard Schutz hatte 1986 bemerkt, was das für die Kosmologie bedeutet. Die Form der Welle verrät die Massen der Sterne, ihre Amplitude dann die Entfernung, ohne Eichung an Cepheiden oder anderen Kerzen. Holz & Hughes nannten solche Quellen 2005 **Standardsirenen**. Kennt man zusätzlich die Rotverschiebung der Wirtsgalaxie, folgt aus dem Hubble-Gesetz $v = H_0 d$ direkt H₀. Abbott et al. (2017) erhielten so aus GW170817 $H_0 = 70.0^{+12.0}_{-8.0}$ km/s/Mpc. Das liegt zwischen Planck und SH0ES, aber mit einem Fehler, der beide zulässt.

Was Forschende damit tun:
- Sie verbinden die Entfernung aus der Welle mit der Rotverschiebung des Gegenstücks.
- Sie brechen die Entartung zwischen Entfernung und Neigung der Bahn mit zusätzlichen Beobachtungen, zum Beispiel dem Jet im Radio.
- Für Ereignisse ohne Gegenstück, die „dunklen Sirenen“, mitteln sie statistisch über die möglichen Wirtsgalaxien in einem Katalog.

Mit jedem weiteren Ereignis mit Gegenstück wird der Fehler kleiner. Seit GW170817 gab es aber kein zweites.

#### **Theorie**

##### *1. Die Entfernung aus der Welle*

Zwei kompakte Objekte auf engen Bahnen strahlen Gravitationswellen ab, verlieren Energie und kommen sich näher; Frequenz und Amplitude steigen (**Chirp**). Wie schnell die Frequenz steigt, hängt nur von der **Chirpmasse** $\mathcal M = (m_1 m_2)^{3/5}/(m_1 + m_2)^{1/5}$ ab, $\dot f \propto \mathcal M^{5/3} f^{11/3}$. Die Amplitude in einem Detektor ist

$$h \propto \frac{\mathcal M^{5/3} f^{2/3}}{d_L}\times\left[\frac{1 + \cos^2\iota}{2}\ \text{für }h_+,\quad \cos\iota\ \text{für }h_\times\right] .$$

Aus $\dot f$ folgt $\mathcal M$, aus der Amplitude dann die Leuchtkraftentfernung $d_L$. Bei 40 Mpc (z ≈ 0.01) unterscheiden sich $d_L$ und die mitbewegte Entfernung nur um etwa 1 %, wenig gegen die Fehler; das Notebook setzt wie Abbott et al. $d = d_L$ im Hubble-Gesetz.

##### *2. Die Entartung von Entfernung und Neigung*

Der Winkel $\iota$ zwischen Bahndrehimpuls und Sichtlinie ändert die Amplitude fast so wie die Entfernung. Eine Bahn, die man von oben sieht ($\cos\iota = \pm 1$), wirkt hell und nah; eine schräg gesehene wirkt schwächer und weiter weg. Nur das Verhältnis von $h_+$ zu $h_\times$, das mehrere Detektoren messen, trennt die beiden etwas. Für GW170817 liegt $\cos\iota$ nahe −1 (die Bahn dreht sich, von uns aus gesehen, im Uhrzeigersinn). Ob −1 oder −0.8 macht in der Entfernung aber rund 20 % aus.

##### *3. Die Geschwindigkeit der Galaxie*

Die gemessene Rotverschiebung enthält neben der Expansion die **Eigenbewegung** der Galaxie: $v_r = H_0 d + v_p$. Bei 40 Mpc ist $H_0 d \approx 3000$ km/s und $v_p$ typisch 300 km/s, also 10 %. Abbott et al. (2017) nehmen für die Gruppe um NGC 4993 im Ruhesystem der Hintergrundstrahlung $v_r = 3327 \pm 72$ km/s. Aus einer Karte der Geschwindigkeiten (6dF) folgt $\langle v_p\rangle = 310$ km/s; für deren Unsicherheit setzen sie vorsichtig 150 km/s an. Das gibt die Hubble-Geschwindigkeit $v_H = 3017 \pm 166$ km/s.

##### *4. Die Bayes-Analyse*

Mit den Stichproben $d_i$ der Parameterschätzung, die mit dem Prior $p(d) \propto d^2$ (gleichmäßig im Volumen) gezogen sind, ist nach Abbott et al. (2017, Gl. 7)

$$p(H_0 \mid \text{Daten}) \propto \frac{p(H_0)}{N_s(H_0)} \sum_i \mathcal N\!\left(v_r - H_0 d_i;\ \langle v_p\rangle,\ \sqrt{\sigma_{v_r}^2 + \sigma_{v_p}^2}\right) .$$

Das Integral über die Eigenbewegung ist dabei schon ausgeführt. Der Prior ist $p(H_0) \propto 1/H_0$ (gleichmäßig im Logarithmus). Die Selektion $N_s(H_0)$ ist konstant, weil bei 40 Mpc nur die Entfernung entscheidet, ob ein Ereignis gefunden wird, nicht H₀. Angegeben werden das Maximum der Posterior und das kürzeste Intervall mit 68.3 % Wahrscheinlichkeit.

##### *5. Der Jet als Winkelmesser*

Bei der Verschmelzung entstand ein schmaler relativistischer Jet entlang der Drehachse. Mit Radio-Interferometrie über Kontinente (VLBI) sahen Mooley et al. (2018) seinen Schwerpunkt mit scheinbarer Überlichtgeschwindigkeit wandern, $\beta_\text{app} \approx 4$. Das gelingt nur, wenn man fast entlang des Jets blickt, und legt den Beobachtungswinkel $\theta_\text{obs} = \pi - \iota$ fest. Weil $\beta_\text{app}$ mit der Entfernung skaliert, gilt die Schranke für das Produkt (Hotokezaka et al. 2019):

$$0.25 < \theta_\text{obs}\,\frac{d}{41\ \text{Mpc}} < 0.45\ \text{rad} .$$

Mit ihr fallen die schräg gesehenen, nahen Lösungen weg, und die Entfernung wird viel genauer.

##### *6. Viele Sirenen*

Unabhängige Ereignisse multiplizieren ihre Likelihoods. Ist jedes für sich gaußisch mit relativem Fehler $\sigma_1$, wird der gemeinsame Fehler $\sigma_1/\sqrt N$. Die Posterior von GW170817 ist aber schief, mit einem langen Ausläufer zu großem H₀ (die schräg gesehenen Lösungen). Erst ab einigen Ereignissen nähert sich das Produkt einer Normalverteilung.

**Fachbegriffe:**
- **Leuchtkraftentfernung $d_L$:** die Entfernung, für die der Fluss wie $1/d_L^2$ abnimmt; für Gravitationswellen fällt die Amplitude wie $1/d_L$.
- **Kilonova:** das Leuchten der radioaktiven schweren Kerne, die bei der Verschmelzung von Neutronensternen entstehen.
- **Parameterschätzung (PE):** Die Messdaten werden mit vielen berechneten Wellenformen verglichen; das Ergebnis sind Stichproben der Posterior aller Parameter.
- **Hubble-Geschwindigkeit $v_H$:** der Teil der Geschwindigkeit, der auf die Expansion zurückgeht, $v_H = H_0 d$.

#### **Daten**

- **Parameterschätzung von GW170817** (LIGO Scientific and Virgo Collaborations, GWTC-1, Abbott et al. 2019): Stichproben der Posterior mit dem Wellenformmodell IMRPhenomPv2NRT, einmal mit kleinen Spins (|χ| < 0.05, wie für Neutronensterne in der Milchstraße) und einmal mit großen Spins (|χ| < 0.89). Die Himmelsposition ist auf NGC 4993 festgehalten. Dazu gehören Stichproben des Priors, mit denen sich prüfen lässt, dass $p(d) \propto d^2$ ist. Eine HDF5-Datei aus dem Dokumentenarchiv von LIGO (DCC P1800370).
- **Geschwindigkeit von NGC 4993** (Abbott et al. 2017): $v_r = 3327 \pm 72$ km/s für die Gruppe im Ruhesystem der Hintergrundstrahlung, Eigenbewegung $\langle v_p\rangle = 310 \pm 150$ km/s.
- **Beobachtungswinkel des Jets** (Mooley et al. 2018; Hotokezaka et al. 2019) und **Vergleichswerte** aus Abbott et al. (2017), Hotokezaka et al. (2019), GWTC-3, Planck, SH0ES und den anderen Notebooks der Reihe; als Konstanten im Code.

**Hintergrund & Fachbegriffe:**
- **IMRPhenomPv2NRT:** eine phänomenologische Wellenform für Einspiralen, Verschmelzung und Abklingen, mit Präzession der Spins und Gezeiteneffekten der Neutronensterne.
- **$\cos\theta_{JN}$:** der Kosinus des Winkels zwischen Gesamtdrehimpuls und Sichtlinie; bei kleinen Spins fast gleich $\cos\iota$.

```python
import os
import urllib.request
from pathlib import Path

ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "astropy").mkdir(parents=True, exist_ok=True)
os.environ["XDG_CACHE_HOME"] = str(ASTRODATA)

import h5py
import numpy as np
import matplotlib.pyplot as plt

PE_URL = "https://dcc.ligo.org/public/0157/P1800370/005/GW170817_GWTC-1.hdf5"
GW_DIR = ASTRODATA / "gwosc"
V_R, SIG_VR = 3327.0, 72.0                                # Gruppe um NGC 4993, CMB-System [km/s]
V_P, SIG_VP = 310.0, 150.0                                # Eigenbewegung (6dF) mit vorsichtigem Fehler
JET = (0.25, 0.45, 41.0)                                  # 0.25 < θ_obs · d/41 Mpc < 0.45 rad
PLANCK, SH0ES = (67.36, 0.54), (73.04, 1.04)
ABBOTT17 = (70.0, 12.0, 8.0)                              # MAP, +, − (68.3 %)
HOTOKEZAKA19 = (68.9, 4.6, 4.5)                           # Median, +, − (hydrodynamisches Jet-Modell)
GWTC3 = {"GWTC-3: Massenverteilung + GW170817": (68.0, 12.0, 8.0), "GWTC-3: Galaxienkatalog": (68.0, 8.0, 6.0)}
OTHERS = {"Inverse Leiter: DESI + BBN": (68.53, 0.60, 0.60), "Kosmische Chronometer (ΛCDM)": (66.99, 5.5, 5.5)}
H = np.linspace(20, 200, 3601)                            # Gitter für H₀ [km/s/Mpc]


def download(url, path, check, timeout=300):
    path = Path(path)
    if not path.exists():
        path.parent.mkdir(parents=True, exist_ok=True)
        with urllib.request.urlopen(url, timeout=timeout) as r:
            data = r.read()
        if not check(data):
            raise OSError(f"{path.name}: unerwarteter Inhalt")
        path.write_bytes(data)
    return path


with h5py.File(download(PE_URL, GW_DIR / "GW170817_GWTC-1.hdf5", lambda d: d[:8] == b"\x89HDF\r\n\x1a\n"), "r") as f:
    PE = {k: f[f"IMRPhenomPv2NRT_{k}Spin_posterior"][()] for k in ("low", "high")}
    PRIOR = f["IMRPhenomPv2NRT_lowSpin_prior"][()]
D = {k: v["luminosity_distance_Mpc"] for k, v in PE.items()}
COSI = {k: v["costheta_jn"] for k, v in PE.items()}

for k in ("low", "high"):
    lo, med, hi = np.percentile(D[k], [5, 50, 95])
    print(f"{'kleine' if k == 'low' else 'große':6s} Spins: {len(D[k])} Stichproben, d = {med:.1f} +{hi - med:.1f}/−{med - lo:.1f} Mpc (90 %), "
          f"cos θ_JN = {np.median(COSI[k]):.2f}, Himmel fest bei RA = {np.degrees(np.median(PE[k]['right_ascension'])):.2f}°, "
          f"Dec = {np.degrees(np.median(PE[k]['declination'])):.2f}°")
print(f"Prior: {len(PRIOR)} Stichproben, d von {PRIOR['luminosity_distance_Mpc'].min():.1f} bis {PRIOR['luminosity_distance_Mpc'].max():.1f} Mpc")
```

**Werkzeuge: Posterior von H₀ und ihre Kennzahlen** (Theorie, Abschnitt 4). `posterior` rechnet $p(H_0)$ auf dem Gitter aus Entfernungsstichproben, wahlweise gewichtet (für den Jet) und mit anderer Eigenbewegung oder anderem Prior. `summary` gibt das Maximum und das kürzeste 68.3-%-Intervall sowie Median und 16. und 84. Perzentil.

```python
def posterior(d, w=None, v_r=V_R, v_p=V_P, sig_vp=SIG_VP, prior="log"):
    # p(H₀) ∝ p(H₀) Σ_i w_i N(v_r − H₀ d_i; ⟨v_p⟩, √(σ_vr² + σ_vp²)); die Stichproben tragen schon den Prior p(d) ∝ d²
    w = np.ones_like(d) if w is None else w
    s = np.hypot(SIG_VR, sig_vp)
    like = np.concatenate([np.exp(-0.5 * ((v_r - np.outer(Hb, d) - v_p) / s) ** 2) @ w for Hb in np.array_split(H, 40)])   # in Blöcken
    p = like / H if prior == "log" else like
    return p / np.trapz(p, H)


def summary(p):
    dh = H[1] - H[0]
    order = np.argsort(p)[::-1]
    inside = order[:np.searchsorted(np.cumsum(p[order]) * dh, 0.683) + 1]
    c = np.cumsum(p) * dh
    med, lo, hi = np.interp([0.5, 0.16, 0.84], c, H)
    return dict(map=H[np.argmax(p)], hpd=(H[inside].min(), H[inside].max()), med=med, q=(lo, hi))


def fmt(s):
    return (f"MAP {s['map']:.1f} +{s['hpd'][1] - s['map']:.1f}/−{s['map'] - s['hpd'][0]:.1f}; "
            f"Median {s['med']:.1f} +{s['q'][1] - s['med']:.1f}/−{s['med'] - s['q'][0]:.1f}")


print(f"Hubble-Geschwindigkeit v_H = v_r − ⟨v_p⟩ = {V_R - V_P:.0f} ± {np.hypot(SIG_VR, SIG_VP):.0f} km/s")
```

##### *1. Entfernung und Neigung aus der Welle*

Wie genau misst die Welle allein die Entfernung, und wie hängt die Entfernung an der Neigung der Bahn? Vorher vorhersagen: Liegen die schräg gesehenen Lösungen ($\cos\iota$ näher an 0) bei kleinerer oder größerer Entfernung?

Links stehen die Posterior der Entfernung für kleine und große Spins und die Verteilung des Priors (∝ d²). Rechts zeigt jede Stichprobe mit kleinen Spins ihre Entfernung gegen $\cos\theta_{JN}$. Der Code gibt für beide Spin-Annahmen Median und 90-%-Intervall der Entfernung, das 68-%-Intervall und die Korrelation zwischen Entfernung und $|\cos\theta_{JN}|$ aus.

**Hintergrund & Fachbegriffe:**
- **90-%-Intervall:** In den Katalogen von LIGO und Virgo üblich; hier vom 5. bis zum 95. Perzentil.
- **Prior ∝ d²:** gleich viele Quellen in jedem Volumenelement; er bevorzugt große Entfernungen, bevor die Daten etwas sagen.

```python
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 5.6))
bins = np.linspace(0, 80, 81)
ax.hist(PRIOR["luminosity_distance_Mpc"], bins=bins, density=True, color="0.8", label="Prior (∝ d²)")
for k, col, lab in [("low", "C0", "kleine Spins"), ("high", "C1", "große Spins")]:
    ax.hist(D[k], bins=bins, density=True, histtype="step", lw=2, color=col, label=f"Posterior, {lab}")
ax.set_xlabel("Leuchtkraftentfernung d [Mpc]")
ax.set_ylabel("Wahrscheinlichkeitsdichte [1/Mpc]")
ax.set_title("Wie weit ist GW170817 entfernt?", fontsize=10)
ax.legend(fontsize=9)
sc = ax2.scatter(COSI["low"], D["low"], s=3, c=np.abs(COSI["low"]), cmap="viridis", alpha=0.6)
ax2.set_xlabel(r"$\cos\theta_{JN}$")
ax2.set_ylabel("Leuchtkraftentfernung d [Mpc]")
ax2.set_title("Je schräger die Bahn, desto näher (kleine Spins)", fontsize=10)
plt.colorbar(sc, ax=ax2, label=r"$|\cos\theta_{JN}|$")
plt.tight_layout()
plt.show()

for k in ("low", "high"):
    p5, p16, p50, p84, p95 = np.percentile(D[k], [5, 16, 50, 84, 95])
    r = np.corrcoef(np.abs(COSI[k]), D[k])[0, 1]
    print(f"{'kleine' if k == 'low' else 'große':6s} Spins: d = {p50:.1f} Mpc, 68 %: {p16:.1f} bis {p84:.1f}, 90 %: {p5:.1f} bis {p95:.1f}; "
          f"Korrelation von d und |cos θ_JN|: {r:+.2f}; Anteil mit |cos θ_JN| > 0.9: {np.mean(np.abs(COSI[k]) > 0.9):.2f}")
```

* Links die Posterior der Entfernung, blau mit kleinen und orange mit großen Spins, grau der Prior ∝ d², der bis 75 Mpc ansteigt. Die Daten ziehen die Entfernung weit unter das, was der Prior bevorzugt: Mit kleinen Spins ist d = 40.0 Mpc (68 %: 30.8 bis 45.3, 90 %: 24.9 bis 47.3), wie im Katalog GWTC-1 (40 +7/−15). Mit großen Spins sind es 41.7 Mpc (29.1 bis 47.5).
* Rechts jede Stichprobe mit ihrer Entfernung und $\cos\theta_{JN}$: Sie liegen auf einem schmalen, schrägen Band, Entfernung und $|\cos\theta_{JN}|$ sind zu +0.98 korreliert. Von uns aus gesehen ist die Bahn gekippt ($\cos\theta_{JN}$ = −0.84 im Median); die Welle allein kann aber nicht sagen, ob um 10° oder 60°.
* Zur Vorhersage: bei kleinerer Entfernung. Eine schräg gesehene Bahn strahlt in unsere Richtung schwächer; damit die gemessene Amplitude herauskommt, muss sie näher sein. Die Entfernung reicht deshalb von 15 Mpc (cos ι = −0.3) bis 50 Mpc (cos ι = −1).

##### *2. Die Galaxie und ihre Geschwindigkeit*

Wo liegt GW170817 im Hubble-Diagramm, und welches H₀ ergibt sich ganz grob? Vorher vorhersagen: Welches H₀ folgt aus $v_H = 3017$ km/s und d = 40 Mpc?

Der Code zeichnet das Hubble-Diagramm der nächsten 100 Mpc mit den Geraden für Planck und SH0ES. GW170817 steht dort einmal mit der gemessenen Geschwindigkeit $v_r$ und einmal mit der Hubble-Geschwindigkeit $v_H = v_r - \langle v_p\rangle$, jeweils mit dem 68-%-Intervall der Entfernung (kleine Spins). Er gibt das grobe H₀ = $v_H/d$ für den Median und die Grenzen des Intervalls aus.

**Hintergrund & Fachbegriffe:**
- **Ruhesystem der Hintergrundstrahlung:** Die Sonne bewegt sich mit 370 km/s gegen die Hintergrundstrahlung; Geschwindigkeiten in der Kosmologie werden auf deren Ruhesystem umgerechnet.
- **Galaxiengruppe:** NGC 4993 gehört zu einer kleinen Gruppe; ihre mittlere Geschwindigkeit ist genauer als die einer einzelnen Galaxie, deren Bewegung in der Gruppe sonst dazukäme.

```python
p16, p50, p84 = np.percentile(D["low"], [16, 50, 84])
v_h = V_R - V_P
dd = np.linspace(0, 100, 2)
fig, ax = plt.subplots(figsize=(11, 6.5))
for H0, col, lab in [(PLANCK[0], "C3", "Planck, H₀ = 67.36"), (SH0ES[0], "C2", "SH0ES, H₀ = 73.04")]:
    ax.plot(dd, H0 * dd, color=col, lw=1.4, label=lab)
ax.errorbar(p50, V_R, xerr=[[p50 - p16], [p84 - p50]], yerr=SIG_VR, fmt="s", color="0.5", ms=8, capsize=4, label="GW170817, gemessene v_r")
ax.errorbar(p50, v_h, xerr=[[p50 - p16], [p84 - p50]], yerr=np.hypot(SIG_VR, SIG_VP), fmt="o", color="C0", ms=9, capsize=4,
            label="GW170817, Hubble-Geschwindigkeit v_H")
ax.set_xlim(0, 80)
ax.set_ylim(0, 6000)
ax.set_xlabel("Entfernung aus der Gravitationswelle [Mpc]")
ax.set_ylabel("Geschwindigkeit im Ruhesystem der Hintergrundstrahlung [km/s]")
ax.set_title("GW170817 im Hubble-Diagramm", fontsize=10)
ax.legend(fontsize=9)
plt.show()

print(f"v_H = {v_h:.0f} ± {np.hypot(SIG_VR, SIG_VP):.0f} km/s (Eigenbewegung {100 * V_P / V_R:.0f} % von v_r)")
print(f"grob H₀ = v_H/d: {v_h / p50:.1f} beim Median d = {p50:.1f} Mpc, {v_h / p84:.1f} bis {v_h / p16:.1f} für das 68-%-Intervall der Entfernung")
print(f"Planck verlangt d = {v_h / PLANCK[0]:.1f} Mpc, SH0ES d = {v_h / SH0ES[0]:.1f} Mpc")
```

* Das Hubble-Diagramm der nächsten 80 Mpc mit den Geraden von Planck (rot) und SH0ES (grün). Grau steht GW170817 mit der gemessenen Geschwindigkeit 3327 km/s, blau mit der Hubble-Geschwindigkeit 3017 ± 166 km/s. Die waagrechten Balken sind das 68-%-Intervall der Entfernung (30.8 bis 45.3 Mpc).
* Die Eigenbewegung macht 9 % der gemessenen Geschwindigkeit aus. Der Fehler der Entfernung ist viel größer als der der Geschwindigkeit; beide Geraden laufen durch den Fehlerbalken. Planck verlangt d = 44.8 Mpc, SH0ES 41.3 Mpc, beide innerhalb des Intervalls.
* Zur Vorhersage: 3017/40 = 75.5 km/s/Mpc. Das Intervall der Entfernung allein lässt 66.5 bis 97.8 zu; die genaue Posterior braucht noch den Fehler der Geschwindigkeit und den Prior (Schritt 3).

##### *3. Vom Prior der Entfernung zum Prior auf H₀*

Wie stark hängt das Ergebnis davon ab, was man vor der Messung über H₀ annimmt? Vorher vorhersagen: Welche Kennzahl ändert sich mit dem Prior stärker, das Maximum oder der Median?

Der Code vergleicht drei Wege:
- **naiv:** Jede Entfernungsstichprobe wird mit $H_0 = v_H/d_i$ umgerechnet und gezählt; der Fehler der Geschwindigkeit fehlt.
- **Standard:** die Analyse aus Theorie, Abschnitt 4, mit Prior $\propto 1/H_0$ und dem Fehler von 166 km/s.
- **flacher Prior:** dieselbe Analyse mit einem gleichmäßigen Prior auf H₀.

Für jeden gibt er Maximum und Median aus.

**Hintergrund & Fachbegriffe:**
- **Ererbter Prior:** Weil die Stichproben mit $p(d) \propto d^2$ gezogen sind und $H_0 \propto 1/d$, steckt in der naiven Umrechnung ein Prior, der zu kleinem H₀ hin wächst. Abbott et al. (2017) zeigen, dass er relativ zur reinen Likelihood der Welle wie $H_0^{-4}$ wirkt.
- **Maximum (MAP) und Median:** Bei schiefen Verteilungen liegen sie auseinander; das Maximum hängt weniger am Ausläufer.

```python
naive = np.histogram(v_h / D["low"], bins=H[::20], density=True)
POST = {"Standard (Prior ∝ 1/H₀)": posterior(D["low"]), "flacher Prior": posterior(D["low"], prior="flat")}
fig, ax = plt.subplots(figsize=(11, 6))
ax.stairs(naive[0], naive[1], color="0.5", lw=1.5, label="naiv: v_H/d_i, ohne Geschwindigkeitsfehler")
for (lab, p), col in zip(POST.items(), ["C0", "C1"]):
    ax.plot(H, p, color=col, lw=1.8, label=lab)
ax.set_xlim(40, 160)
ax.set_xlabel("H₀ [km/s/Mpc]")
ax.set_ylabel("Posterior-Dichte")
ax.set_title("Was der Prior aus derselben Messung macht (kleine Spins)", fontsize=10)
ax.legend(fontsize=9)
plt.show()

c = naive[1][:-1] + np.diff(naive[1]) / 2
print(f"naiv: Maximum bei {c[np.argmax(naive[0])]:.1f}, Median {np.median(v_h / D['low']):.1f}")
for lab, p in POST.items():
    print(f"{lab:24s}: {fmt(summary(p))}")
```

* Grau das naive Histogramm von $v_H/d_i$, blau die Standard-Posterior mit Prior ∝ 1/H₀, orange dieselbe mit flachem Prior. Alle drei steigen steil bei 60 an und haben einen langen Ausläufer bis über 150, der von den schräg gesehenen, nahen Lösungen kommt.
* Das naive Histogramm und die Standard-Posterior liegen übereinander: Das Histogramm ist die Standard-Analyse ohne den Fehler der Geschwindigkeit, der die Spitze nur verschmiert. Der Prior ∝ 1/H₀ steckt schon in den Stichproben (über p(d) ∝ d²). Maximum: naiv 67.5, Standard 68.5, flach 69.2; Median: 75.5, 75.7 und 79.3.
* Zur Vorhersage: Der Median verschiebt sich mit dem Prior (um 3.6), das Maximum kaum (um 0.7). Der flache Prior gibt dem Ausläufer mehr Gewicht. Deshalb nennen Abbott et al. das Maximum; Hotokezaka et al. geben für die Messung von 2017 den Median 74 +16/−8 an.

Welche Hubble-Konstante ergibt GW170817, und stimmt sie mit dem veröffentlichten Wert überein? Vorher vorhersagen: Lassen die Fehler Planck und SH0ES beide zu?

Der Code rechnet die Standard-Posterior (Theorie, Abschnitt 4) für die Stichproben mit kleinen und mit großen Spins. Er zeichnet beide neben den Bändern von Planck und SH0ES und gibt Maximum, kürzestes 68.3-%-Intervall und Median aus, dazu den Wert von Abbott et al. (2017) und die relative Wahrscheinlichkeitsdichte der Posterior bei 67.36 und 73.04.

```python
P4 = {k: posterior(D[k]) for k in ("low", "high")}
S4 = {k: summary(p) for k, p in P4.items()}
fig, ax = plt.subplots(figsize=(11, 6))
for k, col, lab in [("low", "C0", "kleine Spins"), ("high", "C1", "große Spins")]:
    ax.plot(H, P4[k], color=col, lw=1.8, label=f"GW170817, {lab}")
    lo, hi = S4[k]["hpd"]
    m = (H >= lo) & (H <= hi)
    ax.fill_between(H[m], 0, P4[k][m], color=col, alpha=0.2)
for (v, e), col, lab in [(PLANCK, "C3", "Planck"), (SH0ES, "C2", "SH0ES")]:
    ax.axvspan(v - e, v + e, color=col, alpha=0.3, label=lab)
ax.set_xlim(40, 140)
ax.set_xlabel("H₀ [km/s/Mpc]")
ax.set_ylabel("Posterior-Dichte")
ax.set_title("Die Hubble-Konstante aus einer Gravitationswelle (68.3 % schattiert)", fontsize=10)
ax.legend(fontsize=9)
plt.show()

for k in ("low", "high"):
    pr = np.interp(SH0ES[0], H, P4[k]) / np.interp(PLANCK[0], H, P4[k])
    print(f"{'kleine' if k == 'low' else 'große':6s} Spins: {fmt(S4[k])}; p(73.04)/p(67.36) = {pr:.2f}")
print(f"Abbott et al. 2017: MAP {ABBOTT17[0]} +{ABBOTT17[1]}/−{ABBOTT17[2]}")
```

##### *5. H₀ und die Neigung der Bahn*

Wie hängen H₀ und die Neigung zusammen, und welche Neigung verlangen Planck oder SH0ES? Vorher vorhersagen: Verlangt das größere H₀ von SH0ES eine stärker oder schwächer geneigte Bahn?

Der Code rechnet die gemeinsame Posterior von H₀ und $\cos\iota$: Jede Stichprobe trägt mit ihrem $\cos\theta_{JN}$ und dem Gewicht $\mathcal N(v_r - H_0 d_i; \langle v_p\rangle, \sigma)/H_0$ bei. Links stehen die Konturen, rechts die Posterior von $\cos\iota$, wenn man H₀ auf den Wert von Planck oder SH0ES festsetzt oder frei lässt. Er gibt die 68-%-Intervalle des Winkels aus.

**Hintergrund & Fachbegriffe:**
- **Bedingte Posterior:** die Verteilung eines Parameters, wenn ein anderer als bekannt gilt; hier $p(\cos\iota \mid H_0)$.
- **Winkel und Kosinus:** $\cos\iota = -0.9$ entspricht ι = 154°, also einer Bahn, die 26° von der Sicht genau von unten abweicht.

```python
ci = COSI["low"]
s = np.hypot(SIG_VR, SIG_VP)
h2 = np.linspace(40, 140, 101)
c2 = np.linspace(-1, -0.2, 41)
W = np.exp(-0.5 * ((V_R - V_P - np.outer(h2, D["low"])) / s) ** 2) / h2[:, None]   # (H₀, Stichprobe)
G = np.array([[W[i, (ci >= c2[j]) & (ci < c2[j + 1])].sum() for j in range(len(c2) - 1)] for i in range(len(h2))])
G /= G.sum()
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 5.8))
v = np.sort(G.ravel())[::-1]
cum = np.cumsum(v)
lev = [v[np.searchsorted(cum, f)] for f in (0.954, 0.683)]
cc = (c2[1:] + c2[:-1]) / 2
ax.contourf(h2, cc, G.T, levels=lev + [G.max()], colors=["lightsteelblue", "C0"], alpha=0.8)
for (val, e), col, lab in [(PLANCK, "C3", "Planck"), (SH0ES, "C2", "SH0ES")]:
    ax.axvspan(val - e, val + e, color=col, alpha=0.35, label=lab)
ax.plot([], [], color="C0", lw=8, label="68.3 %")
ax.plot([], [], color="lightsteelblue", lw=8, label="95.4 %")
ax.set_xlabel("H₀ [km/s/Mpc]")
ax.set_ylabel(r"$\cos\iota$")
ax.set_title("H₀ und Neigung gemeinsam (kleine Spins)", fontsize=10)
ax.legend(fontsize=9, loc="upper right")
ANG = {}
for lab, w_, col in [("H₀ frei", W.sum(axis=0), "C0"),
                     ("H₀ = 67.36 (Planck)", np.exp(-0.5 * ((V_R - V_P - PLANCK[0] * D["low"]) / s) ** 2), "C3"),
                     ("H₀ = 73.04 (SH0ES)", np.exp(-0.5 * ((V_R - V_P - SH0ES[0] * D["low"]) / s) ** 2), "C2")]:
    hist, e_ = np.histogram(ci, bins=np.linspace(-1, -0.2, 33), weights=w_, density=True)
    ax2.stairs(hist, e_, color=col, lw=1.8, label=lab)
    order = np.argsort(ci)
    cw = np.cumsum(w_[order]) / w_.sum()
    lo, hi = np.interp([0.16, 0.84], cw, ci[order])
    ANG[lab] = (lo, hi)
ax2.set_xlabel(r"$\cos\iota$")
ax2.set_ylabel("Posterior-Dichte")
ax2.set_title("Welche Neigung zu welchem H₀ gehört", fontsize=10)
ax2.legend(fontsize=9)
plt.tight_layout()
plt.show()

for lab, (lo, hi) in ANG.items():
    print(f"{lab:20s}: cos ι von {lo:.3f} bis {hi:.3f} (68 %), also ι von {np.degrees(np.arccos(hi)):.0f}° bis {np.degrees(np.arccos(lo)):.0f}°")
```

* Links die gemeinsame Posterior von H₀ und $\cos\iota$ (68.3 und 95.4 %): ein schräges Band. Kleine H₀ gehören zu Bahnen, die man fast von unten sieht (cos ι ≈ −1, große Entfernung), große H₀ zu schräg gesehenen. Rechts die Posterior von $\cos\iota$, frei und für H₀ fest bei Planck oder SH0ES.
* Ohne Festlegung liegt ι mit 68 % zwischen 129° und 163°. Mit H₀ = 67.36 wird es 152° bis 168°, mit H₀ = 73.04 145° bis 159°. Eine unabhängige Messung des Winkels kann also umgekehrt H₀ festlegen (Schritt 6).
* Zur Vorhersage: stärker geneigt. Ein größeres H₀ heißt bei fester Geschwindigkeit eine kleinere Entfernung, und die passt nur zu einer schrägeren Bahn.

##### *8. Wie viele Sirenen braucht man?*

Wie viele Ereignisse wie GW170817 braucht man für eine Genauigkeit von 2 %, mit der man die Hubble-Spannung entscheiden könnte? Vorher vorhersagen: Wenn ein Ereignis H₀ auf etwa 15 % misst, wie viele braucht man nach der Faustregel $\sigma/\sqrt N$ für 2 %?

Der Code tut so, als gäbe es N unabhängige Ereignisse, jedes mit derselben Form der Posterior wie GW170817, aber um einen Zufallswert verschoben. Die Verschiebung zieht er aus der Posterior selbst, um den Median herum, im Logarithmus. Die Posterioren der N Ereignisse multipliziert er. Für N = 1 bis 200 wiederholt er das 100-mal und gibt die mittlere halbe Breite des 68-%-Intervalls relativ zum Maximum an, einmal ohne und einmal mit der Information über den Jet. Zum Vergleich stehen die Geraden $\sigma_1/\sqrt N$.

**Hintergrund & Fachbegriffe:**
- **Grobe Annahme:** Fernere Ereignisse hätten größere Fehler in der Entfernung, nähere kleinere; die Ereignisse hier sind alle wie GW170817. Die Rechnung zeigt das Prinzip, keine Vorhersage für eine bestimmte Beobachtungskampagne.
- **Schiefe Posterioren:** Der Ausläufer zu großem H₀ in jedem einzelnen Ereignis fällt im Produkt weg; anfangs wird die Breite deshalb schneller kleiner als $1/\sqrt N$.

```python
rng = np.random.default_rng(8)
NS = np.array([1, 2, 3, 5, 10, 20, 50, 100, 200])
lnH = np.log(H)


def draw(p, n):
    c = np.cumsum(p)
    return np.interp(rng.random(n), c / c[-1], H)


FORECAST = {}
for lab, p in [("nur Gravitationswelle", P4["low"]), ("mit dem Jet", JET_RESULTS["low"][1])]:
    med = summary(p)["med"]
    logp = np.log(p + 1e-300)
    widths = []
    for n in NS:
        w_ = []
        for _ in range(100):
            shifts = np.log(draw(p, n) / med)
            tot = sum(np.interp(lnH - s_, lnH, logp, left=-700, right=-700) for s_ in shifts)
            q = np.exp(tot - tot.max())
            q /= np.trapz(q, H)
            sm = summary(q)
            w_.append((sm["hpd"][1] - sm["hpd"][0]) / 2 / sm["map"])
        widths.append(np.mean(w_))
    FORECAST[lab] = np.array(widths)

fig, ax = plt.subplots(figsize=(11, 6))
for (lab, w_), col in zip(FORECAST.items(), ["C0", "C3"]):
    ax.plot(NS, 100 * w_, "o-", color=col, ms=6, label=lab)
    ax.plot(NS, 100 * w_[0] / np.sqrt(NS), color=col, ls=":", lw=1.2, label=f"{lab}: σ₁/√N")
ax.axhline(2, color="k", lw=1)
ax.text(1.05, 2.15, "2 %", fontsize=10)
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Zahl der Ereignisse N")
ax.set_ylabel("relativer Fehler von H₀ (68 %, halbe Breite) [%]")
ax.set_title("Wie viele Ereignisse wie GW170817 für 2 %?", fontsize=10)
ax.legend(fontsize=9)
plt.show()

for lab, w_ in FORECAST.items():
    n2 = np.interp(np.log(0.02), np.log(w_[::-1]), np.log(NS[::-1]))
    print(f"{lab:22s}: " + ", ".join(f"N = {n}: {100 * x:.1f} %" for n, x in zip(NS, w_)) + f"; 2 % bei N ≈ {np.exp(n2):.0f}")
```

* Der relative Fehler von H₀ gegen die Zahl der Ereignisse, blau ohne und rot mit Information über den Winkel, gepunktet $\sigma_1/\sqrt N$. Ein Ereignis wie GW170817 allein gibt 18.7 % (halbe Breite des 68-%-Intervalls relativ zum Maximum), mit dem Jet 6.9 %.
* Ohne Winkel fällt der Fehler anfangs schneller als $1/\sqrt N$: Schon das zweite Ereignis schneidet den Ausläufer weg (10.4 %). Ab etwa 20 Ereignissen folgt er der Geraden. Für 2 % braucht man etwa 83 Ereignisse ohne und 12 mit Winkel. Hotokezaka et al. schätzen etwa 15 Ereignisse mit Radiobildern gegen 50 bis 100 ohne; Chen, Fishbach & Holz (2018) erwarten 2 % innerhalb von fünf Jahren.
* Zur Vorhersage: $(15/2)^2 \approx 56$ Ereignisse. Die Simulation braucht mehr, weil ein Ereignis wie GW170817 ohne Winkel eher 19 % als 15 % misst. Seit 2017 gab es aber kein zweites Ereignis mit Gegenstück, und die Annahme gleich guter Ereignisse ist optimistisch für fernere Quellen.

##### *9. Gegen die veröffentlichten Werte und die anderen Wege*

Wie passen die eigenen Werte zu den Veröffentlichungen, und wo liegen die Sirenen zwischen Planck und SH0ES? Vorher vorhersagen: Liegt der Wert mit dem Jet näher an Planck oder an SH0ES, und wie viele σ von SH0ES entfernt?

Der Code stellt die eigenen Werte (ohne und mit Jet) neben Abbott et al. (2017), Hotokezaka et al. (2019) und die beiden Ergebnisse der GWTC-3-Kosmologie. Dazu kommen Planck, SH0ES und zwei Werte aus den anderen Notebooks der Reihe. Für die eigenen Werte rechnet er den Abstand zu Planck und SH0ES. Weil die Posterioren schief sind, nimmt er dafür die halbe Breite auf der Seite, die zum jeweiligen Vergleichswert zeigt.

**Hintergrund & Fachbegriffe:**
- **Dunkle Sirenen:** Ereignisse ohne Gegenstück, für die man über alle Galaxien im Fehlerbereich mittelt (Galaxienkatalog) oder die Rotverschiebung aus der Form der Massenverteilung schließt (Massenverteilung).
- **Abstand bei schiefen Fehlern:** eine Näherung; genauer wäre die Wahrscheinlichkeit, dass beide Werte gleich sind, aus der vollen Verteilung.

```python
own = {"eigen: nur Gravitationswelle": (S4["low"]["map"], S4["low"]["hpd"][1] - S4["low"]["map"], S4["low"]["map"] - S4["low"]["hpd"][0]),
       "eigen: mit dem Jet": (JET_RESULTS["low"][2]["map"], JET_RESULTS["low"][2]["hpd"][1] - JET_RESULTS["low"][2]["map"],
                              JET_RESULTS["low"][2]["map"] - JET_RESULTS["low"][2]["hpd"][0])}
rows = [(k, *v, "C0") for k, v in own.items()]
rows += [("Abbott et al. 2017", *ABBOTT17, "k"), ("Hotokezaka et al. 2019 (Jet)", *HOTOKEZAKA19, "k")]
rows += [(k, *v, "k") for k, v in GWTC3.items()]
rows += [(k, *v, "0.5") for k, v in OTHERS.items()]
rows += [("Planck 2018", PLANCK[0], PLANCK[1], PLANCK[1], "C3"), ("SH0ES", SH0ES[0], SH0ES[1], SH0ES[1], "C2")]
fig, ax = plt.subplots(figsize=(12, 6.5))
for i, (lab, v, ep, em, col) in enumerate(rows):
    ax.errorbar(v, -i, xerr=[[em], [ep]], fmt="o", color=col, ms=7, capsize=4)
ax.set_yticks(-np.arange(len(rows)))
ax.set_yticklabels([r[0] for r in rows], fontsize=9.5)
ax.axvspan(PLANCK[0] - PLANCK[1], PLANCK[0] + PLANCK[1], color="C3", alpha=0.15)
ax.axvspan(SH0ES[0] - SH0ES[1], SH0ES[0] + SH0ES[1], color="C2", alpha=0.15)
ax.set_xlabel("H₀ [km/s/Mpc]")
ax.set_title("Standardsirenen zwischen frühem Universum und Entfernungsleiter", fontsize=10)
plt.tight_layout()
plt.show()

for lab, (v, ep, em) in own.items():
    to_p = (v - PLANCK[0]) / np.hypot(em if v > PLANCK[0] else ep, PLANCK[1])
    to_s = (v - SH0ES[0]) / np.hypot(ep if v < SH0ES[0] else em, SH0ES[1])
    print(f"{lab:30s}: H₀ = {v:.1f} +{ep:.1f}/−{em:.1f}; gegen Planck {to_p:+.1f} σ, gegen SH0ES {to_s:+.1f} σ")
```

* Alle Werte untereinander: blau die eigenen, schwarz die veröffentlichten Sirenen, grau zwei Wege aus den anderen Notebooks, rot Planck und grün SH0ES. Alle Sirenen liegen zwischen 67 und 70, mit Fehlern von 4.5 bis 18 km/s/Mpc.
* Ohne Jet ist der eigene Wert 0.1σ von Planck und 0.3σ von SH0ES entfernt, er entscheidet nichts. Mit Jet sind es −0.1σ und −1.2σ, also eine leichte Bevorzugung von Planck. Die GWTC-3-Kosmologie mit 47 Ereignissen kommt auf 68 +12/−8 (Massenverteilung, mit GW170817) und 68 +8/−6 (Galaxienkatalog); die dunklen Sirenen sind noch weniger genau als das eine helle Ereignis mit Jet.
* Zur Vorhersage: näher an Planck, aber nur 1.2σ von SH0ES. Für eine Entscheidung mit 3σ müsste der Fehler unter etwa 2 km/s/Mpc fallen, also rund 3 %, und das verlangt nach Schritt 8 etwa fünf bis zehn Ereignisse mit Jet.

#### **Wiederholung**

Erst selbst antworten, dann aufklappen.

<details><summary>1. Warum braucht eine Standardsirene keine Eichung an Cepheiden?</summary>

Die allgemeine Relativitätstheorie sagt voraus, wie stark eine Quelle mit gegebener Chirpmasse abstrahlt. Die Chirpmasse folgt aus dem Anstieg der Frequenz, die Entfernung dann aus der Amplitude. Es gibt keine Sprosse dazwischen, die geeicht werden müsste.
</details>

<details><summary>2. Warum misst die Welle die Entfernung nur auf etwa 15 bis 20 %?</summary>

Die Amplitude hängt außer von der Entfernung auch von der Neigung der Bahn ab. Eine schräg gesehene, nahe Quelle sieht fast aus wie eine von unten gesehene, ferne. Diese Entartung, nicht das Rauschen, bestimmt den Fehler.
</details>

<details><summary>3. Was trägt das elektromagnetische Gegenstück bei?</summary>

Es zeigt die Wirtsgalaxie und damit die Rotverschiebung, die die Welle allein nicht misst. Erst Entfernung und Geschwindigkeit zusammen geben H₀.
</details>

<details><summary>4. Warum muss man die Eigenbewegung der Galaxie abziehen, und wie viel macht das aus?</summary>

Die gemessene Geschwindigkeit ist $v_r = H_0 d + v_p$. Bei 40 Mpc ist $v_p$ ≈ 310 km/s etwa 10 % von $v_r$; ohne Korrektur käme H₀ = 74 statt 67 heraus.
</details>

<details><summary>5. Warum geben Abbott et al. das Maximum der Posterior an und nicht den Median?</summary>

Die Posterior ist schief, mit einem langen Ausläufer zu großem H₀. Der Median hängt stark davon ab, wie der Prior diesen Ausläufer gewichtet (75.7 oder 79.3), das Maximum kaum (68.5 oder 69.2).
</details>

<details><summary>6. Welchen Prior auf H₀ erbt man aus p(d) ∝ d²?</summary>

Mit $H_0 = v_H/d$ und $p(d) \propto d^2$ wird aus der Likelihood der Welle eine Posterior mit einem Prior, der wie $H_0^{-4}$ wirkt. Das naive Histogramm von $v_H/d_i$ ist deshalb dasselbe wie die Standard-Analyse ohne Fehler der Geschwindigkeit.
</details>

<details><summary>7. Warum spielt die Selektion bei GW170817 keine Rolle?</summary>

Ob ein Ereignis entdeckt wird, hängt bei 40 Mpc nur von seiner Entfernung ab, nicht von H₀. Weil der Prior auf die Entfernung gesetzt ist, hängt die Normierung $N_s$ dann nicht von H₀ ab. Bei fernen Ereignissen und Rotverschiebungs-Priors ist das anders.
</details>

<details><summary>8. Wie misst ein Jet die Neigung der Bahn?</summary>

Der Jet läuft entlang der Drehachse. Sieht man ihn unter kleinem Winkel, scheint sein Schwerpunkt mit Überlichtgeschwindigkeit zu wandern; aus $\beta_\text{app} \approx 4$ folgt ein Beobachtungswinkel von etwa 20°. Weil $\beta_\text{app}$ mit der Entfernung skaliert, legt die Messung das Produkt aus Winkel und Entfernung fest.
</details>

<details><summary>9. Warum verlangt ein größeres H₀ eine schräger gesehene Bahn?</summary>

Bei fester Geschwindigkeit heißt größeres H₀ kleinere Entfernung. Dieselbe gemessene Amplitude aus kleinerer Entfernung passt nur, wenn die Bahn schräger gesehen wird und in unsere Richtung schwächer strahlt.
</details>

<details><summary>10. Was ist mit dem Jet der größte Fehler?</summary>

Die Eigenbewegung: Ihre Unsicherheit von 150 km/s macht 5 % von 3000 km/s aus. Ein engeres Band für den Winkel hilft kaum, ein genaueres Geschwindigkeitsfeld schon.
</details>

<details><summary>11. Warum fällt der Fehler mit den ersten Ereignissen schneller als 1/√N?</summary>

Jede einzelne Posterior hat einen langen Ausläufer. Im Produkt zweier Posterioren bleibt nur, was beide zulassen, und die Ausläufer, die an verschiedenen Stellen liegen, verschwinden. Erst wenn das Produkt gaußisch ist, gilt $1/\sqrt N$.
</details>

<details><summary>12. Was sind dunkle Sirenen, und warum sind sie weniger genau?</summary>

Ereignisse ohne Gegenstück. Man mittelt über alle Galaxien im Fehlerbereich am Himmel (Galaxienkatalog) oder schließt aus der Massenverteilung der schwarzen Löcher auf die Rotverschiebung. Beides verwischt die Rotverschiebung stark; GWTC-3 erreicht mit 47 Ereignissen etwa 10 %.
</details>

#### **Nächste Schritte**

So würde eine Analyse in der Forschung weitergehen:
- **Die volle Modellierung des Jets:** Lichtkurve und Bewegung des Schwerpunkts mit einem Modell des Jets anpassen (Hotokezaka et al. 2019) und die Posterior von Winkel und Entfernung gemeinsam mit den Stichproben der Welle verwenden, statt eines harten Bandes.
- **Das Geschwindigkeitsfeld verbessern:** Mit neueren Rekonstruktionen (Cosmicflows-4) und Entfernungen der Wirtsgalaxie aus anderen Methoden (Fundamentalebene, Oberflächenhelligkeits-Fluktuationen) den größten Fehler mit dem Jet verkleinern.
- **Dunkle Sirenen:** Für die Ereignisse ohne Gegenstück die Galaxien im Fehlerbereich aus einem Katalog (GLADE+) gewichten, die Unvollständigkeit des Katalogs berücksichtigen und die Selektion von H₀ abhängig rechnen; oder die Rotverschiebung aus Merkmalen der Massenverteilung schwarzer Löcher schließen.
- **Mehr Ereignisse:** O4 und O5 von LIGO, Virgo und KAGRA sollten weitere Verschmelzungen von Neutronensternen mit Gegenstück finden; Detektoren der nächsten Generation (Einstein Telescope, Cosmic Explorer) Tausende.
- **Kosmologie bei großem z:** Ferne Sirenen messen $d_L(z)$ bis z ≈ 1 und mehr und prüfen dort dasselbe wie die BAO, ohne Lineal und ohne Kerze.
