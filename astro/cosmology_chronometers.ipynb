# Kosmische Chronometer

Das Notebook misst die Expansionsrate H(z) mit kosmischen Chronometern, also dem Altersunterschied passiver Galaxien. Grundlage sind 15 Messungen von Moresco et al. von z = 0.18 bis 1.97, mit der Kovarianz der Systematik nach Moresco et al. (2020). Daraus folgen:
- H₀ aus ΛCDM und, ohne Modell, aus einem Gauß-Prozess,
- der Beginn der beschleunigten Expansion,
- zusammen mit den BAO von DESI und den Supernovae von Pantheon+ der Schallhorizont $r_d$ und die absolute Helligkeit der Supernovae, beides ohne Hintergrundstrahlung und ohne Cepheiden.

Wie misst man die Expansion des Universums mit einer Uhr statt mit einem Lineal, und wie lang ist das Lineal der BAO, wenn man es mit dieser Uhr eicht?

#### **Hintergrund**

Fast alle Messungen der Expansion benutzen Entfernungen: Supernovae als Kerzen bekannter Helligkeit, die BAO als Lineal bekannter Länge. Beide brauchen eine Eichung, die Kerzen durch Cepheiden, das Lineal durch die Physik vor der Rekombination. Jimenez & Loeb schlugen 2002 einen dritten Weg vor: Man misst nicht, wie weit etwas entfernt ist, sondern wie viel Zeit zwischen zwei Rotverschiebungen vergangen ist. Die Expansionsrate ist

$$H(z) = -\frac{1}{1+z}\,\frac{dz}{dt} ,$$

und $dt$ lässt sich am Alter von Galaxien ablesen, wenn man Galaxien findet, die alle gleichzeitig entstanden sind und seither nur altern. Das sind die massivsten **passiven Galaxien**: Sie haben ihre Sterne vor über zehn Milliarden Jahren in kurzer Zeit gebildet und seither keine neuen. Ihre Spektren wirken wie Uhren, die zur selben Zeit gestellt wurden.

Was Forschende damit tun:
- Sie wählen aus großen spektroskopischen Durchmusterungen (SDSS, BOSS, zCOSMOS) Tausende passive Galaxien aus und messen die Stärke des Sprungs bei 4000 Å im Spektrum (D4000). Er wächst mit dem Alter der Sterne.
- Sie eichen D4000 an Modellen von Sternpopulationen und rechnen die Differenz zwischen benachbarten Rotverschiebungen in $dt$ um.
- Den größten systematischen Fehler, die Wahl des Modells der Sternpopulationen, haben Moresco et al. (2020) als Kovarianzmatrix angegeben.

Die Chronometer sind weniger genau als Supernovae oder BAO, aber unabhängig von beiden: Sie brauchen weder Cepheiden noch die Physik des frühen Universums. Damit können sie zwischen den beiden Leitern der Hubble-Spannung vermitteln.

#### **Theorie**

##### *1. Die Uhr*

Aus $1 + z = a_0/a(t)$ folgt $dz/dt = -(1+z)\,H(z)$, also $H(z) = -\frac{1}{1+z}\frac{dz}{dt}$. Misst man das Alter $t_\text{gal}$ passiver Galaxien bei zwei nahen Rotverschiebungen, und sind alle Galaxien gleichzeitig entstanden, dann ist $\Delta t_\text{gal} = \Delta t$, und

$$H(z) \approx -\frac{1}{1+z}\,\frac{\Delta z}{\Delta t} .$$

Nur der Unterschied der Alter zählt, nicht ihr absoluter Wert; ein gemeinsamer Fehler aller Alter fällt heraus. Als Altersmaß dient der Sprung bei 4000 Å: $D4000 = A(Z, \text{SFH})\,t_\text{gal} + B$, mit einer Steigung A, die von der Metallizität Z und der Sternentstehungsgeschichte abhängt. Dann ist

$$H(z) = -\frac{A}{1+z}\,\frac{dz}{d\,D4000} .$$

Eine Größenordnung: Bei z = 0.5 ist $1/H \approx 11$ Gyr. Zwischen z = 0.45 und 0.55 vergehen dann etwa 0.73 Gyr. Für H auf 5 % muss dieser Unterschied auf 40 Millionen Jahre genau bekannt sein.

##### *2. Die Fehler und ihre Kovarianz*

Moresco et al. (2020) trennen die Fehler von $H(z)$ in Teile:
- **statistisch und Metallizität:** von Rotverschiebung zu Rotverschiebung unabhängig, also diagonal;
- **Modell der Sternpopulationen (SPS):** Welches Modell man für A nimmt, verschiebt alle Messungen in dieselbe Richtung, im Mittel um 4.5 %, wenn man das am stärksten abweichende Modell weglässt;
- **Anfangsmassenfunktion (IMF):** ebenso gemeinsam, aber unter 0.5 %.

Gemeinsame Fehler gehen als voll korrelierte Terme in die Kovarianz ein:

$$C_{ij} = \sigma_i^2\,\delta_{ij} + \sum_{X \in \{\text{SPS}, \text{IMF}\}} \eta_X(z_i)\,H(z_i)\;\eta_X(z_j)\,H(z_j) ,$$

mit dem relativen Fehler $\eta_X(z)$ aus ihrer Tabelle. Ein solcher Term lässt sich in einem Fit nicht herausmitteln: Er verschiebt alle Punkte zugleich.

##### *3. ΛCDM und die Verzögerung*

In flachem ΛCDM (Strahlung vernachlässigt) ist $H(z) = H_0\sqrt{\Omega_m (1+z)^3 + 1 - \Omega_m}$. Der **Verzögerungsparameter**

$$q(z) = -\frac{\ddot a\, a}{\dot a^2} = (1+z)\,\frac{H'(z)}{H(z)} - 1$$

ist positiv, solange die Expansion gebremst wird, und negativ, wenn sie beschleunigt. In ΛCDM wechselt er das Vorzeichen bei $z_t = [2(1-\Omega_m)/\Omega_m]^{1/3} - 1$, für $\Omega_m = 0.315$ bei $z_t = 0.63$.

##### *4. Ein Gauß-Prozess*

Ein **Gauß-Prozess** ist eine Wahrscheinlichkeitsverteilung über glatte Funktionen. Je zwei Funktionswerte sind gemeinsam normalverteilt, mit der Kovarianz $k(z, z') = \sigma_f^2 \exp[-(z - z')^2/(2\ell^2)]$. Die Länge $\ell$ sagt, über welche Strecke in z die Funktion glatt ist, $\sigma_f$, wie weit sie schwanken darf. Mit Messwerten $\mathbf y$ und ihrer Kovarianz C ist die Vorhersage an einer neuen Stelle $z_\star$

$$\mu(z_\star) = \mathbf k_\star^T (K + C)^{-1} \mathbf y, \qquad \sigma^2(z_\star) = k(z_\star, z_\star) - \mathbf k_\star^T (K + C)^{-1} \mathbf k_\star .$$

$\sigma_f$ und $\ell$ werden so gewählt, dass die **Randwahrscheinlichkeit** der Daten am größten ist. Für die Ableitung $H'(z)$ ersetzt man k durch seine Ableitung nach z; das gibt $q(z)$ ohne ein kosmologisches Modell. Der Preis: Bei $z_\star = 0$, außerhalb der Daten (z ≥ 0.18), hängt die Extrapolation von der Wahl des Kerns ab.

##### *5. Die Uhr eicht Lineal und Kerze*

Die BAO messen $D_H(z)/r_d = c/[H(z)\,r_d]$. Mit $H(z)$ aus den Chronometern bei derselben Rotverschiebung folgt die Länge des Lineals direkt:

$$r_d = \frac{c}{H(z)\,[D_H/r_d](z)} ,$$

ohne Annahme über das frühe Universum. Ebenso legen die Supernovae die Form von $D_L(z)$ fest; die Chronometer setzen den Maßstab, und daraus folgt die absolute Helligkeit $M_B$, ohne Cepheiden. Die Hubble-Spannung lässt sich so von einer dritten Seite prüfen: Ist das Lineal kürzer als 147 Mpc oder die Kerze heller als −19.25 mag?

**Fachbegriffe:**
- **Passive Galaxie:** eine Galaxie ohne nennenswerte Sternentstehung, deren Licht von alten Sternen kommt; im Spektrum rot, mit starkem Sprung bei 4000 Å.
- **D4000:** das Verhältnis der mittleren Flussdichte knapp über und knapp unter 4000 Å; der Sprung entsteht durch Absorptionslinien von Metallen in alten Sternen.
- **Modell der Sternpopulationen (SPS):** berechnet das Spektrum einer Sternpopulation gegebenen Alters und gegebener Metallizität (Bruzual & Charlot 2003, Maraston & Strömbäck 2011 u. a.).

```python
import os
import urllib.request
from pathlib import Path

ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "astropy").mkdir(parents=True, exist_ok=True)
os.environ["XDG_CACHE_HOME"] = str(ASTRODATA)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import optimize, stats

C_KMS, OMEGA_R = 299792.458, 4.18e-5                     # c [km/s]; ω_r von Photonen und Neutrinos
CC_URL = "https://gitlab.com/mmoresco/CCcovariance/-/raw/master/data/"
DESI_URL = "https://raw.githubusercontent.com/CobayaSampler/bao_data/master/desi_bao_dr2/"
PANTHEON_URL = "https://raw.githubusercontent.com/PantheonPlusSH0ES/DataRelease/main/Pantheon%2B_Data/4_DISTANCES_AND_COVAR/"
CC_DIR, DESI_DIR, PANTHEON = ASTRODATA / "chronometers", ASTRODATA / "desi", ASTRODATA / "pantheon"
PLANCK = dict(H0=(67.36, 0.54), Om=(0.3153, 0.0073), rd=(147.09, 0.26))
H0_SH0ES = (73.04, 1.04)
M_B_CEPHEID = (-19.249, 0.030)                            # Das Hubble-Diagramm (Cepheiden, SH0ES)
HRD_BAO = (101.55, 0.74)                                  # h·r_d aus den BAO allein (Baryonische akustische Oszillationen)
RD_HJV = (142.8, 3.7)                                     # Heavens, Jimenez & Verde 2014 (Uhren und BAO)


def download(url, path, check, timeout=300):
    path = Path(path)
    if not path.exists():
        path.parent.mkdir(parents=True, exist_ok=True)
        with urllib.request.urlopen(url, timeout=timeout) as r:
            data = r.read()
        if not check(data):
            raise OSError(f"{path.name}: unerwarteter Inhalt")
        path.write_bytes(data)
    return path


cc = np.genfromtxt(download(CC_URL + "HzTable_MM_BC03.dat", CC_DIR / "HzTable_MM_BC03.dat", lambda d: d.startswith(b"# z")),
                   delimiter=",", comments="#", dtype=None, encoding="utf-8")
Z_CC = np.array([r[0] for r in cc]); H_CC = np.array([r[1] for r in cc]); E_CC = np.array([r[2] for r in cc])
REF_CC = np.array([r[5] for r in cc])
eta = np.genfromtxt(download(CC_URL + "data_MM20.dat", CC_DIR / "data_MM20.dat", lambda d: d.startswith(b"# z")), comments="#")
ETA = {name: np.interp(Z_CC, eta[:, 0], eta[:, col]) / 100 for name, col in [("IMF", 1), ("Sternbibliothek", 2), ("SPS", 3), ("SPS ohne Ausreißer", 4)]}
COV_PARTS = {"statistisch + Metallizität": np.diag(E_CC**2),
             "IMF": np.outer(H_CC * ETA["IMF"], H_CC * ETA["IMF"]),
             "Modell (SPS)": np.outer(H_CC * ETA["SPS ohne Ausreißer"], H_CC * ETA["SPS ohne Ausreißer"])}
COV_CC = sum(COV_PARTS.values())                          # wie Moresco et al. 2020 (CC_covariance.ipynb)
IC_CC, IC_CC_DIAG = np.linalg.inv(COV_CC), np.diag(1 / E_CC**2)

name = "desi_gaussian_bao_ALL_GCcomb_"
tab = np.loadtxt(download(DESI_URL + name + "mean.txt", DESI_DIR / (name + "mean.txt"), lambda d: d.startswith(b"# [z]")), dtype=str)
Z_BAO, Y_BAO, KIND = tab[:, 0].astype(float), tab[:, 1].astype(float), tab[:, 2]
COV_BAO = np.loadtxt(download(DESI_URL + name + "cov.txt", DESI_DIR / (name + "cov.txt"), lambda d: d[:1].isdigit()))
IC_BAO = np.linalg.inv(COV_BAO)

SN = pd.read_csv(download(PANTHEON_URL + "Pantheon%2BSH0ES.dat", PANTHEON / "Pantheon+SH0ES.dat", lambda d: d.startswith(b"CID")), sep=r"\s+")
N = len(SN)
COV = np.loadtxt(download(PANTHEON_URL + "Pantheon%2BSH0ES_STAT%2BSYS.cov", PANTHEON / "Pantheon+SH0ES_STAT+SYS.cov",
                          lambda d: d.split(b"\n", 1)[0].strip() == b"1701"), skiprows=1).reshape(N, N)
COV = (COV + COV.T) / 2
use = SN["zHD"].values > 0.01
Z_SN, ZHEL_SN, M_SN = SN["zHD"].values[use], SN["zHEL"].values[use], SN["m_b_corr"].values[use]
IC_SN = np.linalg.inv(COV[np.ix_(use, use)])
IC_SN_SUM = IC_SN.sum()

print(f"Chronometer: {len(Z_CC)} Messungen bei z = {Z_CC.min()} bis {Z_CC.max()}, Fehler {100 * (E_CC / H_CC).min():.0f} bis {100 * (E_CC / H_CC).max():.0f} %")
print(f"DESI: {len(Y_BAO)} BAO-Werte; Pantheon+: {use.sum()} Supernovae mit z > 0.01")
```

**Werkzeuge: Expansion, Entfernungen, χ², Fits und der Gauß-Prozess** (Theorie, Abschnitte 3 bis 5).
- `E` ist H(z)/H₀ in flachem ΛCDM mit Strahlung, `comoving` die mitbewegte Entfernung auf einem feinen Gitter.
- Die χ²-Funktionen vergleichen mit Chronometern, BAO ($r_d$ frei) und Supernovae ($M_B$ analytisch herausgerechnet, auf Wunsch ausgegeben).
- `fit` sucht das Minimum, `profile` das Profil-χ² eines Parameters.
- `gp_fit` wählt $\sigma_f$ und $\ell$ des Gauß-Prozesses, `gp_predict` gibt Mittel und Kovarianz von H(z) und H'(z) auf einem Gitter.

```python
ZG = np.linspace(0, 2.5, 2501)


def E(z, H0, Om):
    o_r = OMEGA_R / (H0 / 100) ** 2
    return np.sqrt(Om * (1 + z) ** 3 + o_r * (1 + z) ** 4 + 1 - Om - o_r)


def comoving(H0, Om):
    f = 1 / E(ZG, H0, Om)
    return C_KMS / H0 * np.concatenate([[0.0], np.cumsum((f[1:] + f[:-1]) / 2 * np.diff(ZG))])


def chi2_cc(H0, Om, ic=None):
    r = H_CC - H0 * E(Z_CC, H0, Om)
    return r @ (IC_CC if ic is None else ic) @ r


def chi2_bao(H0, Om, rd):
    DM = np.interp(Z_BAO, ZG, comoving(H0, Om))
    DH = C_KMS / (H0 * E(Z_BAO, H0, Om))
    DV = (Z_BAO * DM**2 * DH) ** (1 / 3)
    r = Y_BAO - np.select([KIND == "DM_over_rs", KIND == "DH_over_rs"], [DM, DH], DV) / rd
    return r @ IC_BAO @ r


def chi2_sn(H0, Om, return_M=False):
    r = M_SN - 5 * np.log10((1 + ZHEL_SN) * np.interp(Z_SN, ZG, comoving(H0, Om))) - 25
    a = IC_SN @ r
    chi = r @ a - a.sum() ** 2 / IC_SN_SUM
    return (chi, a.sum() / IC_SN_SUM) if return_M else chi


def total(p, data):
    H0, Om, rd = p
    if not (20 < H0 < 150 and 0.01 < Om < 1.0 and 50 < rd < 300):
        return 1e10
    return (chi2_cc(H0, Om) if "CC" in data else 0) + (chi2_bao(H0, Om, rd) if "BAO" in data else 0) \
        + (chi2_sn(H0, Om) if "SN" in data else 0)


def fit(data, start=(68.0, 0.3, 147.0)):
    best = None
    p = np.array(start)
    for _ in range(3):
        r = optimize.minimize(total, p, args=(data,), method="Nelder-Mead", options=dict(xatol=1e-7, fatol=1e-8, maxiter=20000))
        p = r.x
        if best is None or r.fun < best.fun - 1e-8:
            best = r
    return best


def profile(data, index, values, best):
    # χ² minimiert über die übrigen Parameter, für jeden Wert des Parameters index, relativ zum Minimum
    free = [i for i in range(3) if i != index]
    out = []
    for v in values:
        def f(q):
            p = np.empty(3)
            p[index], p[free] = v, q
            return total(p, data)
        out.append(optimize.minimize(f, best.x[free], method="Nelder-Mead", options=dict(xatol=1e-6, fatol=1e-7)).fun)
    return np.array(out) - best.fun


def interval(values, dchi2, level=1.0):
    ok = values[dchi2 <= level]
    return ok.min(), ok.max()


def kern(a, b, sf, l):
    return sf**2 * np.exp(-((a[:, None] - b[None, :]) ** 2) / (2 * l**2))


def gp_fit(cov):
    # σ_f und ℓ mit der größten Randwahrscheinlichkeit (Mittelwert null)
    def nlml(th):
        sf, l = np.exp(th)
        L = np.linalg.cholesky(kern(Z_CC, Z_CC, sf, l) + cov)
        alpha = np.linalg.solve(L.T, np.linalg.solve(L, H_CC))
        return 0.5 * H_CC @ alpha + np.log(np.diag(L)).sum()
    r = optimize.minimize(nlml, np.log([150.0, 2.0]), method="Nelder-Mead")
    return np.exp(r.x)


def gp_predict(zs, cov, sf, l):
    # gemeinsames Mittel und Kovarianz von (H(zs), H'(zs)) unter dem Gauß-Prozess
    Ki = np.linalg.inv(kern(Z_CC, Z_CC, sf, l) + cov)
    d = zs[:, None] - Z_CC[None, :]
    k0 = kern(zs, Z_CC, sf, l)
    k1 = -d / l**2 * k0                                    # Kovarianz von H'(z*) und H(z_i)
    dd = zs[:, None] - zs[None, :]
    kss = kern(zs, zs, sf, l)
    K11 = kss * (1 - dd**2 / l**2) / l**2                  # Kovarianz von H'(z*) und H'(z*')
    K01 = dd / l**2 * kss                                  # Kovarianz von H(z*) und H'(z*')
    A = np.vstack([k0, k1])
    mean = A @ Ki @ H_CC
    prior = np.block([[kss, K01], [K01.T, K11]])
    return mean, prior - A @ Ki @ A.T


print(f"Kovarianz: Determinante {np.linalg.det(COV_CC):.3e}, Bedingungszahl {np.linalg.cond(COV_CC):.0f}")
```

##### *1. Die Messungen*

Wie sehen die 15 Messungen von H(z) aus, und passen sie eher zu Planck oder zu SH0ES? Vorher vorhersagen: Wächst oder fällt $\dot a = H/(1+z)$ mit der Rotverschiebung, wenn die Expansion heute beschleunigt?

Der Code zeichnet links H(z) mit den Fehlern aus der Tabelle, gefärbt nach der Veröffentlichung. Dazu kommen zwei ΛCDM-Kurven: Planck (H₀ = 67.36, $\Omega_m$ = 0.315) und dieselbe Form mit H₀ = 73.04. Rechts steht $\dot a/a_0 = H(z)/(1+z)$, die Geschwindigkeit, mit der der Skalenfaktor wächst. Er gibt die Messwerte und die Abweichung von beiden Kurven in Fehlereinheiten aus.

**Hintergrund & Fachbegriffe:**
- **$\dot a$:** die zeitliche Ableitung des Skalenfaktors; nimmt sie mit der Zeit zu (also mit z ab), beschleunigt die Expansion.
- **χ² mit Diagonalfehlern:** hier nur zur Orientierung; die korrelierten Fehler kommen in Schritt 2 dazu.

```python
zz = np.linspace(0, 2.1, 300)
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 6))
for ref, col in zip(dict.fromkeys(REF_CC), ["C0", "C1", "C2"]):
    m = REF_CC == ref
    for a_, y in [(ax, H_CC), (ax2, H_CC / (1 + Z_CC))]:
        a_.errorbar(Z_CC[m], y[m], yerr=E_CC[m] / (1 if a_ is ax else 1 + Z_CC[m]), fmt="o", color=col, ms=7, capsize=3, label=ref)
for H0, col, lab in [(PLANCK["H0"][0], "k", "Planck-ΛCDM (H₀ = 67.36)"), (H0_SH0ES[0], "C3", "gleiche Form, H₀ = 73.04")]:
    ax.plot(zz, H0 * E(zz, H0, PLANCK["Om"][0]), color=col, lw=1.4, label=lab)
    ax2.plot(zz, H0 * E(zz, H0, PLANCK["Om"][0]) / (1 + zz), color=col, lw=1.4, label=lab)
ax.set_xlabel("Rotverschiebung z")
ax.set_ylabel("H(z) [km/s/Mpc]")
ax.set_title("Expansionsrate aus dem Alter passiver Galaxien", fontsize=10)
ax.legend(fontsize=9)
ax2.set_xlabel("Rotverschiebung z")
ax2.set_ylabel(r"$H(z)/(1+z) = \dot a/a_0$ [km/s/Mpc]")
ax2.set_title("Wächst der Skalenfaktor heute schneller als früher?", fontsize=10)
ax2.legend(fontsize=9)
plt.tight_layout()
plt.show()

for H0, lab in [(PLANCK["H0"][0], "Planck"), (H0_SH0ES[0], "H₀ = 73.04")]:
    r = (H_CC - H0 * E(Z_CC, H0, PLANCK["Om"][0])) / E_CC
    print(f"gegen {lab:10s}: χ² (nur Diagonale) = {np.sum(r**2):.1f} bei {len(r)} Werten, mittlere Abweichung {r.mean():+.2f} σ")
for zi, h, e, ref in zip(Z_CC, H_CC, E_CC, REF_CC):
    print(f"z = {zi:.4f}: H = {h:6.1f} ± {e:4.1f} ({100 * e / h:4.1f} %)  {ref}")
```

* Links die 15 Messungen von H(z), gefärbt nach Veröffentlichung, mit Planck-ΛCDM (schwarz) und derselben Form mit H₀ = 73.04 (rot). Die Fehler liegen zwischen 5 % (z = 0.18) und 27 % (z = 1.97). Die meisten Punkte liegen zwischen den beiden Kurven oder näher an Planck.
* Mit den Diagonalfehlern ist χ² = 6.2 gegen Planck und 14.8 gegen H₀ = 73, bei 15 Werten; im Mittel liegen die Punkte +0.11σ über Planck und −0.61σ unter der anderen Kurve. Dass χ² gegen Planck nur 6.2 ist, zeigt, dass die Tabellenfehler eher vorsichtig sind.
* Rechts $\dot a = H/(1+z)$: Er fällt von z = 2 bis etwa 0.6 und steigt danach wieder, in beiden Kurven. Zur Vorhersage: Bei beschleunigter Expansion wächst $\dot a$ mit der Zeit, fällt also mit z; das Minimum markiert den Beginn der Beschleunigung. Die Punkte deuten das Minimum an, sind aber zu ungenau, um es allein zu zeigen (Schritt 6).

##### *2. Der Fehlerhaushalt und die Kovarianz*

Wie groß sind die einzelnen Fehlerquellen, und wie stark sind die Messungen untereinander korreliert? Vorher vorhersagen: Wird der Fehler eines Mittelwerts aus 15 Punkten durch die korrelierten Teile kleiner, gleich oder größer, als wenn alle Fehler unabhängig wären?

Links stehen die relativen Fehler jedes Punktes nach Quelle (Theorie, Abschnitt 2): der Tabellenfehler aus Statistik und Metallizität, die IMF, das Modell der Sternpopulationen ohne das am stärksten abweichende Modell, und zum Vergleich mit ihm. Rechts steht die Korrelationsmatrix der vollen Kovarianz. Zum Schluss schätzt der Code ein gewichtetes Mittel des Verhältnisses $H/H_\text{Planck}$ einmal mit der Diagonale und einmal mit der vollen Matrix.

**Hintergrund & Fachbegriffe:**
- **Korrelationsmatrix:** $\rho_{ij} = C_{ij}/\sqrt{C_{ii} C_{jj}}$; 1 heißt, zwei Fehler bewegen sich immer gemeinsam.
- **Gewichtetes Mittel mit Kovarianz (GLS):** $\hat x = (\mathbf 1^T C^{-1} \mathbf y)/(\mathbf 1^T C^{-1} \mathbf 1)$ mit dem Fehler $(\mathbf 1^T C^{-1}\mathbf 1)^{-1/2}$.

```python
rel = {"Tabelle (Statistik + Metallizität)": E_CC / H_CC, "IMF": ETA["IMF"], "Modell (SPS, ohne Ausreißer)": ETA["SPS ohne Ausreißer"],
       "Modell (SPS, alle Modelle)": ETA["SPS"], "Sternbibliothek (nicht verwendet)": ETA["Sternbibliothek"]}
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 6))
for (k, v), st in zip(rel.items(), ["o-", "s-", "D-", "D--", "^:"]):
    ax.plot(Z_CC, 100 * v, st, ms=5, label=k)
ax.set_xlabel("Rotverschiebung z")
ax.set_ylabel("relativer Fehler von H(z) [%]")
ax.set_title("Woher die Fehler kommen", fontsize=10)
ax.legend(fontsize=9)
corr = COV_CC / np.sqrt(np.outer(np.diag(COV_CC), np.diag(COV_CC)))
im = ax2.imshow(corr, cmap="viridis", vmin=0, vmax=1)
ax2.set_xticks(range(len(Z_CC)))
ax2.set_xticklabels([f"{x:.2f}" for x in Z_CC], rotation=90, fontsize=8)
ax2.set_yticks(range(len(Z_CC)))
ax2.set_yticklabels([f"{x:.2f}" for x in Z_CC], fontsize=8)
ax2.set_xlabel("z")
ax2.set_ylabel("z")
ax2.set_title("Korrelation der Fehler (volle Kovarianz)", fontsize=10)
plt.colorbar(im, ax=ax2, shrink=0.85, label="Korrelationskoeffizient")
plt.tight_layout()
plt.show()

y = H_CC / (PLANCK["H0"][0] * E(Z_CC, PLANCK["H0"][0], PLANCK["Om"][0]))
for lab, cv in [("nur Diagonale", np.diag(E_CC**2)), ("volle Kovarianz", COV_CC)]:
    cy = cv / np.outer(H_CC / y, H_CC / y)
    ci = np.linalg.inv(cy)
    one = np.ones(len(y))
    print(f"H/H_Planck gemittelt, {lab:16s}: {(one @ ci @ y) / (one @ ci @ one):.3f} ± {1 / np.sqrt(one @ ci @ one):.3f}")
off = corr[~np.eye(len(Z_CC), dtype=bool)]
print(f"Korrelationen außerhalb der Diagonale: {off.min():.2f} bis {off.max():.2f}, Median {np.median(off):.2f}")
print(f"mittlere relative Fehler: Tabelle {100 * np.mean(E_CC / H_CC):.1f} %, IMF {100 * np.mean(ETA['IMF']):.2f} %, "
      f"SPS {100 * np.mean(ETA['SPS ohne Ausreißer']):.1f} % (mit allen Modellen {100 * np.mean(ETA['SPS']):.1f} %)")
```

* Links die relativen Fehler je Quelle. Der Tabellenfehler (Statistik und Metallizität) liegt im Mittel bei 13.4 % und wächst zu großem z. Das Modell der Sternpopulationen trägt im Mittel 4.9 % bei, die IMF nur 0.41 %. Mit allen Modellen, auch dem am stärksten abweichenden, wären es 9.8 %; die Sternbibliothek (gepunktet) ist in der Kovarianz nicht enthalten, wie bei Moresco et al.
* Rechts die Korrelationsmatrix: Zwischen den Punkten liegt sie bei 0.01 bis 0.47, im Median 0.13. Am stärksten korreliert sind die beiden genauesten Punkte bei z = 0.18 und 0.20, weil dort der gemeinsame Modellfehler gegen den eigenen Fehler am meisten zählt.
* Zur Vorhersage: größer. Ein gewichtetes Mittel von $H/H_\text{Planck}$ hat mit der Diagonale den Fehler ±0.027, mit der vollen Matrix ±0.058, mehr als doppelt so viel. Ein gemeinsamer Fehler von 5 % lässt sich durch mehr Punkte nicht verkleinern. Beide Mittel sind mit 1.005 und 1.008 nahe 1, also nahe Planck.

##### *3. Wie genau muss die Uhr gehen?*

Wie groß sind die Altersunterschiede, die die Chronometer messen müssen, und wie stark unterscheiden sich Planck und SH0ES darin? Vorher vorhersagen: Um wie viele Millionen Jahre unterscheidet sich der Altersunterschied zwischen z = 0.45 und 0.55, wenn H₀ 73 statt 67.4 ist?

Der Code rechnet in flachem ΛCDM das Alter des Universums $t(z) = \int_z^\infty dz'/[(1+z')H(z')]$, einmal mit H₀ = 67.36 und einmal mit 73.04, beide mit $\Omega_m$ = 0.315. Links stehen die Alter, rechts die Zeit, die zwischen $z - 0.05$ und $z + 0.05$ vergeht, und der Unterschied zwischen den beiden Modellen.

**Hintergrund & Fachbegriffe:**
- **Hubble-Zeit:** $1/H_0$, für H₀ = 70 km/s/Mpc etwa 14 Milliarden Jahre; Umrechnung $1/H$ [Gyr] = 977.8/(H in km/s/Mpc).
- **Altersunterschied statt Alter:** Die Chronometer brauchen nur $\Delta t$; ein Fehler, der alle Alter gleich verschiebt, hebt sich auf.

```python
GYR = 977.8                                                # 1/(km/s/Mpc) in Gyr
zt = np.linspace(0, 3, 601)


def age(z, H0, Om, zmax=1e4):
    zz_ = np.concatenate([np.linspace(z, 20, 4000), np.logspace(np.log10(20.01), np.log10(zmax), 2000)])
    f = 1 / ((1 + zz_) * H0 * E(zz_, H0, Om))
    return GYR * np.sum((f[1:] + f[:-1]) / 2 * np.diff(zz_))


fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 5.6))
DT = {}
for H0, col in [(PLANCK["H0"][0], "k"), (H0_SH0ES[0], "C3")]:
    t = np.array([age(x, H0, PLANCK["Om"][0]) for x in zt])
    dt = np.array([age(x - 0.05, H0, PLANCK["Om"][0]) - age(x + 0.05, H0, PLANCK["Om"][0]) for x in zt[(zt >= 0.05)]])
    DT[H0] = (zt[zt >= 0.05], dt)
    ax.plot(zt, t, color=col, lw=1.6, label=f"H₀ = {H0}")
    ax2.plot(zt[zt >= 0.05], 1e3 * dt, color=col, lw=1.6, label=f"H₀ = {H0}")
ax2.plot(DT[PLANCK["H0"][0]][0], 1e3 * (DT[PLANCK["H0"][0]][1] - DT[H0_SH0ES[0]][1]), color="C0", ls="--", lw=1.4, label="Unterschied")
for a_ in (ax, ax2):
    a_.set_xlabel("Rotverschiebung z")
    a_.legend(fontsize=9)
ax.set_ylabel("Alter des Universums [Gyr]")
ax.set_title("Wie alt war das Universum bei z?", fontsize=10)
ax2.set_ylabel("Zeit zwischen z − 0.05 und z + 0.05 [Myr]")
ax2.set_title("Welche Altersunterschiede die Uhr auflösen muss", fontsize=10)
plt.tight_layout()
plt.show()

for H0 in (PLANCK["H0"][0], H0_SH0ES[0]):
    print(f"H₀ = {H0}: Alter heute {age(0, H0, PLANCK['Om'][0]):.2f} Gyr, bei z = 0.5 {age(0.5, H0, PLANCK['Om'][0]):.2f} Gyr, "
          f"bei z = 1.97 {age(1.965, H0, PLANCK['Om'][0]):.2f} Gyr")
for z0 in (0.2, 0.5, 1.0, 1.5):
    k = np.argmin(np.abs(DT[PLANCK["H0"][0]][0] - z0))
    a1, a2 = DT[PLANCK["H0"][0]][1][k], DT[H0_SH0ES[0]][1][k]
    print(f"z = {z0}: Δt = {1e3 * a1:.0f} Myr (Planck), {1e3 * a2:.0f} Myr (73.04), Unterschied {1e3 * (a1 - a2):.0f} Myr; "
          f"5 % von Δt sind {50 * a1:.0f} Myr")
```

* Links das Alter des Universums bei der Rotverschiebung z: heute 13.80 Gyr mit H₀ = 67.36 und 12.72 Gyr mit 73.04, bei z = 0.5 8.58 und 7.92 Gyr, bei z = 1.97 3.33 und 3.07 Gyr. Die Galaxien der höchsten Messung sind also höchstens drei Milliarden Jahre alt.
* Rechts die Zeit zwischen z − 0.05 und z + 0.05: bei z = 0.2 rund 1.1 Gyr, bei z = 1.5 nur 245 Myr. Der Unterschied zwischen den beiden H₀ (blau gestrichelt) ist 85 Myr bei z = 0.2 und 19 Myr bei z = 1.5.
* Zur Vorhersage: Bei z = 0.5 vergehen 732 Myr (Planck) oder 675 Myr (73.04), 57 Myr Unterschied. Um zwischen den beiden zu unterscheiden, müsste der Altersunterschied von Galaxien, die selbst mehrere Milliarden Jahre alt sind, auf etwa 40 Myr genau bekannt sein, weniger als 1 % ihres Alters.

* Mit voller Kovarianz ist H₀ = 66.99 (68 %: 61.5 bis 72.5, also ± 5.5) und $\Omega_m$ = 0.337, mit χ² = 6.11 bei 13 Freiheitsgraden. Planck liegt 0.1σ, SH0ES 1.1σ entfernt. Mit der Diagonale bleibt das Minimum fast gleich (66.84), aber der Fehler fällt auf ± 3.8.

##### *6. Wann die Beschleunigung begann*

Ab welcher Rotverschiebung beschleunigt die Expansion, und sehen die Uhren das auch ohne Modell? Vorher vorhersagen: Bei welchem z wechselt $q$ in ΛCDM mit $\Omega_m$ = 0.315 das Vorzeichen?

Aus dem Gauß-Prozess mit voller Kovarianz zieht der Code 4000 Stichproben der gemeinsamen Verteilung von H(z) und H'(z) und rechnet für jede $q(z) = (1+z)H'/H - 1$ (Theorie, Abschnitt 3). Die Rotverschiebung, bei der q durch null geht, ist $z_t$. Zum Vergleich steht $q(z)$ des ΛCDM-Fits aus Schritt 4; dessen $z_t$ und Fehler kommen aus dem Profil von $\Omega_m$.

**Hintergrund & Fachbegriffe:**
- **Verzögerungsparameter $q$:** positiv bei gebremster, negativ bei beschleunigter Expansion; heute in ΛCDM $q_0 = \tfrac32\Omega_m - 1 \approx -0.53$.
- **Ableitung eines Gauß-Prozesses:** Auch die Ableitung ist gaußverteilt, mit der abgeleiteten Kovarianz; sie ist unsicherer als die Funktion selbst.

```python
sf, l, mean, cv, _ = GP["volle Kovarianz"]
draws = np.random.default_rng(5).multivariate_normal(mean, cv + 1e-9 * np.eye(len(mean)), size=4000)
Hd, dHd = draws[:, :n], draws[:, n:]
q = (1 + zs) * dHd / Hd - 1


def crossing(row):
    # erste Stelle von z = 0 aus, an der q von negativ zu positiv wechselt, linear interpoliert
    k = np.where((row[:-1] < 0) & (row[1:] >= 0))[0]
    if len(k) == 0:
        return np.nan
    k = k[0]
    return zs[k] - row[k] * (zs[k + 1] - zs[k]) / (row[k + 1] - row[k])


zt_s = np.array([crossing(row) for row in q])
x = FIT4["volle Kovarianz"][0]
q_l = (1 + zs) * np.gradient(E(zs, *x), zs) / E(zs, *x) - 1
om_p = np.linspace(0.05, 0.9, 341)
prof_om = np.array([optimize.minimize_scalar(lambda h: chi2_cc(h, o), bounds=(30, 120), method="bounded").fun for o in om_p])
prof_om -= prof_om.min()
zt_of = lambda Om: (2 * (1 - Om) / Om) ** (1 / 3) - 1
om_lo, om_hi = interval(om_p, prof_om)

fig, ax = plt.subplots(figsize=(12, 6))
lo, med, hi = np.percentile(q, [16, 50, 84], axis=0)
ax.fill_between(zs, lo, hi, color="C0", alpha=0.3, label="Gauß-Prozess, 68 %")
ax.plot(zs, med, color="C0", lw=1.6, label="Gauß-Prozess, Median")
ax.plot(zs, q_l, color="k", ls="--", lw=1.4, label=f"ΛCDM-Fit, Ω_m = {x[1]:.2f}")
ax.plot(zs, 1.5 * PLANCK["Om"][0] * (1 + zs) ** 3 / (PLANCK["Om"][0] * (1 + zs) ** 3 + 1 - PLANCK["Om"][0]) - 1, color="C3", lw=1.2, label="Planck-ΛCDM")
ax.axhline(0, color="0.5", lw=0.8)
ax.set_ylim(-1.5, 1.5)
ax.set_xlabel("Rotverschiebung z")
ax.set_ylabel("Verzögerungsparameter q(z)")
ax.set_title("Gebremste Expansion (q > 0) und Beschleunigung (q < 0)", fontsize=10)
ax.legend(fontsize=9)
plt.show()

ok = np.isfinite(zt_s)
print(f"Gauß-Prozess: q₀ = {np.median(q[:, 0]):.2f} ({np.percentile(q[:, 0], 16):.2f} bis {np.percentile(q[:, 0], 84):.2f}); "
      f"Vorzeichenwechsel in {100 * ok.mean():.0f} % der Stichproben, z_t = {np.median(zt_s[ok]):.2f} "
      f"({np.percentile(zt_s[ok], 16):.2f} bis {np.percentile(zt_s[ok], 84):.2f})")
print(f"ΛCDM-Fit: Ω_m = {x[1]:.3f} ({om_lo:.3f} bis {om_hi:.3f}), z_t = {zt_of(x[1]):.2f} ({zt_of(om_hi):.2f} bis {zt_of(om_lo):.2f}), q₀ = {1.5 * x[1] - 1:.2f}")
print(f"Planck-ΛCDM: z_t = {zt_of(PLANCK['Om'][0]):.2f}, q₀ = {1.5 * PLANCK['Om'][0] - 1:.2f}")
```

* Der Verzögerungsparameter $q(z)$: blau der Gauß-Prozess (Median und 68 %), schwarz gestrichelt der ΛCDM-Fit, rot Planck-ΛCDM. In ΛCDM ist $q$ heute −0.49 (Fit) bzw. −0.53 (Planck) und wird bei z ≈ 0.6 positiv.
* Der Gauß-Prozess lässt beides offen. Sein Median bleibt bei allen z knapp unter null, das Band reicht von −0.5 bis +0.1 bei z = 0. Nur 46 % seiner Stichproben wechseln das Vorzeichen; wenn, dann bei $z_t$ = 0.60 (0.30 bis 0.90). Die Ableitung einer Kurve aus 15 verrauschten Punkten ist viel unsicherer als die Kurve selbst.
* Zur Vorhersage: $z_t$ = 0.63 für $\Omega_m$ = 0.315. Der ΛCDM-Fit an die Uhren gibt $z_t$ = 0.58 (0.36 bis 0.81), aus $\Omega_m$ = 0.337 (0.253 bis 0.445). Dass die Expansion beschleunigt, zeigen die Uhren mit ΛCDM; ohne Modell sind sie dafür noch zu ungenau.

##### *7. Die Uhr eicht das Lineal, Punkt für Punkt*

Wie lang ist das Lineal der BAO, wenn man es an jeder Rotverschiebung einzeln mit der Uhr misst? Vorher vorhersagen: Welches $r_d$ erwartet man, wenn H₀ = 73 stimmt und die BAO $h\,r_d$ = 101.55 Mpc messen?

Für die fünf Stichproben von DESI mit $D_H/r_d$ im Bereich der Chronometer (z = 0.51 bis 1.48) nimmt der Code H(z) aus dem Gauß-Prozess (Schritt 5) und rechnet $r_d = c/[H(z)\,(D_H/r_d)]$ (Theorie, Abschnitt 5). Die Fehler von H(z) und von $D_H/r_d$ werden quadratisch addiert. Ein gemeinsamer Wert folgt aus einem gewichteten Mittel, das die Korrelation der Gauß-Prozess-Werte zwischen den fünf z berücksichtigt.

**Hintergrund & Fachbegriffe:**
- **Unabhängig vom frühen Universum:** Weder Planck noch die Nukleosynthese gehen ein, und keine Annahme über die Dunkle Energie, weil Uhr und Lineal bei derselben Rotverschiebung verglichen werden.
- **Korrelation über den Gauß-Prozess:** Die fünf Werte von H(z) stammen aus derselben glatten Kurve und sind deshalb stark korreliert; ein Mittel ohne diese Korrelation wäre zu genau.

```python
mh = KIND == "DH_over_rs"
z_dh, dh, e_dh = Z_BAO[mh], Y_BAO[mh], np.sqrt(np.diag(COV_BAO))[mh]
inside = z_dh < Z_CC.max()                                # Lyman-α bei z = 2.33 liegt jenseits der Uhren
z_dh, dh, e_dh = z_dh[inside], dh[inside], e_dh[inside]
sf, l, _, _, cov = GP["volle Kovarianz"]
mean_dh, cv_dh = gp_predict(z_dh, cov, sf, l)
H_dh, CH = mean_dh[:len(z_dh)], cv_dh[:len(z_dh), :len(z_dh)]
rd_z = C_KMS / (H_dh * dh)
J = np.diag(rd_z / H_dh)                                  # ∂r_d/∂H
C_rd = J @ CH @ J + np.diag((rd_z * e_dh / dh) ** 2)
ci = np.linalg.inv(C_rd)
one = np.ones(len(rd_z))
RD_POINT = ((one @ ci @ rd_z) / (one @ ci @ one), 1 / np.sqrt(one @ ci @ one))

fig, ax = plt.subplots(figsize=(11, 6))
ax.errorbar(z_dh, rd_z, yerr=np.sqrt(np.diag(C_rd)), fmt="o", color="C0", ms=8, capsize=4, label="Chronometer (Gauß-Prozess) + DESI D_H/r_d")
ax.axhspan(RD_POINT[0] - RD_POINT[1], RD_POINT[0] + RD_POINT[1], color="C0", alpha=0.2, label=f"gewichtetes Mittel {RD_POINT[0]:.1f} ± {RD_POINT[1]:.1f} Mpc")
ax.axhline(PLANCK["rd"][0], color="k", lw=1.4, label="Planck (frühes Universum)")
rd73 = HRD_BAO[0] / (H0_SH0ES[0] / 100)
ax.axhline(rd73, color="C2", ls="--", lw=1.4, label=f"h·r_d der BAO mit H₀ = 73.04: {rd73:.1f} Mpc")
ax.set_xlabel("Rotverschiebung z")
ax.set_ylabel(r"$r_d$ [Mpc]")
ax.set_title("Das Lineal der BAO, gemessen mit der Uhr der Chronometer", fontsize=10)
ax.legend(fontsize=9)
plt.show()

for zi, r, e, h in zip(z_dh, rd_z, np.sqrt(np.diag(C_rd)), H_dh):
    print(f"z = {zi:5.3f}: H = {h:6.1f} km/s/Mpc, r_d = {r:6.1f} ± {e:4.1f} Mpc")
print(f"Mittel: r_d = {RD_POINT[0]:.1f} ± {RD_POINT[1]:.1f} Mpc; Planck {PLANCK['rd'][0]} ({(RD_POINT[0] - PLANCK['rd'][0]) / RD_POINT[1]:+.1f} σ), "
      f"mit H₀ = 73.04 nötig {rd73:.1f} ({(RD_POINT[0] - rd73) / RD_POINT[1]:+.1f} σ)")
```

* Für jede der fünf Stichproben von DESI zwischen z = 0.51 und 1.48 ein Wert von $r_d$ aus der Uhr: 149.7, 149.8, 146.2, 152.3 und 158.1 Mpc, jeweils auf ±10 bis ±17 Mpc. Das blaue Band ist das gewichtete Mittel, schwarz Planck, grün gestrichelt das $r_d$, das die BAO bei H₀ = 73.04 verlangen.
* Mittel: $r_d$ = 148.9 ± 9.3 Mpc, nur 0.2σ neben Planck (147.09) und 1.1σ über 139.0 Mpc. Die fünf Werte sind über den Gauß-Prozess stark korreliert; das Mittel ist deshalb kaum genauer als der beste Einzelwert.
* Zur Vorhersage: $r_d$ = 101.55/0.7304 = 139.0 Mpc. Die Uhren bevorzugen leicht das längere Lineal von Planck, können aber das kürzere nicht ausschließen.

##### *8. Uhren und BAO zusammen in ΛCDM*

Wie genau bestimmen Uhren und BAO gemeinsam $r_d$ und H₀, wenn man ΛCDM annimmt, und hilft die Form aus den Supernovae? Vorher vorhersagen: Wird $r_d$ genauer als im Punkt-für-Punkt-Vergleich?

Der Code passt H₀, $\Omega_m$ und $r_d$ an zwei Kombinationen an: Chronometer und BAO, und zusätzlich Pantheon+ (nur die Form, $M_B$ frei). Für $r_d$ und H₀ rechnet er das Profil-χ² und liest die 68-%-Grenzen bei Δχ² = 1 ab.

**Hintergrund & Fachbegriffe:**
- **Wer was beiträgt:** Die BAO legen $\Omega_m$ und $h\,r_d$ fest, die Chronometer den Maßstab von H(z). Ihre Kombination trennt h und $r_d$.
- **Profil-χ²:** Für jeden Wert von $r_d$ wird χ² über H₀ und $\Omega_m$ minimiert; die Kurve zeigt, wie stark die Daten $r_d$ allein festlegen.

```python
FIT8, PROF8 = {}, {}
rds, hs = np.linspace(125, 175, 101), np.linspace(55, 85, 121)
for key, data in [("Uhren + BAO", ("CC", "BAO")), ("Uhren + BAO + Pantheon+", ("CC", "BAO", "SN"))]:
    r = fit(data)
    FIT8[key] = r
    PROF8[key] = (profile(data, 2, rds, r), profile(data, 0, hs, r))
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 5.6))
for (key, (pr, ph)), col in zip(PROF8.items(), ["C0", "C3"]):
    ax.plot(rds, pr, color=col, lw=1.6, label=key)
    ax2.plot(hs, ph, color=col, lw=1.6, label=key)
ax.axvline(PLANCK["rd"][0], color="k", lw=1.2, label="Planck")
ax.axvline(HRD_BAO[0] / (H0_SH0ES[0] / 100), color="C2", ls="--", lw=1.2, label="nötig für H₀ = 73.04")
ax2.axvspan(H0_SH0ES[0] - H0_SH0ES[1], H0_SH0ES[0] + H0_SH0ES[1], color="C2", alpha=0.25, label="SH0ES")
ax2.axvspan(PLANCK["H0"][0] - PLANCK["H0"][1], PLANCK["H0"][0] + PLANCK["H0"][1], color="0.6", alpha=0.4, label="Planck")
for a_, lab in [(ax, r"$r_d$ [Mpc]"), (ax2, "H₀ [km/s/Mpc]")]:
    a_.axhline(1, color="0.5", ls=":")
    a_.set_ylim(0, 9)
    a_.set_xlabel(lab)
    a_.set_ylabel(r"$\Delta\chi^2$ (Profil)")
    a_.legend(fontsize=9)
ax.set_title("Länge des Lineals ohne frühes Universum", fontsize=10)
ax2.set_title("H₀ ohne Cepheiden und ohne frühes Universum", fontsize=10)
plt.tight_layout()
plt.show()

RD8 = {}
for key, r in FIT8.items():
    pr, ph = PROF8[key]
    (rl, rh), (hl, hh) = interval(rds, pr), interval(hs, ph)
    RD8[key] = (r.x, (rh - rl) / 2, (hh - hl) / 2)
    print(f"{key:24s}: r_d = {r.x[2]:.1f} ({rl:.1f} bis {rh:.1f}), H₀ = {r.x[0]:.2f} ({hl:.1f} bis {hh:.1f}), Ω_m = {r.x[1]:.3f}, "
          f"h r_d = {r.x[0] / 100 * r.x[2]:.2f}, χ² = {r.fun:.1f}")
x, e, _ = RD8["Uhren + BAO"]
print(f"Uhren + BAO: r_d gegen Planck {(x[2] - PLANCK['rd'][0]) / e:+.1f} σ, gegen das für H₀ = 73 nötige {(x[2] - HRD_BAO[0] / 0.7304) / e:+.1f} σ")
```

* Links das Profil-χ² von $r_d$, rechts das von H₀, blau Uhren und BAO, rot zusätzlich Pantheon+. Die Kurven liegen fast aufeinander: Die Supernovae ändern die Form kaum, weil die BAO $\Omega_m$ schon festlegen.
* Uhren + BAO: $r_d$ = 147.8 (140.0 bis 156.5) Mpc, H₀ = 68.70 (64.8 bis 72.5), $\Omega_m$ = 0.298, $h\,r_d$ = 101.52 Mpc wie bei den BAO allein. Mit Pantheon+: $r_d$ = 147.6, H₀ = 68.41. Gegen Planck 0.1σ, gegen 139 Mpc 1.1σ.
* Zur Vorhersage: kaum genauer (±8.2 statt ±9.3 Mpc). Den Fehler setzt in beiden Fällen der gemeinsame Modellfehler der Uhren von rund 5 %, nicht die Zahl der Punkte. Die beiden Hypothesen, 147 und 139 Mpc, liegen nur 1.0 eigene Fehler auseinander.

##### *9. Die Uhr eicht die Kerzen*

Wie hell sind Supernovae vom Typ Ia, wenn man sie mit der Uhr statt mit Cepheiden eicht? Vorher vorhersagen: Ist das $M_B$ aus den Uhren heller (negativer) oder schwächer als das der Cepheiden, wenn die Uhren ein H₀ nahe 67 bevorzugen?

Der Code passt H₀ und $\Omega_m$ an Chronometer und Pantheon+ zusammen an. Die Supernovae legen dabei die Form fest, die Uhren den Maßstab. Aus dem besten Modell folgt die absolute Helligkeit $M_B$, die die Supernovae mit diesen Entfernungen haben müssen. Im Bild stehen das Profil von H₀ und, als zweite Achse, das zugehörige $M_B$, neben dem Wert der Cepheiden.

**Hintergrund & Fachbegriffe:**
- **$M_B$ und H₀:** Für die Supernovae ist nur $M_B - 5\log_{10} H_0$ bestimmt; ein um 0.2 mag helleres $M_B$ entspricht einem um 10 % kleineren H₀.
- **Drei Eichungen:** Cepheiden (SH0ES), das frühe Universum über $r_d$ (inverse Leiter) und hier die Uhren; jede gibt ein $M_B$.

```python
r9 = fit(("CC", "SN"))
hs9 = np.linspace(55, 85, 121)
prof9 = profile(("CC", "SN"), 0, hs9, r9)
lo9, hi9 = interval(hs9, prof9)
M9 = chi2_sn(r9.x[0], r9.x[1], return_M=True)[1]
M_of_H0 = lambda H0: M9 + 5 * np.log10(H0 / r9.x[0])     # bei fester Form
fig, ax = plt.subplots(figsize=(11, 6))
ax.plot(hs9, prof9, color="C0", lw=1.8, label="Uhren + Pantheon+ (Profil)")
ax.axhline(1, color="0.5", ls=":")
ax.axvspan(H0_SH0ES[0] - H0_SH0ES[1], H0_SH0ES[0] + H0_SH0ES[1], color="C2", alpha=0.25, label="SH0ES (M_B aus Cepheiden)")
ax.axvspan(PLANCK["H0"][0] - PLANCK["H0"][1], PLANCK["H0"][0] + PLANCK["H0"][1], color="0.6", alpha=0.4, label="Planck")
ax.set_ylim(0, 9)
ax.set_xlabel("H₀ [km/s/Mpc]")
ax.set_ylabel(r"$\Delta\chi^2$ (Profil)")
ax.set_title("Supernovae mit der Uhr geeicht", fontsize=10, pad=42)
top = ax.secondary_xaxis("top", functions=(M_of_H0, lambda M: r9.x[0] * 10 ** ((M - M9) / 5)))
top.set_xlabel(r"absolute Helligkeit $M_B$ der Supernovae [mag]")
ax.legend(fontsize=9)
plt.show()

eM = 5 / np.log(10) * (hi9 - lo9) / 2 / r9.x[0]
print(f"Uhren + Pantheon+: H₀ = {r9.x[0]:.2f} ({lo9:.1f} bis {hi9:.1f}), Ω_m = {r9.x[1]:.3f}, χ² = {r9.fun:.1f}")
print(f"M_B = {M9:.3f} ± {eM:.3f} mag; Cepheiden {M_B_CEPHEID[0]} ± {M_B_CEPHEID[1]} → Unterschied {M9 - M_B_CEPHEID[0]:+.3f} mag "
      f"({(M9 - M_B_CEPHEID[0]) / np.hypot(eM, M_B_CEPHEID[1]):+.1f} σ)")
```

* Das Profil-χ² von H₀ für Uhren und Supernovae zusammen; oben die zugehörige absolute Helligkeit $M_B$ der Supernovae. Grau Planck, grün SH0ES.
* H₀ = 67.23 (63.5 bis 71.0), $\Omega_m$ = 0.332, und daraus $M_B$ = −19.439 ± 0.121 mag. Die Cepheiden geben −19.249 ± 0.030; der Unterschied ist −0.19 mag, 1.5σ. Mit der Uhr geeicht wären die Supernovae also etwas heller als mit Cepheiden, so viel, wie die Lücke zwischen den beiden Leitern verlangt (0.18 mag, siehe [Baryonische akustische Oszillationen](https://github.com/deltorobarba/science/blob/main/astro/cosmology_bao.ipynb), Schritt 3).
* Zur Vorhersage: heller. Bevorzugt die Uhr ein kleineres H₀, müssen die Supernovae weiter weg sein, also bei gleicher scheinbarer Helligkeit leuchtkräftiger.

#### **Wiederholung**

Erst selbst antworten, dann aufklappen.

<details><summary>1. Warum ist H(z) = −dz/dt/(1+z)?</summary>

Aus $1 + z = 1/a$ folgt $dz/dt = -\dot a/a^2 = -(1+z)\,\dot a/a = -(1+z)\,H$. Kennt man also, wie viel Zeit zwischen zwei Rotverschiebungen vergeht, kennt man H ohne jede Entfernung.
</details>

<details><summary>2. Warum braucht man passive Galaxien, und warum die massivsten?</summary>

Ihre Sterne sind alle früh und in kurzer Zeit entstanden; seither altern sie nur. Dann ist der Altersunterschied zweier solcher Galaxien gleich der kosmischen Zeit zwischen ihren Rotverschiebungen. Die massivsten haben am frühesten und schnellsten gebildet und kaum später Sterne hinzubekommen.
</details>

<details><summary>3. Warum fällt ein Fehler im absoluten Alter der Galaxien heraus?</summary>

Gemessen wird nur der Unterschied der Alter bei benachbarten Rotverschiebungen. Ein Fehler, der alle Alter um denselben Betrag verschiebt, hebt sich in der Differenz auf; nur Fehler in der Steigung A der Beziehung zwischen D4000 und Alter bleiben.
</details>

<details><summary>4. Warum ist der Fehler der Modellwahl voll korreliert?</summary>

Die Wahl des Modells der Sternpopulationen legt die Steigung A für alle Galaxien fest. Ein anderes Modell verschiebt deshalb alle Werte von H(z) in dieselbe Richtung, um etwa denselben relativen Betrag.
</details>

<details><summary>5. Warum wird ein Mittel über 15 Punkte durch korrelierte Fehler nicht viel genauer als ein einzelner Punkt?</summary>

Unabhängige Fehler mitteln sich wie $1/\sqrt N$ weg, ein gemeinsamer Fehler nicht. Bei 4.9 % gemeinsamem Fehler bleibt das Mittel immer mindestens etwa so unsicher, egal wie viele Punkte man hat.
</details>

<details><summary>6. Wie genau muss die Uhr gehen, um Planck und SH0ES zu unterscheiden?</summary>

Zwischen z = 0.45 und 0.55 vergehen mit Planck 732 Myr, mit H₀ = 73 nur 675 Myr. Die Altersunterschiede müssen also auf einige zehn Millionen Jahre genau sein, bei Galaxien, die mehrere Milliarden Jahre alt sind.
</details>

<details><summary>7. Warum ist H₀ aus den Uhren eine Extrapolation, und was folgt daraus?</summary>

Die nächste Messung liegt bei z = 0.18. H₀ folgt aus der Form der Kurve, die man annimmt: in ΛCDM die Form von $E(z)$, im Gauß-Prozess eine glatte Kurve ohne feste Form. Ohne die Punkte bei z ≈ 0.2 wird H₀ fast unbestimmt.
</details>

<details><summary>8. Was macht ein Gauß-Prozess, und wofür stehen σ_f und ℓ?</summary>

Er ist eine Normalverteilung über glatte Funktionen; die Daten wählen aus ihr die passenden aus. $\ell$ ist die Strecke in z, über die die Kurve glatt bleibt, $\sigma_f$ die typische Größe ihrer Werte. Beide werden nach der Randwahrscheinlichkeit der Daten gewählt.
</details>

<details><summary>9. Warum ist der Verzögerungsparameter ohne Modell so unsicher?</summary>

$q = (1+z)H'/H - 1$ braucht die Ableitung von H. Eine Ableitung aus verrauschten Punkten verstärkt das Rauschen; bei 15 Punkten mit 5 bis 27 % Fehler reicht es nicht, das Vorzeichen von q sicher zu bestimmen.
</details>

<details><summary>10. Wie misst die Uhr die Länge des Lineals?</summary>

Die BAO messen $D_H/r_d = c/(H r_d)$. Die Uhr liefert H bei derselben Rotverschiebung, und damit ist $r_d = c/[H \cdot (D_H/r_d)]$, ohne Annahme über das frühe Universum oder die Dunkle Energie.
</details>

<details><summary>11. Welches r_d verlangt H₀ = 73, und warum?</summary>

Die BAO messen $h\,r_d$ = 101.55 Mpc. Mit h = 0.7304 folgt $r_d$ = 139.0 Mpc, 5.5 % weniger als die 147.09 Mpc, die Planck aus der Physik vor der Rekombination berechnet.
</details>

<details><summary>12. Warum entscheiden die Uhren die Hubble-Spannung noch nicht?</summary>

Ihr Fehler auf $r_d$ und H₀ ist durch den gemeinsamen Modellfehler von rund 5 % begrenzt. Die beiden Hypothesen (147 und 139 Mpc, 67 und 73 km/s/Mpc) liegen etwa einen solchen Fehler auseinander. Für 3σ müsste der Modellfehler unter 2 % fallen.
</details>

#### **Nächste Schritte**

So würde eine Analyse in der Forschung weitergehen:
- **Mehr und bessere Uhren:** Neuere Messungen mit anderen Altersmaßen (volle Anpassung der Spektren, Lick-Indizes) und neuen Durchmusterungen (LEGA-C, VANDELS) erweitern die Sammlung auf über 30 Punkte. Euclid und DESI werden Hunderttausende passive Galaxien bis z ≈ 2 liefern.
- **Den Modellfehler verkleinern:** Die Modelle der Sternpopulationen mit Sternhaufen bekannten Alters eichen, damit der gemeinsame Fehler von 5 % unter 2 % fällt. Erst dann können die Uhren zwischen den Leitern entscheiden.
- **Die volle Kovarianz für alle Messungen:** Für die Messungen aus anderen Arbeiten gibt es noch keine gemeinsame Kovarianz. Sie zu bestimmen, ist Voraussetzung, um sie mit den 15 hier zu verbinden.
- **Krümmung ohne Modell:** H(z) aus den Uhren und $D_M(z)$ aus Supernovae oder BAO zusammen messen die Krümmung des Raums, ohne ΛCDM anzunehmen, über $D_M = (c/H_0\sqrt{\Omega_k})\sinh(\sqrt{\Omega_k}\,H_0 \int dz/H)$.
- **Dunkle Energie:** Die Uhren mit BAO, CMB und Supernovae in w₀w_aCDM verbinden; als einzige direkte Messung von H(z) ohne Lineal könnten sie die Bevorzugung durch DESI unabhängig prüfen.
