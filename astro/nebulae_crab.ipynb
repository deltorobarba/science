# Der Krebsnebel

Das Notebook misst die Ausdehnung des Krebsnebels über 60 Jahre: Rote Palomar-Fotoplatten von 1951 und 1991 und Pan-STARRS-Bilder von 2009 bis 2014 werden an Gaia DR3 angeschlossen, auf ein gemeinsames Raster gebracht und Stück für Stück mit einer maskierten Kreuzkorrelation verglichen. Aus den Verschiebungen folgen Expansionszentrum, Konvergenzjahr, Beschleunigung und eine Expansionsparallaxe; 44 Jahre Pulsar-Timing aus Jodrell Bank liefern Leistung, Bremsindex und Anfangsperiode des Pulsars und damit die Energiebilanz.

Kann man aus Fotos, die 60 Jahre auseinanderliegen, das Jahr der Supernova von 1054 ablesen, und warum kommt dabei ein späteres Jahr heraus?

#### **Hintergrund**

Am 4. Juli 1054 verzeichneten Astronomen am chinesischen Kaiserhof einen „Gaststern“ im Sternbild Stier. Er war 23 Tage lang am hellen Tag zu sehen und verschwand erst nach fast zwei Jahren vom Nachthimmel. An derselben Stelle steht heute der Krebsnebel, Messier 1. Dass beides zusammengehört, wurde 1921 klar: John C. Duncan verglich Fotoplatten des Mount-Wilson-Observatoriums, die zwölf Jahre auseinanderlagen, und fand, dass der Nebel wächst, und Knut Lundmark brachte ihn im selben Jahr mit dem Gaststern in Verbindung. 1968 fand man im Zentrum einen Pulsar, der sich 30-mal pro Sekunde dreht. Er war der Beweis, dass Supernovae Neutronensterne hinterlassen, und er erklärte, woher der Nebel seine Energie hat.

Der Krebsnebel ist seitdem das am gründlichsten untersuchte Objekt außerhalb des Sonnensystems. Forschende messen die Bewegung seiner Filamente auf alten und neuen Bildern, bauen aus Spektren dreidimensionale Karten, verfolgen den Pulsar seit über 50 Jahren mit Radioteleskopen und benutzen den Nebel in der Röntgen- und Gammaastronomie als Eichquelle. Ein Befund hält sich dabei seit Trimble (1968): Rechnet man die heutigen Geschwindigkeiten der Filamente zurück, landet man nicht bei 1054, sondern bei einem Jahr zwischen etwa 1090 und 1170. Die übliche Erklärung ist, dass der Pulsarwind die Filamente seither beschleunigt hat.

Dieses Notebook misst das mit drei Bildern aus öffentlichen Archiven nach: einer Palomar-Platte von 1951, einer von 1991 und Pan-STARRS-Aufnahmen von 2009 bis 2014. Alle drei werden an Gaia angeschlossen und Ausschnitt für Ausschnitt verglichen. Dann prüft es die Erklärung mit der Energie, die der Pulsar nach 44 Jahren Messungen in Jodrell Bank abgibt.

#### **Theorie**

##### *1. Ein Supernova-Überrest mit Pulsarwindnebel*

Ein Stern von etwa 8 bis 10 Sonnenmassen hat 1054 sein Leben beendet: Sein Kern ist zu einem Neutronenstern von etwa 12 km Radius zusammengestürzt, die Hülle wurde mit einigen tausend km/s ins All geschleudert. In den meisten Überresten fegt die Hülle das umgebende Gas zusammen, und die Stoßwelle heizt es auf Millionen Kelvin; man sieht dann eine Schale (Cas A, Tychos Supernova). Der Krebsnebel ist anders, er ist innen gefüllt.

Der junge Neutronenstern dreht sich schnell und hat ein starkes Magnetfeld. Er gibt seine Rotationsenergie als **Pulsarwind** ab, einen Strom aus Elektronen, Positronen und Magnetfeld. Dieser Wind bläst innerhalb der Hülle eine Blase auf, den **Pulsarwindnebel**. Die schnellen Elektronen kreisen darin im Magnetfeld und strahlen **Synchrotronlicht** ab, von Radiowellen bis zu Gammastrahlen. Im sichtbaren Licht ist das der bläulich-weiße, strukturlose Teil des Nebels.

Die Blase drückt von innen gegen die Hülle. Das Gas der Hülle wird an ihrem Rand zusammengeschoben und bricht durch die Rayleigh-Taylor-Instabilität (schweres Gas über leichtem, von unten beschleunigt) in Finger auf. Das sind die **Filamente**: Gas aus dem Stern (Wasserstoff, viel Helium, Stickstoff, Sauerstoff, Schwefel), das vom ultravioletten Synchrotronlicht ionisiert wird und in denselben Emissionslinien leuchtet wie eine H-II-Region (Hα, [N II], [S II], [O I], [O III]; siehe [Elektronentemperatur von H-II-Regionen](https://github.com/deltorobarba/science/blob/main/astro/nebulae_temperature.ipynb)).

Das Synchrotronspektrum ist ein Potenzgesetz, $F_\nu \propto \nu^{-\alpha}$, im Sichtbaren mit $\alpha \approx 0.6$ bis $0.9$ (flacher in der Mitte, steiler am Rand, weil die energiereichsten Elektronen auf dem Weg nach außen ihre Energie abstrahlen). Für die Filter r und i bei 617 und 750 nm erwartet man ohne Staub

$$\frac{F_\nu(r)}{F_\nu(i)} = \left(\frac{\nu_r}{\nu_i}\right)^{-\alpha} = \left(\frac{617}{750}\right)^{\alpha} \approx 0.86 \quad (\alpha = 0.8).$$

Staub zwischen uns und dem Nebel schwächt r stärker als i und drückt das Verhältnis um den Faktor $10^{-0.4(A_r - A_i)}$ weiter (Schritt 2).

**Fachbegriffe:**
- **Supernova-Überrest:** alles, was nach einer Supernova zurückbleibt, die ausgeworfene Hülle (Ejekta), das von ihr aufgeheizte Gas und gegebenenfalls ein Neutronenstern.
- **Plerion:** ältere Bezeichnung für einen Überrest, der innen von einem Pulsarwindnebel gefüllt ist; der Krebsnebel ist das Vorbild.
- **Ejekta:** das Material, das die Explosion aus dem Stern geschleudert hat.

##### *2. Fotoplatten und digitalisierte Himmelsdurchmusterungen*

Eine Fotoplatte ist eine Glasplatte mit einer Schicht aus Silberhalogenid-Körnern in Gelatine. Licht macht die Körner entwickelbar; nach dem Entwickeln ist die Platte dort schwarz, wo viel Licht ankam. Gemessen wird die **Schwärzung** (Dichte) $D = \log_{10}(1/T)$ mit der Durchlässigkeit $T$. Wie die Schwärzung von der Belichtung abhängt, beschreibt die **Schwärzungskurve**: Bei wenig Licht spricht die Platte kaum an, im mittleren Bereich wächst $D$ etwa linear mit dem Logarithmus der Belichtung, bei viel Licht sättigt sie. Eine Platte misst Helligkeit also nicht linear; im hellen Zentrum des Krebsnebels ist sie gesättigt.

Das 48-Zoll-Schmidt-Teleskop auf dem Palomar Mountain bildet ein Feld von 6.5° × 6.5° auf eine Platte von 36 cm Kantenlänge ab, 67.2″ pro Millimeter. Die Platte wird dazu in die gewölbte Bildfläche des Teleskops gebogen; deshalb braucht die Plattenlösung Glieder höherer Ordnung. Das Space Telescope Science Institute hat die Platten für den Leitsternkatalog des Hubble-Teleskops eingescannt: POSS-I mit 25 µm (1.7″) für den ersten **Digitized Sky Survey** (DSS1), POSS-II mit 15 µm (1.0″) für den zweiten (DSS2).

Für Eigenbewegungen zählt vor allem die Zeitbasis. Eine CCD-Aufnahme von heute ist viel genauer als eine Platte von 1951, aber nur die Platte kann zeigen, wo die Filamente vor 60 Jahren waren.

**Fachbegriffe:**
- **Schmidt-Teleskop:** ein Spiegelteleskop mit einer dünnen Korrekturplatte vor dem Kugelspiegel; es bildet ein sehr großes Feld scharf ab, aber auf eine gewölbte Fläche.
- **Emulsion:** die lichtempfindliche Schicht; 103aE (POSS-I) ist bis etwa 670 nm empfindlich, IIIaF (POSS-II) bis etwa 690 nm und feinkörniger.

##### *3. Astrometrie: Tangentialebene, Plattenlösung, Eigenbewegung*

**Tangentialebene.** Um Positionen am Himmel wie auf einer Landkarte zu vergleichen, projiziert man die Himmelskugel auf eine Ebene, die sie im Bezugspunkt $(\alpha_0, \delta_0)$ berührt (gnomonische Projektion):

$$\xi = \frac{\cos\delta\,\sin(\alpha - \alpha_0)}{\cos c},\qquad \eta = \frac{\sin\delta\cos\delta_0 - \cos\delta\sin\delta_0\cos(\alpha - \alpha_0)}{\cos c},$$

mit $\cos c = \sin\delta\sin\delta_0 + \cos\delta\cos\delta_0\cos(\alpha - \alpha_0)$, dem Kosinus des Winkelabstands vom Bezugspunkt. $\xi$ zeigt nach Osten, $\eta$ nach Norden; beide rechnet der Code in Bogensekunden.

**Plattenlösung.** Für jedes Bild werden $\xi$ und $\eta$ als Polynome der Pixelkoordinaten angesetzt,

$$\xi = \sum_{i + j \le 2} a_{ij}\, x^i y^j, \qquad \eta = \sum_{i + j \le 2} b_{ij}\, x^i y^j,$$

je sechs Koeffizienten, angepasst an einige hundert Sterne mit bekannten $(\xi, \eta)$. Für das Umrechnen auf ein gemeinsames Raster braucht man die Umkehrung $(x, y)$ als Polynom von $(\xi, \eta)$; sie wird auf dieselbe Weise angepasst.

**Eigenbewegung.** Die Bezugssterne bewegen sich selbst. Gaia gibt ihre Orte für 2016.0; für eine Platte zur Zeit $t$ gilt in guter Näherung

$$\alpha(t) = \alpha_{2016} + \frac{\mu_{\alpha*}}{\cos\delta}\,(t - 2016), \qquad \delta(t) = \delta_{2016} + \mu_\delta\,(t - 2016).$$

Ein Stern mit 5 mas/Jahr wandert in den 64 Jahren bis 1951 um 0.3″, so viel wie der ganze Restfehler einer Platte. Ohne diese Korrektur würde man die Bewegung der Sterne für eine Verzerrung der Platte halten.

**Fachbegriffe:**
- **ICRS:** das internationale himmelsfeste Bezugssystem, festgelegt durch ferne Quasare; Gaia DR3 verwirklicht es im Sichtbaren auf etwa 0.02 mas.
- **Relative und absolute Astrometrie:** Die Filamente werden relativ zu den Sternen im selben Bild gemessen; weil die Sterne an Gaia hängen, sind die Verschiebungen damit auch absolut.

##### *4. Kreuzkorrelation mit Maske*

Ein kleines Beispiel: Ein Ausschnitt von 21 × 21 Pixeln aus dem Bild von 2011 zeigt ein Stück Filament. Man legt ihn auf das Bild von 1951, zuerst an dieselbe Stelle, dann um 1, 2, … Pixel verschoben, und fragt jedes Mal, wie ähnlich die beiden Ausschnitte sind. Wo die Ähnlichkeit am größten ist, lag das Filament 1951.

Das Maß der Ähnlichkeit ist der Korrelationskoeffizient der Pixelwerte, die **normierte Kreuzkorrelation**: Für die Vorlage $T$ und das um $\vec s$ verschobene Stück $R_{\vec s}$ des anderen Bildes ist

$$\rho(\vec s) = \frac{\sum_{p \in V} (T_p - \bar T)(R_{\vec s, p} - \bar R_{\vec s})}{\sqrt{\sum_{p \in V} (T_p - \bar T)^2 \; \sum_{p \in V} (R_{\vec s, p} - \bar R_{\vec s})^2}}.$$

Die Mittelwerte werden über dieselbe Pixelmenge $V$ gebildet. Weil $\rho$ Mittelwert und Streuung abzieht, spielt es keine Rolle, dass Platte und CCD verschieden hell und verschieden kontrastreich sind.

**Die Maske.** Sterne bewegen sich in 60 Jahren kaum. Ein Stern, der in beiden Bildern an derselben Stelle steht, zieht das Maximum von $\rho$ zur Verschiebung null. Deshalb werden Sterne maskiert. Setzt man maskierte Pixel aber einfach auf null, stehen die Nullen in beiden Bildern an derselben Stelle und erzeugen denselben Fehler. Richtig ist, für jede Verschiebung nur die Pixel zu benutzen, die in **beiden** Bildern gültig sind: $V(\vec s)$ = gültig in der Vorlage und gültig im verschobenen Stück. So wird die Maske mitverschoben und erzeugt kein eigenes Signal (maskierte Kreuzkorrelation, Padfield 2012). Variation 1 zeigt, was mit Nullen passiert.

**Subpixel.** Das Maximum liegt bei einem ganzen Pixel $s_0$. Durch $\rho(s_0 - 1)$, $\rho(s_0)$, $\rho(s_0 + 1)$ legt man eine Parabel; ihr Scheitel liegt bei

$$\hat s = s_0 + \frac{\rho(s_0 - 1) - \rho(s_0 + 1)}{2\,\big(\rho(s_0 - 1) - 2\rho(s_0) + \rho(s_0 + 1)\big)}.$$

**Fachbegriffe:**
- **Vorlage** (template): der Ausschnitt, der gesucht wird; **Suchbereich**: das größere Stück des anderen Bildes, in dem gesucht wird.
- **Template Matching:** dieses Verfahren in der Bildverarbeitung; dieselbe Idee steckt in der Bewegungsschätzung von Videokompression.

##### *5. Homologe Expansion, Expansionszentrum und Konvergenzjahr*

Wenn jedes Filament seit dem Jahr $t_0$ geradlinig mit konstanter Geschwindigkeit vom Zentrum $\vec c$ wegfliegt, ist sein Abstand vom Zentrum proportional zur verstrichenen Zeit:

$$\vec p(t) - \vec c = \big(\vec p(t_2) - \vec c\big)\,\frac{t - t_0}{t_2 - t_0}.$$

Zwischen zwei Epochen $t_1 < t_2$ ist die Verschiebung dann

$$\Delta\vec p = \vec p(t_2) - \vec p(t_1) = k\,\big(\vec p(t_2) - \vec c\big), \qquad k = \frac{t_2 - t_1}{t_2 - t_0}.$$

Alle Filamente teilen denselben Faktor $k$: Die Verschiebung wächst linear mit dem Abstand. Schreibt man $k\vec c$ als eigene Unbekannte, ist $\Delta\xi = k\,\xi - (k c_\xi)$ und $\Delta\eta = k\,\eta - (k c_\eta)$ linear in den drei Unbekannten $k$, $kc_\xi$, $kc_\eta$. Aus $k$ folgt das **Konvergenzjahr**

$$t_0 = t_2 - \frac{t_2 - t_1}{k}.$$

Teilt man durch $\Delta t$, wird daraus ein Hubble-Gesetz: Geschwindigkeit $= e \cdot$ Abstand mit der Expansionsrate $e = 1/(t_2 - t_0)$. Genau wie das Hubble-Alter $1/H_0$ des Universums ist $1/e$ das Alter, das man bei konstanter Geschwindigkeit erhält.

**Warum die Methode der kleinsten Streuung zu späte Jahre gibt.** Rechnet man jeden Punkt mit seiner gemessenen Geschwindigkeit $\vec v_i$ zurück, $\vec p_i(t) = \vec p_i - \vec v_i\,(t_2 - t)$, dann ist der mittlere quadratische Radius der Punktwolke

$$R^2(t) = R_2^2\,\frac{(t - t_0)^2}{(t_2 - t_0)^2} + 2\sigma_v^2\,(t_2 - t)^2,$$

wobei $R_2$ der Radius heute ist und $\sigma_v$ der Messfehler jeder Geschwindigkeitskomponente. Der erste Term ist bei $t_0$ null, der zweite wächst, je weiter man zurückrechnet. Das Minimum der Summe liegt deshalb nicht bei $t_0$, sondern näher an heute: Setzt man die Ableitung null, folgt

$$t_\text{min} = t_0 + (t_2 - t_0)\,\frac{B}{A + B}, \qquad A = \frac{R_2^2}{(t_2 - t_0)^2},\quad B = 2\sigma_v^2.$$

Der Fit dagegen benutzt die Fehler nur als Streuung um die Gerade; ihre Steigung bleibt im Mittel richtig.

**Fachbegriffe:**
- **Homolog:** formgleich; der Nebel wird mit der Zeit nur größer, seine Form bleibt.
- **Scheinbares Alter** $1/e$: die Zeit, die die Filamente mit heutiger Geschwindigkeit für ihren heutigen Abstand gebraucht hätten.

##### *6. Beschleunigung durch den Pulsarwind*

Das Konvergenzjahr liegt seit der ersten Messung nach 1054. Das heißt: Die Filamente sind heute schneller als im Mittel seit der Explosion. Der Pulsarwindnebel drückt von innen gegen die Hülle und schiebt sie an. Für eine Blase, die eine Quelle mit konstanter Leistung in frei expandierende Ejekta bläst, wächst der Radius wie $r \propto t^{6/5}$ (Chevalier 1977); weil der Pulsar langsamer wird und die Filamente schwer sind, ist der Exponent für die Filamente kleiner.

Man beschreibt die Beschleunigung mit einem Potenzgesetz $r \propto (t - t_\text{SN})^\beta$. Dann ist die Geschwindigkeit $v = \dot r = \beta\, r/(t - t_\text{SN})$, das scheinbare Alter $r/v = (t - t_\text{SN})/\beta$, und das Konvergenzjahr

$$t_0 = t - \frac{t - t_\text{SN}}{\beta}.$$

Zum Beispiel: Mit $\beta = 1.05$ und $t - t_\text{SN} = 957$ Jahren ist das scheinbare Alter 911 Jahre, das Konvergenzjahr also 46 Jahre nach 1054. Zwischen zwei Epochen gilt für den Skalenfaktor

$$1 - k = \frac{r(t_1)}{r(t_2)} = \left(\frac{t_1 - t_\text{SN}}{t_2 - t_\text{SN}}\right)^\beta,$$

woraus der Code $\beta$ berechnet. Die heutige Geschwindigkeit ist $\beta$-mal so groß wie die mittlere seit der Explosion.

**Fachbegriffe:**
- **Exponent** $\beta$: $\beta = 1$ heißt unbeschleunigt, $\beta > 1$ beschleunigt, $\beta < 1$ gebremst (so bei Überresten, die Gas aus der Umgebung aufsammeln).

##### *7. Entfernung: Expansionsparallaxe und trigonometrische Parallaxe*

Ein Objekt in der Entfernung $d$ (in Parsec), das sich quer zur Sichtlinie mit $v_\perp$ (in km/s) bewegt, wandert am Himmel mit

$$\mu = \frac{v_\perp}{4.74\, d} \quad [″/\text{Jahr}].$$

Beim Krebsnebel misst man $\mu$ in den Bildern und die Geschwindigkeit entlang der Sichtlinie $v_r$ aus der Doppler-Verschiebung der Linien. Um beides zu verbinden, braucht man eine Annahme über die Form. Trimble (1973) nahm an, dass der Nebel ein Rotationsellipsoid ist, dessen lange Achse in der Himmelsebene liegt; dann ist er entlang der Sichtlinie so tief wie entlang seiner kurzen Achse $b$. Die schnellsten Filamente entlang der Sichtlinie bewegen sich dann so schnell wie die am Rand der kurzen Achse quer dazu, und mit $\mu_b = e\,b$ folgt

$$d = \frac{v_{r,\max}}{4.74\; e\; b}.$$

Das ist die **Expansionsparallaxe**. Unabhängig davon misst Gaia die trigonometrische Parallaxe $\varpi$ des Pulsars, $d = 1/\varpi$. Bei einem relativen Fehler von 15 % ist $1/\varpi$ eine leicht verzerrte Schätzung; man gibt deshalb besser die ganze Verteilung an (Bailer-Jones 2015), wie in der Abbildung zu Schritt 10.

**Fachbegriffe:**
- **Parsec:** die Entfernung, aus der 1 AE unter 1″ erscheint, 3.26 Lichtjahre.
- **Doppler-Verschiebung:** $\Delta\lambda/\lambda = v_r/c$; ein Filament, das mit 1500 km/s auf uns zukommt, zeigt Hα bei 653.6 statt 656.3 nm.

##### *8. Wie ein Pulsar bremst*

Ein Neutronenstern mit Trägheitsmoment $I$ und Rotationsfrequenz $\nu$ hat die Rotationsenergie $E = \tfrac12 I\,(2\pi\nu)^2 = 2\pi^2 I \nu^2$. Wird er langsamer, verliert er die Leistung

$$\dot E = -\frac{dE}{dt} = -4\pi^2 I\,\nu\,\dot\nu.$$

Ein rotierender magnetischer Dipol im Vakuum strahlt $\dot E = \frac{B^2 R^6 (2\pi\nu)^4 \sin^2\alpha}{6c^3}$ ab. Gleichsetzen ergibt $\dot\nu \propto -\nu^3$ und, für $R = 10$ km, $I = 10^{45}$ g cm² und senkrechte Achsen ($\alpha = 90°$), das Feld am Äquator $B \approx 3.2\cdot10^{19}\sqrt{P\dot P}$ Gauß.

Allgemein schreibt man das Bremsgesetz als $\dot\nu = -K\nu^n$ mit dem **Bremsindex** $n$. Leitet man einmal nach der Zeit ab, folgt $\ddot\nu = -Kn\nu^{n-1}\dot\nu$ und daraus

$$n = \frac{\nu\,\ddot\nu}{\dot\nu^2}.$$

Der Dipol gibt $n = 3$. Integriert man das Bremsgesetz von der Geburt mit der Anfangsperiode $P_0$ bis heute, ergibt sich das Alter

$$t = \frac{P}{(n - 1)\,\dot P}\left[1 - \left(\frac{P_0}{P}\right)^{n-1}\right].$$

Für $P_0 \ll P$ und $n = 3$ bleibt das **charakteristische Alter** $\tau_c = P/(2\dot P) = \nu/(2|\dot\nu|)$. Kennt man das wahre Alter wie beim Crab, lässt sich die Gleichung nach $P_0$ auflösen. Mit $P_0$ folgt auch die Energie, die der Pulsar seit seiner Geburt abgegeben hat: $2\pi^2 I\,(\nu_0^2 - \nu^2)$.

**Glitches** unterbrechen das gleichmäßige Bremsen. Die Frequenz springt in Sekunden um $\Delta\nu$ nach oben, danach erholt sich $\dot\nu$ über Wochen. Beim Crab bleibt nach jedem größeren Glitch ein kleiner Rest von $\dot\nu$ dauerhaft stärker. Über viele Glitches hinweg sieht das wie ein kleinerer Bremsindex aus.

**Fachbegriffe:**
- **Periode und Periodenzunahme:** $P = 1/\nu$, $\dot P = -\dot\nu/\nu^2$ (dimensionslos, Sekunden pro Sekunde).
- **Charakteristisches Alter** $\tau_c$: das Alter, das man aus $P$ und $\dot P$ allein schätzt. Es stimmt nur, wenn $n = 3$ und der Pulsar viel schneller geboren wurde, als er heute ist.

##### *9. Energiebilanz*

Die Bewegungsenergie der Filamente ist $E_\text{kin} = \tfrac12 M \langle v^2\rangle$. Gemessen wird nur die Geschwindigkeit quer zur Sichtlinie; bewegen sich die Filamente in alle Richtungen gleichermaßen, so ist

$$\langle v^2\rangle = \langle v_x^2 + v_y^2 + v_z^2\rangle = \tfrac32\,\langle v_\perp^2\rangle .$$

Ist die heutige Geschwindigkeit $\beta$-mal so groß wie die mittlere seit der Explosion, dann hätten die Filamente ohne Antrieb nur $E_\text{kin}/\beta^2$. Die zugeführte Energie ist mindestens

$$\Delta E = E_\text{kin}\left(1 - \frac{1}{\beta^2}\right) \approx 2(\beta - 1)\,E_\text{kin} \quad (\beta \text{ nahe } 1).$$

Dagegen steht, was der Pulsar abgibt: heute $\dot E$, seit 1054 die Differenz der Rotationsenergien (Abschnitt 8). Die Leistung verteilt sich auf Synchrotronstrahlung (heute etwa ein Drittel), auf die Arbeit, die die Blase an der Hülle verrichtet, und auf die Energie der Teilchen und des Magnetfelds, die in der Blase gespeichert bleibt.

**Fachbegriffe:**
- **Leuchtkraft:** die gesamte abgestrahlte Leistung in erg/s; die Sonne hat $3.83\cdot10^{33}$ erg/s.

#### **Daten**

- **Fotoplatten der Palomar-Durchmusterungen** in digitalisierter Form (Digitized Sky Survey, DSS): ein 12′ × 12′ großer Ausschnitt der roten Platte von POSS-I (Belichtung im November 1951) vom Archiv des Space Telescope Science Institute und der roten Platte von POSS-II (Oktober 1991) vom DSS-Server der ESO. Der Code fragt jeweils den anderen Server, wenn der erste nicht antwortet.
- **Pan-STARRS1** (PS1), gestapelte Bilder in den Filtern r und i aus Belichtungen von Dezember 2009 bis März 2014, als Ausschnitt vom Cutout-Dienst in Baltimore. Der Nebel liegt auf der Grenze zweier „Skycells“; der Code holt beide und setzt sie zusammen. Der Server mittelt dabei schon auf 0.5″ pro Pixel (8 statt 33 MB pro Bild).
- **Gaia DR3**: Positionen und Eigenbewegungen von 1171 Sternen mit G < 20 im Feld und die Parallaxe des Pulsars, über die TAP-Dienste der ESA oder, falls die nicht antwortet, des Astronomischen Rechen-Instituts in Heidelberg (GAVO).
- **Jodrell Bank**: die monatlichen Ephemeriden des Crab-Pulsars seit Februar 1982 (Frequenz ν und ihre Ableitung ν̇), zwei Textdateien von zusammen 50 kB (Lyne, Pritchard & Graham-Smith 1993, laufend ergänzt).

Alle Downloads landen einmalig im Ordner `astrodata/crab` neben dem Notebook (zusammen etwa 35 MB).

**Hintergrund & Fachbegriffe:**
- **POSS:** Palomar Observatory Sky Survey. POSS-I (1949 bis 1958) fotografierte den Nordhimmel mit dem 48-Zoll-Schmidt-Teleskop auf Glasplatten in Blau und Rot, POSS-II (1985 bis 2000) wiederholte das mit feinkörnigeren Emulsionen in Blau, Rot und nahem Infrarot.
- **Skycell:** Pan-STARRS teilt den Himmel in Projektionszellen von etwa 4° × 4°, diese in 10 × 10 Skycells. Alle Skycells einer Projektionszelle haben dieselbe Tangentialprojektion, ihre Pixel liegen auf einem gemeinsamen Gitter; deshalb lassen sich zwei Ausschnitte Pixel für Pixel zusammensetzen.
- **TAP und ADQL:** Table Access Protocol, eine Web-Schnittstelle für Kataloge; die Abfragesprache ADQL ist SQL mit Funktionen für Himmelskoordinaten (`CONTAINS`, `CIRCLE`).
- **Ephemeride:** die Tabelle der Größen, mit denen man die Ankunftszeit jedes Pulses vorausberechnen kann; hier Rotationsfrequenz ν und ihre zeitliche Ableitung ν̇ zur Mitte jedes Monats.

```python
import calendar
import datetime
import os
import re
import urllib.parse
import urllib.request
import warnings
from pathlib import Path

# Downloads landen im Ordner "astrodata" neben dem Notebook statt in den Caches im Home-Verzeichnis.
# Das muss vor dem ersten Import von astropy passieren; sonst den Kernel neu starten.
ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "astropy").mkdir(parents=True, exist_ok=True)
os.environ["XDG_CACHE_HOME"] = str(ASTRODATA)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from numpy.lib.stride_tricks import sliding_window_view
from scipy import ndimage, optimize, spatial
from astropy.io import fits
from astropy.time import Time
from astropy.wcs import WCS, FITSFixedWarning

warnings.simplefilter("ignore", FITSFixedWarning)     # die DSS-Köpfe enthalten veraltete Schlüsselwörter
RA0, DE0 = 83.63308, 22.01450      # Bezugspunkt aller Bilder: Crab-Pulsar (ICRS, gerundet)
HALF = 360                         # halbe Kantenlänge des gemeinsamen Rasters [″], 1″ pro Pixel (Schritt 5)
FWHM_COMMON = 4.4                  # gemeinsame Bildschärfe nach dem Angleichen [″] (Schritt 5)
HP_SIGMA = 6.0                     # Breite des Hochpasses [″]: Strukturen größer als das werden entfernt
PATCH, SEARCH, STEP = 10, 20, 8    # halbe Ausschnittgröße, Suchradius und Abstand der Ausschnitte [″] (Schritt 6)
PEAK_MIN = 0.6                     # Mindestkorrelation, ab der eine Verschiebung zählt
T_SN = 1054.5                      # Supernova: erstmals gesehen am 4. Juli 1054
I_NS = 1e45                        # Trägheitsmoment des Neutronensterns [g cm²]
V_R_MAX = 1500.0                   # größte Radialgeschwindigkeit der Filamente [km/s] (Martin et al. 2021)
M_FIL = (4.6, 1.8)                 # Masse der Filamente und Fehler [Sonnenmassen] (Fesen et al. 1997)
YEAR = 365.25 * 86400              # Sekunden pro Jahr
M_SUN = 1.989e33                   # Sonnenmasse [g]
CRAB = ASTRODATA / "crab"
rng = np.random.default_rng(42)
pd.set_option("display.width", 160)


def download(urls, path, check=None):
    # lädt eine Datei vom ersten Server, der antwortet, und liest sie danach aus dem Cache
    path = Path(path)
    if not path.exists():
        path.parent.mkdir(parents=True, exist_ok=True)
        for url in urls:
            try:
                with urllib.request.urlopen(url, timeout=120) as r:
                    data = r.read()
            except OSError:
                continue
            if check is None or check(data):
                path.write_bytes(data)
                break
        else:
            raise OSError(f"Kein Server liefert {path.name}")
    return path


fits_ok = lambda d: d[:6] == b"SIMPLE"


def tap(adql, path):
    # Gaia DR3 über TAP: erst ESA, dann GAVO in Heidelberg (dort heißt die Tabelle gaia.dr3lite)
    services = [("https://gea.esac.esa.int/tap-server/tap/sync", "gaiadr3.gaia_source"),
                ("https://dc.zah.uni-heidelberg.de/tap/sync", "gaia.dr3lite")]
    urls = [s + "?" + urllib.parse.urlencode({"REQUEST": "doQuery", "LANG": "ADQL", "FORMAT": "csv",
                                              "QUERY": adql.format(table=t)}) for s, t in services]
    return pd.read_csv(download(urls, path, check=lambda d: d.startswith(b"source_id")))


def plate_epoch(h):
    # Datum der Belichtung als Dezimaljahr; die Server schreiben 1951-11-07, 1991/10/18 oder 08/11/51
    s = str(h["DATE-OBS"]).strip()
    m = re.match(r"(\d{4})[-/](\d{2})[-/](\d{2})", s)
    if m:
        y, mo, d = map(int, m.groups())
    else:
        d, mo, y = map(int, re.match(r"(\d{2})/(\d{2})/(\d{2})", s).groups())
        y += 1900
    day = (datetime.date(y, mo, d) - datetime.date(y, 1, 1)).days + 0.5
    return y + day / (366 if calendar.isleap(y) else 365)


def load_plate(path):
    with fits.open(path) as h:
        h.verify("silentfix")                        # ESO schreibt eine Kopfzeile, die nicht dem Standard folgt
        return h[0].data.astype(float), h[0].header.copy()


# Fotoplatten: POSS-I rot (1951) und POSS-II rot (1991), 12′ × 12′ um den Pulsar
STSCI = "https://archive.stsci.edu/cgi-bin/dss_search?r={}&d={}&e=J2000&h=12&w=12&f=fits&c=none&v={}"
ESO = "https://archive.eso.org/dss/dss/image?ra={}&dec={}&equinox=J2000&x=12&y=12&mime-type=application/x-fits&Sky-Survey={}"
PLATE_FILES = {
    "POSS-I E": download([STSCI.format(RA0, DE0, "poss1_red"), ESO.format(RA0, DE0, "DSS1")], CRAB / "poss1_red.fits", fits_ok),
    "POSS-II F": download([ESO.format(RA0, DE0, "DSS2-red"), STSCI.format(RA0, DE0, "poss2ukstu_red")], CRAB / "poss2_red.fits", fits_ok),
}

# Pan-STARRS1: zwei Skycells pro Filter, 12′ × 12′, auf 0.5″ pro Pixel gemittelt
PS1 = "https://ps1images.stsci.edu/cgi-bin/"


def ps1_part(band, dec):
    # Skycell, die den Punkt (RA0, dec) enthält, ausgeschnitten um den Pulsar
    lst = download([f"{PS1}ps1filenames.py?ra={RA0}&dec={dec}&filters={band}"], CRAB / f"ps1_{band}_{dec:.3f}.txt")
    fn = pd.read_csv(lst, sep=r"\s+")["filename"].iloc[0]
    p = download([f"{PS1}fitscut.cgi?ra={RA0}&dec={DE0}&size=2880&output_size=1440&format=fits&red={fn}"],
                 CRAB / Path(fn).name, fits_ok)
    with fits.open(p) as h:
        hdr = h[0].header.copy()
        img = h[0].data.astype(float) / hdr["EXPTIME"]           # Zählrate: gleiche Skala in beiden Skycells
    mjd = np.array([50000 + int(hdr[k][1:5]) for k in hdr if k.startswith("INP_")])   # o5814g... = MJD 55814
    texp = np.array([hdr[k] for k in hdr if re.fullmatch(r"EXP_\d{4}", k)])
    return img, hdr, mjd, texp


def ps1_mosaic(band):
    north, hdr, mjd, texp = ps1_part(band, DE0)
    south, _, mjd_s, texp_s = ps1_part(band, DE0 - 0.08)       # die Grenze liegt 160″ südlich des Pulsars
    both = np.isfinite(north) & np.isfinite(south)
    p = np.polyfit(south[both], north[both], 1)                # Untergrund und Skala im Überlapp angleichen
    img = np.where(np.isfinite(north), north, np.polyval(p, south))
    img[~np.isfinite(img)] = np.nanmedian(img)
    epoch = Time(np.average(mjd, weights=texp), format="mjd").decimalyear
    return img, hdr, epoch, dict(mjd=mjd, texp=texp, mjd_s=mjd_s, texp_s=texp_s, overlap=both.mean(), scale=p[0])


# Gaia DR3: Feldsterne und Pulsar
gaia = tap("SELECT source_id, ra, dec, pmra, pmdec, parallax, phot_g_mean_mag FROM {table} "
           "WHERE CONTAINS(POINT(ra, dec), CIRCLE(83.63308, 22.0145, 0.125)) = 1 AND phot_g_mean_mag < 20",
           CRAB / "gaia_field.csv")
PSR_ID = 3403818172572314624                                  # Crab-Pulsar in Gaia DR3
PSR = tap("SELECT source_id, ra, dec, parallax, parallax_error, pmra, pmra_error, pmdec, pmdec_error, "
          f"phot_g_mean_mag, ruwe FROM {{table}} WHERE source_id = {PSR_ID}", CRAB / "gaia_pulsar.csv").iloc[0]

# Jodrell Bank: monatliche Ephemeriden des Pulsars (1982 bis heute)
JB_FILES = [download([f"https://www.jb.man.ac.uk/pulsar/crab/{f}"], CRAB / f) for f in ("crab.txt", "crab2.txt")]
print("Geladen:", ", ".join(p.name for p in sorted(CRAB.iterdir())))
```

```python
# Werkzeuge 1: Sterne finden, Astrometrie gegen Gaia, Umrechnen auf ein gemeinsames Raster

def tangent(ra, dec):
    # gnomonische Projektion um (RA0, DE0) in Bogensekunden: xi nach Osten, eta nach Norden
    a, d, a0, d0 = map(np.radians, (ra, dec, RA0, DE0))
    c = np.sin(d) * np.sin(d0) + np.cos(d) * np.cos(d0) * np.cos(a - a0)
    xi = np.cos(d) * np.sin(a - a0) / c
    eta = (np.sin(d) * np.cos(d0) - np.cos(d) * np.sin(d0) * np.cos(a - a0)) / c
    return np.degrees(xi) * 3600, np.degrees(eta) * 3600


def gaia_at(epoch, gmax=20.0):
    # Gaia-Positionen (Epoche 2016.0) mit der Eigenbewegung auf eine andere Epoche geschoben
    g = gaia[gaia["phot_g_mean_mag"] < gmax]
    dt = epoch - 2016.0
    dec = g["dec"] + g["pmdec"].fillna(0) * dt / 3.6e6
    ra = g["ra"] + g["pmra"].fillna(0) * dt / 3.6e6 / np.cos(np.radians(g["dec"]))
    xi, eta = tangent(ra.values, dec.values)
    return ra.values, dec.values, xi, eta, g["phot_g_mean_mag"].values


def find_stars(img, fwhm, nsig=6, edge=10):
    # lokale Maxima im geglätteten, hochpassgefilterten Bild; Lage mit gaußgewichtetem Schwerpunkt
    bg = ndimage.gaussian_filter(img, 4 * fwhm)
    hp = ndimage.gaussian_filter(img, fwhm / 2.355) - bg
    mad = 1.4826 * np.median(np.abs(hp - np.median(hp)))
    peak = (hp == ndimage.maximum_filter(hp, size=int(2 * fwhm) + 1)) & (hp > nsig * mad)
    y, x = np.nonzero(peak)
    keep = (x > edge) & (y > edge) & (x < img.shape[1] - edge) & (y < img.shape[0] - edge)
    x, y = x[keep].astype(float), y[keep].astype(float)
    s = fwhm / 2.355
    r = int(np.ceil(3 * s))
    yy, xx = np.mgrid[-r:r + 1, -r:r + 1]
    for _ in range(10):
        nx, ny = x.copy(), y.copy()
        for i, (xi, yi) in enumerate(zip(x, y)):
            ix, iy = int(round(xi)), int(round(yi))
            cut = np.clip(img[iy - r:iy + r + 1, ix - r:ix + r + 1] - bg[iy - r:iy + r + 1, ix - r:ix + r + 1], 0, None)
            dx, dy = xx + ix - xi, yy + iy - yi
            w = np.exp(-(dx ** 2 + dy ** 2) / (2 * s * s)) * cut
            if w.sum() > 0:
                nx[i] += 2 * (w * dx).sum() / w.sum()
                ny[i] += 2 * (w * dy).sum() / w.sum()
        x, y = np.clip(nx, edge, img.shape[1] - edge - 1), np.clip(ny, edge, img.shape[0] - edge - 1)
    return x, y


def design(u, v, order):
    # Spalten u^i v^j mit i + j <= order (Polynom zweiten Grades: 6 Spalten)
    return np.vstack([u ** i * v ** j for i in range(order + 1) for j in range(order + 1 - i)]).T


def fit_poly(u, v, a, b, order=2, clip=3.0):
    # (a, b) als Polynom von (u, v) mit Kleinste-Quadrate und Sigma-Clipping
    keep = np.ones(len(u), bool)
    for _ in range(5):
        A = design(u[keep], v[keep], order)
        ca = np.linalg.lstsq(A, a[keep], rcond=None)[0]
        cb = np.linalg.lstsq(A, b[keep], rcond=None)[0]
        res = np.hypot(design(u, v, order) @ ca - a, design(u, v, order) @ cb - b)
        keep = res < clip * np.sqrt(np.mean(res[keep] ** 2))
    return ca, cb, res, keep


def astrometry(img, wcs, epoch, fwhm, pix, gmin, gmax=19.0, rmin=250.0, order=2):
    # Sterne im Bild finden, mit Gaia zur Epoche des Bildes paaren, Pixel -> (xi, eta) als Polynom anpassen.
    # Nur Sterne außerhalb von rmin, damit der Nebel die Schwerpunkte nicht verfälscht.
    x, y = find_stars(img, fwhm)
    ra, dec, xi, eta, G = gaia_at(epoch)
    gx, gy = wcs.world_to_pixel_values(ra, dec)
    dist, j = spatial.cKDTree(np.c_[x, y]).query(np.c_[gx, gy], distance_upper_bound=3.0 / pix)
    sel = np.isfinite(dist) & (G > gmin) & (G < gmax) & (np.hypot(xi, eta) > rmin)
    xs, ys = x[j[sel]], y[j[sel]]
    ca, cb, res, keep = fit_poly(xs, ys, xi[sel], eta[sel], order)
    ia, ib, _, _ = fit_poly(xi[sel][keep], eta[sel][keep], xs[keep], ys[keep], order)   # Umkehrung für das Raster
    xi_h, eta_h = tangent(*wcs.pixel_to_world_values(xs, ys))   # was die Plattenlösung im Kopf aus den Sternen macht
    A = design(xs, ys, order)
    return dict(fwd=(ca, cb), inv=(ia, ib), order=order, res=res, keep=keep, G=G[sel], xi=xi[sel], eta=eta[sel],
                hdr_vec=np.c_[xi_h - xi[sel], eta_h - eta[sel]], fit_vec=np.c_[A @ ca - xi[sel], A @ cb - eta[sel]],
                rms=np.sqrt(np.mean(res[keep] ** 2)), n=keep.sum(), n_found=len(x))


def resample(img, ast, presmooth=0.0):
    # Bild auf das gemeinsame Raster (1″ pro Pixel, Osten links, Norden oben) umrechnen
    t = np.arange(-HALF, HALF + 1.0)
    XI, ETA = np.meshgrid(-t, t)
    A = design(XI.ravel(), ETA.ravel(), ast["order"])
    x = (A @ ast["inv"][0]).reshape(XI.shape)
    y = (A @ ast["inv"][1]).reshape(XI.shape)
    src = ndimage.gaussian_filter(img, presmooth) if presmooth else img
    return ndimage.map_coordinates(src, [y, x], order=3, mode="nearest")


def to_grid(xi, eta):
    # (xi, eta) in ″ -> (Spalte, Zeile) im gemeinsamen Raster
    return HALF - np.asarray(xi), HALF + np.asarray(eta)


EXTENT = [HALF + 0.5, -HALF - 0.5, -HALF - 0.5, HALF + 0.5]      # für imshow: Osten links


def show(ax, im, lo=5, hi=99.7, gamma=0.6, cmap="gray"):
    v0, v1 = np.percentile(im, [lo, hi])
    return ax.imshow(np.clip((im - v0) / (v1 - v0), 0, 1) ** gamma, origin="lower", cmap=cmap, extent=EXTENT)
```

```python
# Werkzeuge 2: Bilder angleichen, Verschiebungen messen, Expansion anpassen

def prep(name, fwhm_target=FWHM_COMMON, hp_sigma=HP_SIGMA, star_scale=1.0):
    # gleiche Schärfe für alle Epochen, Hochpass, Sterne ausmaskieren (Maske True = Pixel nicht benutzen)
    im = GRID[name]
    s = np.sqrt(max(fwhm_target ** 2 - FWHM[name] ** 2, 0.0)) / 2.355
    im = ndimage.gaussian_filter(im, s) if s > 0 else im
    hp = im - ndimage.gaussian_filter(im, hp_sigma)
    _, _, xi, eta, G = gaia_at(EPOCH[name])
    col, row = to_grid(xi, eta)
    rad = star_scale * np.where(G > 17, 4.0, np.where(G > 14, 6.0, 9.0))
    yy, xx = np.mgrid[0:2 * HALF + 1, 0:2 * HALF + 1]
    mask = np.zeros(im.shape, bool)
    for c, r, q in zip(col, row, rad):
        i0, i1 = int(max(r - q, 0)), int(min(r + q + 1, 2 * HALF + 1))
        j0, j1 = int(max(c - q, 0)), int(min(c + q + 1, 2 * HALF + 1))
        if i0 < i1 and j0 < j1:
            mask[i0:i1, j0:j1] |= (xx[i0:i1, j0:j1] - c) ** 2 + (yy[i0:i1, j0:j1] - r) ** 2 <= q * q
    return hp, mask


def masked_ncc(tpl, tpl_ok, region, region_ok, min_frac=0.6):
    # normierte Kreuzkorrelation für jede Verschiebung; es zählen nur Pixel, die in BEIDEN Bildern gültig sind
    W = sliding_window_view(region, tpl.shape)
    V = sliding_window_view(region_ok, tpl.shape) & tpl_ok
    n = V.sum((-1, -2)).astype(float)
    T = np.where(V, tpl, 0.0)
    R = np.where(V, W, 0.0)
    st, sr = T.sum((-1, -2)), R.sum((-1, -2))
    with np.errstate(divide="ignore", invalid="ignore"):
        cov = (T * R).sum((-1, -2)) - st * sr / n
        var = ((T * T).sum((-1, -2)) - st * st / n) * ((R * R).sum((-1, -2)) - sr * sr / n)
        ncc = cov / np.sqrt(np.clip(var, 1e-30, None))
    ncc[n < min_frac * tpl.size] = -1
    return ncc


def subpixel(ncc, iy, ix):
    # Parabel durch das Maximum und seine Nachbarn, getrennt in beiden Richtungen
    def vertex(a, b, c):
        d = a - 2 * b + c
        return 0.5 * (a - c) / d if d < 0 else 0.0
    return iy + vertex(ncc[iy - 1, ix], ncc[iy, ix], ncc[iy + 1, ix]), ix + vertex(ncc[iy, ix - 1], ncc[iy, ix], ncc[iy, ix + 1])


def measure(late, early, patch=PATCH, search=SEARCH, step=STEP, use_mask=True):
    # Ausschnitte des späten Bildes im frühen Bild suchen; Ergebnis: Verschiebung früh -> spät in ″ (Osten, Norden)
    (h2, m2), (h1, m1) = late, early
    rows = []
    centers = np.arange(search + patch, 2 * HALF + 1 - search - patch, step)
    for r in centers:
        for c in centers:
            if not NEB[r, c]:
                continue
            ok2 = ~m2[r - patch:r + patch + 1, c - patch:c + patch + 1]
            if ok2.mean() < 0.7:
                continue
            tpl = h2[r - patch:r + patch + 1, c - patch:c + patch + 1]
            win = (slice(r - patch - search, r + patch + search + 1), slice(c - patch - search, c + patch + search + 1))
            if use_mask:
                ncc = masked_ncc(tpl, ok2, h1[win], ~m1[win])
            else:                                             # Variation: Sterne als Nullen statt ausgelassen
                ncc = masked_ncc(np.where(ok2, tpl, 0.0), np.ones_like(ok2), np.where(m1[win], 0.0, h1[win]),
                                 np.ones_like(m1[win]))
            iy, ix = np.unravel_index(np.argmax(ncc), ncc.shape)
            if not (0 < iy < ncc.shape[0] - 1 and 0 < ix < ncc.shape[1] - 1):
                continue                                      # Maximum am Rand des Suchbereichs: unsicher
            sy, sx = subpixel(ncc, iy, ix)
            rows.append(dict(xi=HALF - c, eta=r - HALF, peak=ncc[iy, ix], dxi=sx - search, deta=-(sy - search)))
    return pd.DataFrame(rows)


def fit_expansion(df, clip=3.0):
    # Verschiebung = k * (Ort - Zentrum): linear in (k, k*cx, k*cy); Ausreißer werden iterativ verworfen
    keep = np.ones(len(df), bool)
    for _ in range(5):
        d = df[keep]
        n = len(d)
        A = np.zeros((2 * n, 3))
        A[:n, 0], A[:n, 1], A[n:, 0], A[n:, 2] = d["xi"], -1, d["eta"], -1
        k, ax, ay = np.linalg.lstsq(A, np.r_[d["dxi"], d["deta"]], rcond=None)[0]
        res = np.hypot(df["dxi"] - (k * df["xi"] - ax), df["deta"] - (k * df["eta"] - ay)).values
        keep = res < clip * 1.4826 * np.median(res[keep])
    return dict(k=k, cx=ax / k, cy=ay / k, keep=keep, res=res, n=keep.sum())


def pair(early, late, n_boot=200, **kw):
    # Verschiebungen zwischen zwei Epochen, Expansion anpassen, Konvergenzjahr mit Bootstrap-Fehler
    H_ = kw.pop("H", H)
    peak_min = kw.pop("peak_min", PEAK_MIN)
    df = measure(H_[late], H_[early], **kw)
    df = df[df["peak"] > peak_min].reset_index(drop=True)
    dt = EPOCH[late] - EPOCH[early]
    f = fit_expansion(df)
    boot = []
    for _ in range(n_boot):
        b = fit_expansion(df.iloc[rng.integers(0, len(df), len(df))].reset_index(drop=True))
        boot.append((EPOCH[late] - dt / b["k"], b["cx"], b["cy"]))
    boot = np.array(boot).reshape(-1, 3)
    err = boot.std(axis=0) if n_boot else np.full(3, np.nan)
    return dict(df=df, dt=dt, t0=EPOCH[late] - dt / f["k"], t0_err=err[0], cx_err=err[1], cy_err=err[2], boot=boot, **f)
```

Wie sehen die drei Bilder des Krebsnebels aus, bevor man sie bearbeitet, und worin unterscheiden sie sich außer im Datum? Vorher vorhersagen: Um wie viele Bogensekunden hat sich ein Filament am Rand des Nebels (gut 3′ vom Zentrum) zwischen 1951 und 2011 bewegt, wenn der Nebel seit 1054 mit konstanter Geschwindigkeit wächst?

Die beiden Fotoplatten und das Pan-STARRS-Mosaik werden geladen, jedes mit seiner eigenen Plattenlösung aus dem FITS-Kopf und seinem Aufnahmedatum. Gezeigt werden die ganzen Ausschnitte in ihren eigenen Pixeln und darunter ein Ausschnitt von 80″ × 80″ um ein helles Filament südlich des Pulsars.

**Hintergrund & Fachbegriffe:**
- **Plattenlösung:** die Umrechnung von Pixel auf Himmelskoordinaten. Für die DSS-Platten steht sie als Polynom im FITS-Kopf (Schlüsselwörter `AMDX`, `AMDY`), für Pan-STARRS als Tangentialprojektion (`CTYPE = RA---TAN`); astropy liest beide.
- **Emulsion und Filter:** Die Emulsion legt fest, bis zu welcher Wellenlänge die Platte empfindlich ist, das Glasfilter davor, ab welcher. 103aE mit rotem Plexiglas (POSS-I) sieht 620 bis 670 nm, IIIaF mit RG610 (POSS-II) 610 bis 690 nm, der Pan-STARRS-Filter r 550 bis 690 nm. In allen drei Bändern liegen die Linien Hα 656 nm, [N II] 655/658 nm und, bei POSS-II und PS1, [S II] 672/673 nm und [O I] 630 nm.
- **Seeing:** die Unschärfe durch die Luftunruhe, angegeben als Halbwertsbreite (FWHM) eines Sternbildes; auf den Platten kommt die Körnung der Emulsion und die Abtastung des Scanners dazu.

```python
DESC = {"POSS-I E": "103aE + rotes Plexiglas, 620–670 nm", "POSS-II F": "IIIaF + RG610, 610–690 nm",
        "PS1 r": "CCD, Filter r, 550–690 nm", "PS1 i": "CCD, Filter i, 690–820 nm"}
RAW = {}
for name, p in PLATE_FILES.items():
    img, hdr = load_plate(p)
    RAW[name] = dict(img=img, hdr=hdr, wcs=WCS(hdr), epoch=plate_epoch(hdr),
                     pix=hdr["XPIXELSZ"] * hdr["PLTSCALE"] / 1000, label=f"Platte {hdr['PLATEID'].strip()}, {hdr['EXPOSURE']:.0f} min")
for band in ("r", "i"):
    img, hdr, epoch, info = ps1_mosaic(band)
    first, last = Time(info["mjd"].min(), format="mjd"), Time(info["mjd"].max(), format="mjd")
    RAW[f"PS1 {band}"] = dict(img=img, hdr=hdr, wcs=WCS(hdr), epoch=epoch, pix=abs(hdr["CDELT1"]) * 3600, info=info,
                              label=f"{len(info['mjd'])} Belichtungen, {info['texp'].sum():.0f} s, {first.iso[:7]} bis {last.iso[:7]}")
EPOCH = {k: v["epoch"] for k, v in RAW.items()}

print(f"{'Bild':10s} {'Epoche':>8s}  {'Pixel':>6s}  {'Größe':>9s}  Band und Belichtung")
for k, v in RAW.items():
    print(f"{k:10s} {v['epoch']:8.2f}  {v['pix']:5.2f}″  {v['img'].shape[1]:4d}×{v['img'].shape[0]:<4d}  {DESC[k]}; {v['label']}")
info = RAW["PS1 r"]["info"]
print(f"PS1-Mosaik: Skycells im Überlapp angeglichen (Skala {info['scale']:.3f}), Überlapp {info['overlap']:.0%} des Ausschnitts")
print(f"Zeitbasis 1951 -> PS1 r: {EPOCH['PS1 r'] - EPOCH['POSS-I E']:.2f} Jahre")

ZOOM = (15.0, -75.0, 40.0)           # Mitte (xi, eta) und halbe Breite des Zooms [″]: helles Filament südlich des Pulsars
fig, axes = plt.subplots(2, 3, figsize=(15, 10), gridspec_kw={"height_ratios": [1, 1]})
for j, name in enumerate(["POSS-I E", "POSS-II F", "PS1 r"]):
    v = RAW[name]
    im = v["img"]
    lo, hi = np.percentile(im, [5, 99.7])
    axes[0, j].imshow(np.clip((im - lo) / (hi - lo), 0, 1) ** 0.6, origin="lower", cmap="gray")
    axes[0, j].set_title(f"{name} ({v['epoch']:.1f}), {v['pix']:.2f}″ pro Pixel", fontsize=10)
    axes[0, j].set_xlabel("x [Pixel]")
    axes[0, j].set_ylabel("y [Pixel]")
    # Zoom: Ecken des Himmelsausschnitts über die Plattenlösung in Pixel umrechnen
    xc, yc, w = ZOOM
    d = w / 3600
    ra_c = RA0 + xc / 3600 / np.cos(np.radians(DE0))
    de_c = DE0 + yc / 3600
    px, py = v["wcs"].world_to_pixel_values([ra_c + d / np.cos(np.radians(DE0)), ra_c - d / np.cos(np.radians(DE0))],
                                            [de_c - d, de_c + d])
    x0, x1 = int(min(px)), int(max(px)) + 1
    y0, y1 = int(min(py)), int(max(py)) + 1
    cut = im[y0:y1, x0:x1]
    lo, hi = np.percentile(cut, [1, 99.8])
    axes[1, j].imshow(np.clip((cut - lo) / (hi - lo), 0, 1), origin="lower", cmap="gray", interpolation="nearest",
                      extent=[x0 - 0.5, x1 - 0.5, y0 - 0.5, y1 - 0.5])
    axes[1, j].set_title(f"Zoom 80″ × 80″ um ξ = {xc:.0f}″, η = {yc:.0f}″", fontsize=10)
    axes[1, j].set_xlabel("x [Pixel]")
    axes[1, j].set_ylabel("y [Pixel]")
    axes[0, j].add_patch(plt.Rectangle((x0, y0), x1 - x0, y1 - y0, fill=False, ec="orange", lw=1.2))
fig.suptitle("Rohdaten: POSS-I (1951), POSS-II (1991), Pan-STARRS (2009–2014), jeweils 12′ × 12′ um den Pulsar", fontsize=11)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Die Platte von POSS-I (070U, 50 min) stammt vom 7. November 1951 (Epoche 1951.85), die von POSS-II (A1YP, 65 min) vom 18. Oktober 1991. Die beiden Skycells von Pan-STARRS überlappen auf 17 % des Ausschnitts und unterscheiden sich in der Zählrate um 0.3 %.

##### *2. Filamente und Synchrotronlicht*

Welches Licht im Bild kommt aus den Filamenten und welches aus der Synchrotronstrahlung des Pulsarwindnebels? Vorher vorhersagen: Ist das Synchrotronlicht im Filter i heller oder schwächer als im Filter r, wenn sein Spektrum wie $F_\nu \propto \nu^{-0.8}$ fällt und der Staub davor E(B−V) = 0.52 mag Rötung erzeugt?

Der Filter i (690 bis 820 nm) enthält keine helle Emissionslinie des Nebels; dort sieht man fast nur das Kontinuum der Synchrotronstrahlung. Im Filter r kommen Hα, [N II], [S II] und [O I] aus den Filamenten dazu. Zieht man vom r-Bild das mit einem Faktor $k$ skalierte i-Bild ab, bleibt das Linienlicht übrig. Den Faktor bestimmt der Code aus den Pixeln, in denen das Verhältnis r/i am kleinsten ist, also fast kein Linienlicht steckt (5. Perzentil im hellen Innern). Er wird mit dem Faktor verglichen, den ein Synchrotronspektrum mit Rötung erwarten lässt (Theorie, Abschnitt 1).

**Hintergrund & Fachbegriffe:**
- **Synchrotronstrahlung:** Licht von Elektronen, die sich fast mit Lichtgeschwindigkeit in einem Magnetfeld auf Schraubenbahnen bewegen. Ein Potenzgesetz der Elektronenenergien ergibt ein Potenzgesetz im Spektrum, $F_\nu \propto \nu^{-\alpha}$ mit dem Spektralindex $\alpha$.
- **Rötung** E(B−V): Staub schwächt blaues Licht mehr als rotes. Für die Pan-STARRS-Filter gilt $A_r = 2.27\,E(B-V)$ und $A_i = 1.68\,E(B-V)$ (Schlafly & Finkbeiner 2011).
- **Zählrate und AB-Magnituden:** Pan-STARRS eicht beide Filter auf denselben Nullpunkt (25 mag bei 1 Zählung pro Sekunde, AB-System). Ein Verhältnis der Zählraten ist deshalb direkt ein Verhältnis der Flussdichten $F_\nu$.

```python
r_img, i_img = RAW["PS1 r"]["img"], RAW["PS1 i"]["img"]          # dasselbe Pixelgitter (0.5″), Zählrate
rs, is_ = ndimage.gaussian_filter(r_img, 2), ndimage.gaussian_filter(i_img, 2)
# Sterne ausmaskieren: rote Vordergrundsterne haben selbst ein kleines r/i
ra_s, de_s, _, _, G_s = gaia_at(RAW["PS1 r"]["epoch"], 21)
gx, gy = RAW["PS1 r"]["wcs"].world_to_pixel_values(ra_s, de_s)
stars = np.zeros(r_img.shape, bool)
stars[np.clip(np.round(gy).astype(int), 0, r_img.shape[0] - 1), np.clip(np.round(gx).astype(int), 0, r_img.shape[1] - 1)] = True
for gmax, rad in [(21, 6), (17, 10), (14, 18)]:                  # Radius in Pixeln zu 0.5″, größer für helle Sterne
    sel = np.zeros(r_img.shape, bool)
    sel[np.clip(np.round(gy[G_s < gmax]).astype(int), 0, r_img.shape[0] - 1), np.clip(np.round(gx[G_s < gmax]).astype(int), 0, r_img.shape[1] - 1)] = True
    stars |= ndimage.binary_dilation(sel, iterations=rad)
inner = (is_ > np.percentile(is_, 97)) & ~stars                     # helles Inneres des Nebels ohne Sterne
K_LINE = np.percentile((rs / is_)[inner], 5)                         # wenig Linienlicht: kleinstes r/i
line = r_img - K_LINE * i_img

# Erwartung für Synchrotronlicht: F_nu ~ nu^-alpha, also r/i = (nu_r / nu_i)^-alpha = (lambda_r / lambda_i)^alpha, dazu die Rötung
LAM_R, LAM_I, EBV = 617.0, 750.0, 0.52                              # Schwerpunkte der Filter [nm], Rötung des Krebsnebels
redden = 10 ** (-0.4 * (2.27 - 1.68) * EBV)
for alpha in (0.6, 0.8, 1.0):
    print(f"Synchrotron mit alpha = {alpha}: r/i = {(LAM_R / LAM_I) ** alpha:.3f} ohne Staub, {(LAM_R / LAM_I) ** alpha * redden:.3f} mit E(B-V) = {EBV}")
alpha_meas = np.log(K_LINE / redden) / np.log(LAM_R / LAM_I)
k10 = np.percentile((rs / is_)[inner], 10)
print(f"gemessen: k = 5. Perzentil von r/i im Inneren (ohne Sterne) = {K_LINE:.3f} -> Spektralindex alpha = {alpha_meas:.2f} (mit Rötung); "
      f"mit dem 10. Perzentil {k10:.3f} -> alpha = {np.log(k10 / redden) / np.log(LAM_R / LAM_I):.2f}")
frac = np.clip(line, 0, None).sum() / np.clip(r_img - np.median(r_img), 0, None).sum()
print(f"Anteil des Linienlichts am r-Licht des Feldes: {frac:.0%}")

w = RAW["PS1 r"]["wcs"]
ny, nx = r_img.shape
pcx, pcy = w.world_to_pixel_values(RA0, DE0)                    # Pixel des Bezugspunkts im PS1-Bild
ext = [(pcx + 0.5) * 0.5, -(nx - pcx - 0.5) * 0.5, -(pcy + 0.5) * 0.5, (ny - pcy - 0.5) * 0.5]   # ″, Osten links
fig, axes = plt.subplots(1, 3, figsize=(16, 5.6))
for ax, im, title, cmap in [(axes[0], i_img, "PS1 i: fast nur Synchrotronlicht", "gray"),
                            (axes[1], r_img, "PS1 r: Synchrotron + Linien", "gray"),
                            (axes[2], line, f"r − {K_LINE:.2f}·i: nur Linienlicht (Filamente)", "magma")]:
    lo, hi = np.percentile(im, [20, 99.8])
    ax.imshow(np.clip((im - lo) / (hi - lo), 0, 1) ** 0.6, origin="lower", cmap=cmap, extent=ext)
    ax.set_xlim(240, -240)
    ax.set_ylim(-200, 200)
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("ξ [″] (Osten links)")
    ax.set_ylabel("η [″]")
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Für ein Synchrotronspektrum mit α = 0.8 erwartet man ohne Staub r/i = 0.86. Weil r stärker gerötet ist als i, sinkt das mit E(B−V) = 0.52 auf 0.65; das Synchrotronlicht ist also im Filter i heller, wie vorhergesagt. Gemessen ist das 5. Perzentil von r/i im hellen Inneren ohne Sterne 0.618, also α = 1.02, mit dem 10. Perzentil 0.643, also α = 0.81. Beides liegt im Bereich, den Spektren des Nebels zeigen (flach in der Mitte, steil am Rand). Ohne Sternmaske läge das 5. Perzentil bei 0.60, weil rote Vordergrundsterne selbst ein kleines r/i haben.
* Die **linke Abbildung** (i) zeigt den Synchrotronnebel als glatte, wolkige Fläche, hell um den Pulsar und nach Nordwesten. Die **mittlere** (r) zeigt dieselbe Fläche, dazu ein Netz aus Fäden: die Filamente.
* Die **rechte Abbildung** (r − 0.62·i) zeigt nur das Linienlicht: Die Filamente bilden eine Hülle mit einem dichten Netz im Norden und dem hellen Filament südlich des Pulsars. Ein Drittel (32 %) des r-Lichts im Feld kommt aus Linien. Nahe dem Pulsar bleiben dunkle Flecken. Vermutlich liegt das an den Wisps, Synchrotronstrukturen, die sich dort mit etwa halber Lichtgeschwindigkeit bewegen (rund 15″ pro Jahr): Die i-Belichtungen sind im Mittel 1.3 Jahre jünger als die r-Belichtungen, die Strukturen passen also nicht aufeinander. Auf den Platten lassen sich die beiden Lichtarten nicht trennen; die Messung in Schritt 6 sieht deshalb beide, und die Filamente geben den Ausschlag, weil sie die feinen Strukturen tragen.

##### *3. Gaia-Sterne als Bezugssystem*

Woran misst man, ob sich ein Filament bewegt hat, wenn jede Platte ihre eigene Verzerrung hat? Vorher vorhersagen: Wie weit wandert ein typischer Feldstern mit G = 17 in 60 Jahren, verglichen mit den etwa 10″ eines Filaments?

Gaia DR3 liefert für jeden Stern im Feld Ort und Eigenbewegung zur Epoche 2016.0. Schiebt man die Sterne mit ihrer Eigenbewegung auf das Datum einer Platte zurück, erhält man ein Gitter von Bezugspunkten, an dem sich jedes Bild ausrichten lässt (Theorie, Abschnitt 3). Die Abbildung zeigt die Sterne auf dem Pan-STARRS-Bild und, wie weit sie sich zwischen 1951 und 2011 bewegt haben. Der Pulsar selbst ist ein Gaia-Stern mit G = 16.5; seine Parallaxe kommt in Schritt 10 dran.

**Hintergrund & Fachbegriffe:**
- **Gaia DR3:** der dritte Datensatz des ESA-Satelliten Gaia (2022), mit Orten, Parallaxen und Eigenbewegungen von 1.5 Milliarden Sternen; für G = 17 ist die Eigenbewegung auf etwa 0.1 mas/Jahr genau.
- **Eigenbewegung** $\mu$: die Winkelgeschwindigkeit eines Sterns am Himmel in Millibogensekunden pro Jahr (mas/Jahr), zerlegt in $\mu_{\alpha*} = \mu_\alpha\cos\delta$ (Osten) und $\mu_\delta$ (Norden).
- **Epoche:** der Zeitpunkt, für den ein Ort gilt. Gaia DR3 gibt Orte für 2016.0; für eine Platte von 1951 muss man 64 Jahre Eigenbewegung zurückrechnen.
- **RUWE:** renormalised unit weight error, ein Gütemaß der Gaia-Astrometrie; Werte nahe 1 bedeuten eine saubere Einzelsternlösung.

```python
G = gaia["phot_g_mean_mag"].values
pm = np.hypot(gaia["pmra"], gaia["pmdec"]).values                   # mas/Jahr
shift60 = pm * (EPOCH["PS1 r"] - EPOCH["POSS-I E"]) / 1000          # ″ in 60 Jahren
xi_g, eta_g = tangent(gaia["ra"].values, gaia["dec"].values)
psr_xi, psr_eta = tangent(PSR["ra"], PSR["dec"])
PSR_1054 = (psr_xi + PSR["pmra"] * (T_SN - 2016.0) / 1000, psr_eta + PSR["pmdec"] * (T_SN - 2016.0) / 1000)   # Ort 1054
print(f"Gaia DR3 im Kreis von 7.5′: {len(gaia)} Sterne mit G < 20, davon {np.sum((G < 19) & np.isfinite(pm))} mit G < 19 und Eigenbewegung")
print(f"Eigenbewegung: Median {np.nanmedian(pm):.1f} mas/Jahr, also {np.nanmedian(shift60):.2f}″ in 60 Jahren; "
      f"{np.sum(shift60 > 0.5)} Sterne wandern mehr als 0.5″, der schnellste {np.nanmax(shift60):.1f}″")
print(f"Pulsar (Gaia DR3 {PSR_ID}): G = {PSR['phot_g_mean_mag']:.2f}, Ort 2016.0 bei ξ = {psr_xi:.2f}″, η = {psr_eta:.2f}″, "
      f"μ = ({PSR['pmra']:.2f} ± {PSR['pmra_error']:.2f}, {PSR['pmdec']:.2f} ± {PSR['pmdec_error']:.2f}) mas/Jahr, "
      f"Parallaxe {PSR['parallax']:.3f} ± {PSR['parallax_error']:.3f} mas, RUWE {PSR['ruwe']:.2f}")
print(f"Pulsar mit seiner Eigenbewegung zurückgerechnet auf 1054: ξ = {PSR_1054[0]:+.1f}″, η = {PSR_1054[1]:+.1f}″")

fig, axes = plt.subplots(1, 2, figsize=(15, 6.5), gridspec_kw={"width_ratios": [1.1, 1]})
ax = axes[0]
r_img = RAW["PS1 r"]["img"]
lo, hi = np.percentile(r_img, [5, 99.7])
ax.imshow(np.clip((r_img - lo) / (hi - lo), 0, 1) ** 0.6, origin="lower", cmap="gray", extent=ext)
sc = ax.scatter(xi_g, eta_g, c=G, s=14, cmap="viridis_r", vmin=12, vmax=20, lw=0)
SCALE = 20                                                          # Pfeile: Bewegung in 60 Jahren, 20-fach
fast = (shift60 > 0.3) & (shift60 < 5)                              # die wenigen schnelleren Sterne würden das Bild überdecken
ax.quiver(xi_g[fast], eta_g[fast], -gaia["pmra"].values[fast] * 60e-3 * SCALE, gaia["pmdec"].values[fast] * 60e-3 * SCALE,
          angles="xy", scale_units="xy", scale=1, color="orange", width=0.003, label=f"Bewegung in 60 Jahren × {SCALE}")
ax.plot(psr_xi, psr_eta, "c*", ms=12, label="Pulsar")
ax.set_xlim(HALF, -HALF)
ax.set_ylim(-HALF, HALF)
ax.set_xlabel("ξ [″] (Osten links)")
ax.set_ylabel("η [″]")
ax.set_title("Gaia-DR3-Sterne auf dem Pan-STARRS-Bild (r)", fontsize=10)
ax.legend(fontsize=8, loc="lower left")
fig.colorbar(sc, ax=ax, label="G [mag]", fraction=0.04)
ax = axes[1]
bins = np.logspace(-3, 1.3, 40)
ax.hist(shift60[np.isfinite(shift60)], bins=bins, color="grey", alpha=0.7, label="Feldsterne")
ax.axvline(np.hypot(PSR["pmra"], PSR["pmdec"]) * 60e-3, color="c", lw=2, label="Pulsar")
ax.axvspan(5, 15, color="tab:red", alpha=0.15, label="Filamente (erwartet)")
ax.set_xscale("log")
ax.set_xlabel("Bewegung zwischen 1951 und 2011 [″]")
ax.set_ylabel("Zahl der Sterne")
ax.set_title("Wie weit wandern die Bezugssterne in 60 Jahren?", fontsize=10)
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Im Kreis von 7.5′ liegen 1171 Gaia-Sterne mit G < 20, 765 davon mit G < 19 und Eigenbewegung. Im Median bewegen sie sich mit 2.4 mas/Jahr, in 60 Jahren also 0.14″. 75 Sterne wandern mehr als 0.5″, der schnellste 14.7″. Der Pulsar (G = 16.53, RUWE 1.03) bewegt sich mit (−11.51, +2.30) mas/Jahr, also nach Westnordwest. Zurückgerechnet stand er 1054 bei ξ = +11.0″, η = −2.2″.
* Die **linke Abbildung** zeigt die Sterne auf dem Pan-STARRS-Bild, farbig nach Helligkeit, die orangen Pfeile ihre Bewegung in 60 Jahren, 20-fach vergrößert. Das Feld liegt 6° unter der Milchstraßenebene in Richtung Antizentrum, deshalb ist es dicht mit Sternen besetzt. Viele Sterne stehen vor oder hinter dem Nebel; sie würden die Messung der Filamente stören und werden in Schritt 5 maskiert.
* Die **rechte Abbildung** beantwortet die Frage: Ein typischer Stern wandert in 60 Jahren 0.1 bis 0.3″, also 30- bis 100-mal weniger als ein Filament (rotes Band). Für die Ausrichtung der Bilder ist das trotzdem nicht vernachlässigbar: 0.14″ ist so viel wie der ganze Restfehler einer Platte (Schritt 4). Ohne die Korrektur wird das Bezugssystem der alten Platte verschoben (Variation 3).

##### *4. Astrometrie jeder Epoche*

Wie genau lässt sich jede Platte an Gaia anschließen, und wie gut war die Plattenlösung im FITS-Kopf? Vorher vorhersagen: Passt die alte Platte von 1951 schlechter zu Gaia als die von 1991, und um welchen Faktor?

In jedem Bild sucht der Code die Sterne (lokale Maxima, Schwerpunkt mit Gaußgewicht), paart sie innerhalb von 3″ mit den auf das Plattendatum geschobenen Gaia-Sternen und passt ein Polynom zweiten Grades von den Pixeln auf die Tangentialebene an (Theorie, Abschnitt 3). Benutzt werden nur Sterne mit 13 < G < 19 (bei Pan-STARRS 14.5 < G < 19, hellere sind gesättigt) außerhalb von 250″ um den Pulsar, damit der helle Nebel die Schwerpunkte nicht verschiebt. Die Abbildung zeigt die Abweichungen der Sterne vor und nach der Anpassung und die Restfehler nach Helligkeit.

**Hintergrund & Fachbegriffe:**
- **Tangentialebene** (ξ, η): die Projektion der Himmelskugel auf eine Ebene, die sie im Bezugspunkt berührt; ξ zeigt nach Osten, η nach Norden, beide in Bogensekunden.
- **Polynom-Plattenlösung:** ξ und η als Polynome der Pixelkoordinaten x, y. Grad 1 erfasst Verschiebung, Drehung, Maßstab und Scherung, Grad 2 zusätzlich eine leichte Verzerrung, etwa durch die gewölbte Platte im Schmidt-Teleskop.
- **Sigma-Clipping:** Sterne, deren Abweichung mehr als dreimal so groß ist wie die mittlere, werden verworfen und die Anpassung wiederholt (Doppelsterne, Plattenfehler, falsche Paare).
- **Magnitudengleichung:** Auf Fotoplatten hängt die gemessene Lage eines Sterns leicht von seiner Helligkeit ab, weil helle Sternbilder größer und oft unsymmetrisch sind.

```python
AST = {}
for name, fwhm, gmin in [("POSS-I E", 2.0, 13.0), ("POSS-II F", 3.0, 13.0), ("PS1 r", 2.5, 14.5), ("PS1 i", 2.5, 14.5)]:
    v = RAW[name]
    AST[name] = a = astrometry(v["img"], v["wcs"], v["epoch"], fwhm, v["pix"], gmin)
    h = a["hdr_vec"][a["keep"]]
    print(f"{name:10s}: {a['n_found']:4d} Sterne gefunden, {a['n']:3d} mit Gaia gepaart und benutzt; "
          f"Kopf-Plattenlösung: Versatz ({np.median(h[:, 0]):+.2f}, {np.median(h[:, 1]):+.2f})″, "
          f"Streuung {np.sqrt(np.mean(np.sum((h - np.median(h, 0)) ** 2, 1))):.2f}″ -> eigene Lösung: Rest {a['rms']:.3f}″")
for order in (1, 3):
    a = astrometry(RAW["POSS-I E"]["img"], RAW["POSS-I E"]["wcs"], EPOCH["POSS-I E"], 2.0, RAW["POSS-I E"]["pix"], 13.0, order=order)
    print(f"Kontrolle POSS-I E mit Polynom vom Grad {order}: Rest {a['rms']:.3f}″ ({a['n']} Sterne)")

fig, axes = plt.subplots(1, 3, figsize=(17, 5.6))
a = AST["POSS-I E"]
k = a["keep"]
S = 150                                                              # Pfeile 150-fach vergrößert
h = a["hdr_vec"][k] - np.median(a["hdr_vec"][k], axis=0)
for ax, vec, title in [(axes[0], h, "POSS-I E: Kopf-Plattenlösung minus Gaia\n(ohne den mittleren Versatz)"),
                       (axes[1], a["fit_vec"][k], "POSS-I E: eigene Lösung (Grad 2) minus Gaia")]:
    ax.quiver(a["xi"][k], a["eta"][k], vec[:, 0] * S, vec[:, 1] * S, angles="xy", scale_units="xy", scale=1,
              color="tab:blue", width=0.003)
    ax.quiver([-330], [-330], [1.0 * S], [0], angles="xy", scale_units="xy", scale=1, color="k", width=0.004)
    ax.text(-330 + S / 2, -315, "1″", ha="center", fontsize=9)
    ax.add_patch(plt.Circle((0, 0), 250, fill=False, ls="--", color="grey"))
    ax.set_xlim(HALF + 20, -HALF - 20)
    ax.set_ylim(-HALF - 20, HALF + 20)
    ax.set_aspect("equal")
    ax.set_xlabel("ξ [″] (Osten links)")
    ax.set_ylabel("η [″]")
    ax.set_title(title + f": Streuung {np.sqrt(np.mean(np.sum(vec ** 2, 1))):.2f}″", fontsize=10)
ax = axes[2]
bins = np.arange(13, 19.5, 1.0)
for name, col in zip(AST, ["tab:red", "tab:orange", "tab:blue", "tab:cyan"]):
    a = AST[name]
    r = np.hypot(*a["fit_vec"].T)
    idx = np.digitize(a["G"], bins)
    rms = [np.sqrt(np.mean(r[(idx == i) & a["keep"]] ** 2)) if np.sum((idx == i) & a["keep"]) > 4 else np.nan
           for i in range(1, len(bins))]
    ax.plot(bins[:-1] + 0.5, rms, "o-", color=col, label=f"{name} ({EPOCH[name]:.1f}): {a['rms']:.3f}″")
ax.set_yscale("log")
ax.set_xlabel("Gaia G [mag]")
ax.set_ylabel("Restfehler pro Stern (rms) [″]")
ax.set_title("Restfehler der eigenen Lösung nach Helligkeit", fontsize=10)
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

Die Plattenlösung im Kopf der Platte von 1951 liegt im Mittel um (+0.79, +0.22)″ daneben und streut um 0.22″; die eigene Lösung bringt den Versatz auf null und die Streuung auf 0.19″. Für 1991 sind es 0.12″, für Pan-STARRS 0.02″. Zur Vorhersage: Die alte Platte ist 1.5-mal schlechter als die von 1991 (gröbere Körnung, 1.7″ statt 1.0″ Abtastung), Pan-STARRS zehnmal besser.

##### *5. Gemeinsames Raster und die Expansion auf einen Blick*

Passen die Bilder von 1951 und 2011 besser zusammen, wenn man das alte Bild ein wenig vergrößert, und um welchen Faktor? Vorher vorhersagen: Wenn der Nebel seit 1054 mit konstanter Geschwindigkeit wächst, um wie viel Prozent ist er zwischen 1951 und 2011 größer geworden?

Mit der Umkehrung der Plattenlösung rechnet der Code jedes Bild auf ein gemeinsames Raster um: 12′ × 12′ in der Tangentialebene um den Pulsar, 1″ pro Pixel, Osten links und Norden oben. Danach liegt jeder Stern in allen Epochen auf demselben Pixel. Die Bilder sind unterschiedlich scharf; der Code misst die Halbwertsbreite der Sterne und glättet jedes Bild mit einem Gauß-Kern auf dieselbe Schärfe von 4.4″. Ein Hochpass (Abzug einer mit 6″ geglätteten Kopie) entfernt den großräumigen Helligkeitsverlauf, sodass nur Filamente und Knoten übrig bleiben. Jeder Gaia-Stern wird maskiert, und eine Nebelmaske legt fest, wo gemessen wird: das geglättete r-Bild über 3 % des Helligkeitsumfangs, also bis zu den schwachen äußeren Filamenten.

Als erste, grobe Probe vergrößert der Code das Bild von 1951 um einen Faktor $s$ um den Geburtsort des Pulsars (Schritt 3) und berechnet für jedes $s$ zwischen 1.00 und 1.12 die Korrelation mit dem Bild von 2011 innerhalb der Nebelmaske, ohne die Sterne. Bei homologer Expansion seit $t_0$ ist $s = (t_2 - t_0)/(t_1 - t_0)$.

**Hintergrund & Fachbegriffe:**
- **Umrechnen (Resampling):** Für jedes Pixel des Zielrasters berechnet die Umkehrung der Plattenlösung, wo es im Originalbild liegt; der Wert dort wird kubisch interpoliert.
- **Angleichen der Bildschärfe:** Faltung zweier Gauß-Kurven ergibt wieder eine Gauß-Kurve mit $\mathrm{FWHM}^2 = \mathrm{FWHM}_1^2 + \mathrm{FWHM}_2^2$; der Glättungskern hat also $\mathrm{FWHM}_\text{Kern} = \sqrt{4.4^2 - \mathrm{FWHM}_\text{Bild}^2}$.
- **Hochpass:** Bild minus geglättete Kopie. Er macht Platten und CCD vergleichbar, denn eine Fotoplatte misst eine Schwärzung, die nicht linear mit der Helligkeit wächst und im hellen Zentrum sättigt (Theorie, Abschnitt 2).

```python
GRID = {}
for name in AST:
    GRID[name] = resample(RAW[name]["img"], AST[name], presmooth=1.0 if name.startswith("PS1") else 0.0)


def fwhm_on_grid(im, epoch):
    # Halbwertsbreite der Sterne im Raster: 2D-Gauß an Gaia-Sterne mit 14 < G < 18 außerhalb des Nebels
    _, _, xi, eta, G = gaia_at(epoch, 18)
    col, row = to_grid(xi, eta)
    ok = (G > 14) & (np.abs(xi) < HALF - 30) & (np.abs(eta) < HALF - 30) & (np.hypot(xi, eta) > 250)
    yy, xx = np.mgrid[-7:8, -7:8]
    out = []
    for c, r in zip(col[ok], row[ok]):
        cut = im[int(round(r)) - 7:int(round(r)) + 8, int(round(c)) - 7:int(round(c)) + 8]
        f = lambda p: (p[0] * np.exp(-((xx - p[1]) ** 2 + (yy - p[2]) ** 2) / (2 * p[3] ** 2)) + p[4] - cut).ravel()
        p = optimize.least_squares(f, [cut.max() - np.median(cut), 0, 0, 1.5, np.median(cut)]).x
        if p[0] > 0 and 0.3 < abs(p[3]) < 5:
            out.append(2.355 * abs(p[3]))
    return np.median(out), len(out)


FWHM = {}
for name in GRID:
    FWHM[name], n = fwhm_on_grid(GRID[name], EPOCH[name])
    print(f"{name:10s}: Sterne im Raster mit FWHM {FWHM[name]:.2f}″ (Median aus {n} Sternen) -> Glättungskern "
          f"{np.sqrt(max(FWHM_COMMON ** 2 - FWHM[name] ** 2, 0)):.2f}″")

EPOCHS3 = ["POSS-I E", "POSS-II F", "PS1 r"]
H = {name: prep(name) for name in EPOCHS3}
smooth = {n: ndimage.gaussian_filter(GRID[n], np.sqrt(max(FWHM_COMMON ** 2 - FWHM[n] ** 2, 0)) / 2.355) for n in EPOCHS3}

# Nebelmaske: geglättetes PS1-r-Bild über 3 % des Helligkeitsumfangs, größte zusammenhängende Fläche, Löcher gefüllt
NEB_LEVEL = 0.03
sm = ndimage.gaussian_filter(GRID["PS1 r"], 8)
lo, hi = np.percentile(sm, [50, 99.5])
lab, _ = ndimage.label(sm > lo + NEB_LEVEL * (hi - lo))
NEB = ndimage.binary_fill_holes(lab == np.bincount(lab.ravel())[1:].argmax() + 1)
print(f"Nebelmaske: {NEB.sum()} Quadratbogensekunden, so groß wie ein Kreis mit {np.sqrt(NEB.sum() / np.pi):.0f}″ Radius; "
      f"Sterne maskiert: {H['PS1 r'][1].mean():.0%} des Rasters")

# Grobe Probe: 1951 um den Geburtsort des Pulsars vergrößern und mit 2011 korrelieren
yy_g, xx_g = np.mgrid[0:2 * HALF + 1, 0:2 * HALF + 1]
XI_G, ETA_G = HALF - xx_g, yy_g - HALF


def enlarge(img, s, c=PSR_1054):
    # Wert bei p kommt von c + (p - c)/s: das Bild wird um c um den Faktor s größer
    col, row = to_grid(c[0] + (XI_G - c[0]) / s, c[1] + (ETA_G - c[1]) / s)
    return ndimage.map_coordinates(img, [row, col], order=1, mode="nearest")


(h1, m1), (h2, m2) = H["POSS-I E"], H["PS1 r"]
S_GRID = np.arange(1.0, 1.1201, 0.0025)
corr = []
for s in S_GRID:
    ok = NEB & ~m2 & ~(enlarge(m1.astype(float), s) > 0.5)
    corr.append(np.corrcoef(enlarge(h1, s)[ok], h2[ok])[0, 1])
corr = np.array(corr)
i = np.argmax(corr)
p = np.polyfit(S_GRID[i - 3:i + 4], corr[i - 3:i + 4], 2)
S_MAX = -p[1] / (2 * p[0])
s_1054 = (EPOCH["PS1 r"] - T_SN) / (EPOCH["POSS-I E"] - T_SN)
t0_glob = EPOCH["PS1 r"] - (EPOCH["PS1 r"] - EPOCH["POSS-I E"]) / (1 - 1 / S_MAX)
print(f"Korrelation 1951 gegen 2011: {corr[0]:.2f} ohne Vergrößerung, {corr.max():.2f} bei s = {S_MAX:.4f} "
      f"(der Nebel ist um {100 * (S_MAX - 1):.1f} % gewachsen) -> grobes Konvergenzjahr {t0_glob:.0f}")
print(f"Erwartung bei konstanter Geschwindigkeit seit 1054: s = {s_1054:.4f} ({100 * (s_1054 - 1):.1f} %)")


def norm(im):
    v = np.percentile(im[NEB], [1, 99.5])
    return np.clip((im - v[0]) / (v[1] - v[0]), 0, 1)


a2 = norm(h2)
fig, axes = plt.subplots(1, 3, figsize=(18, 6.2), gridspec_kw={"width_ratios": [1, 1, 0.9]})
for ax, a, title in [(axes[0], norm(h1), "rot = 1951, türkis = 2011 (Hochpass)"),
                     (axes[1], norm(enlarge(h1, S_MAX)), f"1951 um {100 * (S_MAX - 1):.1f} % vergrößert: Filamente decken sich"),
                     ]:
    ax.imshow(np.dstack([a, a2, a2]), origin="lower", extent=EXTENT)
    ax.contour(NEB, levels=[0.5], colors="yellow", linewidths=0.6, extent=EXTENT, origin="lower")
    ax.set_xlim(230, -230)
    ax.set_ylim(-200, 200)
    ax.set_xlabel("ξ [″] (Osten links)")
    ax.set_ylabel("η [″]")
    ax.set_title(title, fontsize=10)
ax = axes[2]
ax.plot(S_GRID, corr, "k-", lw=2)
ax.axvline(S_MAX, color="tab:red", label=f"Maximum s = {S_MAX:.4f}")
ax.axvline(s_1054, color="grey", ls=":", lw=2, label=f"konstant seit 1054: s = {s_1054:.4f}")
ax.set_xlabel("Vergrößerung s des Bildes von 1951")
ax.set_ylabel("Korrelation mit 2011 (Nebel ohne Sterne)")
ax.set_title("Wie stark ist der Nebel in 60 Jahren gewachsen?", fontsize=10)
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Im gemeinsamen Raster haben die Sterne eine Halbwertsbreite von 4.17″ (1951), 3.66″ (1991) und 1.82″ (Pan-STARRS); geglättet wird mit Kernen von 1.4″, 2.5″ und 4.0″. Die Nebelmaske umfasst 82 800 Quadratbogensekunden, so viel wie ein Kreis mit 162″ Radius; 11 % des Rasters sind als Sterne maskiert. Die Korrelation zwischen 1951 und 2011 im Nebel steigt von 0.18 auf 0.40, wenn man das Bild von 1951 um 6.9 % vergrößert. Zur Vorhersage: Konstante Geschwindigkeit seit 1054 gibt (2011.86 − 1054.5)/(1951.85 − 1054.5) = 1.067, also 6.7 %.
* Die **linke Abbildung** legt die Hochpassbilder übereinander, 1951 rot und 2011 türkis. Ohne Vergrößerung sind die Farben getrennt: rote und türkise Fäden liegen nebeneinander. Die Sterne erscheinen weiß, weil sie sich nicht bewegt haben.
* Die **mittlere Abbildung** zeigt dasselbe mit dem um 6.9 % vergrößerten Bild von 1951: Jetzt decken sich die Filamente in vielen Bereichen, dafür stehen die Sterne doppelt, rot und türkis versetzt, denn die Vergrößerung schiebt auch sie nach außen. Die **rechte Abbildung** zeigt die Korrelation als glatte Kurve mit einem Maximum knapp rechts der Erwartung für 1054. Dieses grobe Maß ergibt ein Konvergenzjahr um 1080; es mittelt über alle Pixel und hält das Zentrum am Geburtsort des Pulsars fest. Die genaue Messung folgt in Schritt 6 bis 8.

##### *6. Wie weit sind die Filamente gewandert?*

Um wie viel und in welche Richtung hat sich jedes Stück des Nebels zwischen 1951 und 2011 verschoben? Vorher vorhersagen: Zeigen die Verschiebungen alle vom Pulsar weg, und wachsen sie mit dem Abstand?

Das Raster wird mit Ausschnitten von 21″ × 21″ belegt, Mittelpunkte alle 8″ innerhalb der Nebelmaske. Für jeden Ausschnitt des Bildes von 2011 sucht der Code im Bild von 1951 die Stelle, die am besten passt: Er verschiebt den Ausschnitt um bis zu ±20″ in beiden Richtungen und berechnet für jede Verschiebung die normierte Kreuzkorrelation, wobei maskierte Sterne in keinem der beiden Bilder mitzählen (Theorie, Abschnitt 4). Das Maximum, auf Bruchteile eines Pixels verfeinert, ist die Verschiebung. Ausschnitte mit einer Korrelation unter 0.6 oder einem Maximum am Rand des Suchbereichs werden verworfen.

**Hintergrund & Fachbegriffe:**
- **Normierte Kreuzkorrelation** (NCC): der Korrelationskoeffizient der Pixelwerte zweier gleich großer Ausschnitte, zwischen −1 und 1; sie ist unempfindlich gegen Helligkeit und Kontrast und damit gegen die verschiedene Schwärzung von Platte und CCD.
- **Subpixel-Verfeinerung:** Durch das Maximum und seine beiden Nachbarn legt man eine Parabel; ihr Scheitel liegt zwischen den Pixeln und gibt die Verschiebung auf etwa 0.1 Pixel genau.
- **Radial und tangential:** Die Verschiebung wird in einen Anteil vom Zentrum weg (radial) und einen senkrecht dazu (tangential) zerlegt. Reine Expansion hat keinen tangentialen Anteil.

```python
P60 = pair("POSS-I E", "PS1 r")
df = P60["df"]
dt60 = P60["dt"]
rr_ = np.hypot(df["xi"], df["eta"])
rad = (df["dxi"] * df["xi"] + df["deta"] * df["eta"]) / rr_
tan_ = (df["deta"] * df["xi"] - df["dxi"] * df["eta"]) / rr_
n_all = len(measure(H["PS1 r"], H["POSS-I E"]))
print(f"Ausschnitte in der Nebelmaske mit Maximum im Suchbereich: {n_all}; mit Korrelation > {PEAK_MIN}: {len(df)}")
print(f"Verschiebung 1951 -> 2011 ({dt60:.2f} Jahre): Median {np.median(np.hypot(df['dxi'], df['deta'])):.2f}″, "
      f"größte {np.hypot(df['dxi'], df['deta']).max():.1f}″; radial im Median {np.median(rad):+.2f}″, tangential {np.median(tan_):+.2f}″")
print(f"Anteil der Verschiebungen, die vom Pulsar weg zeigen: {np.mean(rad > 0):.0%}")

# Beispiel: ein Ausschnitt ohne maskierte Sterne, hohe Korrelation, große Verschiebung
(h2, m2), (h1, m1) = H["PS1 r"], H["POSS-I E"]
clean = np.array([not m2[int(HALF + e) - PATCH:int(HALF + e) + PATCH + 1, int(HALF - x) - PATCH:int(HALF - x) + PATCH + 1].any()
                  for x, e in zip(df["xi"], df["eta"])])
cand = df[clean & (df["peak"] > 0.85)]
ex = cand.loc[np.hypot(cand["dxi"], cand["deta"]).idxmax()].copy()
ex["r"] = np.hypot(ex["xi"], ex["eta"])
c, r = int(HALF - ex["xi"]), int(HALF + ex["eta"])
win = (slice(r - PATCH - SEARCH, r + PATCH + SEARCH + 1), slice(c - PATCH - SEARCH, c + PATCH + SEARCH + 1))
ncc_ex = masked_ncc(h2[r - PATCH:r + PATCH + 1, c - PATCH:c + PATCH + 1], ~m2[r - PATCH:r + PATCH + 1, c - PATCH:c + PATCH + 1],
                    h1[win], ~m1[win])
print(f"Beispiel bei ξ = {ex['xi']:.0f}″, η = {ex['eta']:.0f}″ (Abstand {ex['r']:.0f}″): Verschiebung ({ex['dxi']:+.2f}, {ex['deta']:+.2f})″, "
      f"Korrelation {ex['peak']:.2f}")

fig = plt.figure(figsize=(16, 11))
gs = fig.add_gridspec(3, 3)
ext_win = [ex["xi"] + PATCH + SEARCH + 0.5, ex["xi"] - PATCH - SEARCH - 0.5, ex["eta"] - PATCH - SEARCH - 0.5, ex["eta"] + PATCH + SEARCH + 0.5]
for k, (n, title) in enumerate([("PS1 r", "2011: Ausschnitt (Rahmen) im Hochpassbild"), ("POSS-I E", "1951: gleiche Stelle (gestrichelt) und Treffer")]):
    ax = fig.add_subplot(gs[k, 0])
    hp_, m_ = H[n]
    v = np.where(m_, np.nan, hp_)[win]
    cm = plt.get_cmap("gray").copy()
    cm.set_bad("tab:red", 0.35)                                       # maskierte Sterne rötlich
    ax.imshow(v, origin="lower", cmap=cm, extent=ext_win, vmin=np.nanpercentile(v, 2), vmax=np.nanpercentile(v, 99.5))
    ax.add_patch(plt.Rectangle((ex["xi"] - PATCH, ex["eta"] - PATCH), 2 * PATCH, 2 * PATCH, fill=False, ec="c", lw=1.5,
                               ls="--" if k else "-"))
    if k:
        ax.add_patch(plt.Rectangle((ex["xi"] - ex["dxi"] - PATCH, ex["eta"] - ex["deta"] - PATCH), 2 * PATCH, 2 * PATCH,
                                   fill=False, ec="orange", lw=1.5))
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("ξ [″]")
    ax.set_ylabel("η [″]")
ax = fig.add_subplot(gs[2, 0])
im = ax.imshow(np.where(ncc_ex > -1, ncc_ex, np.nan), origin="lower", cmap="viridis",
               extent=[SEARCH + 0.5, -SEARCH - 0.5, -SEARCH - 0.5, SEARCH + 0.5])
ax.plot(-ex["dxi"], -ex["deta"], "r+", ms=14, mew=2)
ax.set_xlabel("Versatz des Ausschnitts in ξ [″]")
ax.set_ylabel("Versatz in η [″]")
ax.set_title("Korrelation für jeden Versatz (Kreuz: Maximum)", fontsize=10)
fig.colorbar(im, ax=ax, fraction=0.05, label="NCC")
ax = fig.add_subplot(gs[:, 1:])
show(ax, smooth["PS1 r"], lo=20, hi=99.8, gamma=0.7)
SC = 5
qv = ax.quiver(df["xi"], df["eta"], df["dxi"] * SC, df["deta"] * SC, df["peak"], angles="xy", scale_units="xy", scale=1,
               cmap="autumn", width=0.0025)
ax.quiver([200], [-190], [5 * SC], [0], angles="xy", scale_units="xy", scale=1, color="w", width=0.003)
ax.text(200 - 2.5 * SC, -180, "5″", color="w", ha="center", fontsize=10)
ax.plot(psr_xi, psr_eta, "c*", ms=12)
ax.set_xlim(240, -240)
ax.set_ylim(-210, 210)
ax.set_xlabel("ξ [″] (Osten links)")
ax.set_ylabel("η [″]")
ax.set_title(f"{len(df)} Verschiebungen 1951 → 2011, Pfeile {SC}-fach (Farbe: Korrelation)", fontsize=10)
fig.colorbar(qv, ax=ax, fraction=0.035, label="Korrelation")
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** In der Nebelmaske liegen 1096 Ausschnitte, deren bestes Maximum im Suchbereich liegt; 847 davon haben eine Korrelation über 0.6. Im Median verschieben sie sich um 8.1″, radial um +7.2″ nach außen und tangential um −0.06″; 86 % zeigen vom Pulsar weg. Die größte gemessene Verschiebung ist 26″, deutlich mehr als die Expansion am Rand erlaubt; das sind falsche Treffer, die die Anpassung in Schritt 7 verwirft.
* Die **linke Spalte** zeigt das Beispiel bei ξ = −166″, η = +62″ am Westrand: oben der Ausschnitt von 2011 im Hochpassbild, in der Mitte das Bild von 1951 mit derselben Stelle (gestrichelt) und dem Treffer (orange), 12.7″ weiter östlich und etwas südlicher, also näher am Zentrum. Unten die Korrelation für jeden Versatz: ein klares Maximum von 0.89 bei (+12, −4)″, umgeben von Werten unter 0.4. Rötliche Flächen sind maskierte Sterne.
* Die **rechte Abbildung** beantwortet die Frage: Die Pfeile zeigen fast überall radial nach außen und werden nach außen länger, das Muster einer Expansion. ##### *7. Das Expansionszentrum und der Geburtsort des Pulsars*

Von welchem Punkt aus dehnt sich der Nebel aus, und stand dort 1054 der Pulsar? Vorher vorhersagen: Der Pulsar bewegt sich nach Gaia mit 11.7 mas/Jahr nach Westnordwest. Wie weit östlich seines heutigen Ortes muss das Zentrum liegen, wenn er in der Explosion entstand?

Bei homologer Expansion ist jede Verschiebung proportional zum Abstand vom Zentrum: $\Delta\vec p = k\,(\vec p - \vec c)$ (Theorie, Abschnitt 5). Schreibt man $k\vec c$ als eigene Unbekannte, ist das ein lineares Problem mit drei Unbekannten ($k$, $kc_\xi$, $kc_\eta$), das die Methode der kleinsten Quadrate in einem Schritt löst; Ausreißer jenseits von drei mittleren Abweichungen werden verworfen. Die Unsicherheit des Zentrums kommt aus 200 Bootstrap-Ziehungen der Ausschnitte. Verglichen wird mit dem Ort des Pulsars heute, mit seinem Ort im Jahr 1054 (Gaia-Eigenbewegung zurückgerechnet) und mit dem Zentrum von Martin et al. (2025) aus fast 20 000 Verschiebungen in CFHT-Bildern.

**Hintergrund & Fachbegriffe:**
- **Homologe Expansion:** Jedes Stück bewegt sich mit konstanter Geschwindigkeit geradlinig vom Ursprung weg; dann ist der Abstand proportional zur Geschwindigkeit, und die Form bleibt erhalten, nur der Maßstab wächst.
- **Bootstrap:** Man zieht aus den Messungen so viele mit Zurücklegen, wie es gibt, wiederholt die Anpassung und nimmt die Streuung der Ergebnisse als Fehler.
- **Kick:** Neutronensterne bekommen in der Explosion oft einen Rückstoß von einigen hundert km/s; der Crab-Pulsar ist mit etwa 110 km/s (bei 2 kpc) eher langsam.

```python
f = P60
cx, cy = f["cx"], f["cy"]
ex_, ey_ = f["cx_err"], f["cy_err"]
# Pulsar: heute (2016.0) und zurückgerechnet auf 1054 (Schritt 3)
psr_1054 = PSR_1054
psr_t = np.linspace(T_SN, 2016.0, 50) - 2016.0
# Martin et al. (2025): Zentrum bei 5h34m32.67s +22°00'46.7" (J2000), Fehler 1.2"
m_xi, m_eta = tangent(15 * (5 + 34 / 60 + 32.67 / 3600), 22 + 0 / 60 + 46.7 / 3600)
print(f"Expansionszentrum (eigene Messung): ξ = {cx:+.1f} ± {ex_:.1f}″, η = {cy:+.1f} ± {ey_:.1f}″ ({f['n']} von {len(f['df'])} Ausschnitten)")
print(f"Pulsar 2016.0: ξ = {psr_xi:+.2f}″, η = {psr_eta:+.2f}″; zurückgerechnet auf 1054: ξ = {psr_1054[0]:+.1f}″, η = {psr_1054[1]:+.1f}″")
print(f"Abstand Zentrum - Pulsar 1054: {np.hypot(cx - psr_1054[0], cy - psr_1054[1]):.1f}″ "
      f"(Δξ = {cx - psr_1054[0]:+.1f}″, Δη = {cy - psr_1054[1]:+.1f}″); Zentrum - Pulsar heute: {np.hypot(cx - psr_xi, cy - psr_eta):.1f}″")
print(f"Martin et al. (2025): ξ = {m_xi:+.1f}″, η = {m_eta:+.1f}″ ± 1.2″; Unterschied zur eigenen Messung "
      f"Δξ = {cx - m_xi:+.1f}″, Δη = {cy - m_eta:+.1f}″")
print(f"Restabweichung pro Ausschnitt nach der Anpassung: {np.sqrt(np.mean(f['res'][f['keep']] ** 2) / 2):.2f}″ pro Achse")

fig, axes = plt.subplots(1, 2, figsize=(15, 6.8))
ax = axes[0]
show(ax, smooth["PS1 r"], lo=20, hi=99.8, gamma=0.7)
d = f["df"][f["keep"]]
for _, row_ in d.sample(150, random_state=1).iterrows():                  # Verschiebungen nach innen verlängert
    ax.plot([row_["xi"], row_["xi"] - row_["dxi"] / f["k"]], [row_["eta"], row_["eta"] - row_["deta"] / f["k"]],
            color="orange", lw=0.4, alpha=0.6)
ax.plot(cx, cy, "w+", ms=16, mew=2.5)
ax.set_xlim(240, -240)
ax.set_ylim(-210, 210)
ax.set_xlabel("ξ [″] (Osten links)")
ax.set_ylabel("η [″]")
ax.set_title("150 Verschiebungen, bis zum Ursprung zurückverlängert", fontsize=10)
ax = axes[1]
show(ax, GRID["PS1 r"], lo=20, hi=99.95, gamma=0.5)
ax.scatter(f["boot"][:, 1], f["boot"][:, 2], s=4, c="orange", alpha=0.5, label="Zentrum, Bootstrap")
ax.errorbar(cx, cy, xerr=ex_, yerr=ey_, fmt="o", color="orange", mec="k", ms=8, capsize=3, label="Zentrum (eigene Messung)")
ax.errorbar(m_xi, m_eta, xerr=1.2, yerr=1.2, fmt="s", color="lime", mec="k", ms=7, capsize=3, label="Zentrum Martin et al. (2025)")
ax.plot(psr_xi + PSR["pmra"] * psr_t / 1000, psr_eta + PSR["pmdec"] * psr_t / 1000, "c-", lw=2, label="Weg des Pulsars seit 1054 (Gaia)")
ax.plot(*psr_1054, "co", ms=8, mec="k", label="Pulsar 1054")
ax.plot(psr_xi, psr_eta, "c*", ms=16, mec="k", label="Pulsar heute")
ax.set_xlim(25, -10)
ax.set_ylim(-17, 12)
ax.set_xlabel("ξ [″] (Osten links)")
ax.set_ylabel("η [″]")
ax.set_title("Zentrum der Expansion und Weg des Pulsars", fontsize=10)
ax.legend(fontsize=8, loc="upper right")
fig.tight_layout()
plt.show()
```

##### *8. Ein Hubble-Diagramm des Nebels und das Konvergenzjahr*

In welchem Jahr wären alle Filamente in einem Punkt gewesen, wenn sie sich immer so schnell bewegt hätten wie heute? Vorher vorhersagen: Liegt dieses Jahr vor oder nach 1054, und um wie viel, wenn die Filamente seit der Explosion um etwa 5 % schneller geworden sind?

Trägt man die Geschwindigkeit vom Zentrum weg gegen den Abstand vom Zentrum auf, liegen die Punkte bei homologer Expansion auf einer Geraden durch den Ursprung, genau wie im Hubble-Diagramm der Galaxien. Ihre Steigung ist die Expansionsrate $e = 1/(t_2 - t_0)$; der Kehrwert ist das scheinbare Alter, und $t_0$ heißt **Konvergenzjahr** (Theorie, Abschnitt 5). Zum Vergleich rechnet der Code dasselbe Jahr mit der älteren Methode der kleinsten Streuung: Jede Position wird mit ihrer eigenen Geschwindigkeit zurückgerechnet, und man sucht das Jahr, in dem die Punktwolke am kleinsten war. Eine Simulation mit perfekter Expansion und dem gemessenen Rauschen zeigt, was diese Methode mit fehlerbehafteten Geschwindigkeiten macht.

**Hintergrund & Fachbegriffe:**
- **Konvergenzjahr** (convergence date): das Jahr, das man erhält, wenn man die heutigen Geschwindigkeiten unverändert in die Vergangenheit verlängert. Es ist gleich dem Explosionsjahr nur, wenn die Filamente nie beschleunigt oder gebremst wurden.
- **Methode der kleinsten Streuung** (closest approach): seit Trimble (1968) üblich. Bietenholz & Nugent (2015) haben gezeigt, dass Messfehler der Geschwindigkeiten die zurückgerechneten Punkte auseinandertreiben, und zwar umso mehr, je weiter man zurückrechnet; das Minimum rutscht dadurch zur Gegenwart hin.
- **Fit-Methode:** die Gerade im Hubble-Diagramm, hier über den linearen Fit von Schritt 7. Fehler in den Geschwindigkeiten streuen die Punkte nur um die Gerade und verschieben ihre Steigung nicht.

```python
t2 = EPOCH["PS1 r"]
d = P60["df"][P60["keep"]].reset_index(drop=True)
px, py = d["xi"] - cx, d["eta"] - cy
r_c = np.hypot(px, py)
v_rad = (d["dxi"] * px + d["deta"] * py) / r_c / dt60                    # ″ pro Jahr vom Zentrum weg
e_rate = P60["k"] / dt60                                                  # Expansionsrate [1/Jahr]
T0 = P60["t0"]
print(f"Steigung im Hubble-Diagramm: e = {e_rate * 100:.4f} % pro Jahr -> scheinbares Alter 1/e = {1 / e_rate:.0f} Jahre "
      f"vor {t2:.1f}, Konvergenzjahr {T0:.1f} ± {P60['t0_err']:.1f}")
print(f"Größte Geschwindigkeit am Himmel: {v_rad.max():.3f}″/Jahr in {r_c[v_rad.idxmax()]:.0f}″ Abstand")

def closest_approach(xi, eta, vx, vy, years):
    # Radius (rms) der zurückgerechneten Punktwolke für jedes Jahr
    return np.array([np.sqrt(np.var(xi - vx * (t2 - t)) + np.var(eta - vy * (t2 - t))) for t in years])

YEARS = np.arange(950, 1301, 1.0)
vx, vy = d["dxi"] / dt60, d["deta"] / dt60
sc_data = closest_approach(d["xi"], d["eta"], vx, vy, YEARS)
t_ca = YEARS[np.argmin(sc_data)]
# Simulation: dieselben Orte, perfekte Expansion aus T0, dazu Rauschen in der Größe der gemessenen Restabweichung
sig = np.sqrt(np.mean(P60["res"][P60["keep"]] ** 2) / 2)
vx_true, vy_true = P60["k"] * px / dt60, P60["k"] * py / dt60
sc_true = closest_approach(d["xi"], d["eta"], vx_true, vy_true, YEARS)
t_sim = []
for _ in range(50):
    nx_, ny_ = rng.normal(0, sig, len(d)) / dt60, rng.normal(0, sig, len(d)) / dt60
    t_sim.append(YEARS[np.argmin(closest_approach(d["xi"], d["eta"], vx_true + nx_, vy_true + ny_, YEARS))])
print(f"Methode der kleinsten Streuung: Minimum {t_ca:.0f} (Radius dort {sc_data.min():.1f}″)")
print(f"Simulation ohne Rauschen: Minimum {YEARS[np.argmin(sc_true)]:.0f}; mit {sig:.2f}″ Rauschen pro Achse: "
      f"{np.mean(t_sim):.0f} ± {np.std(t_sim):.0f} (50 Ziehungen) -> Verschiebung um {np.mean(t_sim) - T0:+.0f} Jahre")

fig, axes = plt.subplots(1, 2, figsize=(15, 5.8))
ax = axes[0]
ax.scatter(r_c, v_rad * 1000, s=6, alpha=0.4, color="tab:blue", label=f"Ausschnitte ({len(d)})")
bins = np.arange(0, 241, 20)
idx = np.digitize(r_c, bins)
med = [np.median(v_rad[idx == i]) * 1000 if np.sum(idx == i) > 5 else np.nan for i in range(1, len(bins))]
ax.plot(bins[:-1] + 10, med, "ko", ms=7, label="Median je 20″")
rg = np.linspace(0, 240, 2)
ax.plot(rg, e_rate * rg * 1000, "r-", lw=2, label=f"Gerade: 1/e = {1 / e_rate:.0f} Jahre (Konvergenz {T0:.0f})")
ax.plot(rg, rg / (t2 - T_SN) * 1000, "k--", lw=1.5, label=f"konstante Geschwindigkeit seit {T_SN:.0f}")
ax.set_xlabel("Abstand vom Expansionszentrum [″]")
ax.set_ylabel("Geschwindigkeit vom Zentrum weg [mas/Jahr]")
ax.set_title("Hubble-Diagramm des Krebsnebels (1951 → 2011)", fontsize=10)
ax.set_xlim(0, 240)
ax.legend(fontsize=9)
ax = axes[1]
ax.plot(YEARS, sc_data, "b-", lw=2, label=f"gemessene Geschwindigkeiten: Minimum {t_ca:.0f}")
ax.plot(YEARS, sc_true, "k--", lw=1.5, label=f"Simulation ohne Rauschen: Minimum {YEARS[np.argmin(sc_true)]:.0f}")
ax.axvline(T0, color="r", lw=1.5, label=f"Fit-Methode {T0:.0f}")
ax.axvline(T_SN, color="grey", ls=":", lw=2, label="Supernova 1054")
ax.axvspan(np.mean(t_sim) - np.std(t_sim), np.mean(t_sim) + np.std(t_sim), color="b", alpha=0.15,
           label=f"Simulation mit Rauschen: {np.mean(t_sim):.0f} ± {np.std(t_sim):.0f}")
ax.set_xlabel("Jahr")
ax.set_ylabel("Radius der zurückgerechneten Punktwolke (rms) [″]")
ax.set_title("Methode der kleinsten Streuung", fontsize=10)
ax.set_ylim(0, None)
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Die Gerade im Hubble-Diagramm hat die Steigung e = 0.1095 % pro Jahr; das scheinbare Alter ist 913 Jahre vor 2011.86, das Konvergenzjahr 1098.5 ± 2.9. Die schnellsten Filamente bewegen sich mit 0.229″ pro Jahr, 200″ vom Zentrum. Die Methode der kleinsten Streuung findet ihr Minimum bei 1109; die Simulation ohne Rauschen findet 1098, also das eingesetzte Jahr, mit dem gemessenen Rauschen von 0.61″ pro Achse aber 1109 ± 3.
* Die **linke Abbildung** ist das Hubble-Diagramm des Nebels: Die 630 Punkte und ihre Mediane liegen auf einer Geraden durch den Ursprung, wie bei homologer Expansion. Die rote Gerade (Konvergenz 1098) ist steiler als die gestrichelte für konstante Geschwindigkeit seit 1054. Zur Vorhersage: Das Konvergenzjahr liegt nach 1054; mit 5 % Beschleunigung sind es nach Theorie, Abschnitt 6, etwa 45 Jahre.
* Die **rechte Abbildung** zeigt den Radius der zurückgerechneten Punktwolke. Ohne Rauschen (gestrichelt) fällt er auf null genau beim Konvergenzjahr des Fits. Mit den gemessenen Geschwindigkeiten (blau) bleibt er überall über 13″, und sein Minimum liegt 10 Jahre später, genau dort, wo die Simulation mit Rauschen es erwartet (blaues Band). Die ältere Methode ist also nicht falsch gerechnet, sondern verzerrt; das erklärt einen Teil der späten Jahre in der Literatur (Schritt 13).

##### *9. Drei Epochen, drei Paare: Wie sicher ist das Jahr?*

Geben die Paare 1951 → 1991 und 1991 → 2011 dasselbe Konvergenzjahr wie das lange Paar, und wie stark hängt das Ergebnis von den Einstellungen der Messung ab? Vorher vorhersagen: Welches Paar reagiert am empfindlichsten auf einen kleinen systematischen Fehler von 0.1″ in den Verschiebungen?

Der Code misst die beiden kürzeren Paare genauso wie in Schritt 6 und 7. Bei homologer Expansion müssen sich die Skalenfaktoren verketten lassen: $(1 - k_{1951\to2011}) = (1 - k_{1951\to1991})(1 - k_{1991\to2011})$. Dann wiederholt er das lange Paar mit geänderten Einstellungen (größere Ausschnitte, anderer Hochpass, andere Schärfe, größere Sternmasken, andere Mindestkorrelation). Die Streuung dieser Ergebnisse ist ein Maß für den systematischen Fehler, den der Bootstrap nicht sieht.

Für die Empfindlichkeit hilft eine kleine Rechnung: Mit $t_0 = t_2 - \Delta t / k$ und $k = \Delta r / r$ verschiebt ein Fehler $\delta$ in der Verschiebung das Konvergenzjahr um

$$\delta t_0 = \frac{(t_2 - t_0)^2}{\Delta t}\,\frac{\delta}{r}.$$

Der Fehler im Jahr fällt also mit $1/\Delta t$: Eine dreimal längere Zeitbasis macht das Ergebnis dreimal unempfindlicher.

**Hintergrund & Fachbegriffe:**
- **Statistischer und systematischer Fehler:** Der statistische Fehler kommt vom Rauschen der einzelnen Messungen und sinkt mit ihrer Zahl. Der systematische kommt von der Methode (Plattenlösung, Filter, Bildschärfe) und betrifft alle Messungen gleich; mehr Ausschnitte helfen dagegen nicht.
- **Zeitbasis** $\Delta t$: der Abstand der beiden Epochen. Eigenbewegungen werden mit der längsten verfügbaren Zeitbasis gemessen; deshalb sind alte Fotoplatten bis heute wertvoll.

```python
PAIRS = {"1951 → 2011": ("POSS-I E", "PS1 r"), "1951 → 1991": ("POSS-I E", "POSS-II F"), "1991 → 2011": ("POSS-II F", "PS1 r")}
RES = {"1951 → 2011": P60}
for lab, (a_, b_) in list(PAIRS.items())[1:]:
    RES[lab] = pair(a_, b_)
for lab, v in RES.items():
    print(f"{lab}: Δt = {v['dt']:5.2f} Jahre, {v['n']:4d} Ausschnitte, k = {v['k']:.5f}, Konvergenzjahr {v['t0']:7.1f} ± {v['t0_err']:4.1f}, "
          f"Zentrum ({v['cx']:+.1f}, {v['cy']:+.1f})″")
k_chain = 1 - (1 - RES["1951 → 1991"]["k"]) * (1 - RES["1991 → 2011"]["k"])
print(f"Verkettet: k = {k_chain:.5f} gegen direkt {P60['k']:.5f} (Unterschied {100 * (k_chain / P60['k'] - 1):+.1f} %)")

VAR = {"Grundeinstellung": P60}
VAR["Ausschnitte 29″ statt 21″"] = pair("POSS-I E", "PS1 r", patch=14, n_boot=0)
for lab, kw in [("Hochpass 4″ statt 6″", dict(hp_sigma=4.0)), ("Hochpass 10″", dict(hp_sigma=10.0)),
                ("Schärfe 5.5″ statt 4.4″", dict(fwhm_target=5.5)), ("Sternmasken × 1.5", dict(star_scale=1.5))]:
    Hv = {n: prep(n, **kw) for n in ("POSS-I E", "PS1 r")}
    VAR[lab] = pair("POSS-I E", "PS1 r", H=Hv, n_boot=0)
VAR["Korrelation > 0.75"] = pair("POSS-I E", "PS1 r", peak_min=0.75, n_boot=0)
VAR["Korrelation > 0.45"] = pair("POSS-I E", "PS1 r", peak_min=0.45, n_boot=0)
t0_var = np.array([v["t0"] for v in VAR.values()])
T0_SYS = t0_var.std()
for lab, v in VAR.items():
    print(f"  {lab:27s}: {v['n']:4d} Ausschnitte, Konvergenzjahr {v['t0']:7.1f}")
print(f"Streuung über die Einstellungen (systematisch): ± {T0_SYS:.1f} Jahre; Spanne {t0_var.min():.0f} bis {t0_var.max():.0f}")
print(f"Ergebnis: Konvergenzjahr {T0:.0f} ± {P60['t0_err']:.0f} (statistisch) ± {T0_SYS:.0f} (systematisch)")

fig, axes = plt.subplots(1, 3, figsize=(19, 5.8), gridspec_kw={"width_ratios": [1, 1.1, 1], "wspace": 0.55})
ax = axes[0]
for i, (lab, v) in enumerate(RES.items()):
    ax.errorbar(i, v["t0"], yerr=v["t0_err"], fmt="o", ms=8, capsize=4, color=["tab:red", "tab:orange", "tab:green"][i])
ax.axhline(T_SN, color="grey", ls=":", lw=2, label="Supernova 1054")
ax.axhspan(T0 - T0_SYS, T0 + T0_SYS, color="tab:red", alpha=0.12, label="1951 → 2011 ± systematisch")
ax.set_xticks(range(3), [f"{lab}\n(Δt = {v['dt']:.0f} J.)" for lab, v in RES.items()])
ax.set_ylabel("Konvergenzjahr")
ax.set_title("Drei Paare von Epochen (Fehler: Bootstrap)", fontsize=10)
ax.legend(fontsize=8, loc="lower right")
ax = axes[1]
labs = list(VAR)
ax.barh(range(len(labs)), [v["t0"] - T0 for v in VAR.values()], color=["tab:red"] + ["grey"] * (len(labs) - 1))
ax.set_yticks(range(len(labs)), [f"{l_} ({VAR[l_]['n']})" for l_ in labs], fontsize=9)
ax.invert_yaxis()
ax.axvline(0, color="k", lw=0.8)
ax.set_xlabel(f"Konvergenzjahr minus Grundeinstellung ({T0:.0f}) [Jahre]")
ax.set_title("1951 → 2011 mit geänderten Einstellungen (Zahl der Ausschnitte)", fontsize=10)
ax = axes[2]
dts = np.linspace(5, 80, 200)
for delta, ls in [(0.05, "--"), (0.1, "-")]:
    ax.plot(dts, (t2 - T0) ** 2 / dts * delta / 120, "k", ls=ls, label=f"Fehler {delta}″ in 120″ Abstand")
for (lab, v), col in zip(RES.items(), ["tab:red", "tab:orange", "tab:green"]):
    ax.plot(v["dt"], (t2 - T0) ** 2 / v["dt"] * 0.1 / 120, "o", color=col, ms=9, label=lab)
ax.set_xlabel("Zeitbasis Δt [Jahre]")
ax.set_ylabel("Verschiebung des Konvergenzjahres [Jahre]")
ax.set_title("Empfindlichkeit gegen systematische Fehler", fontsize=10)
ax.set_ylim(0, 60)
ax.legend(fontsize=8)
plt.show()
```

* Die **Ausgabe:** Die drei Paare geben 1098.5 ± 2.9 (1951 → 2011), 1097.3 ± 3.1 (1951 → 1991) und 1113.2 ± 5.3 (1991 → 2011). Die verketteten Skalenfaktoren der beiden kurzen Paare ergeben k = 0.06598, der direkte 0.06570, 0.4 % Unterschied. Mit geänderten Einstellungen liegt das lange Paar zwischen 1088 und 1099, mit einer Streuung von ±3.7 Jahren. Das Ergebnis ist 1098 ± 3 (statistisch) ± 4 (systematisch).
* Die **linke Abbildung:** Die beiden Paare mit 1951 stimmen überein, das kurze Paar 1991 → 2011 liegt 15 Jahre später, drei seiner Bootstrap-Fehler. In der **mittleren Abbildung** verschieben die größeren Sternmasken das Jahr am stärksten (−10 Jahre, nur 442 Ausschnitte), vermutlich, weil dann in den dichten Sternfeldern vor dem Nebel viele Ausschnitte wegfallen und die übrigen anders über den Nebel verteilt sind. Größere Ausschnitte (−8 Jahre) mitteln über mehr Filamente mit verschiedenen Abständen vom Zentrum. Die übrigen Einstellungen ändern das Jahr um höchstens 4 Jahre.
* Die **rechte Abbildung** beantwortet die Vorhersage: Ein systematischer Fehler von 0.1″ in 120″ Abstand verschiebt das Jahr bei 20 Jahren Zeitbasis um 35 Jahre, bei 60 Jahren nur um 12. Das kurze Paar reagiert also dreimal so empfindlich; schon ein systematischer Fehler von 0.04″, ein Drittel des Restfehlers eines einzelnen Sterns auf der Platte von 1991, erklärt seine 15 Jahre. Deshalb zählt das lange Paar.

##### *10. Wie weit ist der Krebsnebel entfernt?*

Wie weit ist der Nebel entfernt, wenn man nur seine Ausdehnung am Himmel und die Geschwindigkeit seiner Filamente entlang der Sichtlinie kennt, und passt das zur Gaia-Parallaxe des Pulsars? Vorher vorhersagen: Ist der Nebel entlang der Sichtlinie eher so tief wie seine lange oder wie seine kurze Achse am Himmel?

Die **Expansionsparallaxe** verbindet zwei Messungen derselben Bewegung (Theorie, Abschnitt 7): Am Himmel misst man eine Winkelgeschwindigkeit $\mu$ in Bogensekunden pro Jahr, im Spektrum eine Geschwindigkeit $v$ in km/s. Beide gehören zusammen über $v = 4.74\,\mu\,d$ mit $d$ in Parsec. Die schnellsten Filamente vor und hinter dem Pulsar bewegen sich mit etwa ±1500 km/s auf uns zu oder von uns weg (Martin et al. 2021). Nimmt man wie Trimble (1973) an, dass der Nebel entlang der Sichtlinie so tief ist wie entlang seiner kurzen Achse, dann bewegen sich die Filamente am Rand der kurzen Achse am Himmel genauso schnell, und es folgt $d = v_{r,\max} / (4.74\, e\, b)$ mit der Expansionsrate $e$ aus Schritt 8 und der halben kurzen Achse $b$. Die Achsen kommen aus den zweiten Momenten der Nebelmaske, wie bei einer gleichmäßig gefüllten Ellipse.

Dagegen steht die trigonometrische Parallaxe des Pulsars aus Gaia DR3 und die Radioparallaxe aus VLBI-Messungen (Lin et al. 2023).

**Hintergrund & Fachbegriffe:**
- **Parallaxe** $\varpi$: die scheinbare jährliche Bewegung eines Sterns durch den Umlauf der Erde. $d$ [pc] $= 1/\varpi$ [″]; 0.5 mas entsprechen 2000 pc.
- **Die Zahl 4.74:** 1 AE pro Jahr in km/s. Ein Objekt in 1 pc Entfernung, das sich mit 1″ pro Jahr bewegt, legt 1 AE pro Jahr zurück, also 4.74 km/s.
- **Nullpunkt der Gaia-Parallaxen:** Gaia DR3 misst Parallaxen im Mittel um etwa 0.02 bis 0.04 mas zu klein, abhängig von Helligkeit, Farbe und Ort (Lindegren et al. 2021). Die Korrektur macht den Pulsar etwa 5 % näher; der Code zeigt den unkorrigierten Wert.
- **VLBI:** Very Long Baseline Interferometry, Radioteleskope auf mehreren Kontinenten, zusammengeschaltet zu einer Auflösung von Millibogensekunden.

```python
yy_, xx_ = np.nonzero(NEB)
cov = np.cov(np.vstack([HALF - xx_, yy_ - HALF]).astype(float))
lam, vec = np.linalg.eigh(cov)
A_AX, B_AX = 2 * np.sqrt(lam[1]), 2 * np.sqrt(lam[0])                   # gleichmäßig gefüllte Ellipse: Varianz = a²/4
PA_AX = np.degrees(np.arctan2(vec[0, 1], vec[1, 1])) % 180            # Lagewinkel der großen Achse, Nord über Ost
mu_b = e_rate * B_AX                                                   # ″ pro Jahr am Rand der kurzen Achse
D_EXP = V_R_MAX / (4.74 * mu_b)
rel = np.sqrt(0.10 ** 2 + 0.10 ** 2 + (P60["t0_err"] / (t2 - T0)) ** 2)  # 10 % für v_r,max, 10 % für b, dazu e
plx, eplx = PSR["parallax"], PSR["parallax_error"]
D_GAIA = 1000 / plx
print(f"Nebelmaske als Ellipse: halbe Achsen a = {A_AX:.0f}″, b = {B_AX:.0f}″, Lagewinkel {PA_AX:.0f}°")
print(f"Expansionsrate e = {e_rate:.3e} pro Jahr -> Rand der kurzen Achse bewegt sich mit {mu_b * 1000:.0f} mas/Jahr")
print(f"Expansionsparallaxe: d = {V_R_MAX:.0f} km/s / (4.74 × {mu_b:.4f}″/Jahr) = {D_EXP:.0f} ± {rel * D_EXP:.0f} pc")
print(f"Gaia DR3: Parallaxe {plx:.3f} ± {eplx:.3f} mas -> d = 1/ϖ = {D_GAIA:.0f} pc "
      f"(68 %: {1000 / (plx + eplx):.0f} bis {1000 / (plx - eplx):.0f} pc)")
print(f"VLBI (Lin et al. 2023): 0.53 ± 0.06 mas -> {1000 / 0.53:.0f} pc")
print(f"Mit d = {D_GAIA:.0f} pc: Rand der langen Achse {4.74 * e_rate * A_AX * D_GAIA:.0f} km/s, der kurzen {4.74 * e_rate * B_AX * D_GAIA:.0f} km/s; "
      f"Größe des Nebels {2 * A_AX * D_GAIA / 206265:.1f} × {2 * B_AX * D_GAIA / 206265:.1f} pc")

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
ax = axes[0]
show(ax, smooth["PS1 r"], lo=20, hi=99.8, gamma=0.7)
ax.contour(NEB, levels=[0.5], colors="yellow", linewidths=1, extent=EXTENT, origin="lower")
xm, ym = (HALF - xx_).mean(), (yy_ - HALF).mean()
tt = np.linspace(0, 2 * np.pi, 300)
pa = np.radians(PA_AX)
ex_xi = xm + A_AX * np.cos(tt) * np.sin(pa) + B_AX * np.sin(tt) * np.cos(pa)
ex_eta = ym + A_AX * np.cos(tt) * np.cos(pa) - B_AX * np.sin(tt) * np.sin(pa)
ax.plot(ex_xi, ex_eta, "c--", lw=1.5, label=f"Ellipse: a = {A_AX:.0f}″, b = {B_AX:.0f}″")
for L, ang, col in [(A_AX, pa, "c"), (B_AX, pa + np.pi / 2, "m")]:
    ax.plot([xm - L * np.sin(ang), xm + L * np.sin(ang)], [ym - L * np.cos(ang), ym + L * np.cos(ang)], color=col, lw=1.5)
ax.set_xlim(260, -260)
ax.set_ylim(-220, 220)
ax.set_xlabel("ξ [″] (Osten links)")
ax.set_ylabel("η [″]")
ax.set_title("Nebelmaske (gelb) und Ellipse gleicher Momente", fontsize=10)
ax.legend(fontsize=9, loc="lower left")
ax = axes[1]
dd = np.linspace(800, 4000, 600)
pdf = lambda p, s: np.exp(-0.5 * ((1000 / dd - p) / s) ** 2) * 1000 / dd ** 2      # Gauß in der Parallaxe -> Dichte in d
for (p, s), lab, col in [((plx, eplx), "Gaia DR3 (Pulsar)", "tab:blue"), ((0.53, 0.06), "VLBI, Lin et al. (2023)", "tab:green")]:
    y = pdf(p, s)
    ax.plot(dd, y / y.max(), color=col, lw=2, label=f"{lab}: {1000 / p:.0f} pc")
y = np.exp(-0.5 * ((dd - D_EXP) / (rel * D_EXP)) ** 2)
ax.plot(dd, y, color="tab:red", lw=2, label=f"Expansionsparallaxe (eigene Messung): {D_EXP:.0f} pc")
ax.axvline(2000, color="grey", ls=":", label="üblicher Wert 2 kpc (Trimble 1973)")
ax.set_xlabel("Entfernung d [pc]")
ax.set_ylabel("Wahrscheinlichkeitsdichte (Maximum = 1)")
ax.set_title("Entfernung des Krebsnebels", fontsize=10)
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Die Nebelmaske entspricht einer Ellipse mit halben Achsen 190″ und 144″, die lange Achse liegt bei 125° (von Nord über Ost). Der Rand der kurzen Achse bewegt sich mit 158 mas pro Jahr. Mit 1500 km/s entlang der Sichtlinie folgt d = 2003 ± 283 pc. Gaia gibt für den Pulsar 0.511 ± 0.079 mas, also 1957 pc (68 %: 1695 bis 2313 pc), VLBI 1887 pc. Bei 1957 pc misst der Nebel 3.6 × 2.7 pc, und sein Rand bewegt sich mit 1465 km/s (kurze Achse) bis 1932 km/s (lange Achse) quer zur Sichtlinie.
* Die **linke Abbildung** zeigt die Maske (gelb) und die Ellipse gleicher Momente (gestrichelt) mit ihren Achsen. Die Maske ist kein glatter Umriss: Im Südwesten und im Osten ragen Ausbuchtungen über die Ellipse, im Nordosten bleibt die Maske darunter. Wie weit die Maske reicht, hängt an der Schwelle; mit 8 % statt 3 % wird die kurze Achse 122″ und die Entfernung 2360 pc (Variation 4).
* Die **rechte Abbildung** beantwortet die Frage: Die Expansionsparallaxe (rot) und die Parallaxen aus Gaia (blau) und VLBI (grün) überlappen; alle drei liegen nahe den 2 kpc, die man seit Trimble (1973) annimmt. Zur Vorhersage: Bei 1957 pc entsprechen ±1500 km/s entlang der Sichtlinie der Geschwindigkeit am Rand der kurzen Achse (1465 km/s), nicht der langen (1932 km/s). Der Nebel ist also entlang der Sichtlinie etwa so tief wie seine kurze Achse, wie Trimble annahm. Die Gaia-Verteilung ist nach rechts schief, weil $d = 1/\varpi$ bei kleinen Parallaxen schnell wächst.

##### *11. Der Pulsar bremst*

Wie schnell dreht sich der Pulsar, wie schnell wird er langsamer, und wie viel Leistung verliert er dabei? Vorher vorhersagen: Wie viele Umdrehungen pro Sekunde hat der Pulsar seit 1982 verloren, ungefähr?

Die Ephemeriden von Jodrell Bank geben für jeden Monat seit Februar 1982 die Rotationsfrequenz $\nu$ und ihre Ableitung $\dot\nu$. Daraus folgen die Größen, die man aus einer einzigen Messung von Periode $P = 1/\nu$ und Periodenzunahme $\dot P = -\dot\nu/\nu^2$ ableitet (Theorie, Abschnitt 8): die Leistung, die der Pulsar an Rotationsenergie verliert, $\dot E = 4\pi^2 I \nu |\dot\nu|$, das charakteristische Alter $\tau_c = \nu / (2|\dot\nu|)$ und das Magnetfeld am Äquator $B \approx 3.2\cdot10^{19}\sqrt{P\dot P}$ Gauß. Sprünge der Frequenz, die Glitches, findet der Code als Monate, in denen $\nu$ um mehr als $10^{-7}$ Hz über der Vorhersage aus dem Vormonat liegt.

**Hintergrund & Fachbegriffe:**
- **Pulsar:** ein schnell rotierender Neutronenstern, dessen Strahlungskegel wie ein Leuchtturm über die Erde streicht; der Crab-Pulsar dreht sich etwa 30-mal pro Sekunde.
- **Trägheitsmoment** $I$: für einen Neutronenstern mit 1.4 Sonnenmassen und 12 km Radius etwa $10^{45}$ g cm²; es geht linear in $\dot E$ ein und ist auf etwa 50 % unsicher.
- **Glitch:** ein plötzlicher Sprung der Rotationsfrequenz um $\Delta\nu/\nu \sim 10^{-9}$ bis $10^{-7}$, vermutlich weil supraflüssige Neutronen im Inneren ihren Drehimpuls schlagartig an die Kruste abgeben. Danach erholt sich $\dot\nu$ über Wochen bis Monate.
- **MJD:** Modified Julian Date, die fortlaufende Tageszählung der Astronomie; MJD 45015 ist der 15. Februar 1982.

```python
rows = []
for p in JB_FILES:
    for line in p.read_text(encoding="latin-1").splitlines():
        t = line.split()
        if len(t) < 8 or not re.fullmatch(r"\d{2}", t[0]) or not re.fullmatch(r"\d{5}", t[3]):
            continue
        k = next((i for i, s in enumerate(t) if re.fullmatch(r"\d{2}\.\d{9,11}", s)), None)   # die Frequenz, z. B. 29.5340803404
        if k is None:
            continue
        ndig = len(t[k].split(".")[1])
        rows.append(dict(mjd=int(t[3]), nu=float(t[k]), s_nu=float(t[k + 1]) * 10.0 ** -ndig,
                         nudot=float(t[k + 2]) * 1e-15, s_nudot=float(t[k + 3]) * 1e-15))
JB = pd.DataFrame(rows).drop_duplicates("mjd").sort_values("mjd").reset_index(drop=True)
JB["year"] = Time(JB["mjd"].values, format="mjd").decimalyear

# Glitches: Frequenz liegt mehr als 1e-7 Hz über der Vorhersage aus dem Vormonat; ein Ereignis pro 60 Tage
dsec = JB["mjd"].diff().values * 86400
pred = JB["nu"].shift(1) + 0.5 * (JB["nudot"].shift(1) + JB["nudot"]) * dsec
JB["jump"] = JB["nu"] - pred
GLITCH = []
for m, j in zip(JB["mjd"], JB["jump"]):
    if j > 1e-7 and (not GLITCH or m - GLITCH[-1] > 60):
        GLITCH.append(m)
GLITCH = np.array(GLITCH)

now = JB.iloc[-1]
nu, nud = now["nu"], now["nudot"]
P, Pdot = 1 / nu, -nud / nu ** 2
EDOT = 4 * np.pi ** 2 * I_NS * nu * -nud
TAU_C = nu / (-2 * nud) / YEAR
B_EQ = 3.2e19 * np.sqrt(P * Pdot)
print(f"{len(JB)} Monatswerte von {Time(JB['mjd'].min(), format='mjd').iso[:10]} bis {Time(JB['mjd'].max(), format='mjd').iso[:10]}; "
      f"Fehler typisch {JB['s_nu'].median():.0e} Hz in ν und {JB['s_nudot'].median():.0e} Hz/s in ν̇")
print(f"ν: {JB['nu'].iloc[0]:.6f} Hz -> {nu:.6f} Hz (Verlust {JB['nu'].iloc[0] - nu:.4f} Hz in {now['year'] - JB['year'].iloc[0]:.1f} Jahren)")
print(f"Heute: P = {P * 1e3:.4f} ms, Ṗ = {Pdot:.3e} s/s, die Periode wächst also um {Pdot * 86400 * 1e9:.0f} Nanosekunden pro Tag")
print(f"Ė = 4π² I ν |ν̇| = {EDOT:.2e} erg/s = {EDOT / 3.828e33:.1e} Sonnenleuchtkräfte (I = {I_NS:.0e} g cm²)")
print(f"charakteristisches Alter τ_c = ν / (2|ν̇|) = {TAU_C:.0f} Jahre (wahres Alter {now['year'] - T_SN:.0f} Jahre); B = {B_EQ:.2e} G")
print(f"Glitches (Sprung > 1e-7 Hz): {len(GLITCH)}, in den Jahren {', '.join(f'{y:.1f}' for y in Time(GLITCH, format='mjd').decimalyear)}")
big = JB.loc[JB["jump"].idxmax()]
print(f"größter Sprung: {big['jump'] * 1e6:.1f} µHz (Δν/ν = {big['jump'] / big['nu']:.1e}) im Jahr {big['year']:.1f}")

fig, axes = plt.subplots(1, 2, figsize=(16, 5.6))
ax = axes[0]
ax.plot(JB["year"], JB["nudot"] * 1e10, "k.", ms=3)
for g in Time(GLITCH, format="mjd").decimalyear:
    ax.axvline(g, color="tab:red", lw=0.8, alpha=0.6)
ax.set_xlabel("Jahr")
ax.set_ylabel("ν̇ [10⁻¹⁰ Hz/s]")
ax.set_title("Abbremsung ν̇ (rote Linien: Glitches)", fontsize=10)
ax = axes[1]
jumps = JB.set_index("mjd").loc[GLITCH, "jump"]
ax.stem(Time(GLITCH, format="mjd").decimalyear, jumps.values / JB["nu"].mean(), linefmt="tab:red", markerfmt="o", basefmt=" ")
ax.set_yscale("log")
ax.set_xlim(JB["year"].min(), JB["year"].max() + 0.5)
ax.set_xlabel("Jahr")
ax.set_ylabel("Δν/ν (Sprung gegen die Vorhersage aus dem Vormonat)")
ax.set_title("Größe der gefundenen Glitches", fontsize=10)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** 559 Monatswerte von Februar 1982 bis September 2026. In 44.6 Jahren hat der Pulsar 0.525 Umdrehungen pro Sekunde verloren, von 30.059 auf 29.534 Hz; das beantwortet die Vorhersage. Heute ist P = 33.86 ms und Ṗ = 4.19·10⁻¹³; die Periode wächst um 36 Nanosekunden pro Tag. Daraus folgen Ė = 4.3·10³⁸ erg/s, gut 100 000 Sonnenleuchtkräfte, ein Magnetfeld von 3.8·10¹² G und ein charakteristisches Alter von 1279 Jahren statt der wahren 972.
* Die **linke Abbildung** zeigt ν̇: Die Bremsung wird langsam schwächer, von −3.81 auf −3.66·10⁻¹⁰ Hz/s, fast auf einer Geraden; ihre Steigung ist ν̈ (Schritt 12). An den roten Linien springt ν̇ nach unten: Der Pulsar bremst nach einem Glitch kurz stärker und erholt sich über einige Monate. Die bleibenden Reste sind zu klein, um sie hier zu sehen; sie zeigen sich im Bremsindex über alle Jahre (Schritt 12).
* Die **rechte Abbildung** zeigt die Größe der 10 Sprünge über 10⁻⁷ Hz, von Δν/ν ≈ 4·10⁻⁹ bis 5·10⁻⁷. Der größte, 2017, ist der größte Glitch, der beim Crab je gemessen wurde. Viele kleinere Glitches (Lyne et al. 2015 zählen 24 bis 2013) bleiben unter der Schwelle, weil die Monatswerte sie nicht auflösen.

##### *12. Bremsindex und Anfangsperiode*

Bremst der Pulsar so, wie ein rotierender Magnet im Vakuum bremsen sollte, und wie schnell drehte er sich bei seiner Geburt? Vorher vorhersagen: Ist der Bremsindex größer oder kleiner als 3, und war der Pulsar 1054 doppelt so schnell wie heute oder nur etwas schneller?

Zwischen zwei Glitches ändert sich $\dot\nu$ fast linear; die Steigung ist $\ddot\nu$. Der Code passt in jedem Abschnitt zwischen zwei Glitches eine Gerade an $\dot\nu(t)$ an, lässt die ersten 150 Tage nach jedem Glitch weg (dort erholt sich der Pulsar noch) und benutzt nur Abschnitte, die länger als 500 Tage sind. Jeder Abschnitt gibt einen Bremsindex $n = \nu\ddot\nu/\dot\nu^2$. Zum Vergleich passt er eine einzige Gerade über alle 44 Jahre an, Glitches eingeschlossen.

Mit dem Bremsindex und dem bekannten Alter folgt die Anfangsperiode $P_0$ (Theorie, Abschnitt 8) und daraus die Rotationsenergie, die der Pulsar seit 1054 abgegeben hat.

**Hintergrund & Fachbegriffe:**
- **Bremsindex** $n$: der Exponent im Bremsgesetz $\dot\nu = -K\nu^n$. Ein Dipolmagnet im Vakuum gibt $n = 3$; ein Wind aus Teilchen, der Drehimpuls mitnimmt, oder ein wachsendes Magnetfeld machen $n$ kleiner.
- **$\ddot\nu$:** die zweite Ableitung der Frequenz; beim Crab etwa $10^{-20}$ Hz/s², über ein Jahr eine Änderung von $\dot\nu$ um 0.1 %.
- **Anfangsperiode** $P_0$: die Periode bei der Geburt des Neutronensterns. Sie verrät, wie viel Drehimpuls der Kern des Sterns beim Kollaps hatte.

```python
def braking_segments(skip=150, min_span=500):
    # in jedem Abschnitt zwischen zwei Glitches: Gerade an nu-dot(t) -> nu-ddot, daraus n = nu nu-ddot / nu-dot²
    edges = np.r_[JB["mjd"].min() - 1, GLITCH, JB["mjd"].max() + 1]
    out = []
    for a_, b_ in zip(edges[:-1], edges[1:]):
        s = JB[(JB["mjd"] >= a_ + (skip if a_ in GLITCH else 0)) & (JB["mjd"] < b_ - 1)]
        if len(s) < 8 or s["mjd"].max() - s["mjd"].min() < min_span:
            continue
        t = (s["mjd"] - s["mjd"].mean()).values * 86400
        p = np.polyfit(t, s["nudot"], 1)
        res = s["nudot"] - np.polyval(p, t)
        err = np.sqrt(np.sum(res ** 2) / (len(s) - 2) / np.sum((t - t.mean()) ** 2))
        nu_m = s["nu"].mean()
        out.append(dict(start=s["year"].min(), end=s["year"].max(), n_pts=len(s), nuddot=p[0],
                        n=nu_m * p[0] / p[1] ** 2, n_err=nu_m * err / p[1] ** 2))
    return pd.DataFrame(out)


SEG = braking_segments()
N_BRAKE = SEG["n"].median()
tw = (JB["mjd"] - JB["mjd"].mean()).values * 86400
pw = np.polyfit(tw, JB["nudot"], 1)
N_ALL = JB["nu"].mean() * pw[0] / pw[1] ** 2
print(SEG.round({"start": 1, "end": 1, "n": 3, "n_err": 3}).to_string(index=False, formatters={"nuddot": "{:.3e}".format}))
print(f"Bremsindex zwischen Glitches: Median {N_BRAKE:.2f} (Abschnitte {SEG['n'].min():.2f} bis {SEG['n'].max():.2f}); "
      f"über alle {now['year'] - JB['year'].iloc[0]:.0f} Jahre mit Glitches: {N_ALL:.2f}")

AGE = now["year"] - T_SN
def p0_from(n):
    # P0 aus P, Ṗ, n und dem bekannten Alter: Alter = P / ((n - 1) Ṗ) · [1 - (P0 / P)^(n - 1)]
    return P * (1 - AGE * YEAR * (n - 1) * Pdot / P) ** (1 / (n - 1))
E_ROT = {}
for n in (N_BRAKE, N_ALL, 3.0):
    P0 = p0_from(n)
    E_ROT[n] = 2 * np.pi ** 2 * I_NS * (1 / P0 ** 2 - 1 / P ** 2)
    print(f"n = {n:.2f}: Anfangsperiode P0 = {P0 * 1e3:.1f} ms ({1 / P0:.0f} Umdrehungen pro Sekunde), "
          f"seit 1054 abgegebene Rotationsenergie {E_ROT[n]:.2e} erg")
P0_BEST = p0_from(N_BRAKE)

fig, axes = plt.subplots(1, 2, figsize=(16, 5.6))
ax = axes[0]
ax.errorbar((SEG["start"] + SEG["end"]) / 2, SEG["n"], xerr=(SEG["end"] - SEG["start"]) / 2, yerr=SEG["n_err"],
            fmt="o", color="k", capsize=3, label="Abschnitt zwischen zwei Glitches")
ax.axhline(3, color="tab:blue", ls="--", label="Dipol im Vakuum: n = 3")
ax.axhline(2.50, color="tab:green", ls=":", lw=2, label="Lyne et al. (2015): n = 2.50 zwischen Glitches")
ax.axhline(N_ALL, color="tab:red", ls="-.", label=f"eine Gerade über alles: n = {N_ALL:.2f}")
for g in Time(GLITCH, format="mjd").decimalyear:
    ax.axvline(g, color="tab:red", lw=0.6, alpha=0.4)
ax.set_xlabel("Jahr")
ax.set_ylabel("Bremsindex n = ν ν̈ / ν̇²")
ax.set_ylim(2.0, 3.1)
ax.set_title("Bremsindex aus 44 Jahren Jodrell Bank", fontsize=10)
ax.legend(fontsize=8, loc="upper left")
ax = axes[1]
yrs = np.linspace(T_SN, now["year"], 300)
for n, col in [(N_BRAKE, "k"), (N_ALL, "tab:red"), (3.0, "tab:blue")]:
    P0 = p0_from(n)
    # P(t) aus dem Bremsgesetz: P^(n-1) wächst linear mit der Zeit
    Pt = (P0 ** (n - 1) + (P ** (n - 1) - P0 ** (n - 1)) * (yrs - T_SN) / AGE) ** (1 / (n - 1))
    ax.plot(yrs, Pt * 1e3, color=col, lw=2, label=f"n = {n:.2f}: P0 = {P0 * 1e3:.1f} ms")
ax.plot(JB["year"], 1e3 / JB["nu"], "o", color="orange", ms=2, label="Jodrell Bank 1982–2026")
ax.set_xlabel("Jahr")
ax.set_ylabel("Periode P [ms]")
ax.set_title("Wie schnell drehte sich der Pulsar seit 1054?", fontsize=10)
ax.legend(fontsize=9)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Acht Abschnitte zwischen Glitches geben Bremsindizes von 2.22 bis 2.55, im Median 2.51; Lyne et al. (2015) fanden 2.50. Eine einzige Gerade über alle 44 Jahre ergibt 2.25. Zur Vorhersage: Der Index ist kleiner als 3. Mit n = 2.51 war die Anfangsperiode 19.3 ms, der Pulsar drehte sich also 52-mal statt heute 30-mal pro Sekunde, und er hat seit 1054 3.6·10⁴⁹ erg Rotationsenergie abgegeben. Mit n = 3 wären es 16.6 ms und 5.5·10⁴⁹ erg.
* Die **linke Abbildung:** Fünf Abschnitte liegen bei 2.5 bis 2.55, nahe dem Wert von Lyne et al. Die Abschnitte 1997 bis 2004, in der Zeit mit den meisten Glitches, liegen bei 2.22, wie Lyne et al. es auch gesehen haben (2.3); der Grund ist offen. Die Fehlerbalken aus der Streuung der Monatswerte sind klein (0.003 bis 0.03), die Unterschiede zwischen den Abschnitten also echt.
* Die **rechte Abbildung** rechnet die Periode mit dem Bremsgesetz bis 1054 zurück. Alle drei Kurven treffen heute die Messungen von Jodrell Bank (orange); sie unterscheiden sich erst weit in der Vergangenheit. Wie schnell der Pulsar geboren wurde, hängt also am Bremsindex, den man nur heute misst; die Anfangsperiode ist auf etwa ±2 ms bekannt.

##### *13. Das Konvergenzjahr gegen die Literatur*

Wo liegt die eigene Messung zwischen den Konvergenzjahren, die seit 1939 veröffentlicht wurden, und wie stark sind die Filamente seit 1054 beschleunigt worden? Vorher vorhersagen: Liegen die älteren Messungen mit der Methode der kleinsten Streuung eher früher oder später als die Fit-Methoden?

Die veröffentlichten Werte stammen aus sehr verschiedenen Daten: Fotoplatten am Mount-Wilson- und Palomar-Teleskop (Duncan 1939, Trimble 1968, Nugent 1998), eine Neuauswertung von Trimbles Messungen (Wyckoff & Murray 1977), eine Bayes-Auswertung der Daten von Nugent (Bietenholz & Nugent 2015), der nördliche Jet allein (Rudie et al. 2008) und fast 20 000 Verschiebungen aus CFHT-Bildern von 2007 bis 2019 (Martin et al. 2025), dort mit beiden Methoden.

Aus dem Konvergenzjahr folgt, wie stark die Filamente beschleunigt wurden (Theorie, Abschnitt 6). Wächst der Radius wie $r \propto (t - t_\text{SN})^\beta$, dann gilt für zwei Epochen $1 - k = \left(\frac{t_1 - t_\text{SN}}{t_2 - t_\text{SN}}\right)^\beta$. Ohne Beschleunigung ist $\beta = 1$; das Verhältnis der heutigen Geschwindigkeit zur mittleren seit 1054 ist $\beta$.

**Hintergrund & Fachbegriffe:**
- **Potenzgesetz der Expansion:** $r \propto (t - t_\text{SN})^\beta$. Bietenholz & Nugent (2015) fanden für die Filamente $\beta = 1.040 \pm 0.039$, für den Synchrotronnebel im Radiobereich $1.264 \pm 0.049$; Martin et al. (2025) geben das Verhältnis $v/v_h = 1.06$ an, mit $v_h$ der Geschwindigkeit bei unbeschleunigter Expansion seit 1054.
- **Bayes-Auswertung:** statt des einen Jahres mit der kleinsten Streuung die Wahrscheinlichkeit jedes Konvergenzjahres bei gegebenen Messfehlern; sie vermeidet die Verschiebung zur Gegenwart.

```python
LIT = pd.DataFrame([
    ("Duncan (1939)", 1172, np.nan, "kleinste Streuung"),
    ("Trimble (1968)", 1140, 15, "kleinste Streuung"),
    ("Wyckoff & Murray (1977)", 1120, 7, "Fit an Trimbles Daten"),
    ("Nugent (1998)", 1130, 16, "kleinste Streuung"),
    ("Rudie et al. (2008), nur Jet", 1055, 24, "Fit"),
    ("Bietenholz & Nugent (2015)", 1091, 34, "Bayes, Daten von Nugent"),
    ("Martin et al. (2025)", 1140.5, 1.4, "kleinste Streuung"),
    ("Martin et al. (2025)", 1105.5, 0.5, "Fit"),
    ("eigene Messung", round(t_ca), np.nan, "kleinste Streuung"),
    ("eigene Messung", round(T0), np.hypot(P60["t0_err"], T0_SYS), "Fit"),
], columns=["Quelle", "Jahr", "Fehler", "Methode"])
fmt_year = lambda x: f"{x:.1f}" if x % 1 else f"{x:.0f}"
fmt_err = lambda x: "–" if not np.isfinite(x) else (f"{x:.1f}" if x < 2 else f"{x:.0f}")
print(LIT.to_string(index=False, na_rep="–", formatters={"Jahr": fmt_year, "Fehler": fmt_err}))

t1 = EPOCH["POSS-I E"]
BETA = np.log(1 - P60["k"]) / np.log((t1 - T_SN) / (t2 - T_SN))
dbeta = np.std([np.log(1 - P60["dt"] / (t2 - t)) / np.log((t1 - T_SN) / (t2 - T_SN)) for t in rng.normal(T0, np.hypot(P60["t0_err"], T0_SYS), 2000)])
v_hom = 1 / (t2 - T_SN)
print(f"Beschleunigung: β = {BETA:.3f} ± {dbeta:.3f}; die Filamente sind heute {100 * (BETA - 1):.1f} % schneller als im Mittel seit 1054")
print(f"Zum Vergleich: Bietenholz & Nugent (2015) β = 1.040 ± 0.039; Martin et al. (2025) v/v_h = 1.06 "
      f"(eigene Messung im selben Maß: {(t2 - T_SN) / (t2 - T0):.3f})")

fig, axes = plt.subplots(1, 2, figsize=(16, 5.8), gridspec_kw={"width_ratios": [1.4, 1]})
ax = axes[0]
cols = {"kleinste Streuung": "tab:blue", "Fit": "tab:red", "Fit an Trimbles Daten": "tab:red", "Bayes, Daten von Nugent": "tab:purple"}
for i, row_ in LIT.iterrows():
    own = row_["Quelle"] == "eigene Messung"
    ax.errorbar(row_["Jahr"], i, xerr=row_["Fehler"] if np.isfinite(row_["Fehler"]) else None, fmt="*" if own else "o",
                ms=14 if own else 7, color=cols[row_["Methode"]], capsize=3)
ax.set_yticks(range(len(LIT)), [f"{q} – {m}" for q, m in zip(LIT["Quelle"], LIT["Methode"])], fontsize=9)
ax.invert_yaxis()
ax.axvline(T_SN, color="grey", ls=":", lw=2)
ax.text(T_SN + 2, len(LIT) - 0.6, "SN 1054", color="grey", fontsize=9)
ax.set_xlabel("Konvergenzjahr")
ax.set_title("Konvergenzjahre der Filamente (blau: kleinste Streuung, rot: Fit)", fontsize=10)
ax = axes[1]
tc_of = lambda b: t2 - P60["dt"] / (1 - ((t1 - T_SN) / (t2 - T_SN)) ** b)  # Konvergenzjahr, das unsere Epochen bei gegebenem β messen
bb = np.linspace(1.0, 1.10, 200)
ax.plot(bb, tc_of(bb), "k-", lw=2, label="Konvergenzjahr bei r ∝ (t − 1054)^β")
ax.errorbar(BETA, T0, xerr=dbeta, yerr=np.hypot(P60["t0_err"], T0_SYS), fmt="*", ms=14, color="tab:red", capsize=3,
            label=f"eigene Messung: β = {BETA:.3f}")
ax.errorbar(1.040, tc_of(1.040), xerr=0.039, fmt="s", color="tab:purple", capsize=3, label="Bietenholz & Nugent (2015): β = 1.040")
ax.plot(1.06, 1105.5, "o", color="tab:green", ms=8, label="Martin et al. (2025): v/v_h = 1.06, Fit 1105.5")
ax.axhline(T_SN, color="grey", ls=":", lw=2)
ax.set_xlabel("Exponent β (heutige Geschwindigkeit / mittlere seit 1054)")
ax.set_ylabel("Konvergenzjahr")
ax.set_title("Wie viel Beschleunigung steckt im Konvergenzjahr?", fontsize=10)
ax.legend(fontsize=8, loc="lower right")
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Die Tabelle stellt die eigene Messung neben neun veröffentlichte Werte. Aus k = 0.0657 und den Epochen folgt β = 1.050 ± 0.006: Die Filamente sind heute 5.0 % schneller als im Mittel seit 1054. Im Maß von Martin et al. (heutige Geschwindigkeit geteilt durch die unbeschleunigte) sind es 1.048 gegen deren 1.06.
* Die **linke Abbildung** beantwortet die Vorhersage: Die blauen Werte (kleinste Streuung) liegen später als die roten (Fit); bei Martin et al. sind es mit denselben Daten 35 Jahre Unterschied, hier 11. Die eigene Fit-Messung (1098) liegt zwischen Bietenholz & Nugent (1091 ± 34) und Martin et al. (1105.5), anderthalb eigene Fehler vor Martin et al. Nur der nördliche Jet (Rudie et al. 2008) zeigt kein Zeichen einer Beschleunigung.
* Die **rechte Abbildung** übersetzt β in das Konvergenzjahr, das unsere beiden Epochen messen würden: β = 1 gibt genau 1054, jedes Prozent Beschleunigung verlegt das Jahr um etwa 9 Jahre. Die eigene Messung, Bietenholz & Nugent (β = 1.040, Konvergenzjahr 1090) und Martin et al. (1.06, Konvergenzjahr 1105.5) liegen alle auf derselben Kurve. Die Unterschiede in der Literatur entsprechen also nur 1 bis 2 Prozent Unterschied in der Beschleunigung.

##### *14. Energiebilanz: Kann der Pulsar die Filamente beschleunigen?*

Reicht die Energie, die der Pulsar seit 1054 verloren hat, um die Filamente auf ihre heutige Geschwindigkeit zu beschleunigen? Vorher vorhersagen: Welcher Anteil der abgegebenen Rotationsenergie steckt in der zusätzlichen Bewegungsenergie der Filamente, eher 1 %, 10 % oder 100 %?

Die Bewegungsenergie der Filamente schätzt der Code aus den gemessenen Verschiebungen: Mit der Gaia-Entfernung wird jede Verschiebung zu einer Geschwindigkeit quer zur Sichtlinie, $v_\perp = 4.74\,\mu\,d$. Bei Bewegungen in alle Richtungen ist das mittlere Quadrat der räumlichen Geschwindigkeit $\tfrac32\langle v_\perp^2\rangle$. Mit der Masse von 4.6 ± 1.8 Sonnenmassen (Fesen et al. 1997) folgt $E_\text{kin} = \tfrac12 M\langle v^2\rangle$. Ohne Beschleunigung hätten die Filamente heute nur die mittlere Geschwindigkeit seit 1054, also $v/\beta$; die zusätzliche Energie ist $\Delta E = E_\text{kin}(1 - 1/\beta^2)$ (Theorie, Abschnitt 9). Dagegen steht die Rotationsenergie, die der Pulsar seit 1054 verloren hat (Schritt 12), und die Leistung, die der Nebel heute als Synchrotronstrahlung abgibt, etwa $1.3\cdot10^{38}$ erg/s (Hester 2008).

**Hintergrund & Fachbegriffe:**
- **erg:** die Energieeinheit im cgs-System, $10^{-7}$ Joule. Eine typische Supernova setzt etwa $10^{51}$ erg als Bewegungsenergie frei, die Sonne strahlt $3.8\cdot10^{33}$ erg pro Sekunde ab.
- **Pulsarwindnebel:** Der Pulsar gibt seine Rotationsenergie als Wind aus Elektronen, Positronen und Magnetfeld ab. Dieser Wind bläst eine Blase, die als Synchrotronnebel leuchtet und von innen gegen die Hülle aus Filamenten drückt.
- **Gewichtung:** Die Ausschnitte stehen für gleich große Flächen am Himmel, nicht für gleich viel Masse; die Schätzung von $\langle v^2\rangle$ ist deshalb nur auf etwa 20 % genau.

```python
d = P60["df"][P60["keep"]]
v_perp = np.hypot(d["dxi"], d["deta"]) / dt60 * 4.74 * D_GAIA                 # km/s quer zur Sichtlinie
v2 = 1.5 * np.mean(v_perp ** 2) * 1e10                                        # (cm/s)², räumlich
E_KIN = 0.5 * M_FIL[0] * M_SUN * v2
E_KIN_ERR = E_KIN * np.hypot(M_FIL[1] / M_FIL[0], 0.2)
DE_ACC = E_KIN * (1 - 1 / BETA ** 2)
DE_ERR = np.hypot(DE_ACC * E_KIN_ERR / E_KIN, E_KIN * 2 * dbeta / BETA ** 3)
L_SYN = 1.3e38
E_R = E_ROT[N_BRAKE]
print(f"Geschwindigkeit quer zur Sichtlinie (d = {D_GAIA:.0f} pc): Median {np.median(v_perp):.0f} km/s, höchste {v_perp.max():.0f} km/s; "
      f"räumlich rms {np.sqrt(v2) / 1e5:.0f} km/s")
print(f"Bewegungsenergie der Filamente: {E_KIN:.1e} ± {E_KIN_ERR:.1e} erg")
print(f"davon durch Beschleunigung (β = {BETA:.3f}): {DE_ACC:.1e} ± {DE_ERR:.1e} erg")
print(f"Rotationsenergie, die der Pulsar seit 1054 verloren hat (n = {N_BRAKE:.2f}): {E_R:.1e} erg "
      f"-> Anteil der Beschleunigung {DE_ACC / E_R:.0%}")
print(f"Synchrotronleistung heute {L_SYN:.1e} erg/s = {L_SYN / EDOT:.0%} von Ė; über {AGE:.0f} Jahre bei heutiger Leistung "
      f"{L_SYN * AGE * YEAR:.1e} erg")

items = [("Rotationsenergie, seit 1054\nverloren (Pulsar)", E_R, 0.5 * E_R, "tab:blue"),
         ("Bewegungsenergie der\nFilamente heute", E_KIN, E_KIN_ERR, "tab:orange"),
         ("davon durch Beschleunigung\nseit 1054", DE_ACC, DE_ERR, "tab:red"),
         ("Synchrotronstrahlung,\nheutige Leistung × Alter", L_SYN * AGE * YEAR, 0.5 * L_SYN * AGE * YEAR, "tab:purple"),
         ("Ė heute × 1 Jahr", EDOT * YEAR, 0, "grey")]
fig, ax = plt.subplots(figsize=(12, 5.2))
for i, (lab, val, err, col) in enumerate(items):
    ax.barh(i, val, color=col, alpha=0.8)
    if err:
        ax.errorbar(val, i, xerr=[[min(err, 0.9 * val)], [err]], color="k", capsize=4)
    ax.text(val * 1.15 + (err or 0), i, f"{val:.1e} erg", va="center", fontsize=9)
ax.set_xscale("log")
ax.set_xlim(1e45, 1e51)
ax.set_yticks(range(len(items)), [it[0] for it in items], fontsize=9)
ax.invert_yaxis()
ax.set_xlabel("Energie [erg]")
ax.set_title("Energiebilanz des Krebsnebels (Fehlerbalken: Masse, Trägheitsmoment, β)", fontsize=10)
fig.tight_layout()
plt.show()
```

* Die **Ausgabe:** Bei 1957 pc bewegen sich die Ausschnitte quer zur Sichtlinie im Median mit 1237 km/s, die schnellsten mit 2130 km/s; räumlich ergibt das rms 1556 km/s. Mit 4.6 Sonnenmassen ist die Bewegungsenergie der Filamente 1.1·10⁵⁰ erg (± 45 %), davon 1.0·10⁴⁹ erg aus der Beschleunigung. Der Pulsar hat seit 1054 3.6·10⁴⁹ erg abgegeben; die Beschleunigung braucht davon 28 %. Die Synchrotronstrahlung nimmt heute 30 % von Ė mit, bei heutiger Leistung über 972 Jahre 4·10⁴⁸ erg.
* Die **Abbildung** (logarithmische Achse) beantwortet die Frage: Die Bewegungsenergie der Filamente (orange) ist dreimal so groß wie alles, was der Pulsar abgegeben hat (blau); sie stammt aus der Explosion selbst, die beim Krebsnebel mit etwa 10⁵⁰ erg ungewöhnlich schwach war (eine typische Supernova hat 10⁵¹ erg). Der Teil, den die Beschleunigung hinzugefügt hat (rot), ist aber kleiner als die Pulsarenergie. Zur Vorhersage: Es sind etwa 10 bis 30 %, nicht 1 % und nicht 100 %.
* Was bleibt, ist ein Budget, das aufgeht: Gut ein Viertel der Pulsarenergie in die Bewegung der Hülle, mindestens ein Zehntel als Synchrotronstrahlung (früher war der Nebel heller, also eher mehr), der Rest als Teilchen und Magnetfeld in der Blase. Am unsichersten ist die Masse der Filamente (± 40 %); schon dadurch schwankt der Anteil der Beschleunigung zwischen etwa 15 und 40 %. Ė heute mal ein Jahr (grau) zeigt den Maßstab: Der Pulsar braucht etwa 2700 Jahre heutiger Leistung für die Energie, die er seit 1054 abgegeben hat, weil er früher viel schneller drehte.

<details><summary>1. Was ist das Konvergenzjahr, und warum ist es nicht dasselbe wie das Jahr der Explosion?</summary>

Das Konvergenzjahr ist das Jahr, in dem alle Filamente in einem Punkt gewesen wären, wenn sie sich immer so schnell bewegt hätten wie heute. Es stimmt nur dann mit dem Explosionsjahr überein, wenn sie nie beschleunigt oder gebremst wurden. Beim Krebsnebel liegt es bei 1098, also 44 Jahre nach 1054: Die Filamente sind schneller geworden.
</details>

<details><summary>2. Warum braucht man Gaia, um eine Platte von 1951 auszurichten, und warum genügt dafür nicht die Plattenlösung im FITS-Kopf?</summary>

Die Plattenlösung im Kopf lag um 0.8″ daneben, mehr als ein Zehntel der typischen Verschiebung von 7″. Gaia liefert für Hunderte Sterne im Feld Orte und Eigenbewegungen; zurückgerechnet auf 1951 sind sie ein Gitter von Bezugspunkten, an das man ein eigenes Polynom anpasst (Rest 0.19″). Ohne die Eigenbewegungen steigt der Rest auf 0.26″, und das Zentrum rückt um 1.8″ (Variation 3).
</details>

<details><summary>3. Warum vergleicht der Code hochpassgefilterte Bilder und nicht die Bilder selbst?</summary>

Eine Fotoplatte misst eine Schwärzung, die nicht linear mit der Helligkeit wächst und im hellen Zentrum sättigt; das CCD-Bild ist linear. Der Hochpass entfernt den großräumigen Helligkeitsverlauf, und die normierte Kreuzkorrelation zieht Mittelwert und Kontrast ab. Übrig bleibt die Lage der Filamente, und die ist in beiden Bildern vergleichbar.
</details>

<details><summary>4. Warum darf man maskierte Sterne nicht einfach auf null setzen?</summary>

Die Nullen stehen in beiden Bildern an derselben Stelle. Bei der Verschiebung null decken sie sich und erhöhen die Korrelation; das zieht jede gemessene Verschiebung zu kleineren Werten. Bei 20 Jahren Zeitbasis verschiebt das das Konvergenzjahr von 1113 auf 1037 (Variation 1). Richtig ist, für jede Verschiebung nur die Pixel zu nehmen, die in beiden Bildern gültig sind.
</details>

<details><summary>5. Wie bekommt man Zentrum und Konvergenzjahr aus einem einzigen linearen Fit?</summary>

Bei homologer Expansion ist jede Verschiebung $\Delta\vec p = k(\vec p - \vec c)$. Mit $k\vec c$ als eigener Unbekannter ist das linear in drei Unbekannten ($k$, $kc_\xi$, $kc_\eta$). Aus $k = \Delta t/(t_2 - t_0)$ folgt $t_0 = t_2 - \Delta t/k$; mit $k = 0.0657$ und $\Delta t = 60$ Jahren ist das 1098.
</details>

<details><summary>6. Warum gibt die Methode der kleinsten Streuung ein späteres Jahr als der Fit?</summary>

Rechnet man jeden Punkt mit seiner fehlerbehafteten Geschwindigkeit zurück, wächst die Streuung durch die Fehler mit dem Quadrat der Zeit, über die man zurückrechnet. Das Minimum der Gesamtstreuung rutscht deshalb zur Gegenwart hin. Hier sind es 10 Jahre (1109 statt 1098), und die Simulation mit dem gemessenen Rauschen gibt genau denselben Wert.
</details>

<details><summary>7. Warum ist eine lange Zeitbasis so wichtig?</summary>

Ein systematischer Fehler $\delta$ in den Verschiebungen verschiebt das Konvergenzjahr um $(t_2 - t_0)^2/\Delta t \cdot \delta/r$. Bei 0.1″ in 120″ Abstand sind das 35 Jahre für 20 Jahre Zeitbasis, aber nur 12 Jahre für 60 Jahre. Deshalb gibt das Paar 1991 → 2011 1113 statt 1098, und deshalb sind die alten Platten so wertvoll.
</details>

<details><summary>8. Was bedeutet β = 1.05?</summary>

Der Radius wächst wie $(t - 1054)^{1.05}$; die heutige Geschwindigkeit ist 5 % größer als die mittlere seit der Explosion. Das scheinbare Alter ist $(t - 1054)/\beta$, also verlegt schon diese kleine Beschleunigung das Konvergenzjahr um gut 40 Jahre nach 1054.
</details>

<details><summary>9. Wie funktioniert die Expansionsparallaxe, und welche Annahme steckt darin?</summary>

Dieselbe Bewegung wird zweimal gemessen: am Himmel als Winkelgeschwindigkeit, im Spektrum als Radialgeschwindigkeit. Wenn der Nebel entlang der Sichtlinie so tief ist wie entlang seiner kurzen Achse, bewegen sich die schnellsten Filamente vorn und hinten (±1500 km/s) so schnell wie die am Rand der kurzen Achse am Himmel (158 mas/Jahr); $d = v/(4.74\,\mu) = 2000$ pc. Die Annahme über die Form und die Lage des Randes (Variation 4) bestimmen das Ergebnis mit.
</details>

<details><summary>10. Woher weiß man, wie viel Leistung der Pulsar abgibt, ohne sie direkt zu messen?</summary>

Aus seiner Rotation: Die Rotationsenergie ist $2\pi^2 I\nu^2$; nimmt $\nu$ ab, sinkt sie um $\dot E = 4\pi^2 I \nu |\dot\nu| = 4.3\cdot10^{38}$ erg/s. Diese Energie muss irgendwo hin; im Krebsnebel geht etwa ein Drittel als Synchrotronstrahlung ab.
</details>

<details><summary>11. Was sagt ein Bremsindex von 2.5 statt 3?</summary>

Ein rotierender Dipolmagnet im Vakuum würde mit $n = 3$ bremsen. Ein kleinerer Index bedeutet, dass zusätzlich etwas anderes Drehimpuls wegträgt oder das Magnetfeld sich ändert, etwa ein Teilchenwind oder ein wachsender Winkel zwischen Magnet- und Drehachse. Über alle 44 Jahre mit Glitches sieht man sogar nur 2.25, weil nach jedem großen Glitch ein Teil der stärkeren Bremsung bleibt.
</details>

<details><summary>12. Warum ist das charakteristische Alter (1279 Jahre) größer als das wahre (972 Jahre)?</summary>

$\tau_c = P/(2\dot P)$ nimmt an, dass $n = 3$ ist und der Pulsar viel schneller geboren wurde, als er heute ist. Beides stimmt nicht: Mit $n = 2.5$ ist die Zeitskala $P/((n-1)\dot P) = 1706$ Jahre, und die Anfangsperiode von 19 ms ist nur 1.75-mal kürzer als die heutige. Die eckige Klammer $1 - (P_0/P)^{n-1}$ verkürzt das Alter auf die bekannten 972 Jahre.
</details>

<details><summary>13. Reicht die Energie des Pulsars, um die Filamente zu beschleunigen?</summary>

Ja. Die Beschleunigung von 5 % hat etwa $E_\text{kin}(1 - 1/\beta^2) \approx 10^{49}$ erg gekostet; der Pulsar hat seit 1054 rund $3.6\cdot10^{49}$ erg abgegeben. Etwa ein Viertel davon ging in die Bewegung der Filamente, der Rest in Strahlung und in die Energie der Teilchen und des Magnetfelds in der Blase.
</details>

<details><summary>14. Warum liegt das Expansionszentrum 10″ östlich des heutigen Pulsars?</summary>

Der Pulsar bewegt sich mit 11.7 mas/Jahr nach Westnordwest. In 962 Jahren seit 1054 sind das 11″. Rechnet man seine Gaia-Eigenbewegung zurück, stand er 1054 nur 1.0″ vom gemessenen Zentrum entfernt, genau dort, wo die Explosion stattfand.
</details>

#### **Nächste Schritte**

So würde eine Analyse in der Forschung weitergehen:
- **Mehr Epochen, schärfere Bilder:** Martin et al. (2025) haben mit CFHT-Bildern von 2007 bis 2019 fast 20 000 Verschiebungen gemessen und die Beschleunigung als Karte gezeigt: Am wenigsten beschleunigt sind die Filamente in den östlichen und westlichen Buchten, vermutlich wegen einer Scheibe aus Gas um den Vorgängerstern. Hubble-Bilder seit 1994 erlauben dasselbe mit 0.1″ Auflösung.
- **Drei Dimensionen:** Radialgeschwindigkeiten aus Integralfeld-Spektroskopie (SITELLE am CFHT, MUSE am VLT) zusammen mit den Eigenbewegungen ergeben 3D-Geschwindigkeiten jedes Filaments; damit braucht die Expansionsparallaxe keine Annahme über die Form mehr.
- **Den Synchrotronnebel getrennt messen:** In Radiobildern wächst der Synchrotronnebel schneller als die Filamente (Konvergenzjahr um 1255, β ≈ 1.26; Bietenholz & Nugent 2015). Der Pulsarwindnebel holt die Hülle also ein; wann und wie, ist eine Frage an Simulationen.
- **Simulationen von Pulsarwind und Hülle:** zwei- und dreidimensionale Rechnungen der Rayleigh-Taylor-Finger, die zeigen, wie viel Energie in Beschleunigung, Strahlung und Magnetfeld geht.
- **Pulsar-Timing:** Glitch-Statistik, die Entwicklung des Bremsindex und die Frage, warum er zwischen 1997 und 2004 auf 2.2 fiel; dazu die Spannung zwischen den Eigenbewegungen aus Gaia und VLBI (Lin et al. 2023).
- **Andere Überreste:** Cassiopeia A (Expansion im Röntgenlicht mit Chandra), die Überreste der Supernovae von Tycho (1572) und Kepler (1604); dort bremst das umgebende Gas die Hülle, β < 1.
