- die Extinktion der Haufensterne aus ihren Farben,

Was sieht man vom Orionnebel in jeder Wellenlänge, und wie viel lässt sich daraus über den einen Stern lernen, der ihn zum Leuchten bringt?

#### **Hintergrund**

Der Orionnebel ist das nächste große Gebiet, in dem gerade massereiche Sterne entstehen, und mit bloßem Auge als verwaschener Fleck im Schwert des Orion zu sehen. In seiner Mitte steht das **Trapez**, vier helle Sterne, von denen einer, θ¹ Orionis C, ein O-Stern mit rund 30 Sonnenmassen ist. Sein ultraviolettes Licht ionisiert den Wasserstoff einer dichten Molekülwolke (OMC-1). Wir sehen auf die Oberfläche dieser Wolke, auf die eine dünne, leuchtende Schicht ionisierten Gases geklebt ist, eine sogenannte **Blase** oder Blister. Hinter dem Nebel, in der Wolke verborgen, entstehen weitere Sterne; im Infraroten leuchtet dort das Kleinmann-Low-Gebiet mit dem Becklin-Neugebauer-Objekt.

Jede Wellenlänge zeigt einen anderen Teil des Geschehens:
- **Röntgen:** junge, aktive Sterne und heißes Gas.
- **Optisch und Hα:** das ionisierte Gas und der Staub, der es teilweise verdeckt.
- **Nahes Infrarot:** die Sterne des Haufens, auch durch den Staub hindurch.
- **Mittleres Infrarot:** warmer Staub und große Moleküle (PAH) am Rand des ionisierten Gebiets.
- **Fernes Infrarot und Submillimeter:** kalter Staub in der Molekülwolke.
- **Radio:** die Frei-frei-Strahlung des ionisierten Gases, ungestört vom Staub.

Was Forschende damit tun:
- Sie messen in jedem Band, wie viel Energie herauskommt.
- Sie bilanzieren, ob der zentrale Stern sie liefern kann.
- Sie bestimmen aus Radio und Hα die Extinktion, aus der Infrarotstrahlung Temperatur und Masse des Staubs.
- Sie zählen die jungen Sterne mit Scheiben. Der Orionnebel ist das wichtigste Labor dafür, wie massereiche Sterne ihre Umgebung formen.

#### **Theorie**

##### *1. Ionisiertes Gas und seine Strahlung*

Ein O-Stern sendet Photonen mit mehr als 13.6 eV aus, die Wasserstoff ionisieren; ihre Zahl pro Sekunde heißt $Q$. Im Gleichgewicht fangen Protonen genau so viele Elektronen wieder ein, wie ionisiert werden. Für ein Gebiet mit Dichte $n$ und Radius $R$ gilt

$$Q = \tfrac{4}{3}\pi R^3\, n^2\, \alpha_B, \qquad R_S = \left(\frac{3Q}{4\pi n^2 \alpha_B}\right)^{1/3},$$

mit dem Rekombinationskoeffizienten $\alpha_B = 2.59\cdot10^{-13}$ cm³/s bei 10⁴ K (Fall B: Rekombinationen direkt in den Grundzustand zählen nicht, weil ihre Photonen sofort wieder ionisieren). $R_S$ ist der **Strömgren-Radius**: Bis dorthin reichen die Photonen des Sterns. Bei 45 % der Rekombinationen entsteht unterwegs ein Hα-Photon, $\alpha_{\text{H}\alpha}^\text{eff}/\alpha_B = 1.17/2.59$.

Die freien Elektronen strahlen beim Vorbeiflug an Protonen auch im Radiobereich (**Frei-frei-Strahlung**). Ist das Gas dünn genug, fällt ihr Fluss nur schwach mit der Frequenz, $S_\nu \propto \nu^{-0.1}$, und ist proportional zur Zahl der Rekombinationen. Nach Condon (1992) folgt aus der Leuchtkraft $L_\nu = 4\pi d^2 S_\nu$

$$Q \ge 6.3\cdot10^{52}\ \text{s}^{-1}\left(\frac{T_e}{10^4\ \text{K}}\right)^{-0.45}\left(\frac{\nu}{\text{GHz}}\right)^{0.1}\frac{L_\nu}{10^{27}\ \text{erg s}^{-1}\text{Hz}^{-1}} .$$

„≥“, weil Photonen, die Staub verschluckt, keinen Wasserstoff mehr ionisieren. Anders als Hα geht Radiostrahlung ungehindert durch Staub.

##### *2. Extinktion*

Aus $Q$ folgt auch der Hα-Fluss, den man ohne Staub sähe, $F_{\text{H}\alpha} = Q\,(\alpha_{\text{H}\alpha}^\text{eff}/\alpha_B)\,h\nu_{\text{H}\alpha}/(4\pi d^2)$. Der Vergleich mit dem gemessenen Fluss gäbe die Extinktion vor dem Gas. Das scheitert hier an der Übersättigung der Hα-Karte (Schritt 6). Die Extinktion vor den Sternen folgt stattdessen aus ihren Farben: Staub schluckt bei J mehr als bei H, und mit $A_J : A_H = 0.282 : 0.175$ in Einheiten von $A_V$ (Rieke & Lebofsky 1985) wird aus der Rötung $E(J-H)$ die Extinktion $A_V = 9.35\,E(J-H)$.

##### *3. Emissionsmaß und Dichte*

Das **Emissionsmaß** $EM = \int n_e^2\,dl$ längs der Sichtlinie bestimmt, wie hell das Gas in Hα leuchtet und wie optisch dick es im Radio ist: $\tau_\nu = 3.28\cdot10^{-7}\,(T_e/10^4\,\text{K})^{-1.35}\,(\nu/\text{GHz})^{-2.1}\,EM/(\text{pc cm}^{-6})$ (Condon & Ransom 2016). Bei niedrigen Frequenzen wird ein dichter Kern optisch dick, bei 30 GHz kaum. Teilt man $Q$ auf eine Kugel vom Radius des Nebels auf, erhält man die **mittlere quadratische Dichte** $n_\text{rms} = \sqrt{3Q/(4\pi R^3\alpha_B)}$. Ist das Gas klumpig, ist die Dichte in den Klumpen $n_\text{lok}$ höher, und der **Füllfaktor** $\epsilon = (n_\text{rms}/n_\text{lok})^2$ sagt, welcher Teil des Volumens leuchtet.

##### *4. Staub*

Staub verschluckt das Licht der Sterne und strahlt es im Infraroten wieder ab, als modifizierter Schwarzkörper $S_\nu \propto \nu^\beta B_\nu(T)$ (siehe [Die Dunkelwolke ρ Ophiuchi](https://github.com/deltorobarba/science/blob/main/astro/nebulae_dust.ipynb), Theorie, Abschnitt 4). Das Integral über alle Frequenzen ist die **Infrarot-Leuchtkraft** $L_\text{IR} = 4\pi d^2 \int S_\nu\,d\nu$. Wird fast alles Sternlicht vom Staub verschluckt, ist $L_\text{IR}$ ungefähr die Leuchtkraft der heizenden Sterne. Aus dem Fluss bei 353 GHz folgt die Masse von Gas und Staub, $M = S_\nu d^2/[\kappa_\nu B_\nu(T)]$, mit der Opazität $\kappa_\nu = 0.1\ \text{cm}^2/\text{g}\,(\nu/1\ \text{THz})^\beta$ pro Gramm Gas (Beckwith et al. 1990, Gas zu Staub 100:1).

##### *5. Junge Sterne mit Scheiben*

Ein junger Stern ist oft noch von einer Scheibe aus Gas und Staub umgeben. Ihr innerer, heißer Rand (rund 1500 K) strahlt im nahen Infrarot zusätzlich zum Stern. Im Diagramm $J-H$ gegen $H-K$ liegen gewöhnliche Sterne auf einem schmalen Streifen, den Staub entlang des **Rötungsvektors** verschiebt ($E(J-H)/E(H-K) = 1.70$, Rieke & Lebofsky 1985). Sterne mit Scheibe liegen rechts davon: zu viel $K$ für ihre Rötung.

**Fachbegriffe:**
- **O-Stern:** Stern mit über 30 000 K Oberflächentemperatur und mehr als etwa 15 Sonnenmassen; Spektraltypen O3 (heißest) bis O9.5.
- **Blister-H-II-Gebiet:** ein ionisiertes Gebiet am Rand einer Molekülwolke, das wie eine Blase nach außen offen ist; das Gas strömt zu uns hin ab.
- **PAH:** polyzyklische aromatische Kohlenwasserstoffe, große Moleküle, die ultraviolettes Licht in Banden bei 3.3, 6.2, 7.7, 8.6 und 11.3 µm umsetzen; sie leuchten an der Grenze zwischen ionisiertem und molekularem Gas.

```python
import io
import os
import time
import warnings
import urllib.parse
import urllib.request
from pathlib import Path

ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "astropy").mkdir(parents=True, exist_ok=True)
os.environ["XDG_CACHE_HOME"] = str(ASTRODATA)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy.io import fits
from scipy import ndimage, optimize

warnings.filterwarnings("ignore", message="Invalid 'BLANK' keyword")   # SkyView schreibt BLANK auch in Gleitkomma-Bilder

RA0, DE0 = 83.8186, -5.3897                                   # θ¹ Ori C (SIMBAD)
ORION = ASTRODATA / "orion"
SKYVIEW = "https://skyview.gsfc.nasa.gov/current/cgi/runquery.pl?"
IRSA_TAP = "https://irsa.ipac.caltech.edu/TAP/sync"
H_PL, K_B, C_SI = 6.62607e-34, 1.380649e-23, 2.99792458e8
PC_CM, LSUN, MSUN, M_H = 3.0857e18, 3.828e33, 1.989e33, 1.6726e-24
ALPHA_B, ALPHA_HA = 2.59e-13, 1.17e-13                         # cm³/s bei 10⁴ K, Fall B
T_E = 1.0e4
WIDE, N_WIDE = 3.0, 360                                        # Übersicht: 3° mit 360 Pixeln (30″)
CLOSE, N_CLOSE = 1.0, 800                                      # Farbbilder: 1° mit 800 Pixeln (4.5″)
APER, BG = 40.0, (60.0, 85.0)                                  # Öffnung und Hintergrundring [′]
# Band: (SkyView-Name, Frequenz [Hz], Einheit, Beschriftung)
BANDS = {"ROSAT": ("RASS-Cnt Broad", None, "Zählrate", "ROSAT 0.1–2.4 keV"),
         "DSS2 R": ("DSS2 Red", None, "Zählwert", "DSS2 rot (0.65 µm)"),
         "Hα": ("SHASSA H", None, "dR", "SHASSA Hα (656 nm)"),
         "2MASS K": ("2MASS-K", None, "Zählwert", "2MASS K (2.2 µm)"),
         "WISE 3.4": ("WISE 3.4", None, "Zählwert", "WISE 3.4 µm"),
         "WISE 22": ("WISE 22", None, "Zählwert", "WISE 22 µm"),
         "IRIS 12": ("IRIS12", 25e12, "MJy/sr", "IRIS 12 µm"),
         "IRIS 25": ("IRIS25", 12e12, "MJy/sr", "IRIS 25 µm"),
         "IRIS 60": ("IRIS60", 5e12, "MJy/sr", "IRIS 60 µm"),
         "IRIS 100": ("IRIS100", 3e12, "MJy/sr", "IRIS 100 µm"),
         "Planck 857": ("Planck 857 I", 857e9, "MJy/sr", "Planck 857 GHz (350 µm)"),
         "Planck 545": ("Planck 545 I", 545e9, "MJy/sr", "Planck 545 GHz (550 µm)"),
         "Planck 353": ("Planck 353 I", 353e9, "K_CMB", "Planck 353 GHz (850 µm)"),
         "Planck 217": ("Planck 217 I", 217e9, "K_CMB", "Planck 217 GHz (1.4 mm)"),
         "Planck 70": ("Planck 070 I", 70.4e9, "K_CMB", "Planck 70 GHz (4.3 mm)"),
         "Planck 44": ("Planck 044 I", 44.1e9, "K_CMB", "Planck 44 GHz (6.8 mm)"),
         "Planck 28": ("Planck 030 I", 28.4e9, "K_CMB", "Planck 28 GHz (1 cm)"),
         "Bonn 1.4": ("1420MHz (Bonn)", 1.42e9, "mK", "Bonn 1.4 GHz (21 cm)"),
         "NVSS": ("NVSS", None, "Jy/beam", "NVSS 1.4 GHz (21 cm)")}
COLOR = {"DSS2": ["DSS2 Red", "DSS2 Mitte", "DSS2 Blue"], "2MASS": ["2MASS-K", "2MASS-H", "2MASS-J"], "WISE": ["WISE 12", "WISE 4.6", "WISE 3.4"]}
DIST_REF = {"Menten et al. 2007 (VLBA)": (414.0, 7.0), "Kounkel et al. 2017 (VLBA)": (388.0, 5.0)}
# Martins et al. 2005, Tabelle 4 (Zwergsterne, beobachtete Temperaturskala): Typ, T_eff, log L/L☉, log Q₀
MARTINS_V = pd.DataFrame([("O3", 44852, 5.84, 49.64), ("O4", 42857, 5.67, 49.44), ("O5", 40862, 5.49, 49.22), ("O5.5", 39865, 5.41, 49.10),
                          ("O6", 38867, 5.32, 48.99), ("O6.5", 37870, 5.23, 48.88), ("O7", 36872, 5.14, 48.75), ("O7.5", 35874, 5.05, 48.61),
                          ("O8", 34877, 4.96, 48.44), ("O8.5", 33879, 4.86, 48.27), ("O9", 32882, 4.77, 48.06), ("O9.5", 31884, 4.68, 47.88)],
                         columns=["typ", "teff", "logL", "logQ"])
DISK_HILLENBRAND = (0.55, 0.90)                               # Anteil mit inneren Scheiben (I−K-Überschuss)


def fetch(url, path, check, tries=5, pause=15, timeout=600):
    path = Path(path)
    if path.exists():
        return path
    path.parent.mkdir(parents=True, exist_ok=True)
    for _ in range(tries):
        try:
            with urllib.request.urlopen(url, timeout=timeout) as r:
                data = r.read()
            if check(data):
                path.write_bytes(data)
                return path
        except OSError:
            pass
        time.sleep(pause)                                     # SkyView lehnt schnelle Folgeanfragen ab („Too Many Processes“)
    raise OSError(f"{path.name}: kein gültiger Download nach {tries} Versuchen")


def skyview(survey, size, npix):
    q = urllib.parse.urlencode({"Position": f"{RA0},{DE0}", "Survey": survey, "Pixels": npix, "Size": size, "Return": "FITS"})
    name = survey.replace(" ", "_").replace("(", "").replace(")", "") + f"_{size:g}_{npix}.fits"
    with fits.open(fetch(SKYVIEW + q, ORION / name, lambda d: d.startswith(b"SIMPLE"))) as h:
        return h[0].data.astype(float)


def tap(adql, path):
    url = IRSA_TAP + "?" + urllib.parse.urlencode({"REQUEST": "doQuery", "LANG": "ADQL", "FORMAT": "csv", "QUERY": adql})
    first = adql.split()[1].rstrip(",").encode()
    return pd.read_csv(fetch(url, path, lambda d: d.startswith(first)))


WIDEMAP = {k: skyview(v[0], WIDE, N_WIDE) for k, v in BANDS.items()}
CLOSEMAP = {s: skyview(s, CLOSE, N_CLOSE) for s in sorted({x for v in COLOR.values() for x in v} - {"DSS2 Mitte"})}
CLOSEMAP["DSS2 Mitte"] = (CLOSEMAP["DSS2 Red"] + CLOSEMAP["DSS2 Blue"]) / 2   # künstlicher Grünkanal aus Rot und Blau
GAIA = tap("SELECT source_id, ra, dec, parallax, parallax_error, ruwe, phot_g_mean_mag FROM gaia_dr3_source "
           f"WHERE CONTAINS(POINT('ICRS', ra, dec), CIRCLE('ICRS', {RA0}, {DE0}, 0.25)) = 1 AND parallax_over_error > 10 AND ruwe < 1.4",
           ORION / "gaia_onc.csv")
COLS = "ra, dec, j_m, h_m, k_m, j_msigcom, h_msigcom, k_msigcom, ph_qual"
TM_ONC = tap(f"SELECT {COLS} FROM fp_psc WHERE CONTAINS(POINT('ICRS', ra, dec), CIRCLE('ICRS', {RA0}, {DE0}, 0.17)) = 1", ORION / "2mass_onc.csv")
TM_CTRL = tap(f"SELECT {COLS} FROM fp_psc WHERE CONTAINS(POINT('ICRS', ra, dec), CIRCLE('ICRS', {RA0 + 6}, {DE0}, 0.17)) = 1", ORION / "2mass_ctrl.csv")
print(f"{len(WIDEMAP)} Übersichtskarten (3°), {len(CLOSEMAP)} Karten für Farbbilder (1°); Gaia: {len(GAIA)} Sterne bis 15′; "
      f"2MASS: {len(TM_ONC)} Quellen bis 10′ um das Trapez, {len(TM_CTRL)} im Vergleichsfeld")
```

**Werkzeuge: Einheiten, Photometrie, Farbbilder und Staub** (Theorie, Abschnitte 1 bis 4).
- `to_jy_sr` rechnet jede geeichte Karte in Jy/sr um.
- `photometry` summiert den Fluss in einer Öffnung um das Trapez und zieht den Median eines Rings als Hintergrund ab; der Fehler kommt aus der Streuung im Ring.
- `rgb` setzt drei Karten mit einer Asinh-Streckung zu einem Farbbild zusammen.
- `planck` und `mbb` sind Schwarzkörper und modifizierter Schwarzkörper.

```python
PIX_WIDE = np.radians(WIDE / N_WIDE)                            # Pixel in rad
yy, xx = np.mgrid[:N_WIDE, :N_WIDE]
R_ARCMIN = np.hypot(xx - N_WIDE / 2 + 0.5, yy - N_WIDE / 2 + 0.5) * WIDE / N_WIDE * 60


def planck(nu, T):
    return 2 * H_PL * nu**3 / C_SI**2 / np.expm1(H_PL * nu / (K_B * T))      # W/m²/Hz/sr


def dB_dT_cmb(nu, T=2.7255):
    x = H_PL * nu / (K_B * T)
    return 2 * K_B * nu**2 / C_SI**2 * x**2 * np.exp(x) / np.expm1(x) ** 2 * 1e26   # Jy/sr pro K


def to_jy_sr(key):
    survey, nu, unit, _ = BANDS[key]
    x = WIDEMAP[key]
    if unit == "MJy/sr":
        return x * 1e6
    if unit == "K_CMB":
        return x * dB_dT_cmb(nu)
    if unit == "mK":
        return x * 1e-3 * 2 * K_B * nu**2 / C_SI**2 * 1e26                       # Rayleigh-Jeans
    raise ValueError(f"{key}: nicht geeicht")


def photometry(img, radius=APER, ring=BG):
    # Fluss in der Öffnung minus Median des Rings; Fehler aus der Streuung des Hintergrunds zwischen 8 Sektoren des Rings
    ring_pix = (R_ARCMIN > ring[0]) & (R_ARCMIN < ring[1])
    inside = R_ARCMIN < radius
    flux = np.sum(img[inside] - np.median(img[ring_pix])) * PIX_WIDE**2
    sector = (np.degrees(np.arctan2(yy - N_WIDE / 2 + 0.5, xx - N_WIDE / 2 + 0.5)) % 360 // 45).astype(int)
    levels = [np.median(img[ring_pix & (sector == s)]) for s in range(8)]
    return flux, np.std(levels) * inside.sum() * PIX_WIDE**2


def stretch(x, lo=1, hi=99.7, soft=0.1, gamma=1.0):
    a, b = np.nanpercentile(x, [lo, hi])
    y = np.clip((x - a) / (b - a), 0, 1)
    return (np.arcsinh(y / soft) / np.arcsinh(1 / soft)) ** gamma


def rgb(keys, **kw):
    # übersättigte Pixel (NaN) als hellster Wert
    return np.dstack([np.nan_to_num(stretch(CLOSEMAP[k], **kw), nan=1.0) for k in keys])


def mbb(nu, lnA, T, beta):
    return np.exp(lnA) * (nu / 1e12) ** beta * planck(nu, T) * 1e26                # Jy


D_FROM_GAIA = None                                               # wird in Schritt 3 gesetzt
print(f"Pixel der Übersicht {np.degrees(PIX_WIDE) * 3600:.0f}″, Öffnung {APER:.0f}′, Hintergrund {BG[0]:.0f}–{BG[1]:.0f}′")
```

##### *3. Die Entfernung aus Gaia*

Wie weit ist der Orionnebel-Haufen entfernt, und stimmt Gaia mit den Parallaxen aus der Radio-Interferometrie überein? Vorher vorhersagen: Wie groß ist die Parallaxe in Millibogensekunden, wenn der Haufen etwa 400 pc entfernt ist?

Der Code nimmt alle Sterne aus Gaia DR3 bis 15′ um das Trapez mit Parallaxe auf 10 % genau und RUWE < 1.4. Mitglieder sind die mit einer Parallaxe zwischen 2.0 und 3.5 mas. Auf jede Parallaxe zählt er den Nullpunkt von Gaia (+0.017 mas) und mittelt gewichtet; den Fehler des Mittels gibt er zusammen mit der Streuung der Sterne aus. Im Bild stehen die Parallaxen aller Sterne und die beiden VLBA-Messungen.

**Hintergrund & Fachbegriffe:**
- **RUWE:** Maß dafür, wie gut eine Einzelstern-Lösung passt; über 1.4 sind Doppelsterne oder Störungen wahrscheinlich.
- **Nullpunkt:** Gaia DR3 misst Parallaxen im Mittel um etwa 0.017 mas zu klein (Lindegren et al. 2021); bei 2.5 mas ist das 0.7 % der Entfernung. Er hängt von Helligkeit, Farbe und Ort ab; als Unsicherheit setzt der Code 0.01 mas an (eigene Annahme).
- **VLBA-Parallaxen:** Radio-Interferometrie über Kontinente misst die Parallaxen einzelner, radiolauter junger Sterne auf etwa 1 % genau.

```python
mem = (GAIA.parallax > 2.0) & (GAIA.parallax < 3.5)
plx = GAIA.parallax[mem] + 0.017
w = 1 / GAIA.parallax_error[mem] ** 2
PLX = (np.average(plx, weights=w), 1 / np.sqrt(w.sum()))
PLX_SYS = 0.010                                                # angenommene Unsicherheit des Nullpunkts [mas]
D_FROM_GAIA = 1000 / PLX[0]
D_ERR = D_FROM_GAIA * np.hypot(PLX[1], PLX_SYS) / PLX[0]
fig, ax = plt.subplots(figsize=(11, 5.6))
ax.hist(GAIA.parallax + 0.017, bins=np.linspace(0, 6, 121), color="0.7", label="alle Sterne (15′, Parallaxe auf 10 %)")
ax.hist(plx, bins=np.linspace(0, 6, 121), color="C0", label=f"Mitglieder (2.0–3.5 mas): {mem.sum()}")
ax.axvline(PLX[0], color="k", lw=1.5, label=f"gewichtetes Mittel: {PLX[0]:.3f} mas = {D_FROM_GAIA:.1f} pc")
for (lab, (d, e)), col in zip(DIST_REF.items(), ["C3", "C2"]):
    ax.axvspan(1000 / (d + e), 1000 / (d - e), color=col, alpha=0.4, label=f"{lab}: {d:.0f} ± {e:.0f} pc")
ax.set_xlabel("Parallaxe [mas] (mit Nullpunkt +0.017)")
ax.set_ylabel("Zahl der Sterne")
ax.set_title("Die Parallaxen der Sterne im Orionnebel", fontsize=10)
ax.legend(fontsize=9)
plt.show()

print(f"Mitglieder {mem.sum()} von {len(GAIA)}; Parallaxe {PLX[0]:.4f} ± {PLX[1]:.4f} mas (Streuung der Sterne {plx.std():.3f} mas)")
print(f"Entfernung {D_FROM_GAIA:.1f} ± {D_ERR:.1f} pc (statistisch ± {D_FROM_GAIA * PLX[1] / PLX[0]:.1f}, Nullpunkt ± {D_FROM_GAIA * PLX_SYS / PLX[0]:.1f}; Median der Sterne {1000 / np.median(plx):.1f} pc; ohne Nullpunkt {1000 / np.average(plx - 0.017, weights=w):.1f} pc)")
for lab, (d, e) in DIST_REF.items():
    print(f"  gegen {lab}: {(D_FROM_GAIA - d) / np.hypot(D_ERR, e):+.1f} σ")
```

* Grau die Parallaxen aller Sterne bis 15′, blau die 915 Mitglieder mit 2.0 bis 3.5 mas, schwarz ihr gewichtetes Mittel, dazu die beiden VLBA-Messungen. Die Mitglieder bilden eine schmale Spitze bei 2.57 mas; die Streuung der Sterne (0.18 mas) ist meist ihr Messfehler und etwas Tiefe des Haufens.
* Ergebnis: 2.5671 mas, also d = 389.5 ± 1.5 pc. Der statistische Fehler ist nur 0.2 pc; der größere Teil (1.5 pc) kommt aus der angenommenen Unsicherheit des Nullpunkts von Gaia. Ohne Nullpunktkorrektur wären es 392.1 pc, mit dem Median der Sterne statt des gewichteten Mittels 396.0 pc.
* Zur Vorhersage: 1000/400 = 2.5 mas. Kounkel et al. (388 ± 5 pc) liegen 0.3σ entfernt, Menten et al. (414 ± 7 pc) 3.4σ. Die ältere VLBA-Messung von Menten beruhte auf wenigen Sternen; mit Gaia ist sie überholt.

Wie viel Strahlung kommt in jedem Band aus der Gegend um den Nebel, und ist die Öffnung groß genug? Vorher vorhersagen: Warum muss die Öffnung im Radio größer sein als im fernen Infrarot?

Der Code rechnet alle geeichten Karten in Jy/sr um und misst den Fluss in Öffnungen von 5′ bis 55′ um das Trapez, jeweils mit dem Hintergrund aus dem Ring von 60′ bis 85′. Der Fehler ist grob aus der Streuung im Ring geschätzt; die Eichung der Karten (einige Prozent) ist darin nicht enthalten.

**Hintergrund & Fachbegriffe:**
- **Wachstumskurve:** der Fluss als Funktion des Öffnungsradius; flacht sie ab, ist die ganze Quelle erfasst.
- **Strahl:** Planck bei 28 GHz hat eine Auflösung von 33′; eine Punktquelle verteilt sich darüber, und erst eine Öffnung von gut 40′ fängt sie ganz ein.
- **Farbkorrektur:** Die Bänder sind für ein bestimmtes Spektrum geeicht; für andere Spektren weichen die Flüsse um einige Prozent ab. Das Notebook lässt das weg.

```python
CAL = [k for k, v in BANDS.items() if v[1] is not None]
radii = np.arange(5, 56, 2.5)
FLUX = {}
fig, ax = plt.subplots(figsize=(11, 5.8))
for key in CAL:
    img = to_jy_sr(key)
    grow = np.array([photometry(img, r)[0] for r in radii])
    FLUX[key] = photometry(img)
    if key in ["IRIS 12", "IRIS 100", "Planck 857", "Planck 353", "Planck 28", "Bonn 1.4"]:
        ax.plot(radii, grow / FLUX[key][0], "o-", ms=4, label=BANDS[key][3])
ax.axvline(APER, color="0.5", ls=":")
ax.axhline(1, color="0.5", lw=0.8)
ax.set_xlabel("Radius der Öffnung [′]")
ax.set_ylabel("Fluss / Fluss bei 40′")
ax.set_title("Wachstumskurven: Wann ist die ganze Strahlung erfasst?", fontsize=10)
ax.legend(fontsize=9)
plt.show()

for key in CAL:
    f, e = FLUX[key]
    print(f"{BANDS[key][3]:26s} ν = {BANDS[key][1] / 1e9:8.1f} GHz: S = {f:10.1f} ± {e:7.1f} Jy")
```

* Wachstumskurven für sechs Bänder: der Fluss in einer Öffnung vom gegebenen Radius, geteilt durch den bei 40′. Bei 12 µm stammt die Hälfte des Flusses aus den innersten 5′, bei 28 GHz nur 6 % und bei 1.4 GHz (Bonn) 2 %, weil diese Karten mit 33′ und 34′ den Nebel verschmieren.
* Bei 40′ sind IRIS 12, IRIS 100 und Planck 28 GHz flach (bis 55′ kommen noch 4 bis 6 % hinzu). Planck 857 und 353 GHz wachsen dagegen weiter (15 bis 18 % bis 55′): Dort leuchtet die ausgedehnte Molekülwolke, und die Grenze der Öffnung ist eine Wahl. Auch Bonn wächst weiter (17 %), ein weiterer Grund, dieser Karte nicht zu trauen. Die Flüsse bei 40′ reichen von 357 Jy bei 28 GHz bis 568 743 Jy bei 100 µm; die Fehler aus der Streuung des Hintergrunds sind 1 bis 13 %.
* Zur Vorhersage: weil die Auflösung im Radio schlechter ist (33′ bei 28 GHz gegen 2′ bei IRIS). Eine kleine Öffnung schneidet dort einen Teil des verschmierten Lichts ab.

##### *5. Frei-frei-Strahlung und der Stern, der den Nebel ionisiert*

Wie viele ionisierende Photonen braucht der Nebel pro Sekunde, und welcher Stern kann sie liefern? Vorher vorhersagen: Fällt der Radiofluss zwischen 28 und 70 GHz, steigt er, oder bleibt er gleich?

Der Code passt an die Flüsse bei 28, 44 und 70 GHz ein Potenzgesetz $S_\nu \propto \nu^\alpha$ an (Bonn bei 1.4 GHz steht zum Vergleich im Bild). Aus dem Fluss bei 28 GHz und der Entfernung aus Schritt 3 rechnet er mit der Formel von Condon (Theorie, Abschnitt 1) die Zahl Q der ionisierenden Photonen pro Sekunde. Rechts vergleicht er Q mit der Kalibration von O-Sternen nach Martins et al. (2005).

**Hintergrund & Fachbegriffe:**
- **Spektralindex α:** die Steigung im doppelt-logarithmischen Spektrum; −0.1 für dünne Frei-frei-Strahlung, +2 für optisch dicke, −0.7 für Synchrotronstrahlung, um +3 bis +4 für Staub.
- **θ¹ Ori C:** der hellste Stern des Trapezes, Spektraltyp O7 V (mit Begleiter); die anderen Trapezsterne sind späte O- und frühe B-Sterne mit viel weniger ionisierendem Licht.

```python
ff_keys = ["Planck 28", "Planck 44", "Planck 70"]
nu_ff = np.array([BANDS[k][1] for k in ff_keys])
S_ff = np.array([FLUX[k][0] for k in ff_keys])
E_ff = np.array([FLUX[k][1] for k in ff_keys])
ALPHA_FF, lnS0 = np.polyfit(np.log(nu_ff / 28.4e9), np.log(S_ff), 1, w=S_ff / E_ff)
d_cm = D_FROM_GAIA * PC_CM
L28 = 4 * np.pi * d_cm**2 * S_ff[0] * 1e-23
Q_ION = 6.3e52 * (T_E / 1e4) ** -0.45 * (28.4) ** 0.1 * L28 / 1e27
Q_ERR = Q_ION * np.hypot(E_ff[0] / S_ff[0], 2 * D_ERR / D_FROM_GAIA)

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 5.6))
nn = np.logspace(9, 11.2, 50)
ax.errorbar(nu_ff / 1e9, S_ff, yerr=E_ff, fmt="o", color="C0", ms=8, capsize=4, label="Planck (Öffnung 40′)")
ax.plot(nn / 1e9, np.exp(lnS0) * (nn / 28.4e9) ** ALPHA_FF, color="C0", lw=1.2, label=f"Potenzgesetz, α = {ALPHA_FF:+.2f}")
ax.plot(nn / 1e9, S_ff[0] * (nn / 28.4e9) ** -0.1, color="0.5", ls=":", label="dünne Frei-frei-Strahlung, α = −0.1")
ax.errorbar(1.42, FLUX["Bonn 1.4"][0], yerr=FLUX["Bonn 1.4"][1], fmt="s", color="C3", ms=8, capsize=4, label="Bonn 1.4 GHz (nicht im Fit)")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_ylim(50, 1000)
ax.set_xlabel("Frequenz [GHz]")
ax.set_ylabel("Flussdichte [Jy]")
ax.set_title("Das Radiospektrum des Orionnebels", fontsize=10)
ax.legend(fontsize=9)
ax2.plot(MARTINS_V.logQ, range(len(MARTINS_V)), "o-", color="0.4", label="O-Zwergsterne (Martins et al. 2005)")
ax2.set_yticks(range(len(MARTINS_V)))
ax2.set_yticklabels(MARTINS_V.typ + " V")
ax2.axvspan(np.log10(Q_ION - Q_ERR), np.log10(Q_ION + Q_ERR), color="C0", alpha=0.3, label=f"Orionnebel: log Q = {np.log10(Q_ION):.2f}")
ax2.set_xlabel(r"log $Q$ [Photonen/s]")
ax2.set_title("Welcher Stern liefert so viele ionisierende Photonen?", fontsize=10)
ax2.legend(fontsize=9)
plt.tight_layout()
plt.show()

typ = MARTINS_V.typ[np.argmin(np.abs(MARTINS_V.logQ - np.log10(Q_ION)))]
print(f"Spektralindex 28–70 GHz: α = {ALPHA_FF:+.3f}; S(28 GHz) = {S_ff[0]:.1f} ± {E_ff[0]:.1f} Jy, Bonn 1.4 GHz {FLUX['Bonn 1.4'][0]:.1f} Jy "
      f"(erwartet mit α = −0.1: {S_ff[0] * (1.42 / 28.4) ** -0.1:.0f} Jy)")
print(f"L_ν(28 GHz) = {L28:.2e} erg/s/Hz, Q = {Q_ION:.2e} ± {Q_ERR:.1e} s⁻¹ (log Q = {np.log10(Q_ION):.2f}); passt zu {typ} V "
      f"(log Q = {MARTINS_V.logQ[MARTINS_V.typ == typ].values[0]})")
```

* Links das Radiospektrum: Die drei Planck-Flüsse bei 28, 44 und 70 GHz liegen bei 357, 350 und 358 Jy, das Potenzgesetz hat α = −0.006 und ist damit fast flach, wie für Frei-frei-Strahlung (−0.1) mit etwas Staub bei 70 GHz. Bonn bei 1.4 GHz liegt mit 141 Jy weit unter den 482 Jy, die man mit α = −0.1 erwartet. Ein Teil kommt daher, dass der Kern bei 1.4 GHz halb optisch dick ist (Schritt 7). Der größere Teil liegt nach meiner Vermutung an der Eichung und Auflösung dieser Karte; die Wachstumskurve (Schritt 4) zeigt, wie stark sie den Nebel verschmiert. Bonn geht deshalb in keine Rechnung ein.
* Rechts: Aus 357 Jy bei 28 GHz und 389.5 pc folgt Q = 5.7·10⁴⁸ ionisierende Photonen pro Sekunde, log Q = 48.76. Ein O7-V-Stern liefert nach Martins et al. (2005) log Q = 48.75, ein O6 V schon 48.99, ein O8 V nur 48.44.
* Zur Vorhersage: fast gleich bleibend. Der Nebel braucht also genau so viel ionisierendes Licht, wie θ¹ Ori C als O7-V-Stern abgibt; die anderen Trapezsterne tragen wenig bei. Dass die Zahl so genau passt, ist auch Glück: Q ist ein unterer Wert (Staub schluckt Photonen), und die Temperatur des Gases ist hier mit 10⁴ K angenommen.

##### *6. Wie viel Staub vor den Sternen liegt*

Wie stark röten Staub und Gas das Licht der Sterne im Haufen, und wie verschieden ist das von Stern zu Stern? Vorher vorhersagen: Liegt die mittlere Extinktion der Haufensterne über oder unter der des Vergleichsfelds, und um wie viel?

Hα lässt sich hier nicht verwenden: SHASSA ist im Kern des Orionnebels übersättigt, alle Pixel bis etwa 18′ um das Trapez stehen auf demselben Wert um 9500 Rayleigh. Der Code nimmt deshalb die Sterne selbst. Für alle Sterne aus 2MASS ohne Infrarot-Überschuss (Kriterium aus Schritt 10) rechnet er aus $J-H$ die Extinktion

$$A_V = \frac{(J-H) - (J-H)_0}{0.282 - 0.175} = 9.35\,\left[(J-H) - (J-H)_0\right],$$

mit der Eigenfarbe $(J-H)_0$ als Median des Vergleichsfelds (Rieke & Lebofsky 1985). Links stehen die Sterne am Himmel, gefärbt nach $A_V$, rechts die Verteilung im Haufen und im Vergleichsfeld.

**Hintergrund & Fachbegriffe:**
- **Übersättigung:** Ein Detektor kann nur bis zu einer Höchstzahl zählen; heller wird es nicht. Eine Fläche mit fast konstantem Wert mitten in der hellsten Gegend ist das typische Zeichen dafür.
- **Warum J − H:** Die Scheibe junger Sterne leuchtet vor allem in K; $J-H$ ist davon weniger betroffen als $H-K$.
- **Vor, in und hinter der Wolke:** Die Sterne des Haufens stehen teils vor, teils in der Molekülwolke OMC-1; deshalb streut ihre Extinktion stark.
- **OMC-1 bis OMC-3:** dichte Kerne der Molekülwolke Orion A; OMC-1 liegt direkt hinter dem Nebel, OMC-2 und OMC-3 etwa 12′ und 20′ nördlich des Trapezes.

```python
def colors(df):
    df = df[df.ph_qual.astype(str).str[:3] == "AAA"]
    jh, hk = df.j_m - df.h_m, df.h_m - df.k_m
    dist = (1.70 * (hk - 0.30) - (jh - 0.61)) / np.hypot(1.70, 1)
    sig = np.hypot(1.70 * np.hypot(df.h_msigcom, df.k_msigcom), np.hypot(df.j_msigcom, df.h_msigcom)) / np.hypot(1.70, 1)
    return df, jh, hk, dist > np.maximum(sig, 0.03)


onc, jh_o, hk_o, ex_o = colors(TM_ONC)
ctl, jh_c, hk_c, ex_c = colors(TM_CTRL)
JH0 = np.median(jh_c[~ex_c])
AV_ONC = 9.35 * (jh_o[~ex_o] - JH0)
AV_CTL = 9.35 * (jh_c[~ex_c] - JH0)
sat_area = np.sum(WIDEMAP["Hα"] > 9.0e4) * (WIDE / N_WIDE * 60) ** 2
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 6))
sc = ax.scatter(-(onc.ra[~ex_o] - RA0) * np.cos(np.radians(DE0)) * 60, (onc.dec[~ex_o] - DE0) * 60, c=AV_ONC, s=10, cmap="viridis", vmin=0, vmax=10)
plt.colorbar(sc, ax=ax, label=r"$A_V$ [mag]")
ax.invert_xaxis()
ax.set_aspect("equal")
ax.set_xlabel("Rektaszension relativ [′] (Osten links)")
ax.set_ylabel("Deklination relativ [′]")
ax.set_title("Extinktion der Haufensterne ohne Überschuss (2MASS J − H)", fontsize=10)
bins = np.linspace(-3, 20, 47)
ax2.hist(AV_CTL, bins=bins, density=True, histtype="step", lw=2, color="0.5", label=f"Vergleichsfeld ({len(AV_CTL)})")
ax2.hist(AV_ONC, bins=bins, density=True, histtype="step", lw=2, color="C3", label=f"Orionnebel-Haufen ({len(AV_ONC)})")
ax2.set_xlabel(r"$A_V$ [mag]")
ax2.set_ylabel("Anteil pro mag")
ax2.set_title("Wie viel Staub vor den Sternen liegt", fontsize=10)
ax2.legend(fontsize=9)
plt.tight_layout()
plt.show()

print(f"SHASSA übersättigt (über 9·10⁴ dR) auf {sat_area:.0f} Quadratbogenminuten")
print(f"Eigenfarbe (J − H)₀ = {JH0:.3f} (Median im Vergleichsfeld)")
print(f"A_V Haufen: Median {np.median(AV_ONC):.2f} mag, 16. bis 84. Perzentil {np.percentile(AV_ONC, 16):.2f} bis {np.percentile(AV_ONC, 84):.2f}, "
      f"Anteil über 5 mag {np.mean(AV_ONC > 5):.2f}; Vergleichsfeld: Median {np.median(AV_CTL):.2f}, Streuung {np.std(AV_CTL):.2f} mag")
print(f"Sterne mit A_V über 5 mag liegen im Median {np.median(np.hypot((onc.ra[~ex_o][AV_ONC > 5] - RA0) * np.cos(np.radians(DE0)), onc.dec[~ex_o][AV_ONC > 5] - DE0)) * 60:.1f}′ "
      f"vom Trapez, die übrigen {np.median(np.hypot((onc.ra[~ex_o][AV_ONC <= 5] - RA0) * np.cos(np.radians(DE0)), onc.dec[~ex_o][AV_ONC <= 5] - DE0)) * 60:.1f}′")
```

##### *7. Dichte, Größe und die Strömgren-Kugel*

Wie dicht ist das ionisierte Gas im Kern, wie groß ist das ionisierte Gebiet, und passt das zu dem, was ein O7-Stern ionisieren kann? Vorher vorhersagen: Ist die Dichte im hellen Kern größer oder kleiner als die mittlere Dichte, die Q und die Größe verlangen?

Der Code nimmt zwei Wege:
- **Kern aus NVSS:** Die hellste Stelle in der NVSS-Karte (Strahl 45″) gibt eine Helligkeitstemperatur $T_b$. Mit $T_b = T_e(1 - e^{-\tau})$ folgt die optische Tiefe τ bei 1.4 GHz. Mit $\tau = 3.28\cdot10^{-7}\,(T_e/10^4\,\text{K})^{-1.35}\,(\nu/\text{GHz})^{-2.1}\,EM$ folgt das Emissionsmaß (Condon & Ransom 2016). Mit einer leuchtenden Schicht von 0.1 pc Dicke ergibt sich daraus die Dichte.
- **Ganzes Gebiet aus Planck 70 GHz:** Eine zweidimensionale Gauß-Glocke liefert die Größe; abzüglich des Strahls (13.3′) folgt der Radius. Q und Radius geben die mittlere quadratische Dichte und die Masse des ionisierten Gases (Theorie, Abschnitte 1 und 3).

Rechts steht der Strömgren-Radius als Funktion der Dichte mit den beiden Messungen.

**Hintergrund & Fachbegriffe:**
- **Helligkeitstemperatur:** die Temperatur, die ein Schwarzkörper im Rayleigh-Jeans-Bereich haben müsste, um so hell zu sein; $T_b = S\,c^2/(2k\nu^2\Omega)$ für den Fluss S im Strahl Ω.
- **Optisch dick:** Bei τ ≳ 1 sieht man nur noch die Oberfläche, und $T_b$ nähert sich $T_e$; die Radiostrahlung zählt dann nicht mehr alle Elektronen.
- **Dicke der leuchtenden Schicht:** Das ionisierte Gas liegt im Orionnebel als dünne Schicht vor der Molekülwolke; für die Rechnung ist sie mit 0.1 pc angenommen (eigene Annahme, Größenordnung aus den Ausmaßen des hellen Kerns).

```python
nv = WIDEMAP["NVSS"]
om_nvss = 1.1331 * np.radians(45 / 3600) ** 2
T_B = nv[R_ARCMIN < 5].max() * 1e-26 * C_SI**2 / (2 * K_B * 1.4e9**2 * om_nvss)
TAU_14 = -np.log(1 - T_B / T_E)
EM_CORE = TAU_14 / (3.28e-7 * (T_E / 1e4) ** -1.35 * 1.4 ** -2.1)
N_CORE = np.sqrt(EM_CORE / 0.1)
z70 = WIDEMAP["Planck 70"] - np.median(WIDEMAP["Planck 70"][(R_ARCMIN > BG[0]) & (R_ARCMIN < BG[1])])


def gauss2d(p):
    A, x0, y0, s1, s2, th = p
    xr = (xx - x0) * np.cos(th) + (yy - y0) * np.sin(th)
    yr = -(xx - x0) * np.sin(th) + (yy - y0) * np.cos(th)
    return A * np.exp(-0.5 * (xr**2 / s1**2 + yr**2 / s2**2))


fit70 = optimize.least_squares(lambda p: (gauss2d(p) - z70)[R_ARCMIN < 30], [z70.max(), N_WIDE / 2, N_WIDE / 2, 15, 15, 0])
fwhm = 2.355 * np.abs(fit70.x[3:5]) * WIDE / N_WIDE * 60
SRC = np.sqrt(np.clip(fwhm**2 - 13.3**2, 0, None))
R_ION = 0.5 * np.sqrt(SRC[0] * SRC[1]) * D_FROM_GAIA * np.radians(1 / 60)    # pc, Halbwertsradius
R_cm = R_ION * PC_CM
N_RMS = np.sqrt(3 * Q_ION / (4 * np.pi * R_cm**3 * ALPHA_B))
M_HII = 4 / 3 * np.pi * R_cm**3 * N_RMS * 1.4 * M_H / MSUN
EPS = (N_RMS / N_CORE) ** 2
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 6))
zoom = R_ARCMIN < 20
c0, c1 = N_WIDE // 2 - 40, N_WIDE // 2 + 40
im = ax.imshow(nv[c0:c1, c0:c1], origin="lower", extent=[20, -20, -20, 20], cmap="magma", vmin=0, vmax=np.percentile(nv[c0:c1, c0:c1], 99.8))
plt.colorbar(im, ax=ax, label="NVSS 1.4 GHz [Jy/Strahl]")
ax.contour(np.linspace(20, -20, 80), np.linspace(-20, 20, 80), z70[c0:c1, c0:c1], levels=[0.5 * z70.max()], colors="cyan", linewidths=1.5)
ax.set_xlabel("Rektaszension relativ [′]")
ax.set_ylabel("Deklination relativ [′]")
ax.set_title("Der Kern bei 1.4 GHz (NVSS); cyan: halbe Höhe bei 70 GHz", fontsize=10)
nn = np.logspace(1, 5, 100)
ax2.loglog(nn, (3 * Q_ION / (4 * np.pi * nn**2 * ALPHA_B)) ** (1 / 3) / PC_CM, color="C0", lw=1.8, label=f"Strömgren-Radius, Q = {Q_ION:.1e} s⁻¹")
ax2.plot(N_RMS, R_ION, "o", color="C3", ms=9, label=f"ganzes Gebiet (70 GHz): R = {R_ION:.2f} pc, n_rms = {N_RMS:.0f}")
ax2.axvline(N_CORE, color="C2", ls="--", label=f"Kern aus NVSS: n ≈ {N_CORE:.0f} cm⁻³")
ax2.set_xlabel(r"Elektronendichte $n_e$ [cm$^{-3}$]")
ax2.set_ylabel("Radius [pc]")
ax2.set_title("Wie groß kann das ionisierte Gebiet sein?", fontsize=10)
ax2.legend(fontsize=9)
plt.tight_layout()
plt.show()

print(f"NVSS: Spitze {nv[R_ARCMIN < 5].max():.1f} Jy/Strahl → T_b = {T_B:.0f} K, τ(1.4 GHz) = {TAU_14:.2f}, EM = {EM_CORE:.2e} pc cm⁻⁶, "
      f"n ≈ {N_CORE:.0f} cm⁻³ (0.1 pc); τ(28 GHz) = {TAU_14 * (28.4 / 1.4) ** -2.1:.4f}")
print(f"70 GHz: FWHM {fwhm[0]:.1f}′ × {fwhm[1]:.1f}′, ohne Strahl {SRC[0]:.1f}′ × {SRC[1]:.1f}′ → R = {R_ION:.2f} pc; n_rms = {N_RMS:.0f} cm⁻³, "
      f"Masse {M_HII:.1f} M☉; Füllfaktor (n_rms/n_Kern)² = {EPS:.2f}")
print(f"Strömgren-Radius: n = 10³ → {(3 * Q_ION / (4 * np.pi * 1e6 * ALPHA_B)) ** (1 / 3) / PC_CM:.2f} pc, n = {N_CORE:.0f} → "
      f"{(3 * Q_ION / (4 * np.pi * N_CORE**2 * ALPHA_B)) ** (1 / 3) / PC_CM:.2f} pc")
```

* Links der Kern bei 1.4 GHz in NVSS, mit der halben Höhe bei 70 GHz (cyan). Der helle Kern ist etwa 5′ groß und sitzt beim Trapez; die Kontur bei 70 GHz ist größer, weil der Strahl dort 13′ beträgt. Rechts der Strömgren-Radius für Q = 5.7·10⁴⁸ s⁻¹ mit den beiden Messungen.
* Kern: Die Spitze von 14.8 Jy pro Strahl entspricht $T_b$ = 4561 K. Mit $T_e$ = 10⁴ K ist τ(1.4 GHz) = 0.61, EM = 3.8·10⁶ pc cm⁻⁶, und mit einer Schicht von 0.1 pc n ≈ 6100 cm⁻³. Bei 28 GHz ist τ nur 0.001, die Planck-Flüsse zählen also alle Elektronen. Ganzes Gebiet: Die Quelle bei 70 GHz ist ohne Strahl 9.5′ × 2.9′ groß, R = 0.30 pc. Damit verlangt Q eine mittlere Dichte von 2600 cm⁻³ und eine Masse von 10 M☉ ionisiertem Gas.
* Zur Vorhersage: größer, aber nur um den Faktor 2.4; der Füllfaktor ist (2600/6100)² = 0.18. Der gemessene Radius liegt fast genau auf der Strömgren-Kurve: Bei der mittleren Dichte reichen die Photonen von θ¹ Ori C bis 0.30 pc, und so groß ist das Gebiet. Die Dichte im Kern hängt an der angenommenen Dicke der Schicht (n ∝ 1/√L).

##### *8. Das Spektrum der ganzen Region*

Bei welcher Wellenlänge strahlt die Region am meisten, wie warm ist ihr Staub, und reicht das Licht der Trapezsterne, um ihn zu heizen? Vorher vorhersagen: Liegt das Maximum der Staubstrahlung bei 12, 60, 100 oder 350 µm?

Der Code zeichnet die Flüsse aus Schritt 4 als Spektrum von 28 GHz bis 25 THz (12 µm). Davon zieht er die Frei-frei-Strahlung aus Schritt 5 ab (mit α = −0.1 fortgesetzt). An den Rest von 60 µm bis 217 GHz passt er einen modifizierten Schwarzkörper an (Theorie, Abschnitt 4). Bei 12 und 25 µm leuchten kleinere, heißere Körner, die dieses Modell nicht beschreibt. Daraus folgen:
- die Infrarot-Leuchtkraft, einmal aus dem Modell und einmal direkt aus den Messpunkten,
- die Masse von Gas und Staub aus 353 GHz.

Zum Vergleich steht die Leuchtkraft eines O7-V-Sterns.

**Hintergrund & Fachbegriffe:**
- **νS_ν:** Trägt man $\nu S_\nu$ auf, zeigt das Maximum, wo pro logarithmischem Frequenzintervall die meiste Energie herauskommt.
- **β kleiner als 1.5:** Ein einzelner modifizierter Schwarzkörper mit kleinem β beschreibt oft eine Mischung aus Staub verschiedener Temperaturen.

```python
keys = [k for k in CAL if k != "Bonn 1.4"]
nu_all = np.array([BANDS[k][1] for k in keys])
S_all = np.array([FLUX[k][0] for k in keys])
E_all = np.array([FLUX[k][1] for k in keys])
ff_all = S_ff[0] * (nu_all / 28.4e9) ** -0.1
dust = S_all - ff_all
fitm = (nu_all <= 5e12) & (nu_all >= 217e9)
res = optimize.least_squares(lambda p: np.log(mbb(nu_all[fitm], *p)) - np.log(dust[fitm]), [np.log(1e-3), 40.0, 1.7])
LNA, T_DUST, BETA = res.x
J = res.jac
cov = np.linalg.inv(J.T @ J) * np.sum(res.fun**2) / max(fitm.sum() - 3, 1)
E_T, E_BETA = np.sqrt(np.diag(cov))[1:]
d_m = D_FROM_GAIA * PC_CM / 100
nn = np.logspace(np.log10(100e9), 14, 3000)
L_MBB = 4 * np.pi * d_m**2 * np.trapz(mbb(nn, *res.x) * 1e-26, nn) / (LSUN / 1e7)
o = np.argsort(nu_all)
L_DATA = 4 * np.pi * d_m**2 * np.trapz(dust[o] * 1e-26, nu_all[o]) / (LSUN / 1e7)
kappa353 = 0.1 * (353e9 / 1e12) ** BETA * 0.1                  # m²/kg Gas
M_GAS = dust[keys.index("Planck 353")] * 1e-26 * d_m**2 / (kappa353 * planck(353e9, T_DUST)) / (MSUN / 1e3)
L_O7 = 10 ** MARTINS_V.logL[MARTINS_V.typ == "O7"].values[0]

fig, ax = plt.subplots(figsize=(12, 6.5))
ax.errorbar(nu_all / 1e9, nu_all * S_all * 1e-26, yerr=nu_all * E_all * 1e-26, fmt="o", color="k", ms=7, capsize=3, label="gemessen (Öffnung 40′)")
ax.plot(nn / 1e9, nn * mbb(nn, *res.x) * 1e-26, color="C3", lw=1.6, label=f"Staub: T = {T_DUST:.1f} K, β = {BETA:.2f}")
nf = np.logspace(9.4, 13.5, 200)
ax.plot(nf / 1e9, nf * S_ff[0] * (nf / 28.4e9) ** -0.1 * 1e-26, color="C0", lw=1.4, label="Frei-frei-Strahlung (α = −0.1)")
ax.plot(nf / 1e9, nf * (mbb(nf, *res.x) + S_ff[0] * (nf / 28.4e9) ** -0.1) * 1e-26, color="0.4", ls="--", lw=1, label="Summe")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_ylim(1e-14, 3e-8)
ax.set_xlabel("Frequenz [GHz]")
ax.set_ylabel(r"$\nu S_\nu$ [W/m²]")
ax.set_title("Das Spektrum des Orionnebels von 1 cm bis 12 µm", fontsize=10)
top = ax.secondary_xaxis("top", functions=(lambda f: 2.998e5 / np.maximum(f, 1e-9), lambda w: 2.998e5 / np.maximum(w, 1e-9)))
top.set_xlabel("Wellenlänge [µm]")
ax.legend(fontsize=9)
plt.show()

print(f"Staub: T = {T_DUST:.1f} ± {E_T:.1f} K, β = {BETA:.2f} ± {E_BETA:.2f}; Maximum von νS_ν bei {2.998e14 / nu_all[np.argmax(nu_all * S_all)]:.0f} µm")
print(f"Infrarot-Leuchtkraft: Modell {L_MBB:.2e} L☉, Messpunkte {L_DATA:.2e} L☉; O7 V nach Martins: {L_O7:.2e} L☉ "
      f"(Verhältnis Messpunkte/O7 V = {L_DATA / L_O7:.1f})")
print(f"Masse von Gas und Staub aus 353 GHz: {M_GAS:.0f} M☉; ionisiertes Gas (Schritt 7): {M_HII:.0f} M☉")
```

* Das Spektrum der Region als $\nu S_\nu$: schwarz gemessen, rot der Staub (modifizierter Schwarzkörper), blau die Frei-frei-Strahlung, gestrichelt die Summe. Unter 100 GHz dominiert das ionisierte Gas, darüber der Staub. Bei 12 und 25 µm liegen die Messpunkte weit über dem Modell: Dort leuchten kleine, heiße Körner und PAH, die eine einzelne Temperatur nicht beschreibt.
* Der Staub hat T = 36.8 ± 2.8 K und β = 1.19 ± 0.14; das Maximum von $\nu S_\nu$ liegt bei 60 µm. Die Infrarot-Leuchtkraft ist 1.3·10⁵ L☉ aus dem Modell und 2.2·10⁵ L☉ direkt aus den Messpunkten (mit 12 und 25 µm), das 1.6-Fache eines O7-V-Sterns (1.4·10⁵ L☉). Aus 353 GHz folgen 1560 M☉ Gas und Staub, 150-mal mehr als ionisiertes Gas.
* Zur Vorhersage: bei 60 µm (genauer 60 bis 100 µm). Die Energiebilanz geht auf: Der O7-Stern allein liefert gut die Hälfte; den Rest geben die übrigen Trapezsterne, die eingebetteten jungen Sterne im Kleinmann-Low-Gebiet und M 43 mit NU Ori.

##### *9. Eine Karte der Staubtemperatur*

Wo ist der Staub warm, wo kalt, und wie fällt die Temperatur mit dem Abstand vom Trapez? Vorher vorhersagen: Wenn ein einzelner Stern heizt, wie sollte die Temperatur mit dem Abstand r fallen?

Für jedes Pixel rechnet der Code aus dem Verhältnis der Flächenhelligkeiten bei 60 und 100 µm (IRIS) die Temperatur, für die ein modifizierter Schwarzkörper mit β = 1.5 dieses Verhältnis hat. Links steht die Karte, rechts die Temperatur gegen den Abstand vom Trapez mit dem Gesetz $T \propto r^{-2/(4+\beta)}$ für die Heizung durch einen Stern (siehe [Die Dunkelwolke ρ Ophiuchi](https://github.com/deltorobarba/science/blob/main/astro/nebulae_dust.ipynb), Theorie, Abschnitt 4). Er passt eine Gerade im doppelt-logarithmischen Bild an.

**Hintergrund & Fachbegriffe:**
- **Farbtemperatur:** die Temperatur aus dem Verhältnis zweier Bänder; sie ist größer als die mittlere Temperatur entlang der Sichtlinie, wenn warmer und kalter Staub gemischt sind.
- **Heizung durch einen Stern:** Absorbierter Fluss ∝ $1/r^2$, abgestrahlte Leistung ∝ $T^{4+\beta}$, also $T \propto r^{-2/(4+\beta)} = r^{-0.36}$ für β = 1.5.

```python
i60, i100 = to_jy_sr("IRIS 60"), to_jy_sr("IRIS 100")
bg60 = np.median(i60[(R_ARCMIN > BG[0]) & (R_ARCMIN < BG[1])])
bg100 = np.median(i100[(R_ARCMIN > BG[0]) & (R_ARCMIN < BG[1])])
ratio = (i60 - 0.5 * bg60) / (i100 - 0.5 * bg100)               # halber Hintergrund abgezogen: Wolke vor und hinter dem Nebel
Tg = np.linspace(10, 150, 1401)
rg = (5e12 / 3e12) ** 1.5 * planck(5e12, Tg) / planck(3e12, Tg)
TMAP = np.interp(ratio, rg, Tg, left=np.nan, right=np.nan)
TMAP[(i100 - 0.5 * bg100) < 3 * np.std(i100[(R_ARCMIN > BG[0]) & (R_ARCMIN < BG[1])])] = np.nan
rr = np.arange(1, 50, 1.0)
tprof = np.array([np.nanmedian(TMAP[(R_ARCMIN >= r - 0.5) & (R_ARCMIN < r + 0.5)]) for r in rr])
ok = (rr >= 3) & (rr <= 30) & np.isfinite(tprof)
SLOPE, lnT0 = np.polyfit(np.log(rr[ok]), np.log(tprof[ok]), 1)
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(16, 6))
im = ax.imshow(TMAP, origin="lower", extent=ext, cmap="coolwarm", vmin=20, vmax=70)
ax.plot(0, 0, "+", color="k", ms=12, mew=1.5)
plt.colorbar(im, ax=ax, label="Farbtemperatur 60/100 µm [K]")
ax.set_xlabel("Rektaszension relativ [′]")
ax.set_ylabel("Deklination relativ [′]")
ax.set_title("Staubtemperatur aus IRIS 60 und 100 µm (β = 1.5)", fontsize=10)
ax2.loglog(rr, tprof, "o", color="C3", ms=5, label="Median im Ring")
ax2.loglog(rr[ok], np.exp(lnT0) * rr[ok] ** SLOPE, color="k", lw=1.4, label=f"Fit 3–30′: T ∝ r^{SLOPE:.2f}")
ax2.loglog(rr[ok], np.exp(lnT0) * rr[ok][0] ** SLOPE * (rr[ok] / rr[ok][0]) ** (-2 / 5.5), color="C0", ls="--", lw=1.2, label="ein Stern: T ∝ r^−0.36")
ax2.set_xlabel("Abstand vom Trapez [′]")
ax2.set_ylabel("Farbtemperatur [K]")
ax2.set_title("Wie die Staubtemperatur nach außen fällt", fontsize=10)
ax2.legend(fontsize=9)
plt.tight_layout()
plt.show()

print(f"Farbtemperatur: Maximum {np.nanmax(TMAP):.1f} K, bei 5′ {np.interp(5, rr, tprof):.1f} K, bei 20′ {np.interp(20, rr, tprof):.1f} K, "
      f"bei 40′ {np.interp(40, rr, tprof):.1f} K; Steigung {SLOPE:.2f} (ein Stern: −0.36)")
```

* Links die Farbtemperatur aus IRIS 60 und 100 µm für jedes Pixel mit genug Signal, rechts der Median in Ringen um das Trapez. Am wärmsten (bis 47 K) ist der Staub nahe dem Kern, nach außen fällt die Temperatur bis 32 K bei 40′. Im Süden folgt die kühlere Wolke Orion A.
* Die Temperatur bleibt in den innersten 10′ bei etwa 39 bis 41 K und fällt danach nur langsam; die Steigung von 3 bis 30′ ist −0.13 statt −0.36 für die Heizung durch einen einzelnen Stern. Im Kern verschmiert die Auflösung (2′) die heißesten Stellen, und weiter außen heizen auch das allgemeine Sternlicht der Milchstraße und Sterne des Haufens. Entlang der Sichtlinie mischt sich zudem Staub vor und hinter dem Nebel.

##### *10. Junge Sterne mit Infrarot-Überschuss*

Wie viele Sterne im Orionnebel-Haufen haben noch eine Scheibe, die man im nahen Infrarot sieht? Vorher vorhersagen: Wie viele Sterne mit scheinbarem Überschuss erwartet man in einem Feld ohne junge Sterne?

Der Code nimmt aus 2MASS alle Quellen mit guter Photometrie in J, H und K (Qualität A in allen drei Bändern) bis 10′ um das Trapez und in einem gleich großen Vergleichsfeld 6° östlich. Ein Stern zählt als Stern mit Überschuss, wenn er im Diagramm $J-H$ gegen $H-K$ weiter rechts von der Rötungsgeraden liegt als sein Farbfehler. Die Gerade geht durch das rote Ende der Zwergsterne ($H-K = 0.30$, $J-H = 0.61$) mit der Steigung 1.70 (Theorie, Abschnitt 5). Der Anteil im Vergleichsfeld zeigt, wie oft das Kriterium fälschlich anschlägt.

**Hintergrund & Fachbegriffe:**
- **Rötungsgerade:** Staub schiebt einen Stern im Farbdiagramm entlang dieser Richtung nach rechts oben; ohne Scheibe kann er nicht rechts von der Geraden durch die rötesten gewöhnlichen Sterne liegen.
- **Klassische T-Tauri-Sterne:** junge Sterne mit Scheibe, die noch Gas aufnehmen; sie liegen auf einem eigenen Streifen rechts der Rötungsgeraden (Meyer et al. 1997).

```python
# jh_o, hk_o, ex_o und jh_c, hk_c, ex_c aus Schritt 6 (Funktion colors)
F_ONC, F_CTRL = ex_o.mean(), ex_c.mean()
F_CORR = (F_ONC - F_CTRL) / (1 - F_CTRL)
fig, axs = plt.subplots(1, 2, figsize=(16, 6.5), sharex=True, sharey=True)
hk_line = np.linspace(-0.3, 2.5, 10)
for ax, (jh, hk, ex, lab) in zip(axs, [(jh_o, hk_o, ex_o, "Orionnebel-Haufen (10′)"), (jh_c, hk_c, ex_c, "Vergleichsfeld 6° östlich (10′)")]):
    ax.scatter(hk[~ex], jh[~ex], s=6, color="0.5", label="ohne Überschuss")
    ax.scatter(hk[ex], jh[ex], s=8, color="C3", label=f"mit Überschuss: {ex.sum()} von {len(ex)} ({100 * ex.mean():.1f} %)")
    ax.plot(hk_line, 0.61 + 1.70 * (hk_line - 0.30), color="k", lw=1.2, label="Rötungsgerade durch das rote Ende der Zwerge")
    ax.plot(hk_line, 0.52 + 0.58 * hk_line, color="C0", ls="--", lw=1.2, label="T-Tauri-Streifen (Meyer et al. 1997)")
    ax.set_xlim(-0.3, 2.2)
    ax.set_ylim(-0.2, 3.5)
    ax.set_xlabel("H − K [mag]")
    ax.set_title(lab, fontsize=10)
    ax.legend(fontsize=8.5, loc="upper left")
axs[0].set_ylabel("J − H [mag]")
plt.tight_layout()
plt.show()

print(f"Orionnebel-Haufen: {ex_o.sum()} von {len(ex_o)} mit Überschuss ({100 * F_ONC:.1f} %); Vergleichsfeld {ex_c.sum()} von {len(ex_c)} "
      f"({100 * F_CTRL:.1f} %); um Fehlalarme korrigiert {100 * F_CORR:.1f} %")
print(f"Hillenbrand et al. 1998 (aus I − K): mindestens {100 * DISK_HILLENBRAND[0]:.0f} %, höchstens {100 * DISK_HILLENBRAND[1]:.0f} %")
```

Im Vergleichsfeld liegen fast alle Sterne in einem kleinen Fleck ohne Rötung.
* Im Haufen haben 249 von 873 Sternen (28.5 %) einen Überschuss, im Vergleichsfeld 13 von 315 (4.1 %), das sind Fehlalarme. Um sie korrigiert hat ein Viertel der Sterne eine heiße innere Scheibe, die man in K sieht: 25.4 %.
* Zur Vorhersage: einige Prozent, hier 4.1 %, wegen der Farbfehler und Sterne mit ungewöhnlichen Farben. Hillenbrand et al. <details><summary>1. Warum sieht der Orionnebel in jeder Wellenlänge anders aus?</summary>

Jedes Band sieht eine andere Zutat: Röntgen und nahes Infrarot die jungen Sterne, Hα und Radio das ionisierte Gas, 12 bis 60 µm warmen Staub um das Trapez und die Orion-Bar, 350 und 850 µm den kalten Staub der Molekülwolke. Wo das hellste Pixel liegt, verrät, welche Zutat das Band zeigt.
</details>

<details><summary>2. Warum ist eine Fläche mit konstantem Wert im hellsten Teil eines Bildes verdächtig?</summary>

Der Himmel hat dort keinen Grund, gleichmäßig zu sein; ein Detektor, der nicht weiter zählen kann, schon. SHASSA steht im Kern auf 1063 Quadratbogenminuten auf demselben Wert, und das hellste Pixel liegt 19′ daneben. Für Hα-Flüsse ist die Karte dort unbrauchbar.
</details>

<details><summary>3. Wie bestimmt Gaia die Entfernung eines Haufens genauer als die eines einzelnen Sterns?</summary>

Die Mitglieder stehen fast alle in derselben Entfernung; das gewichtete Mittel von 915 Parallaxen hat einen Fehler von nur 0.0005 mas. Dann bestimmt der Nullpunkt von Gaia den Fehler, nicht die Statistik.
</details>

<details><summary>4. Warum misst die Radiostrahlung die Zahl der ionisierenden Photonen?</summary>

Im Gleichgewicht wird jedes ionisierende Photon durch eine Rekombination ausgeglichen, und beides geht wie $n_e^2$. Die Frei-frei-Strahlung wächst ebenfalls mit $n_e^2$. Ist sie optisch dünn und nicht vom Staub geschluckt, zählt ihr Fluss also die Rekombinationen und damit die Photonen.
</details>

<details><summary>5. Woran erkennt man Frei-frei-Strahlung im Spektrum?</summary>

Optisch dünn ist ihr Spektrum fast flach, $S_\nu \propto \nu^{-0.1}$, optisch dick steigt es wie $\nu^2$. Synchrotronstrahlung fällt steiler (−0.7), Staub steigt viel steiler (+3 bis +4).
</details>

<details><summary>6. Warum ist Q ein unterer Wert für die Photonen des Sterns?</summary>

Ein Teil der ionisierenden Photonen wird vom Staub im ionisierten Gas geschluckt, bevor er ein Atom ionisiert, und ein Teil entweicht durch Lücken in der Wolke. Der Radiofluss sieht nur die Photonen, die tatsächlich ionisiert haben.
</details>

<details><summary>7. Wie liefern J, H und K die Extinktion eines Sterns?</summary>

Staub rötet: Er schwächt J stärker als H. Die meisten Sterne haben fast dieselbe Eigenfarbe in J − H; jeder Überschuss darüber, geteilt durch die Differenz der Extinktionen in J und H (0.107 in Einheiten von $A_V$), ist die Extinktion.
</details>

<details><summary>8. Was sagt der Füllfaktor von 0.18?</summary>

Das ionisierte Gas füllt das Gebiet nicht gleichmäßig. Die Dichte im Kern (6100 cm⁻³) ist höher als die mittlere Dichte, die Q für das ganze Gebiet verlangt (2600 cm⁻³). Als Faustregel wäre das Gas in 18 % des Volumens konzentriert.
</details>

<details><summary>9. Warum liegen die Punkte bei 12 und 25 µm über dem modifizierten Schwarzkörper?</summary>

Dort leuchten sehr kleine Körner und PAH, die einzelne Photonen kurz auf Hunderte Kelvin erhitzen. Eine einzige Temperatur von 37 K beschreibt nur die großen Körner im Gleichgewicht.
</details>

<details><summary>10. Warum ist die Infrarot-Leuchtkraft größer als die eines O7-V-Sterns?</summary>

Der Staub schluckt das Licht aller Sterne im Gebiet: der übrigen Trapezsterne, der eingebetteten jungen Sterne im Kleinmann-Low-Gebiet und von NU Ori in M 43. Der O7-Stern liefert gut die Hälfte.
</details>

<details><summary>11. Warum fällt die Staubtemperatur nach außen langsamer als $r^{-0.36}$?</summary>

Die Formel gilt für einen einzelnen Stern in einer gleichmäßigen Wolke. Hier heizen außen auch das Sternlicht der Milchstraße und die Sterne des Haufens, die Auflösung verschmiert den Kern, und entlang der Sichtlinie mischen sich warmer und kalter Staub.
</details>

<details><summary>12. Warum sieht man mit J, H und K weniger Scheiben als mit I − K?</summary>

Nur der heiße innere Rand einer Scheibe (um 1500 K) strahlt in K merklich; Scheiben mit großem inneren Loch oder um sehr leuchtschwache Sterne bleiben im JHK-Diagramm unauffällig. I − K vergleicht mit einem Band, in dem nur der Stern leuchtet, und ist deshalb empfindlicher.
</details>

#### **Nächste Schritte**

So würde eine Analyse in der Forschung weitergehen:
- **Hα ohne Sättigung:** die Karte von Hα aus kalibrierten Aufnahmen mit kurzer Belichtung (z. B. Hubble) und daraus mit dem Radiofluss die Extinktion vor dem Gas messen, Pixel für Pixel.
- **Radio mit voller Auflösung:** Interferometer mit vielen kurzen Basislinien (VLA in mehreren Konfigurationen, MeerKAT) zeigen das Gas bis zu großen Strukturen; dann folgt τ für jedes Pixel und eine echte Dichteverteilung statt eines Füllfaktors.
- **Die Temperatur des Gases messen:** Radio-Rekombinationslinien (z. B. H64α) geben mit dem Kontinuum $T_e$ direkt, statt 10⁴ K anzunehmen; das ändert Q über $T_e^{-0.45}$.
- **Staub mit mehreren Komponenten:** ein Modell mit Körnern verschiedener Größe (Draine & Li) statt eines einzelnen modifizierten Schwarzkörpers; Herschel-Karten mit 6 bis 36″ Auflösung trennen den warmen Staub um das Trapez vom kalten der Wolke.
- **Scheiben zählen:** mit Spitzer (3.6 bis 8 µm) oder JWST den Anteil der Sterne mit Scheiben nach Masse und Abstand vom Trapez messen; nahe θ¹ Ori C verdampft das UV-Licht die Scheiben (Proplyds).
