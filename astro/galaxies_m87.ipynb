# Der Schatten von M 87*

Das Notebook lädt die öffentlichen Daten des Event Horizon Telescope von 2017 (M 87*, 230 GHz, vier Nächte, zwei Frequenzbänder), zeigt mit Closure-Phasen, wie sich die vom Wetter verfälschten Phasen der Stationen herausheben, und bestimmt den Durchmesser des Rings direkt aus den Visibility-Amplituden mit einem Ringmodell. Aus dem Durchmesser folgt die Masse des Schwarzen Lochs, verglichen mit der EHT-Kollaboration und mit den Massen aus der Bewegung von Sternen und Gas. Mit der echten (u,v)-Abdeckung simuliert es dann die Abbildung: Dirty Image, Högbom-CLEAN und wie viele Punktquellen sich aus 21 Basislinien wiederfinden lassen.

Wie misst ein Netz aus sieben Radioteleskopen einen Ring von 40 Mikrobogensekunden, obwohl es nie ein Bild aufnimmt?

#### **Hintergrund**

M 87 ist eine riesige elliptische Galaxie im Virgo-Haufen, 16.8 Mpc entfernt. In ihrem Zentrum sitzt ein Schwarzes Loch von einigen Milliarden Sonnenmassen, das einen Jet von mehreren tausend Lichtjahren Länge antreibt. Sein Ereignishorizont hat einen Radius von etwa 120 Astronomischen Einheiten; aus 16.8 Mpc erscheint das unter einem Winkel von wenigen Mikrobogensekunden (µas), so groß wie eine Orange auf dem Mond.

Ein Schwarzes Loch selbst leuchtet nicht. Das heiße Gas um es herum strahlt aber im Radiobereich, und die Gravitation lenkt diese Strahlung ab. Von fern gesehen erscheint deshalb ein heller Ring um einen dunklen Fleck, den **Schatten**: die Fläche, aus der kein Lichtstrahl zu uns kommt, weil alle Strahlen mit diesem Stoßparameter im Loch enden.

Kein einzelnes Teleskop kann das auflösen; bei 1.3 mm Wellenlänge bräuchte man eine Schüssel von der Größe der Erde. Das **Event Horizon Telescope** (EHT) verbindet deshalb Radioteleskope auf mehreren Kontinenten zu einem Interferometer (Very Long Baseline Interferometry, VLBI). Jede Station zeichnet das Signal mit einer Atomuhr auf Festplatten auf; erst danach werden die Aufzeichnungen paarweise korreliert. Im April 2017 beobachteten sieben Stationen M 87: ALMA und APEX in Chile, das SMT in Arizona, JCMT und SMA auf Hawaii, das LMT in Mexiko und das IRAM-30-m-Teleskop in Spanien. 2019 veröffentlichte die Kollaboration das erste Bild eines Schwarzen Lochs: einen Ring von 42 µas Durchmesser, im Süden heller als im Norden.

#### **Herleitung**

**1. Auflösung.** Ein Teleskop vom Durchmesser $B$ trennt zwei Punkte, die mehr als etwa $\lambda/B$ auseinanderliegen. Für ein Interferometer zählt der Abstand der Teleskope, die Basislinie. Bei $\lambda = 1.3$ mm und $B = 10\,000$ km ist $\lambda/B = 1.3 \cdot 10^{-10}$ rad $= 27$ µas. Basislinien gibt man oft in Wellenlängen an: $B/\lambda = 8 \cdot 10^9 = 8$ Gλ.

**2. Was ein Interferometer misst.** Eine ebene Welle aus der Richtung $\hat s$ erreicht das zweite Teleskop um die Zeit $\tau = \vec B \cdot \hat s/c$ später als das erste. Multipliziert man die beiden Signale und mittelt über die Zeit, bleibt für einen Punkt der Quelle mit der Helligkeit $I(\hat s)$ der Anteil $I(\hat s)\,e^{-2\pi i \nu \tau} = I(\hat s)\,e^{-2\pi i \vec B \cdot \hat s/\lambda}$. Verschiedene Punkte der Quelle strahlen unabhängig voneinander (inkohärent), ihre Beiträge addieren sich. Mit kleinen Winkeln $(l, m)$ um die Mitte des Feldes und der Basislinie $(u, v)$ in Wellenlängen, projiziert auf die Himmelsebene, ist die gemessene **Visibility**
$$V(u, v) = \iint I(l, m)\,e^{-2\pi i (u l + v m)}\,dl\,dm$$
(Satz von van Cittert und Zernike): die zweidimensionale Fourier-Transformierte des Himmelsbildes an der Stelle $(u, v)$. Jedes Teleskoppaar misst zu jedem Zeitpunkt einen Punkt in der Fourier-Ebene. Weil sich die Erde dreht, wandert die projizierte Basislinie über die Nacht und zeichnet eine Spur, die (u,v)-Abdeckung.

**3. Hermitesche Symmetrie.** Das Bild $I$ ist reell. Dann gilt $V(-u, -v) = V^*(u, v)$; jede Messung liefert auch den gespiegelten Punkt.

**4. Ein dünner Ring.** Ein Ring vom Durchmesser $d$ mit dem Gesamtfluss $F$ hat in Polarkoordinaten die Helligkeit $I(r) = \frac{F}{\pi d}\,\delta(r - d/2)$. Für eine kreissymmetrische Quelle hängt $V$ nur von $\rho = \sqrt{u^2 + v^2}$ ab. Mit $u l + v m = \rho\,r\cos\varphi$:
$$V(\rho) = \int_0^{2\pi}\!\!\int_0^\infty I(r)\,e^{-2\pi i \rho r \cos\varphi}\,r\,dr\,d\varphi = \frac{F}{\pi d}\cdot\frac{d}{2}\int_0^{2\pi} e^{-i\pi d \rho \cos\varphi}\,d\varphi = F\,J_0(\pi d \rho),$$
mit der Bessel-Funktion $J_0(x) = \frac{1}{2\pi}\int_0^{2\pi} e^{-i x \cos\varphi}\,d\varphi$. $J_0$ hat seine erste Nullstelle bei $x = 2.405$. Die Amplitude verschwindet also bei
$$\rho_0 = \frac{2.405}{\pi d} ;$$
für $d = 42$ µas $= 2.04 \cdot 10^{-10}$ rad liegt sie bei 3.8 Gλ. Die Lage dieser Nullstelle misst den Durchmesser, ganz ohne Bild.

**5. Ein Ring mit Breite.** Ein echter Ring ist nicht unendlich dünn. Faltet man den dünnen Ring mit einer Gaußfunktion der Halbwertsbreite $w$ (also $\sigma = w/\sqrt{8\ln 2}$), multipliziert sich nach dem Faltungssatz die Fourier-Transformierte mit der Transformierten der Gaußfunktion, $e^{-2\pi^2\sigma^2\rho^2}$:
$$|V(\rho)| = F\,|J_0(\pi d \rho)|\,\exp\left(-\frac{(\pi w \rho)^2}{4 \ln 2}\right) .$$

**6. Dirty Image und Dirty Beam.** Gemessen ist $V$ nur an den Punkten $(u_k, v_k)$, beschrieben durch die Abtastfunktion $S(u, v) = \sum_k \delta(u - u_k, v - v_k)$. Transformiert man die gemessenen Visibilities zurück, erhält man nach dem Faltungssatz
$$I_D = \mathcal{F}^{-1}[S \cdot V] = I * B, \qquad B = \mathcal{F}^{-1}[S],$$
das wahre Bild gefaltet mit dem **Dirty Beam** $B$, dem Bild einer Punktquelle bei dieser Abdeckung. Bei 21 Basislinien hat $B$ starke Nebenkeulen, und das Dirty Image ist voller Artefakte.

**7. CLEAN.** Högbom (1974) nimmt an, dass der Himmel aus wenigen Punktquellen besteht, $I = \sum_j a_j\,\delta(x - x_j)$. Dann ist $I_D = \sum_j a_j B(x - x_j)$. Der Algorithmus sucht im Restbild $R$ (anfangs $I_D$) das hellste Pixel $x_j$, speichert eine Komponente $\gamma R(x_j)$ mit einem kleinen Faktor $\gamma \approx 0.1$ und zieht $\gamma R(x_j)\,B(x - x_j)$ von $R$ ab. Das wiederholt er, bis $R$ nur noch Rauschen enthält. Am Ende faltet er die Komponenten mit einer Gaußfunktion von der Breite der Hauptkeule und addiert $R$. CLEAN ist eine gierige Suche nach den $k$ hellsten Quellen, dieselbe Struktur wie Matching Pursuit (Mallat & Zhang 1993) mit verschobenen Dirty Beams als Wörterbuch.

**8. Closure-Phasen.** Die Atmosphäre über jeder Station verzögert das Signal zufällig um Bruchteile einer Wellenlänge; jede Station $i$ bekommt einen eigenen Phasenfehler $\theta_i(t)$. Die gemessene Phase auf der Basislinie $ij$ ist $\phi_{ij}^\text{beob} = \phi_{ij} + \theta_i - \theta_j$. Summiert man über ein geschlossenes Dreieck,
$$\phi_{ij}^\text{beob} + \phi_{jk}^\text{beob} + \phi_{ki}^\text{beob} = \phi_{ij} + \phi_{jk} + \phi_{ki} + (\theta_i - \theta_j) + (\theta_j - \theta_k) + (\theta_k - \theta_i),$$
fallen die Fehler der Stationen heraus. Diese **Closure-Phase** misst die Quelle allein. Ist das Bild punktsymmetrisch, ist $V$ reell, jede Phase 0° oder 180° und die Closure-Phase auch. Jeder andere Wert zeigt, dass das Bild asymmetrisch ist.

**9. Der Schatten.** Für Licht in der Umgebung einer Masse $M$ (Schwarzschild-Metrik) gilt mit $u = 1/r$, dem Stoßparameter $b$ und $r_g = GM/c^2$
$$\left(\frac{du}{d\varphi}\right)^2 = \frac{1}{b^2} - W(u), \qquad W(u) = u^2\,(1 - 2 r_g u) .$$
Ein Strahl von weit her läuft nach innen, bis $1/b^2 = W(u)$; dort kehrt er um. $W$ hat sein Maximum bei $dW/du = 2u - 6 r_g u^2 = 0$, also bei $r = 3 r_g$ (der Photonensphäre), mit $W_\text{max} = \frac{1}{9 r_g^2}\left(1 - \frac{2}{3}\right) = \frac{1}{27 r_g^2}$. Ist $1/b^2 > W_\text{max}$, gibt es keinen Umkehrpunkt, und der Strahl fällt ins Loch. Alle Strahlen mit
$$b < \sqrt{27}\,r_g \approx 5.2\,r_g$$
enden im Loch; von fern sieht man eine dunkle Scheibe mit dem Winkeldurchmesser
$$d_\text{Schatten} = 2\sqrt{27}\,\theta_g \approx 10.4\,\theta_g, \qquad \theta_g = \frac{G M}{c^2 D} .$$
Für $M = 6.5 \cdot 10^9\,M_\odot$ und $D = 16.8$ Mpc ist $\theta_g = 3.8$ µas und $d_\text{Schatten} = 40$ µas. Der leuchtende Ring liegt etwas außerhalb des Schattens; Simulationen ergeben $d \approx 11\,\theta_g$. Der Drehimpuls des Lochs ändert den Durchmesser nur um wenige Prozent.

#### **Daten**

Die EHT-Kollaboration hat mit dem ersten Bild die kalibrierten Visibilities veröffentlicht (Datensatz `2019-D01-01` auf GitHub): M 87 am 5., 6., 10. und 11. April 2017 (Tag 095, 096, 100 und 101 des Jahres), jeweils in zwei Frequenzbändern um 227 und 229 GHz („lo“, „hi“). Die Daten sind über die Zeit gemittelt, die Amplituden mit den kurzen Basislinien innerhalb eines Standorts geeicht („netcal“). Pro Messung enthalten die Dateien die Polarisationen RR und LL; ihr Mittel ist die Gesamtintensität (Stokes I).

Die acht Dateien (zusammen etwa 4 MB) landen in `astrodata/eht`; mit `REFRESH = True` lädt das Notebook sie neu.

```python
import urllib.request
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
from scipy.special import j0
from astropy.io import fits

ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "eht").mkdir(parents=True, exist_ok=True)
REFRESH = False                      # True: Dateien neu laden
EHT_URL = "https://raw.githubusercontent.com/eventhorizontelescope/2019-D01-01/master/uvfits/"
DAYS = {"095": "5. April", "096": "6. April", "100": "10. April", "101": "11. April"}
EHT_FILES = [f"SR1_M87_2017_{day}_{band}_hops_netcal_StokesI.uvfits" for day in DAYS for band in ["lo", "hi"]]
STATIONS = {"AA": "ALMA (Chile)", "AP": "APEX (Chile)", "AZ": "SMT (Arizona)", "JC": "JCMT (Hawaii)",
            "LM": "LMT (Mexiko)", "PV": "IRAM 30 m (Spanien)", "SM": "SMA (Hawaii)"}
UAS = np.pi / 180 / 3600e6           # 1 Mikrobogensekunde in Radiant
G, C, MSUN, MPC = 6.674e-11, 2.998e8, 1.989e30, 3.0857e22
pd.set_option("display.width", 160)
```

##### *1. Die Visibilities*

Welche Punkte der Fourier-Ebene haben die sieben Stationen in vier Nächten gemessen, und was sieht man in den Amplituden, bevor man irgendein Modell anpasst?

```python
rows = []
for fname in EHT_FILES:
    path = ASTRODATA / "eht" / fname
    if REFRESH or not path.exists():
        urllib.request.urlretrieve(EHT_URL + fname, path)
    with fits.open(path) as hdul:
        grp = hdul[0]
        freq = grp.header["CRVAL4"]
        names = [n.strip() for n in hdul["AIPS AN"].data["ANNAME"]]
        d = grp.data.data[:, 0, 0, 0, 0, :2, :]                  # RR und LL, je [Realteil, Imaginärteil, Gewicht]
        w = np.clip(d[..., 2], 0, None)
        wsum = w.sum(axis=1)
        ok = wsum > 0
        vis = ((d[..., 0] + 1j * d[..., 1]) * w).sum(axis=1)[ok] / wsum[ok]   # gewichtetes Mittel = Stokes I
        bl = grp.data.par("BASELINE").astype(int)[ok]
        rows.append(pd.DataFrame({"day": fname[13:16], "band": fname[17:19],
                                  "u": grp.data.par("UU---SIN")[ok] * freq, "v": grp.data.par("VV---SIN")[ok] * freq,
                                  "vis": vis, "sigma": 1 / np.sqrt(wsum[ok]), "time": grp.data.par("DATE")[ok],
                                  "st1": [names[a - 1] for a in bl // 256], "st2": [names[a - 1] for a in bl % 256]}))
eht = pd.concat(rows, ignore_index=True)
eht["rho"] = np.hypot(eht["u"], eht["v"])
eht["amp"] = np.abs(eht["vis"])
n_bl = eht[["st1", "st2"]].drop_duplicates().shape[0]
print(f"{len(eht):,} Visibilities, {n_bl} Basislinien zwischen {len(set(eht.st1) | set(eht.st2))} Stationen")
print("Pro Nacht:", {DAYS[k]: v for k, v in eht.groupby("day").size().items()})
print(f"Längste Basislinie {eht['rho'].max() / 1e9:.2f} Gλ -> Auflösung λ/B ≈ {1 / eht['rho'].max() / UAS:.0f} µas")
short = eht.groupby(["st1", "st2"])["rho"].max().sort_values()
print("Kürzeste Basislinien:", ", ".join(f"{a}-{b} {r / 1e6:.1f} Mλ" for (a, b), r in short.head(2).items()),
      "(innerhalb eines Standorts)")
print(f"Amplituden: auf kurzen Basislinien (< 0.1 Gλ) im Median {eht.loc[eht['rho'] < 1e8, 'amp'].median():.2f} Jy, "
      f"auf langen (> 1 Gλ) {eht.loc[eht['rho'] > 1e9, 'amp'].median():.2f} Jy")

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))
for (s1, s2), g in eht.groupby(["st1", "st2"]):
    pts = axes[0].plot(g["u"] / 1e9, g["v"] / 1e9, ".", ms=1.5, label=f"{s1}-{s2}")
    axes[0].plot(-g["u"] / 1e9, -g["v"] / 1e9, ".", ms=1.5, c=pts[0].get_color())
axes[0].set_aspect("equal")
axes[0].invert_xaxis()
axes[0].set_xlabel("u [Gλ]")
axes[0].set_ylabel("v [Gλ]")
axes[0].set_title("(u,v)-Abdeckung des EHT 2017 (mit gespiegelten Punkten)", fontsize=10)
axes[0].legend(fontsize=6, ncol=3, markerscale=5, loc="lower left")
axes[1].scatter(eht["rho"] / 1e9, eht["amp"], s=1, c="grey", alpha=0.3)
axes[1].set_ylim(0, 1.4)
axes[1].set_xlabel("Länge der Basislinie ρ [Gλ]")
axes[1].set_ylabel("|V| [Jy]")
axes[1].set_title("M 87*: Amplitude gegen Länge der Basislinie", fontsize=10)
fig.tight_layout()
plt.show()
```

#### **Reduktion**

##### *2. Phasen und Closure-Phasen*

Wie stark verfälscht die Atmosphäre die Phasen einzelner Basislinien, und bleibt die Closure-Phase eines Dreiecks trotzdem stabil?

```python
DAY, BAND, TRI = "095", "lo", ("AA", "AZ", "LM")      # ALMA, SMT, LMT

def circ_stats(phase_deg):
    # Mittelwert und Streuung für Winkel: über Einheitszeiger mitteln
    z = np.mean(np.exp(1j * np.deg2rad(phase_deg)))
    return np.rad2deg(np.angle(z)), np.rad2deg(np.sqrt(-2 * np.log(np.abs(z))))

sub = eht[(eht["day"] == DAY) & (eht["band"] == BAND)].copy()
sub["t"] = np.round((sub["time"] - sub["time"].min()) * 86400).astype(int)       # Sekunden seit Beginn
def series(a, b):
    s = sub[(sub["st1"] == a) & (sub["st2"] == b)].set_index("t")["vis"]
    return s[~s.index.duplicated()]
a, b, c = TRI
v_ab, v_bc, v_ac = series(a, b), series(b, c), series(a, c)
common = v_ab.index.intersection(v_bc.index).intersection(v_ac.index)
closure = np.angle(v_ab[common] * v_bc[common] * np.conj(v_ac[common]), deg=True)   # φ_ab + φ_bc + φ_ca
phase_ab = np.angle(v_ab[common], deg=True)

print(f"Dreieck {'-'.join(TRI)} am {DAYS[DAY]}, Band {BAND}: {len(common)} gemeinsame Zeitpunkte")
for label, ph in [(f"Phase {a}-{b}", phase_ab), (f"Phase {b}-{c}", np.angle(v_bc[common], deg=True)),
                  (f"Phase {a}-{c}", np.angle(v_ac[common], deg=True)), ("Closure-Phase", closure)]:
    mean, std = circ_stats(ph)
    print(f"  {label:16s} Mittel {mean:+7.1f}°, zirkuläre Streuung {std:6.1f}°")
cl_all = {}
for day in DAYS:
    sub = eht[(eht["day"] == day) & (eht["band"] == BAND)].copy()
    sub["t"] = np.round((sub["time"] - sub["time"].min()) * 86400).astype(int)
    x, y, z = series(a, b), series(b, c), series(a, c)
    cm = x.index.intersection(y.index).intersection(z.index)
    cl_all[day] = circ_stats(np.angle(x[cm] * y[cm] * np.conj(z[cm]), deg=True)) + (len(cm),)
print("Closure-Phase in allen Nächten:", ", ".join(f"{DAYS[k]}: {m:+.0f}° ± {s:.0f}° ({n})" for k, (m, s, n) in cl_all.items()))

hours = np.asarray(common) / 3600
fig, ax = plt.subplots(figsize=(11, 4))
ax.plot(hours, phase_ab, ".", ms=3, c="tab:blue", label=f"Phase der Basislinie {a}-{b}")
ax.plot(hours, closure, ".", ms=3, c="tab:red", label=f"Closure-Phase {'-'.join(TRI)}")
ax.axhline(0, color="black", lw=0.6)
ax.set_ylim(-180, 180)
ax.set_yticks(range(-180, 181, 90))
ax.set_xlabel(f"Zeit seit Beginn der Beobachtung am {DAYS[DAY]} [h]")
ax.set_ylabel("Phase [°]")
ax.set_title("Die Stationsfehler heben sich in der Closure-Phase heraus", fontsize=10)
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

#### **Eigene Messung**

##### *3. Der Ringdurchmesser aus den Amplituden*

Wie groß ist der Ring, wenn man das Modell aus der Herleitung (Punkt 5) an die Amplituden anpasst, und wie stark schwankt das Ergebnis zwischen den Nächten? Basislinien unter 1 Gλ bleiben weg, weil die kurzen Basislinien innerhalb eines Standorts auch den großräumigen Jet sehen.

```python
def ring_amp(rho, F, d_uas, w_uas):
    return F * np.abs(j0(np.pi * d_uas * UAS * rho)) * np.exp(-(np.pi * w_uas * UAS * rho) ** 2 / (4 * np.log(2)))

def fit_ring(df):
    sig = df["sigma"] + 0.05 * df["amp"]               # 5 % systematischer Fehler der Amplitudeneichung
    p, cov = curve_fit(ring_amp, df["rho"], df["amp"], p0=[0.6, 40, 10], sigma=sig, bounds=([0, 10, 0], [3, 100, 60]))
    return p, cov

long_bl = eht["rho"] > 1e9
p_ring, c_ring = fit_ring(eht[long_bl])
F_ring, d_ring, w_ring = p_ring
rho_null = 2.405 / (np.pi * d_ring * UAS)                # erste Nullstelle von J0
per_day = {day: fit_ring(eht[long_bl & (eht["day"] == day)])[0] for day in DAYS}
d_days = np.array([p[1] for p in per_day.values()])

print(f"Ringfit an {long_bl.sum():,} Amplituden auf Basislinien über 1 Gλ")
print(f"  Durchmesser d = {d_ring:.1f} µas (formaler Fehler ± {np.sqrt(c_ring[1, 1]):.2f}), Breite w = {w_ring:.1f} µas, "
      f"kompakter Fluss F = {F_ring:.2f} Jy")
print(f"  Erste Nullstelle der Amplitude bei {rho_null / 1e9:.2f} Gλ")
for day, p in per_day.items():
    print(f"  {DAYS[day]:>10s}: d = {p[1]:.1f} µas, w = {p[2]:.1f} µas, F = {p[0]:.2f} Jy")
print(f"  Streuung zwischen den Nächten: ± {d_days.std(ddof=1):.1f} µas")

fig, ax = plt.subplots(figsize=(10, 5))
ax.scatter(eht["rho"] / 1e9, eht["amp"], s=1, c="grey", alpha=0.3, label="Amplituden")
rr = np.linspace(0.01, 9e9, 500)
ax.plot(rr / 1e9, ring_amp(rr, *p_ring), c="tab:red", lw=2, label=f"Ringmodell: d = {d_ring:.1f} µas, w = {w_ring:.1f} µas")
ax.axvline(rho_null / 1e9, color="tab:red", ls="--", lw=1, label=f"erste Nullstelle {rho_null / 1e9:.2f} Gλ")
ax.axvline(1, color="black", lw=0.6, ls=":")
ax.set_ylim(0, 0.8)
ax.set_xlabel("Länge der Basislinie ρ [Gλ]")
ax.set_ylabel("|V| [Jy]")
ax.set_title("M 87*: Ringmodell an die Amplituden", fontsize=10)
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

##### *4. Dirty Beam und Dirty Image*

Wie sieht eine Punktquelle aus, wenn man nur an den Punkten der EHT-Abdeckung misst, und was bleibt von einem Ring mit dem gemessenen Durchmesser übrig? Die Simulation nimmt perfekte Phasen an; für die Rechenzeit werden die (u,v)-Punkte pro Basislinie über 10 Minuten gemittelt.

```python
eht["tbin"] = np.floor((eht["time"] - eht["time"].min()) * 24 * 6)        # 10-Minuten-Intervalle
uv = eht.groupby(["day", "st1", "st2", "tbin"])[["u", "v"]].mean().to_numpy()
uv = np.vstack([uv, -uv])                                                 # Herleitung, Punkt 3
N_PIX, FOV_UAS = 64, 160
PIX_UAS = FOV_UAS / N_PIX
lm = (np.arange(N_PIX) - N_PIX // 2) * PIX_UAS * UAS
EX = np.exp(2j * np.pi * np.outer(lm, uv[:, 0]))                          # Spalten: l (Osten)
EY = np.exp(2j * np.pi * np.outer(lm, uv[:, 1]))                          # Zeilen: m (Norden)

def sky_visibilities(sources):
    # Visibilities von Punktquellen [(Fluss, l_µas, m_µas), ...] an den EHT-Punkten (Herleitung, Punkt 2)
    return sum(F * np.exp(-2j * np.pi * (uv[:, 0] * l + uv[:, 1] * m) * UAS) for F, l, m in sources)

def dirty_image(vis):
    return ((EY * vis) @ EX.T).real / len(vis)                            # Rücktransformation an den gemessenen Punkten

# Dirty Beam auf einem doppelt so großen Raster, damit er beim CLEAN an jedes Pixel verschoben werden kann
lm2 = (np.arange(2 * N_PIX) - N_PIX) * PIX_UAS * UAS
BEAM2 = (np.exp(2j * np.pi * np.outer(lm2, uv[:, 1])) @ np.exp(2j * np.pi * np.outer(lm2, uv[:, 0])).T).real / len(uv)
dirty_beam = BEAM2[N_PIX // 2:3 * N_PIX // 2, N_PIX // 2:3 * N_PIX // 2]
side = np.abs(dirty_beam[np.hypot(*np.mgrid[-N_PIX // 2:N_PIX // 2, -N_PIX // 2:N_PIX // 2]) * PIX_UAS > 25]).max()

ring_src = [(F_ring / 72, d_ring / 2 * np.cos(t), d_ring / 2 * np.sin(t)) for t in np.linspace(0, 2 * np.pi, 72, endpoint=False)]
ring_dirty = dirty_image(sky_visibilities(ring_src))
print(f"{len(uv) // 2} gemittelte (u,v)-Punkte (dazu die gespiegelten), Bild {N_PIX} x {N_PIX} Pixel zu {PIX_UAS:.1f} µas")
print(f"Höchste Nebenkeule des Dirty Beam außerhalb von 25 µas: {side:.0%} der Hauptkeule")

ext = [(-N_PIX // 2 - 0.5) * PIX_UAS, (N_PIX // 2 - 0.5) * PIX_UAS] * 2           # Pixelmitten auf den Koordinaten lm
def show(ax, img, title, cmap="afmhot"):
    ax.imshow(img, origin="lower", cmap=cmap, extent=ext)
    ax.invert_xaxis()                                                     # Osten links
    ax.set_title(title, fontsize=10)
    ax.set_xlabel("Δα [µas]")
    ax.set_ylabel("Δδ [µas]")

fig, axes = plt.subplots(1, 2, figsize=(11, 5))
show(axes[0], dirty_beam, "Dirty Beam (EHT-Abdeckung 2017)", "viridis")
show(axes[1], ring_dirty, f"Dirty Image eines Rings von {d_ring:.0f} µas")
axes[1].add_patch(plt.Circle((0, 0), d_ring / 2, fill=False, ec="cyan", ls="--", lw=0.8))
fig.tight_layout()
plt.show()
```

##### *5. CLEAN*

Holt CLEAN den Ring und ein Feld aus einzelnen Punktquellen mit Rauschen aus dem Dirty Image zurück?

```python
def hogbom_clean(dirty, gain=0.1, n_iter=1000, threshold=0.02):
    # Herleitung, Punkt 7: hellstes Restpixel suchen, Komponente speichern, verschobenen Dirty Beam abziehen
    res, comps = dirty.copy(), np.zeros_like(dirty)
    for _ in range(n_iter):
        iy, ix = np.unravel_index(np.argmax(res), res.shape)
        if res[iy, ix] < threshold * dirty.max():
            break
        comps[iy, ix] += gain * res[iy, ix]
        res -= gain * res[iy, ix] * BEAM2[N_PIX - iy:2 * N_PIX - iy, N_PIX - ix:2 * N_PIX - ix]
    fwhm_pix = 1 / eht["rho"].max() / UAS / PIX_UAS                     # Restaurierungsstrahl ≈ λ/B_max
    yy_, xx_ = np.mgrid[:N_PIX, :N_PIX] - N_PIX // 2
    clean_beam = np.exp(-4 * np.log(2) * (xx_ ** 2 + yy_ ** 2) / fwhm_pix ** 2)
    restored = np.real(np.fft.ifft2(np.fft.fft2(comps) * np.fft.fft2(np.fft.ifftshift(clean_beam)))) + res
    return comps, res, restored

def top_k_peaks(img, k, min_sep_pix=4):
    # Orte der k hellsten lokalen Maxima (gierig, mit Mindestabstand)
    work, peaks = img.copy(), []
    for _ in range(k):
        iy, ix = np.unravel_index(np.argmax(work), work.shape)
        peaks.append((ix, iy))
        work[max(iy - min_sep_pix, 0):iy + min_sep_pix + 1, max(ix - min_sep_pix, 0):ix + min_sep_pix + 1] = -np.inf
    return np.array(peaks)

def pix_to_uas(p):
    return (p - N_PIX // 2) * PIX_UAS

rng = np.random.default_rng(11)
K_EX, NOISE = 5, 0.02
def random_sky(k):
    return [(rng.uniform(0.5, 1.0), rng.uniform(-60, 60), rng.uniform(-60, 60)) for _ in range(k)]
def noisy(vis):
    n = NOISE * (rng.normal(size=len(vis) // 2) + 1j * rng.normal(size=len(vis) // 2))
    return vis + np.r_[n, np.conj(n)]                                      # Rauschen hermitesch, damit das Bild reell bleibt

ring_comps, _, ring_restored = hogbom_clean(ring_dirty)
sparse_sky = random_sky(K_EX)
sp_dirty = dirty_image(noisy(sky_visibilities(sparse_sky)))
sp_comps, sp_res, sp_restored = hogbom_clean(sp_dirty)
yy_, xx_ = np.mgrid[:N_PIX, :N_PIX] - N_PIX // 2
r_pix = np.hypot(xx_, yy_) * PIX_UAS
centre_ratio = ring_restored[r_pix < 5].mean() / ring_restored[np.abs(r_pix - d_ring / 2) < 3].mean()
print(f"Ring: {np.count_nonzero(ring_comps)} CLEAN-Komponenten; Helligkeit in der Mitte / auf dem Ring = {centre_ratio:.2f}")
print(f"Punktquellen: {np.count_nonzero(sp_comps)} Komponenten, Rest im Restbild {np.abs(sp_res).max() / sp_dirty.max():.1%} des Dirty-Maximums")

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
show(axes[0], ring_restored, "Ring nach CLEAN")
axes[0].add_patch(plt.Circle((0, 0), d_ring / 2, fill=False, ec="cyan", ls="--", lw=0.8))
for ax, img, title in [(axes[1], sp_dirty, f"Dirty Image: {K_EX} Punktquellen mit Rauschen"),
                       (axes[2], sp_restored, "nach CLEAN")]:
    show(ax, img, title)
    ax.scatter([l for _, l, _ in sparse_sky], [m for _, _, m in sparse_sky], marker="+", s=120, c="cyan", label="wahre Quellen")
    pk = top_k_peaks(img, K_EX)
    ax.scatter(pix_to_uas(pk[:, 0]), pix_to_uas(pk[:, 1]), marker="o", s=80, facecolors="none", edgecolors="lime",
               label=f"die {K_EX} hellsten Maxima")
    ax.legend(fontsize=7, loc="lower left")
fig.tight_layout()
plt.show()
```

##### *6. Wie viele Punktquellen findet man wieder?*

Bis zu wie vielen Quellen findet man die richtigen Orte, wenn man die $k$ hellsten Maxima des Dirty Image nimmt, und wie viel hilft CLEAN?

```python
def recall(sources, img, tol_uas=8):
    # Anteil der wahren Quellen, neben denen eines der k hellsten Maxima liegt
    truth = np.array([(l, m) for _, l, m in sources])
    found = pix_to_uas(top_k_peaks(img, len(sources)))
    d = np.hypot(truth[:, None, 0] - found[None, :, 0], truth[:, None, 1] - found[None, :, 1])
    return np.mean(d.min(axis=1) < tol_uas)

K_LIST, N_TRIALS = [1, 2, 3, 5, 8, 12], 30
rec = {"dirty": [], "clean": []}
for k in K_LIST:
    r_d, r_c = [], []
    for _ in range(N_TRIALS):
        s = random_sky(k)
        dimg = dirty_image(noisy(sky_visibilities(s)))
        r_d.append(recall(s, dimg))
        r_c.append(recall(s, hogbom_clean(dimg)[2]))
    rec["dirty"].append((np.mean(r_d), np.std(r_d) / np.sqrt(N_TRIALS)))
    rec["clean"].append((np.mean(r_c), np.std(r_c) / np.sqrt(N_TRIALS)))
print(f"Trefferquote (Abstand unter 8 µas), {N_TRIALS} zufällige Himmel pro k, Rauschen {NOISE} Jy pro Visibility")
for k, (rd, ed), (rc, ec) in zip(K_LIST, rec["dirty"], rec["clean"]):
    print(f"  k = {k:2d}:  Dirty Image {rd:.2f} ± {ed:.2f}   |   nach CLEAN {rc:.2f} ± {ec:.2f}")

fig, ax = plt.subplots(figsize=(8, 4.5))
for key, mk, label in [("dirty", "o--", "k hellste Maxima des Dirty Image"), ("clean", "s-", "k hellste Maxima nach CLEAN")]:
    m_, e_ = np.array(rec[key]).T
    ax.errorbar(K_LIST, m_, e_, fmt=mk, capsize=3, label=label)
ax.set_ylim(0, 1.05)
ax.set_xlabel("Zahl der Punktquellen k")
ax.set_ylabel("Trefferquote")
ax.set_title(f"Wiederfinden von Punktquellen mit der EHT-Abdeckung ({N_TRIALS} Himmel pro k)", fontsize=10)
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

#### **Eigene Messung gegen den Katalog**

##### *7. Durchmesser und Masse im Vergleich*

Passt der gemessene Durchmesser zum veröffentlichten Ring, und welche Masse folgt daraus im Vergleich mit den Massen aus der Bewegung von Sternen und Gas?

```python
D_M87 = 16.8                                     # Mpc, wie in EHT Collaboration (2019)
def theta_g(m_sun, d_mpc=D_M87):
    return G * m_sun * MSUN / C ** 2 / (d_mpc * MPC) / UAS          # µas

alpha_eht = 42 / theta_g(6.5e9)                  # Verhältnis Ring / θ_g im veröffentlichten Ergebnis
d_err = d_days.std(ddof=1)
m_own = d_ring / alpha_eht / theta_g(1.0)
m_err = m_own * d_err / d_ring
m_shadow = d_ring / (2 * np.sqrt(27)) / theta_g(1.0)               # falls der Ring genau der Schatten wäre

masses = [("EHT Collaboration (2019), Ring", 6.5e9, "6.5 ± 0.7"),
          ("Sterne (Gebhardt et al. 2011)", 6.6e9 * D_M87 / 17.9, "6.6 ± 0.4 bei 17.9 Mpc"),
          ("Gas (Walsh et al. 2013)", 3.5e9 * D_M87 / 17.9, "3.5 (+0.9/−0.3) bei 17.9 Mpc")]
print(f"Ringdurchmesser: eigen {d_ring:.1f} ± {d_err:.1f} µas (Streuung der Nächte), EHT Collaboration (2019): 42 ± 3 µas")
print(f"θ_g für 6.5·10⁹ M_sun bei {D_M87} Mpc: {theta_g(6.5e9):.2f} µas; veröffentlichtes Verhältnis d / θ_g = {alpha_eht:.1f}, "
      f"Schatten allein 2√27 = {2 * np.sqrt(27):.1f}")
print(f"Eigene Masse mit d / θ_g = {alpha_eht:.1f}: ({m_own / 1e9:.2f} ± {m_err / 1e9:.2f})·10⁹ M_sun (nur Streuung der Nächte); "
      f"wäre der Ring der Schatten selbst: {m_shadow / 1e9:.1f}·10⁹ M_sun")
print("\nVorhergesagter Ringdurchmesser für die veröffentlichten Massen (umgerechnet auf 16.8 Mpc):")
for label, m, quote in masses:
    print(f"  {label:32s} M = {m / 1e9:.1f}·10⁹ M_sun ({quote})  ->  d = {alpha_eht * theta_g(m):.0f} µas")

fig, ax = plt.subplots(figsize=(10, 3.2))
for j, (label, m, _) in enumerate(masses):
    ax.plot(alpha_eht * theta_g(m), j, "s", ms=8, c="black")
ax.errorbar(42, 0, xerr=3, fmt="none", c="black", capsize=4)
ax.axvline(d_ring, color="tab:red", label=f"eigene Messung {d_ring:.1f} µas")
ax.set_yticks(range(len(masses)), [f"{label}: {m / 1e9:.1f}·10⁹ M_sun" for label, m, _ in masses], fontsize=8)
ax.set_ylim(-0.6, len(masses) - 0.4)
ax.set_xlabel("erwarteter Ringdurchmesser d = 11.0 θ_g [µas]")
ax.set_title("Erwarteter Ring für die veröffentlichten Massen gegen den gemessenen Ring", fontsize=10)
ax.legend(fontsize=8, loc="center left")
fig.tight_layout()
plt.show()
```

Ergebnis beim Lauf vom 2026-10-12:

- **Daten:** 51 119 Visibilities auf 21 Basislinien zwischen 7 Stationen, in den vier Nächten 12 911, 17 726, 4 977 und 15 505. Die längste Basislinie hat 8.33 Gλ, also eine Auflösung von 25 µas. Die beiden kürzesten liegen innerhalb eines Standorts (JCMT–SMA 0.1 Mλ, ALMA–APEX 1.9 Mλ); sie sehen die ganze Quelle mit dem Jet und messen 1.14 Jy. Auf den langen Basislinien bleiben im Median nur 0.09 Jy. Schon im Rohplot fallen die Amplituden bei etwa 3.6 und 8.3 Gλ fast auf null.
- **Closure-Phasen:** Die Phase der Basislinie ALMA–SMT läuft im Lauf der Nacht einmal um den ganzen Kreis. Die Closure-Phase des Dreiecks ALMA–SMT–LMT bleibt dagegen zwischen etwa +30° und +75°, im Mittel +52° mit 20° Streuung, und in allen vier Nächten bei +47° bis +58°. Sie ist weder 0° noch 180°: Das Bild ist nicht punktsymmetrisch, was zur helleren Südseite des veröffentlichten Rings passt.
- **Ringdurchmesser:** Das Ringmodell ergibt $d = 43.8$ µas mit der ersten Nullstelle bei 3.60 Gλ und einem kompakten Fluss von 0.30 Jy. Die einzelnen Nächte liegen zwischen 43.5 und 44.2 µas (Streuung 0.3 µas); der formale Fehler von 0.02 µas ist bedeutungslos. Die Breite ist kaum bestimmt (0 bis 10 µas je nach Nacht). Das Modell verfehlt die Daten dort, wo es null wird: Die gemessenen Amplituden fallen an den Nullstellen nicht ganz auf null, denn ein asymmetrischer Ring füllt sie auf.
- **Dirty Beam und Dirty Image:** Die höchste Nebenkeule des Dirty Beam außerhalb von 25 µas erreicht 65 % der Hauptkeule. Im Dirty Image des Rings sieht man nur ein dunkles Zentrum zwischen hellen Nebenkeulen; ohne Rekonstruktion ist der Ring nicht zu erkennen.
- **CLEAN:** Mit 87 Komponenten erscheint der Ring wieder, sein Zentrum ist aber zu 65 % gefüllt, weil der Restaurierungsstrahl (25 µas) so groß ist wie der Radius des Rings (22 µas). Im Beispiel mit fünf Punktquellen liegen im Dirty Image drei der fünf hellsten Maxima auf Nebenkeulen; nach CLEAN sitzen vier an echten Quellen. Zwei Quellen oben rechts liegen nur wenige µas auseinander und verschmelzen zu einem Fleck.
- **Trefferquote:** Mit den $k$ hellsten Maxima des Dirty Image findet man bei 2 Quellen 87 %, bei 3 Quellen 73 % der Orte, nach CLEAN 98 % und 94 %. Bei 5 Quellen steigt die Quote von 63 % auf 73 %, bei 12 Quellen nur von 53 % auf 61 %. CLEAN hilft bei wenigen Quellen am meisten; bei vielen begrenzen die 21 Basislinien beide Verfahren.
- **Vergleich:** Der eigene Durchmesser von 43.8 µas liegt im Fehlerbereich der EHT-Kollaboration (42 ± 3 µas). Mit dem veröffentlichten Verhältnis $d/\theta_g = 11.0$ folgt eine Masse von $6.8 \cdot 10^9\,M_\odot$ (EHT: $6.5 \pm 0.7$); die Unsicherheit steckt im Verhältnis $d/\theta_g$ aus Simulationen und in der Entfernung, nicht in der Messung. Wäre der Ring genau der Schatten, wären es $7.2 \cdot 10^9\,M_\odot$. Für die Masse aus der Bewegung der Sterne ($6.2 \cdot 10^9\,M_\odot$ bei 16.8 Mpc) wäre ein Ring von 40 µas zu erwarten, für die Masse aus der Bewegung des Gases ($3.3 \cdot 10^9\,M_\odot$) einer von 21 µas, halb so groß wie gemessen. Der Ring entscheidet also zwischen den beiden Methoden, zugunsten der Sterne.

#### **Variationen**

Jeweils zuerst die Vorhersage notieren, dann den Code ändern und prüfen:

1. **Triviales Dreieck:** Setze `TRI = ("AA", "JC", "SM")`. JCMT und SMA stehen 160 m voneinander entfernt und lösen die Quelle nicht auf. *(Vorhersage: Die Closure-Phase liegt nahe 0°, weil eine der drei Basislinien nur einen Punkt sieht; so prüft die Kollaboration die Kalibrierung.)*
2. **Kurze Basislinien im Fit:** Nimm in Schritt 3 alle Basislinien statt nur derer über 1 Gλ. *(Vorhersage: Der Fit versucht, den hohen Fluss der kurzen Basislinien zu erklären, den der großräumige Jet beiträgt, und wird schlechter.)*
3. **Mehr Rauschen:** `NOISE = 0.1`. Ab wie vielen Quellen bricht die Trefferquote ein, und hilft CLEAN noch?
4. **Größerer Schritt beim CLEAN:** `gain=0.5`. *(Vorhersage: CLEAN ist schneller, aber die Komponenten sitzen öfter an Nebenkeulen statt an echten Quellen.)*
5. **Asymmetrischer Ring:** Gib den 72 Punktquellen des Rings eine Helligkeit, die nach Süden ansteigt, etwa $1 + 0.5\sin\varphi$, berechne Closure-Phasen der Simulation für das Dreieck aus Schritt 2 und vergleiche mit den gemessenen.

#### **Nächste Schritte**

So würde eine Analyse in der Forschung weitergehen:
- **Bilder aus echten Daten:** Mit Closure-Phasen und Closure-Amplituden statt der verfälschten Phasen, mit Selbstkalibrierung (CLEAN in DIFMAP) oder mit regularisierter Maximum-Likelihood-Rekonstruktion (`eht-imaging`), die Glattheit, Sparsity und Gesamtfluss als Nebenbedingungen nimmt.
- **Geometrische Modelle mit Asymmetrie** (Ring mit Helligkeitsgefälle, Bayessche Fits mit Themis oder DPI) und Kalibrierung des Verhältnisses von Ring zu $\theta_g$ an Simulationen des einfallenden Plasmas (GRMHD).
- **Polarisation:** die Magnetfelder am Rand des Lochs (EHT 2021).
- **Mehrere Jahre:** Der Ring blieb 2018 gleich groß, der helle Teil wanderte; das trennt die Raumzeit (fest) vom Plasma (veränderlich).
- **Höhere Auflösung:** mehr Stationen und höhere Frequenzen (ngEHT), Basislinien in den Weltraum, um den dünnen Photonenring zu messen. Ein Vorschlag nutzt Quantenrepeater, um das Licht entfernter Stationen direkt zu interferieren, statt es aufzuzeichnen (Gottesman, Jennewein & Croke 2012).
