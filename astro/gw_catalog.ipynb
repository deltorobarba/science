# Der Katalog der Gravitationswellen-Ereignisse

Das Notebook lädt alle Ereignisse aus dem Gravitational Wave Open Science Center (GWOSC), wählt die bestätigten Ereignisse der Kataloge GWTC-1 bis GWTC-5 aus und untersucht Massen, Entfernungen und Signalstärken: Es teilt die Ereignisse in Paare aus Schwarzen Löchern und Neutronensternen ein, prüft, dass das Signal-Rausch-Verhältnis wie $\mathcal{M}^{5/6}/D_L$ mit Chirp-Masse und Entfernung skaliert, misst daraus die gewachsene Reichweite der Detektoren von O1 bis O4 und korrigiert die Massenverteilung mit der $1/V_{\max}$-Methode um den Auswahleffekt. Ein beliebiges Ereignis aus der Tabelle lässt sich auswählen und in den Rohdaten als Q-Transformation ansehen.

Welche Doppelsysteme haben die Detektoren bisher gefunden, und wie stark verzerrt ihre Empfindlichkeit das Bild der Massen?

#### **Hintergrund**

Seit September 2015 haben die LIGO-Detektoren in Hanford (H1) und Livingston (L1), seit 2017 auch Virgo (V1) bei Pisa und zeitweise KAGRA (K1) in Japan einige hundert Signale verschmelzender kompakter Objekte gemessen. Sie beobachten in Kampagnen: O1 (2015–2016), O2 (2016–2017), O3 (2019–2020) und O4 (ab Mai 2023). Nach jeder Kampagne veröffentlicht die LIGO-Virgo-KAGRA-Kollaboration (LVK) einen Katalog, den Gravitational-Wave Transient Catalog (GWTC), und GWOSC stellt die Parameter und Rohdaten öffentlich bereit.

Für jedes Ereignis schätzt die Kollaboration aus dem Signal die Massen der beiden Objekte, die Entfernung, die Spins und die Lage am Himmel, jeweils als Median mit 90-%-Intervall. Daraus ergeben sich drei Klassen:
- **BBH:** zwei Schwarze Löcher, der weitaus größte Teil;
- **BNS:** zwei Neutronensterne (GW170817, mit Gammablitz und Kilonova, und GW190425);
- **NSBH:** ein Neutronenstern und ein Schwarzes Loch.

Zwei Lücken in der Massenverteilung sind theoretisch erwartet. Zwischen etwa 3 und 5 Sonnenmassen fand man in Röntgendoppelsternen weder Neutronensterne noch Schwarze Löcher (untere Lücke). Und Sterne, deren Heliumkern in einem bestimmten Massenbereich liegt, werden durch Paarinstabilität vollständig zerrissen; dort sollten keine Schwarzen Löcher von etwa 50 bis 130 Sonnenmassen entstehen (obere Lücke). Einzelne Ereignisse liegen in beiden Lücken.

Ein Detektor sieht schwere Systeme viel weiter als leichte. Der Katalog enthält deshalb nicht die Häufigkeiten in der Natur, sondern die Häufigkeiten mal der Reichweite hoch drei.

#### **Herleitung**

**1. Chirp-Masse.** In der Einspiralphase hängt die Frequenzänderung des Signals nur von einer Kombination der Massen ab,
$$\dot f = \frac{96}{5}\,\pi^{8/3}\left(\frac{G\mathcal{M}}{c^3}\right)^{5/3} f^{11/3}, \qquad \mathcal{M} = \frac{(m_1 m_2)^{3/5}}{(m_1 + m_2)^{1/5}} .$$
Deshalb ist $\mathcal{M}$ die am besten gemessene Masse (Herleitung in `gw150914.ipynb`).

**2. Rotverschiebung.** In den Gleichungen der Allgemeinen Relativitätstheorie tritt die Masse nur in der Kombination $GM/c^3$ auf, einer Zeit. Die Expansion des Universums dehnt alle Zeiten des Signals um den Faktor $1+z$. Ein System der Masse $M$ bei der Rotverschiebung $z$ sieht deshalb genauso aus wie ein nahes System der Masse
$$M_{det} = (1+z)\,M_{src} .$$
Der Katalog gibt die Masse im Ruhesystem der Quelle ($M_{src}$) an; für die Signalstärke zählt die gemessene Masse $M_{det}$. Die Amplitude fällt mit der Leuchtkraftentfernung $D_L$.

**3. Das Signal-Rausch-Verhältnis.** Für einen Detektor mit der spektralen Rauschdichte $S_n(f)$ ist das optimale Signal-Rausch-Verhältnis
$$\rho^2 = 4\int_{f_{\min}}^{f_{\max}} \frac{|\tilde h(f)|^2}{S_n(f)}\,df .$$
In der Einspiralphase gibt die Näherung der stationären Phase
$$|\tilde h(f)| \propto \frac{\mathcal{M}_{det}^{5/6}}{D_L}\,f^{-7/6} .$$
Eingesetzt:
$$\rho = \frac{\mathcal{M}_{det}^{5/6}}{D_L}\,K, \qquad K \propto \sqrt{\int f^{-7/3}/S_n(f)\,df} ,$$
also $\rho\,D_L \propto \mathcal{M}_{det}^{5/6}$. Das gilt, solange die Einspiralphase durch das empfindliche Band läuft. Die Einspiralphase endet ungefähr bei der Frequenz der innersten stabilen Kreisbahn, $f_{ISCO} \approx 4400\,\text{Hz} \cdot M_\odot/M_{det}$. Für $M_{det} = 60\,M_\odot$ sind das 73 Hz, unterhalb des empfindlichsten Bereichs (100 bis 300 Hz). Bei schweren Systemen trägt dann vor allem die Verschmelzung zum Signal bei, und $\rho\,D_L$ wächst langsamer als $\mathcal{M}^{5/6}$. Lage am Himmel und Neigung der Bahn ändern $\rho$ zusätzlich um einen Faktor von etwa zwei.

**4. Reichweite der Detektoren.** Ein Ereignis gilt ungefähr ab $\rho \approx 8$ als nachweisbar. Bei gegebener Masse ist die Reichweite also proportional zu $K$, und das Verhältnis der $K$-Werte zweier Kampagnen ist das Verhältnis ihrer Reichweiten. Wir passen
$$\log_{10}(\rho\,D_L) = s\,\log_{10}\mathcal{M}_{det} + a_{\text{Kampagne}}$$
an und lesen die Reichweite relativ zu O3 als $10^{a - a_{O3}}$ ab.

**5. Der Auswahleffekt und die $1/V_{\max}$-Methode.** Das Ereignis $i$ wäre bis zur Entfernung $D_{\max,i} = D_{L,i}\,\rho_i/8$ nachweisbar gewesen, also in einem Volumen $V_{\max,i} \propto D_{\max,i}^3$. Schwere Systeme werden in einem viel größeren Volumen gefunden: $V_{\max} \propto \mathcal{M}^{5/2}$ in der Einspiralnäherung. Gewichtet man jedes Ereignis mit $1/V_{\max,i}$ (Schmidt 1968), schätzt die gewichtete Verteilung die Häufigkeit pro Volumen. Die Methode nimmt einen euklidischen Raum und gleiche Beobachtungszeiten an; bei $z \sim 1$ wächst das Volumen langsamer als $D^3$, die Korrektur ist dort nur grob.

```python
import os, json, urllib.request
from pathlib import Path

# Downloads landen im Ordner "astrodata" neben dem Notebook; vor dem ersten Import von astropy setzen
ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "astropy").mkdir(parents=True, exist_ok=True)
(ASTRODATA / "gwosc").mkdir(parents=True, exist_ok=True)
os.environ["XDG_CACHE_HOME"] = str(ASTRODATA)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from astropy.time import Time

REFRESH = False                                  # True: Katalog neu von GWOSC laden
CONFIDENT = ["GWTC-5.0", "GWTC-4.1", "GWTC-4.0", "GWTC-3-confident", "GWTC-2.1-confident", "GWTC-1-confident",
             "O4_Discovery_Papers", "O3_Discovery_Papers"]          # Reihenfolge: neuester Katalog zuerst

path = ASTRODATA / "gwosc" / "allevents.json"
if REFRESH or not path.exists():
    urllib.request.urlretrieve("https://gwosc.org/eventapi/json/allevents/", path)
raw = json.loads(path.read_text())["events"]
COLS = {"commonName": "name", "version": "version", "catalog.shortName": "katalog", "GPS": "gps",
        "mass_1_source": "m1", "mass_2_source": "m2", "chirp_mass_source": "mc", "total_mass_source": "m_ges",
        "final_mass_source": "m_end", "network_matched_filter_snr": "snr", "luminosity_distance": "dl",
        "redshift": "z", "chi_eff": "chi_eff", "far": "far", "p_astro": "p_astro"}
allev = pd.DataFrame([{v: e.get(k) for k, v in COLS.items()} for e in raw.values()])
print(f"{len(allev)} Einträge, {allev.name.nunique()} verschiedene Ereignisse")
print(allev.katalog.value_counts().to_string())
```

#### **Reduktion**

##### *1. Den Katalog bereinigen*

Wie viele bestätigte Ereignisse bleiben, wie verteilen sie sich auf die Kampagnen, und für wie viele gibt es Parameter?

```python
cat = allev[allev.katalog.isin(CONFIDENT)].copy()
cat["prio"] = cat.katalog.map({k: i for i, k in enumerate(CONFIDENT)})
cat = cat.sort_values(["prio", "version"], ascending=[True, False]).drop_duplicates("name").drop(columns="prio")
cat["jahr"] = Time(cat.gps.values.astype(float), format="gps").decimalyear
cat["kampagne"] = pd.cut(cat.jahr, [2015, 2016.5, 2018, 2021, 2030], labels=["O1", "O2", "O3", "O4"])
cat["typ"] = np.select([cat.m1.isna(), cat.m1 < 3, cat.m2 < 3], ["ohne Parameter", "BNS", "NSBH"], "BBH")
cat["mc_det"] = cat.mc * (1 + cat.z)
cat = cat.sort_values("gps").reset_index(drop=True)

print(f"Bestätigte Ereignisse: {len(cat)}, davon mit Parametern {cat.m1.notna().sum()}")
print(pd.crosstab(cat.kampagne, cat.typ, margins=True, margins_name="Summe").to_string())
print("\nNeuester Katalog je Ereignis:", cat.katalog.value_counts().to_dict())
nopar = cat[cat.typ == "ohne Parameter"]
print(f"Ohne Parameter: {len(nopar)} Kandidaten aus {sorted(nopar.katalog.unique())}, "
      f"Fehlalarmrate Median {nopar.far.median():.2g} pro Jahr (mit Parametern: {cat[cat.m1.notna()].far.median():.2g})")
ev = cat[cat.m1.notna()].copy()                     # ab hier nur Ereignisse mit Parametern
```

##### *5. Ein Ereignis aus dem Katalog ansehen*

Wie sieht das Signal eines frei gewählten Ereignisses in den Rohdaten aus? Der Name kommt aus den Tabellen oben; voreingestellt ist das lauteste Ereignis bisher, GW250114.

```python
from gwosc.datasets import event_gps, event_detectors
from gwpy.timeseries import TimeSeries

SELECTED_EVENT = "GW250114_082203"               # Name aus der Tabelle, z. B. "GW150914", "GW190521" oder "GW170817"
PADDING = 16                                     # Sekunden Rohdaten vor und nach dem Ereignis
row = cat[cat.name == SELECTED_EVENT].iloc[0]
gps = event_gps(SELECTED_EVENT)
dets = sorted(event_detectors(SELECTED_EVENT))
DET = "H1" if "H1" in dets else dets[0]          # L1 hatte bei GW170817 eine Störung (Glitch) kurz vor dem Signal
print(f"{SELECTED_EVENT} ({row.katalog}): GPS {gps}, Detektoren {dets}, verwendet {DET}")
print(f"m1 = {row.m1:.2f}, m2 = {row.m2:.2f} M☉, Chirp-Masse {row.mc:.3f} M☉ (im Detektor {row.mc_det:.3f}), "
      f"D_L = {row.dl:.0f} Mpc, SNR = {row.snr:.1f}")

# Dauer im Band ab 30 Hz aus der Herleitung (Newtonsche Näherung, Integration von df/dt)
G_C3 = 4.925e-6                                  # G M☉ / c³ in Sekunden
tau30 = 5 / 256 * (np.pi * 30) ** (-8 / 3) * (G_C3 * row.mc_det) ** (-5 / 3)
print(f"Zeit von 30 Hz bis zur Verschmelzung: {tau30:.2f} s")

strain = TimeSeries.fetch_open_data(DET, gps - PADDING, gps + PADDING, sample_rate=4096, cache=True)
white = strain.whiten()
window = min(max(2 * tau30, 0.5), PADDING - 2)
q = white.q_transform(qrange=(4, 64), frange=(20, 512), outseg=(gps - window, gps + 0.3))
plot = q.plot(figsize=(12, 4.5))
ax = plot.gca()
ax.set_xscale("seconds", epoch=gps)
ax.set_yscale("log")
ax.set_ylim(20, 512)
ax.set_ylabel("Frequenz [Hz]")
ax.set_xlabel(f"Zeit relativ zum Ereignis [s]")
ax.set_title(f"{SELECTED_EVENT}: Q-Transformation der weißgerechneten Daten ({DET})")
ax.colorbar(label="normierte Energie", clim=(0, 25))
plot.show()
plt.style.use("default")                         # gwpy stellt beim Import den Plotstil um; zurücksetzen
```

#### **Modell und Fit**

##### *6. Signalstärke, Chirp-Masse und die Reichweite der Detektoren*

Wächst $\rho\,D_L$ wie $\mathcal{M}_{det}^{5/6}$, und um wie viel ist die Reichweite der Detektoren von O1 bis O4 gewachsen?

```python
RUNS = ["O1", "O2", "O3", "O4"]
def fit_scaling(d):
    X = np.column_stack([np.log10(d.mc_det)] + [(d.kampagne == r).astype(float) for r in RUNS])
    y = np.log10(d.snr * d.dl)
    coef, res, *_ = np.linalg.lstsq(X, y, rcond=None)
    resid = y - X @ coef
    cov = np.linalg.inv(X.T @ X) * resid.var(ddof=X.shape[1])
    return coef, np.sqrt(np.diag(cov)), resid.std()

g = ev.dropna(subset=["mc_det", "dl", "snr"])
for lim in (10, 20, 1000):
    d = g[g.mc_det < lim]
    coef, err, sd = fit_scaling(d)
    print(f"M_det < {lim:>4} M☉ (n = {len(d):>3}): Steigung s = {coef[0]:.2f} ± {err[0]:.2f}  (erwartet 5/6 = 0.83), "
          f"Streuung ±{sd:.2f} dex")
coef, err, sd = fit_scaling(g[g.mc_det < 20])
reach = {r: 10 ** (coef[i + 1] - coef[3]) for i, r in enumerate(RUNS)}
print("\nReichweite relativ zu O3 (aus den Ereignissen mit M_det < 20 M☉): "
      + ", ".join(f"{r} {v:.2f}" for r, v in reach.items()))

fig, ax = plt.subplots(figsize=(9, 5))
for r, col in zip(RUNS, ["tab:purple", "tab:green", "tab:blue", "tab:orange"]):
    s = g[g.kampagne == r]
    ax.scatter(s.mc_det, s.snr * s.dl, s=12, c=col, label=r)
    xx = np.logspace(0, 2.3, 20)
    ax.plot(xx, 10 ** (coef[0] * np.log10(xx) + coef[RUNS.index(r) + 1]), c=col, lw=1)
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Chirp-Masse im Detektor M_det [M☉]")
ax.set_ylabel("ρ · D_L [Mpc]")
ax.set_title(f"ρ·D_L ∝ M^s mit s = {coef[0]:.2f} (Fit für M_det < 20 M☉, verlängert)")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

##### *7. Der Auswahleffekt*

Wie sieht die Verteilung der Massen der Schwarzen Löcher aus, wenn man jedes Ereignis mit $1/V_{\max}$ gewichtet?

```python
bbh = ev[ev.typ == "BBH"].dropna(subset=["dl", "snr"]).copy()
bbh["d_max"] = bbh.dl * bbh.snr / 8
bbh["w"] = 1 / bbh.d_max ** 3
bins = np.array([5, 7, 9, 12, 15, 20, 25, 30, 35, 40, 50, 65, 80, 150])
obs, _ = np.histogram(bbh.m1, bins)
wgt, _ = np.histogram(bbh.m1, bins, weights=bbh.w)
width = np.diff(bins)
tab = pd.DataFrame({"m1": [f"{a}-{b}" for a, b in zip(bins[:-1], bins[1:])], "beobachtet": obs,
                    "Anteil beobachtet": obs / obs.sum(), "Anteil gewichtet": wgt / wgt.sum(),
                    "gewichtet pro M☉": wgt / wgt.sum() / width})
print(tab.round(4).to_string(index=False))
print(f"\nAnteil m1 < 15 M☉: beobachtet {np.mean(bbh.m1 < 15):.0%}, gewichtet {bbh.w[bbh.m1 < 15].sum() / bbh.w.sum():.0%}")
print(f"Reichweite D_max: Median {bbh.d_max.median():.0f} Mpc; für m1 < 15 M☉ {bbh.d_max[bbh.m1 < 15].median():.0f} Mpc, "
      f"für m1 > 30 M☉ {bbh.d_max[bbh.m1 > 30].median():.0f} Mpc")

fig, ax = plt.subplots(figsize=(9, 4.2))
centers = np.sqrt(bins[:-1] * bins[1:])
ax.step(bins, np.r_[obs / obs.sum() / width, 0], where="post", c="0.5", label="beobachtet")
ax.step(bins, np.r_[wgt / wgt.sum() / width, 0], where="post", c="tab:red", label="mit 1/V_max gewichtet")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("m1 [M☉]")
ax.set_ylabel("Anteil pro M☉")
ax.set_title("Massenverteilung der Schwarzen Löcher: beobachtet und um den Auswahleffekt korrigiert")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

##### *8. Vergleich mit den Katalogwerten und der Literatur*

Stimmt die selbst berechnete Chirp-Masse mit dem Katalog überein, und passen Reichweiten und gewichtete Massenverteilung zu den Angaben der Kollaboration?
