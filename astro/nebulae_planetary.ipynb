#### **Theorie**

##### *1. Was ein Planetarischer Nebel ist*

Ein Stern mit 0.8 bis 8 Sonnenmassen endet nicht als Supernova. Nach dem Wasserstoff- und dem Heliumbrennen im Kern bläht er sich auf dem **asymptotischen Riesenast** (AGB) zu einem Roten Riesen von einigen hundert Sonnenradien auf und verliert in einem langsamen Wind (10 bis 20 km/s) den größten Teil seiner Hülle. Übrig bleibt ein Kern aus Kohlenstoff und Sauerstoff von 0.55 bis 0.9 Sonnenmassen, umgeben von einer dünnen Schicht Wasserstoff, in der das Brennen weitergeht.

Ist die Hülle fast ganz verloren, schrumpft der Kern und wird bei fast gleicher Leuchtkraft immer heißer, von 5000 K auf über 100 000 K, in wenigen tausend bis zehntausend Jahren. Ab etwa 25 000 K sendet er genug Photonen über 13.6 eV aus, um das abgestoßene Gas zu ionisieren: Es leuchtet in Emissionslinien, als **Planetarischer Nebel** (der Name stammt von William Herschel, dem die runden Scheibchen im Fernrohr wie Planeten erschienen). Ein schneller Wind des heißen Sterns (über 1000 km/s) schiebt das Gas dabei zu einer Schale zusammen, die mit 10 bis 40 km/s wächst.

Nach etwa 20 000 bis 50 000 Jahren ist die Schale so groß und dünn, dass sie im Gas der Milchstraße verschwimmt; der Stern hat seine Wasserstoffschale aufgebraucht und kühlt als **Weißer Zwerg** ab. Ein Planetarischer Nebel ist also eine Momentaufnahme: Stern und Nebel sind beide Uhren für dieselbe kurze Phase.

**Fachbegriffe:**
- **Zentralstern:** der freigelegte Kern im Zentrum des Nebels, im Englischen CSPN (central star of a planetary nebula).
- **AGB:** asymptotic giant branch, die letzte Phase der Riesensterne mit zwei brennenden Schalen (Wasserstoff und Helium) um einen entarteten Kern.

##### *2. Rekombination, Fall B und Rötung*

Im Nebel stellt sich ein Gleichgewicht ein: Jede Sekunde ionisiert der Stern $Q(\mathrm{H^0})$ Wasserstoffatome, und genauso viele Protonen fangen wieder ein Elektron ein. Ist der Nebel optisch dick, gilt

$$Q(\mathrm{H^0}) = \int_{\nu_0}^{\infty} \frac{L_\nu}{h\nu}\, d\nu = \alpha_B \, n_e n_p V,$$

mit der Ionisationskante $\nu_0$ (13.6 eV), dem Volumen $V$ und dem Rekombinationskoeffizienten im **Fall B**, $\alpha_B = 2.54\cdot10^{-13}$ cm³ s⁻¹ bei 10 000 K. Fall B heißt: Rekombinationen direkt in den Grundzustand zählen nicht, weil ihr Photon sofort ein anderes Atom ionisiert.

Ein Teil der Rekombinationen läuft über den Übergang n = 4 → 2 und erzeugt ein Hβ-Photon. Mit dem effektiven Koeffizienten $\alpha_\text{eff}(\mathrm{H}\beta)$ ist

$$L(\mathrm{H}\beta) = h\nu_{\mathrm{H}\beta}\,\alpha_\text{eff}(\mathrm{H}\beta)\, n_e n_p V = h\nu_{\mathrm{H}\beta}\,\frac{\alpha_\text{eff}(\mathrm{H}\beta)}{\alpha_B}\, Q(\mathrm{H^0}).$$

Bei 10 000 K ist $\alpha_\text{eff}/\alpha_B \approx 0.12$: Etwa jede achte Rekombination liefert ein Hβ-Photon. Die Emissivitäten stammen aus den Tabellen von Storey & Hummer (1995), die auch $\mathrm{H}\alpha/\mathrm{H}\beta = 2.86$ liefern.

Staub schwächt Licht der Wellenlänge λ um $A_\lambda = k(\lambda)\,E(B-V)$ Magnituden, blaues stärker als rotes ($k(\mathrm{H}\beta) = 3.61$, $k(\mathrm{H}\alpha) = 2.53$ für $R_V = 3.1$). Aus dem beobachteten Verhältnis folgt

$$E(B-V) = \frac{2.5}{k(\mathrm{H}\beta) - k(\mathrm{H}\alpha)}\,\log_{10}\frac{(\mathrm{H}\alpha/\mathrm{H}\beta)_\text{beob}}{2.86}, \qquad c(\mathrm{H}\beta) = 0.4\,k(\mathrm{H}\beta)\,E(B-V),$$

und jeder Fluss wird mit $10^{0.4\,k(\lambda)E(B-V)}$ entrötet.

**Fachbegriffe:**
- **Rekombination:** das Einfangen eines freien Elektrons durch ein Ion; das Elektron fällt dann Stufe für Stufe in den Grundzustand und sendet dabei Linien aus.
- **Emissivität:** die Leistung, die ein Kubikzentimeter Gas in einer Linie abstrahlt, geteilt durch $n_e n_p$.

##### *3. Die Zanstra-Methode*

Teilt man den Hβ-Fluss des Nebels durch den Fluss des Sterns bei der Wellenlänge des V-Filters, fallen Entfernung und Sternradius heraus:

$$\frac{F(\mathrm{H}\beta)}{F_\nu(V)} = \nu_{\mathrm{H}\beta}\,\frac{\alpha_\text{eff}(\mathrm{H}\beta)}{\alpha_B}\;\frac{\int_{\nu_0}^{\infty} B_\nu(T)/\nu \; d\nu}{B_\nu(\nu_V, T)}.$$

Rechts steht nur noch die Temperatur $T$ des Sterns, wenn man sein Spektrum als Schwarzkörper $B_\nu(T)$ annimmt. Der Zähler zählt die ionisierenden Photonen, der Nenner das sichtbare Licht; je heißer der Stern, desto mehr seiner Strahlung liegt jenseits der Kante, und desto größer ist das Verhältnis. Bei sehr hohen Temperaturen liegt V im Rayleigh-Jeans-Teil des Spektrums ($B_\nu \propto T$), die Photonenzahl wächst wie $T^3$, das Verhältnis also wie $T^2$.

Mit He II 4686 und der Kante bei 54.4 eV (vierfache Frequenz, weil He⁺ die Kernladung 2 hat) gibt dieselbe Rechnung eine zweite Temperatur. He⁺ ist wasserstoffähnlich: $\alpha_B(\mathrm{He^+}, T) = 2\,\alpha_B(\mathrm{H}, T/4)$.

Die Methode setzt voraus, dass der Nebel alle ionisierenden Photonen schluckt. Ist er optisch dünn, gibt sie eine **untere Grenze** der Temperatur.

**Fachbegriffe:**
- **Schwarzkörper:** ein idealer Strahler, dessen Spektrum nur von der Temperatur abhängt (Planck-Funktion $B_\nu(T)$).
- **Ionisationskante:** die kleinste Photonenenergie, die ein Atom oder Ion ionisieren kann: 13.6 eV für H, 24.6 eV für He, 54.4 eV für He⁺.

##### *4. Leuchtkraft eines heißen Sterns*

Der Fluss im V-Band, $F_\nu(V) = 3.63\cdot10^{-20}\,\text{erg s}^{-1}\text{cm}^{-2}\text{Hz}^{-1} \cdot 10^{-0.4\,V_0}$ für die entrötete Helligkeit $V_0$, legt zusammen mit der Temperatur das ganze Schwarzkörperspektrum fest. Die gesamte Leuchtkraft ist

$$L = 4\pi d^2 \, F_\nu(V)\,\frac{\sigma T^4/\pi}{B_\nu(\nu_V, T)}.$$

Der Bruch ist die **bolometrische Korrektur** in linearer Form: wie viel mehr Licht der Stern insgesamt abgibt als im V-Band. Bei 100 000 K ist sie riesig, weil fast alles im Ultravioletten liegt; ein Fehler in $T$ geht deshalb etwa mit der dritten Potenz in $L$ ein.

##### *5. Parallaxe und Entfernung*

Die Parallaxe $\varpi$ in Millibogensekunden gibt die Entfernung $d = 1000/\varpi$ pc. Gaia-EDR3-Parallaxen haben einen Nullpunktfehler: Quasare, die keine messbare Parallaxe haben, zeigen im Mittel $-0.017$ mas (Lindegren et al. 2021). Korrigiert wird $\varpi \to \varpi + 0.017$ mas.

Der Kehrwert einer verrauschten Größe ist verzerrt: Bei relativem Fehler $f = \sigma_\varpi/\varpi$ gibt eine um $-f$ zu kleine Parallaxe eine um $f/(1-f)$ zu große Entfernung, eine um $+f$ zu große nur eine um $f/(1+f)$ zu kleine. Unterhalb von etwa S/N = 5 braucht man deshalb eine Annahme über die Verteilung der Entfernungen (einen Prior), wie es Chornay & Walton (2021) mit der Flächenhelligkeit des Nebels tun.

##### *6. Das kinematische Alter*

Der Radius des Nebels in Parsec ist $R = \theta\, d$ mit dem Winkelradius $\theta$ im Bogenmaß. Wächst die Schale seit ihrer Entstehung mit der Geschwindigkeit $v$, ist ihr Alter

$$t_\text{kin} = \frac{R}{v}, \qquad t_\text{kin} = 978\,\text{Jahre}\cdot\frac{R/0.001\,\text{pc}}{v/1\,\text{km s}^{-1}} .$$

Die Geschwindigkeit aus der Linienaufspaltung misst die Materie, die in [O III] leuchtet, also eher innen; der Rand bewegt sich oft schneller, und die Schale wird mit der Zeit beschleunigt. Das kinematische Alter ist deshalb nur auf etwa einen Faktor 2 genau (Schönberner et al. 2005).

##### *7. Entwicklung des Zentralsterns*

Auf dem waagrechten Ast im HR-Diagramm brennt Wasserstoff in einer dünnen Schale auf dem Kern. Die Leuchtkraft hängt dabei fast nur von der Masse des Kerns ab, die **Kernmasse-Leuchtkraft-Beziehung** (Paczyński 1970):

$$\frac{L}{L_\odot} \approx 59\,250\,\left(\frac{M_c}{M_\odot} - 0.495\right).$$

Ein Kern von 0.6 M☉ leuchtet also mit etwa 6000 L☉. Die Zeit bis zum Knie ist die Masse der restlichen Wasserstoffschale geteilt durch die Rate, mit der sie verbrennt. Schwere Kerne haben dünnere Schalen und verbrennen sie schneller; die Zeitskala fällt deshalb extrem steil mit der Masse: von etwa 10⁴ Jahren bei 0.57 M☉ auf einige 10² Jahre bei 0.9 M☉. Wer aus der Lage im HR-Diagramm ein Alter ablesen will, muss die Masse sehr genau kennen.

Vassiliadis & Wood (1994) haben solche Wege für Sterne mit 1 bis 5 Sonnenmassen gerechnet; die Zeit zählt ab $\log T = 4$. Neuere Rechnungen (Miller Bertolami 2016) gehen 3- bis 10-mal schneller, vor allem für leichte Kerne.

**Fachbegriffe:**
- **Kernmasse:** Masse des Kohlenstoff-Sauerstoff-Kerns, die später die Masse des Weißen Zwergs ist; Weiße Zwerge der Milchstraße haben im Mittel 0.6 M☉.
- **Knie:** der Punkt höchster Temperatur, an dem das Schalenbrennen erlischt und der Stern bei fallender Leuchtkraft abkühlt.

##### *8. Die [O III]-Leuchtkraftfunktion*

In einem jungen, dichten Nebel kühlt das Gas vor allem über die Linie [O III] 5007; bis zu 10 % der Energie des Sterns gehen in diese eine Linie. Ihre Helligkeit wird in Magnituden angegeben (Jacoby 1989):

$$m_{5007} = -2.5\,\log_{10} F(5007) - 13.74 \qquad (F \text{ in erg s}^{-1}\,\text{cm}^{-2}).$$

Zählt man in einer Galaxie die Nebel nach ihrer Helligkeit, ergibt sich die **Leuchtkraftfunktion** (PNLF), nach Ciardullo et al. (1989)

$$N(M) \propto e^{0.307 M}\left(1 - e^{3(M^\star - M)}\right) \quad \text{für } M > M^\star .$$

Der erste Faktor beschreibt Nebel, deren Helligkeit langsam abnimmt; der zweite schneidet die Verteilung bei $M^\star \approx -4.5$ scharf ab. Warum dieser Abbruch in alten elliptischen und jungen Spiralgalaxien fast gleich liegt, ist nicht ganz verstanden; empirisch ist $M^\star$ an Galaxien mit Cepheiden-Entfernung geeicht, $M^\star = -4.53 \pm 0.06$ (Ciardullo 2012). Misst man in einer Galaxie die Abbruchhelligkeit $m^\star$, folgt der **Entfernungsmodul**

$$\mu = m^\star - A_{5007} - M^\star, \qquad d = 10^{\mu/5 + 1}\ \text{pc}.$$

Mit einigen Dutzend Nebeln im hellsten Magnitudenbereich reicht das bis etwa 20 Mpc, zum Virgo- und Fornax-Haufen.

##### *9. Maximum Likelihood mit Fremdanteil*

Gegeben die Helligkeiten $m_i$ zwischen $m_\text{lo}$ und $m_\text{hi}$, ist die Wahrscheinlichkeitsdichte des Modells

$$p(m \mid m^\star, f) = (1 - f)\,\frac{(N \ast G_\sigma)(m)}{\int_{m_\text{lo}}^{m_\text{hi}} (N \ast G_\sigma)\,dm} + \frac{f}{m_\text{hi} - m_\text{lo}},$$

die PNLF $N$ gefaltet mit einer Gauß-Kurve $G_\sigma$ für den Messfehler plus einem gleichverteilten Anteil $f$ für Fremdobjekte (Hintergrundgalaxien, H-II-Regionen, falsch photometrierte Nebel). Gesucht ist das Maximum von $\ln L = \sum_i \ln p(m_i)$. Ohne den Fremdanteil wäre ein einziges zu helles Objekt fatal: Die Dichte ist oberhalb der Kante null, $\ln L$ würde $-\infty$, und der Fit müsste $m^\star$ unter dieses Objekt schieben.

Den Fehler von $m^\star$ gibt die **Profil-Likelihood**: Für jeden festen Wert von $m^\star$ wird $f$ optimiert; wo $\ln L$ um 0.5 unter dem Maximum liegt, ist der 1σ-Rand. Für Leser aus dem maschinellen Lernen: Das ist ein Mischmodell aus Signal und Ausreißern mit einem Störparameter, der herausoptimiert wird.

```python
import io
import os
import re
import urllib.request
from pathlib import Path

# Downloads landen im Ordner "astrodata" neben dem Notebook statt in den Caches im Home-Verzeichnis.
# Das muss vor dem ersten Import von astropy passieren; sonst den Kernel neu starten.
ASTRODATA = Path.cwd() / "astrodata"
(ASTRODATA / "astropy").mkdir(parents=True, exist_ok=True)
os.environ["XDG_CACHE_HOME"] = str(ASTRODATA)

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
import matplotlib.path as mpath
from scipy import integrate, optimize
from scipy.interpolate import griddata
from scipy.ndimage import gaussian_filter1d
from astropy.io import fits

SNR_PLX = 5.0          # Mindest-S/N der Parallaxe (Schritt 3)
PLX_ZP = -0.017        # Nullpunkt der Gaia-Parallaxen [mas]: wahre = gemessene − PLX_ZP (Lindegren et al. 2021)
RUWE_MAX = 1.4         # darüber ist die Astrometrie eines Einzelsterns unzuverlässig
REL_MIN = 0.8          # Mindestzuverlässigkeit der Zuordnung Gaia-Stern = Zentralstern (Chornay & Walton 2021)
R_V = 3.1              # Staubgesetz von Cardelli, Clayton & Mathis (1989) mit R_V = A_V / E(B−V) (Schritt 4)
T_NEB = 1.0e4          # Elektronentemperatur der Nebel [K] für die Rekombinationskoeffizienten (Schritt 6)
LAM_V = 5480.0         # effektive Wellenlänge des V-Filters [Å]
FNU_V0 = 3.63e-20      # F_ν eines Sterns mit V = 0 [erg s⁻¹ cm⁻² Hz⁻¹] (Bessell et al. 1998)
Z_TRACKS = 0.016       # Metallizität der Entwicklungswege (Vassiliadis & Wood 1994), etwa solar
M_CORE = {1.0: 0.569, 1.5: 0.597, 2.0: 0.633, 2.5: 0.677, 3.5: 0.754, 5.0: 0.900}   # Anfangs- -> Kernmasse [M☉]
R0_PC = 8178.0         # Entfernung zum Galaktischen Zentrum [pc] (GRAVITY Collaboration 2019) (Schritt 10)
C_BULGE_MAX = 1.5      # nur Bulge-Nebel mit mäßiger Rötung, c(Hβ) < 1.5 (Schritt 10)
SIG_BULGE = 0.3        # Streuung von M5007 im Bulge [mag]: Rötungsfehler und Tiefe des Bulge
M_STAR_LIT = -4.53     # Abbruchhelligkeit der PNLF, an Cepheiden geeicht (Ciardullo 2012)
EBV_M31 = 0.057        # Vordergrundrötung vor M 31 (Schlafly & Finkbeiner 2011, Mittel 2.5° bis 3.3° um M 31)
SIG_C89, SIG_M06 = 0.07, 0.15   # Messfehler von m5007 bei Ciardullo et al. und Merrett et al. [mag] (Schritt 11)
M5007_LIM = {"Ciardullo 1989": 22.0, "Merrett 2006": 22.5}   # Grenze der Vollständigkeit [mag] (Schritt 12)
rng = np.random.default_rng(42)
pd.set_option("display.width", 160)
np.seterr(divide="ignore", invalid="ignore")      # log von fehlenden Werten -> NaN

H_PLANCK, C_CGS, K_B, SIGMA_SB = 6.62607e-27, 2.99792458e10, 1.380649e-16, 5.6704e-5
L_SUN, PC, YEAR, KM = 3.828e33, 3.0857e18, 3.156e7, 1e5
NU_H = 13.598 * 1.602177e-12 / H_PLANCK          # Ionisationskante des Wasserstoffs (91.2 nm) [Hz]
NU_HE2 = 4 * NU_H                                # Ionisationskante von He⁺ (54.4 eV, 22.8 nm) [Hz]

VIZIER = ["https://vizier.cds.unistra.fr", "https://vizier.cfa.harvard.edu", "http://vizier.nao.ac.jp"]
PYNEB = "https://raw.githubusercontent.com/Morisset/PyNeb_devel/master/pyneb/atomic_data_fits/"
PS1 = "https://ps1images.stsci.edu/cgi-bin/"


def download(urls, path, check=None, timeout=60, rounds=2):
    # lädt eine Datei vom ersten Server, der antwortet, und liest sie danach aus dem Cache;
    # die Spiegel fallen manchmal für Minuten aus, deshalb wird die Liste zweimal durchlaufen
    path = Path(path)
    if path.exists():
        return path
    path.parent.mkdir(parents=True, exist_ok=True)
    for url in urls * rounds:
        try:
            with urllib.request.urlopen(url, timeout=timeout) as r:
                data = r.read()
        except OSError:
            continue
        if check is None or check(data):
            path.write_bytes(data)
            return path
    raise OSError(f"Kein Server liefert {path.name}")


def vizier_ok(d):
    # eine Strichzeile unter dem Spaltenkopf und danach Daten; ein gestörter Spiegel liefert nur Fehlermeldungen
    lines = d.splitlines()
    dash = [i for i, line in enumerate(lines) if line.startswith(b"---")]
    return bool(dash) and any(line.strip() and not line.startswith(b"#") for line in lines[dash[0] + 1:])


def vizier_table(source, numeric=()):
    # eine VizieR-Tabelle als Text; Zahlenspalten werden umgewandelt, leere Felder werden NaN
    query = f"/viz-bin/asu-tsv?-source={source}&-out.max=unlimited&-out.all"
    p = download([m + query for m in VIZIER], ASTRODATA / "vizier" / (source.replace("/", "_") + ".tsv"), vizier_ok,
                 timeout=30)
    lines = [l for l in p.read_text().splitlines(keepends=True) if l.strip() and not l.startswith("#")]
    df = pd.read_csv(io.StringIO("".join(lines)), sep="\t", skiprows=[1, 2], dtype=str)
    df.columns = [c.strip() for c in df.columns]
    df = df.apply(lambda s: s.str.strip())
    for c in numeric:
        df[c] = pd.to_numeric(df[c], errors="coerce")
    return df


def ps1_color(ra, dec, size_arcsec, path, stretch=""):
    # Farbbild aus i, r, g (rot, grün, blau), serverseitig auf höchstens 800 Pixel verkleinert
    lst = download([f"{PS1}ps1filenames.py?ra={ra}&dec={dec}&filters=gri"], path.with_suffix(".txt"))
    files = pd.read_csv(lst, sep=r"\s+").set_index("filter")["filename"]
    size = int(size_arcsec / 0.25)                                       # 0.25″ pro Pixel
    url = (f"{PS1}fitscut.cgi?ra={ra}&dec={dec}&size={size}&output_size={min(size, 800)}&format=jpg"
           f"&red={files['i']}&green={files['r']}&blue={files['g']}{stretch}")
    return mpimg.imread(download([url], path, check=lambda d: d[:3] == b"\xff\xd8\xff", timeout=300))


# Zentralsterne in Gaia EDR3
CW = vizier_table("J/A+A/656/A110/tablea1", ["RA_ICRS", "DE_ICRS", "Rel", "BP-RP", "Gmag", "plx", "e_plx", "RUWE",
                                             "pnRad", "rcomb", "b_rcomb", "B_rcomb"])
# Strasbourg-ESO-Katalog: Namen, Durchmesser, Hβ-Flüsse, Linien, Geschwindigkeiten, Zentralsterne
V84 = {"main": vizier_table("V/84/main"),
       "diam": vizier_table("V/84/diam", ["oDiam"]),
       "hbeta": vizier_table("V/84/hbeta", ["log(Fbeta)"]),
       "intens": vizier_table("V/84/intens", ["I4686", "I5007", "I6563"]),
       "vel": vizier_table("V/84/vel", ["ExpVel"]),
       "cstar": vizier_table("V/84/cstar", ["Vmag"])}
# Entwicklungswege, Literaturwerte der Zentralsterne, Nebel in M 31
VW = vizier_table("J/ApJS/92/125/table3", ["M", "Z", "Time", "log(Teff)", "log(L)"])
WEID = (vizier_table("J/A+A/640/A10/ph-para", ["logTeff", "logL"])
        .merge(vizier_table("J/A+A/640/A10/catalog")[["PNG", "SpType1"]].drop_duplicates("PNG"), on="PNG", how="left")
        .drop_duplicates("PNG"))
C89 = vizier_table("J/ApJ/339/53/table2", ["PN", "m5007", "e_m5007"])
M06 = vizier_table("J/MNRAS/369/120/table3", ["m5007", "HV", "C89"])
# Rekombination im Fall B (Storey & Hummer 1995), Zeile 0 = Elektronendichte 100 cm⁻³
fits_ok = lambda d: d[:6] == b"SIMPLE"
SH_H = fits.open(download([PYNEB + "h_i_rec_SH95.fits"], ASTRODATA / "pyneb" / "h_i_rec_SH95.fits", fits_ok))[1].data
SH_HE = fits.open(download([PYNEB + "he_ii_rec_SH95.fits"], ASTRODATA / "pyneb" / "he_ii_rec_SH95.fits", fits_ok))[1].data

for name, df in [("Chornay & Walton 2021", CW), ("Strasbourg-ESO: Nebel", V84["main"]),
                 ("  mit Durchmesser", V84["diam"].dropna(subset=["oDiam"])), ("  mit Hβ-Fluss", V84["hbeta"]),
                 ("  mit Linienstärken", V84["intens"]), ("  mit Geschwindigkeiten", V84["vel"]),
                 ("  mit Zentralstern", V84["cstar"]), ("Vassiliadis & Wood 1994", VW), ("Weidmann et al. 2020", WEID),
                 ("Ciardullo et al. 1989", C89), ("Merrett et al. 2006", M06)]:
    print(f"{name:26s} {len(df):5d} Zeilen")
```

**Werkzeuge 1: Rötung, Rekombination, Zanstra-Methode und Leuchtkraft** (Theorie, Abschnitte 2 bis 5). `deredden` rechnet aus dem Balmer-Dekrement die Rötung und korrigiert Hβ, [O III] 5007 und He II 4686; `zanstra_T` kehrt die Zanstra-Beziehung für H oder He II um; `luminosity` rechnet aus V, Entfernung und Temperatur die Leuchtkraft eines Schwarzkörpers.

```python
def k_ccm(lam, Rv=R_V):
    # Staubgesetz von Cardelli, Clayton & Mathis (1989) im Optischen: k(λ) = A_λ / E(B−V), λ in Å
    x = 1e4 / np.asarray(lam, float)
    y = x - 1.82
    a = 1 + 0.17699 * y - 0.50447 * y**2 - 0.02427 * y**3 + 0.72085 * y**4 + 0.01979 * y**5 - 0.77530 * y**6 + 0.32999 * y**7
    b = 1.41338 * y + 2.28305 * y**2 + 1.07233 * y**3 - 5.38434 * y**4 - 0.62251 * y**5 + 5.30260 * y**6 - 2.09002 * y**7
    return Rv * a + b


LAM = {"Hb": 4861.33, "Ha": 6562.80, "O3": 5006.84, "He2": 4685.68, "V": LAM_V}
K = {k: float(k_ccm(v)) for k, v in LAM.items()}
nu_of = lambda lam: C_CGS / (lam * 1e-8)


def j_rec(tab, trans, T):
    # Emissivität 4πj / (n_e n_ion) [erg cm³ s⁻¹] bei n_e = 100 cm⁻³; "4_2" = Hβ, "3_2" = Hα, He II "4_3" = 4686
    return np.interp(T, tab["TEMP"][0], tab[trans][0])


def alpha_B(T):
    # Rekombinationskoeffizient des Wasserstoffs im Fall B [cm³ s⁻¹] (Draine 2011, Gl. 14.6)
    t4 = T / 1e4
    return 2.54e-13 * t4 ** (-0.8163 - 0.0208 * np.log(t4))


HA_HB = j_rec(SH_H, "3_2", T_NEB) / j_rec(SH_H, "4_2", T_NEB)                 # Balmer-Dekrement ohne Staub
A_EFF = {"H": j_rec(SH_H, "4_2", T_NEB) / (H_PLANCK * nu_of(LAM["Hb"])),       # effektive Rekombination in Hβ
         "He": j_rec(SH_HE, "4_3", T_NEB) / (H_PLANCK * nu_of(LAM["He2"]))}    # ... und in He II 4686
A_B = {"H": alpha_B(T_NEB), "He": 2 * alpha_B(T_NEB / 4)}                      # He⁺ ist wasserstoffähnlich mit Z = 2
LINE = {"H": (nu_of(LAM["Hb"]), NU_H), "He": (nu_of(LAM["He2"]), NU_HE2)}       # Linienfrequenz, Ionisationskante
NU_V = nu_of(LAM_V)


def deredden(df):
    # Rötung aus Hα/Hβ (Theorie, Abschnitt 2) und entrötete Flüsse [erg s⁻¹ cm⁻²]
    df = df.copy()
    df["EBV"] = (np.log10(df["I6563"] / 100 / HA_HB) / (0.4 * (K["Hb"] - K["Ha"]))).clip(lower=0)
    df["c"] = 0.4 * K["Hb"] * df["EBV"]                                        # c(Hβ) = log10 der Schwächung von Hβ
    df["F_Hb"] = 10 ** df["log(Fbeta)"] * 10 ** df["c"]
    df["F_O3"] = df["I5007"] / 100 * df["F_Hb"] * 10 ** (0.4 * df["EBV"] * (K["O3"] - K["Hb"]))
    df["F_He2"] = df["I4686"] / 100 * df["F_Hb"] * 10 ** (0.4 * df["EBV"] * (K["He2"] - K["Hb"]))
    df["m5007"] = -2.5 * np.log10(df["F_O3"]) - 13.74                           # [O III]-Helligkeit (Jacoby 1989)
    return df


def planck_nu(nu, T):
    return 2 * H_PLANCK * nu**3 / C_CGS**2 / np.expm1(H_PLANCK * nu / (K_B * T))


def photon_flux(T, nu0):
    # ∫ B_ν / ν dν von nu0 bis unendlich [erg s⁻¹ cm⁻² sr⁻¹]; geteilt durch h ist das die Zahl ionisierender Photonen
    x0 = H_PLANCK * nu0 / (K_B * T)
    I = integrate.quad(lambda x: x * x / np.expm1(x), x0, x0 + 60)[0]
    return 2 * H_PLANCK / C_CGS**2 * (K_B * T / H_PLANCK) ** 3 * I


def zanstra_ratio(T, ion):
    # erwartetes F(Linie) / F_ν(V) [Hz] für einen Schwarzkörper der Temperatur T (Theorie, Abschnitt 3)
    nu_line, nu0 = LINE[ion]
    return nu_line * A_EFF[ion] / A_B[ion] * photon_flux(T, nu0) / planck_nu(NU_V, T)


LOGT = np.linspace(np.log10(1.2e4), np.log10(6e5), 500)
ZR = {ion: np.log10([zanstra_ratio(10**t, ion) for t in LOGT]) for ion in ("H", "He")}   # steigt monoton mit T


def zanstra_T(ratio, ion):
    # Umkehrung: die Temperatur, bei der das erwartete Verhältnis dem gemessenen gleicht
    return 10 ** np.interp(np.log10(np.asarray(ratio, float)), ZR[ion], LOGT, left=np.nan, right=np.nan)


def luminosity(fnu_v, d_pc, T):
    # Leuchtkraft [L☉] eines Schwarzkörpers mit dem Fluss F_ν(V) in der Entfernung d (Theorie, Abschnitt 4)
    return 4 * np.pi * (d_pc * PC) ** 2 * fnu_v * SIGMA_SB * T**4 / np.pi / planck_nu(NU_V, T) / L_SUN


print(f"k(λ) = A_λ/E(B−V): Hβ {K['Hb']:.2f}, Hα {K['Ha']:.2f}, 5007 {K['O3']:.2f}, 4686 {K['He2']:.2f}, V {K['V']:.2f}")
print(f"Hα/Hβ ohne Staub = {HA_HB:.3f};  α_eff(Hβ)/α_B = {A_EFF['H'] / A_B['H']:.3f};  "
      f"α_eff(4686)/α_B(He⁺) = {A_EFF['He'] / A_B['He']:.3f}")
```

**Werkzeuge 2: Entwicklungswege und Leuchtkraftfunktion** (Theorie, Abschnitte 6 und 8). `track_mass` liest aus den Entwicklungswegen Kernmasse und Alter an einem Punkt des HR-Diagramms ab und markiert Sterne unter der leichtesten Spur; `fit_pnlf` passt die Leuchtkraftfunktion nach der Maximum-Likelihood-Methode an, mit einem kleinen gleichverteilten Anteil für Fremdobjekte.

```python
TRACKS = VW[VW["Z"] == Z_TRACKS].copy()
TRACKS["Mc"] = TRACKS["M"].map(M_CORE)
lightest = TRACKS[TRACKS["Mc"] == TRACKS["Mc"].min()].sort_values("Time")
# Die Bucht innerhalb der leichtesten Spur (waagrechter Ast oben, Abkühlungsast links, zum Anfang geschlossen):
# Sterne dort sind leichter als 0.569 M☉, oder ihre Leuchtkraft ist unterschätzt
BELOW = mpath.Path(lightest[["log(Teff)", "log(L)"]].values)


def track_mass(logT, logL):
    # Kernmasse und Alter seit log T = 4 durch lineare Interpolation zwischen den Spuren
    pts = TRACKS[["log(Teff)", "log(L)"]].values
    xy = np.column_stack([logT, logL])
    mc = griddata(pts, TRACKS["Mc"].values, xy)
    age = 10 ** griddata(pts, np.log10(TRACKS["Time"].clip(lower=10)).values, xy)
    under =np.array([BELOW.contains_point(p) for p in xy])
    mc[under], age[under] = np.nan, np.nan
    return mc, age, under


def pnlf_pdf(m, m_star, sig, frac, lo, hi, dm=0.005):
    # Wahrscheinlichkeitsdichte der Helligkeiten in [lo, hi]: PNLF gefaltet mit dem Messfehler sig plus Fremdanteil frac
    g = np.arange(lo - 2, hi + 2, dm)
    p = np.where(g > m_star, np.exp(0.307 * g) * (1 - np.exp(3 * (m_star - g))), 0.0)
    if sig > 0:
        p = gaussian_filter1d(p, sig / dm, mode="constant")
    inside = (g >= lo) & (g <= hi)
    p = p / np.trapz(p[inside], g[inside])
    return (1 - frac) * np.interp(m, g, p) + frac / (hi - lo)


def fit_pnlf(m, sig, lo, hi, start, free_frac=True):
    # Maximum Likelihood für m* (und den Fremdanteil); Fehler aus der Profil-Likelihood (Δ ln L = 0.5)
    m = m[(m >= lo) & (m <= hi)]
    if not free_frac:                       # ohne Fremdanteil muss die Kante unter dem hellsten Objekt liegen
        start = min(start, m.min() - 4 * sig - 0.05)
    to_frac = lambda t: 1 / (1 + np.exp(-t)) if free_frac else 0.0
    nll = lambda th: -np.sum(np.log(pnlf_pdf(m, th[0], sig, to_frac(th[1]), lo, hi)))
    best = optimize.minimize(nll, [start, -4.0], method="Nelder-Mead", options=dict(xatol=1e-4, fatol=1e-4))
    grid = np.arange(best.x[0] - 1.0, best.x[0] + 1.0, 0.01)
    prof = np.array([optimize.minimize_scalar(lambda t: nll([g, t]), bounds=(-12, 3), method="bounded").fun
                     if free_frac else nll([g, 0.0]) for g in grid])
    prof -= min(prof.min(), best.fun)
    inside = grid[prof < 0.5]
    return dict(m_star=best.x[0], frac=to_frac(best.x[1]), n=len(m), lo=inside.min(), hi=inside.max(),
                grid=grid, prof=prof)


print(f"Spuren bei Z = {Z_TRACKS}: Anfangsmassen {sorted(TRACKS['M'].unique())} M☉, "
      f"Kernmassen {sorted(TRACKS['Mc'].unique())} M☉, {len(TRACKS)} Punkte")
for M, g in TRACKS.groupby("M"):
    L_plateau = 10 ** g["log(L)"].max()
    print(f"  M_i = {M:.1f}: Plateau log L = {np.log10(L_plateau):.2f}, Kernmasse aus L = 59250 (M_c − 0.495): "
          f"{0.495 + L_plateau / 59250:.3f} (Tabelle {M_CORE[M]:.3f})")
```

Wie sehen Planetarische Nebel verschiedenen Alters aus, und wo sitzt der Stern, der sie zum Leuchten bringt? Vorher vorhersagen: Welcher der vier Nebel ist der jüngste, wenn alle Hüllen mit etwa 20 km/s wachsen?

Für vier bekannte Nebel holt der Code Farbbilder aus Pan-STARRS (i als Rot, r als Grün, g als Blau) und markiert den Gaia-Stern, den Chornay & Walton als Zentralstern zuordnen, sowie den Radius aus HASH. **Hintergrund & Fachbegriffe:**
- **Farben der Linien:** Das Bild zeigt den Filter i als Rot, r als Grün und g als Blau. Im Filter g liegt [O III] 4959/5007 (Sauerstoff, zweifach ionisiert), blau ist also stark ionisiertes Gas; im Filter r liegen Hα und [N II] 6548/6584, grün ist also Wasserstoff und einfach ionisierter Stickstoff. Im Filter i leuchten vor allem kühle Sterne, sie erscheinen rot oder gelb.
- **Zentralstern:** der heiße, freigelegte Kern des Sterns, der die Hülle abgestoßen hat (Theorie, Abschnitt 1). Er ist im Optischen oft schwach, weil er den größten Teil seiner Energie im Ultravioletten abgibt.
- **Parallaxe:** die scheinbare jährliche Verschiebung eines Sterns durch den Umlauf der Erde; Entfernung in Parsec = 1000 / Parallaxe in Millibogensekunden (Theorie, Abschnitt 5).

```python
SHOW = [("NGC 7662", "106.5-17.6", 60, "&asinh=True&autoscale=99.99"),   # Name, PNG, Bildgröße [″], Streckung
        ("M 57 (NGC 6720)", "063.1+13.9", 150, ""),
        ("M 27 (NGC 6853)", "060.8-03.6", 540, ""),
        ("Abell 39", "047.0+42.4", 240, "")]
cw_by_png = CW.set_index("PNG")
fig, axes = plt.subplots(1, 4, figsize=(18, 5))
rows = []
for ax, (name, png, size, stretch) in zip(axes, SHOW):
    r = cw_by_png.loc[png]
    img = ps1_color(r["RA_ICRS"], r["DE_ICRS"], size, ASTRODATA / "pn" / f"ps1_{png}.jpg", stretch)
    h = size / 2
    ax.imshow(img, extent=[h, -h, -h, h])                       # Norden oben, Osten links
    ax.plot(0, 0, "o", mfc="none", mec="yellow", ms=16, mew=1.3)
    ax.add_patch(plt.Circle((0, 0), r["pnRad"], fill=False, ls="--", color="white", lw=0.8))
    d = 1000 / (r["plx"] - PLX_ZP)
    bar = 0.1 * 206265 / d                                      # 0.1 pc in Bogensekunden
    ax.plot([0.9 * h, 0.9 * h - bar], [-0.85 * h] * 2, color="white", lw=2.5)
    ax.text(0.9 * h - bar / 2, -0.78 * h, "0.1 pc", color="white", ha="center", fontsize=9)
    ax.set_title(f"{name}\nd = {d:.0f} pc, Radius {r['pnRad']:.0f}″", fontsize=10)
    ax.set_xlabel("Δα cos δ [″] (Osten links)")
    ax.set_ylabel("Δδ [″]")
    rows.append(dict(Nebel=name, PNG=png, G=r["Gmag"], Parallaxe_mas=r["plx"], SN=r["plx"] / r["e_plx"], d_pc=d,
                     Radius_as=r["pnRad"], Radius_pc=r["pnRad"] / 206265 * d))
fig.suptitle("Vier Planetarische Nebel in Pan-STARRS (i, r, g); gelber Kreis: Zentralstern nach Gaia, gestrichelt: Radius nach HASH",
             fontsize=11)
plt.tight_layout()
plt.show()
print(pd.DataFrame(rows).to_string(index=False, float_format=lambda x: f"{x:.2f}"))
```

**Hintergrund & Fachbegriffe:**

##### *3. Entfernungen aus Gaia*

Wie weit sind die Nebel entfernt, und ab wann darf man die Entfernung einfach als 1/Parallaxe rechnen? Vorher vorhersagen: Weicht 1/Parallaxe bei schlechten Parallaxen eher zu großen oder zu kleinen Entfernungen von der sorgfältigeren Schätzung ab?

Die Entfernung ist 1000 / (ϖ − ϖ₀) pc mit dem Nullpunkt ϖ₀ = −0.017 mas: Gaia-EDR3-Parallaxen sind im Mittel um 0.017 mas zu klein. Zum Vergleich dient die Entfernung, die Chornay & Walton aus der Parallaxe zusammen mit einer statistischen Entfernung berechnen (Median der kombinierten Wahrscheinlichkeitsverteilung). Links stehen Parallaxe und Fehler aller sicher zugeordneten Zentralsterne, rechts der Vergleich der beiden Entfernungen für S/N über 5 und für S/N zwischen 2 und 5.

**Hintergrund & Fachbegriffe:**
- **Parallaxen-Nullpunkt:** ein systematischer Fehler aller Parallaxen, gemessen an Quasaren, die keine Parallaxe haben (Lindegren et al. 2021); er hängt etwas von Helligkeit, Farbe und Ort ab.
- **Verzerrung von 1/ϖ:** Eine gemessene Parallaxe streut symmetrisch um den wahren Wert, ihr Kehrwert aber nicht; kleine Parallaxen werden zu sehr großen Entfernungen. Bei S/N über 5 ist der Effekt klein (Theorie, Abschnitt 5).
- **Statistische Entfernung:** eine Entfernung aus einer empirischen Beziehung zwischen Flächenhelligkeit und Radius des Nebels (Frew, Parker & Bojičić 2016); sie streut um etwa 30 %, gilt aber für alle Nebel.

```python
S["d"] = 1000 / (S["plx"] - PLX_ZP)
sure = GAIA[GAIA["sure"]].copy()
sure["d_inv"] = 1000 / (sure["plx"] - PLX_ZP)
hi_q = sure[sure["snr"] > SNR_PLX]
mid_q = sure[(sure["snr"] > 2) & (sure["snr"] <= SNR_PLX)]

fig, axes = plt.subplots(1, 2, figsize=(14, 5.2))
ax = axes[0]
ok = sure["plx"] > 0
ax.scatter(sure.loc[ok, "e_plx"], sure.loc[ok, "plx"], s=4, c=np.where(sure.loc[ok, "snr"] > SNR_PLX, "C0", "0.6"))
e = np.logspace(-2, 0.5, 50)
for k, ls in [(SNR_PLX, "-"), (2, "--"), (1, ":")]:
    ax.plot(e, k * e, "k", ls=ls, lw=1, label=f"S/N = {k:.0f}")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Fehler der Parallaxe [mas]")
ax.set_ylabel("Parallaxe [mas]")
ax.set_title(f"{len(sure)} sicher zugeordnete Zentralsterne ({ok.sum()} mit ϖ > 0); blau: S/N > {SNR_PLX:.0f}", fontsize=10)
ax.legend(fontsize=9)
ax = axes[1]
for df, col, lab in [(mid_q, "C1", "2 < S/N ≤ 5"), (hi_q, "C0", f"S/N > {SNR_PLX:.0f}")]:
    v = df.dropna(subset=["rcomb"])
    ax.scatter(v["rcomb"], v["d_inv"], s=8, c=col, label=f"{lab} ({len(v)})")
x = np.logspace(1.8, 4.3, 10)
ax.plot(x, x, "k-", lw=1, label="gleich")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Entfernung aus Parallaxe und Flächenhelligkeit (Chornay & Walton) [pc]")
ax.set_ylabel("1000 / (ϖ − ϖ₀) [pc]")
ax.set_title("Wann genügt 1/Parallaxe?", fontsize=10)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

for lab, df in [(f"S/N > {SNR_PLX:.0f}", hi_q), ("2 < S/N ≤ 5", mid_q)]:
    v = df.dropna(subset=["rcomb"])
    q = np.log10(v["d_inv"] / v["rcomb"])
    print(f"{lab:10s}: {len(df):4d} Sterne, log10(1/ϖ ÷ kombiniert) Median {q.median():+.3f}, "
          f"Anteil 1/ϖ größer: {np.mean(q > 0):.0%}, typischer Fehler der Entfernung {np.median(1 / df['snr']):.0%}")
print(f"Arbeitsstichprobe: {len(S)} Nebel, Entfernung Median {S['d'].median():.0f} pc "
      f"(10 % bis 90 %: {S['d'].quantile(0.1):.0f} bis {S['d'].quantile(0.9):.0f} pc)")
print("Die nächsten fünf:", ", ".join(f"{n} {d:.0f} pc" for n, d in S.nsmallest(5, "d")[["Name", "d"]].values))
```

* Links Parallaxe gegen Fehler aller 1134 sicher zugeordneten Zentralsterne, mit Linien für S/N = 5, 2 und 1. Die meisten Zentralsterne sind schwach, ihre Fehler liegen zwischen 0.03 und 1 mas; nur 291 liegen über der Linie S/N = 5.
* Rechts 1/ϖ gegen die kombinierte Entfernung von Chornay & Walton. Für S/N > 5 stimmen beide im Median auf 5 % (−0.020 dex), der typische Fehler ist 8 %. Für 2 < S/N ≤ 5 ist 1/ϖ im Median 21 % kleiner (−0.099 dex), nur 21 % der Sterne liegen darüber. Die Vorhersage „zu groß“ aus der Theorie trifft hier also nicht zu. Vermutlich liegt das an der Auswahl: Wer nach S/N > 2 auswählt, nimmt bevorzugt Sterne, deren Parallaxe zufällig nach oben gestreut ist, und eine zu große Parallaxe gibt eine zu kleine Entfernung. Dazu zieht der Prior bei schwachen Parallaxen zur statistischen Entfernung hin.

##### *4. Rötung aus dem Balmer-Dekrement*

Wie viel Licht schluckt der Staub zwischen uns und jedem Nebel, und wie misst man das, ohne den Staub zu sehen? Vorher vorhersagen: Wie hängt die Rötung von der Entfernung ab, und welche Nebel sind am stärksten gerötet?

Ohne Staub ist das Verhältnis Hα/Hβ durch die Atomphysik festgelegt: 2.86 bei 10 000 K (Theorie, Abschnitt 2). Staub schwächt blaues Licht stärker als rotes; das beobachtete Verhältnis ist deshalb größer. Aus dem Überschuss folgt die Farbexzess E(B−V) und daraus mit dem Staubgesetz die Schwächung bei jeder Wellenlänge. Der Code korrigiert so den Hβ-Fluss, die Linien [O III] 5007 und He II 4686 und die V-Helligkeit des Zentralsterns. Verhältnisse unter 2.86 (Messfehler) setzt er auf E(B−V) = 0.

**Hintergrund & Fachbegriffe:**
- **Balmer-Dekrement:** das Verhältnis der Balmer-Linien Hα/Hβ (oder Hγ/Hβ), dessen Wert ohne Staub aus der Rekombinationsphysik bekannt ist.
- **Farbexzess E(B−V):** wie viel röter ein Objekt durch Staub erscheint, gemessen als Zunahme der Farbe B − V in Magnituden.
- **c(Hβ):** der Zehnerlogarithmus des Faktors, um den Staub Hβ schwächt; c = 1 heißt, nur ein Zehntel des Hβ-Lichts kommt an. Es gilt c = 0.4 · k(Hβ) · E(B−V) ≈ 1.44 · E(B−V).
- **Staubgesetz:** wie die Schwächung A_λ von der Wellenlänge abhängt, hier nach Cardelli, Clayton & Mathis (1989) mit R_V = 3.1, dem typischen Wert in der Milchstraße.

```python
r = S.dropna(subset=["EBV"])
ratio = r["I6563"] / 100
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = axes[0]
ax.hist(ratio, bins=np.arange(1.5, 12.5, 0.25), color="C0", alpha=0.8)
ax.axvline(HA_HB, color="k", ls="--", label=f"ohne Staub: {HA_HB:.2f}")
ax.set_xlabel("beobachtetes Hα / Hβ")
ax.set_ylabel("Zahl der Nebel")
ax.set_title(f"Balmer-Dekrement von {len(r)} Nebeln der Arbeitsstichprobe", fontsize=10)
ax.legend()
ax = axes[1]
sc = ax.scatter(r["d"], r["EBV"], c=np.abs(r["b"]), cmap="plasma_r", s=14, vmax=30)
plt.colorbar(sc, ax=ax, label="|galaktische Breite b| [°]")
ax.set_xscale("log")
ax.set_xlabel("Entfernung [pc]")
ax.set_ylabel("E(B−V) [mag]")
ax.set_title("Rötung gegen Entfernung", fontsize=10)
for _, row in r.nlargest(4, "EBV").iterrows():
    ax.annotate(row["Name"], (row["d"], row["EBV"]), fontsize=8, xytext=(4, 2), textcoords="offset points")
plt.tight_layout()
plt.show()

print(f"{len(r)} Nebel: E(B−V) Median {r['EBV'].median():.2f} mag, c(Hβ) Median {r['c'].median():.2f}; "
      f"{np.mean(ratio < HA_HB):.0%} haben Hα/Hβ < {HA_HB:.2f} (E = 0 gesetzt)")
near, far = r[r["d"] < 1500], r[r["d"] >= 1500]
print(f"E(B−V) Median: näher als 1.5 kpc {near['EBV'].median():.2f} ({len(near)}), weiter {far['EBV'].median():.2f} ({len(far)}); "
      f"|b| < 5°: {r.loc[np.abs(r['b']) < 5, 'EBV'].median():.2f}, |b| > 15°: {r.loc[np.abs(r['b']) > 15, 'EBV'].median():.2f}")
print(f"Schwächung bei E(B−V) = 1: Hβ {K['Hb']:.2f} mag, [O III] 5007 {K['O3']:.2f} mag, V {K['V']:.2f} mag")
```

* Rechts die Rötung gegen die Entfernung, eingefärbt nach galaktischer Breite. Im Median ist E(B−V) = 0.33 mag (c(Hβ) = 0.47); näher als 1.5 kpc 0.20, weiter weg 0.36. ##### *5. Größe und kinematisches Alter*

Wie groß sind die Nebel wirklich, und wie lange wachsen sie schon? Vorher vorhersagen: Wenn ein Nebel 0.3 pc Radius hat und mit 20 km/s wächst, wie alt ist er dann?

Der Winkelradius aus HASH mal Entfernung gibt den Radius in Parsec. Die Expansionsgeschwindigkeit aus dem Strasbourg-ESO-Katalog stammt aus der Aufspaltung der Linie [O III] 5007: Vorder- und Rückseite der Hülle bewegen sich auf uns zu und von uns weg, die Linie spaltet sich in zwei Teile, und ihr halber Abstand ist die Geschwindigkeit. Radius geteilt durch Geschwindigkeit ist das kinematische Alter (Theorie, Abschnitt 6). Links stehen Radius und Geschwindigkeit mit Linien gleichen Alters, rechts die Verteilung der Alter.

**Hintergrund & Fachbegriffe:**
- **Kinematisches Alter:** t = R / v unter der Annahme, dass die Hülle immer gleich schnell gewachsen ist; eine Uhr, die nur Größe und Geschwindigkeit braucht.
- **Linienaufspaltung:** Dopplerverschiebung von Vorder- und Rückseite einer expandierenden Schale; man misst sie mit hochauflösenden Spektren (Auflösung etwa 5 km/s).
- **1 pc / (1 km/s):** Eine Hülle, die 1 pc mit 1 km/s zurücklegt, braucht 978 000 Jahre; 0.1 pc mit 20 km/s sind also knapp 4900 Jahre.

```python
S["R_pc"] = S["pnRad"] / 206265 * S["d"]
S["t_kin"] = S["R_pc"] * PC / (S["ExpVel"] * KM) / YEAR
kin = S.dropna(subset=["t_kin"])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = axes[0]
sc = ax.scatter(kin["ExpVel"], kin["R_pc"], c=np.log10(kin["d"]), cmap="viridis", s=18)
plt.colorbar(sc, ax=ax, label="log₁₀(Entfernung / pc)")
v = np.logspace(0.4, 2.0, 50)
for t, lab in [(1e3, "1000 J"), (3e3, "3000 J"), (1e4, "10 000 J"), (3e4, "30 000 J"), (1e5, "100 000 J")]:
    ax.plot(v, v * KM * t * YEAR / PC, "k:", lw=0.8)
    ax.text(v[-1] * 1.03, v[-1] * KM * t * YEAR / PC, lab, fontsize=8, va="center")
for name, off in [("NGC 7662", (5, -3)), ("NGC 6720", (5, 2)), ("NGC 6853", (5, -10)), ("NGC 7293", (-48, 4))]:
    m = kin["Name"] == name
    if m.any():
        ax.annotate(name, (kin.loc[m, "ExpVel"].iloc[0], kin.loc[m, "R_pc"].iloc[0]), fontsize=8,
                    xytext=off, textcoords="offset points")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlim(2.5, 190)
ax.set_xlabel("Expansionsgeschwindigkeit aus [O III] [km/s]")
ax.set_ylabel("Radius [pc]")
ax.set_title(f"{len(kin)} Nebel; punktiert: gleiches kinematisches Alter in Jahren", fontsize=10)
ax = axes[1]
ax.hist(np.log10(kin["t_kin"]), bins=np.arange(2.6, 5.4, 0.1), color="C0", alpha=0.8)
ax.axvline(np.log10(kin["t_kin"].median()), color="k", ls="--", label=f"Median {kin['t_kin'].median():.0f} Jahre")
ax.set_xlabel("log₁₀(kinematisches Alter / Jahre)")
ax.set_ylabel("Zahl der Nebel")
ax.set_title("Wie lange wachsen die Hüllen schon?", fontsize=10)
ax.legend()
plt.tight_layout()
plt.show()

print(f"{len(kin)} Nebel: Radius Median {kin['R_pc'].median():.3f} pc ({kin['R_pc'].quantile(0.1):.3f} bis "
      f"{kin['R_pc'].quantile(0.9):.3f}), Geschwindigkeit Median {kin['ExpVel'].median():.1f} km/s, "
      f"Alter Median {kin['t_kin'].median():.0f} Jahre (10 % bis 90 %: {kin['t_kin'].quantile(0.1):.0f} bis {kin['t_kin'].quantile(0.9):.0f})")
rho = kin[["R_pc", "ExpVel"]].corr(method="spearman").iloc[0, 1]
print(f"Rangkorrelation Radius–Geschwindigkeit: {rho:+.2f}")
print(kin.set_index("Name").loc[[n for n in ["NGC 7662", "NGC 6720", "NGC 6853", "NGC 7293"] if n in set(kin["Name"])],
                                ["d", "pnRad", "R_pc", "ExpVel", "t_kin"]].round(3).to_string())
```

* Links Radius gegen Expansionsgeschwindigkeit, die Farbe ist die Entfernung, die punktierten Linien zeigen gleiches kinematisches Alter. Die Radien reichen von 0.07 bis 0.85 pc (Median 0.19 pc), die Geschwindigkeiten liegen meist zwischen 8 und 45 km/s (Median 19.8 km/s).
* Zur Vorhersage: 0.3 pc mit 20 km/s sind 0.3 × 978 000 / 20 ≈ 14 700 Jahre. Rechts die Verteilung aller 92 Alter: Median 9200 Jahre, 10 % bis 90 % zwischen 4400 und 47 000 Jahren. Ältere Nebel als etwa 50 000 Jahre gibt es kaum; sie sind so dünn geworden, dass man sie nicht mehr sieht.
* Größere Nebel wachsen im Mittel schneller (Rangkorrelation +0.37). Das passt zu Modellen, in denen der Wind des Sterns die Schale mit der Zeit beschleunigt (Theorie, Abschnitt 6). NGC 7662 ist 4500 Jahre alt, M 57 6200, M 27 29 000 und die Helix (NGC 7293) 33 000 Jahre.

##### *6. Die Zanstra-Temperatur aus Wasserstoff*

Wie heiß ist ein Stern, dessen ultraviolettes Licht man gar nicht sieht? Vorher vorhersagen: Ein Zentralstern hat V = 13 und sein Nebel einen Hβ-Fluss von 10⁻¹⁰ erg s⁻¹ cm⁻²; ist der Stern eher 30 000 K, 60 000 K oder 100 000 K heiß?

Jedes Photon über 13.6 eV, das der Stern aussendet, ionisiert im Nebel ein Wasserstoffatom, und jede Rekombination danach liefert mit fester Wahrscheinlichkeit ein Hβ-Photon. Der Hβ-Fluss des Nebels zählt so die ionisierenden Photonen des Sterns; die V-Helligkeit misst seinen sichtbaren Fluss. Das Verhältnis der beiden steigt steil mit der Temperatur, und für einen Schwarzkörper gibt es genau eine Temperatur, die passt (Theorie, Abschnitt 3). Links steht die berechnete Kurve mit den gemessenen Verhältnissen vier bekannter Nebel, rechts die Verteilung aller Zanstra-Temperaturen.

Ausgelassen werden Sterne mit Photometrie der Güte C oder D und Zentralsterne, deren V-Licht ein kühler Begleiter liefert (Spektraltyp A bis M oder B5 bis B9 im Katalog von Weidmann et al.); bei ihnen misst V den Begleiter, nicht den heißen Stern.

**Hintergrund & Fachbegriffe:**
- **Zanstra-Methode:** Herman Zanstra (1927) bestimmte so die ersten Temperaturen von Zentralsternen; sie setzt voraus, dass der Nebel alle ionisierenden Photonen schluckt (optisch dick).
- **Fall B:** Rekombinationen direkt in den Grundzustand erzeugen wieder ein ionisierendes Photon, das gleich wieder geschluckt wird; gezählt werden nur Rekombinationen in angeregte Zustände, Koeffizient α_B.
- **Effektiver Rekombinationskoeffizient α_eff(Hβ):** der Teil der Rekombinationen, der über den Übergang n = 4 → 2 läuft und ein Hβ-Photon erzeugt; bei 10 000 K etwa 12 % von α_B.

```python
WEID_COLS = ["PNG", "SpType1", "TpTeff", "logTeff", "logL"]
S = S.drop(columns=[c for c in WEID_COLS[1:] if c in S.columns]).merge(WEID[WEID_COLS], on="PNG", how="left")
S["cool"] = S["SpType1"].fillna("").str.match(r"^\s*([AFGKM]|B[5-9])")
S["V0"] = S["Vmag"] - K["V"] * S["EBV"]                           # entrötete V-Helligkeit
S["FnuV"] = FNU_V0 * 10 ** (-0.4 * S["V0"])                        # F_ν bei 548 nm [erg s⁻¹ cm⁻² Hz⁻¹]
usable = S["Vmag"].notna() & S["F_Hb"].notna() & ~S["Quality"].isin(["C", "D"]) & ~S["cool"]
S["T_H"] = np.where(usable, zanstra_T(S["F_Hb"] / S["FnuV"], "H"), np.nan)

EXAMPLES = ["NGC 6853", "NGC 7293", "NGC 3242", "NGC 6543"]
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = axes[0]
ax.plot(10 ** LOGT / 1e3, 10 ** ZR["H"], "C0-", label="Hβ: Photonen über 13.6 eV")
ax.plot(10 ** LOGT / 1e3, 10 ** ZR["He"], "C3-", label="He II 4686: Photonen über 54.4 eV")
for i, name in enumerate(EXAMPLES):
    row = S[S["Name"] == name]
    if len(row) and np.isfinite(row["T_H"].iloc[0]):
        q, T = (row["F_Hb"] / row["FnuV"]).iloc[0], row["T_H"].iloc[0]
        ax.axhline(q, color=f"C{i + 4}", lw=0.8, ls=":")
        ax.plot(T / 1e3, q, "o", color=f"C{i + 4}", label=f"{name}: {T / 1e3:.0f} kK")
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlim(12, 600)
ax.set_xticks([20, 50, 100, 200, 500], labels=["20", "50", "100", "200", "500"])
ax.xaxis.set_minor_formatter(plt.NullFormatter())
ax.set_ylim(1e10, 1e17)
ax.set_xlabel("Temperatur des Sterns [1000 K]")
ax.set_ylabel("F(Linie) / F_ν(V) [Hz]")
ax.set_title("Zanstra-Beziehung für einen Schwarzkörper", fontsize=10)
ax.legend(fontsize=8, loc="lower right")
ax = axes[1]
th = S["T_H"].dropna()
ax.hist(th / 1e3, bins=np.arange(10, 160, 5), color="C0", alpha=0.8)
ax.axvline(th.median() / 1e3, color="k", ls="--", label=f"Median {th.median() / 1e3:.0f} kK")
ax.set_xlabel("Zanstra-Temperatur aus Wasserstoff [1000 K]")
ax.set_ylabel("Zahl der Zentralsterne")
ax.set_title(f"{len(th)} Zentralsterne", fontsize=10)
ax.legend()
plt.tight_layout()
plt.show()

print(f"Zanstra-Temperatur (H) für {len(th)} Sterne: Median {th.median():.0f} K, 10 % bis 90 %: "
      f"{th.quantile(0.1):.0f} bis {th.quantile(0.9):.0f} K")
print("Ausgelassen wegen kühlem Begleiter:", ", ".join(f"{n} ({s})" for n, s in
                                                     S.loc[S["cool"] & S["Vmag"].notna(), ["Name", "SpType1"]].values))
print(S.set_index("Name").loc[[n for n in EXAMPLES if n in set(S["Name"])], ["Vmag", "EBV", "V0", "F_Hb", "T_H"]]
      .to_string(float_format=lambda x: f"{x:.3g}"))
```

Hantel- und Helixnebel haben 127 000 und 116 000 K, NGC 3242 56 000 K und der Katzenaugennebel NGC 6543 45 000 K.

##### *7. Ein zweites Thermometer: He II*

Stimmt die Temperatur aus Wasserstoff mit der aus ionisiertem Helium überein, und wenn nicht, was verrät der Unterschied über den Nebel? Vorher vorhersagen: Bei welchen Nebeln, kleinen oder großen, gibt Helium die höhere Temperatur?

Dasselbe Prinzip wie in Schritt 6 mit der Linie He II 4686: Sie entsteht, wenn He⁺⁺ rekombiniert, zählt also die Photonen über 54.4 eV, die He⁺ ionisieren können. Ein großer, dünner Nebel lässt einen Teil der Photonen über 13.6 eV entweichen (er ist für Wasserstoff optisch dünn); dann unterschätzt Hβ die Photonenzahl, und die Wasserstoff-Temperatur wird zu niedrig. Die viel selteneren Photonen über 54.4 eV werden dagegen meist vollständig geschluckt. Links steht T(He II) gegen T(H), eingefärbt nach dem Radius des Nebels, rechts das Verhältnis gegen den Radius.

Für die folgenden Schritte gilt: Wo He II gemessen ist, zählt die höhere der beiden Temperaturen; sonst die Wasserstoff-Temperatur als untere Grenze.

**Hintergrund & Fachbegriffe:**
- **Optisch dick und dünn:** Ein Nebel ist für eine Strahlung optisch dick, wenn er sie vollständig absorbiert, und optisch dünn, wenn ein Teil durchkommt. Ob der Nebel die Photonen über 13.6 eV ganz schluckt, hängt von seiner Masse und Dichte ab.
- **Zanstra-Diskrepanz:** T(He II) > T(H), schon 1956 von Harman & Seaton bemerkt; ein Zeichen dafür, dass der Nebel für Wasserstoff ionisierende Strahlung durchlässig ist.

```python
S["T_He"] = np.where(usable, zanstra_T(S["F_He2"] / S["FnuV"], "He"), np.nan)
S["T"] = np.where(S["T_He"].notna(), np.fmax(S["T_He"], S["T_H"]), S["T_H"])
S["T_limit"] = S["T_He"].isna() & S["T_H"].notna()                  # nur Wasserstoff: untere Grenze
both = S.dropna(subset=["T_H", "T_He"])
both_r = both.dropna(subset=["R_pc"])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = axes[0]
sc = ax.scatter(both["T_H"] / 1e3, both["T_He"] / 1e3, c=np.log10(both["R_pc"]), cmap="viridis", s=22)
plt.colorbar(sc, ax=ax, label="log₁₀(Radius / pc)")
ax.plot([15, 250], [15, 250], "k-", lw=1, label="gleich")
ax.set_xscale("log")
ax.set_yscale("log")
for setter in (ax.set_xticks, ax.set_yticks):
    setter([20, 30, 50, 100, 200], labels=["20", "30", "50", "100", "200"])
ax.xaxis.set_minor_formatter(plt.NullFormatter())
ax.yaxis.set_minor_formatter(plt.NullFormatter())
ax.set_xlabel("T aus Wasserstoff [1000 K]")
ax.set_ylabel("T aus He II [1000 K]")
ax.set_title(f"{len(both)} Zentralsterne mit beiden Temperaturen", fontsize=10)
ax.legend()
ax = axes[1]
q = both_r["T_He"] / both_r["T_H"]
ax.scatter(both_r["R_pc"], q, s=22, c="C0")
for _, row in both_r.assign(q=q).nlargest(3, "q").iterrows():
    ax.annotate(row["Name"], (row["R_pc"], row["q"]), fontsize=8, xytext=(4, 2), textcoords="offset points")
ax.axhline(1, color="k", lw=1)
ax.set_xscale("log")
ax.set_xlabel("Radius des Nebels [pc]")
ax.set_ylabel("T(He II) / T(H)")
ax.set_title("Große Nebel lassen Wasserstoff-Photonen entweichen", fontsize=10)
plt.tight_layout()
plt.show()

rho = both_r[["R_pc"]].assign(q=q).corr(method="spearman").iloc[0, 1]
small, large = both_r[both_r["R_pc"] < 0.1], both_r[both_r["R_pc"] >= 0.1]
print(f"{len(both)} Sterne mit beiden Temperaturen: T(He II)/T(H) Median {np.median(both['T_He'] / both['T_H']):.2f}; "
      f"Radius < 0.1 pc: {np.median(small['T_He'] / small['T_H']):.2f} ({len(small)}), "
      f"≥ 0.1 pc: {np.median(large['T_He'] / large['T_H']):.2f} ({len(large)}); Rangkorrelation mit dem Radius {rho:+.2f}")
print(f"Temperatur für die folgenden Schritte: {S['T'].notna().sum()} Sterne, davon {int(S['T_limit'].sum())} nur aus "
      f"Wasserstoff (untere Grenze); Median mit He II {S.loc[~S['T_limit'] & S['T'].notna(), 'T'].median():.0f} K")
```

* Links T(He II) gegen T(H) für 39 Zentralsterne; alle liegen über der Diagonale, im Median ist T(He II) 1.96-mal so hoch. Die Farbe zeigt den Radius des Nebels: Die großen (gelbgrün) liegen weiter über der Diagonale als die kleinen (violett).
* Rechts das Verhältnis gegen den Radius: 1.64 für Nebel unter 0.1 pc, 2.21 darüber, Rangkorrelation +0.61. Große, alte, dünne Nebel lassen mehr Photonen über 13.6 eV entweichen; Hβ zählt nur einen Teil, und T(H) wird zu niedrig. Das bestätigt die Vorhersage.
* Auch kleine Nebel haben T(He II) / T(H) = 1.6. Vermutlich lassen auch sie etwas Strahlung entweichen, und dazu ist der Schwarzkörper nur eine Näherung für das Spektrum eines heißen Sterns (Schritt 13). Für die weiteren Schritte gelten 39 Temperaturen mit He II (Median 70 000 K) und 38 nur aus Wasserstoff, als untere Grenzen.

##### *8. Das HR-Diagramm der Zentralsterne*

Wo stehen die Zentralsterne im Hertzsprung-Russell-Diagramm, und auf welchem Weg wandern sie zu Weißen Zwergen? Vorher vorhersagen: Haben Zentralsterne mit großen, alten Nebeln eine höhere oder niedrigere Leuchtkraft als solche mit kleinen?

Mit Temperatur und entröteter V-Helligkeit ist das Spektrum des Schwarzkörpers festgelegt; mit der Gaia-Entfernung folgt die Leuchtkraft L = 4π d² · F_ν(V) · σT⁴ / (π B_ν(V, T)) (Theorie, Abschnitt 4). Dazu zeichnet der Code die Entwicklungswege von Vassiliadis & Wood (1994) für Kernmassen von 0.569 bis 0.900 Sonnenmassen, mit Punkten nach 10³, 10⁴ und 10⁵ Jahren seit T = 10 000 K. Links das HR-Diagramm, rechts die Leuchtkraft gegen den Radius des Nebels.

**Hintergrund & Fachbegriffe:**
- **HR-Diagramm:** Leuchtkraft gegen Temperatur, die Temperatur nach Konvention nach links steigend.
- **Post-AGB-Weg:** Nach dem Abstoßen der Hülle brennt im Stern noch eine dünne Wasserstoffschale; die Leuchtkraft bleibt fast konstant, während der Stern schrumpft und heißer wird (waagrechter Ast). Ist die Schale verbraucht, kühlt er bei fallender Leuchtkraft ab (Knie und Abkühlungsast).
- **Kernmasse:** die Masse des freigelegten Kohlenstoff-Sauerstoff-Kerns, die spätere Masse des Weißen Zwergs; sie legt die Leuchtkraft auf dem waagrechten Ast fest (Theorie, Abschnitt 7).

```python
S["L"] = luminosity(S["FnuV"], S["d"], S["T"])
hr = S.dropna(subset=["T", "L"])
hr_he, hr_h = hr[~hr["T_limit"]], hr[hr["T_limit"]]

fig, axes = plt.subplots(1, 2, figsize=(15, 5.5))
ax = axes[0]
cmap = plt.get_cmap("copper_r")
for i, (M, g) in enumerate(TRACKS.groupby("M")):
    col = cmap(0.2 + 0.8 * i / 5)
    ax.plot(g["log(Teff)"], g["log(L)"], "-", color=col, lw=1.2, label=f"{M_CORE[M]:.3f} M☉")
    for t, mk in [(1e3, "o"), (1e4, "s"), (1e5, "^")]:
        j = np.argmin(np.abs(np.log10(g["Time"].clip(lower=1)) - np.log10(t)))
        ax.plot(g["log(Teff)"].iloc[j], g["log(L)"].iloc[j], mk, color=col, ms=4)
ax.scatter(np.log10(hr_he["T"]), np.log10(hr_he["L"]), s=24, c="C3", zorder=3, label=f"T aus He II ({len(hr_he)})")
ax.scatter(np.log10(hr_h["T"]), np.log10(hr_h["L"]), s=24, facecolors="none", edgecolors="C0", zorder=3,
           label=f"T nur aus H, untere Grenze ({len(hr_h)})")
ax.invert_xaxis()
ax.set_xlim(5.45, 4.2)
ax.set_ylim(0.3, 4.8)
ax.set_xlabel("log₁₀(T / K)")
ax.set_ylabel("log₁₀(L / L☉)")
ax.set_title("Kreis, Quadrat, Dreieck: 10³, 10⁴, 10⁵ Jahre seit T = 10 000 K", fontsize=10)
ax.legend(fontsize=7, loc="lower left", ncol=2)
ax = axes[1]
for df, kw in [(hr_he, dict(c="C3", s=24)), (hr_h, dict(facecolors="none", edgecolors="C0", s=24))]:
    v = df.dropna(subset=["R_pc"])
    ax.scatter(v["R_pc"], v["L"], **kw)
ax.set_xscale("log")
ax.set_yscale("log")
ax.set_xlabel("Radius des Nebels [pc]")
ax.set_ylabel("Leuchtkraft des Zentralsterns [L☉]")
ax.set_title("Mit wachsendem Nebel wird der Stern schwächer", fontsize=10)
plt.tight_layout()
plt.show()

hv = hr.dropna(subset=["R_pc"])
rho = hv[["R_pc", "L"]].corr(method="spearman").iloc[0, 1]
print(f"{len(hr)} Sterne im HR-Diagramm; log L Median {np.log10(hr_he['L']).median():.2f} (mit He II), "
      f"{np.log10(hr_h['L']).median():.2f} (nur H); Rangkorrelation Radius–Leuchtkraft {rho:+.2f}")
print(f"Radius < 0.1 pc: L Median {hv.loc[hv['R_pc'] < 0.1, 'L'].median():.0f} L☉, "
      f"> 0.3 pc: {hv.loc[hv['R_pc'] > 0.3, 'L'].median():.0f} L☉")
```

* Links das HR-Diagramm mit den Entwicklungswegen von 0.569 bis 0.900 M☉; Kreis, Quadrat und Dreieck markieren 10³, 10⁴ und 10⁵ Jahre. Die 39 Sterne mit He-II-Temperatur (rot) liegen bei log T 4.7 bis 5.2 und log L 1.4 bis 4.5, entlang des waagrechten Asts und am Knie der leichten Spuren; viele sitzen in der Bucht unter der leichtesten Spur. Die 38 offenen Kreise stehen zu kühl, weil ihre Temperatur eine untere Grenze ist, und mit ihr ist auch L zu klein (L wächst etwa wie T³).
* Rechts Leuchtkraft gegen Radius des Nebels: Rangkorrelation −0.64. Zentralsterne kleiner Nebel (unter 0.1 pc) leuchten im Median mit 5300 L☉, die großer Nebel (über 0.3 pc) mit 330 L☉. Die absolute [O III]-Helligkeit ist M5007 = m5007 − 5 log₁₀(d / 10 pc), mit m5007 = −2.5 log₁₀ F(5007) − 13.74 aus dem entröteten Fluss (Theorie, Abschnitt 8). Zwei Stichproben: die nahen Nebel der Arbeitsstichprobe mit Gaia-Entfernung und die Nebel des Bulge. Für den Bulge gilt die klassische Auswahl (Stasińska et al. 1991): |l| < 10°, 2° < |b| < 10°, Durchmesser unter 20″; alle stehen dann in etwa der Entfernung des Galaktischen Zentrums, 8.18 kpc. Weil die Rötung zum Bulge groß ist, zählen für die Eichung nur Nebel mit c(Hβ) < 1.5. Angepasst wird die Leuchtkraftfunktion mit Messfehler 0.3 mag und einem gleichverteilten Anteil für Fremdobjekte (Theorie, Abschnitt 9). Links die Helligkeiten, rechts die Profil-Likelihood von $M^\star$.

**Hintergrund & Fachbegriffe:**
- **Bulge:** die Verdickung um das Zentrum der Milchstraße aus alten Sternen; seine Planetarischen Nebel sind eine Stichprobe in fast gleicher, bekannter Entfernung.
- **Fremdobjekte:** Vordergrundnebel der Scheibe, die zufällig in Richtung Bulge stehen und heller erscheinen, und Nebel mit zu groß gemessener Rötung.
- **Profil-Likelihood:** Für jeden Wert von $M^\star$ wird der Fremdanteil optimiert; die Kurve Δ ln L zeigt, welche $M^\star$ die Daten zulassen (Δ ln L < 0.5 entspricht 1σ).

```python
S["M5007"] = S["m5007"] - 5 * np.log10(S["d"] / 10)
BULGE_LIM = -2.5                                                   # nur die hellsten 2 mag gehen in den Fit
bulge = ALL[(ALL["l"].abs() < 10) & ALL["b"].abs().between(2, 10) & (ALL["oDiam"] < 20)].dropna(subset=["m5007"]).copy()
bulge["M5007"] = bulge["m5007"] - 5 * np.log10(R0_PC / 10)
calib = bulge[bulge["c"] < C_BULGE_MAX]
FIT_BULGE = fit_pnlf(calib["M5007"].values, SIG_BULGE, -8.0, BULGE_LIM, -4.6)
FIT_BULGE_ALL = fit_pnlf(bulge["M5007"].values, SIG_BULGE, -8.0, BULGE_LIM, -4.6)
M_STAR_OWN = FIT_BULGE["m_star"]
M_STAR_OWN_ERR = (FIT_BULGE["hi"] - FIT_BULGE["lo"]) / 2
local = S.dropna(subset=["M5007"])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = axes[0]
bins = np.arange(-8, 2.01, 0.25)
ax.hist(bulge["M5007"], bins=bins, color="0.75", label=f"Bulge, alle ({len(bulge)})")
ax.hist(calib["M5007"], bins=bins, color="C0", alpha=0.8, label=f"Bulge, c(Hβ) < {C_BULGE_MAX} ({len(calib)})")
ax.hist(local["M5007"], bins=bins, histtype="step", color="C3", lw=2, label=f"Gaia-Stichprobe ({len(local)})")
g = np.linspace(-7, BULGE_LIM, 300)
nfit = FIT_BULGE["n"] * (1 - FIT_BULGE["frac"])
ax.plot(g, nfit * 0.25 * pnlf_pdf(g, M_STAR_OWN, SIG_BULGE, 0, -8.0, BULGE_LIM), "k-", lw=1.5, label="angepasste PNLF")
ax.axvline(M_STAR_OWN, color="C0", ls="--", label=f"M* Bulge = {M_STAR_OWN:.2f}")
ax.axvline(M_STAR_LIT, color="k", ls=":", label=f"M* Ciardullo 2012 = {M_STAR_LIT}")
ax.set_xlabel("M5007 [mag]")
ax.set_ylabel("Zahl der Nebel pro 0.25 mag")
ax.set_title("Absolute [O III]-Helligkeiten", fontsize=10)
ax.legend(fontsize=8)
ax = axes[1]
for f, col, lab in [(FIT_BULGE, "C0", f"c(Hβ) < {C_BULGE_MAX}"), (FIT_BULGE_ALL, "0.5", "alle Bulge-Nebel")]:
    ax.plot(f["grid"], f["prof"], color=col, label=f"{lab}: M* = {f['m_star']:.2f}, Fremdanteil {f['frac']:.0%}")
ax.axhline(0.5, color="k", ls=":", lw=1)
ax.axvline(M_STAR_LIT, color="k", ls=":")
ax.set_ylim(0, 6)
ax.set_xlim(-6.6, -4.0)
ax.set_xlabel("M* [mag]")
ax.set_ylabel("Δ ln L")
ax.set_title("Profil-Likelihood der Abbruchhelligkeit", fontsize=10)
ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

print("Hellste Nebel mit Gaia-Entfernung:", ", ".join(f"{n} {m:.2f}" for n, m in local.nsmallest(5, "M5007")[["Name", "M5007"]].values))
print(f"Bulge: {len(bulge)} Nebel, c(Hβ) Median {bulge['c'].median():.2f}; mit c < {C_BULGE_MAX}: {len(calib)}")
print("Hellste Bulge-Nebel (alle):", ", ".join(f"{n} {m:.2f} (c = {c:.1f})" for n, m, c in
                                             bulge.nsmallest(5, "M5007")[["Name", "M5007", "c"]].values))
for lab, f in [(f"c < {C_BULGE_MAX}", FIT_BULGE), ("alle", FIT_BULGE_ALL)]:
    print(f"Fit Bulge {lab:8s}: M* = {f['m_star']:.2f} (1σ: {f['lo']:.2f} bis {f['hi']:.2f}), Fremdanteil {f['frac']:.1%}, "
          f"{f['n']} Nebel heller als {BULGE_LIM}")
```

Von den nahen Nebeln erreicht nur IC 1747 den Abbruch (−4.83), der nächste ist NGC 3242 mit −3.88; die Vorhersage „nein“ trifft also fast zu. Im hellsten Magnitudenbereich liegen nur wenige Prozent aller Nebel, und 103 nahe Nebel sind zu wenige. IC 1747 selbst ist mit c = 1.75 so stark gerötet wie die verdächtigen Nebel im Bulge.

##### *11. Planetarische Nebel in M 31*

Wie sieht die Helligkeitsverteilung der Planetarischen Nebel in der Andromedagalaxie aus, und wie gut stimmen zwei unabhängige Messungen überein? Vorher vorhersagen: Bei welcher scheinbaren Helligkeit m5007 erwartest du die hellsten Nebel in M 31, wenn die Abbruchhelligkeit −4.5 ist, M 31 etwa 770 kpc entfernt ist und der Staub der Milchstraße 0.2 mag schluckt?

Ciardullo et al. (1989) haben im Bulge von M 31 mit Schmalbandfiltern 429 Nebel photometriert; die Teilstichproben A und B sind bis zu einer festen Helligkeit vollständig. Merrett et al. (2006) haben mit einem Spaltlos-Spektrographen über die ganze Scheibe 3305 Emissionslinienobjekte gefunden, mit Geschwindigkeit und m5007. Der Code nimmt von Merrett nur die Objekte, die zu M 31 gehören (nicht zu den Begleitern M 32, NGC 205 und anderen), vergleicht die Photometrie der gemeinsamen Nebel und zeigt links die Karte, eingefärbt nach Geschwindigkeit, rechts die Helligkeitsverteilungen.

**Hintergrund & Fachbegriffe:**
- **Schmalbandfilter:** ein Filter von wenigen nm Breite um 5007 Å (bei der Geschwindigkeit von M 31); Nebel leuchten darin hell, Sterne kaum. Ein zweites Bild daneben im Kontinuum trennt die Nebel von Sternen.
- **Planetary Nebula Spectrograph (PN.S):** ein Spaltlos-Spektrograph am William-Herschel-Teleskop, der Nebel findet und gleichzeitig ihre Geschwindigkeit misst.
- **Vollständigkeit:** bis zu welcher Helligkeit eine Durchmusterung alle Objekte findet; darunter fehlen schwache Nebel, und die Verteilung fällt künstlich ab.

```python
m31 = M06[(M06["Host"] == "") & M06["m5007"].notna()].copy()
ra = m31["RAJ2000"].str.split(expand=True).astype(float)
de = m31["DEJ2000"].str.split(expand=True).astype(float)
m31["ra"] = 15 * (ra[0] + ra[1] / 60 + ra[2] / 3600)
sgn = np.where(m31["DEJ2000"].str.startswith("-"), -1, 1)
m31["dec"] = sgn * (de[0].abs() + de[1] / 60 + de[2] / 3600)
RA_M31, DE_M31, V_M31 = 10.6847, 41.2690, -300.0                  # Zentrum und Systemgeschwindigkeit [km/s]
m31["x"] = (m31["ra"] - RA_M31) * np.cos(np.radians(DE_M31))
m31["y"] = m31["dec"] - DE_M31
c89 = C89[C89["Sample"].str.contains("A|B", na=False) & C89["m5007"].notna()]
match = M06.dropna(subset=["C89", "m5007"]).merge(C89.dropna(subset=["m5007"]), left_on="C89", right_on="PN",
                                                  suffixes=("_M06", "_C89"))
diff = match["m5007_M06"] - match["m5007_C89"]
SIG_DIFF = 1.4826 * np.median(np.abs(diff - diff.median()))

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))
ax = axes[0]
sc = ax.scatter(m31["x"], m31["y"], c=m31["HV"] - V_M31, cmap="coolwarm", s=3, vmin=-300, vmax=300)
plt.colorbar(sc, ax=ax, label="Geschwindigkeit relativ zu M 31 [km/s]")
ax.set_xlim(2.2, -2.2)
ax.set_aspect("equal")
ax.set_xlabel("Δα cos δ [°] (Osten links)")
ax.set_ylabel("Δδ [°]")
ax.set_title(f"{len(m31)} Nebel in M 31 (Merrett et al. 2006)", fontsize=10)
ax = axes[1]
bins = np.arange(19.0, 27.01, 0.2)
ax.hist(m31["m5007"], bins=bins, histtype="step", lw=1.8, color="C0", label=f"Merrett et al. 2006 ({len(m31)})")
ax.hist(c89["m5007"], bins=bins, color="C1", alpha=0.7, label=f"Ciardullo et al. 1989, Stichproben A, B ({len(c89)})")
for k, col in [("Ciardullo 1989", "C1"), ("Merrett 2006", "C0")]:
    ax.axvline(M5007_LIM[k], color=col, ls=":", lw=1)
ax.set_yscale("log")
ax.set_xlabel("m5007 [mag]")
ax.set_ylabel("Zahl pro 0.2 mag")
ax.set_title("Helligkeitsverteilung; punktiert: Grenze für den Fit", fontsize=10)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

print(f"Merrett et al.: {len(M06)} Objekte, davon {len(m31)} in M 31; Begleiter und andere: "
      + ", ".join(f"{k} {v}" for k, v in M06.loc[M06['Host'] != '', 'Host'].value_counts().head(5).items()))
print("Hellste fünf in M 31 (Merrett):", np.sort(m31["m5007"].values)[:5], "; Ciardullo A/B:", np.sort(c89["m5007"].values)[:5])
print(f"Gemeinsame Nebel: {len(match)}; Merrett − Ciardullo: Median {diff.median():+.3f} mag, robuste Streuung {SIG_DIFF:.3f} mag")
print(f"Geschwindigkeit: Nordosten (x > 0.5°) Median {np.median(m31.loc[m31['x'] > 0.5, 'HV']):.0f} km/s, "
      f"Südwesten (x < −0.5°) {np.median(m31.loc[m31['x'] < -0.5, 'HV']):.0f} km/s")
```

* Links die 3126 Nebel von M 31, eingefärbt nach Geschwindigkeit relativ zur Galaxie (−300 km/s). Der Nordosten entfernt sich (Median −106 km/s, also +194 relativ), der Südwesten kommt auf uns zu (−465 km/s): Die Nebel zeigen die Rotation der Scheibe. Einzelne Objekte weit außerhalb gehören zu Gezeitenströmen und zum Halo.
* Rechts die Helligkeitsverteilungen. Bei Merrett steigt die Zahl ab m5007 ≈ 20.0 steil an, bleibt zwischen 23 und 25 etwa gleich und fällt dann ab, weil schwache Nebel nicht mehr gefunden werden. ##### *12. Die Leuchtkraftfunktion von M 31*

Wo bricht die Helligkeitsverteilung der Nebel in M 31 ab, und welche Entfernung folgt daraus? Vorher vorhersagen: Wenn die hellsten Nebel bei m5007 ≈ 20.2 stehen und $M^\star$ = −4.53 ist, wie groß ist der Entfernungsmodul ohne Staub?

Für beide Kataloge passt der Code die Leuchtkraftfunktion an: Form nach Ciardullo et al. (1989), gefaltet mit dem Messfehler (0.07 mag bei Ciardullo, 0.15 mag bei Merrett aus dem Vergleich in Schritt 11), dazu ein gleichverteilter Anteil für Fremdobjekte. Gefittet werden nur Nebel bis zur Vollständigkeitsgrenze (22.0 bzw. 22.5 mag). Der Entfernungsmodul ist dann μ = $m^\star$ − A5007 − $M^\star$, mit A5007 = k(5007) · E(B−V) für den Staub der Milchstraße vor M 31. Links Daten und Modelle, rechts die Profil-Likelihood von $m^\star$.

**Hintergrund & Fachbegriffe:**
- **Entfernungsmodul:** μ = m − M = 5 log₁₀(d / 10 pc); μ = 24.4 sind 760 kpc.
- **Maximum-Likelihood mit scharfer Kante:** Die Lage einer Kante bestimmen die hellsten Objekte; ein einziger zu heller Fremdkörper würde den Fit ohne Fremdanteil unter die Kante ziehen (Variation 5).

```python
A5007_M31 = K["O3"] * EBV_M31
FIT = {"Ciardullo 1989": fit_pnlf(c89["m5007"].values, SIG_C89, 18.5, M5007_LIM["Ciardullo 1989"], 20.2),
       "Merrett 2006": fit_pnlf(m31["m5007"].values, SIG_M06, 17.0, M5007_LIM["Merrett 2006"], 20.2)}
SAMPLES = {"Ciardullo 1989": (c89["m5007"].values, SIG_C89, 18.5), "Merrett 2006": (m31["m5007"].values, SIG_M06, 17.0)}

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = axes[0]
for (k, f), col in zip(FIT.items(), ["C1", "C0"]):
    m, sig, lo = SAMPLES[k]
    hi = M5007_LIM[k]
    sel = m[(m >= 19.0) & (m <= hi)]
    h, e = np.histogram(sel, bins=np.arange(19.0, hi + 0.01, 0.1))
    ax.errorbar(0.5 * (e[1:] + e[:-1]), h, yerr=np.sqrt(h), fmt="o", ms=3, color=col, label=f"{k} ({f['n']} Nebel)")
    g = np.linspace(19.0, hi, 400)
    ax.plot(g, f["n"] * 0.1 * pnlf_pdf(g, f["m_star"], sig, f["frac"], lo, hi), color=col, lw=1.5,
            label=f"m* = {f['m_star']:.2f}, Fremdanteil {f['frac']:.1%}")
ax.set_yscale("log")
ax.set_ylim(0.5, 300)
ax.set_xlabel("m5007 [mag]")
ax.set_ylabel("Zahl pro 0.1 mag")
ax.set_title("Helle Enden der Leuchtkraftfunktion", fontsize=10)
ax.legend(fontsize=8)
ax = axes[1]
for (k, f), col in zip(FIT.items(), ["C1", "C0"]):
    ax.plot(f["grid"], f["prof"], color=col, label=f"{k}: {f['lo']:.2f} bis {f['hi']:.2f}")
ax.axhline(0.5, color="k", ls=":", lw=1)
ax.set_xlim(19.7, 20.7)
ax.set_ylim(0, 6)
ax.set_xlabel("m* [mag]")
ax.set_ylabel("Δ ln L")
ax.set_title("Profil-Likelihood; punktiert: 1σ", fontsize=10)
ax.legend(fontsize=9)
plt.tight_layout()
plt.show()

print(f"Vordergrund vor M 31: E(B−V) = {EBV_M31}, A5007 = {A5007_M31:.3f} mag")
for k, f in FIT.items():
    mu = f["m_star"] - A5007_M31 - M_STAR_LIT
    print(f"{k:15s}: m* = {f['m_star']:.3f} (+{f['hi'] - f['m_star']:.3f} / −{f['m_star'] - f['lo']:.3f}), "
          f"Fremdanteil {f['frac']:.1%} von {f['n']}; μ = {mu:.2f} mit M* = {M_STAR_LIT}, d = {10 ** (mu / 5 + 1) / 1e3:.0f} kpc")
```

##### *13. Temperaturen und Leuchtkräfte gegen die Literatur*

Vorher vorhersagen: Welche Gruppe weicht stärker ab, die Sterne mit He-II-Temperatur oder die mit Wasserstoff-Temperatur als untere Grenze?

Weidmann et al. (2020) sammeln für jeden Zentralstern die beste Temperatur und Leuchtkraft aus der Literatur mit der Methode („eff“ = Spektralanalyse mit Modellatmosphäre, „z(HeII)“ = Zanstra mit He II und andere). Der Code vergleicht log T und log L für alle Sterne, die in beiden Listen stehen, getrennt nach der eigenen Methode. Links die Temperaturen, rechts die Leuchtkräfte; die Literaturwerte der Leuchtkraft beruhen auf älteren Entfernungen.

**Hintergrund & Fachbegriffe:**
- **Modellatmosphäre:** ein Rechenmodell der äußeren Schichten eines Sterns, das Linienprofile (etwa von He II 4686 und Hβ im Stern selbst) aus Temperatur, Schwerebeschleunigung und Zusammensetzung vorhersagt; an das Spektrum angepasst, gibt es T unabhängig vom Nebel.
- **Schwarzkörper gegen Stern:** Ein heißer Stern strahlt jenseits der Ionisationskanten anders als ein Schwarzkörper derselben Temperatur; die Zanstra-Temperatur ist eine Schwarzkörper-Temperatur und kann deshalb systematisch abweichen.

```python
cmp = S[S["T"].notna() & S["logTeff"].notna()].copy()
cmp["dT"] = np.log10(cmp["T"]) - cmp["logTeff"]
cmp["dL"] = np.log10(cmp["L"]) - cmp["logL"]
groups = [(~cmp["T_limit"], "C3", "T aus He II"), (cmp["T_limit"], "C0", "T nur aus H")]

fig, axes = plt.subplots(1, 2, figsize=(14, 5.2))
ax = axes[0]
for m, col, lab in groups:
    ax.scatter(cmp.loc[m, "logTeff"], np.log10(cmp.loc[m, "T"]), s=22, c=col, label=f"{lab} ({m.sum()})")
ax.plot([4.2, 5.4], [4.2, 5.4], "k-", lw=1)
ax.set_xlabel("log₁₀ T, Weidmann et al. 2020")
ax.set_ylabel("log₁₀ T, Zanstra (dieses Notebook)")
ax.set_title("Temperaturen", fontsize=10)
ax.legend()
ax = axes[1]
for m, col, lab in groups:
    v = cmp[m].dropna(subset=["dL"])
    ax.scatter(v["logL"], np.log10(v["L"]), s=22, c=col, label=f"{lab} ({len(v)})")
ax.plot([0.5, 4.6], [0.5, 4.6], "k-", lw=1)
ax.set_xlabel("log₁₀ L / L☉, Weidmann et al. 2020")
ax.set_ylabel("log₁₀ L / L☉ (dieses Notebook)")
ax.set_title("Leuchtkräfte", fontsize=10)
ax.legend()
plt.tight_layout()
plt.show()

for m, _, lab in groups:
    v = cmp[m]
    print(f"{lab:12s}: {len(v):3d} Sterne, Δ log T Median {v['dT'].median():+.3f}, Streuung (robust) "
          f"{1.4826 * np.median(np.abs(v['dT'] - v['dT'].median())):.3f}; Δ log L Median {v['dL'].median():+.2f}, "
          f"Streuung {1.4826 * np.median(np.abs(v['dL'].dropna() - v['dL'].median())):.2f}")
eff = cmp[~cmp["T_limit"] & (cmp["TpTeff"] == "eff")]
print(f"Nur gegen Spektralanalysen („eff“) mit He-II-Temperatur: {len(eff)} Sterne, Δ log T Median {eff['dT'].median():+.3f}")
print("Größte Abweichungen:", ", ".join(f"{n} {d:+.2f}" for n, d in cmp.reindex(cmp["dT"].abs().sort_values(ascending=False).index)
                                         [["Name", "dT"]].head(4).values))
```

* Links die Temperaturen gegen Weidmann et al. (2020). Die 35 Sterne mit He-II-Temperatur (rot) liegen im Median nur 0.020 dex (5 %) darunter, robuste Streuung 0.076 dex; gegen reine Spektralanalysen (22 Sterne) sind es −0.039 dex. Die 34 Sterne mit Wasserstoff-Temperatur (blau) liegen 0.106 dex (22 %) darunter, mit doppelter Streuung; wie vorhergesagt weichen sie stärker ab, weil T(H) eine untere Grenze ist.
* Die größten Abweichungen haben Th 2-A (−0.85 dex), Abell 30 (−0.67), Abell 43 (−0.63) und Sa 1-8 (−0.43). Th 2-A und Abell 30 haben wasserstoffarme Zentralsterne ([WO]- und [WC]-PG1159-Typ), Abell 43 ist ein „hybrider“ Stern mit wenig Wasserstoff; Abell 30 und Abell 43 haben große, dünne Nebel. Wo Wasserstoff fehlt oder der Nebel durchlässig ist, versagt die Wasserstoff-Methode.
* Rechts die Leuchtkräfte: +0.10 dex für die He-II-Gruppe, −0.36 dex für die Wasserstoff-Gruppe, Streuung jeweils fast 0.5 dex. Hier mischen sich Temperaturfehler (L wächst etwa wie T³) und die Entfernungen: Die Literaturwerte beruhen meist auf älteren Entfernungen ohne Gaia.

#### **Wiederholung**

Erst selbst antworten, dann aufklappen.

<details><summary>1. Was ist ein Planetarischer Nebel, und was hat er mit Planeten zu tun?</summary>

Die abgestoßene Hülle eines Sterns von 0.8 bis 8 Sonnenmassen am Ende des asymptotischen Riesenasts, ionisiert vom freigelegten, heißen Kern. Mit Planeten hat er nichts zu tun; William Herschel nannte die runden Scheibchen so, weil sie im Fernrohr wie Uranus aussahen.
</details>

<details><summary>2. Wie misst man die Rötung eines Nebels, ohne den Staub zu sehen?</summary>

Mit dem Balmer-Dekrement: Ohne Staub ist Hα/Hβ durch die Rekombinationsphysik auf 2.86 festgelegt. Staub schwächt Hβ stärker als Hα, das beobachtete Verhältnis ist größer, und aus dem Überschuss folgt E(B−V). Im Median der Stichprobe sind es 0.33 mag, nahe der Milchstraßenebene 0.74.
</details>

<details><summary>3. Warum ist 1/Parallaxe bei schlechten Parallaxen eine schlechte Entfernung, und in welche Richtung lag sie hier daneben?</summary>

Der Kehrwert einer verrauschten Größe ist verzerrt; eine nach unten gestreute Parallaxe gibt eine viel zu große Entfernung. Hier lag 1/ϖ bei 2 < S/N ≤ 5 aber im Median 21 % zu niedrig, vermutlich weil die Auswahl nach S/N > 2 bevorzugt nach oben gestreute Parallaxen nimmt. Ab S/N = 5 stimmt 1/ϖ auf wenige Prozent.
</details>

<details><summary>4. Was misst das kinematische Alter, und warum ist es nur auf einen Faktor 2 genau?</summary>

Radius geteilt durch Expansionsgeschwindigkeit, also die Zeit, die die Schale mit der heutigen Geschwindigkeit gebraucht hätte. Die Geschwindigkeit aus [O III] misst eher innere Materie, der Rand ist schneller, und die Schale wird mit der Zeit beschleunigt. Im Median sind die Nebel 9200 Jahre alt.
</details>

<details><summary>5. Wie kann man die Temperatur eines Sterns messen, dessen Strahlung man fast nicht sieht?</summary>

Mit der Zanstra-Methode: Jedes Photon über 13.6 eV ionisiert im Nebel ein Atom, jede achte Rekombination gibt ein Hβ-Photon. Hβ zählt so die ionisierenden Photonen, V misst das sichtbare Licht; ihr Verhältnis hängt beim Schwarzkörper nur von der Temperatur ab.
</details>

<details><summary>6. Warum gibt He II 4686 fast immer eine höhere Temperatur als Hβ?</summary>

Weil viele Nebel nicht alle Photonen über 13.6 eV schlucken; dann zählt Hβ zu wenige, und die Wasserstoff-Temperatur wird zu niedrig. Die Photonen über 54.4 eV werden dagegen meist ganz geschluckt. Je größer und dünner der Nebel, desto größer der Unterschied: 1.64-fach bei kleinen, 2.21-fach bei großen Nebeln.
</details>

<details><summary>7. Warum geht ein Fehler der Temperatur so stark in die Leuchtkraft ein?</summary>

Gemessen ist nur der Fluss im V-Band, also auf der Rayleigh-Jeans-Seite des Spektrums, wo B_ν proportional zu T ist. Die Gesamtstrahlung wächst wie T⁴; bei festem V wächst L deshalb etwa wie T³. 0.08 dex in T sind 0.25 dex in L.
</details>

<details><summary>8. Was sagt die Kernmasse-Leuchtkraft-Beziehung, und warum macht sie das Alter auf dem Entwicklungsweg so unsicher?</summary>

Auf dem waagrechten Ast legt die Masse des Kerns die Leuchtkraft fest, L ≈ 59 250 (M_c − 0.495) L☉. Die Zeit bis zum Knie fällt aber extrem steil mit der Masse: 26 000 Jahre bei 0.569 M☉, 500 Jahre bei 0.754 M☉. Ein kleiner Fehler in L verschiebt die Masse etwas und das Alter um eine Größenordnung.
</details>

<details><summary>9. Warum werden die Zentralsterne großer Nebel schwächer?</summary>

Während der Nebel wächst, verbraucht der Stern die Wasserstoffschale auf seinem Kern; danach erlischt das Brennen, und er kühlt bei fallender Leuchtkraft zum Weißen Zwerg ab. Zentralsterne von Nebeln über 0.3 pc leuchten im Median mit 330 L☉, die kleiner Nebel mit 5300 L☉.
</details>

<details><summary>10. Was ist die [O III]-Leuchtkraftfunktion, und warum taugt sie als Entfernungsmesser?</summary>

Die Verteilung der Nebel einer Galaxie nach ihrer Helligkeit in [O III] 5007. Sie bricht bei $M^\star$ ≈ −4.5 scharf ab, in alten wie jungen Galaxien fast gleich. Misst man die Abbruchhelligkeit $m^\star$, folgt der Entfernungsmodul μ = $m^\star$ − A5007 − $M^\star$.
</details>

<details><summary>11. Warum braucht der Fit der Leuchtkraftfunktion einen Fremdanteil?</summary>

Oberhalb der Kante ist die Dichte des Modells null; ein einziges zu helles Objekt (eine H-II-Region, eine Hintergrundgalaxie, ein Messfehler) würde die Likelihood auf null setzen und den Fit unter dieses Objekt zwingen. Ohne Fremdanteil fällt $m^\star$ bei Merrett von 20.21 auf 18.34, und M 31 rückte von 809 auf 340 kpc (Variation 5).
</details>

<details><summary>12. Warum ist die Eichung im Bulge so empfindlich gegen die Rötung?</summary>

Die Nebel des Bulge sind stark gerötet (c(Hβ) im Median 1.47). Bei großer Rötung macht ein kleiner Fehler im Balmer-Dekrement große Fehler in der Helligkeit, und weil die Verteilung zur schwachen Seite steil ansteigt, füllen überkorrigierte Nebel die helle Spitze. Mit allen Nebeln wird $M^\star$ = −5.83, mit c < 1.5 −4.72, mit c < 1.0 −4.56.
</details>

<details><summary>13. Wie weit ist M 31 nach den Planetarischen Nebeln entfernt, und wie unabhängig ist das von den Cepheiden?</summary>

Mit $M^\star$ = −4.53 sind es 791 kpc (μ = 24.49 ± 0.10), 4 % mehr als die Cepheiden mit 761 kpc. Ganz unabhängig ist das nicht: $M^\star$ ist an Cepheiden-Galaxien geeicht und wurde ursprünglich an M 31 bestimmt. Mit der unabhängigen Eichung im Bulge sind es 862 kpc, mit großem Fehler.
</details>

#### **Nächste Schritte**

So würde eine Analyse in der Forschung weitergehen:
- **Temperaturen aus Spektren der Sterne:** NLTE-Modellatmosphären (TMAP, PoWR) an UV- und optische Spektren angepasst geben T und Schwerebeschleunigung ohne die Annahmen der Zanstra-Methode; mit Gaia-Entfernungen werden daraus Massen, die auf wenige hundertstel Sonnenmassen genau sind.
- **Neuere Entwicklungswege:** Die Spuren von Miller Bertolami (2016) reichen bis 0.53 M☉ und laufen 3- bis 10-mal schneller; sie würden die Sterne unter der leichtesten Spur erfassen und die beiden Uhren anders vergleichen. Dazu gehört ein Modell der Schale, die beschleunigt wird, statt R/v.
- **Ganze Populationen modellieren:** Statt jeden Stern einzeln einzuordnen, simuliert man die Nebel einer Galaxie (Anfangsmassen, Entwicklung, Sichtbarkeit) und vergleicht die vorhergesagte Leuchtkraftfunktion mit der gemessenen; so versucht man zu verstehen, warum $M^\star$ in alten Galaxien so hell ist.
- **Bessere Photometrie im Bulge:** Einheitliche [O III]-Aufnahmen (Kovacevic et al. 2011) und Rötung aus Radiokontinuum statt aus dem Balmer-Dekrement umgehen die Fehler der Spektren bei starker Rötung; Gaia DR4 wird Vordergrundnebel aussortieren.
- **Weiter hinaus:** Mit Integralfeld-Spektrographen (MUSE am VLT) findet man Planetarische Nebel bis in 40 Mpc Entfernung; Roth et al. (2021) haben damit Entfernungen zu Galaxien gemessen, die in die Bestimmung der Hubble-Konstante eingehen, als Gegenprobe zu Cepheiden und Supernovae vom Typ Ia.
