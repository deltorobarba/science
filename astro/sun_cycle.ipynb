# Der Sonnenfleckenzyklus

Das Notebook liest die monatliche Sonnenfleckenzahl von SILSO seit 1749, glättet sie selbst über 13 Monate und bestimmt daraus Minima, Maxima, Zykluslängen und mit einem Lomb-Scargle-Periodogramm die mittlere Periode. Jedem der 25 Zyklen wird die Kurvenform von Hathaway, Wilson & Reichmann (1994) angepasst; daraus folgen der Waldmeier-Effekt und eine Prognose für Zyklus 25 aus seinen ersten Jahren, verglichen mit der offiziellen geglätteten Reihe und der Vorhersage des NOAA/NASA-Panels von 2019.

Wie regelmäßig ist der elfjährige Zyklus der Sonne, und wie früh in einem Zyklus lässt sich sagen, wie stark er wird?

#### **Hintergrund**

Sonnenflecken sind dunkle Stellen in der Photosphäre, an denen starke Magnetfelder (einige tausend Gauß) die Konvektion bremsen und weniger Wärme nach oben kommt. Ihre Zahl schwankt mit einer Periode von etwa elf Jahren. Mit ihr schwanken fast alle Formen der Sonnenaktivität: Flares, koronale Massenauswürfe, die UV- und Röntgenstrahlung und damit auch das Weltraumwetter an der Erde.

Der Zyklus entsteht im Dynamo der Sonne. Differentielle Rotation wickelt ein schwaches Nord-Süd-Feld (poloidal) zu einem starken Ost-West-Feld (toroidal) auf; Flussröhren aus diesem Feld steigen auf und erscheinen als Fleckengruppen. Aus den zerfallenden Gruppen baut sich das poloidale Feld mit umgekehrter Polarität neu auf (Babcock-Leighton-Mechanismus). Nach elf Jahren hat das Magnetfeld deshalb die umgekehrte Polarität; der vollständige magnetische Zyklus (Hale-Zyklus) dauert 22 Jahre.

Die Sonnenfleckenzahl ist die längste fortlaufende Messreihe der Astronomie. Das World Data Center SILSO (Sunspot Index and Long-term Solar Observations) in Brüssel führt sie; es hat sie 2015 als Version 2 neu kalibriert. Die Zyklen werden ab dem Minimum von 1755 gezählt; Zyklus 25 begann im Dezember 2019.

#### **Herleitung**

**1. Die Sonnenfleckenzahl.** Rudolf Wolf führte 1848 die Größe
$$R = k\,(10\,g + f)$$
ein, mit der Zahl der Fleckengruppen $g$, der Zahl der Einzelflecken $f$ und einem Faktor $k$, der Teleskop, Beobachter und Seeing ausgleicht. Die Gruppen zählen zehnfach, weil eine neue Gruppe ein neues aktives Gebiet bedeutet, ein weiterer Fleck in einer bestehenden Gruppe nicht. SILSO bildet heute aus den Meldungen von rund 80 Stationen einen gewichteten Tagesmittelwert. In Version 2 entfiel der historische Zürcher Faktor 0.6, deshalb liegen die heutigen Werte etwa um den Faktor 1/0.6 ≈ 1.67 über älteren Veröffentlichungen (Clette et al. 2014).

**2. Die 13-Monats-Glättung.** SILSO glättet die Monatsmittel $R_m$ mit
$$\bar R_m = \frac{1}{24}\,R_{m-6} + \frac{1}{12}\sum_{j=-5}^{5} R_{m+j} + \frac{1}{24}\,R_{m+6} .$$
Das ist der Mittelwert zweier gleitender 12-Monats-Mittel, die um einen Monat gegeneinander verschoben sind; so liegt der Mittelpunkt genau auf dem Monat $m$. Ein gleitendes Mittel über $N = 12$ Werte multipliziert eine Schwingung mit der Frequenz $f$ (in 1/Monat) mit
$$H_{12}(f) = \frac{\sin(12\pi f)}{12\,\sin(\pi f)} ,$$
die Verschiebung um einen halben Monat nach links und rechts und das Mitteln der beiden Mittel ergeben einen weiteren Faktor $\cos(\pi f)$:
$$H(f) = \cos(\pi f)\,\frac{\sin(12\pi f)}{12\,\sin(\pi f)} .$$
$H$ verschwindet bei $f = k/12$, das heißt für Perioden von 12, 6, 4, 3 und 2.4 Monaten, und bei $f = 1/2$ (2 Monate). Für kleine $f$ gilt nach Taylorentwicklung $H \approx 1 - (\pi f)^2\,(N^2 + 2)/6 = 1 - 24.3\,(\pi f)^2$. Eine Schwingung mit 11 Jahren ($f = 1/132$) behält damit 98.6 % ihrer Amplitude. Der Preis: Für die letzten sechs Monate gibt es noch keinen geglätteten Wert. Ein Maximum steht deshalb frühestens ein halbes Jahr später fest.

**3. Das Lomb-Scargle-Periodogramm.** Für jede Testfrequenz $f$ passt man
$$y(t) = A\cos(2\pi f t) + B\sin(2\pi f t) + C$$
mit kleinsten Quadraten an die Daten an. Ist $\chi^2_0$ die Summe der quadrierten Abweichungen vom Mittelwert und $\chi^2(f)$ die Summe nach dem Fit, dann ist die normierte Leistung
$$P(f) = \frac{\chi^2_0 - \chi^2(f)}{\chi^2_0} ,$$
der Anteil der Varianz, den eine Sinusschwingung mit dieser Frequenz erklärt (Lomb 1976, Scargle 1982). Bei gleichmäßig abgetasteten Daten ist das bis auf die Normierung das klassische Periodogramm. Eine Reihe der Länge $T$ trennt Frequenzen, die mehr als etwa $1/T$ auseinanderliegen. Für $T = 277$ Jahre und $P = 11$ Jahre ist die Breite des Peaks in der Periode $\Delta P \approx P^2/T \approx 0.4$ Jahre.

**4. Die Kurvenform eines Zyklus.** Hathaway, Wilson & Reichmann (1994) beschreiben jeden Zyklus mit
$$F(t) = \frac{a\,x^3}{\exp(x^2/b^2) - c}, \qquad x = t - t_0 \ge 0 .$$
Kurz nach dem Start $t_0$ ist $\exp(x^2/b^2) \approx 1 + x^2/b^2$, also $F \approx a\,x^3/(1-c)$: Der Zyklus steigt kubisch an. Spät dominiert die Exponentialfunktion, und $F$ fällt wie $x^3 e^{-x^2/b^2}$ ab. $c$ legt fest, wie schief die Kurve ist; wie die Autoren halten wir es für alle Zyklen fest ($c = 0.71$). Für das Maximum setzt man die Ableitung null. Mit $u = x^2/b^2$ ist
$$\frac{dF}{dx} = a\,x^2\,\frac{3\left(e^u - c\right) - 2u\,e^u}{\left(e^u - c\right)^2} = 0 \quad\Longrightarrow\quad e^u\,(3 - 2u) = 3c .$$
Für $c = 0.71$ ist die Lösung $u^* = 1.17$. Damit liegt das Maximum bei $x_{\max} = \sqrt{u^*}\,b = 1.08\,b$ nach dem Start und hat die Höhe
$$F_{\max} = a\,b^3\,\frac{(u^*)^{3/2}}{e^{u^*} - c} = 0.50\,a\,b^3 .$$
Die Anstiegszeit hängt also nur von $b$ ab, die Höhe von $a\,b^3$.

**5. Der Waldmeier-Effekt.** Waldmeier (1935) bemerkte, dass starke Zyklen schneller ansteigen als schwache. Im Modell heißt das: Große $a$ gehen mit kleinen $b$ einher. Hathaway et al. fanden einen engen Zusammenhang der Form $b = b_0 + b_1\,a^{-1/4}$. Kennt man ihn, hat die Kurve nur noch die Parameter $a$ und $t_0$, und die ersten Jahre eines Zyklus legen seine Höhe fest. Ob eine Korrelation $r$ aus $n$ Wertepaaren zufällig sein kann, prüft
$$t = r\,\sqrt{\frac{n-2}{1-r^2}} ,$$
das ohne echten Zusammenhang einer Student-$t$-Verteilung mit $n-2$ Freiheitsgraden folgt.

#### **Daten**

Zwei Reihen von SILSO (Royal Observatory of Belgium), beide in Version 2:
- **Monatsmittel** der täglichen Sonnenfleckenzahl ab Januar 1749 (`SN_m_tot_V2.0`), mit Standardabweichung, Zahl der Beobachtungen und einer Marke, ob der Wert endgültig oder vorläufig ist.
- **Offizielle 13-Monats-Glättung** derselben Reihe (`SN_ms_tot_V2.0`), die SILSO für die Daten der Minima und Maxima verwendet. Fehlende Werte stehen als −1 in der Datei.

Die Dateien landen beim ersten Lauf in `astrodata/silso`. SILSO ergänzt am Anfang jedes Monats den Vormonat; mit `REFRESH = True` lädt das Notebook die Reihen neu.

```python
import urllib.request
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.signal import find_peaks
from scipy.optimize import curve_fit, brentq
from scipy.stats import t as student_t
from astropy.timeseries import LombScargle

DATA = Path.cwd() / "astrodata" / "silso"
DATA.mkdir(parents=True, exist_ok=True)
REFRESH = False                       # True: Reihen neu von SILSO laden

def silso(name):
    """Eine SILSO-Reihe lesen; fehlende Werte (-1) werden zu NaN."""
    path = DATA / f"{name}.csv"
    if REFRESH or not path.exists():
        urllib.request.urlretrieve(f"https://www.sidc.be/SILSO/INFO/{name}.php", path)
    df = pd.read_csv(path, sep=";", names=["jahr", "monat", "t", "sn", "sd", "n_beob", "definitiv"])
    df.loc[df.sn < 0, "sn"] = np.nan
    return df

monthly = silso("snmtotcsv")          # Monatsmittel
official = silso("snmstotcsv")        # offizielle 13-Monats-Glättung
t, sn = monthly.t.values, monthly.sn.values

last = monthly.iloc[-1]
last_def = monthly[monthly.definitiv == 1].iloc[-1]
last_sm = official.dropna(subset=["sn"]).iloc[-1]
def ym(row):
    return f"{int(row.jahr)}-{int(row.monat):02d}"
print(f"Monatsmittel:  {len(monthly)} Monate, {ym(monthly.iloc[0])} bis {ym(last)} (letzter Wert {last.sn:.1f})")
print(f"Endgültig bis: {ym(last_def)}; danach vorläufig")
print(f"Offiziell geglättet bis {ym(last_sm)} ({last_sm.sn:.1f})")
print(f"Höchstes Monatsmittel: {np.nanmax(sn):.1f} ({t[np.nanargmax(sn)]:.2f})")
```

#### **Reduktion**

##### *1. Die eigene 13-Monats-Glättung*

Liefert die Formel aus der Herleitung dieselben Werte wie die offizielle Reihe, und was kostet die Glättung an Amplitude?

```python
w = np.r_[0.5, np.ones(11), 0.5] / 12
own = np.convolve(sn, w, mode="same")
own[:6] = own[-6:] = np.nan               # am Rand fehlen Nachbarmonate

diff = own - official.sn.values
print(f"Eigene minus offizielle Glättung: max |Δ| = {np.nanmax(np.abs(diff)):.3f}, "
      f"RMS = {np.sqrt(np.nanmean(diff ** 2)):.3f}  (SILSO rundet auf 0.1)")

# Übertragungsfunktion aus der Herleitung
def H(f):
    return np.cos(np.pi * f) * np.sin(12 * np.pi * f) / (12 * np.sin(np.pi * f))
for months in (132, 60, 24, 12, 6):
    print(f"Periode {months:>3} Monate: H = {H(1 / months):+.3f}")
peak_month, peak_smooth = np.nanmax(sn), np.nanmax(own)
print(f"Höchstes Monatsmittel {peak_month:.1f} gegen höchsten geglätteten Wert {peak_smooth:.1f}")

fig, axes = plt.subplots(1, 2, figsize=(13, 3.8), gridspec_kw={"width_ratios": [2, 1]})
sel = t > 1996
axes[0].plot(t[sel], sn[sel], c="0.7", lw=0.8, label="Monatsmittel")
axes[0].plot(t[sel], own[sel], c="black", lw=1.8, label="eigene 13-Monats-Glättung")
axes[0].plot(official.t[sel], official.sn[sel], c="tab:orange", lw=1, ls="--", label="offizielle Glättung")
axes[0].set_xlabel("Jahr")
axes[0].set_ylabel("Sonnenfleckenzahl")
axes[0].set_title("Zyklen 23 bis 25")
axes[0].legend(fontsize=8)
f = np.linspace(1e-4, 0.5, 2000)
axes[1].plot(1 / f, np.abs(H(f)), c="black")
axes[1].set_xscale("log")
axes[1].set_xlabel("Periode [Monate]")
axes[1].set_ylabel("|H|")
axes[1].set_title("Übertragungsfunktion der Glättung")
for p in (12, 6, 4, 3):
    axes[1].axvline(p, c="tab:red", lw=0.5, ls=":")
fig.tight_layout()
plt.show()
```

#### **Eigene Messung**

##### *2. Minima, Maxima und Zykluslängen*

Wann lagen die Minima und Maxima der letzten 277 Jahre, und wie stark schwanken Länge und Höhe der Zyklen?

```python
ok = np.isfinite(own)
tt, yy = t[ok], own[ok]
# Ein Zyklus dauert mindestens 7 Jahre; die Prominenz 30 trennt echte Extrema von Zacken
i_max, _ = find_peaks(yy, distance=7 * 12, prominence=30)
i_min, _ = find_peaks(-yy, distance=7 * 12, prominence=30)
t_min, a_min = tt[i_min], yy[i_min]
t_max, a_max = tt[i_max], yy[i_max]

rows = []
for k, tm in enumerate(t_min):                       # Minimum k eröffnet Zyklus k+1 (Zyklus 1 ab 1755)
    nxt = t_min[k + 1] if k + 1 < len(t_min) else np.nan
    m = (t_max > tm) & ((t_max < nxt) | np.isnan(nxt))
    rows.append(dict(zyklus=k + 1, t_min=tm, sn_min=a_min[k],
                     t_max=t_max[m][0] if m.any() else np.nan, sn_max=a_max[m][0] if m.any() else np.nan,
                     laenge=nxt - tm))
cyc = pd.DataFrame(rows)
cyc["anstieg"] = cyc.t_max - cyc.t_min
cyc["abfall"] = cyc.laenge - cyc.anstieg
pd.set_option("display.width", 160)
print(cyc.round(2).to_string(index=False))

done = cyc.dropna(subset=["laenge"])
print(f"\n{len(done)} abgeschlossene Zyklen: Länge {done.laenge.mean():.2f} ± {done.laenge.std(ddof=1):.2f} Jahre "
      f"(kürzester {done.laenge.min():.1f}, längster {done.laenge.max():.1f})")
print(f"Anstieg {done.anstieg.mean():.2f} ± {done.anstieg.std(ddof=1):.2f} Jahre, "
      f"Abfall {done.abfall.mean():.2f} ± {done.abfall.std(ddof=1):.2f} Jahre")
print(f"Maxima: {cyc.sn_max.min():.0f} (Zyklus {cyc.zyklus[cyc.sn_max.idxmin()]}) bis "
      f"{cyc.sn_max.max():.0f} (Zyklus {cyc.zyklus[cyc.sn_max.idxmax()]})")

fig, ax = plt.subplots(figsize=(14, 4))
ax.plot(t, sn, c="0.75", lw=0.5)
ax.plot(t, own, c="black", lw=1.2)
ax.plot(t_min, a_min, "v", c="tab:blue", ms=5, label="Minimum")
ax.plot(t_max, a_max, "^", c="tab:red", ms=5, label="Maximum")
for _, r in cyc.dropna(subset=["t_max"]).iterrows():
    ax.text(r.t_max, r.sn_max + 25, f"{r.zyklus:.0f}", ha="center", fontsize=8)
ax.axvspan(1790, 1830, color="tab:blue", alpha=0.08, label="Dalton-Minimum")
ax.set_xlabel("Jahr")
ax.set_ylabel("Sonnenfleckenzahl")
ax.set_title("Sonnenfleckenzahl 1749 bis heute mit Zyklusnummern")
ax.legend(fontsize=8, loc="upper left")
fig.tight_layout()
plt.show()
```

##### *3. Das Periodogramm*

Welche Periode findet ein Sinusfit über die ganze Reihe, und wie passt sie zur Streuung der einzelnen Zykluslängen?

```python
good = np.isfinite(sn)
freq = np.linspace(1 / 150, 1 / 3, 20000)          # Perioden 3 bis 150 Jahre
power = LombScargle(t[good], sn[good]).power(freq)
P_best = 1 / freq[np.argmax(power)]
half = power > power.max() / 2
near = np.abs(1 / freq - P_best) < 3
fwhm = np.ptp(1 / freq[half & near])
T_total = np.ptp(t[good])
print(f"Höchster Peak: P = {P_best:.2f} Jahre, Leistung {power.max():.2f}")
print(f"Halbwertsbreite des Peaks: {fwhm:.2f} Jahre; erwartet etwa P²/T = {P_best ** 2 / T_total:.2f} Jahre")
print(f"Mittlere Zykluslänge (Schritt 2): {done.laenge.mean():.2f} Jahre, Streuung ± {done.laenge.std(ddof=1):.2f}")

# weitere lokale Maxima
pk, _ = find_peaks(power, prominence=0.01)
top = pk[np.argsort(power[pk])[::-1][:6]]
print("Die sechs höchsten Peaks:", ", ".join(f"{1 / freq[i]:.1f} J ({power[i]:.2f})" for i in top))

fig, ax = plt.subplots(figsize=(11, 3.8))
ax.plot(1 / freq, power, c="black", lw=1)
ax.axvspan(done.laenge.min(), done.laenge.max(), color="tab:orange", alpha=0.15,
           label="Spanne der einzelnen Zykluslängen")
ax.axvline(P_best, c="tab:red", ls="--", lw=1, label=f"Peak {P_best:.2f} Jahre")
ax.set_xscale("log")
ax.set_xlabel("Periode [Jahre]")
ax.set_ylabel("Lomb-Scargle-Leistung")
ax.set_title("Periodogramm der monatlichen Sonnenfleckenzahl")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
```

#### **Modell und Fit**

##### *4. Die Kurvenform jedes Zyklus*

Beschreibt die Formel von Hathaway et al. alle Zyklen von 1755 bis heute, und wie unterscheiden sich die angepassten Parameter?

```python
C_SHAPE = 0.71

def hathaway(x_t, a, b, t0, c=C_SHAPE):
    x = np.clip(x_t - t0, 0, None)
    with np.errstate(over="ignore"):
        return a * x ** 3 / (np.exp(x ** 2 / b ** 2) - c)

# Maximum der Kurve aus der Herleitung: e^u (3 - 2u) = 3c
u_star = brentq(lambda u: np.exp(u) * (3 - 2 * u) - 3 * C_SHAPE, 0.1, 1.49)
k_rise, k_amp = np.sqrt(u_star), u_star ** 1.5 / (np.exp(u_star) - C_SHAPE)
print(f"u* = {u_star:.3f}:  Maximum bei t0 + {k_rise:.3f}·b,  Höhe {k_amp:.3f}·a·b³")

fits = []
for k, r in cyc.iterrows():
    hi = r.t_min + r.laenge if np.isfinite(r.laenge) else t[-1] + 0.01
    m = (t >= r.t_min) & (t < hi) & np.isfinite(sn)
    p, cov = curve_fit(hathaway, t[m], sn[m], p0=[3, 4, r.t_min - 0.3], maxfev=20000)
    resid = sn[m] - hathaway(t[m], *p)
    fits.append(dict(zyklus=r.zyklus, a=p[0], b=p[1], t0=p[2], a_err=np.sqrt(cov[0, 0]),
                     t_max_fit=p[2] + k_rise * p[1], A_fit=k_amp * p[0] * p[1] ** 3,
                     rms=np.sqrt(np.mean(resid ** 2)), n=m.sum()))
fits = pd.DataFrame(fits)
print(fits.round(2).to_string(index=False))
print(f"\nMittel über die Zyklen 1-24: b = {fits.b[:24].mean():.2f} ± {fits.b[:24].std(ddof=1):.2f} Jahre, "
      f"Streuung der Monatswerte um die Kurve {fits.rms[:24].mean():.0f}")

fig, axes = plt.subplots(5, 5, figsize=(15, 11), sharey=True)
for ax, (_, r), (_, fp) in zip(axes.flat, cyc.iterrows(), fits.iterrows()):
    hi = r.t_min + r.laenge if np.isfinite(r.laenge) else t[-1]
    m = (t >= r.t_min - 1) & (t <= hi + 1)
    ax.plot(t[m], sn[m], c="0.7", lw=0.6)
    ax.plot(t[m], own[m], c="black", lw=1)
    tf = np.linspace(r.t_min, hi, 300)
    ax.plot(tf, hathaway(tf, fp.a, fp.b, fp.t0), c="tab:red", lw=1.5)
    ax.set_title(f"Zyklus {r.zyklus:.0f} ({r.t_min:.0f})", fontsize=9)
    ax.tick_params(labelsize=7)
fig.suptitle("Monatsmittel (grau), 13-Monats-Glättung (schwarz) und Hathaway-Kurve (rot)")
fig.tight_layout()
plt.show()
```

##### *5. Der Waldmeier-Effekt*

Steigen starke Zyklen schneller an, und lässt sich das als Zusammenhang zwischen $a$ und $b$ fassen?

```python
d = cyc.merge(fits, on="zyklus")[:24]               # nur abgeschlossene Zyklen
n = len(d)

def corr_test(x, y):
    r = np.corrcoef(x, y)[0, 1]
    tval = r * np.sqrt((n - 2) / (1 - r ** 2))
    return r, 2 * student_t.sf(abs(tval), n - 2)

r1, p1 = corr_test(d.anstieg, d.sn_max)
r2, p2 = corr_test(d.laenge, d.sn_max)
print(f"Anstiegszeit gegen Maximum: r = {r1:+.2f}, p = {p1:.1e}  (n = {n})")
print(f"Zykluslänge gegen Maximum:  r = {r2:+.2f}, p = {p2:.2f}")

X = np.c_[np.ones(n), d.a ** -0.25]
(b0, b1), *_ = np.linalg.lstsq(X, d.b, rcond=None)
b_rms = np.std(d.b - X @ np.r_[b0, b1], ddof=2)
print(f"Zusammenhang b(a) = {b0:.2f} + {b1:.2f}·a^(-1/4)  [Jahre], Streuung ± {b_rms:.2f} Jahre")

fig, axes = plt.subplots(1, 2, figsize=(12, 4.2))
axes[0].scatter(d.sn_max, d.anstieg, c="black", s=18)
for _, r in d.iterrows():
    axes[0].annotate(f"{r.zyklus:.0f}", (r.sn_max, r.anstieg), fontsize=7, xytext=(3, 2), textcoords="offset points")
cf = np.polyfit(d.sn_max, d.anstieg, 1)
xs = np.linspace(d.sn_max.min(), d.sn_max.max(), 10)
axes[0].plot(xs, np.polyval(cf, xs), c="tab:red", lw=1, label=f"Gerade, r = {r1:.2f}")
axes[0].set_xlabel("Maximum der geglätteten Sonnenfleckenzahl")
axes[0].set_ylabel("Anstieg Minimum → Maximum [Jahre]")
axes[0].set_title("Waldmeier-Effekt")
axes[0].legend(fontsize=8)
axes[1].scatter(d.a, d.b, c="black", s=18)
aa = np.linspace(d.a.min() * 0.9, d.a.max() * 1.05, 100)
axes[1].plot(aa, b0 + b1 * aa ** -0.25, c="tab:red", lw=1, label="b = b0 + b1·a^(-1/4)")
axes[1].set_xlabel("a [Sonnenflecken / Jahr³]")
axes[1].set_ylabel("b [Jahre]")
axes[1].set_title("Die Formparameter hängen zusammen")
axes[1].legend(fontsize=8)
fig.tight_layout()
plt.show()
```

##### *6. Prognose für Zyklus 25 aus seinen ersten Jahren*

Mit dem Zusammenhang $b(a)$ aus Schritt 5 hat die Kurve nur noch zwei freie Parameter, $a$ und $t_0$. Wie gut sagen die ersten 1.5, 2, 3 und 4 Jahre nach dem Minimum vom Dezember 2019 das Maximum voraus?

```python
def hathaway1(x_t, a, t0):
    return hathaway(x_t, a, b0 + b1 * a ** -0.25, t0)

t_start25 = cyc.t_min.iloc[24]
full = fits.iloc[24]
truth_t, truth_A = cyc.t_max.iloc[24], cyc.sn_max.iloc[24]
tf = np.linspace(t_start25, t_start25 + 11, 600)
fig, ax = plt.subplots(figsize=(11, 4.2))
m_all = t >= t_start25 - 0.5
ax.plot(t[m_all], sn[m_all], c="0.7", lw=0.8, label="Monatsmittel")
ax.plot(t[m_all], own[m_all], c="black", lw=1.5, label="13-Monats-Glättung")
forecasts = []
for years, col in zip((1.5, 2, 3, 4), ("tab:purple", "tab:blue", "tab:green", "tab:orange")):
    m = (t >= t_start25) & (t < t_start25 + years)
    p, cov = curve_fit(hathaway1, t[m], sn[m], p0=[3, t_start25 - 0.3], bounds=([0.05, t_start25 - 3], [50, t_start25 + 2]))
    f = hathaway1(tf, *p)
    # Unsicherheit der Höhe aus der Kovarianz (Monte Carlo über a und t0)
    draws = np.random.default_rng(1).multivariate_normal(p, cov, 400)
    draws = draws[draws[:, 0] > 0]
    peaks = [hathaway1(tf, *q).max() for q in draws]
    forecasts.append(dict(jahre=years, a=p[0], t0=p[1], maximum=f.max(), fehler=np.std(peaks), zeitpunkt=tf[np.argmax(f)]))
    ax.plot(tf, f, c=col, lw=1.3, label=f"Prognose aus {years} Jahren: {f.max():.0f} ± {np.std(peaks):.0f}")
    ax.axvline(t_start25 + years, c=col, lw=0.6, ls=":")
ax.plot(tf, hathaway(tf, full.a, full.b, full.t0), c="tab:red", lw=1, ls="--", label=f"Fit an alle Daten: {full.A_fit:.0f}")
ax.plot(2025 + 6.5 / 12, 115, "s", c="tab:brown", ms=8, label="NOAA/NASA-Panel 2019: 115 im Juli 2025")
ax.set_xlabel("Jahr")
ax.set_ylabel("Sonnenfleckenzahl")
ax.set_title("Zyklus 25: Prognosen aus den ersten Jahren")
ax.legend(fontsize=8, loc="upper left")
fig.tight_layout()
plt.show()

fc = pd.DataFrame(forecasts)
print(fc.round(2).to_string(index=False))
print(f"\nBeobachtet (geglättet): Maximum {truth_A:.1f} bei {truth_t:.2f};  Fit an alle Daten: {full.A_fit:.1f} bei {full.t_max_fit:.2f}")
```

- **Kurvenform:** Die Hathaway-Kurve beschreibt alle 25 Zyklen; die Monatswerte streuen im Mittel um 25 um sie (12 bis 36). $b$ liegt bei 4.35 ± 0.55 Jahren, $a$ schwankt um den Faktor zehn (1.3 bis 13.7). So würde eine Analyse in der Forschung weitergehen:
- **Vorläufer-Methoden:** Die Stärke des polaren Magnetfelds im Minimum ist der beste bekannte Vorhersagewert für das folgende Maximum (Messreihe des Wilcox Solar Observatory seit 1976). Sie wirkt Jahre vor dem Kurvenfit.
- **Schmetterlingsdiagramm:** Mit den Positionen einzelner Fleckengruppen (Royal Greenwich Observatory 1874–1976, danach USAF/NOAA) sieht man, wie die Flecken im Lauf eines Zyklus von etwa 30° Breite zum Äquator wandern (Spörers Gesetz).
