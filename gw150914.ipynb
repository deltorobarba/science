# GW150914

Das Notebook lädt die LIGO-Rohdaten von H1 und L1 über GWOSC, macht das Signal mit Rauschspektrum, Whitening, Bandpass und Q-Transformation sichtbar und misst die Chirp-Masse, indem es die Newtonsche Chirp-Kurve an die Q-Transformation anpasst. Das Ergebnis wird mit dem Katalogwert verglichen.

Wie sieht das erste direkt gemessene Gravitationswellensignal in den Daten der beiden LIGO-Detektoren aus, welche Schritte machen es sichtbar, und welche Chirp-Masse ergibt sich aus dem Frequenzverlauf?

#### **Hintergrund**

- Am 14. September 2015 um 09:50:45 UTC registrierten beide LIGO-Detektoren, Hanford (H1) und Livingston (L1), das Signal zweier verschmelzender schwarzer Löcher. Die ursprüngliche Veröffentlichung nennt etwa 36 und 29 Sonnenmassen und eine Entfernung von etwa 410 Mpc.
- Gemessen wird der Strain $h(t) = \Delta L/L$, die relative Längenänderung der 4 km langen Arme. Das Signal erreicht $h\approx10^{-21}$.
- Die Detektoren liegen etwa 3000 km auseinander. Ein Signal mit Lichtgeschwindigkeit braucht von einem zum anderen höchstens etwa 10 ms.
- Beim Einspiralen steigt die Frequenz (Chirp). In niedrigster, Newtonscher Ordnung gilt mit der Zeit bis zur Verschmelzung $\tau = t_c - t$
$$f(\tau) = \frac1\pi\left(\frac{5}{256\,\tau}\right)^{3/8}\left(\frac{G\mathcal M}{c^3}\right)^{-5/8},\qquad \mathcal M = \frac{(m_1m_2)^{3/5}}{(m_1+m_2)^{1/5}} .$$
- Daraus folgt, dass $f^{-8/3}$ linear in $t$ fällt:
$$f^{-8/3}(t) = \frac{256}{5}\,\pi^{8/3}\left(\frac{G\mathcal M}{c^3}\right)^{5/3}(t_c - t).$$
  Die Steigung dieser Geraden bestimmt die Chirp-Masse $\mathcal M$. Gemessen wird so die Chirp-Masse im Detektorsystem; sie ist um den Faktor $(1+z)$ größer als die im Ruhesystem der Quelle.
- Im Code wird die ganze Kurve $f(t)$ an die Q-Transformation angepasst. Für jedes Paar $(\mathcal M, t_c)$ wird die Energie entlang der vorhergesagten Kurve gemittelt, und das Paar mit der größten mittleren Energie ist die Schätzung. Das ist eine einfache Hough-Transformation; Matched Filtering macht dasselbe mit der vollen Wellenform statt nur mit ihrer Frequenz.

#### **Herleitung**

**1. Warum das Signal in den Rohdaten unsichtbar ist.** Das Rauschspektrum (Abschnitt 2) liegt bei 10 Hz bei etwa $3\cdot10^{-20}/\sqrt{\mathrm{Hz}}$ und bei 100 Hz bei etwa $10^{-23}/\sqrt{\mathrm{Hz}}$, mehr als drei Größenordnungen tiefer. Die Rohdaten werden deshalb von den tiefen Frequenzen beherrscht, vor allem von seismischem Rauschen: Ihre Standardabweichung ist etwa $2\cdot10^{-19}$, rund 200-mal mehr als das Signal mit $h\approx10^{-21}$. Im Band von 35 bis 350 Hz, in dem das Signal liegt, ergibt das Integral über das Spektrum, $\big(\int S(f)\,df\big)^{1/2}$ mit der spektralen Dichte $S$, ein Rauschen von etwa $10^{-21}$, also so viel wie das Signal. Whitening und Bandpass entfernen die tiefen Frequenzen und teilen jede Frequenz durch ihr Rauschen. Danach ragt das Signal über mehrere Zyklen zwei- bis dreimal über das Rauschen hinaus, in beiden Detektoren zur gleichen Zeit.

**2. Die Gerade.** Mit $X = G\mathcal M/c^3$ und $\tau = t_c - t$ ergibt die Formel $f = \frac1\pi\left(\frac{5}{256\,\tau}\right)^{3/8}X^{-5/8}$, mit $-8/3$ potenziert,
$$f^{-8/3} = \pi^{8/3}\,\frac{256\,\tau}{5}\,X^{5/3} = \frac{256}{5}\,\pi^{8/3}X^{5/3}\,(t_c - t).$$
Das ist eine Gerade in $t$ mit der Steigung $s = -\frac{256}{5}\pi^{8/3}X^{5/3}$. Aufgelöst:
$$X = \left(-\frac{5\,s}{256\,\pi^{8/3}}\right)^{3/5},\qquad \mathcal M = \frac{c^3}{G}\,X.$$

**3. Der Chirp aus Kepler-Gesetz und Quadrupolformel.** Zwei Massen $m_1, m_2$ umkreisen einander im Abstand $a$; $M = m_1 + m_2$ ist die Gesamtmasse und $\mu = m_1m_2/M$ die reduzierte Masse.
- Kepler: $\omega^2 = GM/a^3$, also $a = (GM)^{1/3}\omega^{-2/3}$.
- Bahnenergie: $E = -\dfrac{GM\mu}{2a}$.
- Abgestrahlte Leistung nach der Quadrupolformel: $P = \dfrac{32}{5}\dfrac{G}{c^5}\mu^2a^4\omega^6 = \dfrac{32}{5}\dfrac{G^4\mu^2M^3}{c^5a^5}$.
- Energiebilanz $\dot E = -P$: $\dfrac{GM\mu}{2a^2}\,\dot a = -\dfrac{32}{5}\dfrac{G^4\mu^2M^3}{c^5a^5}$, also $\dot a = -\dfrac{64}{5}\dfrac{G^3\mu M^2}{c^5a^3}$. Der Abstand schrumpft.
- Aus $\omega\propto a^{-3/2}$ folgt $\dot\omega = -\frac32\,\omega\,\dot a/a = \frac{96}{5}\frac{G^3\mu M^2}{c^5a^4}\,\omega$. Mit $a^4 = (GM)^{4/3}\omega^{-8/3}$ wird daraus
$$\dot\omega = \frac{96}{5}\,\frac{G^{5/3}\mu M^{2/3}}{c^5}\,\omega^{11/3} = \frac{96}{5}\,X^{5/3}\,\omega^{11/3},\qquad X = \frac{G\mathcal M}{c^3},\quad \mathcal M^{5/3} = \mu M^{2/3}.$$
  Hier entsteht die Chirp-Masse: Nur die Kombination $\mu M^{2/3}$ bestimmt den Frequenzanstieg. Aus $\mathcal M = \mu^{3/5}M^{2/5}$ folgt die Formel im Hintergrund.
- Trennung der Variablen: $\omega^{-11/3}\,d\omega = \frac{96}{5}X^{5/3}\,dt$ ergibt $-\frac38\,\omega^{-8/3} = \frac{96}{5}X^{5/3}\,t + C$, also $\omega^{-8/3} = \frac{256}{5}X^{5/3}(t_c - t)$ mit der Verschmelzungszeit $t_c$.
- Die Welle schwingt mit der doppelten Bahnfrequenz: $f = 2\cdot\frac{\omega}{2\pi} = \omega/\pi$. Mit $\omega = \pi f$ folgt $f^{-8/3} = \frac{256}{5}\pi^{8/3}X^{5/3}\tau$, und aufgelöst die Formel für $f(\tau)$.

**4. Zeitversatz und Vorzeichen.** Hanford und Livingston liegen etwa 3000 km auseinander. Eine Welle mit Lichtgeschwindigkeit braucht für diese Strecke höchstens $3\cdot10^6\,\mathrm m/(3\cdot10^8\,\mathrm{m/s}) = 10$ ms; das Maximum gilt, wenn sie entlang der Verbindungslinie einfällt. Bei GW150914 betrug der Versatz etwa 7 ms, L1 zuerst. Zum Vorzeichen: Ein Detektor misst die Differenz der Längenänderungen seiner beiden Arme. Die Arme von H1 und L1 sind gegeneinander um etwa 90° gedreht, dazu kommt die Erdkrümmung zwischen den Standorten. Vertauscht man die Rollen der Arme, kehrt sich das Vorzeichen um; für die Himmelsrichtung von GW150914 haben die beiden Detektoren deshalb nahezu entgegengesetzte Antwort, und das L1-Signal erscheint invertiert.

```python
import numpy as np
import matplotlib.pyplot as plt
from gwosc.api import fetch_event_json
from gwosc.datasets import event_gps
from gwpy.timeseries import TimeSeries

EREIGNIS = "GW150914"
gps = event_gps(EREIGNIS)
print("GPS-Zeit des Ereignisses:", gps)

daten = {}
for ifo in ["H1", "L1"]:
    daten[ifo] = TimeSeries.fetch_open_data(ifo, gps - 16, gps + 16, sample_rate=4096, cache=True)
    print(f"{ifo}: {daten[ifo].duration.value:.0f} s, {daten[ifo].sample_rate.value:.0f} Hz")
```

##### *1. Rohdaten*

Ist das Signal in den Rohdaten zu sehen?

```python
fig, ax = plt.subplots(figsize=(10, 3.5))
for ifo in daten:
    h = daten[ifo]
    ax.plot(h.times.value - gps, h.value, lw=0.6, label=ifo)
ax.set_xlabel("Zeit relativ zum Ereignis [s]")
ax.set_ylabel("Strain")
ax.set_title("GW150914: Rohdaten")
ax.legend()
plt.show()
```

##### *2. Rauschspektrum*

Bei welchen Frequenzen sind die Detektoren am empfindlichsten, und welche schmalen Linien fallen auf?

```python
fig, ax = plt.subplots(figsize=(9, 4.5))
for ifo in daten:
    asd = daten[ifo].asd(fftlength=4, overlap=2)
    ax.loglog(asd.frequencies.value, asd.value, lw=0.8, label=ifo)
ax.set_xlim(10, 2000)
ax.set_ylim(1e-24, 1e-19)
ax.set_xlabel("Frequenz [Hz]")
ax.set_ylabel("Amplitudenspektraldichte [1/√Hz]")
ax.set_title("Rauschspektrum um GW150914")
ax.grid(True, which="both", alpha=0.3)
ax.legend()
plt.show()
```

##### *3. Whitening und Bandpass*

Was bleibt übrig, wenn man jede Frequenz durch ihr Rauschen teilt (Whitening) und nur 35 bis 350 Hz behält? Ist das Signal in beiden Detektoren zu sehen, mit dem erwarteten Zeitversatz?

```python
gefiltert = {ifo: daten[ifo].whiten(4, 2).bandpass(35, 350) for ifo in daten}
VERSATZ = 0.0069        # Sekunden: L1 sah das Signal etwa 7 ms vor H1

h1 = gefiltert["H1"].crop(gps - 0.2, gps + 0.05)
l1 = gefiltert["L1"].crop(gps - 0.2, gps + 0.05)

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(h1.times.value - gps, h1.value, label="H1")
ax.plot(l1.times.value - gps + VERSATZ, -l1.value, label="L1, um 6.9 ms verschoben und invertiert")
ax.set_xlim(-0.2, 0.05)
ax.set_xlabel("Zeit relativ zum Ereignis [s]")
ax.set_ylabel("Strain, geweißt (Einheiten des Rauschens)")
ax.set_title("GW150914 nach Whitening und Bandpass")
ax.legend()
plt.show()
```

##### *4. Zeit-Frequenz-Bild*

Wie steigt die Frequenz bis zur Verschmelzung an?

```python
qt = daten["H1"].q_transform(outseg=(gps - 0.3, gps + 0.05), qrange=(4, 64), frange=(20, 500))

plot = qt.plot(figsize=(10, 4.5))
ax = plot.gca()
ax.set_xscale("seconds", epoch=gps)
ax.set_yscale("log")
ax.set_ylabel("Frequenz [Hz]")
ax.set_title("H1: Q-Transformation um GW150914")
ax.colorbar(label="Normierte Energie", clim=(0, 25))
plt.show()
```

##### *5. Eigene Messung: die Chirp-Masse*

Welche Newtonsche Chirp-Kurve passt am besten zur Q-Transformation, und welche Chirp-Masse gehört zu ihr?

```python
G, c, M_SONNE = 6.674e-11, 2.998e8, 1.989e30


def chirp_frequenz(tau, M_chirp):
    """Newtonscher Chirp: Frequenz tau Sekunden vor der Verschmelzung, Chirp-Masse in Sonnenmassen."""
    X = G * M_chirp * M_SONNE / c**3                  # G·M/c³ in Sekunden
    return (1 / np.pi) * (5 / (256 * tau)) ** (3 / 8) * X ** (-5 / 8)


def chirp_suche(qt, massen, zeiten_c, tau_min=0.01, tau_max=0.15):
    """Mittlere Energie der Q-Transformation entlang der vorhergesagten Kurve, für jedes Paar (M, t_c)."""
    t = qt.times.value - gps
    f = qt.frequencies.value
    E = qt.value                                       # Form: (Zeit, Frequenz)
    S = np.zeros((len(massen), len(zeiten_c)))
    for a, M in enumerate(massen):
        for b, tc in enumerate(zeiten_c):
            tau = tc - t
            m = (tau > tau_min) & (tau < tau_max)      # nur der Teil vor der Verschmelzung
            fi = np.searchsorted(f, chirp_frequenz(tau[m], M)).clip(0, len(f) - 1)
            S[a, b] = E[np.where(m)[0], fi].mean()
    return S


massen = np.arange(15, 60.1, 0.5)                      # Chirp-Masse in Sonnenmassen
zeiten_c = np.arange(-0.02, 0.05, 0.001)               # Verschmelzungszeit relativ zu gps in Sekunden
S = chirp_suche(qt, massen, zeiten_c)
a, b = np.unravel_index(np.argmax(S), S.shape)
M_chirp, t_c = massen[a], zeiten_c[b]
print(f"H1: Chirp-Masse im Detektorsystem {M_chirp:.1f} Sonnenmassen, Verschmelzung bei {1000 * t_c:.0f} ms")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
im = ax1.pcolormesh(1000 * zeiten_c, massen, S, shading="auto")
ax1.plot(1000 * t_c, M_chirp, "w+", ms=14, mew=2)
ax1.set_xlabel("Verschmelzungszeit t_c [ms]")
ax1.set_ylabel("Chirp-Masse [Sonnenmassen]")
ax1.set_title("Mittlere Energie entlang der Chirp-Kurve")
fig.colorbar(im, ax=ax1)

t = qt.times.value - gps
ax2.pcolormesh(t, qt.frequencies.value, qt.value.T, shading="auto", vmax=25)
tau = np.linspace(0.15, 0.01, 200)
ax2.plot(t_c - tau, chirp_frequenz(tau, M_chirp), "w--", lw=2, label=f"Newtonscher Chirp, {M_chirp:.1f} Sonnenmassen")
ax2.set_yscale("log")
ax2.set_ylim(20, 500)
ax2.set_xlabel("Zeit relativ zum Ereignis [s]")
ax2.set_ylabel("Frequenz [Hz]")
ax2.set_title("H1: Q-Transformation mit der besten Chirp-Kurve")
ax2.legend(loc="upper left")
plt.tight_layout()
plt.show()
```

**Grobe Unsicherheit.** Dieselbe Suche für beide Detektoren und zwei Zeitbereiche vor der Verschmelzung. Vorher überlegen: Welche Werte erwartest du, und welcher Bereich reicht näher an die Verschmelzung heran?

```python
werte = []
for ifo in ["H1", "L1"]:
    q = daten[ifo].q_transform(outseg=(gps - 0.3, gps + 0.05), qrange=(4, 64), frange=(20, 500))
    for tau_min, tau_max in [(0.01, 0.15), (0.02, 0.2)]:
        S_ = chirp_suche(q, massen, zeiten_c, tau_min, tau_max)
        M_ = massen[np.unravel_index(np.argmax(S_), S_.shape)[0]]
        werte.append(M_)
        print(f"{ifo}, tau von {tau_min} bis {tau_max} s: {M_:.1f} Sonnenmassen")
M_mittel, M_streuung = np.mean(werte), np.std(werte)
print(f"Eigene Messung: {M_mittel:.1f} ± {M_streuung:.1f} Sonnenmassen (Mittelwert und Streuung der vier Werte)")
```

##### *6. Eigene Messung gegen den Katalog*

Wie weit liegt die eigene Chirp-Masse vom Katalogwert entfernt? Mögliche Gründe vorher notieren: Gültigkeit der Newtonschen Näherung kurz vor der Verschmelzung, Zeit- und Frequenzauflösung der Q-Transformation, Spins und höhere Ordnungen, Rotverschiebung.

```python
eintraege = fetch_event_json(EREIGNIS)["events"]
schluessel = sorted(eintraege)[-1]                     # neueste Version des Katalogeintrags
Mc_quelle, z = eintraege[schluessel]["chirp_mass_source"], eintraege[schluessel]["redshift"]
print(f"Katalog ({schluessel}): Chirp-Masse der Quelle {Mc_quelle:.1f}, Rotverschiebung z = {z:.2f}")
print(f"Katalog im Detektorsystem, (1 + z) · M: {(1 + z) * Mc_quelle:.1f} Sonnenmassen")
print(f"Eigene Messung:                        {M_mittel:.1f} ± {M_streuung:.1f} Sonnenmassen")
```
