# GitHub MCP (Fremd-MCP)

UC2: Ein Agent liest Issues und Repositories über den MCP-Server von GitHub, also über ein Tool, das nicht uns gehört.

##### <font color="blue">*Plan (noch nicht begonnen)*

**Was neu ist gegenüber UC1 und UC4:** Der MCP-Server gehört GitHub, nicht uns. Daraus folgen drei Themen:

* **Fremde Geheimnisse:** GitHub verlangt ein eigenes Token. Es kommt in den Secret Manager, nie ins Notebook. Deployed bekommt nur die Agent Identity dieses Agents `roles/secretmanager.secretAccessor`, und zwar nur auf dieses eine Secret.
* **Rechte auf Tool-Ebene:** Der GitHub-Server bietet viele Tools, auch schreibende (Issues anlegen, Code ändern). Mit `tool_filter` bekommt der Agent nur die Lese-Tools.
* **Prompt Injection durch fremde Inhalte:** Issue-Texte schreiben Fremde. Ein Issue kann Anweisungen enthalten („ignoriere deine Regeln …“). Der Agent muss sie als Daten behandeln, nicht als Auftrag.

**Stufen**

1. Lokal: GitHub-MCP mit Token aus dem Secret Manager, nur Lese-Tools (`tool_filter`).
2. Deployed auf Agent Runtime mit Agent Identity, Secret-Zugriff nur für diesen Agent.
3. Später: Auth Manager (OAuth im Namen des Nutzers statt eines festen Tokens).

**Offen:** genaue URL und Anmeldung des GitHub-MCP-Servers prüfen; welche Lese-Tools wir freigeben.
