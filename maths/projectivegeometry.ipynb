# Projektive Geometrie

Das Notebook baut die projektive Geometrie von der Perspektive bis zum Raum der Quantenzustände auf und rechnet alles mit numpy nach: homogene Koordinaten, Punkte im Unendlichen und Dualität, Homographien und die Hierarchie euklidisch–ähnlich–affin–projektiv, das Doppelverhältnis und Möbius-Abbildungen, die Sätze von Desargues, Pappus, Pascal und Brianchon, Kegelschnitte mit Pol und Polare, hyperbolische Geometrie als Cayley–Klein-Geometrie, die Schätzung von Homographien in der Computer Vision, endliche projektive Ebenen mit Fano-Ebene und Hamming-Code, die projektive Gerade über $\mathbb F_p$ und gegenseitig unverzerrte Basen, und der projektive Hilbertraum: Bloch-Kugel als $\mathbb CP^1$, Fubini–Study-Metrik, geometrische Phase, Hopf-Faserung, Segre-Einbettung und Majorana-Sterne.

Was gewinnt man, wenn man der Ebene Punkte im Unendlichen hinzufügt, und warum ist der Raum der reinen Quantenzustände genau ein solcher projektiver Raum?

#### **Hintergrund**

Ein Maler, der eine gerade Straße zeichnet, lässt ihre parallelen Ränder am Horizont in einem Punkt zusammenlaufen. Die projektive Geometrie nimmt dieses Bild ernst: Sie fügt der Ebene für jede Richtung einen **Punkt im Unendlichen** hinzu, und alle diese Punkte bilden eine **Gerade im Unendlichen**, den Horizont. Danach schneiden sich zwei verschiedene Geraden immer in genau einem Punkt, Ausnahmen gibt es nicht mehr. Die Rechnung wird linear: Punkte und Geraden sind Vektoren in $\mathbb R^3$, Abbildungen sind $3\times3$-Matrizen, und Schnittpunkte sind Kreuzprodukte.

Dieselbe Konstruktion, „Vektoren bis auf einen Faktor“, beschreibt in der Quantenmechanik die reinen Zustände: $\vert\psi\rangle$ und $c\vert\psi\rangle$ sind derselbe physikalische Zustand. Der Zustandsraum ist deshalb der **projektive Hilbertraum** $\mathbb P(\mathcal H)$, für ein Qubit $\mathbb CP^1$, die Bloch-Kugel.

#### **Theorie**

##### *1. Der projektive Raum*

Der **projektive Raum** über einem Körper $K$ ($\mathbb R$, $\mathbb C$ oder ein endlicher Körper $\mathbb F_q$) ist die Menge der Geraden durch den Ursprung von $K^{n+1}$:
$$\mathbb P^n(K) = \big(K^{n+1}\setminus\{0\}\big)/\sim,\qquad x\sim\lambda x\ \text{für }\lambda\ne0 .$$
Ein Punkt wird in **homogenen Koordinaten** $(x_0 : x_1 : \dots : x_n)$ geschrieben; nur die Verhältnisse zählen. In der Ebene ($n = 2$) schreibt man $(x : y : w)$.
- **Affine Karte:** Die Punkte mit $w\ne0$ sind $(x/w : y/w : 1)$, also die gewöhnliche Ebene $\mathbb R^2$.
- **Punkte im Unendlichen:** Die Punkte mit $w = 0$, $(x : y : 0)$, sind Richtungen. Sie bilden die **Gerade im Unendlichen** $w = 0$. Also $\mathbb P^2 = \mathbb R^2\cup\mathbb P^1$, allgemein $\mathbb P^n = K^n\cup\mathbb P^{n-1}$.
- **Modelle:** $\mathbb RP^1$ ist ein Kreis, $\mathbb RP^2$ die Kugel mit identifizierten Gegenpunkten (eine nicht orientierbare Fläche), $\mathbb CP^1$ die Riemann-Zahlenkugel $\mathbb C\cup\{\infty\}$, eine gewöhnliche Kugel $S^2$.

##### *2. Punkte, Geraden, Dualität*

Eine Gerade der projektiven Ebene ist ebenfalls ein Vektor $l = (a, b, c)$ bis auf einen Faktor; der Punkt $x$ liegt auf ihr, wenn $l^\top x = ax + by + cw = 0$. Daraus folgt:
- **Verbindungsgerade** zweier Punkte: $l = p\times q$. **Schnittpunkt** zweier Geraden: $x = l\times m$.
- Drei Punkte liegen genau dann auf einer Geraden, wenn $\det[p\ q\ r] = 0$; drei Geraden gehen genau dann durch einen Punkt, wenn $\det[l\ m\ n] = 0$.
- Zwei parallele Geraden $ax + by + c = 0$, $ax + by + c' = 0$ schneiden sich im Punkt $(b : -a : 0)$ im Unendlichen.
- **Dualität:** Punkte und Geraden gehorchen denselben Formeln. Jeder Satz über Inzidenz bleibt wahr, wenn man „Punkt“ und „Gerade“, „liegt auf“ und „geht durch“ vertauscht.

##### *3. Projektive Abbildungen und die Hierarchie der Geometrien*

Eine **projektive Abbildung** (Homographie, Kollineation) ist $x\mapsto Hx$ mit invertierbarem $H$, wieder bis auf einen Faktor; die Gruppe heißt $\mathrm{PGL}(n+1, K)$, in der Ebene hat sie $9 - 1 = 8$ Parameter. Geraden gehen über in Geraden ($l\mapsto H^{-\top}l$), Kegelschnitte in Kegelschnitte ($C\mapsto H^{-\top}CH^{-1}$). **Hauptsatz:** Vier Punkte in allgemeiner Lage (keine drei auf einer Geraden) lassen sich durch genau eine Homographie auf vier beliebige andere Punkte in allgemeiner Lage abbilden.

Kleins **Erlanger Programm** (1872): Eine Geometrie ist durch ihre Transformationsgruppe bestimmt und studiert deren Invarianten.

| Gruppe | Abbildung der affinen Ebene | Freiheitsgrade | Invarianten |
| --- | --- | --- | --- |
| euklidisch | $x\mapsto Rx + t$ ($R$ Drehung) | 3 | Längen, Winkel, Flächen |
| Ähnlichkeit | $x\mapsto sRx + t$ | 4 | Winkel, Längenverhältnisse |
| affin | $x\mapsto Ax + t$ | 6 | Parallelität, Teilverhältnisse auf einer Geraden, Flächenverhältnisse |
| projektiv | $x\mapsto(Ax + t)/(v^\top x + 1)$ | 8 | Inzidenz, Doppelverhältnis |

In homogenen Koordinaten sind das $3\times3$-Matrizen; bei den ersten drei ist die letzte Zeile $(0, 0, 1)$, bei der projektiven $(v^\top, 1)$.

Eine affine Abbildung lässt die Gerade im Unendlichen fest; eine projektive darf sie bewegen, deshalb geht Parallelität verloren.

##### *4. Das Doppelverhältnis*

Für vier Punkte $A, B, C, D$ einer Geraden mit Koordinaten $a, b, c, d$ ist
$$(A, B; C, D) = \frac{(c - a)(d - b)}{(c - b)(d - a)} .$$
Es ist unter allen Homographien der Geraden invariant und die einzige Invariante von vier Punkten: Zwei Quadrupel lassen sich genau dann ineinander abbilden, wenn ihre Doppelverhältnisse gleich sind. Vertauscht man die Punkte, entstehen höchstens sechs Werte $\lambda, 1/\lambda, 1-\lambda, 1/(1-\lambda), \lambda/(\lambda-1), (\lambda-1)/\lambda$. **Harmonische Lage:** $(A, B; C, D) = -1$. Auf $\mathbb CP^1$ sind die Homographien die **Möbius-Abbildungen** $z\mapsto(az + b)/(cz + d)$; sie bilden Kreise und Geraden (Kreise durch $\infty$) auf Kreise und Geraden ab, und vier Punkte liegen genau dann auf einem Kreis, wenn ihr Doppelverhältnis reell ist.

##### *5. Klassische Sätze*

- **Desargues:** Liegen zwei Dreiecke perspektiv zu einem Punkt (die Verbindungsgeraden entsprechender Ecken gehen durch einen Punkt), dann liegen die drei Schnittpunkte entsprechender Seiten auf einer Geraden.
- **Pappus:** Liegen $A, B, C$ auf einer Geraden und $a, b, c$ auf einer anderen, dann liegen $Ab\cap aB$, $Ac\cap aC$, $Bc\cap bC$ auf einer Geraden.
- **Pascal:** Für sechs Punkte auf einem Kegelschnitt liegen die drei Schnittpunkte gegenüberliegender Seiten des Sechsecks auf einer Geraden (der Pascal-Geraden). Pappus ist der Grenzfall eines Kegelschnitts, der in zwei Geraden zerfällt.
- **Brianchon** (dual zu Pascal): Für ein Sechseck, dessen Seiten einen Kegelschnitt berühren, gehen die drei Hauptdiagonalen durch einen Punkt.

##### *6. Kegelschnitte*

Ein Kegelschnitt ist $x^\top Cx = 0$ mit symmetrischer $3\times3$-Matrix $C$.
- **Tangente** im Punkt $x$ des Kegelschnitts: $l = Cx$. Allgemein heißt $l = Cx$ die **Polare** von $x$; liegt $x$ außerhalb, schneidet die Polare den Kegelschnitt in den Berührpunkten der beiden Tangenten durch $x$.
- **Dualer Kegelschnitt:** Die Tangenten erfüllen $l^\top C^{-1}l = 0$.
- **Klassifikation:** Über $\mathbb R$ sind alle nicht entarteten, nicht leeren Kegelschnitte projektiv gleich (Signatur $(2,1)$). Ellipse, Parabel und Hyperbel unterscheiden sich nur in ihrer Lage zur Geraden im Unendlichen: kein, ein (berührender) oder zwei Schnittpunkte.
- **Kreispunkte:** Jeder Kreis geht durch die beiden imaginären Punkte $I, J = (1 : \pm i : 0)$. Euklidische Geometrie ist projektive Geometrie mit diesem festgehaltenen Punktepaar; Winkel ergeben sich daraus über die Formel von Laguerre, $\theta = \tfrac1{2i}\ln(l, m; i, j)$, mit den Geraden $i, j$ durch den Schnittpunkt von $l, m$ und die Kreispunkte.

##### *7. Cayley–Klein-Geometrien*

Cayley (1859) und Klein (1871): Hält man einen Kegelschnitt fest (das **Absolute**), macht das Doppelverhältnis daraus einen Abstand. Liegen $p, q$ im Inneren eines reellen Kegelschnitts und trifft ihre Verbindungsgerade ihn in $a, b$, dann ist
$$d(p, q) = \tfrac12\left\vert\ln(a, b; p, q)\right\vert .$$
Im Inneren des Einheitskreises ist das die **hyperbolische Geometrie** (Beltrami–Klein-Modell): Geodäten sind Sehnen, der Rand ist unendlich weit weg, und durch einen Punkt außerhalb einer Geraden gehen unendlich viele Geraden, die sie nicht schneiden. Die Isometrien sind die Homographien, die den Kreis festhalten, die Gruppe $\mathrm{SO}(2,1)$ (Lorentz-Gruppe in 2+1 Dimensionen). Mit einem imaginären Absoluten ($x^2 + y^2 + w^2 = 0$) entsteht die elliptische Geometrie, mit dem entarteten Paar der Kreispunkte die euklidische.

##### *8. Computer Vision*

Eine Lochkamera ist eine projektive Abbildung $\mathbb P^3\to\mathbb P^2$, $x = PX$ mit der $3\times4$-Matrix $P = K[R\,\vert\,t]$ ($K$: Brennweite und Bildmitte, $R, t$: Lage der Kamera). Bilder einer Ebene hängen über eine **Homographie** zusammen, $H = K[r_1\ r_2\ t]$. Der **Fluchtpunkt** einer Richtung $d$ ist das Bild ihres Punktes im Unendlichen, $v = KRd$; die Fluchtpunkte aller Richtungen einer Ebene mit Normale $n$ liegen auf dem **Horizont** $l = K^{-\top}Rn$. Aus $N\ge4$ Punktpaaren schätzt man $H$ mit dem **Direct Linear Transform** (DLT): Jedes Paar gibt über $x'\times Hx = 0$ zwei lineare Gleichungen für die 9 Einträge von $H$, gelöst als kleinster Singulärvektor. Vorher normiert man beide Punktmengen (Schwerpunkt in den Ursprung, mittlerer Abstand $\sqrt2$). Ohne Normierung stehen in den Gleichungen Einträge der Größe 1 neben solchen der Größe (Pixel)², das System wird schlecht konditioniert, und das Ergebnis hängt davon ab, wie man die Koordinaten wählt (Hartley 1997).

##### *9. Endliche projektive Ebenen*

Über $\mathbb F_q$ ($q$ Primzahlpotenz) hat $\mathrm{PG}(2, q)$ genau $q^2 + q + 1$ Punkte und ebenso viele Geraden; jede Gerade hat $q + 1$ Punkte, durch jeden Punkt gehen $q + 1$ Geraden, zwei Punkte bestimmen genau eine Gerade, zwei Geraden schneiden sich in genau einem Punkt. Die kleinste ist die **Fano-Ebene** $\mathrm{PG}(2, 2)$ mit 7 Punkten. Allgemein heißt eine Inzidenzstruktur mit diesen Axiomen projektive Ebene der Ordnung $n$. Für jede Primzahlpotenz gibt es eine; Ordnung 6 gibt es nicht (Bruck–Ryser 1949), Ordnung 10 auch nicht (Lam et al. 1989, Computerbeweis), Ordnung 12 ist offen. Die projektive Gerade $\mathrm{PG}(1, d)$ hat $d + 1$ Punkte: die $d+1$ Geraden durch den Ursprung von $\mathbb F_d^2$. Für Primzahlen $d$ gehört zu jeder davon eine Klasse vertauschender Weyl–Heisenberg-Operatoren $X^aZ^b$ und deren Eigenbasis; diese $d + 1$ Basen sind **gegenseitig unverzerrt** (Wootters–Fields 1989), $\vert\langle u\vert v\rangle\vert^2 = 1/d$ für Vektoren aus verschiedenen Basen.

##### *10. Der projektive Hilbertraum*

Physikalisch unterscheidbar sind nur **Strahlen**: $\vert\psi\rangle\sim c\vert\psi\rangle$ für $c\in\mathbb C\setminus\{0\}$, denn alle Wahrscheinlichkeiten $\vert\langle\phi\vert\psi\rangle\vert^2/(\langle\phi\vert\phi\rangle\langle\psi\vert\psi\rangle)$ hängen nur vom Strahl ab. Der Zustandsraum eines $N$-dimensionalen Systems ist $\mathbb P(\mathbb C^N) = \mathbb CP^{N-1}$, eine reelle Mannigfaltigkeit der Dimension $2N - 2$.
- **Qubit:** $(\alpha : \beta)\mapsto z = \beta/\alpha\in\mathbb C\cup\{\infty\}$, und die stereographische Projektion macht daraus die **Bloch-Kugel**. Für $\cos\tfrac\theta2\vert0\rangle + e^{i\varphi}\sin\tfrac\theta2\vert1\rangle$ ist $z = \tan\tfrac\theta2\,e^{i\varphi}$. Eine Unitäre $U$ wirkt als Möbius-Abbildung von $z$ und als Drehung der Kugel: $\mathrm{PU}(2) = \mathrm{SU}(2)/\{\pm1\}\cong\mathrm{SO}(3)$.
- **Satz von Wigner:** Jede Bijektion von $\mathbb P(\mathcal H)$, die alle Übergangswahrscheinlichkeiten erhält, kommt von einem unitären oder antiunitären Operator.
- **Fubini–Study-Metrik:** $d_{FS}(\psi,\phi) = \arccos\dfrac{\vert\langle\psi\vert\phi\rangle\vert}{\Vert\psi\Vert\,\Vert\phi\Vert}$, infinitesimal $ds^2 = \langle d\psi\vert d\psi\rangle - \vert\langle\psi\vert d\psi\rangle\vert^2$ (normiertes $\psi$). Beim Qubit ist $d_{FS}$ der halbe Winkel zwischen den Bloch-Vektoren; $\mathbb CP^1$ ist eine Kugel vom Radius ½, und allgemein ist das Volumen von $\mathbb CP^n$ gleich $\pi^n/n!$. Das zugehörige Maß ist das einzige unitär invariante, das der Haar-zufälligen Zustände. Die Quanten-Fisher-Information reiner Zustände ist das Vierfache der Metrik: $F_Q\,d\theta^2 = 4\,ds^2$.
- **Geometrische Phase:** Die **Bargmann-Invariante** $\langle\psi_1\vert\psi_2\rangle\langle\psi_2\vert\psi_3\rangle\langle\psi_3\vert\psi_1\rangle$ hängt nicht von den Phasen der Vektoren ab. Ihr Argument ist beim Qubit der halbe (orientierte) Raumwinkel $\Omega$ des geodätischen Dreiecks auf der Bloch-Kugel; die geometrische Phase beim Umlauf $1\to2\to3\to1$ ist $\gamma = -\Omega/2$ (Pancharatnam 1956, Berry 1984).
- **Hopf-Faserung:** Normierte Vektoren bilden $S^3\subset\mathbb C^2$; die Abbildung auf die Bloch-Kugel $S^3\to S^2$ hat als Fasern die Kreise $\{e^{i\chi}\psi\}$, die globalen Phasen. Je zwei Fasern sind miteinander verschlungen.
- **Segre-Einbettung:** $\mathbb CP^1\times\mathbb CP^1\to\mathbb CP^3$, $((a_0 : a_1), (b_0 : b_1))\mapsto(a_0b_0 : a_0b_1 : a_1b_0 : a_1b_1)$. Ihr Bild, die Produktzustände zweier Qubits, ist die Quadrik $c_{00}c_{11} - c_{01}c_{10} = 0$. Die **Concurrence** $C = 2\vert c_{00}c_{11} - c_{01}c_{10}\vert$ misst, wie weit ein Zustand von ihr entfernt ist.
- **Veronese-Einbettung und Majorana-Sterne:** Symmetrische Zustände von $n$ Qubits (Spin $j = n/2$) bilden $\mathbb CP^n$, und jeder ist (bis auf einen Faktor) ein symmetrisiertes Produkt von $n$ Qubit-Zuständen. Er ist also durch $n$ Punkte auf der Bloch-Kugel bestimmt, die **Majorana-Sterne** (Majorana 1932): die Nullstellen eines Polynoms vom Grad $n$. Kohärente Zustände haben alle $n$ Sterne an einer Stelle.

#### **Herleitung**

**1. Verbindung und Schnitt über das Kreuzprodukt.** Für $l = p\times q$ ist $l^\top p = (p\times q)\cdot p = \det[p\ q\ p] = 0$ und ebenso $l^\top q = 0$: Beide Punkte liegen auf $l$. Sind $p, q$ verschiedene Punkte (linear unabhängig), ist $l\ne0$, und $l$ ist die einzige Gerade durch beide, denn die Lösungen von $l^\top p = l^\top q = 0$ bilden einen eindimensionalen Raum. Wegen der Symmetrie der Gleichung $l^\top x = 0$ gilt dieselbe Rechnung für den Schnittpunkt $x = l\times m$ zweier Geraden; das ist die Dualität.

**2. Parallelen schneiden sich im Unendlichen.** $(a, b, c)\times(a, b, c') = \big(bc' - cb,\ ca - ac',\ ab - ba\big) = (c' - c)\,(b, -a, 0)$. Die dritte Koordinate ist null: Der Schnittpunkt ist der Punkt $(b : -a : 0)$ im Unendlichen, und er hängt nur von der Richtung $(b, -a)$ der Geraden ab. Alle Punkte im Unendlichen liegen auf $l_\infty = (0, 0, 1)$, denn $l_\infty^\top(x, y, 0) = 0$.

**3. Vier Punkte bestimmen eine Homographie.** Seien $p_1, \dots, p_4$ in allgemeiner Lage. Weil $p_1, p_2, p_3$ linear unabhängig sind, gibt es $\lambda_i$ mit $p_4 = \lambda_1p_1 + \lambda_2p_2 + \lambda_3p_3$, und keines ist null (sonst lägen drei Punkte auf einer Geraden). Die Matrix $M = [\lambda_1p_1\ \lambda_2p_2\ \lambda_3p_3]$ bildet $e_1, e_2, e_3$ auf $p_1, p_2, p_3$ (bis auf Faktoren) und $e_1 + e_2 + e_3$ auf $p_4$ ab. Mit derselben Konstruktion $M'$ für die Zielpunkte ist $H = M'M^{-1}$ die gesuchte Abbildung. Eindeutig: Eine Homographie, die $e_1, e_2, e_3$ festhält, ist diagonal, und hält sie auch $(1, 1, 1)$ fest, sind alle Diagonaleinträge gleich, sie ist also die Identität in $\mathrm{PGL}$.

**4. Das Doppelverhältnis ist invariant.** Schreibe die Punkte der Geraden homogen, $A = (a_1 : a_2)$ usw., und $[AC] = \det\begin{pmatrix}a_1 & c_1\\ a_2 & c_2\end{pmatrix}$. Mit $a_2 = c_2 = 1$ ist $[AC] = a - c$, also $(A, B; C, D) = \frac{[AC][BD]}{[BC][AD]}$. Eine Homographie $H$ der Geraden ($2\times2$-Matrix) gibt $[HA\,HC] = \det H\cdot[AC]$; im Bruch kürzt sich $(\det H)^2$. Ersetzt man einen Vertreter durch $\lambda A$, multipliziert sich $[AC]$ und $[AD]$ mit $\lambda$, also Zähler und Nenner gleich. Das Doppelverhältnis hängt also weder von den Vertretern noch von $H$ ab. Teilverhältnisse wie $(c-a)/(c-b)$ enthalten dagegen den Punkt im Unendlichen als versteckten vierten Punkt; sie sind nur affin invariant.

**5. Desargues aus dem Raum.** Liegen die beiden Dreiecke in verschiedenen Ebenen $E, E'$ des Raums und perspektiv zu einem Punkt $O$, dann liegen $A, B, A', B'$ in einer Ebene (der durch $O, A, B$), also schneiden sich die Geraden $AB$ und $A'B'$. Ihr Schnittpunkt liegt in $E$ und in $E'$, also auf der Schnittgeraden $E\cap E'$; ebenso die beiden anderen Schnittpunkte. Der ebene Fall folgt durch Projektion dieses räumlichen Bildes.

**6. Tangente und Polare.** Liegt $x$ auf dem Kegelschnitt ($x^\top Cx = 0$), dann liegt $x$ auf $l = Cx$, denn $l^\top x = x^\top Cx = 0$. Für einen weiteren Punkt $y$ auf $l$ ($x^\top Cy = 0$) liegen die Punkte $x + ty$ der Geraden genau dann auf dem Kegelschnitt, wenn $(x + ty)^\top C(x + ty) = 2t\,x^\top Cy + t^2\,y^\top Cy = t^2\,y^\top Cy = 0$. Die Lösung $t = 0$ ist doppelt: $l$ berührt den Kegelschnitt in $x$. Für einen Punkt $p$ außerhalb: Ist $x$ Berührpunkt einer Tangente durch $p$, dann $p^\top Cx = 0$, also $x^\top(Cp) = 0$: Die Berührpunkte liegen auf der Polaren $Cp$.

**7. Abbildung von Kegelschnitten und Kreispunkte.** Aus $x' = Hx$ folgt $x^\top Cx = x'^\top H^{-\top}CH^{-1}x'$, also $C' = H^{-\top}CH^{-1}$. Ein Kreis $x^2 + y^2 + Dxw + Eyw + Fw^2 = 0$ trifft die Gerade $w = 0$ dort, wo $x^2 + y^2 = 0$, also in $(1 : \pm i : 0)$, unabhängig von $D, E, F$. Umgekehrt ist ein Kegelschnitt durch beide Punkte ein Kreis: Die Koeffizienten von $x^2$ und $y^2$ müssen gleich sein, und der von $xy$ muss verschwinden.

**8. Der Abstand im Klein-Modell.** Liegen $a, p, q, r, b$ in dieser Reihenfolge auf einer Sehne, dann ist $(a, b; p, q)\cdot(a, b; q, r) = (a, b; p, r)$ (im Produkt kürzen sich die Faktoren mit $q$), also $d(p, q) + d(q, r) = d(p, r)$: Abstände addieren sich längs Geodäten. Homographien, die den Kreis festhalten, erhalten Sehnen, Randpunkte und das Doppelverhältnis, also den Abstand. Für $p = 0$, $q = \rho$ auf der x-Achse ist $a = -1$, $b = 1$ und $d = \tfrac12\ln\frac{(\rho + 1)\cdot1}{1\cdot(1 - \rho)} = \operatorname{artanh}\rho$, unendlich für $\rho\to1$.

**9. Zählen in PG(2, q).** $\mathbb F_q^3$ hat $q^3 - 1$ Vektoren ungleich null, jeder Punkt hat $q - 1$ Vertreter: $(q^3 - 1)/(q - 1) = q^2 + q + 1$ Punkte. Eine Gerade ist ein zweidimensionaler Unterraum mit $q^2 - 1$ Vektoren ungleich null, also $q + 1$ Punkten. Zwei verschiedene Punkte spannen genau einen zweidimensionalen Unterraum auf. Für die Inzidenzmatrix $N$ (Zeilen Punkte, Spalten Geraden) heißt das $NN^\top = qI + J$ ($J$: Einsmatrix): Auf der Diagonale steht die Zahl $q + 1$ der Geraden durch einen Punkt, außerhalb die eine gemeinsame Gerade.

**10. Bloch-Kugel als stereographische Projektion.** Für $\psi = (\cos\tfrac\theta2, e^{i\varphi}\sin\tfrac\theta2)$ ist $z = \beta/\alpha = \tan\tfrac\theta2\,e^{i\varphi}$. Die stereographische Projektion vom Südpol $(0, 0, -1)$ auf die Äquatorebene bildet $(\sin\theta\cos\varphi, \sin\theta\sin\varphi, \cos\theta)$ auf $\frac{\sin\theta}{1 + \cos\theta}e^{i\varphi} = \tan\tfrac\theta2\,e^{i\varphi}$ ab, also genau auf $z$. Eine Unitäre $U = \begin{pmatrix}a & b\\ c & d\end{pmatrix}$ macht aus $(\alpha, \beta)$ den Vektor $(a\alpha + b\beta, c\alpha + d\beta)$, also $z\mapsto(c + dz)/(a + bz)$: eine Möbius-Abbildung. Weil $U$ Übergangswahrscheinlichkeiten $\tfrac12(1 + \vec r\cdot\vec s)$ erhält, erhält sie die Winkel zwischen Bloch-Vektoren; sie ist eine Drehung.

**11. Fubini–Study beim Qubit.** $\vert\langle\psi\vert\phi\rangle\vert^2 = \mathrm{Tr}(\rho\sigma) = \tfrac12(1 + \vec r\cdot\vec s) = \cos^2\tfrac\gamma2$ mit dem Winkel $\gamma$ zwischen den Bloch-Vektoren, also $d_{FS} = \gamma/2$. Die Metrik ist $ds^2 = \tfrac14(d\theta^2 + \sin^2\theta\,d\varphi^2)$, die Metrik einer Kugel vom Radius ½ mit Fläche $\pi$. Für $\psi_\theta = e^{-i\theta G}\psi$ ist $d\psi = -iG\psi\,d\theta$, also $ds^2 = (\langle G^2\rangle - \langle G\rangle^2)\,d\theta^2$ und $F_Q = 4\,\mathrm{Var}(G) = 4\,ds^2/d\theta^2$. **Haar-Zustände:** $\psi = g/\Vert g\Vert$ mit komplex-gaußschem $g$ ist unitär invariant verteilt. $\vert g_0\vert^2$ ist exponentialverteilt, die Summe der übrigen $N-1$ Beiträge gammaverteilt; der Quotient $x = \vert\langle0\vert\psi\rangle\vert^2$ folgt der Beta-Verteilung $\mathrm{Beta}(1, N-1)$ mit Dichte $(N-1)(1-x)^{N-2}$. Für $N = 2$ ist das die Gleichverteilung: Die Bloch-Vektoren Haar-zufälliger Qubits sind gleichmäßig auf der Kugel verteilt.

**12. Geometrische Phase.** Die Bargmann-Invariante ändert sich nicht, wenn man $\psi_k$ durch $e^{i\chi_k}\psi_k$ ersetzt, denn jeder Vektor kommt einmal als Ket und einmal als Bra vor. Für einen fein unterteilten geschlossenen Weg ist $\langle\psi_k\vert\psi_{k+1}\rangle\approx1 + \langle\psi\vert d\psi\rangle = 1 - iA$ mit dem Berry-Zusammenhang $A = i\langle\psi\vert d\psi\rangle$, also $\arg\prod_k\langle\psi_k\vert\psi_{k+1}\rangle = -\oint A = -\gamma$. Für $\psi = (\cos\tfrac\theta2, e^{i\varphi}\sin\tfrac\theta2)$ ist $\langle\psi\vert\partial_\varphi\psi\rangle = i\sin^2\tfrac\theta2$, also $A = -\sin^2\tfrac\theta2\,d\varphi$, und mit dem Satz von Stokes $\gamma = \iint dA = -\tfrac12\iint\sin\theta\,d\theta\,d\varphi = -\Omega/2$.

**13. Segre-Quadrik und Concurrence.** $\psi = \sum_{ij}c_{ij}\vert ij\rangle$ ist genau dann ein Produkt $a\otimes b$, wenn die Matrix $c = ab^\top$ Rang 1 hat, also $\det c = c_{00}c_{11} - c_{01}c_{10} = 0$. Die Singulärwerte $\sigma_0\ge\sigma_1$ von $c$ (Schmidt-Koeffizienten) erfüllen $\sigma_0^2 + \sigma_1^2 = 1$ und $\sigma_0\sigma_1 = \vert\det c\vert = C/2$. Der nächste Produktzustand hat den Überlapp $\sigma_0$ (Eckart–Young), also $d_{FS}(\psi, \text{Segre}) = \arccos\sigma_0$ mit $\sigma_0^2 = \tfrac12\big(1 + \sqrt{1 - C^2}\big)$.

**14. Majorana-Sterne.** Ein symmetrischer Zustand mit Koeffizienten $a_k$ vor den Dicke-Zuständen $\vert D_k\rangle$ ($k$ Einsen) gehört zum homogenen Polynom $f(x, y) = \sum_k a_k\sqrt{\tbinom nk}\,x^{n-k}y^k$. Für den Produktzustand $(\alpha\vert0\rangle + \beta\vert1\rangle)^{\otimes n}$ ist $a_k = \sqrt{\tbinom nk}\alpha^{n-k}\beta^k$ und $f = (\alpha x + \beta y)^n$; die Symmetrisierung von $n$ verschiedenen Faktoren gibt das Produkt $\prod_j(\alpha_jx + \beta_jy)$. Jeder Faktor verschwindet bei $y/x = r_j = -\alpha_j/\beta_j$; der zugehörige Zustand hat $z_j = \beta_j/\alpha_j = -1/r_j$. Die Sterne sind also die Punkte $z_j = -1/r_j$, auf die Kugel gebracht wie in Herleitung 10. Beispiele: GHZ $= (\vert0\cdots0\rangle + \vert1\cdots1\rangle)/\sqrt2$ gibt $f\propto x^n + y^n$, also $r^n = -1$, $\vert z\vert = 1$: $n$ Sterne gleichmäßig auf dem Äquator. Der Dicke-Zustand $\vert D_k\rangle$ gibt $f\propto x^{n-k}y^k$: $k$ Sterne am Südpol ($\vert1\rangle$) und $n - k$ am Nordpol.

#### **Code**

Werkzeuge: homogene Koordinaten, Verbindung und Schnitt, Homographien aus vier Punkten und aus vielen Punktpaaren (DLT), Doppelverhältnis, und für Quantenzustände Bloch-Vektor, Riemann-Koordinate, Haar-Zustände und Fubini–Study-Abstand.

```python
# numpy, scipy und matplotlib sind in Colab vorinstalliert; es wird nichts nachinstalliert
import itertools
import math
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

rng = np.random.default_rng(2026)


def verbindung(p, q):
    """Gerade durch zwei Punkte (homogen): l = p × q."""
    return np.cross(p, q)


def schnitt(l, m):
    """Schnittpunkt zweier Geraden: x = l × m."""
    return np.cross(l, m)


def kollinear(p, q, r):
    """|det| der normierten Vektoren; null genau dann, wenn die drei Punkte (oder Geraden) inzident sind."""
    return abs(np.linalg.det(np.array([v / np.linalg.norm(v) for v in (p, q, r)])))


def h(x, y):
    """Affiner Punkt -> homogene Koordinaten."""
    return np.array([x, y, 1.0])


def anwenden(H, pts):
    """Homographie H auf affine Punkte der Form (N, 2) anwenden."""
    P = np.column_stack([pts, np.ones(len(pts))]) @ H.T
    return P[:, :2] / P[:, 2:3]


def homographie_aus_4(src, dst):
    """Die Homographie, die vier Punkte src auf dst abbildet (Herleitung 3, über den Standardrahmen)."""
    def rahmen(pts):
        P = np.column_stack([pts, np.ones(4)]).T                 # Spalten p1, ..., p4
        lam = np.linalg.solve(P[:, :3], P[:, 3])                 # p4 = λ1 p1 + λ2 p2 + λ3 p3
        return P[:, :3] * lam
    H = rahmen(np.asarray(dst, float)) @ np.linalg.inv(rahmen(np.asarray(src, float)))
    return H / H[2, 2]


def dlt(src, dst, normieren=True, kondition=False):
    """Homographie aus N ≥ 4 Punktpaaren (Direct Linear Transform), wahlweise mit Hartley-Normierung.
    Mit kondition=True zusätzlich die Konditionszahl σ_max/σ_(n−1) des Gleichungssystems."""
    def T(pts):
        c = pts.mean(axis=0)
        s = np.sqrt(2) / np.mean(np.linalg.norm(pts - c, axis=1))
        return np.array([[s, 0, -s * c[0]], [0, s, -s * c[1]], [0, 0, 1]])
    T1, T2 = (T(src), T(dst)) if normieren else (np.eye(3), np.eye(3))
    a, b_ = anwenden(T1, src), anwenden(T2, dst)
    A = []
    for (x, y), (u, v) in zip(a, b_):                            # zwei Zeilen von x' × Hx = 0 pro Paar
        A.append([-x, -y, -1, 0, 0, 0, u * x, u * y, u])
        A.append([0, 0, 0, -x, -y, -1, v * x, v * y, v])
    _, sv, Vt = np.linalg.svd(np.array(A))
    H = Vt[-1].reshape(3, 3)                                     # kleinster Singulärvektor
    H = np.linalg.inv(T2) @ H @ T1
    return (H / H[2, 2], sv[0] / sv[-2]) if kondition else H / H[2, 2]


def doppelverhaeltnis(a, b, c, d):
    """(A, B; C, D) für Koordinaten auf einer Geraden (reell oder komplex)."""
    return ((c - a) * (d - b)) / ((c - b) * (d - a))


def kegelschnitt_zeichnen(ax, C, grenzen=(-3, 3, -3, 3), **kw):
    """Die Nullstellenmenge von (x, y, 1) C (x, y, 1)ᵀ in der affinen Karte als Höhenlinie zeichnen."""
    x = np.linspace(grenzen[0], grenzen[1], 600)
    y = np.linspace(grenzen[2], grenzen[3], 600)
    Xg, Yg = np.meshgrid(x, y)
    P = np.stack([Xg, Yg, np.ones_like(Xg)])
    Q = np.einsum("i...,ij,j...->...", P, C, P)
    ax.contour(Xg, Yg, Q, levels=[0], **kw)


def gerade_zeichnen(ax, l, xlim, **kw):
    """Gerade l = (a, b, c) in der affinen Karte zeichnen."""
    a, b_, c = l
    if abs(b_) > 1e-12:
        x = np.array(xlim)
        ax.plot(x, -(a * x + c) / b_, **kw)
    else:
        ax.axvline(-c / a, **kw)


# Quantenzustände als Strahlen
def bloch(psi):
    """Bloch-Vektor eines (nicht unbedingt normierten) Qubit-Vektors."""
    a, b_ = psi / np.linalg.norm(psi)
    return np.array([2 * (np.conj(a) * b_).real, 2 * (np.conj(a) * b_).imag, abs(a) ** 2 - abs(b_) ** 2])


def ket(r):
    """Ein Qubit-Vektor zum Bloch-Vektor r (Phase so gewählt, dass die erste Komponente reell ist)."""
    th, ph = np.arccos(np.clip(r[2], -1, 1)), np.arctan2(r[1], r[0])
    return np.array([np.cos(th / 2), np.exp(1j * ph) * np.sin(th / 2)])


def riemann_z(psi):
    """Projektive Koordinate z = β/α ∈ ℂ ∪ {∞} des Strahls (α : β)."""
    return np.inf if abs(psi[0]) < 1e-14 else psi[1] / psi[0]


def kugel_aus_z(z):
    """Inverse stereographische Projektion vom Südpol: z -> Punkt der Bloch-Kugel."""
    if np.isinf(z):
        return np.array([0, 0, -1.0])
    return np.array([2 * z.real, 2 * z.imag, 1 - abs(z) ** 2]) / (1 + abs(z) ** 2)


def haar_zustand(N):
    g = rng.normal(size=N) + 1j * rng.normal(size=N)
    return g / np.linalg.norm(g)


def fs_abstand(psi, phi):
    """Fubini-Study-Abstand arccos |<ψ|φ>| / (|ψ| |φ|)."""
    ov = abs(np.vdot(psi, phi)) / (np.linalg.norm(psi) * np.linalg.norm(phi))
    return np.arccos(np.clip(ov, 0, 1))


def kugel_gitter(ax, alpha=0.15):
    u, v = np.mgrid[0:2 * np.pi:40j, 0:np.pi:20j]
    ax.plot_wireframe(np.cos(u) * np.sin(v), np.sin(u) * np.sin(v), np.cos(v), color="0.6", lw=0.4, alpha=alpha)
    ax.set_box_aspect((1, 1, 1))
```

##### *1. Perspektive: Fluchtpunkte und Horizont*

Eine Kamera steht 1.6 m über einer Ebene mit einem Gitter aus Linien und blickt 8° nach unten.
```python
f_px, cx, cy = 800.0, 640.0, 360.0                               # Brennweite und Bildmitte [Pixel]
K = np.array([[f_px, 0, cx], [0, f_px, cy], [0, 0, 1]])
neigung = np.radians(8)                                          # Blick 8° nach unten (y zeigt nach unten)
R = np.array([[1, 0, 0], [0, np.cos(neigung), -np.sin(neigung)], [0, np.sin(neigung), np.cos(neigung)]])
P = K @ np.column_stack([R, np.zeros(3)])                        # Kamera im Ursprung: P = K [R | 0]
hoehe = 1.6


def bild(X):
    """Weltpunkte (N, 3) oder ein Punkt im Unendlichen (Richtung, w = 0) ins Bild projizieren."""
    X = np.atleast_2d(X)
    x = np.column_stack([X, np.ones(len(X))]) @ P.T
    return x[:, :2] / x[:, 2:3]


def fluchtpunkt(d):
    v = P @ np.append(d, 0.0)                                    # Bild des Punktes im Unendlichen (d, 0)
    return v[:2] / v[2]


horizont = np.linalg.inv(K).T @ R @ np.array([0, 1.0, 0])        # l = K^(−T) R n, n = Normale der Bodenebene
richtungen = {"vorwärts (0, 0, 1)": np.array([0, 0, 1.0]), "diagonal (1, 0, 1)": np.array([1, 0, 1.0]),
              "diagonal (−1, 0, 1)": np.array([-1, 0, 1.0])}

fig, ax = plt.subplots(figsize=(11, 6.8))
for x0 in np.arange(-4, 4.5, 1.0):                               # Linien in Blickrichtung
    seg = bild(np.array([[x0, hoehe, 2.0], [x0, hoehe, 400.0]]))
    ax.plot(seg[:, 0], seg[:, 1], color="tab:blue", lw=1)
for z0 in [2, 3, 4, 6, 8, 12, 20, 40]:                           # Querlinien
    seg = bild(np.array([[-4, hoehe, z0], [4, hoehe, z0]]))
    ax.plot(seg[:, 0], seg[:, 1], color="tab:blue", lw=0.6, alpha=0.6)
for x0 in np.arange(-12, 5, 2.0):                                # Diagonalen in Richtung (1, 0, 1) und (−1, 0, 1)
    seg = bild(np.array([[x0, hoehe, 2.0], [x0 + 300, hoehe, 302.0]]))
    ax.plot(seg[:, 0], seg[:, 1], color="tab:orange", lw=0.7, alpha=0.7)
    seg = bild(np.array([[-x0, hoehe, 2.0], [-x0 - 300, hoehe, 302.0]]))
    ax.plot(seg[:, 0], seg[:, 1], color="tab:green", lw=0.7, alpha=0.7)
xs = np.array([-600, 1900])
ax.plot(xs, -(horizont[0] * xs + horizont[2]) / horizont[1], "k--", lw=1, label="Horizont l = K^(−T) R n")
for (name, d), m, farbe in zip(richtungen.items(), ["o", "s", "^"], ["tab:blue", "tab:orange", "tab:green"]):
    v = fluchtpunkt(d)
    ax.plot(*v, m, ms=10, color=farbe, mec="k", label=f"Fluchtpunkt {name}: ({v[0]:.0f}, {v[1]:.0f})")
    print(f"Fluchtpunkt {name:22s}: ({v[0]:8.1f}, {v[1]:6.1f}) px, liegt auf dem Horizont: "
          f"{abs(horizont @ np.append(v, 1)) / np.linalg.norm(horizont[:2]) < 1e-9}")
ax.add_patch(plt.Rectangle((0, 0), 1280, 720, fill=False, ec="0.3", lw=1.5))
ax.set_xlim(-600, 1900); ax.set_ylim(950, 150)                   # Bildkoordinaten: v nach unten
ax.set_aspect("equal")
ax.set_xlabel("Bildkoordinate u [Pixel]"); ax.set_ylabel("Bildkoordinate v [Pixel]")
ax.set_title("Bodengitter durch eine Lochkamera (Rahmen: Bildsensor 1280 × 720)")
ax.legend(fontsize=8, loc="upper center", bbox_to_anchor=(0.5, -0.14), ncol=2)
plt.show()
```

##### *3. Die Hierarchie der Transformationen*

Was bleibt unter euklidischen, Ähnlichkeits-, affinen und projektiven Abbildungen eines Quadratgitters mit Kreis erhalten: Längen, Winkel, Parallelität, Teilverhältnisse, Doppelverhältnis? Die projektive Abbildung wird aus vier Eckpunkten konstruiert (Herleitung 3). Vorher die Tabelle der Invarianten aus der Theorie zur Hand nehmen und für jede Spalte vorhersagen.

```python
quadrat = np.array([[0, 0], [1, 0], [1, 1], [0, 1.0]])
w = np.radians(30)
Rot = np.array([[np.cos(w), -np.sin(w)], [np.sin(w), np.cos(w)]])
abbildungen = {
    "euklidisch": np.block([[Rot, np.array([[0.4], [0.2]])], [np.zeros((1, 2)), np.ones((1, 1))]]),
    "Ähnlichkeit": np.block([[1.5 * Rot, np.array([[0.4], [0.2]])], [np.zeros((1, 2)), np.ones((1, 1))]]),
    "affin": np.array([[1.2, 0.6, 0.1], [0.1, 0.8, 0.2], [0, 0, 1.0]]),
    "projektiv": homographie_aus_4(quadrat, np.array([[0, 0], [1.6, 0.2], [1.1, 1.0], [0.2, 0.8]])),
}
print("projektive Matrix bildet die vier Ecken richtig ab:",
      np.allclose(anwenden(abbildungen["projektiv"], quadrat), [[0, 0], [1.6, 0.2], [1.1, 1.0], [0.2, 0.8]]))

gitter = [np.column_stack([np.full(50, s), np.linspace(0, 1, 50)]) for s in np.linspace(0, 1, 6)] + \
         [np.column_stack([np.linspace(0, 1, 50), np.full(50, s)]) for s in np.linspace(0, 1, 6)]
kreis = 0.5 + 0.3 * np.column_stack([np.cos(np.linspace(0, 2 * np.pi, 200)), np.sin(np.linspace(0, 2 * np.pi, 200))])
punkte = np.array([[0, 0], [0.25, 0], [0.5, 0], [1, 0.0]])        # vier Punkte auf der unteren Kante


def messen(H):
    A, B, D = anwenden(H, np.array([[0, 0], [1, 0], [0, 1.0]]))
    winkel = np.degrees(np.arccos((B - A) @ (D - A) / np.linalg.norm(B - A) / np.linalg.norm(D - A)))
    links, rechts = anwenden(H, np.array([[0, 0], [0, 1.0]])), anwenden(H, np.array([[1, 0], [1, 1.0]]))
    r1, r2 = links[1] - links[0], rechts[1] - rechts[0]
    parallel = abs(r1[0] * r2[1] - r1[1] * r2[0]) / np.linalg.norm(r1) / np.linalg.norm(r2) < 1e-9
    q = anwenden(H, punkte)
    s = np.linalg.norm(q - q[0], axis=1)                         # Koordinate längs der Bildgeraden
    teil = s[1] / s[2]                                           # Teilverhältnis AB/AC (im Original 0.5)
    return np.linalg.norm(B - A), winkel, parallel, teil, doppelverhaeltnis(*s)


print(f"\n{'':12s} {'Länge':>7s} {'Winkel':>8s} {'parallel':>9s} {'Teilverh.':>10s} {'Doppelverh.':>12s}")
print(f"{'Original':12s} {1:7.3f} {90:8.2f} {'ja':>9s} {0.5:10.3f} {doppelverhaeltnis(0, 0.25, 0.5, 1):12.4f}")
fig, axes = plt.subplots(1, 4, figsize=(17, 4.4))
for ax, (name, H) in zip(axes, abbildungen.items()):
    L, W, par, teil, dv = messen(H)
    print(f"{name:12s} {L:7.3f} {W:8.2f} {'ja' if par else 'nein':>9s} {teil:10.3f} {dv:12.4f}")
    for g in gitter:
        ax.plot(*anwenden(np.eye(3), g).T, color="0.8", lw=0.6)
        ax.plot(*anwenden(H, g).T, color="tab:blue", lw=0.8)
    ax.plot(*anwenden(H, kreis).T, color="tab:red")
    ax.plot(*anwenden(H, punkte).T, "ko", ms=4)
    ax.set_aspect("equal"); ax.set_title(name)
    ax.set_xlabel("x"); ax.set_ylabel("y")
plt.suptitle("Quadratgitter mit Kreis (grau: Original); schwarz: vier Punkte für das Doppelverhältnis", y=1.02)
plt.tight_layout()
plt.show()
```

##### *5. Desargues, Pappus, Pascal, Brianchon*

Gelten die vier Sätze für zufällige Konfigurationen, gemessen an der Determinante $\vert\det[p\ q\ r]\vert$ der drei Punkte (bzw. Geraden), die auf einer Geraden liegen (bzw. durch einen Punkt gehen) sollen? Vorher überlegen: Welche Größe in der Größenordnung erwartet man bei Rechengenauigkeit?

```python
def desargues(r):
    O = h(*r.normal(size=2))
    A, B, C = (h(*r.normal(size=2)) for _ in range(3))
    A2, B2, C2 = (O + r.uniform(0.5, 2.5) * (X - O) for X in (A, B, C))   # perspektiv zu O
    S = [schnitt(verbindung(A, B), verbindung(A2, B2)), schnitt(verbindung(B, C), verbindung(B2, C2)),
         schnitt(verbindung(C, A), verbindung(C2, A2))]
    return kollinear(*S), dict(O=O, Dreieck1=(A, B, C), Dreieck2=(A2, B2, C2), S=S)


def pappus(r):
    g1, g2 = [h(*r.normal(size=2)), h(*r.normal(size=2))], [h(*r.normal(size=2)), h(*r.normal(size=2))]
    A, B, C = (g1[0] + t * (g1[1] - g1[0]) for t in r.uniform(-1, 2, 3))
    a, b_, c = (g2[0] + t * (g2[1] - g2[0]) for t in r.uniform(-1, 2, 3))
    S = [schnitt(verbindung(A, b_), verbindung(a, B)), schnitt(verbindung(A, c), verbindung(a, C)),
         schnitt(verbindung(B, c), verbindung(b_, C))]
    return kollinear(*S), dict(Punkte=(A, B, C, a, b_, c), S=S)


ELLIPSE = np.diag([1 / 2.0 ** 2, 1 / 1.2 ** 2, -1.0])            # x²/4 + y²/1.44 = 1


def auf_ellipse(phi):
    return np.array([2.0 * np.cos(phi), 1.2 * np.sin(phi), 1.0])


def pascal(r):
    P = [auf_ellipse(t) for t in np.sort(r.uniform(0, 2 * np.pi, 6))]
    S = [schnitt(verbindung(P[i], P[i + 1]), verbindung(P[i + 3], P[(i + 4) % 6])) for i in range(3)]
    return kollinear(*S), dict(P=P, S=S)


def brianchon(r):
    P = [auf_ellipse(t) for t in np.sort(r.uniform(0, 2 * np.pi, 6))]
    T = [ELLIPSE @ p for p in P]                                 # Tangenten (Herleitung 6)
    V = [schnitt(T[i], T[(i + 1) % 6]) for i in range(6)]        # Ecken des umbeschriebenen Sechsecks
    D = [verbindung(V[i], V[i + 3]) for i in range(3)]           # Hauptdiagonalen
    return kollinear(*D), dict(P=P, V=V, D=D)


saetze = {"Desargues": desargues, "Pappus": pappus, "Pascal": pascal, "Brianchon": brianchon}
for name, satz in saetze.items():
    werte = [satz(rng)[0] for _ in range(2000)]
    print(f"{name:10s}: |det| über 2000 zufällige Konfigurationen höchstens {max(werte):.1e}")

aff = lambda X: X[:2] / X[2]


def kompakt(satz, schluessel, grenze):
    """Erste Konfiguration (Startwert 0, 1, ...), deren Schnittpunkte im Bildausschnitt liegen."""
    for start in range(500):
        _, k = satz(np.random.default_rng(start))
        if all(np.all(np.abs(aff(X)) < grenze) for X in k[schluessel]):
            return k


fig, axes = plt.subplots(2, 2, figsize=(13, 11))
# Desargues
k = kompakt(desargues, "S", 4)
ax = axes[0, 0]
for D, farbe in [(k["Dreieck1"], "tab:blue"), (k["Dreieck2"], "tab:orange")]:
    ax.add_patch(Polygon([aff(X) for X in D], fill=False, ec=farbe, lw=1.5))
for X, X2 in zip(k["Dreieck1"], k["Dreieck2"]):
    ax.plot(*np.array([aff(k["O"]), aff(X2)]).T, color="0.6", lw=0.8)
ax.plot(*aff(k["O"]), "ko"); ax.annotate("O", aff(k["O"]), xytext=(4, 4), textcoords="offset points")
S = np.array([aff(s) for s in k["S"]])
ax.plot(*S.T, "rs-", label="Schnittpunkte entsprechender Seiten")
ax.set_title("Desargues"); ax.legend(fontsize=8)
# Pappus
k = kompakt(pappus, "S", 3)
ax = axes[0, 1]
A, B, C, a, b_, c = [aff(X) for X in k["Punkte"]]
ax.plot(*np.array([A, B, C]).T, "o-", color="tab:blue"); ax.plot(*np.array([a, b_, c]).T, "o-", color="tab:orange")
for X, Y in [(A, b_), (a, B), (A, c), (a, C), (B, c), (b_, C)]:
    ax.plot(*np.array([X, Y]).T, color="0.6", lw=0.8)
S = np.array([aff(s) for s in k["S"]])
ax.plot(*S.T, "rs-", label="Pappus-Gerade"); ax.set_title("Pappus"); ax.legend(fontsize=8)
# Pascal
k = kompakt(pascal, "S", 5)
ax = axes[1, 0]
kegelschnitt_zeichnen(ax, ELLIPSE, (-6, 6, -6, 6), colors="k")
Pp = np.array([aff(p) for p in k["P"]])
for i in range(6):
    p, q = k["P"][i], k["P"][(i + 1) % 6]
    ax.plot(*np.array([aff(p), aff(q)]).T, color="tab:blue")
S = np.array([aff(s) for s in k["S"]])
for i in range(3):                                               # Seiten bis zu den Schnittpunkten verlängern
    for p in (k["P"][i], k["P"][i + 1], k["P"][i + 3], k["P"][(i + 4) % 6]):
        ax.plot(*np.array([aff(p), S[i]]).T, color="0.75", lw=0.6, ls=":")
ax.plot(*Pp.T, "ko"); ax.plot(*S.T, "rs-", label="Pascal-Gerade")
ax.set_xlim(-6, 6); ax.set_ylim(-6, 6); ax.set_aspect("equal"); ax.set_title("Pascal"); ax.legend(fontsize=8)
# Brianchon
_, k = brianchon(np.random.default_rng(4))
ax = axes[1, 1]
kegelschnitt_zeichnen(ax, ELLIPSE, (-6, 6, -6, 6), colors="k")
V = np.array([aff(v) for v in k["V"]])
ax.add_patch(Polygon(V, fill=False, ec="tab:blue"))
for i in range(3):
    ax.plot(*np.array([V[i], V[i + 3]]).T, color="tab:orange")
M = schnitt(k["D"][0], k["D"][1])
ax.plot(*aff(M), "rs", ms=8, label="gemeinsamer Punkt der Diagonalen")
ax.plot(*np.array([aff(p) for p in k["P"]]).T, "ko", ms=4)
ax.set_xlim(-6, 6); ax.set_ylim(-6, 6); ax.set_aspect("equal"); ax.set_title("Brianchon"); ax.legend(fontsize=8)
for ax in axes.flat:
    ax.set_xlabel("x"); ax.set_ylabel("y")
plt.tight_layout()
plt.show()
```

##### *7. Cayley–Klein: hyperbolische Geometrie im Kreis*

Im Inneren des Einheitskreises wird der Abstand über das Doppelverhältnis mit den Randpunkten gemessen. Wie sehen Kreise mit gleichem hyperbolischen Radius aus, wie weit ist der Rand entfernt, und gilt das Parallelenaxiom? Vorher überlegen: Welche euklidischen Abstände haben Punkte, die hyperbolisch gleich weit auseinanderliegen?

```python
def klein_abstand(p, q):
    """Hyperbolischer Abstand im Klein-Modell: ½ ln (a, b; p, q) mit den Randpunkten a, b der Sehne durch p, q."""
    d = q - p
    A, B, C = d @ d, 2 * p @ d, p @ p - 1                      # |p + t d|² = 1
    t1, t2 = np.sort(np.roots([A, B, C]).real)                   # t1 < 0 < 1 < t2: Randpunkte a, b
    return 0.5 * np.log(((1 - t1) * t2) / ((0 - t1) * (t2 - 1)))


def boost(u):
    """Homographie, die den Einheitskreis festhält: Lorentz-Boost in x-Richtung (Form x² + y² − w²)."""
    return np.array([[np.cosh(u), 0, np.sinh(u)], [0, 1, 0], [np.sinh(u), 0, np.cosh(u)]])


p, q, r_ = np.array([-0.3, 0.1]), np.array([0.2, 0.25]), np.array([0.7, 0.4])
r_ = p + 1.6 * (q - p)                                           # dritter Punkt auf derselben Sehne
print(f"Additivität: d(p,q) + d(q,r) = {klein_abstand(p, q) + klein_abstand(q, r_):.6f}, d(p,r) = {klein_abstand(p, r_):.6f}")
print(f"d(0, ρ) gegen artanh ρ: {klein_abstand(np.zeros(2), np.array([0.9, 0])):.6f} gegen {np.arctanh(0.9):.6f}")
B = boost(0.8)
for _ in range(3):
    a, c = (rng.uniform(-0.6, 0.6, 2) for _ in range(2))
    a2, c2 = anwenden(B, np.array([a, c]))
    print(f"Boost: hyperbolisch {klein_abstand(a, c):.5f} -> {klein_abstand(a2, c2):.5f}, "
          f"euklidisch {np.linalg.norm(a - c):.4f} -> {np.linalg.norm(a2 - c2):.4f}")


def hyperbolischer_kreis(m, rad, n=200):
    """Punkte mit hyperbolischem Abstand rad von m (Bisektion längs Strahlen)."""
    pts = []
    for phi in np.linspace(0, 2 * np.pi, n):
        e = np.array([np.cos(phi), np.sin(phi)])
        lo, hi = 0.0, (-(m @ e) + np.sqrt((m @ e) ** 2 - m @ m + 1)) * (1 - 1e-12)
        for _ in range(60):
            mid = (lo + hi) / 2
            lo, hi = (mid, hi) if klein_abstand(m, m + mid * e) < rad else (lo, mid)
        pts.append(m + lo * e)
    return np.array(pts)


fig, axes = plt.subplots(1, 2, figsize=(13, 6.2))
ax = axes[0]
ax.add_patch(plt.Circle((0, 0), 1, fill=False, color="k"))
for x0, farbe in zip([0, 0.5, 0.8, 0.95], ["tab:blue", "tab:orange", "tab:green", "tab:red"]):
    kreis = hyperbolischer_kreis(np.array([x0, 0.0]), 0.5)
    ax.plot(*kreis.T, color=farbe, label=f"Mittelpunkt x = {x0}")
    ax.plot(x0, 0, ".", color=farbe)
schritte = np.tanh(0.5 * np.arange(0, 7))                        # hyperbolisch gleich weit: Abstand 0.5
ax.plot(schritte, -0.6 * np.ones_like(schritte), "k|", ms=12)
ax.plot(-schritte, -0.6 * np.ones_like(schritte), "k|", ms=12)
ax.text(-0.95, -0.72, "Marken im hyperbolischen Abstand 0.5 (auf y = −0.6 verschoben)", fontsize=8)
ax.set_aspect("equal"); ax.set_xlim(-1.05, 1.05); ax.set_ylim(-1.05, 1.05)
ax.set_title("Kreise mit hyperbolischem Radius 0.5"); ax.legend(fontsize=8, loc="upper left")
ax = axes[1]
ax.add_patch(plt.Circle((0, 0), 1, fill=False, color="k"))
g = [np.array([-0.6, -0.8]), np.array([0.6, -0.8])]              # eine Gerade (Sehne)
ax.plot(*np.array(g).T, color="tab:blue", lw=2, label="Gerade g")
P_ = np.array([0.0, 0.3])
ax.plot(*P_, "ko"); ax.annotate("P", P_, xytext=(5, 5), textcoords="offset points")
for i, phi in enumerate(np.linspace(-0.55, 0.55, 9)):          # Sehnen durch P, die g im Inneren nicht treffen
    e = np.array([np.cos(phi), np.sin(phi)])
    t = np.roots([1, 2 * P_ @ e, P_ @ P_ - 1]).real
    ax.plot(*np.array([P_ + t[0] * e, P_ + t[1] * e]).T, color="tab:orange", lw=0.8,
            label="Geraden durch P, die g nicht schneiden" if i == 0 else None)
ax.set_aspect("equal"); ax.set_xlim(-1.05, 1.05); ax.set_ylim(-1.05, 1.05)
ax.set_title("Unendlich viele Parallelen durch P"); ax.legend(fontsize=8, loc="upper left")
for ax in axes:
    ax.set_xlabel("x"); ax.set_ylabel("y")
plt.tight_layout()
plt.show()
```

##### *9. Endliche projektive Ebenen: Fano-Ebene und PG(2, q)*

##### *10. Die projektive Gerade über $\mathbb F_p$ und gegenseitig unverzerrte Basen*

Für $p = 5$ hat $\mathrm{PG}(1, 5)$ die sechs Punkte $(0 : 1)$ und $(1 : k)$, $k = 0, \dots, 4$, die Geraden durch den Ursprung von $\mathbb F_5^2$.
```python
p = 5
omega = np.exp(2j * np.pi / p)
Xop = np.roll(np.eye(p), 1, axis=0)                              # X|j> = |j+1>
Zop = np.diag(omega ** np.arange(p))                             # Z|j> = ω^j |j>
punkte_pg1 = [(0, 1)] + [(1, k) for k in range(p)]
geraden = {pt: sorted({((t * pt[0]) % p, (t * pt[1]) % p) for t in range(p)}) for pt in punkte_pg1}
print("Punkte von PG(1, 5) = Geraden durch 0 in F_5²:")
for pt, g in geraden.items():
    print(f"  ({pt[0]} : {pt[1]}): {g}")
alle = {v for g in geraden.values() for v in g}
print(f"jeder der {p * p - 1} Vektoren ≠ 0 liegt auf genau einer Geraden: {len(alle) == p * p and sum(len(g) - 1 for g in geraden.values()) == p * p - 1}")

basen = []
for a, b_ in punkte_pg1:
    O = np.linalg.matrix_power(Xop, a) @ np.linalg.matrix_power(Zop, b_)
    _, V = np.linalg.eig(O)
    basen.append(V / np.linalg.norm(V, axis=0))
V = np.hstack(basen)
U2 = np.abs(V.conj().T @ V) ** 2
block = np.kron(np.eye(p + 1), np.ones((p, p))).astype(bool)
print(f"innerhalb einer Basis: größte Abweichung von δ {np.abs(U2[block] - np.eye(p * (p + 1))[block]).max():.1e}; "
      f"zwischen Basen: größte Abweichung von 1/p {np.abs(U2[~block] - 1 / p).max():.1e}")

fig, ax = plt.subplots(figsize=(7, 6))
im = ax.imshow(U2, cmap="viridis")
plt.colorbar(im, ax=ax, label="|⟨u|v⟩|²")
ticks = np.arange(p // 2, p * (p + 1), p)
ax.set_xticks(ticks, [f"({a}:{b_})" for a, b_ in punkte_pg1]); ax.set_yticks(ticks, [f"({a}:{b_})" for a, b_ in punkte_pg1])
ax.set_xlabel("Basis zum Punkt von PG(1, 5)"); ax.set_ylabel("Basis zum Punkt von PG(1, 5)")
ax.set_title("Überlappe der 6 Eigenbasen von X^a Z^b (je 5 Vektoren)")
plt.show()
```

##### *13. Geometrische Phase und Hopf-Faserung*

Ist das Argument der Bargmann-Invariante $\langle\psi_1\vert\psi_2\rangle\langle\psi_2\vert\psi_3\rangle\langle\psi_3\vert\psi_1\rangle$ der halbe Raumwinkel des Dreiecks auf der Bloch-Kugel, unabhängig von den Phasen der Vektoren? Und wie sehen die Phasen-Kreise (Fasern) über der Bloch-Kugel aus, wenn man $S^3$ stereographisch in den $\mathbb R^3$ projiziert? Vorher vorhersagen: Wie liegen zwei Fasern zueinander?

```python
def raumwinkel(r1, r2, r3):
    """Orientierter Raumwinkel des geodätischen Dreiecks (Formel von Van Oosterom und Strackee)."""
    return 2 * np.arctan2(r1 @ np.cross(r2, r3), 1 + r1 @ r2 + r2 @ r3 + r3 @ r1)


phase, halber = [], []
for _ in range(1000):
    r = rng.normal(size=(3, 3)); r /= np.linalg.norm(r, axis=1, keepdims=True)
    k = [ket(x) * np.exp(1j * rng.uniform(0, 2 * np.pi)) for x in r]    # beliebige Phasen
    B = np.vdot(k[0], k[1]) * np.vdot(k[1], k[2]) * np.vdot(k[2], k[0])
    phase.append(np.angle(B)); halber.append(raumwinkel(*r) / 2)
phase, halber = np.array(phase), np.array(halber)
print(f"arg(Bargmann) − Ω/2: größte Abweichung {np.max(np.abs(np.angle(np.exp(1j * (phase - halber))))):.1e} "
      "(geometrische Phase des Umlaufs 1→2→3→1: γ = −Ω/2)")

fig = plt.figure(figsize=(16, 5.6))
ax = fig.add_subplot(1, 3, 1)
ax.plot(halber, phase, ".", ms=3)
ax.plot([-np.pi, np.pi], [-np.pi, np.pi], "k--", lw=1)
ax.set_xlabel("Ω/2: halber orientierter Raumwinkel [rad]"); ax.set_ylabel("arg⟨1|2⟩⟨2|3⟩⟨3|1⟩ [rad]")
ax.set_title("Bargmann-Invariante zufälliger Dreiecke\n(Phasen der Vektoren zufällig)")

breiten = {np.radians(45): "Blues", np.radians(90): "Reds"}
ax_b = fig.add_subplot(1, 3, 2, projection="3d")
ax_h = fig.add_subplot(1, 3, 3, projection="3d")
kugel_gitter(ax_b)
chi = np.linspace(0, 2 * np.pi, 200)
for th, cmap in breiten.items():
    farben = plt.get_cmap(cmap)(np.linspace(0.45, 0.95, 12))
    for ph, farbe in zip(np.linspace(0, 2 * np.pi, 12, endpoint=False), farben):
        v = ket(np.array([np.sin(th) * np.cos(ph), np.sin(th) * np.sin(ph), np.cos(th)]))
        ax_b.scatter(*bloch(v), color=farbe, s=25)
        faser = np.exp(1j * chi)[:, None] * v[None, :]           # alle Vektoren e^(iχ) ψ: ein Kreis in S³
        x4 = np.column_stack([faser[:, 0].real, faser[:, 0].imag, faser[:, 1].real, faser[:, 1].imag])
        xyz = x4[:, :3] / (1 - x4[:, 3:4])                        # stereographische Projektion S³ -> ℝ³
        ax_h.plot(*xyz.T, color=farbe, lw=0.9)
# Fasern über den Polen: über |0⟩ der Einheitskreis in der x-y-Ebene, über |1⟩ die z-Achse (durch den Projektionspunkt)
ax_b.scatter([0, 0], [0, 0], [1, -1], color="k", s=40)
ax_h.plot(np.cos(chi), np.sin(chi), 0 * chi, color="k", lw=2, label="Faser über |0⟩ (Nordpol)")
ax_h.plot([0, 0], [0, 0], [-3, 3], color="k", lw=2, ls="--", label="Faser über |1⟩ (Südpol)")
ax_h.legend(fontsize=8, loc="upper left")
ax_b.set_title("Basispunkte auf zwei Breitenkreisen und die Pole", fontsize=10)
ax_h.set_title("Ihre Fasern (globale Phasen) in S³, stereographisch in den ℝ³\nprojiziert: jeder Breitenkreis gibt einen Torus", fontsize=10)
for a_ in (ax_b, ax_h):
    a_.set_xlabel("x"); a_.set_ylabel("y"); a_.set_zlabel("z")
lim = 3.0
ax_h.set_xlim(-lim, lim); ax_h.set_ylim(-lim, lim); ax_h.set_zlim(-lim, lim); ax_h.set_box_aspect((1, 1, 1))
ax_h.view_init(elev=28, azim=40)
plt.tight_layout()
plt.show()

# Verschlingung: zwei Fasern haben die Verschlingungszahl ±1 (Gauß-Integral, numerisch)
def faser(r, n=400):
    v = ket(r)
    f_ = np.exp(1j * np.linspace(0, 2 * np.pi, n, endpoint=False))[:, None] * v[None, :]
    x4 = np.column_stack([f_[:, 0].real, f_[:, 0].imag, f_[:, 1].real, f_[:, 1].imag])
    return x4[:, :3] / (1 - x4[:, 3:4])


def verschlingung(c1, c2):
    d1, d2 = np.roll(c1, -1, axis=0) - c1, np.roll(c2, -1, axis=0) - c2
    m1, m2 = c1 + d1 / 2, c2 + d2 / 2
    diff = m1[:, None, :] - m2[None, :, :]
    return np.sum(np.einsum("ijk,ijk->ij", diff, np.cross(d1[:, None, :], d2[None, :, :]))
                  / np.linalg.norm(diff, axis=2) ** 3) / (4 * np.pi)


r1, r2 = np.array([1.0, 0, 0]), np.array([0, 0.6, 0.8])
print(f"Gauß-Verschlingungszahl zweier Fasern: {verschlingung(faser(r1), faser(r2)):+.3f}")
```

##### *14. Segre und Veronese: Verschränkung und Majorana-Sterne*

Wie weit ist ein zufälliger Zwei-Qubit-Zustand (im Fubini–Study-Abstand) von der Segre-Quadrik der Produktzustände entfernt, und hängt das nur von der Concurrence ab? Wie sehen die Majorana-Sterne eines kohärenten, eines GHZ-, eines Dicke- und eines zufälligen symmetrischen Zustands aus? Vorher die Sterne von GHZ mit fünf Qubits vorhersagen.

```python
conc, abstand = [], []
for _ in range(2000):
    c = haar_zustand(4).reshape(2, 2)
    conc.append(2 * abs(np.linalg.det(c)))
    abstand.append(np.arccos(np.clip(np.linalg.svd(c, compute_uv=False)[0], 0, 1)))   # Überlapp mit dem nächsten Produkt
a_, b2 = haar_zustand(2), haar_zustand(2)
print(f"Produktzustand: c00 c11 − c01 c10 = {abs(np.linalg.det(np.outer(a_, b2))):.1e}")
Cs = np.linspace(0, 1, 200)
print(f"größte Abweichung von arccos √((1 + √(1−C²))/2): "
      f"{np.max(np.abs(np.array(abstand) - np.arccos(np.sqrt((1 + np.sqrt(1 - np.array(conc) ** 2)) / 2)))):.1e}")


def majorana_sterne(a):
    """Majorana-Sterne eines symmetrischen Zustands mit Koeffizienten a_k vor |D_k> (k Einsen), Herleitung 14."""
    n = len(a) - 1
    koef = np.array([a[k] * np.sqrt(math.comb(n, k)) for k in range(n + 1)], dtype=complex)
    koef[np.abs(koef) < 1e-12] = 0
    hoch = koef[::-1]                                            # Polynom in r = y/x, höchster Grad zuerst
    fuehrend = int(np.argmax(np.abs(hoch) > 0))                  # fehlende Grade: Nullstellen bei r = ∞, also z = 0
    wurzeln = np.roots(hoch[fuehrend:])
    zs = [0j] * fuehrend + [np.inf if abs(r) < 1e-12 else -1 / r for r in wurzeln]
    return np.array([kugel_aus_z(z) for z in zs])


u = ket(np.array([np.sin(1.0) * np.cos(0.5), np.sin(1.0) * np.sin(0.5), np.cos(1.0)]))
beispiele = {
    "kohärent, n = 4": np.array([np.sqrt(math.comb(4, k)) * u[0] ** (4 - k) * u[1] ** k for k in range(5)]),
    "GHZ, n = 5": np.array([1, 0, 0, 0, 0, 1]) / np.sqrt(2),
    "Dicke |D_2⟩, n = 4": np.array([0, 0, 1, 0, 0.0]),
    "zufällig symmetrisch, n = 6": haar_zustand(7),
}
fig = plt.figure(figsize=(18, 5))
ax = fig.add_subplot(1, 5, 1)
ax.plot(conc, abstand, ".", ms=2, label="2000 Haar-zufällige Zustände")
ax.plot(Cs, np.arccos(np.sqrt((1 + np.sqrt(1 - Cs ** 2)) / 2)), "k-", lw=1, label="arccos √((1+√(1−C²))/2)")
ax.set_xlabel("Concurrence C"); ax.set_ylabel("d_FS zur Segre-Quadrik [rad]")
ax.set_title("Abstand zu den Produktzuständen", fontsize=10); ax.legend(fontsize=7)
for i, (name, a) in enumerate(beispiele.items()):
    s = majorana_sterne(a)
    ax3 = fig.add_subplot(1, 5, i + 2, projection="3d")
    kugel_gitter(ax3, alpha=0.1)
    ax3.scatter(*s.T, s=60, color="tab:red", depthshade=False)
    ax3.set_title(name, fontsize=10)
    ax3.set_xticks([]); ax3.set_yticks([]); ax3.set_zticks([])
    print(f"{name:28s}: Sterne (x, y, z) = {np.round(s, 2).tolist()}")
print("Bloch-Vektor des kohärenten Zustands:", np.round(bloch(u), 2).tolist())
plt.tight_layout()
plt.show()
```

#### **Wiederholung**

Erst selbst antworten, dann aufklappen.

<details><summary>1. Was ist ein Punkt von ℙⁿ(K), und was ist ein Punkt im Unendlichen?</summary>

Eine Gerade durch den Ursprung von $K^{n+1}$, geschrieben als $(x_0 : \dots : x_n)$ bis auf einen Faktor. In der Ebene sind die Punkte mit $w = 0$ die Punkte im Unendlichen, eine Richtung $(x : y : 0)$; zusammen bilden sie die Gerade $w = 0$.
</details>

<details><summary>2. Wie berechnet man die Verbindungsgerade zweier Punkte und den Schnittpunkt zweier Geraden?</summary>

Beides mit dem Kreuzprodukt: $l = p\times q$ und $x = l\times m$. Parallele Geraden geben einen Schnittpunkt mit $w = 0$.
</details>

<details><summary>3. Wie viele Punkte bestimmen eine Homographie der Ebene, und warum?</summary>

Vier in allgemeiner Lage: $\mathrm{PGL}(3)$ hat 8 Parameter, jeder Punkt gibt 2 Bedingungen. Konstruktion über den Standardrahmen $e_1, e_2, e_3, (1,1,1)$.
</details>

<details><summary>4. Welche Größen erhalten affine, welche projektive Abbildungen?</summary>

Affin: Parallelität, Teilverhältnisse auf einer Geraden, Flächenverhältnisse. Projektiv nur noch Inzidenz und das Doppelverhältnis.
</details>

<details><summary>5. Warum ist das Doppelverhältnis invariant?</summary>

Es ist ein Quotient von $2\times2$-Determinanten, in dem jeder Punkt einmal oben und einmal unten steht. Eine Homographie multipliziert jede Determinante mit $\det H$, das kürzt sich, ebenso die Faktoren der Vertreter.
</details>

<details><summary>6. Was sagen Pascal und Brianchon, und wie hängen sie zusammen?</summary>

Pascal: Für sechs Punkte auf einem Kegelschnitt liegen die Schnittpunkte gegenüberliegender Seiten auf einer Geraden. Brianchon: Für sechs Tangenten gehen die Hauptdiagonalen durch einen Punkt. Sie sind zueinander dual.
</details>

<details><summary>7. Warum sind Ellipse, Parabel und Hyperbel projektiv dasselbe?</summary>

Eine Homographie kann jede Gerade zur Geraden im Unendlichen machen. Der Typ hängt nur davon ab, ob diese Gerade den Kegelschnitt nicht, berührend oder zweimal schneidet.
</details>

<details><summary>8. Wie wird aus projektiver Geometrie die hyperbolische?</summary>

Man hält einen Kegelschnitt fest und misst $d(p, q) = \tfrac12\vert\ln(a, b; p, q)\vert$ mit den Randpunkten $a, b$ der Sehne. Isometrien sind die Homographien, die den Kegelschnitt festhalten ($\mathrm{SO}(2,1)$).
</details>

<details><summary>9. Wie viele Punkte hat PG(2, q), und was ist die Fano-Ebene?</summary>

$q^2 + q + 1$; jede Gerade hat $q + 1$ Punkte. Die Fano-Ebene ist $\mathrm{PG}(2, 2)$ mit 7 Punkten und 7 Geraden; ihre Geraden erzeugen den Hamming-Code [7, 4, 3].
</details>

<details><summary>10. Was hat die projektive Gerade über 𝔽_p mit gegenseitig unverzerrten Basen zu tun?</summary>

Ihre $p + 1$ Punkte sind die Geraden durch den Ursprung von $\mathbb F_p^2$. Jede gibt eine Klasse vertauschender Weyl–Heisenberg-Operatoren $X^aZ^b$, und deren Eigenbasen sind gegenseitig unverzerrt.
</details>

<details><summary>11. Warum ist der Zustandsraum eines Qubits ℂP¹, und wie kommt man zur Bloch-Kugel?</summary>

Vektoren, die sich nur um einen komplexen Faktor unterscheiden, geben dieselben Wahrscheinlichkeiten. $(\alpha : \beta)\mapsto z = \beta/\alpha$, dann inverse stereographische Projektion; Unitäre wirken als Möbius-Abbildungen von $z$, also als Drehungen.
</details>

<details><summary>12. Was ist die Fubini–Study-Metrik, und wie hängt sie mit der Quanten-Fisher-Information zusammen?</summary>

$d_{FS} = \arccos\vert\langle\psi\vert\phi\rangle\vert$, infinitesimal $ds^2 = \langle d\psi\vert d\psi\rangle - \vert\langle\psi\vert d\psi\rangle\vert^2$; beim Qubit der halbe Bloch-Winkel. Für reine Zustände ist $F_Q = 4\,ds^2/d\theta^2$.
</details>

<details><summary>13. Was ist die geometrische Phase eines Dreiecks auf der Bloch-Kugel?</summary>

$\arg\langle\psi_1\vert\psi_2\rangle\langle\psi_2\vert\psi_3\rangle\langle\psi_3\vert\psi_1\rangle = \Omega/2$, unabhängig von den Phasen der Vektoren; die Phase des Umlaufs ist $-\Omega/2$.
</details>

<details><summary>14. Welche Zustände bilden die Segre-Quadrik, welche haben zusammenfallende Majorana-Sterne?</summary>

Die Produktzustände zweier Qubits, $c_{00}c_{11} - c_{01}c_{10} = 0$. Zusammenfallende Sterne haben die kohärenten (Produkt-)Zustände $\vert u\rangle^{\otimes n}$.
</details>

#### **Nächste Schritte**

- **Algebraische Geometrie:** projektive Varietäten, Satz von Bézout (zwei ebene Kurven vom Grad $m$ und $n$ schneiden sich in $mn$ Punkten, wenn man im Projektiven und über $\mathbb C$ zählt), elliptische Kurven als ebene Kubiken mit dem Punkt im Unendlichen als neutralem Element (Notebook [Quadriken und elliptische Kurven](https://github.com/deltorobarba/science/blob/main/maths/quadrics_ellipticcurves.ipynb)).
- **Mehrbildgeometrie:** Epipolargeometrie und Fundamentalmatrix ($x'^\top Fx = 0$), Triangulation, Structure from Motion.
- **Grassmann- und Plücker-Koordinaten:** Geraden im $\mathbb P^3$ als Punkte einer Quadrik im $\mathbb P^5$; die Grassmann-Mannigfaltigkeit $\mathrm{Gr}(k, n)$ als Verallgemeinerung von $\mathbb P^{n-1}$.
- **Endliche Geometrie und Quanteninformation:** gegenseitig unverzerrte Basen für Primzahlpotenzen über $\mathbb F_{p^k}$, symplektische Geometrie über $\mathbb F_p$ und Stabilisatorzustände (Notiz [Stabilisatorzustände](https://github.com/deltorobarba/science/blob/main/quantum/stabilizer.md)), SIC-POVMs.
- **Geometrische Quantenmechanik:** Schrödinger-Dynamik als Hamiltonscher Fluss auf $\mathbb CP^n$ mit der Fubini–Study-Form als symplektischer Form (Kähler-Struktur), Quantengeschwindigkeitsgrenzen (Mandelstam–Tamm) als Geodätenlängen.
- **Berry-Phase und Topologie:** Berry-Krümmung, Chern-Zahlen, die Hopf-Faserung als einfachstes nichttriviales Bündel.
