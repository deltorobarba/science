#### **Theorie**

Hier stehen die allgemeinen Begriffe und der Katalog der wichtigsten Ringe und Körper; die Theorie zu jedem einzelnen Thema steht in seinem Experiment, direkt nach den Fachbegriffen.

##### *1. Ringe*

Ein **Ring** ist eine Menge $R$ mit zwei Verknüpfungen $+$ und $\cdot$, so dass
- $(R, +)$ eine abelsche Gruppe ist (mit neutralem Element 0 und Negativen $-a$),
- die Multiplikation assoziativ ist und ein Einselement 1 hat,
- die **Distributivgesetze** $a(b + c) = ab + ac$ und $(a + b)c = ac + bc$ gelten.

Gilt zusätzlich $ab = ba$, heißt der Ring **kommutativ**. Weitere Begriffe:
- **Einheit:** ein Element mit multiplikativem Inversen; die Einheiten bilden die Gruppe $R^\times$.
- **Nullteiler:** ein Element $a\neq0$, für das es ein $b\neq0$ mit $ab = 0$ gibt. Ein kommutativer Ring ohne Nullteiler heißt **Integritätsbereich**; in ihm darf man kürzen ($ab = ac$, $a\neq0\Rightarrow b = c$).
- **Nilpotent:** $a^k = 0$ für ein $k$; **idempotent:** $a^2 = a$.
- **Körper:** ein kommutativer Ring, in dem jedes Element $\neq0$ eine Einheit ist. **Schiefkörper:** dasselbe ohne Kommutativität (die Quaternionen).
- **Charakteristik:** die kleinste Zahl $n$ mit $1 + 1 + \dots + 1 = 0$ ($n$ Summanden), oder 0, wenn es keine gibt. Die Charakteristik eines Körpers ist 0 oder eine Primzahl.
- **Ringhomomorphismus:** eine Abbildung, die Summe, Produkt und 1 erhält, etwa $\mathbb Z\to\mathbb Z_n$, $a\mapsto a\bmod n$.

##### *2. Ideale und Faktorringe*

- **Ideal:** eine Teilmenge $I$, die unter Addition abgeschlossen ist und Produkte mit beliebigen Ringelementen „schluckt“: $a\in I$, $r\in R\Rightarrow ra\in I$. Beispiel: die Vielfachen von $n$, das **Hauptideal** $(n) = n\mathbb Z$. Ideale spielen für Ringe die Rolle der Normalteiler für Gruppen.
- **Faktorring** $R/I$: Man rechnet „modulo $I$“, also mit Restklassen $a + I$. Beispiele: $\mathbb Z/(n) = \mathbb Z_n$; $\mathbb R[x]/(x^2 + 1)\cong\mathbb C$ (man rechnet mit Polynomen und setzt $x^2 = -1$, also $x = i$); $\mathbb F_2[x]/(x^4 + x + 1) = \mathbb F_{16}$ (Exp. 5).
- **Primideal:** $ab\in P\Rightarrow a\in P$ oder $b\in P$; genau dann ist $R/P$ ein Integritätsbereich. **Maximales Ideal:** kein größeres echtes Ideal; genau dann ist $R/M$ ein Körper. So entsteht jeder endliche Körper: $\mathbb F_p[x]/(f)$ mit irreduziblem $f$.
- **Homomorphiesatz:** Für einen Ringhomomorphismus $\varphi$ ist $R/\ker\varphi\cong\varphi(R)$.
- **Chinesischer Restsatz:** Sind $I$ und $J$ teilerfremd ($I + J = R$), so ist $R/(I\cap J)\cong R/I\times R/J$; für Zahlen $\mathbb Z_{mn}\cong\mathbb Z_m\times\mathbb Z_n$ bei $\mathrm{ggT}(m, n) = 1$ (Exp. 2).

##### *3. Die Hierarchie der kommutativen Ringe*

Jede Zeile ist ein Spezialfall der vorigen:

| Klasse | Definierende Eigenschaft | Beispiel | Gegenbeispiel aus der Zeile darüber |
| --- | --- | --- | --- |

##### *5. Körper und Körpererweiterungen*

- **Körpererweiterung** $L/K$: ein Körper $L$, der $K$ enthält, etwa $\mathbb Q\subset\mathbb Q(\sqrt2)\subset\mathbb R$. $L$ ist ein Vektorraum über $K$; seine Dimension ist der **Grad** $[L:K]$. Grade multiplizieren sich in Türmen: $[M:K] = [M:L]\,[L:K]$.
- **Algebraisch** heißt ein Element $\alpha$ über $K$, wenn es Nullstelle eines Polynoms mit Koeffizienten in $K$ ist; sein **Minimalpolynom** ist das normierte Polynom kleinsten Grades mit dieser Eigenschaft, und $[K(\alpha):K]$ ist dessen Grad. Nicht algebraische Zahlen heißen **transzendent** ($\pi$, $e$).
- **Zerfällungskörper** eines Polynoms: der kleinste Körper, in dem es in Linearfaktoren zerfällt. **Algebraisch abgeschlossen** ist ein Körper, in dem jedes Polynom zerfällt; $\mathbb C$ ist es (Fundamentalsatz der Algebra).
- **Endliche Körper:** Zu jeder Primzahlpotenz $q = p^k$ gibt es bis auf Isomorphie genau einen Körper $\mathbb F_q$, und es gibt keine anderen endlichen Körper. $\mathbb F_q^\times$ ist zyklisch (Exp. 5).
- **Galois-Theorie:** Die **Galois-Gruppe** einer Erweiterung sind die Automorphismen von $L$, die $K$ festlassen. Bei einer Galois-Erweiterung entsprechen ihre Untergruppen genau den Zwischenkörpern (Hauptsatz); eine Gleichung ist durch Wurzeln lösbar genau dann, wenn ihre Galois-Gruppe auflösbar ist (Exp. 9).

##### *6. Kommutative Algebra und Geometrie*

- **Noethersch** heißt ein Ring, in dem jedes Ideal endlich erzeugt ist (gleichwertig: jede aufsteigende Kette von Idealen wird stationär). **Hilberts Basissatz:** Mit $R$ ist auch $R[x]$ noethersch; also ist jedes Polynomgleichungssystem in endlich vielen Variablen zu endlich vielen Gleichungen gleichwertig.
- **Varietät:** die Nullstellenmenge $V(I)$ eines Ideals von Polynomen. **Hilberts Nullstellensatz:** Über $\mathbb C$ hat ein Gleichungssystem genau dann keine Lösung, wenn $1$ im erzeugten Ideal liegt, und die Polynome, die auf $V(I)$ verschwinden, sind genau die, von denen eine Potenz in $I$ liegt. Geometrie (Punktmengen) und Algebra (Ideale) entsprechen sich.
- **Gröbner-Basis:** ein besonderes Erzeugendensystem eines Ideals, mit dem man Zugehörigkeit zum Ideal testet und Gleichungssysteme eliminiert (Exp. 7).
- **Moduln:** „Vektorräume über einem Ring“; über einem Hauptidealring zerfällt jeder endlich erzeugte Modul in zyklische Moduln (Smith-Normalform, Exp. 13).
- **Lokalisierung und Vervollständigung:** Man macht Elemente invertierbar (aus $\mathbb Z$ wird $\mathbb Q$) oder betrachtet einen Ring „nahe bei einer Primzahl“ (aus $\mathbb Z$ wird $\mathbb Z_p$, Exp. 10).
- **Dimension** (Krull): die Länge der längsten Kette von Primidealen; $K[x_1, \dots, x_n]$ hat Dimension $n$, passend zur Geometrie des $n$-dimensionalen Raums.

Drei Werkzeuge, die mehrere Experimente brauchen: `solve_mod_p` löst lineare Gleichungssysteme über dem Körper $\mathbb F_p$ mit Gauß-Elimination (für die Reed-Solomon-Dekodierung), `rank_mod_p` berechnet den Rang einer Matrix über $\mathbb F_p$ (für invertierbare Matrizen), und `polymul_mod` multipliziert im Ring $\mathbb Z_q[x]/(x^n + 1)$ (für die Gitterkryptographie).
```python
# Alles hier (numpy, scipy, sympy, matplotlib) ist in Colab vorinstalliert
import itertools, time, warnings
from math import gcd, isqrt, log
import numpy as np
import matplotlib.pyplot as plt
from sympy import (symbols, Poly, Matrix, ZZ, factor_list, groebner, minimal_polynomial, sqrt, cos, pi, primerange,
                   isprime, factorint, totient, mobius, Rational, real_roots, solve)
from sympy.polys.numberfields.galoisgroups import galois_group
from sympy.matrices.normalforms import smith_normal_form
from sympy.ntheory import sqrt_mod, primitive_root, n_order
from sympy.functions.combinatorial.numbers import legendre_symbol
from sympy.utilities.exceptions import SymPyDeprecationWarning
warnings.filterwarnings("ignore", category=SymPyDeprecationWarning)   # sympy vergleicht intern Restklassen (factor_list modulo p); harmlos
x, y = symbols("x y")

def solve_mod_p(A, b, p):
    """Löst A·z = b über dem Körper 𝔽_p mit Gauß-Elimination; freie Variablen werden 0 gesetzt. Gibt None zurück, wenn das System widersprüchlich ist."""
    M = np.concatenate([np.array(A, dtype=np.int64) % p, np.array(b, dtype=np.int64).reshape(-1, 1) % p], axis=1)
    rows, cols = M.shape[0], M.shape[1] - 1
    pivots, r = [], 0
    for c in range(cols):
        nz = [i for i in range(r, rows) if M[i, c]]
        if not nz:
            continue
        M[[r, nz[0]]] = M[[nz[0], r]]
        M[r] = M[r] * pow(int(M[r, c]), -1, p) % p
        for i in range(rows):
            if i != r and M[i, c]:
                M[i] = (M[i] - M[i, c] * M[r]) % p
        pivots.append(c)
        r += 1
        if r == rows:
            break
    if any(M[i, -1] for i in range(r, rows)):
        return None
    z = np.zeros(cols, dtype=np.int64)
    for i, c in enumerate(pivots):
        z[c] = M[i, -1]
    return z

def rank_mod_p(M, p):
    """Rang einer ganzzahligen Matrix über 𝔽_p (Gauß-Elimination)."""
    M = np.array(M, dtype=np.int64) % p
    r = 0
    for c in range(M.shape[1]):
        nz = [i for i in range(r, M.shape[0]) if M[i, c]]
        if not nz:
            continue
        M[[r, nz[0]]] = M[[nz[0], r]]
        M[r] = M[r] * pow(int(M[r, c]), -1, p) % p
        for i in range(M.shape[0]):
            if i != r and M[i, c]:
                M[i] = (M[i] - M[i, c] * M[r]) % p
        r += 1
    return r

def polymul_mod(a, b, q, n):
    """Produkt im Ring ℤ_q[x]/(xⁿ + 1): gewöhnliches Produkt, dann xⁿ = −1 einsetzen."""
    full = np.convolve(a, b)
    out = full[:n].copy()
    out[:len(full) - n] -= full[n:]
    return out % q

print("Werkzeuge bereit: lineare Gleichungssysteme und Rang über 𝔽_p, Multiplikation in ℤ_q[x]/(xⁿ + 1)")
```

#### **Experimente**

##### *1. Die Restklassenringe ℤₙ: Einheiten, Nullteiler, Körper*

Wann kann man in den Restklassen modulo $n$ dividieren? Vorher vorhersagen: Wie viele Elemente von $\mathbb Z_{12}$ haben ein multiplikatives Inverses, und wie viele von $\mathbb Z_{13}$?

Für $n = 2$ bis 60 zählt der Code in $\mathbb Z_n$ die Einheiten, die Nullteiler, die nilpotenten und die idempotenten Elemente. Links und in der Mitte stehen die Multiplikationstafeln von $\mathbb Z_{12}$ und $\mathbb Z_{13}$ (Farbe = Wert von $a\cdot b\bmod n$, rote Kreuze bei $a\cdot b = 0$ mit $a, b\neq0$), rechts die Anzahlen gegen $n$.

**Hintergrund & Fachbegriffe:**
- **Restklassenring** $\mathbb Z_n$: die Zahlen $0, 1, \dots, n - 1$ mit Addition und Multiplikation modulo $n$.
- **Einheit:** ein Element $a$ mit einem $b$, so dass $ab = 1$.
- **Nullteiler:** ein Element $a\neq0$, für das es ein $b\neq0$ mit $ab = 0$ gibt.
- **Nilpotent:** eine Potenz ist 0; **idempotent:** $a^2 = a$.
- **Eulersche $\varphi$-Funktion:** die Zahl der zu $n$ teilerfremden Zahlen in $1, \dots, n$.
- **Radikal** $\mathrm{rad}(n)$: das Produkt der verschiedenen Primfaktoren von $n$; $\omega(n)$ ist ihre Anzahl.

**Theorie:**

- **Einheit oder Nullteiler:** Ist $\mathrm{ggT}(a, n) = 1$, gibt es nach Bézout (Exp. 2) $s, t$ mit $sa + tn = 1$, also $sa\equiv1$: $a$ ist eine Einheit. Ist $d = \mathrm{ggT}(a, n) > 1$, dann ist $a\cdot(n/d)\equiv0$: $a$ ist ein Nullteiler. In einem endlichen kommutativen Ring ist also jedes Element $\neq0$ genau eines von beiden, und es gibt $\varphi(n)$ Einheiten.
- **Körper:** $\mathbb Z_n$ ist genau dann ein Körper, wenn $n$ eine Primzahl ist; dann ist jedes $a\neq0$ teilerfremd zu $n$.
- **Nilpotente und idempotente Elemente:** $a$ ist nilpotent genau dann, wenn $\mathrm{rad}(n)$ die Zahl $a$ teilt; es gibt $n/\mathrm{rad}(n)$ davon. Idempotente Elemente gibt es $2^{\omega(n)}$, eines zu jeder Auswahl von Primfaktoren (chinesischer Restsatz: in jedem Faktor $\mathbb Z_{p^k}$ ist ein idempotentes Element 0 oder 1).

```python
def ring_stats(n):
    units = sum(1 for a in range(n) if gcd(a, n) == 1)
    zero_div = sum(1 for a in range(1, n) if any(a * b % n == 0 for b in range(1, n)))
    nilpotent = sum(1 for a in range(n) if pow(a, n, n) == 0)
    idempotent = sum(1 for a in range(n) if a * a % n == a)
    return units, zero_div, nilpotent, idempotent

ns = np.arange(2, 61)
stats_n = np.array([ring_stats(int(m)) for m in ns])
fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
for ax, m in zip(axes[:2], (12, 13)):
    T = np.outer(np.arange(m), np.arange(m)) % m
    ax.imshow(T, cmap="viridis")
    zr, zc = np.where((T == 0) & (np.arange(m)[:, None] > 0) & (np.arange(m)[None, :] > 0))
    ax.plot(zc, zr, "rx", ms=7, label="a·b = 0 mit a, b ≠ 0")
    ax.set_title(f"Multiplikationstafel von ℤ_{m}" + (" (Körper)" if isprime(m) else ""), fontsize=10)
    ax.set_xlabel("b"); ax.set_ylabel("a")
    ax.legend(fontsize=8, loc="lower right")
prime_mask = np.array([isprime(int(m)) for m in ns])
axes[2].plot(ns, stats_n[:, 0], "o-", ms=3, label="Einheiten φ(n)")
axes[2].plot(ns, stats_n[:, 1], "s-", ms=3, label="Nullteiler ≠ 0")
axes[2].plot(ns, stats_n[:, 2], "^-", ms=3, label="nilpotente Elemente")
axes[2].plot(ns[prime_mask], stats_n[prime_mask, 0], "k*", ms=8, label="n prim: alles ≠ 0 ist Einheit")
axes[2].set_xlabel("n")
axes[2].set_ylabel("Anzahl")
axes[2].set_title("Einheiten und Nullteiler in ℤₙ", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

rad = lambda m: int(np.prod(list(factorint(m).keys())))
print("n  Einheiten  Nullteiler≠0  nilpotent  idempotent")
for m in (12, 13, 30, 32, 60):
    u, z, nil, idem = ring_stats(m)
    print(f"{m:2d} {u:9d} {z:13d} {nil:10d} {idem:11d}   (φ = {int(totient(m))}, n/rad(n) = {m // rad(m)}, 2^ω(n) = {2 ** len(factorint(m))})")
print(f"Jedes Element ≠ 0 ist Einheit oder Nullteiler: {all(stats_n[:, 0] + stats_n[:, 1] == ns - 1)};  ℤₙ ist Körper genau für n prim: "
      f"{all((stats_n[:, 1] == 0) == prime_mask)}")
```

* Die **beiden Tafeln** links zeigen den Unterschied: In $\mathbb Z_{12}$ gibt es viele rote Kreuze ($2\cdot6 = 3\cdot4 = 0$ usw.), und die Zeilen zu 2, 3, 4, 6, 8, 9, 10 enthalten nicht alle Werte. In $\mathbb Z_{13}$ ist jede Zeile außer der ersten eine Permutation von $0, \dots, 12$: Jedes Element $\neq0$ hat ein Inverses. Laut **Ausgabe** hat $\mathbb Z_{12}$ genau 4 Einheiten (1, 5, 7, 11), $\mathbb Z_{13}$ hat 12; das beantwortet die Vorhersage.
* Im **rechten Bild** springt die Zahl der Einheiten: Bei Primzahlen (Sterne) ist sie $n - 1$, bei Zahlen mit vielen kleinen Faktoren klein. Laut Ausgabe ist jedes Element $\neq0$ Einheit oder Nullteiler, und $\mathbb Z_n$ ist Körper genau für $n$ prim. Die nilpotenten Elemente folgen $n/\mathrm{rad}(n)$ (in $\mathbb Z_{32}$ sind 16 nilpotent: alle geraden), die idempotenten $2^{\omega(n)}$ (8 in $\mathbb Z_{30}$ und $\mathbb Z_{60}$).
* **Wo es vorkommt:** **RSA** verschlüsselt in $\mathbb Z_n$ mit $n = pq$ und braucht dort Einheiten; die Prüfziffern der ISBN-10 rechnen in $\mathbb Z_{11}$, einem Körper, damit jede vertauschte Ziffer erkannt wird. Computer rechnen mit festen Wortbreiten in $\mathbb Z_{2^{32}}$ und $\mathbb Z_{2^{64}}$ (Überlauf ist Rechnen modulo $2^{64}$); dort sind alle geraden Zahlen Nullteiler, deshalb funktionieren multiplikative Hash-Funktionen nur mit ungeraden Faktoren.

##### *2. Der euklidische Algorithmus, Bézout und der chinesische Restsatz*

Wie berechnet man den größten gemeinsamen Teiler und Inverse modulo $n$ schnell, und wie setzt man Reste zusammen? Vorher vorhersagen: Wie viele Divisionsschritte braucht Euklid ungefähr für zwei zwölfstellige Zahlen?

Für Paare zufälliger Zahlen bis $10^2, \dots, 10^{12}$ wird die mittlere Zahl der Divisionsschritte gemessen und mit aufeinanderfolgenden Fibonacci-Zahlen verglichen, dem schlechtesten Fall (links). In der Mitte steht der chinesische Restsatz als Bild: Jede Zahl $0, \dots, 14$ wird an die Stelle ($x\bmod3$, $x\bmod5$) geschrieben. Rechts dasselbe für $\mathbb Z_{24}$ mit den nicht teilerfremden Moduln 4 und 6. Die Ausgabe zeigt Bézout-Koeffizienten, ein Inverses modulo $2^{127} - 1$ und eine Lösung des Restsatzes.

**Hintergrund & Fachbegriffe:**
- **Euklidischer Algorithmus:** $\mathrm{ggT}(a, b) = \mathrm{ggT}(b, a\bmod b)$, wiederholt, bis der Rest 0 ist.
- **Lemma von Bézout:** Es gibt ganze Zahlen $s, t$ mit $\mathrm{ggT}(a, b) = sa + tb$; der **erweiterte** euklidische Algorithmus liefert sie.
- **Modulares Inverses:** $a^{-1}\bmod n$, das $b$ mit $ab\equiv1$.
- **Chinesischer Restsatz:** Simultane Kongruenzen zu teilerfremden Moduln haben genau eine Lösung modulo des Produkts.
- **Fibonacci-Zahlen:** $1, 1, 2, 3, 5, 8, \dots$, jede die Summe der beiden vorigen.

**Theorie:**

- **Laufzeit:** Jeder Schritt ersetzt $(a, b)$ durch $(b, a\bmod b)$, und nach zwei Schritten hat sich die größere Zahl mindestens halbiert. Am langsamsten ist es, wenn jeder Quotient 1 ist: bei benachbarten Fibonacci-Zahlen (Lamé 1844: höchstens fünfmal so viele Schritte wie die kleinere Zahl Dezimalstellen hat). Im Mittel braucht man etwa $\frac{12\ln2}{\pi^2}\ln N\approx0.843\ln N$ Schritte (Heilbronn, Dixon 1969).
- **Bézout gibt Inverse:** Aus $sa + tn = 1$ folgt $sa\equiv1\bmod n$. Python rechnet so mit `pow(a, -1, n)`.
- **Chinesischer Restsatz als Ringisomorphismus:** $\mathbb Z_{mn}\to\mathbb Z_m\times\mathbb Z_n$, $x\mapsto(x\bmod m, x\bmod n)$, ist für teilerfremde $m, n$ bijektiv und verträglich mit $+$ und $\cdot$. Rechnen modulo $mn$ ist also Rechnen in zwei kleineren Ringen zugleich. Ohne Teilerfremdheit trifft die Abbildung nur einen Teil der Paare.

```python
rng = np.random.default_rng(2)

def euclid_steps(a, b):
    k = 0
    while b:
        a, b, k = b, a % b, k + 1
    return k

def extended_euclid(a, b):
    """Gibt (g, s, t) mit g = ggT(a, b) = s·a + t·b zurück."""
    s0, s1, t0, t1 = 1, 0, 0, 1
    while b:
        qq = a // b
        a, b = b, a - qq * b
        s0, s1, t0, t1 = s1, s0 - qq * s1, t1, t0 - qq * t1
    return a, s0, t0

sizes = [10 ** k for k in range(2, 13)]
mean_steps = [np.mean([euclid_steps(int(rng.integers(1, Nn)), int(rng.integers(1, Nn))) for _ in range(4000)]) for Nn in sizes]
fib = [1, 1]
while len(fib) < 60:
    fib.append(fib[-1] + fib[-2])
worst = [(fib[k + 1], euclid_steps(fib[k + 1], fib[k])) for k in range(1, 58)]

fig, axes = plt.subplots(1, 3, figsize=(17, 4.6))
axes[0].semilogx(sizes, mean_steps, "o", label="Mittel über zufällige Paare < N")
lnN = np.log(np.array(sizes, dtype=float))
slope = 12 * np.log(2) / np.pi ** 2
axes[0].semilogx(sizes, slope * lnN + (np.mean(mean_steps) - slope * lnN.mean()), "k--", label="Steigung 12 ln 2 / π² = 0.843 pro ln N")
axes[0].semilogx([w[0] for w in worst], [w[1] for w in worst], "r.", ms=4, label="Fibonacci-Paare (schlechtester Fall)")
axes[0].set_xlim(50, 3e12)
axes[0].set_xlabel("Größe N der Zahlen")
axes[0].set_ylabel("Zahl der Divisionsschritte")
axes[0].set_title("Euklid braucht nur logarithmisch viele Schritte", fontsize=10)
axes[0].legend(fontsize=8)
m1, m2 = 3, 5
grid = np.full((m1, m2), -1)
for v in range(m1 * m2):
    grid[v % m1, v % m2] = v
axes[1].imshow(grid, cmap="viridis")
for i in range(m1):
    for j in range(m2):
        axes[1].text(j, i, grid[i, j], ha="center", va="center", color="w", fontsize=12)
axes[1].set_xticks(range(m2)); axes[1].set_yticks(range(m1))
axes[1].set_xlabel("x mod 5"); axes[1].set_ylabel("x mod 3")
axes[1].set_title("ℤ₁₅ ≅ ℤ₃ × ℤ₅: jedes Paar genau einmal", fontsize=10)
m1, m2 = 4, 6
grid2 = np.full((m1, m2), -1)
for v in range(24):
    grid2[v % m1, v % m2] = v if grid2[v % m1, v % m2] < 0 else grid2[v % m1, v % m2]
axes[2].imshow(np.where(grid2 < 0, np.nan, grid2), cmap="viridis")
for i in range(m1):
    for j in range(m2):
        axes[2].text(j, i, "–" if grid2[i, j] < 0 else grid2[i, j], ha="center", va="center", color="k" if grid2[i, j] < 0 else "w", fontsize=12)
axes[2].set_xticks(range(m2)); axes[2].set_yticks(range(m1))
axes[2].set_xlabel("x mod 6"); axes[2].set_ylabel("x mod 4")
axes[2].set_title("ggT(4, 6) = 2: nur die Hälfte der Paare wird getroffen", fontsize=10)
plt.tight_layout()
plt.show()

a_, b_ = 240, 46
g_, s_, t_ = extended_euclid(a_, b_)
print(f"ggT({a_}, {b_}) = {g_} = {s_}·{a_} + {t_}·{b_}  (Bézout)")
big = 2 ** 127 - 1
inv = pow(65537, -1, big)
print(f"Inverses von 65537 modulo 2¹²⁷ − 1: {inv}, Probe 65537·inv mod m = {65537 * inv % big};  Schritte: {euclid_steps(big, 65537)}")
print(f"Gemessene Steigung der mittleren Schrittzahl: {np.polyfit(lnN, mean_steps, 1)[0]:.3f} pro ln N (Theorie 12 ln 2/π² = {slope:.3f})")
print(f"Schlechtester Fall: F₅₈ = {fib[57]} und F₅₇ brauchen {euclid_steps(fib[57], fib[56])} Schritte (Lamé: höchstens 5 × Zahl der Dezimalstellen = {5 * len(str(fib[56]))})")
from sympy.ntheory.modular import crt
print(f"Chinesischer Restsatz: x ≡ 2 (mod 3), x ≡ 3 (mod 5), x ≡ 2 (mod 7) → x = {crt([3, 5, 7], [2, 3, 2])[0]} (mod 105)")
```

* Im **linken Bild** liegt die mittlere Schrittzahl auf einer Geraden über $\log N$: Sie wächst nur logarithmisch. Laut **Ausgabe** ist die Steigung 0.844 pro $\ln N$ (Theorie 0.843); für zwölfstellige Zahlen sind es im Mittel etwa 23 Schritte, das beantwortet die Vorhersage. Die roten Punkte (Fibonacci-Paare) liegen deutlich höher: $F_{58} = 591\,286\,729\,879$ und $F_{57}$ brauchen 56 Schritte, unter Lamés Schranke 60.
* Im **mittleren Bild** steht jede der 15 Zahlen an einer eigenen Stelle der $3\times5$-Tabelle, und jede Stelle ist besetzt: $\mathbb Z_{15}\cong\mathbb Z_3\times\mathbb Z_5$. Im **rechten Bild** (Moduln 4 und 6, nicht teilerfremd) bleibt die Hälfte der Stellen leer: Ein Paar wie ($x\equiv0\bmod4$, $x\equiv1\bmod6$) hat keine Lösung, weil beide Bedingungen die Parität festlegen. Laut Ausgabe ist $\mathrm{ggT}(240, 46) = 2 = -9\cdot240 + 47\cdot46$, das Inverse von 65537 modulo $2^{127} - 1$ ist in 3 Schritten gefunden, und $x\equiv2, 3, 2$ modulo $3, 5, 7$ hat die Lösung 23.
* **Wo es vorkommt:** **RSA** braucht das Inverse des öffentlichen Exponenten modulo $\varphi(n)$ (oft $e = 65537$) und entschlüsselt mit dem chinesischen Restsatz etwa viermal schneller (getrennt modulo $p$ und $q$). Der Restsatz steckt in der Kalenderrechnung, in der schnellen Arithmetik großer Zahlen (Restzahlsysteme in Hardware), in Shamirs Geheimnisteilung und in der Kristallographie; der euklidische Algorithmus in der Kettenbruchentwicklung (Shors Periodenfindung) und im Kürzen von Brüchen in jedem Computeralgebrasystem.

##### *3. Gaußsche Zahlen, eindeutige Zerlegung und Klassenzahlen*

Ist die Zerlegung in Primfaktoren in jedem Zahlbereich eindeutig? Vorher vorhersagen: Welche Primzahlen lassen sich als Summe zweier Quadrate schreiben?

Links stehen die Gaußschen Primzahlen $a + bi$ mit $\vert a\vert, \vert b\vert\leq40$. Rechts wird für alle Diskriminanten $-D$ mit $D\leq2000$ die **Klassenzahl** $h(-D)$ des imaginärquadratischen Zahlkörpers $\mathbb Q(\sqrt{-D})$ berechnet, durch Zählen reduzierter quadratischer Formen. Die Ausgabe prüft Fermats Zwei-Quadrate-Satz und zeigt die nicht eindeutige Zerlegung von 6 in $\mathbb Z[\sqrt{-5}]$.

**Hintergrund & Fachbegriffe:**
- **Gaußsche Zahlen** $\mathbb Z[i]$: die Zahlen $a + bi$ mit ganzen $a, b$. **Norm** $N(a + bi) = a^2 + b^2$; sie ist multiplikativ, $N(zw) = N(z)N(w)$.
- **Irreduzibel:** nicht als Produkt zweier Nichteinheiten schreibbar. **Prim:** teilt es ein Produkt, so teilt es einen Faktor.
- **Zahlkörper** $\mathbb Q(\sqrt{-D})$: die Zahlen $a + b\sqrt{-D}$ mit rationalen $a, b$; sein **Ganzheitsring** sind die darin enthaltenen „ganzen“ Zahlen.
- **Klassenzahl** $h$: die Zahl der Idealklassen; $h = 1$ heißt, dass jedes Ideal ein Hauptideal ist und die Zerlegung in Primelemente eindeutig.
- **Reduzierte quadratische Form:** $ax^2 + bxy + cy^2$ mit $b^2 - 4ac = -D$ und $\vert b\vert\leq a\leq c$; jede Idealklasse entspricht genau einer davon.

**Theorie:**

- **Gaußsche Primzahlen:** $\mathbb Z[i]$ ist euklidisch (Division mit Rest bezüglich der Norm), also faktoriell. Eine rationale Primzahl $p$ bleibt prim in $\mathbb Z[i]$, wenn $p\equiv3\bmod4$; sie zerfällt als $p = (a + bi)(a - bi) = a^2 + b^2$, wenn $p\equiv1\bmod4$; und $2 = -i(1 + i)^2$ ist verzweigt. Daraus folgt **Fermats Zwei-Quadrate-Satz**: Eine ungerade Primzahl ist genau dann Summe zweier Quadrate, wenn sie $\equiv1\bmod4$ ist.
- **$\mathbb Z[\sqrt{-5}]$:** $6 = 2\cdot3 = (1 + \sqrt{-5})(1 - \sqrt{-5})$. Die Normen sind $4, 9, 6, 6$; ein echter Teiler von 2 oder 3 müsste Norm 2 oder 3 haben, und $a^2 + 5b^2$ nimmt diese Werte nie an. Alle vier Faktoren sind also irreduzibel, aber nicht prim, und die Zerlegung ist nicht eindeutig.
- **Ideale retten die Eindeutigkeit:** In jedem Ganzheitsring eines Zahlkörpers zerfällt jedes Ideal eindeutig in Primideale (Dedekind); in $\mathbb Z[\sqrt{-5}]$ ist $(6) = \mathfrak p_2^2\,\mathfrak p_3\,\mathfrak p_3'$. Die Klassenzahl misst, wie weit Ideale von Zahlen entfernt sind.
- **Klassenzahl 1:** Genau für $D = 3, 4, 7, 8, 11, 19, 43, 67, 163$ (Gauß vermutet; bewiesen von Heegner 1952, Baker und Stark 1967). Deshalb ist $e^{\pi\sqrt{163}}$ fast ganzzahlig. Im Mittel wächst $h(-D)$ wie $\sqrt D$ (Siegel).

```python
def is_gaussian_prime(a, b):
    """a + bi ist Gaußsche Primzahl: Norm prim, oder eine Komponente 0 und die andere prim ≡ 3 mod 4."""
    if a == 0 or b == 0:
        c = abs(a + b)
        return isprime(c) and c % 4 == 3
    return isprime(a * a + b * b)

R = 40
gp = [(a, b) for a in range(-R, R + 1) for b in range(-R, R + 1) if is_gaussian_prime(a, b)]

def class_number(D):
    """Klassenzahl h(−D): Zahl der reduzierten primitiven quadratischen Formen ax² + bxy + cy² mit b² − 4ac = −D."""
    h = 0
    for a in range(1, isqrt(D // 3) + 1):
        for b in range(-a + 1, a + 1):
            if (b * b + D) % (4 * a):
                continue
            c = (b * b + D) // (4 * a)
            if c < a or (c == a and b < 0) or gcd(gcd(a, abs(b)), c) != 1:
                continue
            h += 1
    return h

def fundamental(D):
    if D % 4 == 3:
        return all(D % (p * p) for p in range(2, isqrt(D) + 1))
    if D % 4 == 0:
        m = D // 4
        return m % 4 in (1, 2) and all(m % (p * p) for p in range(2, isqrt(m) + 1))
    return False

Ds = [D for D in range(3, 2001) if fundamental(D)]
hs = np.array([class_number(D) for D in Ds])
fig, axes = plt.subplots(1, 2, figsize=(16, 5.2), gridspec_kw={"width_ratios": [1, 1.5]})
g_arr = np.array(gp)
axes[0].plot(g_arr[:, 0], g_arr[:, 1], "s", ms=2.5)
axes[0].set_aspect("equal")
axes[0].set_xlabel("Realteil a"); axes[0].set_ylabel("Imaginärteil b")
axes[0].set_title(f"Gaußsche Primzahlen a + bi mit |a|, |b| ≤ {R}", fontsize=10)
axes[1].plot(Ds, hs, ".", ms=4, label="h(−D)")
h1 = [D for D, h in zip(Ds, hs) if h == 1]
axes[1].plot(h1, [1] * len(h1), "r*", ms=10, label=f"Klassenzahl 1: D = {h1}")
DD = np.linspace(3, 2000, 200)
axes[1].plot(DD, np.sqrt(DD) / np.pi, "k--", lw=1, label="√D / π (Größenordnung)")
axes[1].set_xlabel("D (Diskriminante −D)")
axes[1].set_ylabel("Klassenzahl")
axes[1].set_title("Wie stark versagt die eindeutige Zerlegung in ℚ(√−D)?", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

primes = list(primerange(3, 1000))
two_squares = {p: next(((a, isqrt(p - a * a)) for a in range(1, isqrt(p) + 1) if isqrt(p - a * a) ** 2 == p - a * a), None) for p in primes}
print(f"Fermat: Primzahl p < 1000 ist Summe zweier Quadrate genau dann, wenn p ≡ 1 mod 4: {all((two_squares[p] is not None) == (p % 4 == 1) for p in primes)};  "
      f"z. B. 13 = {two_squares[13][0]}² + {two_squares[13][1]}², 97 = {two_squares[97][0]}² + {two_squares[97][1]}²")
norms5 = {a * a + 5 * b * b for a in range(-5, 6) for b in range(-3, 4)}
print(f"ℤ[√−5]: 6 = 2·3 = (1 + √−5)(1 − √−5); Normen 4·9 = 6·6; Elemente mit Norm 2 oder 3: {'gefunden' if (2 in norms5 or 3 in norms5) else 'keine'} → 2, 3, 1 ± √−5 sind unzerlegbar, die Zerlegung ist nicht eindeutig")
print(f"Klassenzahlen: h(−4) = {class_number(4)} (ℤ[i]), h(−20) = {class_number(20)} (ℤ[√−5]), h(−163) = {class_number(163)};  Klassenzahl 1 für D ≤ 2000: {h1}")
```

* Im **linken Bild** bilden die Gaußschen Primzahlen ein Muster mit der Symmetrie des Quadrats (Multiplikation mit den Einheiten $\pm1, \pm i$ dreht um 90°). Auf den Achsen liegen nur die Primzahlen $3, 7, 11, 19, \dots\equiv3\bmod4$; die Primzahlen $\equiv1\bmod4$ fehlen dort, weil sie zerfallen. Laut **Ausgabe** ist eine Primzahl $p < 1000$ genau dann Summe zweier Quadrate, wenn $p\equiv1\bmod4$, etwa $13 = 2^2 + 3^2$ und $97 = 4^2 + 9^2$; das beantwortet die Vorhersage.
* Im **rechten Bild** streuen die Klassenzahlen stark und wachsen im Mittel wie $\sqrt D/\pi$. Klassenzahl 1 (rote Sterne) haben laut Ausgabe genau $D = 3, 4, 7, 8, 11, 19, 43, 67, 163$, darüber bis 2000 keine einzige. $h(-4) = 1$ für $\mathbb Z[i]$, $h(-20) = 2$ für $\mathbb Z[\sqrt{-5}]$ (dort scheitert die eindeutige Zerlegung, Elemente mit Norm 2 oder 3 gibt es laut Ausgabe keine), $h(-163) = 1$.
* **Wo es vorkommt:** Kummers Versuch, **Fermats letzten Satz** über die Zerlegung in $\mathbb Z[e^{2\pi i/p}]$ zu beweisen, scheiterte genau an Klassenzahlen größer 1 und führte zur Idealtheorie. Klassengruppen imaginärquadratischer Körper sind ein Baustein von Kryptoverfahren ohne vertrauenswürdige Einrichtung (verifizierbare Verzögerungsfunktionen). Gaußsche Zahlen beschreiben Gitterpunkte auf Kreisen (wie viele Punkte $(a, b)$ liegen auf $a^2 + b^2 = n$?) und das Quadratgitter in der Kristallographie.

##### *4. Polynomringe: irreduzible Polynome über endlichen Körpern*

Wie viele Polynome lassen sich nicht weiter zerlegen, und hängt das vom Körper ab? Vorher vorhersagen: Gibt es eine Primzahl $p$, modulo derer $x^4 + 1$ irreduzibel ist?

Für $\mathbb F_2$ (Grad bis 10), $\mathbb F_3$ (bis 6) und $\mathbb F_5$ (bis 4) werden alle normierten Polynome durchprobiert und die irreduziblen gezählt, im Vergleich mit Gauß' Formel und $p^n/n$ (links). Rechts wird $x^4 + 1$ modulo aller Primzahlen $3\leq p < 400$ faktorisiert und nach den Graden der Faktoren sortiert.

**Hintergrund & Fachbegriffe:**
- **Polynomring** $K[x]$: Polynome mit Koeffizienten in $K$, mit der üblichen Addition und Multiplikation.
- **Normiert:** der höchste Koeffizient ist 1.
- **Irreduzibel:** nicht als Produkt zweier Polynome kleineren Grades schreibbar; die „Primzahlen“ des Polynomrings.
- **Möbius-Funktion** $\mu(d)$: $(-1)^r$, wenn $d$ ein Produkt von $r$ verschiedenen Primzahlen ist, sonst 0.
- **Eisenstein-Kriterium:** Teilt eine Primzahl $p$ alle Koeffizienten außer dem höchsten und $p^2$ nicht den konstanten, so ist das Polynom über $\mathbb Q$ irreduzibel.

**Theorie:**

- **$K[x]$ ist euklidisch:** Division mit Rest nach dem Grad, deshalb gibt es ggT, Bézout und eindeutige Zerlegung in irreduzible Polynome, wie bei $\mathbb Z$.
- **Gauß' Formel:** Die Zahl der normierten irreduziblen Polynome vom Grad $n$ über $\mathbb F_p$ ist $\frac1n\sum_{d\mid n}\mu(d)\,p^{n/d}\approx p^n/n$. Grund: Das Polynom $x^{p^n} - x$ ist das Produkt aller normierten irreduziblen Polynome, deren Grad $n$ teilt; Gradvergleich und Möbius-Umkehr ergeben die Formel. Ein zufälliges Polynom vom Grad $n$ ist also mit Wahrscheinlichkeit etwa $1/n$ irreduzibel, wie eine zufällige Zahl der Größe $N$ mit Wahrscheinlichkeit $1/\ln N$ prim ist.
- **Über $\mathbb Q$ irreduzibel, modulo $p$ zerlegbar:** $x^4 + 1$ ist über $\mathbb Q$ irreduzibel (Eisenstein nach Einsetzen von $x + 1$), zerfällt aber modulo jeder Primzahl. Grund: $\mathbb F_{p^2}$ enthält die achten Einheitswurzeln, weil 8 die Zahl $p^2 - 1$ teilt; deshalb haben alle Faktoren höchstens Grad 2. Galois-theoretisch: Die Galois-Gruppe $\mathbb Z_2\times\mathbb Z_2$ enthält keinen 4-Zyklus (Exp. 9). Irreduzibilität modulo $p$ beweist Irreduzibilität über $\mathbb Q$, die Umkehrung gilt nicht.

```python
def count_irreducible(p, n):
    """Zählt die normierten irreduziblen Polynome vom Grad n über 𝔽_p durch Ausprobieren."""
    cnt = 0
    for coeffs in itertools.product(range(p), repeat=n):
        if Poly([1, *coeffs], x, modulus=p).is_irreducible:
            cnt += 1
    return cnt

def gauss_formula(p, n):
    return sum(int(mobius(d)) * p ** (n // d) for d in range(1, n + 1) if n % d == 0) // n

cases = {2: range(1, 11), 3: range(1, 7), 5: range(1, 5)}
brute = {p: [count_irreducible(p, n) for n in ns_] for p, ns_ in cases.items()}
patterns = {}
for p in primerange(3, 400):
    degs = tuple(sorted(Poly(f, x).degree() for f, _ in factor_list(x ** 4 + 1, modulus=p)[1]))
    patterns.setdefault(degs, []).append(p)

fig, axes = plt.subplots(1, 2, figsize=(16, 4.6))
for (p, ns_), c in zip(cases.items(), ["tab:blue", "tab:orange", "tab:green"]):
    ns_ = list(ns_)
    axes[0].semilogy(ns_, brute[p], "o", c=c, ms=6, label=f"𝔽_{p}: durchgezählt")
    axes[0].semilogy(ns_, [gauss_formula(p, n) for n in ns_], "-", c=c, lw=1, label=f"𝔽_{p}: Formel (1/n) Σ μ(d) p^(n/d)")
    axes[0].semilogy(ns_, [p ** n / n for n in ns_], ":", c=c, lw=1)
axes[0].set_xlabel("Grad n")
axes[0].set_ylabel("Zahl der normierten irreduziblen Polynome")
axes[0].set_title("Irreduzible Polynome: etwa pⁿ/n (gepunktet)", fontsize=10)
axes[0].legend(fontsize=7)
labels = [" + ".join(map(str, k)) for k in patterns]
axes[1].bar(labels, [len(v) for v in patterns.values()])
axes[1].set_xlabel("Grade der Faktoren von x⁴ + 1 modulo p")
axes[1].set_ylabel("Zahl der Primzahlen p < 400")
axes[1].set_title("x⁴ + 1 ist über ℚ irreduzibel, modulo jeder Primzahl aber zerlegbar", fontsize=10)
plt.tight_layout()
plt.show()

for p, ns_ in cases.items():
    print(f"𝔽_{p}: durchgezählt {brute[p]};  Formel stimmt: {brute[p] == [gauss_formula(p, n) for n in ns_]}")
print(f"x⁴ + 1 über ℚ irreduzibel: {Poly(x ** 4 + 1, x).is_irreducible};  Zerlegungstypen modulo p < 400: " + ", ".join(f"{k}: {len(v)}" for k, v in patterns.items())
      + f";  irreduzibel modulo einer ungeraden Primzahl: {(4,) in patterns}")
print(f"Beispiele: x⁴ + 1 ≡ {' · '.join(str(f.as_expr()) for f, _ in factor_list(x ** 4 + 1, modulus=7)[1])} (mod 7), ≡ {' · '.join(str(f.as_expr()) for f, _ in factor_list(x ** 4 + 1, modulus=17)[1])} (mod 17)")
```

* Im **linken Bild** (logarithmische Achse) liegen die durchgezählten Werte genau auf Gauß' Formel: über $\mathbb F_2$ $2, 1, 2, 3, 6, 9, 18, 30, 56, 99$ irreduzible Polynome der Grade 1 bis 10, über $\mathbb F_3$ $3, 3, 8, 18, 48, 116$, über $\mathbb F_5$ $5, 10, 40, 150$; laut **Ausgabe** stimmt die Formel in allen Fällen. Die gepunkteten Kurven $p^n/n$ liegen für großes $n$ fast darauf.
* Im **rechten Bild** zerfällt $x^4 + 1$ modulo jeder Primzahl $3\leq p < 400$: 62-mal in zwei quadratische Faktoren und 15-mal in vier lineare (genau für $p\equiv1\bmod8$), nie bleibt es irreduzibel. Das beantwortet die Vorhersage: keine. Laut Ausgabe ist es über $\mathbb Q$ irreduzibel; modulo 7 ist es $(x^2 + 3x + 1)(x^2 - 3x + 1)$, modulo 17 zerfällt es in vier Linearfaktoren.
* **Wo es vorkommt:** Irreduzible Polynome über $\mathbb F_2$ definieren endliche Körper (AES benutzt $x^8 + x^4 + x^3 + x + 1$), **CRC-Prüfsummen** in Ethernet, ZIP und USB (Division durch ein Generatorpolynom über $\mathbb F_2$) und Pseudozufallsgeneratoren mit linearen Schieberegistern (primitive Polynome geben die maximale Periode $2^n - 1$). Faktorisierung von Polynomen modulo $p$ und Hochheben nach $\mathbb Z$ (Berlekamp, Zassenhaus) ist der Standardweg, wie Computeralgebrasysteme Polynome über $\mathbb Q$ faktorisieren.

##### *5. Endliche Körper: 𝔽₁₆ ist nicht ℤ₁₆*

Wie baut man einen Körper mit 16 Elementen, wenn $\mathbb Z_{16}$ keiner ist? Vorher vorhersagen: Wie viele Elemente von $\mathbb F_{16}$ erzeugen die ganze multiplikative Gruppe?

$\mathbb F_{16}$ wird als $\mathbb F_2[x]/(x^4 + x + 1)$ gebaut: Elemente sind Polynome vom Grad $< 4$ mit Koeffizienten 0 oder 1, gespeichert als 4-Bit-Zahlen; Addition ist bitweises XOR, multipliziert wird als Polynom und dann modulo $x^4 + x + 1$ reduziert. Links und in der Mitte stehen die Multiplikationstafeln von $\mathbb Z_{16}$ und $\mathbb F_{16}$, rechts die Elementordnungen in $\mathbb F_{16}^\times$. Die Ausgabe listet Potenzen von $x$, Nullteiler, den Frobenius und den Teilkörper $\mathbb F_4$.

**Hintergrund & Fachbegriffe:**
- **Endlicher Körper** $\mathbb F_q$ (Galois-Körper $GF(q)$): ein Körper mit $q$ Elementen; $q$ muss eine Primzahlpotenz sein.
- **$\mathbb F_2[x]/(f)$:** Polynome über $\mathbb F_2$, gerechnet modulo $f$; ist $f$ irreduzibel vom Grad $k$, ist das ein Körper mit $2^k$ Elementen.
- **Primitives Element:** ein Erzeuger der zyklischen Gruppe $\mathbb F_q^\times$.
- **Frobenius-Abbildung:** $a\mapsto a^p$ in Charakteristik $p$; sie ist ein Körperautomorphismus, weil $(a + b)^p = a^p + b^p$ gilt.
- **Teilkörper:** ein Körper innerhalb eines größeren; $\mathbb F_{p^m}\subseteq\mathbb F_{p^k}$ genau dann, wenn $m$ die Zahl $k$ teilt.

**Theorie:**

- **Konstruktion:** $(x^4 + x + 1)$ ist ein maximales Ideal in $\mathbb F_2[x]$, weil das Polynom irreduzibel ist, also ist der Faktorring ein Körper (Theorie, Abschnitt 2). $\mathbb Z_{16}$ ist dagegen kein Körper: $4\cdot4 = 0$. $\mathbb F_{16}$ hat Charakteristik 2 ($a + a = 0$), $\mathbb Z_{16}$ Charakteristik 16.
- **Zyklische Einheitengruppe:** In jedem endlichen Körper ist $\mathbb F_q^\times$ zyklisch der Ordnung $q - 1$; es gibt $\varphi(q - 1)$ primitive Elemente, für $q = 16$ also $\varphi(15) = 8$. Hier ist $x$ selbst primitiv (das Polynom heißt deshalb **primitiv**). Mit einer Tabelle der Potenzen („Logarithmen“) wird Multiplikation zur Addition der Exponenten modulo 15.
- **Eindeutigkeit:** Jeder Körper mit 16 Elementen ist isomorph zu diesem; ein anderes irreduzibles Polynom vom Grad 4 gibt nur eine andere Beschriftung (Variation 4).
- **Galois-Gruppe:** Die Automorphismen von $\mathbb F_{p^k}$ sind die Potenzen des Frobenius, eine zyklische Gruppe der Ordnung $k$. Die Teilkörper sind die Fixkörper: $\mathbb F_4 = \lbrace a: a^4 = a\rbrace$ (Hauptsatz der Galois-Theorie im Kleinen).

```python
MOD = 0b10011                                                     # x⁴ + x + 1, irreduzibel über 𝔽₂

def gf16_mul(a, b):
    """Multiplikation in 𝔽₁₆ = 𝔽₂[x]/(x⁴ + x + 1); Elemente als 4-Bit-Zahlen (Bit i = Koeffizient von xⁱ)."""
    r = 0
    while b:
        if b & 1:
            r ^= a
        b >>= 1
        a <<= 1
        if a & 0b10000:
            a ^= MOD
    return r

T16 = np.array([[gf16_mul(a, b) for b in range(16)] for a in range(16)])
Z16 = np.outer(np.arange(16), np.arange(16)) % 16
powers = [1]
for _ in range(15):
    powers.append(gf16_mul(powers[-1], 0b10))                      # Potenzen von x
orders = {}
for a in range(1, 16):
    k, v = 1, a
    while v != 1:
        v, k = gf16_mul(v, a), k + 1
    orders[a] = k
frob = [gf16_mul(a, a) for a in range(16)]
subfield4 = [a for a in range(16) if gf16_mul(gf16_mul(a, a), gf16_mul(a, a)) == a]

def name(a):
    terms = [("x³", 8), ("x²", 4), ("x", 2), ("1", 1)]
    return " + ".join(t for t, bit in terms if a & bit) or "0"

fig, axes = plt.subplots(1, 3, figsize=(17, 4.8))
axes[0].imshow(Z16, cmap="viridis")
zr, zc = np.where((Z16 == 0) & (np.arange(16)[:, None] > 0) & (np.arange(16)[None, :] > 0))
axes[0].plot(zc, zr, "rx", ms=6, label="Nullteiler")
axes[0].set_title("ℤ₁₆: Multiplikation modulo 16", fontsize=10)
axes[0].legend(fontsize=8, loc="lower right")
axes[1].imshow(T16, cmap="viridis")
axes[1].set_title("𝔽₁₆: Multiplikation von Polynomen mod x⁴ + x + 1", fontsize=10)
for ax in axes[:2]:
    ax.set_xlabel("b"); ax.set_ylabel("a")
d15 = [1, 3, 5, 15]
axes[2].bar([str(d) for d in d15], [sum(1 for v in orders.values() if v == d) for d in d15], alpha=0.6, label="gemessen")
axes[2].plot([str(d) for d in d15], [int(totient(d)) for d in d15], "ko", label="φ(d)")
axes[2].set_xlabel("Ordnung in der multiplikativen Gruppe 𝔽₁₆^×")
axes[2].set_ylabel("Zahl der Elemente")
axes[2].set_title("𝔽₁₆^× ist zyklisch mit 15 Elementen", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

print("Potenzen von x in 𝔽₁₆: " + ", ".join(f"x^{k} = {name(v)}" for k, v in enumerate(powers[:8])) + ", …")
print(f"x hat Ordnung {orders[2]} (primitives Element); alle 15 Elemente ≠ 0 kommen als Potenz vor: {sorted(powers[:15]) == list(range(1, 16))}")
print(f"Nullteiler: ℤ₁₆ hat {len(zr)} Paare (a, b ≠ 0) mit a·b = 0, 𝔽₁₆ hat {int(np.sum(T16[1:, 1:] == 0))};  Charakteristik 2: a + a = 0 für alle a")
print(f"Frobenius a ↦ a² ist bijektiv: {sorted(frob) == list(range(16))};  Teilkörper 𝔽₄ = {{a : a⁴ = a}} = {[name(a) for a in subfield4]}")
```

* Die **linke Tafel** ($\mathbb Z_{16}$) hat laut **Ausgabe** 17 Paare $a, b\neq0$ mit $ab = 0$ (rote Kreuze, etwa $4\cdot4$, $2\cdot8$), die Zeilen der geraden Zahlen enthalten nur gerade Werte. In der **mittleren Tafel** ($\mathbb F_{16}$) ist jede Zeile außer der ersten eine Permutation aller 16 Elemente, keine Nullteiler. Die Muster wirken zufälliger, weil Multiplikation von Polynomen mit Reduktion modulo $x^4 + x + 1$ keine Ordnung der Zahlen $0, \dots, 15$ respektiert.
* Laut Ausgabe durchlaufen die Potenzen von $x$ alle 15 Elemente $\neq0$ ($x^4 = x + 1$, $x^5 = x^2 + x$, …); $x$ hat Ordnung 15. Im **rechten Bild** gibt es von jeder Ordnung $d\mid15$ genau $\varphi(d)$ Elemente: 1, 2, 4 und 8 Elemente der Ordnungen 1, 3, 5, 15. Das beantwortet die Vorhersage: 8 primitive Elemente. Der Frobenius $a\mapsto a^2$ ist bijektiv, und $\mathbb F_4 = \lbrace0, 1, x^2 + x, x^2 + x + 1\rbrace$ ist ein Teilkörper.
* **Wo es vorkommt:** **AES** rechnet in $\mathbb F_{256} = \mathbb F_2[x]/(x^8 + x^4 + x^3 + x + 1)$: Die S-Box ist das Inverse in diesem Körper (mit einer affinen Abbildung), und MixColumns multipliziert in ihm. **Reed-Solomon-Codes** in QR-Codes, CDs und RAID 6 rechnen in $\mathbb F_{256}$, BCH-Codes in $\mathbb F_{2^m}$, elliptische Kurven für Kryptographie über $\mathbb F_p$ oder $\mathbb F_{2^m}$; Quantencodes benutzen $\mathbb F_4$ (Pauli-Operatoren als Elemente von $\mathbb F_4$).

##### *6. Reed-Solomon-Codes: Fehlerkorrektur mit Polynomen*

Wie kann ein Empfänger verfälschte Daten reparieren, ohne zu wissen, wo die Fehler sind? Vorher vorhersagen: Ein Codewort hat 30 Symbole und trägt 20 Nutzsymbole. Wie viele verfälschte Symbole lassen sich sicher korrigieren?

Ein **Reed-Solomon-Code** über $\mathbb F_{929}$ (dem Körper des Barcodes PDF417): Die 20 Nachrichtensymbole sind die Koeffizienten eines Polynoms $P$ vom Grad $< 20$, das Codewort sind seine 30 Werte $P(1), \dots, P(30)$. Für 0 bis 10 verfälschte Symbole an zufälligen Stellen wird je 150-mal mit dem **Berlekamp-Welch-Verfahren** dekodiert (links). Rechts steht ein Codewort mit 5 Fehlern vor und nach der Dekodierung.

**Hintergrund & Fachbegriffe:**
- **Code:** eine Menge zulässiger Wörter; durch Redundanz lassen sich Fehler erkennen und korrigieren.
- **Mindestabstand** $d$: die kleinste Zahl von Stellen, in denen sich zwei Codewörter unterscheiden; ein Code korrigiert $\lfloor(d - 1)/2\rfloor$ Fehler.
- **Singleton-Schranke:** $d\leq n - k + 1$ für jeden Code mit Länge $n$ und $k$ Nutzsymbolen; Codes, die sie erreichen, heißen **MDS-Codes**.
- **Berlekamp-Welch:** ein Dekodierverfahren, das ein lineares Gleichungssystem über dem Körper löst.
- **Fehlerortungspolynom** $E$: ein Polynom, das an den verfälschten Stellen 0 ist.

**Theorie:**

- **Interpolation:** Ein Polynom vom Grad $< k$ ist durch $k$ Werte eindeutig bestimmt (über jedem Körper, weil ein Polynom vom Grad $< k$ höchstens $k - 1$ Nullstellen hat). Zwei verschiedene Nachrichten unterscheiden sich deshalb in höchstens $k - 1$ der $n$ Werte, stimmen also in höchstens $k - 1$ überein: Der Mindestabstand ist $n - k + 1 = 11$, und es werden $t = (n - k)/2 = 5$ Fehler korrigiert.
- **Berlekamp-Welch:** Ist $E$ das normierte Polynom vom Grad $t$ mit Nullstellen an den Fehlerstellen und $Q = PE$, so gilt an allen Stellen $Q(a_i) = r_i\,E(a_i)$, auch an den falschen (dort ist $E(a_i) = 0$). Das sind $n$ lineare Gleichungen für die $n$ unbekannten Koeffizienten von $Q$ und $E$; aus einer Lösung folgt $P = Q/E$. Bei mehr als $t$ Fehlern geht die Division nicht auf, und der Decoder meldet einen Fehler.
- **Warum ein Körper:** Die Interpolation braucht Division; über $\mathbb Z_n$ mit zusammengesetztem $n$ hätten Polynome zu viele Nullstellen ($x^2 - 1$ hat in $\mathbb Z_8$ vier).

```python
rng = np.random.default_rng(6)
p, n_rs, k_rs = 929, 30, 20                                       # 𝔽₉₂₉ wie im Barcode PDF417; Codewortlänge 30, Nachricht 20 Symbole
t_rs = (n_rs - k_rs) // 2
alphas = np.arange(1, n_rs + 1)

def rs_encode(msg):
    """Codewort = Werte des Nachrichtenpolynoms m₀ + m₁a + … an n Stellen a."""
    V = np.array([[pow(int(a), j, p) for j in range(k_rs)] for a in alphas])
    return V @ msg % p

def poly_divmod(num, den):
    """Polynomdivision über 𝔽_p (Koeffizienten aufsteigend)."""
    num = list(num); out = [0] * max(len(num) - len(den) + 1, 1)
    inv_lead = pow(int(den[-1]), -1, p)
    for i in range(len(num) - len(den), -1, -1):
        coef = num[i + len(den) - 1] * inv_lead % p
        out[i] = coef
        for j, dj in enumerate(den):
            num[i + j] = (num[i + j] - coef * dj) % p
    return np.array(out), np.array(num[:len(den) - 1])

def berlekamp_welch(r):
    """Dekodiert mit Berlekamp-Welch: sucht E (normiert, Grad t) und Q (Grad < k + t) mit Q(aᵢ) = rᵢ·E(aᵢ), dann P = Q/E."""
    A, b = [], []
    for a, ri in zip(alphas, r):
        a = int(a)
        rowQ = [pow(a, j, p) for j in range(k_rs + t_rs)]
        rowE = [(-int(ri) * pow(a, j, p)) % p for j in range(t_rs)]
        A.append(rowQ + rowE)
        b.append(int(ri) * pow(a, t_rs, p) % p)
    z = solve_mod_p(A, b, p)
    if z is None:
        return None
    Q, E = z[:k_rs + t_rs], np.r_[z[k_rs + t_rs:], 1]
    P, rem = poly_divmod(Q, E)
    return P[:k_rs] if not np.any(rem % p) else None

results = {e: {"richtig": 0, "Fehler erkannt": 0, "falsch dekodiert": 0} for e in range(0, 11)}
for e, _ in itertools.product(range(0, 11), range(150)):
    msg = rng.integers(0, p, k_rs)
    cw = rs_encode(msg)
    r = cw.copy()
    pos = rng.choice(n_rs, e, replace=False)
    r[pos] = (r[pos] + rng.integers(1, p, e)) % p
    dec = berlekamp_welch(r)
    if dec is None:
        results[e]["Fehler erkannt"] += 1
    elif np.array_equal(dec % p, msg):
        results[e]["richtig"] += 1
    else:
        results[e]["falsch dekodiert"] += 1

msg = rng.integers(0, p, k_rs); cw = rs_encode(msg); r = cw.copy()
pos = rng.choice(n_rs, t_rs, replace=False); r[pos] = (r[pos] + rng.integers(1, p, t_rs)) % p
fig, axes = plt.subplots(1, 2, figsize=(16, 4.6))
es = list(results)
bottom = np.zeros(len(es))
for lab, c in [("richtig", "tab:green"), ("Fehler erkannt", "tab:orange"), ("falsch dekodiert", "tab:red")]:
    vals = np.array([results[e][lab] / 150 for e in es])
    axes[0].bar(es, vals, bottom=bottom, color=c, label=lab)
    bottom += vals
axes[0].axvline(t_rs + 0.5, color="k", ls="--", label=f"Grenze t = (n − k)/2 = {t_rs}")
axes[0].set_xlabel("Zahl der verfälschten Symbole")
axes[0].set_ylabel("Anteil von 150 Versuchen")
axes[0].set_title(f"Reed-Solomon über 𝔽_{p}: n = {n_rs}, k = {k_rs}", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].plot(alphas, cw, "o-", ms=4, label="gesendetes Codewort")
axes[1].plot(alphas[pos], r[pos], "rx", ms=10, label=f"{t_rs} verfälschte Symbole")
axes[1].plot(alphas, rs_encode(berlekamp_welch(r)), "k.", ms=5, label="nach der Dekodierung")
axes[1].set_xlabel("Stelle a (Auswertungspunkt)")
axes[1].set_ylabel("Symbolwert (0 … 928)")
axes[1].set_title("Ein Codewort mit 5 Fehlern wird vollständig repariert", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f"Parameter: n = {n_rs}, k = {k_rs}, Mindestabstand n − k + 1 = {n_rs - k_rs + 1} (Singleton-Schranke erreicht), korrigiert t = {t_rs} Fehler")
for e in (0, 3, 5, 6, 8, 10):
    print(f"{e:2d} Fehler: " + ", ".join(f"{lab} {v}" for lab, v in results[e].items()))
```

* Im **linken Bild** wird bis zu 5 Fehlern jedes Codewort richtig repariert (grün), ab 6 Fehlern erkennt der Decoder in allen 150 Versuchen, dass er nicht reparieren kann (orange); falsch dekodiert wurde nichts. Laut **Ausgabe**: Mindestabstand 11, $t = 5$. Das beantwortet die Vorhersage: 5 Fehler. Dass nie falsch dekodiert wurde, liegt am großen Körper: Ein falsches Polynom, das zufällig zu 25 von 30 Werten passt, ist extrem unwahrscheinlich.
* Im **rechten Bild** liegen die 5 verfälschten Werte (rote Kreuze) weit neben dem gesendeten Codewort; nach der Dekodierung (schwarze Punkte) stimmen alle 30 Werte wieder mit dem Original überein. Der Decoder wusste nicht, wo die Fehler sind; er hat sie über das Fehlerortungspolynom gefunden.
* **Wo es vorkommt:** **QR-Codes** (bis 30 % der Fläche dürfen zerstört sein), **CDs und DVDs** (verschachtelte Reed-Solomon-Codes gleichen Kratzer aus), Festplatten-**RAID 6**, DSL, digitales Fernsehen (DVB), Datenübertragung der Raumsonden Voyager und Cassini, Speicher in Rechenzentren (Erasure Coding). Dieselbe Idee „Polynom durch Werte festlegen“ steckt in **Shamirs Geheimnisteilung** (jedes Teilgeheimnis ist ein Wert eines Polynoms) und in modernen **Zero-Knowledge-Beweisen**.

##### *7. Gröbner-Basen: Polynomgleichungen systematisch lösen*

Wie löst man ein System von Polynomgleichungen in mehreren Unbekannten? Vorher vorhersagen: Wie viele Stellungen hat ein ebener Roboterarm mit zwei Gelenken, deren Hand einen vorgegebenen Punkt erreicht?

Zwei Gleichungssysteme werden mit sympy in eine **Gröbner-Basis** (lexikographische Ordnung) umgerechnet. Erstens Kreis und Hyperbel, $x^2 + y^2 = 4$ und $xy = 1$. Zweitens die inverse Kinematik eines Arms mit Gliedlängen 2 und 1.5, dessen Hand den Punkt $(2.2, 1.4)$ erreichen soll: Unbekannt sind $\cos$ und $\sin$ der beiden Gelenkwinkel, mit den Nebenbedingungen $c^2 + s^2 = 1$. Die Bilder zeigen die gefundenen Lösungen.

**Hintergrund & Fachbegriffe:**
- **Ideal eines Gleichungssystems:** alle Kombinationen $\sum h_if_i$ der Gleichungen $f_i = 0$ mit Polynomfaktoren $h_i$; jede Lösung des Systems ist Nullstelle aller Polynome im Ideal.
- **Monomordnung:** eine Regel, welcher Term eines Polynoms der „führende“ ist; die **lexikographische** Ordnung $x > y$ bevorzugt jede Potenz von $x$ vor allen Potenzen von $y$.
- **Gröbner-Basis:** ein Erzeugendensystem des Ideals, bei dem die Division mit Rest eindeutige Reste liefert.
- **Elimination:** Mit der lexikographischen Ordnung enthält die Gröbner-Basis Polynome, in denen die vorderen Variablen fehlen; das letzte hängt nur von einer Variablen ab.
- **Inverse Kinematik:** aus der gewünschten Position eines Roboterarms die Gelenkwinkel berechnen.

**Theorie:**

- **Division mit Rest in mehreren Variablen:** Anders als in $K[x]$ hängt der Rest im Allgemeinen von der Reihenfolge der Divisoren ab. Eine Gröbner-Basis ist genau so gewählt, dass er es nicht tut; dann liegt ein Polynom genau dann im Ideal, wenn der Rest 0 ist (Zugehörigkeitsproblem).
- **Buchbergers Algorithmus** (1965): Man bildet aus je zwei Basispolynomen ihr **S-Polynom** (die führenden Terme heben sich auf), reduziert es und nimmt nicht verschwindende Reste in die Basis auf, bis alle S-Polynome zu 0 reduzieren. Dass das endet, folgt aus Hilberts Basissatz.
- **Dreiecksform:** Für die lexikographische Ordnung $x_1 > \dots > x_n$ ist die Gröbner-Basis eines Systems mit endlich vielen Lösungen „dreieckig“: ein Polynom nur in $x_n$, dann Polynome in $x_{n-1}, x_n$, und so weiter. Man löst rückwärts, wie beim Gauß-Verfahren; Gröbner-Basen sind dessen nichtlineare Verallgemeinerung.
- **Nullstellensatz:** Ist $1$ in der Gröbner-Basis, hat das System keine komplexe Lösung.

```python
G1 = groebner([x ** 2 + y ** 2 - 4, x * y - 1], x, y, order="lex")
ys = [float(rt) for rt in real_roots(Poly(G1.exprs[-1], y))]
sols1 = [(1 / yy, yy) for yy in ys]                                # aus xy = 1 folgt x = 1/y

c1, s1, c2, s2 = symbols("c1 s1 c2 s2")
l1, l2, tx, ty = 2, Rational(3, 2), Rational(11, 5), Rational(7, 5)
eqs = [c1 ** 2 + s1 ** 2 - 1, c2 ** 2 + s2 ** 2 - 1,
       l1 * c1 + l2 * (c1 * c2 - s1 * s2) - tx, l1 * s1 + l2 * (s1 * c2 + c1 * s2) - ty]
G2 = groebner(eqs, c1, s1, s2, c2, order="lex")
c2_val = float(solve(G2.exprs[-1], c2)[0])                          # letztes Basispolynom ist linear in c2
arms = []
for s2v in (np.sqrt(1 - c2_val ** 2), -np.sqrt(1 - c2_val ** 2)):
    th2 = np.arctan2(s2v, c2_val)
    th1 = np.arctan2(float(ty), float(tx)) - np.arctan2(float(l2) * s2v, float(l1) + float(l2) * c2_val)
    arms.append((th1, th2))

fig, axes = plt.subplots(1, 2, figsize=(15, 5.2))
tt = np.linspace(0, 2 * np.pi, 400)
axes[0].plot(2 * np.cos(tt), 2 * np.sin(tt), label="x² + y² = 4")
xs_ = np.linspace(0.2, 2.5, 200)
axes[0].plot(xs_, 1 / xs_, "tab:orange", label="xy = 1"); axes[0].plot(-xs_, -1 / xs_, "tab:orange")
axes[0].plot([s[0] for s in sols1], [s[1] for s in sols1], "ko", ms=8, label="Lösungen aus der Gröbner-Basis")
axes[0].set_aspect("equal"); axes[0].set_xlim(-2.6, 2.6); axes[0].set_ylim(-2.6, 2.6)
axes[0].set_title("Kreis und Hyperbel: vier Schnittpunkte", fontsize=10)
axes[0].legend(fontsize=8)
for (th1, th2), c, lab in zip(arms, ["tab:blue", "tab:red"], ["Ellbogen unten (θ₂ > 0)", "Ellbogen oben (θ₂ < 0)"]):
    j1 = np.array([float(l1) * np.cos(th1), float(l1) * np.sin(th1)])
    j2 = j1 + float(l2) * np.array([np.cos(th1 + th2), np.sin(th1 + th2)])
    axes[1].plot([0, j1[0], j2[0]], [0, j1[1], j2[1]], "o-", c=c, lw=3, ms=8, label=lab)
axes[1].plot(float(tx), float(ty), "k*", ms=16, label="Ziel (2.2, 1.4)")
axes[1].set_aspect("equal"); axes[1].set_xlim(-0.5, 3.2); axes[1].set_ylim(-0.8, 2.6)
axes[1].set_title("Inverse Kinematik eines Roboterarms: zwei Lösungen", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print("Gröbner-Basis (lex, x > y) für {x² + y² − 4, xy − 1}:", [str(g) for g in G1.exprs])
print(f"Letztes Polynom hängt nur von y ab; seine Nullstellen y = {[round(v, 4) for v in ys]};  Probe x² + y² = {[round(s[0] ** 2 + s[1] ** 2, 10) for s in sols1]}")
print("Gröbner-Basis des Roboterarms (lex, c1 > s1 > s2 > c2):")
for g in G2.exprs:
    print("   ", g)
print(f"Daraus c2 = {c2_val:.4f}, s2 = ±{np.sqrt(1 - c2_val ** 2):.4f};  Winkel (θ₁, θ₂) in Grad: " + ", ".join(f"({np.degrees(a):.1f}, {np.degrees(b):.1f})" for a, b in arms))
```

* Im **linken Bild** schneiden sich Kreis und Hyperbel in vier Punkten (schwarz). Laut **Ausgabe** ist die Gröbner-Basis $\lbrace x + y^3 - 4y,\ y^4 - 4y^2 + 1\rbrace$: Das zweite Polynom hängt nur von $y$ ab, hat die Nullstellen $y = \pm0.5176, \pm1.9319$, und das erste liefert $x = 4y - y^3$. Alle vier Punkte erfüllen $x^2 + y^2 = 4$ exakt.
* Im **rechten Bild** erreicht der Arm den Zielpunkt (Stern) in zwei Stellungen: mit dem Ellbogen unten (blau, Winkel $-2.5°$ und $84.7°$) und oben (rot, $67.4°$ und $-84.7°$). Laut Ausgabe ist das letzte Polynom der Basis linear in $\cos\theta_2$ und gibt $\cos\theta_2 = 0.0917$; daraus folgen $\sin\theta_2 = \pm0.9958$ und die beiden Lösungen. Das beantwortet die Vorhersage: zwei Stellungen (wenn der Punkt erreichbar und nicht am Rand des Arbeitsbereichs liegt).
* **Wo es vorkommt:** **Robotik** (inverse Kinematik, Bahnplanung, Mechanismen mit geschlossenen Ketten), **Computeralgebra** (jedes System zum Lösen von Polynomgleichungen), geometrisches Beweisen (Sätze der Elementargeometrie als Idealzugehörigkeit), algebraische Statistik, Kryptanalyse (algebraische Angriffe auf Chiffren über $\mathbb F_2$), chemische Reaktionsnetzwerke (Gleichgewichte) und die Kalibrierung von Kameras in der Computer Vision (Minimalprobleme wie der Fünf-Punkte-Algorithmus).

##### *8. Körpererweiterungen und Konstruktionen mit Zirkel und Lineal*

Welche Zahlen lassen sich mit Zirkel und Lineal konstruieren, und welche regelmäßigen Vielecke? Vorher vorhersagen: Kann man das regelmäßige Siebeneck konstruieren, und das Siebzehneck?

Für $n = 3$ bis 60 wird der Grad von $\cos(2\pi/n)$ über $\mathbb Q$ berechnet (Formel $\varphi(n)/2$, für einige $n$ mit dem Minimalpolynom aus sympy geprüft); grün sind die $n$, für die er eine Zweierpotenz ist (links). Rechts steht das Siebzehneck mit dem Wert $\cos(2\pi/17)$ aus Gauß' Formel mit verschachtelten Quadratwurzeln. Die Ausgabe gibt die Minimalpolynome der drei klassischen Probleme.

**Hintergrund & Fachbegriffe:**
- **Konstruierbar:** eine Zahl (Streckenlänge), die man aus der Strecke 1 mit Zirkel und Lineal in endlich vielen Schritten erhält.
- **Grad** $[\mathbb Q(\alpha):\mathbb Q]$: der Grad des Minimalpolynoms von $\alpha$.
- **Minimalpolynom:** das normierte Polynom kleinsten Grades mit rationalen Koeffizienten, das $\alpha$ als Nullstelle hat.
- **Fermatsche Primzahl:** eine Primzahl der Form $2^{2^k} + 1$; bekannt sind nur 3, 5, 17, 257, 65 537.
- **Drei klassische Probleme:** Würfelverdopplung ($\sqrt[3]2$), Winkeldreiteilung, Quadratur des Kreises ($\sqrt\pi$).

**Theorie:**

- **Konstruktionen sind quadratische Erweiterungen:** Schnittpunkte von Geraden und Kreisen bekommt man durch lineare und quadratische Gleichungen. Jeder Schritt erweitert den Körper der bisherigen Koordinaten höchstens um eine Quadratwurzel, also um den Grad 2. Eine konstruierbare Zahl liegt in einem Turm von Erweiterungen vom Grad 2, und nach der Gradformel ist ihr Grad eine **Zweierpotenz** (Wantzel 1837).
- **Die klassischen Probleme:** $\sqrt[3]2$ hat Minimalpolynom $x^3 - 2$, Grad 3: Die Würfelverdopplung ist unmöglich. $\cos20°$ hat $8x^3 - 6x - 1$, Grad 3: Der Winkel 60° lässt sich nicht dreiteilen. $\pi$ ist transzendent (Lindemann 1882): Die Quadratur des Kreises ist unmöglich.
- **Gauß-Wantzel:** $\cos(2\pi/n)$ hat Grad $\varphi(n)/2$; das regelmäßige $n$-Eck ist genau dann konstruierbar, wenn $\varphi(n)$ eine Zweierpotenz ist, also wenn $n = 2^k\,p_1\cdots p_r$ mit verschiedenen Fermatschen Primzahlen $p_i$. Gauß fand 1796, mit 18 Jahren, die Konstruktion des Siebzehnecks: $\varphi(17) = 16$; die Galois-Gruppe $\mathbb Z_{16}$ hat eine Kette von Untergruppen mit Index 2, und jede Stufe ist eine Quadratwurzel.

```python
ngon = np.arange(3, 61)
deg = np.array([int(totient(int(m))) // 2 for m in ngon])          # Grad von cos(2π/n) über ℚ
constructible = np.array([d & (d - 1) == 0 for d in deg])
check_deg = {m: Poly(minimal_polynomial(cos(2 * pi / m), x), x).degree() for m in (5, 7, 9, 17)}
examples = {"√2 + √3": sqrt(2) + sqrt(3), "∛2 (Würfelverdopplung)": 2 ** Rational(1, 3), "cos 20° (Dreiteilung von 60°)": cos(pi / 9),
            "cos(2π/7) (Siebeneck)": cos(2 * pi / 7), "cos(2π/17) (Siebzehneck)": cos(2 * pi / 17)}
minpolys = {k: minimal_polynomial(v, x) for k, v in examples.items()}
s17 = np.sqrt(17)
gauss17 = (-1 + s17 + np.sqrt(34 - 2 * s17) + 2 * np.sqrt(17 + 3 * s17 - np.sqrt(34 - 2 * s17) - 2 * np.sqrt(34 + 2 * s17))) / 16

fig, axes = plt.subplots(1, 2, figsize=(16, 4.8), gridspec_kw={"width_ratios": [2, 1]})
axes[0].bar(ngon, deg, color=np.where(constructible, "tab:green", "tab:red"))
axes[0].set_yscale("log", base=2)
axes[0].set_xlabel("Eckenzahl n des regelmäßigen Vielecks")
axes[0].set_ylabel("Grad [ℚ(cos 2π/n) : ℚ] = φ(n)/2")
axes[0].set_title("Grün: Grad ist Zweierpotenz → mit Zirkel und Lineal konstruierbar", fontsize=10)
for m, dx in ((7, -10), (9, 10), (17, 0)):
    axes[0].annotate(f"n = {m}", (m, deg[m - 3]), textcoords="offset points", xytext=(dx, 6), ha="center", fontsize=8)
ang = 2 * np.pi * np.arange(18) / 17
axes[1].plot(np.cos(ang), np.sin(ang), "o-", c="tab:green", ms=4)
axes[1].plot([0, gauss17], [0, 0], "k--", lw=1)
axes[1].plot(gauss17, 0, "k|", ms=20)
axes[1].set_aspect("equal")
axes[1].set_title("Gauß 1796: das Siebzehneck", fontsize=10)
axes[1].set_xticks([]); axes[1].set_yticks([])
plt.tight_layout()
plt.show()

print("Konstruierbare Vielecke bis 60:", list(ngon[constructible]))
print("Gradformel φ(n)/2 gegen Minimalpolynom von cos(2π/n):", {m: (int(totient(m)) // 2, d) for m, d in check_deg.items()})
for k, mp in minpolys.items():
    d_ = Poly(mp, x).degree()
    print(f"{k:32s}: Minimalpolynom {mp}, Grad {d_} → {'konstruierbar' if d_ & (d_ - 1) == 0 else 'nicht konstruierbar'}")
print(f"Gauß' Formel für cos(2π/17) mit verschachtelten Quadratwurzeln: {gauss17:.15f}, direkt {np.cos(2 * np.pi / 17):.15f}")
```

* Im **linken Bild** (logarithmische Achse) sind die Grade der konstruierbaren Vielecke grün: laut **Ausgabe** $n = 3, 4, 5, 6, 8, 10, 12, 15, 16, 17, 20, 24, 30, 32, 34, 40, 48, 51, 60$ bis 60. Das Siebeneck und Neuneck haben Grad 3 (rot), das Siebzehneck Grad 8. Das beantwortet die Vorhersage: Siebeneck nein, Siebzehneck ja. Die Formel $\varphi(n)/2$ stimmt für $n = 5, 7, 9, 17$ mit dem Grad des Minimalpolynoms aus sympy überein.
* Laut Ausgabe hat $\sqrt2 + \sqrt3$ das Minimalpolynom $x^4 - 10x^2 + 1$ (Grad 4, konstruierbar), $\sqrt[3]2$ das Polynom $x^3 - 2$, $\cos20°$ das Polynom $8x^3 - 6x - 1$, $\cos(2\pi/7)$ das Polynom $8x^3 + 4x^2 - 4x - 1$ (alle Grad 3, nicht konstruierbar), $\cos(2\pi/17)$ ein Polynom vom Grad 8. Gauß' Formel aus Quadratwurzeln (Marke im **rechten Bild**) trifft $\cos(2\pi/17) = 0.932472229404356$ auf alle 15 Stellen.
* **Wo es vorkommt:** Die Gradformel ist das Werkzeug für alle Unmöglichkeitsbeweise dieser Art; dieselbe Idee (welche Zahlen erreicht man mit welchen Operationen?) steckt in der Frage, welche Gleichungen durch Wurzeln lösbar sind (Exp. 9), und welche Zahlen ein Computer exakt darstellen kann (algebraische Zahlen in Computeralgebra). Geometrische Konstruktionen mit Zirkel und Lineal entsprechen genau den Erweiterungen vom Grad $2^k$; mit Origami (Falten) kommen Grad-3-Schritte hinzu, damit lassen sich Winkel dreiteilen und Würfel verdoppeln.

##### *9. Galois-Gruppen und die Zerlegung modulo Primzahlen*

Was verrät die Zerlegung eines Polynoms modulo Primzahlen über seine Galois-Gruppe? Vorher vorhersagen: Für welchen Anteil der Primzahlen bleibt $x^5 - x - 1$ modulo $p$ irreduzibel?

Für fünf Polynome berechnet sympy die Galois-Gruppe (als Permutationsgruppe ihrer Nullstellen) und daraus, welcher Anteil der Gruppenelemente welchen **Zyklentyp** hat. Dann wird jedes Polynom modulo aller Primzahlen $3\leq p < 6000$ faktorisiert, die seine Diskriminante nicht teilen, und die Grade der Faktoren werden gezählt. Die Bilder vergleichen beide Verteilungen für $x^3 - 2$ und $x^5 - x - 1$.

**Hintergrund & Fachbegriffe:**
- **Galois-Gruppe eines Polynoms:** die Permutationen seiner komplexen Nullstellen, die alle algebraischen Beziehungen zwischen ihnen erhalten.
- **Zyklentyp:** die Längen der Zyklen einer Permutation, etwa $(2, 3)$ für einen 2-Zyklus und einen 3-Zyklus.
- **Diskriminante:** eine Zahl, die genau dann 0 ist, wenn das Polynom eine doppelte Nullstelle hat; Primzahlen, die sie teilen, sind die Ausnahmen.
- **Auflösbar:** Die Gruppe hat eine Kette von Untergruppen mit abelschen Faktoren (siehe Gruppentheorie).
- **Dichte:** der Grenzwert des Anteils der Primzahlen mit einer Eigenschaft.

**Theorie:**

- **Satz von Dedekind:** Zerfällt $f$ modulo $p$ (für $p$ ohne doppelte Nullstellen) in irreduzible Faktoren der Grade $d_1, \dots, d_r$, so enthält die Galois-Gruppe ein Element mit dem Zyklentyp $(d_1, \dots, d_r)$: den Frobenius zu $p$.
- **Dichtesatz von Chebotarev** (1922): Jeder Zyklentyp kommt bei einem Anteil der Primzahlen vor, der seinem Anteil in der Galois-Gruppe entspricht. Die Primzahlen „würfeln“ also gleichmäßig Elemente der Galois-Gruppe aus. Für $x^5 - x - 1$ mit Galois-Gruppe $S_5$ sind 24 der 120 Elemente 5-Zyklen; also bleibt das Polynom modulo $1/5$ aller Primzahlen irreduzibel.
- **Konsequenzen:** Enthält die Galois-Gruppe keinen $n$-Zyklus, ist das Polynom modulo keiner Primzahl irreduzibel (so bei $x^4 + 1$ und $x^4 - 10x^2 + 1$, Exp. 4). Umgekehrt verrät die Faktorisierung modulo weniger Primzahlen die Galois-Gruppe; so berechnen Computeralgebrasysteme sie.
- **Auflösbarkeit:** $x^3 - 2$ ($S_3$), $x^5 - 2$ (die Frobenius-Gruppe $F_{20}$ der Ordnung 20) und $x^4 - 10x^2 + 1$ ($\mathbb Z_2^2$) haben auflösbare Gruppen, ihre Nullstellen lassen sich mit Wurzeln schreiben. $x^5 - x - 1$ hat $S_5$, nicht auflösbar: Es gibt keine Formel aus Wurzeln für seine Nullstellen.

```python
polys9 = {"x³ − 2": x ** 3 - 2, "x⁴ − 10x² + 1": x ** 4 - 10 * x ** 2 + 1, "x⁴ + 1": x ** 4 + 1, "x⁵ − x − 1": x ** 5 - x - 1, "x⁵ − 2": x ** 5 - 2}
primes9 = list(primerange(3, 6000))

def cycle_type(perm, n):
    seen, out = set(), []
    for i in range(n):
        if i not in seen:
            L, j = 0, i
            while j not in seen:
                seen.add(j); j = perm(j); L += 1
            out.append(L)
    return tuple(sorted(out))

data9 = {}
for name_, f in polys9.items():
    P = Poly(f, x)
    G, _ = galois_group(P)
    deg_ = P.degree()
    elements = list(G.generate())
    types = {}
    for g in elements:
        ct = cycle_type(lambda i: g.array_form[i] if i < len(g.array_form) else i, deg_)
        types[ct] = types.get(ct, 0) + 1
    disc = int(P.discriminant())
    seen_types = {}
    for p in primes9:
        if disc % p == 0:
            continue
        ct = tuple(sorted(Poly(g_, x).degree() for g_, e_ in factor_list(f, modulus=p)[1] for _ in range(e_)))
        seen_types[ct] = seen_types.get(ct, 0) + 1
    total = sum(seen_types.values())
    data9[name_] = (G, len(elements), types, {k: v / total for k, v in seen_types.items()})

fig, axes = plt.subplots(1, 2, figsize=(17, 4.6), gridspec_kw={"width_ratios": [1, 2]})
for ax, nm in zip(axes, ["x³ − 2", "x⁵ − x − 1"]):
    G, order_, types, freq = data9[nm]
    keys = sorted(types, key=lambda k: (len(k), k))
    lab = [" ".join(map(str, k)) for k in keys]
    ax.bar(np.arange(len(keys)) - 0.2, [types[k] / order_ for k in keys], 0.4, label="Anteil in der Galois-Gruppe")
    ax.bar(np.arange(len(keys)) + 0.2, [freq.get(k, 0) for k in keys], 0.4, label=f"Anteil der Primzahlen < 6000")
    ax.set_xticks(range(len(keys)), lab)
    ax.set_xlabel("Grade der Faktoren mod p = Zyklentyp")
    ax.set_ylabel("Anteil")
    ax.set_title(f"{nm}: Galois-Gruppe der Ordnung {order_}", fontsize=10)
    ax.legend(fontsize=8)
plt.tight_layout()
plt.show()

for nm, (G, order_, types, freq) in data9.items():
    dev = max(abs(types.get(k, 0) / order_ - freq.get(k, 0)) for k in set(types) | set(freq))
    print(f"{nm:14s}: Galois-Gruppe Ordnung {order_:3d}, auflösbar {G.is_solvable};  Zyklentypen {dict(sorted(types.items()))};  größte Abweichung der Primzahl-Anteile {dev:.3f}")
print(f"x⁵ − x − 1: Anteil der Primzahlen, modulo derer das Polynom irreduzibel bleibt: {data9['x⁵ − x − 1'][3].get((5,), 0):.3f} (Theorie 24/120 = 0.200)")
```

* Im **linken Bild** ($x^3 - 2$, Galois-Gruppe $S_3$ der Ordnung 6) bleibt das Polynom modulo etwa $1/3$ der Primzahlen irreduzibel (Typ 3, die zwei 3-Zyklen), zerfällt bei der Hälfte in Grad 1 und 2 (die drei Transpositionen) und bei $1/6$ vollständig (Identität). Die gemessenen Anteile treffen die Gruppenanteile. Im **rechten Bild** ($x^5 - x - 1$, $S_5$) passen alle sieben Zyklentypen von $5$ (24 Elemente) bis $1\,1\,1\,1\,1$ (nur die Identität, 1/120) zu den Primzahlanteilen.
* Laut **Ausgabe** ist die größte Abweichung bei allen fünf Polynomen höchstens 0.012; $x^5 - x - 1$ bleibt modulo 0.201 der Primzahlen irreduzibel (Theorie $24/120 = 0.200$), das beantwortet die Vorhersage. Auflösbar sind die Galois-Gruppen von $x^3 - 2$ (Ordnung 6), $x^4 - 10x^2 + 1$ und $x^4 + 1$ (Ordnung 4, nur die Zyklentypen $1\,1\,1\,1$ und $2\,2$), $x^5 - 2$ (Ordnung 20), nicht aber die von $x^5 - x - 1$ (Ordnung 120).
* **Wo es vorkommt:** Chebotarev ist ein Grundwerkzeug der algebraischen Zahlentheorie (Verteilung von Primzahlen in Zahlkörpern, Primzahlen der Form $x^2 + ny^2$). Computeralgebrasysteme bestimmen Galois-Gruppen über die Faktorisierung modulo Primzahlen; dieselbe Statistik prüft, ob ein Polynom wohl irreduzibel ist. In der Kryptographie hängt die Sicherheit mancher Verfahren davon ab, wie ein Polynom modulo $p$ zerfällt (Zahlkörpersieb zum Faktorisieren großer Zahlen).

##### *10. p-adische Zahlen und das Henselsche Lemma*

Was passiert, wenn man Zahlen nach ihrer Teilbarkeit durch $p$ misst statt nach ihrer Größe? Vorher vorhersagen: Was ergibt die unendliche Summe $1 + 2 + 4 + 8 + \dots$, wenn „nah“ bedeutet „durch eine hohe Zweierpotenz teilbar“?

Links wird mit dem Newton-Verfahren modulo $7, 7^2, 7^4, \dots$ eine Lösung von $x^2 = 2$ in den 7-adischen Zahlen berechnet (Hensel-Lifting, Startwert $3^2 = 9\equiv2\bmod7$); gezeigt wird der 7-adische Betrag des Fehlers $x_k^2 - 2$. Rechts stehen die Partialsummen von $1 + 2 + 4 + \dots$, einmal mit dem gewöhnlichen und einmal mit dem 2-adischen Abstand zu $-1$. Die Ausgabe gibt die Ziffern von $\sqrt2$ in Basis 7 und prüft $\sqrt{-1}$ in $\mathbb Z_5$ und $\mathbb Z_3$.

**Hintergrund & Fachbegriffe:**
- **p-adische Bewertung** $v_p(n)$: wie oft $p$ die Zahl $n$ teilt. **p-adischer Betrag** $\vert n\vert_p = p^{-v_p(n)}$: eine Zahl ist klein, wenn sie durch eine hohe Potenz von $p$ teilbar ist.
- **p-adische Zahlen** $\mathbb Z_p$: unendliche Ziffernfolgen $a_0 + a_1p + a_2p^2 + \dots$ nach links; $\mathbb Q_p$ ist ihr Bruchkörper.
- **Vervollständigung:** das Hinzufügen aller Grenzwerte von Cauchy-Folgen, wie $\mathbb R$ aus $\mathbb Q$ entsteht.
- **Henselsches Lemma:** Eine einfache Nullstelle modulo $p$ lässt sich eindeutig zu einer Nullstelle modulo jeder Potenz $p^k$ hochheben.
- **Ultrametrik:** $\vert a + b\vert_p\leq\max(\vert a\vert_p, \vert b\vert_p)$, stärker als die Dreiecksungleichung.

**Theorie:**

- **Zwei Vervollständigungen von $\mathbb Q$:** Nach dem Satz von Ostrowski gibt es auf $\mathbb Q$ bis auf Äquivalenz nur den gewöhnlichen Betrag und die $p$-adischen Beträge. Vervollständigt man nach dem gewöhnlichen, erhält man $\mathbb R$, nach einem $p$-adischen $\mathbb Q_p$. Viele Fragen der Zahlentheorie zerlegt man in je eine Frage über $\mathbb R$ und über alle $\mathbb Q_p$ (Lokal-Global-Prinzip, Hasse-Minkowski für quadratische Formen).
- **Hensel:** Ist $f(a)\equiv0\bmod p$ und $f'(a)\not\equiv0\bmod p$, so konvergiert das Newton-Verfahren $a\mapsto a - f(a)/f'(a)$ $p$-adisch, und jeder Schritt verdoppelt die Zahl der richtigen Stellen (quadratische Konvergenz). Deshalb hat $x^2 = 2$ in $\mathbb Z_7$ eine Lösung (weil $3^2\equiv2\bmod7$), $x^2 = -1$ in $\mathbb Z_5$ (weil $2^2\equiv-1$), aber $x^2 = -1$ in $\mathbb Z_3$ nicht (kein Quadrat modulo 3 ist $-1$).
- **$1 + 2 + 4 + \dots = -1$:** Die Partialsummen sind $2^m - 1$, und $\vert(2^m - 1) - (-1)\vert_2 = \vert2^m\vert_2 = 2^{-m}\to0$. In $\mathbb Z_2$ konvergiert die Reihe also gegen $-1$; so stellen Computer im Zweierkomplement $-1$ als „lauter Einsen“ dar.

```python
def vp(n, p):
    """p-adische Bewertung: wie oft p die ganze Zahl n ≠ 0 teilt."""
    if n == 0:
        return np.inf
    k = 0
    while n % p == 0:
        n //= p; k += 1
    return k

p7, K = 7, 40
xk, mod, newton = 3, 7, []                                         # 3² = 9 ≡ 2 mod 7
for it in range(6):
    newton.append((it, mod, vp(xk * xk - 2, p7)))
    mod = mod * mod
    xk = (xk - (xk * xk - 2) * pow(2 * xk, -1, mod)) % mod         # Newton-Schritt modulo p^(2^k)
root = xk % 7 ** K
digits7 = []
r_ = root
for _ in range(K):
    digits7.append(r_ % 7); r_ //= 7
partial = [sum(2 ** j for j in range(m)) for m in range(1, 41)]

fig, axes = plt.subplots(1, 2, figsize=(16, 4.6))
axes[0].semilogy([n_[0] for n_ in newton], [7.0 ** -n_[2] for n_ in newton], "o-", label="|x_k² − 2|₇ = 7^(−v₇)")
axes[0].set_xlabel("Newton-Schritt k")
axes[0].set_ylabel("7-adischer Betrag des Fehlers")
axes[0].set_title("Hensel-Lifting: die Genauigkeit verdoppelt sich", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].semilogy(range(1, 41), [2.0 ** -vp(s + 1, 2) for s in partial], "o", ms=4, label="|1 + 2 + … + 2^(m−1) − (−1)|₂")
axes[1].semilogy(range(1, 41), [float(s + 1) for s in partial], "s", ms=3, c="grey", label="gewöhnlicher Abstand zu −1")
axes[1].set_xlabel("Zahl der Summanden m")
axes[1].set_ylabel("Abstand")
axes[1].set_title("1 + 2 + 4 + 8 + … = −1 in ℚ₂", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print("Hensel für x² = 2 in ℤ₇: (Schritt, Modul, v₇(x² − 2)) =", [(a, f"7^{vp(m, 7)}", v) for a, m, v in newton])
print(f"√2 in ℤ₇, letzte Stellen zuerst: …{''.join(map(str, digits7[::-1][-20:]))} (Basis 7);  Probe: (√2)² − 2 durch 7^{K} teilbar: {(root * root - 2) % 7 ** K == 0}")
i5 = sqrt_mod(-1, 5 ** 20, all_roots=True)
print(f"√−1 existiert in ℤ₅ (2² ≡ −1 mod 5): Lösungen modulo 5²⁰: {sorted(i5)[:2]};  in ℤ₃ nicht: {sqrt_mod(-1, 3) is None}")
```

* Im **linken Bild** fällt der 7-adische Betrag des Fehlers bei jedem Newton-Schritt auf das Quadrat: Laut **Ausgabe** ist $v_7(x_k^2 - 2) = 1, 2, 4, 8, 16, 33$, die Genauigkeit verdoppelt sich. Die Ausgabe zeigt die 20 niedrigsten Ziffern von $\sqrt2\in\mathbb Z_7$ in Basis 7 ($\dots64112011266421216213$, die letzte Ziffer 3 ist der Startwert), und $(\sqrt2)^2 - 2$ ist durch $7^{40}$ teilbar. $\sqrt{-1}$ existiert in $\mathbb Z_5$, in $\mathbb Z_3$ nicht.
* Im **rechten Bild** wächst der gewöhnliche Abstand der Partialsummen zu $-1$ ohne Grenze (grau), der 2-adische fällt wie $2^{-m}$ (blau). Das beantwortet die Vorhersage: $1 + 2 + 4 + \dots = -1$ in $\mathbb Q_2$. Beides ist richtig, nur in verschiedenen Abstandsbegriffen.
* **Wo es vorkommt:** Hensel-Lifting ist der Standardweg, wie Computeralgebrasysteme Polynome über $\mathbb Z$ faktorisieren (modulo $p$ faktorisieren, dann hochheben) und Gleichungen exakt lösen ($p$-adische Lösungen linearer Systeme mit Dixons Verfahren). Das Zweierkomplement der Rechner ist 2-adische Arithmetik, und Inverse modulo $2^{64}$ berechnet man mit Newton-Hensel. In der Zahlentheorie sind $p$-adische Zahlen unverzichtbar (Wiles' Beweis, Langlands-Programm), in der Physik gibt es $p$-adische Modelle der Stringtheorie.

##### *11. Quadratische Reste, Reziprozität und Gauß-Summen*

Wann ist eine Zahl ein Quadrat modulo $p$, und wie groß werden Summen von Einheitswurzeln? Vorher vorhersagen: Wie groß ist $\vert\sum_{x=0}^{p-1}e^{2\pi ix^2/p}\vert$ für eine Primzahl $p$?

Links steht für $p = 101$, welche Zahlen quadratische Reste sind (Legendre-Symbol $+1$) und welche nicht ($-1$). Rechts werden für alle Primzahlen $5\leq p < 3000$ die quadratische und die kubische **Gauß-Summe** $\sum_xe^{2\pi ix^k/p}$ berechnet und durch $\sqrt p$ geteilt. Die Ausgabe prüft das quadratische Reziprozitätsgesetz für alle Paare ungerader Primzahlen unter 400.

**Hintergrund & Fachbegriffe:**
- **Quadratischer Rest** modulo $p$: eine Zahl $a\not\equiv0$, die modulo $p$ ein Quadrat ist.
- **Legendre-Symbol** $(a/p)$: $+1$ für einen quadratischen Rest, $-1$ für einen Nichtrest, 0 für $p\mid a$.
- **Quadratisches Reziprozitätsgesetz:** verknüpft $(p/q)$ und $(q/p)$ für ungerade Primzahlen.
- **Gauß-Summe:** eine Summe von Einheitswurzeln $e^{2\pi if(x)/p}$ über alle $x$ modulo $p$.
- **Weil-Schranke:** eine obere Schranke für solche Summen mit Polynomen $f$.

**Theorie:**

- **Genau die Hälfte:** $x\mapsto x^2$ ist auf $\mathbb F_p^\times$ zwei zu eins ($x$ und $-x$), also gibt es $(p - 1)/2$ quadratische Reste. Nach Euler ist $(a/p)\equiv a^{(p-1)/2}\bmod p$.
- **Reziprozität** (Gauß 1796, er gab acht Beweise): $(p/q)(q/p) = (-1)^{\frac{p-1}2\cdot\frac{q-1}2}$. Dazu kommen $(-1/p) = +1$ genau für $p\equiv1\bmod4$ und $(2/p) = +1$ genau für $p\equiv\pm1\bmod8$. Damit lässt sich jedes Legendre-Symbol schnell berechnen, ohne Quadrate zu suchen.
- **Gauß-Summen:** Eine Summe von $p$ zufälligen Einheitswurzeln hätte typisch den Betrag $\sqrt p$, aber hier herrscht Struktur: Die quadratische Gauß-Summe hat **genau** den Betrag $\sqrt p$ (Gauß; sie ist $\sqrt p$ oder $i\sqrt p$). Für $x^3$ ist die Summe 0, wenn $p\equiv2\bmod3$ (dann ist $x\mapsto x^3$ bijektiv und die Summe läuft über alle Einheitswurzeln), und höchstens $2\sqrt p$ für $p\equiv1\bmod3$ (**Weil-Schranke** $(d - 1)\sqrt p$ für Grad $d$, eine Folge der Riemann-Vermutung für Kurven über endlichen Körpern, Weil 1948).

```python
odd_primes = list(primerange(3, 400))
recip = all(legendre_symbol(p, q) * legendre_symbol(q, p) == (-1) ** (((p - 1) // 2) * ((q - 1) // 2))
            for p, q in itertools.combinations(odd_primes, 2))
ps = np.array(list(primerange(5, 3000)))

def gauss_sum(p, k):
    xs_ = np.arange(p, dtype=np.int64)
    return np.abs(np.exp(2j * np.pi * (pow_mod_array(xs_, k, p)) / p).sum())

def pow_mod_array(a, k, p):
    out = np.ones_like(a)
    for _ in range(k):
        out = out * a % p
    return out

g2 = np.array([gauss_sum(int(p), 2) for p in ps]) / np.sqrt(ps)
g3 = np.array([gauss_sum(int(p), 3) for p in ps]) / np.sqrt(ps)
p101 = 101
leg = [legendre_symbol(a, p101) for a in range(1, p101)]

fig, axes = plt.subplots(1, 2, figsize=(16, 4.6))
axes[0].bar(range(1, p101), leg, color=["tab:blue" if v > 0 else "tab:red" for v in leg], width=1.0)
axes[0].set_xlabel("a")
axes[0].set_ylabel("Legendre-Symbol (a/101)")
axes[0].set_title("Quadratische Reste modulo 101: genau die Hälfte", fontsize=10)
axes[1].plot(ps, g2, ".", ms=3, label="|Σ e^(2πi x²/p)| / √p")
mask1 = ps % 3 == 1
axes[1].plot(ps[mask1], g3[mask1], ".", ms=3, label="|Σ e^(2πi x³/p)| / √p, p ≡ 1 mod 3")
axes[1].plot(ps[~mask1], g3[~mask1], ".", ms=3, label="dasselbe, p ≡ 2 mod 3")
axes[1].axhline(2, color="k", ls="--", lw=1, label="Weil-Schranke 2")
axes[1].set_xlabel("Primzahl p")
axes[1].set_ylabel("Betrag / √p")
axes[1].set_title("Gauß-Summen: quadratisch genau √p, kubisch höchstens 2√p", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f"Quadratisches Reziprozitätsgesetz für alle {len(odd_primes) * (len(odd_primes) - 1) // 2} Paare ungerader Primzahlen < 400: {recip}")
print(f"Quadratische Reste mod 101: {sum(1 for v in leg if v > 0)} von 100;  (−1/101) = {legendre_symbol(-1 % 101, 101)} (101 ≡ 1 mod 4), (2/101) = {legendre_symbol(2, 101)} (101 ≡ 5 mod 8)")
print(f"Quadratische Gauß-Summe / √p: min {g2.min():.6f}, max {g2.max():.6f};  kubisch: p ≡ 2 mod 3 max {g3[~mask1].max():.1e}, p ≡ 1 mod 3 max {g3[mask1].max():.4f} (≤ 2)")
```

* Im **linken Bild** sind genau 50 der 100 Zahlen modulo 101 quadratische Reste (blau). Das Muster ist symmetrisch um 50.5, weil $(-1/101) = +1$ (101 $\equiv1\bmod4$): mit $a$ ist auch $-a$ ein Rest. Laut **Ausgabe** ist $(2/101) = -1$, passend zu $101\equiv5\bmod8$, und das Reziprozitätsgesetz gilt für alle 2926 Paare ungerader Primzahlen unter 400.
* Im **rechten Bild** liegt die quadratische Gauß-Summe für jede Primzahl genau bei $\sqrt p$ (blaue Linie bei 1; laut Ausgabe min und max 1.000000). Das beantwortet die Vorhersage. Die kubische Summe ist für $p\equiv2\bmod3$ null (grün, bis auf Rundung $10^{-14}$) und für $p\equiv1\bmod3$ über das ganze Band zwischen 0 und $2\sqrt p$ verteilt (orange), mit Maximum 1.9997: Die Weil-Schranke wird nie überschritten, aber fast erreicht.
* **Wo es vorkommt:** Quadratische Reste stecken im **Rabin-Kryptosystem** und im Goldwasser-Micali-Verfahren, im Solovay-Strassen-Primzahltest (über das Jacobi-Symbol), in den **Quadratische-Reste-Codes** (der Golay-Code aus dem Notebook Gruppentheorie ist einer) und in Hadamard-Matrizen (Paley-Konstruktion). Gauß- und Weil-Summen begrenzen die Korrelationen von Pseudozufallsfolgen, etwa der Spreizcodes in GPS und Mobilfunk, und sind ein Werkzeug der analytischen Zahlentheorie und der Theorie der Zufallsextraktoren.

##### *12. Nichtkommutative Ringe: Matrizen und Quaternionen*

Wie sehen Ringe aus, in denen die Reihenfolge der Faktoren zählt? Vorher vorhersagen: Welcher Anteil aller $n\times n$-Matrizen über $\mathbb F_2$ ist für großes $n$ invertierbar?

Links steht die Multiplikationstafel des Rings $M_2(\mathbb F_2)$ aller 16 Matrizen $2\times2$ mit Einträgen 0 und 1; die Ausgabe zählt Einheiten, Nullteiler, nilpotente und idempotente Elemente. Rechts wird für $n = 1$ bis 8 der Anteil invertierbarer Matrizen über $\mathbb F_2$ und $\mathbb F_3$ in je 4000 Zufallsstichproben gemessen und mit der Formel verglichen. Zuletzt prüft die Ausgabe die Rechenregeln der Quaternionen.

**Hintergrund & Fachbegriffe:**
- **Matrixring** $M_n(K)$: alle $n\times n$-Matrizen mit Einträgen in $K$, mit Matrixaddition und -multiplikation; für $n\geq2$ nicht kommutativ.
- **$GL(n, q)$:** die invertierbaren Matrizen über $\mathbb F_q$, die Einheitengruppe von $M_n(\mathbb F_q)$.
- **Quaternionen** $\mathbb H$: Zahlen $a + bi + cj + dk$ mit $i^2 = j^2 = k^2 = ijk = -1$.
- **Schiefkörper:** ein Ring, in dem jedes Element $\neq0$ invertierbar ist, ohne dass die Multiplikation kommutativ sein muss.
- **Norm einer Quaternion:** $\vert q\vert^2 = a^2 + b^2 + c^2 + d^2$; das Inverse ist $\bar q/\vert q\vert^2$ mit $\bar q = a - bi - cj - dk$.

**Theorie:**

- **Matrizen haben Nullteiler:** Die Matrix mit einer einzigen 1 oben rechts ist nilpotent, ihr Quadrat ist 0; in $M_2(\mathbb F_2)$ sind 6 Matrizen invertierbar ($GL(2, 2)\cong S_3$), die übrigen 9 von Null verschiedenen sind Nullteiler. Trotzdem ist $M_n(K)$ ein **einfacher** Ring (keine zweiseitigen Ideale außer 0 und dem ganzen Ring).
- **Anteil der Einheiten:** $\vert GL(n, q)\vert = \prod_{k=0}^{n-1}(q^n - q^k)$ (die Zeilen müssen der Reihe nach linear unabhängig sein), also ist der Anteil $\prod_{k=1}^n(1 - q^{-k})$; für $q = 2$ und $n\to\infty$ strebt er gegen 0.2888.
- **Quaternionen:** Hamilton 1843. Die Norm ist multiplikativ, $\vert ab\vert = \vert a\vert\,\vert b\vert$, deshalb hat jedes $q\neq0$ ein Inverses: $\mathbb H$ ist ein Schiefkörper. Einheitsquaternionen drehen Vektoren durch $v\mapsto qv\bar q$ (die Gruppe $SU(2)$ aus der Gruppentheorie).
- **Satz von Frobenius:** Die einzigen endlichdimensionalen assoziativen Divisionsalgebren über $\mathbb R$ sind $\mathbb R$, $\mathbb C$ und $\mathbb H$. **Satz von Wedderburn:** Jeder endliche Schiefkörper ist kommutativ, also ein Körper; nichtkommutative Schiefkörper sind immer unendlich.

```python
mats = [np.array(m).reshape(2, 2) for m in itertools.product([0, 1], repeat=4)]   # alle 16 Matrizen über 𝔽₂
idx = {m.tobytes(): i for i, m in enumerate(mats)}
MT = np.array([[idx[(A @ B % 2).tobytes()] for B in mats] for A in mats])
units = [i for i, A in enumerate(mats) if round(np.linalg.det(A)) % 2 == 1]
zero_div = [i for i, A in enumerate(mats) if i not in units and A.any()]
nilp = [i for i, A in enumerate(mats) if not (A @ A % 2).any()]
idem = [i for i, A in enumerate(mats) if np.array_equal(A @ A % 2, A)]
commuting = sum(1 for i in range(16) for j in range(16) if MT[i, j] == MT[j, i])

def frac_invertible(q, n):
    return np.prod([1 - q ** (-k) for k in range(1, n + 1)])

rng = np.random.default_rng(12)
sampled = {}
for q in (2, 3):
    sampled[q] = []
    for n_ in range(1, 9):
        cnt = sum(rank_mod_p(rng.integers(0, q, (n_, n_)), q) == n_ for _ in range(4000))
        sampled[q].append(cnt / 4000)

def qmul(a, b):
    """Quaternionenprodukt (w, x, y, z)."""
    w1, x1, y1, z1 = a; w2, x2, y2, z2 = b
    return np.array([w1 * w2 - x1 * x2 - y1 * y2 - z1 * z2, w1 * x2 + x1 * w2 + y1 * z2 - z1 * y2,
                     w1 * y2 - x1 * z2 + y1 * w2 + z1 * x2, w1 * z2 + x1 * y2 - y1 * x2 + z1 * w2])

fig, axes = plt.subplots(1, 2, figsize=(16, 4.8))
axes[0].imshow(MT, cmap="viridis")
axes[0].set_title(f"M₂(𝔽₂): Multiplikationstafel der 16 Matrizen ({commuting} von 256 Paaren vertauschen)", fontsize=10)
axes[0].set_xlabel("B"); axes[0].set_ylabel("A")
for q, c in [(2, "tab:blue"), (3, "tab:orange")]:
    axes[1].plot(range(1, 9), sampled[q], "o", c=c, label=f"𝔽_{q}: Stichprobe")
    axes[1].plot(range(1, 9), [frac_invertible(q, n_) for n_ in range(1, 9)], "-", c=c, label=f"𝔽_{q}: Π (1 − q^(−k))")
axes[1].axhline(frac_invertible(2, 60), color="tab:blue", ls=":", lw=1)
axes[1].set_xlabel("Matrixgröße n")
axes[1].set_ylabel("Anteil invertierbarer Matrizen")
axes[1].set_title("Einheiten in Mₙ(𝔽_q): |GL(n, q)| / q^(n²)", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f"M₂(𝔽₂): {len(units)} Einheiten (= |GL(2,2)| = 6 ≅ S₃), {len(zero_div)} Nullteiler ≠ 0, {len(nilp)} nilpotente (A² = 0), {len(idem)} idempotente (A² = A)")
i_, j_, k_ = np.array([0, 1, 0, 0]), np.array([0, 0, 1, 0]), np.array([0, 0, 0, 1])
print(f"Quaternionen: ij = {qmul(i_, j_)}, ji = {qmul(j_, i_)} (nicht kommutativ), i² = j² = k² = ijk = {qmul(qmul(i_, j_), k_)[0]:.0f}")
qa = np.array([1.0, 2.0, -1.0, 0.5]); qb = np.array([0.3, -1.0, 2.0, 1.0])
norm = lambda q_: np.sqrt(q_ @ q_)
print(f"Norm ist multiplikativ: |ab| = {norm(qmul(qa, qb)):.6f} = |a|·|b| = {norm(qa) * norm(qb):.6f} → jedes q ≠ 0 hat das Inverse q*/|q|² (Schiefkörper)")
print(f"Anteil invertierbarer n×n-Matrizen über 𝔽₂ für n → ∞: {frac_invertible(2, 60):.4f}")
```

* Die **linke Tafel** ist nicht symmetrisch: laut Bildtitel vertauschen nur 88 der 256 Paare. Laut **Ausgabe** hat $M_2(\mathbb F_2)$ 6 Einheiten ($= \vert GL(2, 2)\vert$, die Gruppe $S_3$), 9 Nullteiler $\neq0$, 4 nilpotente Matrizen ($A^2 = 0$, die Nullmatrix eingeschlossen) und 8 idempotente ($A^2 = A$, die Projektionen). Die Zeilen der Einheiten sind Permutationen aller 16 Elemente, die übrigen nicht.
* Im **rechten Bild** liegen die Stichproben auf der Formel $\prod(1 - q^{-k})$: Über $\mathbb F_2$ ist eine $1\times1$-Matrix zur Hälfte invertierbar, eine große nur zu 28.9 % (gepunktete Grenze 0.2888 laut Ausgabe); über $\mathbb F_3$ sind es etwa 56 %. Das beantwortet die Vorhersage. Laut Ausgabe ist bei den Quaternionen $ij = k$, aber $ji = -k$, $i^2 = j^2 = k^2 = ijk = -1$, und $\vert ab\vert = \vert a\vert\,\vert b\vert = 6.169481$.
* **Wo es vorkommt:** Matrixringe sind die Algebra der linearen Abbildungen und der **Quantenmechanik** (Observable sind Matrizen, $XY\neq YX$). Zufällige Matrizen über $\mathbb F_2$ sind invertierbar mit Wahrscheinlichkeit 0.289; das bestimmt, wie viele Pakete ein **Netzwerkcode** oder ein Fountain-Code zur Dekodierung braucht. **Quaternionen** speichern Orientierungen in Spiele-Engines, Robotik und Raumfahrt (keine Kardansperre wie bei Euler-Winkeln); die **dualen Quaternionen** beschreiben Drehung und Verschiebung zugleich (Skinning in der Computeranimation).

##### *13. Moduln über ℤ: Smith-Normalform und Homologie*

Wie zerlegt man eine abelsche Gruppe, die durch Relationen gegeben ist, und wie zählt man die Löcher einer Fläche? Vorher vorhersagen: Hat die projektive Ebene ein „Loch“, und wenn ja, welcher Art?

Erstens wird die Gruppe $\mathbb Z^3/\langle\text{Zeilen von }A\rangle$ mit der **Smith-Normalform** von $A$ zerlegt. Zweitens werden die Homologiegruppen von drei triangulierten Flächen berechnet: Kugel (Rand eines Tetraeders), Torus (7 Ecken, 14 Dreiecke) und projektive Ebene (6 Ecken, 10 Dreiecke). Dafür werden die **Randmatrizen** aufgestellt und in Smith-Normalform gebracht. Links steht die Randmatrix $\partial_2$ der projektiven Ebene, rechts die Betti-Zahlen der drei Flächen.

**Hintergrund & Fachbegriffe:**
- **Modul** über einem Ring: wie ein Vektorraum, aber mit Skalaren aus einem Ring; ein $\mathbb Z$-Modul ist dasselbe wie eine abelsche Gruppe.
- **Smith-Normalform:** Jede ganzzahlige Matrix lässt sich durch invertierbare ganzzahlige Zeilen- und Spaltenumformungen auf Diagonalform $\mathrm{diag}(d_1, d_2, \dots)$ mit $d_1\mid d_2\mid\dots$ bringen.
- **Triangulierung:** eine Zerlegung einer Fläche in Dreiecke, die entlang ganzer Kanten aneinanderstoßen.
- **Randmatrix** $\partial_k$: ordnet jedem $k$-Simplex (Kante, Dreieck) die Summe seiner Seiten mit Vorzeichen zu.
- **Homologie** $H_k$: Zyklen (Ketten ohne Rand) modulo Ränder; **Betti-Zahl** $b_k$: der Rang von $H_k$, die Zahl der unabhängigen $k$-dimensionalen Löcher. **Torsion:** endliche zyklische Anteile von $H_k$.
- **Euler-Charakteristik:** Ecken − Kanten + Flächen $= b_0 - b_1 + b_2$.

**Theorie:**

- **Struktursatz für endlich erzeugte Moduln über einem Hauptidealring:** Jeder ist $\cong R^r\oplus R/(d_1)\oplus\dots\oplus R/(d_s)$ mit $d_1\mid\dots\mid d_s$. Die Smith-Normalform berechnet diese Zerlegung: $\mathbb Z^n/\mathrm{Bild}(A)\cong\bigoplus\mathbb Z/d_i\oplus\mathbb Z^{n - \mathrm{rang}}$. Für $K[x]$ statt $\mathbb Z$ liefert derselbe Satz die Jordan-Normalform von Matrizen.
- **Warum über $\mathbb Z$ und nicht über $\mathbb Q$:** Über $\mathbb Q$ würde Gauß-Elimination nur den Rang liefern; ganzzahlig bleiben auch die Teiler $d_i$ sichtbar. Das ist der Unterschied zwischen „wie viele Löcher“ (Betti-Zahlen) und „welcher Art“ (Torsion). Über $\mathbb Z_n$ mit zusammengesetztem $n$ versagt Gauß-Elimination ganz, weil nicht jedes Element $\neq0$ invertierbar ist; dort braucht man Hermite- und Smith-Normalform.
- **Homologie:** $b_k = \dim\ker\partial_k - \mathrm{rang}\,\partial_{k+1}$, und die Torsion von $H_k$ sind die Teiler $d_i > 1$ der Smith-Normalform von $\partial_{k+1}$. Kugel: $b = (1, 0, 1)$; Torus: $(1, 2, 1)$, zwei unabhängige Schleifen; projektive Ebene: $(1, 0, 0)$ mit $H_1 = \mathbb Z_2$, eine Schleife, die nicht zusammenziehbar ist, deren doppelter Umlauf aber schon.

```python
A13 = Matrix([[2, 4, 4], [-6, 6, 12], [10, 4, 16]])
S13 = smith_normal_form(A13, domain=ZZ)

def boundary(simplices_k, simplices_km1):
    """Randmatrix ∂: Kanten bzw. Dreiecke → Seiten, mit Vorzeichen (−1)^i für die weggelassene i-te Ecke."""
    index = {s: i for i, s in enumerate(simplices_km1)}
    D = np.zeros((len(simplices_km1), len(simplices_k)), dtype=int)
    for j, s in enumerate(simplices_k):
        for i in range(len(s)):
            D[index[s[:i] + s[i + 1:]], j] = (-1) ** i
    return D

def homology(triangles):
    tri = sorted(tuple(sorted(t)) for t in triangles)
    edges = sorted({e for t in tri for e in itertools.combinations(t, 2)})
    verts = sorted({(v,) for t in tri for v in t})
    d1, d2 = boundary(edges, verts), boundary(tri, edges)
    snf1 = [abs(int(v)) for v in smith_normal_form(Matrix(d1), domain=ZZ).diagonal() if v != 0]
    snf2 = [abs(int(v)) for v in smith_normal_form(Matrix(d2), domain=ZZ).diagonal() if v != 0]
    r1, r2 = len(snf1), len(snf2)
    betti = (len(verts) - r1, len(edges) - r1 - r2, len(tri) - r2)
    torsion1 = [d for d in snf2 if d > 1]
    return dict(V=len(verts), E=len(edges), F=len(tri), betti=betti, torsion=torsion1, euler=len(verts) - len(edges) + len(tri), d2=d2)

surfaces = {
    "Kugel (Tetraederrand)": [(0, 1, 2), (0, 1, 3), (0, 2, 3), (1, 2, 3)],
    "Torus (7 Ecken)": [(i, (i + 1) % 7, (i + 3) % 7) for i in range(7)] + [(i, (i + 2) % 7, (i + 3) % 7) for i in range(7)],
    "projektive Ebene (6 Ecken)": [(1, 2, 3), (1, 3, 4), (1, 4, 5), (1, 5, 6), (1, 6, 2), (2, 3, 5), (3, 4, 6), (4, 5, 2), (5, 6, 3), (6, 2, 4)],
}
H = {nm: homology(tr) for nm, tr in surfaces.items()}

fig, axes = plt.subplots(1, 2, figsize=(16, 4.8), gridspec_kw={"width_ratios": [1.2, 1]})
axes[0].imshow(H["projektive Ebene (6 Ecken)"]["d2"], cmap="RdBu", vmin=-1, vmax=1)
axes[0].set_xlabel("Dreieck"); axes[0].set_ylabel("Kante")
axes[0].set_title("Randmatrix ∂₂ der projektiven Ebene (rot −1, blau +1)", fontsize=10)
names_ = list(H)
w_ = 0.25
for k_, c in zip(range(3), ["tab:blue", "tab:orange", "tab:green"]):
    axes[1].bar(np.arange(3) + (k_ - 1) * w_, [H[nm]["betti"][k_] for nm in names_], w_, color=c, label=f"Betti-Zahl b_{k_}")
for i, nm in enumerate(names_):
    if H[nm]["torsion"]:
        axes[1].annotate(f"Torsion ℤ_{H[nm]['torsion'][0]} in H₁", (i, 0.1), ha="center", fontsize=9, color="tab:red")
axes[1].set_xticks(range(3), [n.split(" (")[0] for n in names_])
axes[1].set_ylabel("Rang")
axes[1].set_title("Homologie aus der Smith-Normalform", fontsize=10)
axes[1].legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f"Relationenmatrix A = {A13.tolist()}: Smith-Normalform diag{tuple(int(v) for v in S13.diagonal())} → ℤ³/Zeilen(A) ≅ ℤ₂ × ℤ₂ × ℤ₁₅₆, Ordnung {int(np.prod([int(v) for v in S13.diagonal()]))} = |det A| = {abs(A13.det())}")
for nm, h in H.items():
    print(f"{nm:28s}: E − K + F = {h['V']} − {h['E']} + {h['F']} = {h['euler']};  Betti (b₀, b₁, b₂) = {h['betti']};  Torsion in H₁: {h['torsion'] or 'keine'}")
```

* Laut **Ausgabe** hat die Relationenmatrix $A$ die Smith-Normalform $\mathrm{diag}(2, 2, 156)$: $\mathbb Z^3/\langle A\rangle\cong\mathbb Z_2\times\mathbb Z_2\times\mathbb Z_{156}$, eine Gruppe mit $624 = \vert\det A\vert$ Elementen. Das **linke Bild** zeigt die Randmatrix $\partial_2$ der projektiven Ebene: 15 Kanten mal 10 Dreiecke, in jeder Spalte drei Einträge $\pm1$ (die drei Seiten eines Dreiecks), in jeder Zeile zwei (jede Kante grenzt an zwei Dreiecke).
* Im **rechten Bild** stehen die Betti-Zahlen: Kugel $(1, 0, 1)$, Torus $(1, 2, 1)$, projektive Ebene $(1, 0, 0)$; laut Ausgabe stimmen die Euler-Charakteristiken $2, 0, 1$ damit überein. Die projektive Ebene hat keine freie Schleife, aber Torsion $\mathbb Z_2$ in $H_1$ (rote Beschriftung) und kein $H_2$, weil sie nicht orientierbar ist. ##### *14. Polynomringe in der Post-Quanten-Kryptographie*

Wie verschlüsselt man so, dass auch ein Quantencomputer es nicht brechen kann? Vorher vorhersagen: Wie groß darf das Rauschen sein, bevor die Entschlüsselung Fehler macht?

Eine vereinfachte Fassung von ML-KEM (Kyber), dem 2024 standardisierten Post-Quanten-Verfahren, im Ring $\mathbb Z_{3329}[x]/(x^{256} + 1)$. Der öffentliche Schlüssel ist $(a, b = as + e)$ mit geheimem kleinem $s$ und kleinem Rauschen $e$; eine Nachricht aus 256 Bits wird als $(u, v) = (ar + e_1,\ br + e_2 + \lceil q/2\rfloor m)$ verschlüsselt und mit $v - us$ entschlüsselt. Links steht das Rauschen nach dem Entschlüsseln für den Rauschparameter $\eta = 2$, rechts die Bitfehlerrate für wachsendes $\eta$. Die Ausgabe prüft außerdem die schnelle Multiplikation mit der zahlentheoretischen Transformation (NTT).

**Hintergrund & Fachbegriffe:**
- **Ring-LWE** (Learning with Errors over Rings): Aus $(a, as + e)$ mit kleinem Rauschen $e$ das Geheimnis $s$ zu finden, gilt als schwer, auch für Quantencomputer; es ist mit schwierigen Gitterproblemen verwandt.
- **Zentrierte Binomialverteilung** mit Parameter $\eta$: Werte zwischen $-\eta$ und $\eta$, Varianz $\eta/2$.
- **Negazyklisch:** In $\mathbb Z_q[x]/(x^n + 1)$ wird $x^n$ durch $-1$ ersetzt; bei der Multiplikation wickeln sich die hohen Koeffizienten mit Vorzeichenwechsel nach vorn.
- **NTT** (zahlentheoretische Transformation): die diskrete Fourier-Transformation über $\mathbb Z_q$ statt $\mathbb C$, mit einer Einheitswurzel modulo $q$ statt $e^{2\pi i/n}$.
- **ML-KEM:** Module-Lattice Key Encapsulation Mechanism, der NIST-Standard FIPS 203 (2024), hervorgegangen aus Kyber.

**Theorie:**

- **Warum Entschlüsseln klappt:** $v - us = (as + e)r + e_2 + \lceil q/2\rfloor m - (ar + e_1)s = er + e_2 - e_1s + \lceil q/2\rfloor m$. Die Terme mit $a$ heben sich auf, weil der Ring kommutativ ist; übrig bleibt kleines Rauschen plus $\lceil q/2\rfloor m$. Ein Koeffizient nahe 0 bedeutet Bit 0, nahe $q/2$ Bit 1. Das klappt, solange jeder Rauschkoeffizient kleiner als $q/4 = 832$ bleibt.
- **Größe des Rauschens:** Jeder Koeffizient von $er$ ist eine Summe von 256 Produkten kleiner Zahlen; bei $\eta = 2$ hat sie Standardabweichung etwa $256^{1/2}\cdot\eta/2 = 16$, zusammen mit $e_1s$ und $e_2$ etwa 22. Der Abstand zur Grenze 832 ist riesig; das echte ML-KEM wählt die Parameter so, dass die Fehlerwahrscheinlichkeit unter $2^{-139}$ liegt.
- **Schnelle Multiplikation:** Für $q = 7681$ gibt es eine 512-te Einheitswurzel $\psi$ ($\psi^{256}\equiv-1$); mit ihr wird die negazyklische Faltung zu einer gewöhnlichen DFT der Länge 256 (Koeffizienten mit $\psi^j$ gewichtet), und die Multiplikation kostet $O(n\log n)$ statt $O(n^2)$. Kyber benutzt mit $q = 3329$ eine Variante, weil dort nur 256-te Einheitswurzeln existieren.
- **Warum Ringe statt Matrizen:** Ein Element von $\mathbb Z_q[x]/(x^n + 1)$ entspricht einer strukturierten $n\times n$-Matrix; so braucht der Schlüssel $n$ statt $n^2$ Zahlen.

```python
rng = np.random.default_rng(14)
nq, q = 256, 3329                                                  # Parameter von ML-KEM (Kyber)

def cbd(eta, size):
    """Zentrierte Binomialverteilung: Summe von eta Münzwürfen minus Summe von eta Münzwürfen."""
    return rng.binomial(eta, 0.5, size) - rng.binomial(eta, 0.5, size)

def centered(v):
    return ((v + q // 2) % q) - q // 2

def rlwe_trial(eta, n_msgs=20):
    """Schlüssel erzeugen, n_msgs Nachrichten zu je 256 Bit verschlüsseln und entschlüsseln; gibt Bitfehler und Rauschen zurück."""
    a = rng.integers(0, q, nq)
    s, e = cbd(eta, nq), cbd(eta, nq)
    b = (polymul_mod(a, s, q, nq) + e) % q                        # öffentlicher Schlüssel (a, b = a·s + e)
    errors, noise = 0, []
    for _ in range(n_msgs):
        m = rng.integers(0, 2, nq)
        r, e1, e2 = cbd(eta, nq), cbd(eta, nq), cbd(eta, nq)
        u = (polymul_mod(a, r, q, nq) + e1) % q
        v = (polymul_mod(b, r, q, nq) + e2 + (q + 1) // 2 * m) % q
        d = centered(v - polymul_mod(u, s, q, nq))                # = Rauschen + ⌈q/2⌋·m
        m_dec = (np.abs(d) > q // 4).astype(int)
        errors += int(np.sum(m_dec != m))
        noise.extend(centered(d - (q + 1) // 2 * m).tolist())
    return errors / (n_msgs * nq), np.array(noise)

err2, noise2 = rlwe_trial(2)
etas = [2, 8, 16, 24, 32, 40, 48, 56, 64]
rates = [rlwe_trial(e, 10)[0] for e in etas]
psi, q_ntt = 7, 7681                                               # NTT-Prüfung mit q = 7681 (hat 512-te Einheitswurzeln)
psi = next(g for g in range(2, q_ntt) if pow(g, 256, q_ntt) == q_ntt - 1)
A_ = rng.integers(0, q_ntt, nq); B_ = rng.integers(0, q_ntt, nq)
w_ = pow(psi, 2, q_ntt)
Vm = np.array([[pow(w_, i * j, q_ntt) for j in range(nq)] for i in range(nq)], dtype=object)
tw = np.array([pow(psi, j, q_ntt) for j in range(nq)], dtype=object)
ntt = lambda a: Vm.dot(a * tw % q_ntt) % q_ntt
inv_n, inv_psi = pow(nq, -1, q_ntt), pow(psi, -1, q_ntt)
Vinv = np.array([[pow(w_, -i * j % (2 * nq), q_ntt) for j in range(nq)] for i in range(nq)], dtype=object)
intt = lambda A: (Vinv.dot(A) % q_ntt * inv_n % q_ntt) * np.array([pow(inv_psi, j, q_ntt) for j in range(nq)], dtype=object) % q_ntt
prod_ntt = intt(ntt(A_.astype(object)) * ntt(B_.astype(object)) % q_ntt).astype(np.int64)

fig, axes = plt.subplots(1, 2, figsize=(16, 4.6))
axes[0].hist(noise2, bins=np.arange(-120, 121, 4), density=True, alpha=0.7, label="Rauschen nach dem Entschlüsseln (η = 2)")
for sgn in (-1, 1):
    axes[0].axvline(sgn * q / 4, color="k", ls="--", lw=1)
axes[0].set_xlim(-q / 2, q / 2)
axes[0].set_xlabel("Rauschwert (zentriert modulo q = 3329)")
axes[0].set_ylabel("Dichte")
axes[0].set_title("Das Rauschen bleibt weit innerhalb ±q/4 (gestrichelt)", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].plot(etas, rates, "o-")
axes[1].set_xlabel("Rauschparameter η")
axes[1].set_ylabel("Anteil falsch entschlüsselter Bits")
axes[1].set_title("Zu viel Rauschen zerstört die Entschlüsselung", fontsize=10)
plt.tight_layout()
plt.show()

print(f"Ring ℤ_{q}[x]/(x^{nq} + 1): 256-Bit-Nachrichten mit η = 2: Bitfehlerrate {err2:.4f};  Rauschen: Standardabweichung {noise2.std():.1f}, größter Betrag {np.abs(noise2).max()} (Grenze q/4 = {q / 4:.0f})")
print("Bitfehlerrate gegen η: " + ", ".join(f"η = {e}: {r_:.4f}" for e, r_ in zip(etas, rates)))
print(f"Negazyklische NTT mit q = {q_ntt}, ψ = {psi} (ψ²⁵⁶ ≡ −1): Produkt über NTT = Produkt direkt: {np.array_equal(prod_ntt, polymul_mod(A_, B_, q_ntt, nq))}")
```

* Im **linken Bild** ist das Rauschen nach dem Entschlüsseln eng um 0 verteilt, weit innerhalb der Grenzen $\pm q/4 = \pm832$ (gestrichelt). Laut **Ausgabe** hat es bei $\eta = 2$ die Standardabweichung 21.8 und den größten Betrag 92; von 20 Nachrichten mit je 256 Bits wurde kein einziges Bit falsch entschlüsselt.
* Im **rechten Bild** steigt die Bitfehlerrate, sobald das Rauschen in die Nähe der Grenze kommt: laut Ausgabe 0 bis $\eta = 16$, dann 0.0016 bei $\eta = 24$, 0.016 bei 32, 0.13 bei 48 und 0.28 bei 64. Das beantwortet die Vorhersage: Das Rauschen muss deutlich unter $q/4$ bleiben; die Sicherheit verlangt aber, dass es nicht zu klein ist. Laut Ausgabe stimmt die Multiplikation über die NTT mit $q = 7681$ und $\psi = 62$ exakt mit der direkten überein.
* **Wo es vorkommt:** **ML-KEM** (FIPS 203) und die Signatur **ML-DSA** (FIPS 204, aus Dilithium) rechnen in genau solchen Ringen; sie sind seit 2024 Standard und werden in TLS (Chrome, Cloudflare), Signal (PQXDH) und iMessage (PQ3) gegen künftige Quantenangriffe eingesetzt. Dieselben Ringe mit der NTT sind die Grundlage der **voll homomorphen Verschlüsselung** (Rechnen auf verschlüsselten Daten, BFV und CKKS), die auch für datenschutzfreundliches maschinelles Lernen benutzt wird.

<details><summary>1. Was unterscheidet einen Ring von einem Körper?</summary>

Im Ring kann man addieren, subtrahieren und multiplizieren; im Körper zusätzlich durch jedes Element außer 0 teilen, und die Multiplikation ist kommutativ. $\mathbb Z$ ist ein Ring, $\mathbb Q$ ein Körper.
</details>

<details><summary>2. Warum ist in ℤₙ jedes Element ≠ 0 entweder Einheit oder Nullteiler?</summary>

Ist $\mathrm{ggT}(a, n) = 1$, liefert Bézout ein Inverses. Ist $d = \mathrm{ggT}(a, n) > 1$, dann ist $a\cdot(n/d)\equiv0$ mit $n/d\not\equiv0$. Deshalb ist $\mathbb Z_n$ genau für prime $n$ ein Körper.
</details>

<details><summary>3. Was ist ein Ideal, und wozu braucht man Faktorringe?</summary>

Eine additive Untergruppe, die Produkte mit beliebigen Ringelementen schluckt. Der Faktorring $R/I$ rechnet „modulo $I$“; so entstehen $\mathbb Z_n$, $\mathbb C = \mathbb R[x]/(x^2 + 1)$ und alle endlichen Körper. Ist $I$ maximal, ist $R/I$ ein Körper.
</details>

<details><summary>4. Warum braucht der euklidische Algorithmus nur logarithmisch viele Schritte?</summary>

Nach zwei Schritten hat sich die größere Zahl mindestens halbiert. Am langsamsten sind benachbarte Fibonacci-Zahlen, im Mittel braucht man etwa $0.843\ln N$ Schritte.
</details>

<details><summary>5. Was sagt der chinesische Restsatz, und wann gilt er nicht?</summary>

Für teilerfremde $m, n$ ist $\mathbb Z_{mn}\cong\mathbb Z_m\times\mathbb Z_n$: Jedes Paar von Resten kommt genau einmal vor. Ohne Teilerfremdheit werden nur manche Paare getroffen.
</details>

<details><summary>6. Warum ist die Zerlegung in ℤ[√−5] nicht eindeutig, und was rettet die Eindeutigkeit?</summary>

$6 = 2\cdot3 = (1 + \sqrt{-5})(1 - \sqrt{-5})$, und alle vier Faktoren sind irreduzibel, weil es keine Elemente mit Norm 2 oder 3 gibt. Zerlegt man Ideale statt Zahlen, ist die Zerlegung in Primideale eindeutig (Dedekind); die Klassenzahl misst den Unterschied.
</details>

<details><summary>7. Welche Primzahlen sind Summen zweier Quadrate, und warum?</summary>

2 und die Primzahlen $\equiv1\bmod4$. Sie zerfallen in den Gaußschen Zahlen als $(a + bi)(a - bi) = a^2 + b^2$; Primzahlen $\equiv3\bmod4$ bleiben dort prim.
</details>

<details><summary>8. Wie viele irreduzible Polynome vom Grad n gibt es über 𝔽_p ungefähr?</summary>

Etwa $p^n/n$, genau $\frac1n\sum_{d\mid n}\mu(d)p^{n/d}$. Ein zufälliges Polynom vom Grad $n$ ist also mit Wahrscheinlichkeit etwa $1/n$ irreduzibel.
</details>

<details><summary>9. Wie baut man einen Körper mit 16 Elementen, und warum geht ℤ₁₆ nicht?</summary>

Als $\mathbb F_2[x]/(f)$ mit einem irreduziblen Polynom $f$ vom Grad 4, etwa $x^4 + x + 1$: Polynome vom Grad $< 4$ über $\mathbb F_2$, multipliziert modulo $f$. $\mathbb Z_{16}$ hat Nullteiler wie $4\cdot4 = 0$.
</details>

<details><summary>10. Warum ist die multiplikative Gruppe eines endlichen Körpers zyklisch, und was ist ein primitives Element?</summary>

Ein Polynom vom Grad $d$ hat in einem Körper höchstens $d$ Nullstellen; deshalb gibt es zu jedem Teiler $d$ von $q - 1$ höchstens $\varphi(d)$ Elemente der Ordnung $d$, und die Abzählung erzwingt Gleichheit, also auch Elemente der Ordnung $q - 1$. Ein primitives Element erzeugt die ganze Gruppe.
</details>

<details><summary>11. Wie viele Fehler korrigiert ein Reed-Solomon-Code mit n Symbolen und k Nutzsymbolen, und warum?</summary>

$\lfloor(n - k)/2\rfloor$. Zwei verschiedene Polynome vom Grad $< k$ stimmen in höchstens $k - 1$ Stellen überein, der Mindestabstand ist also $n - k + 1$.
</details>

<details><summary>12. Was leistet eine Gröbner-Basis?</summary>

Sie ist ein Erzeugendensystem eines Polynomideals, bei dem die Division mit Rest eindeutig ist. Mit lexikographischer Ordnung bringt sie ein Gleichungssystem in Dreiecksform: ein Polynom in einer Variablen, dann in zweien, und so weiter.
</details>

<details><summary>13. Warum lässt sich das regelmäßige Siebeneck nicht mit Zirkel und Lineal konstruieren, das Siebzehneck aber schon?</summary>

Konstruierbare Zahlen haben einen Grad $2^k$ über $\mathbb Q$. $\cos(2\pi/7)$ hat Grad 3, $\cos(2\pi/17)$ hat Grad 8 $= 2^3$; allgemein ist das $n$-Eck konstruierbar genau dann, wenn $\varphi(n)$ eine Zweierpotenz ist.
</details>

<details><summary>14. Was sagt der Dichtesatz von Chebotarev?</summary>

Wie ein Polynom modulo $p$ zerfällt, entspricht dem Zyklentyp eines Elements seiner Galois-Gruppe, und jeder Zyklentyp kommt bei einem Anteil der Primzahlen vor, der seinem Anteil in der Gruppe entspricht.
</details>

<details><summary>15. Warum gibt es für x⁵ − x − 1 keine Lösungsformel mit Wurzeln?</summary>

Seine Galois-Gruppe ist $S_5$, und $S_5$ ist nicht auflösbar (sie enthält die einfache Gruppe $A_5$). Durch Wurzeln lösbar sind genau die Gleichungen mit auflösbarer Galois-Gruppe.
</details>

<details><summary>16. Was sind p-adische Zahlen, und was besagt das Henselsche Lemma?</summary>

Zahlen, bei denen „klein“ bedeutet „durch eine hohe Potenz von $p$ teilbar“; die Vervollständigung von $\mathbb Q$ bezüglich dieses Betrags. Hensel: Eine einfache Nullstelle modulo $p$ lässt sich zu einer $p$-adischen Nullstelle hochheben; das Newton-Verfahren verdoppelt die Genauigkeit pro Schritt.
</details>

<details><summary>17. Was sagt das quadratische Reziprozitätsgesetz?</summary>

Für ungerade Primzahlen $p\neq q$ ist $(p/q)(q/p) = (-1)^{\frac{p-1}2\cdot\frac{q-1}2}$: Ob $p$ ein Quadrat modulo $q$ ist, hängt eng damit zusammen, ob $q$ eines modulo $p$ ist.
</details>

<details><summary>18. Welche endlichdimensionalen Divisionsalgebren über ℝ gibt es, und gibt es endliche Schiefkörper?</summary>

Nur $\mathbb R$, $\mathbb C$ und die Quaternionen $\mathbb H$ (Frobenius). Endliche Schiefkörper sind immer kommutativ, also Körper (Wedderburn).
</details>

<details><summary>19. Was berechnet die Smith-Normalform, und warum braucht man sie für Homologie?</summary>

Eine Diagonalform $d_1\mid d_2\mid\dots$ einer ganzzahligen Matrix durch ganzzahlige Umformungen; sie zerlegt die abelsche Gruppe $\mathbb Z^n/\mathrm{Bild}$ in zyklische Teile. Für die Homologie liefert sie neben den Betti-Zahlen auch die Torsion, etwa $\mathbb Z_2$ bei der projektiven Ebene, die über $\mathbb Q$ unsichtbar wäre.
</details>

<details><summary>20. Warum funktioniert die Entschlüsselung bei Ring-LWE, und was passiert bei zu viel Rauschen?</summary>

In $v - us$ heben sich die Terme mit $a$ weg, weil der Ring kommutativ ist; übrig bleiben kleines Rauschen und $\lceil q/2\rfloor m$. Solange jeder Rauschkoeffizient kleiner als $q/4$ ist, wird jedes Bit richtig gerundet; größeres Rauschen erzeugt Bitfehler.
</details>
