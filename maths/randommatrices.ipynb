# Zufallsmatrizen

Das Notebook zieht große Zufallsmatrizen aus den Gaußschen Ensembles GOE und GUE und prüft numerisch die zentralen Ergebnisse der Theorie: den Wigner-Halbkreis, die Abstoßung benachbarter Eigenwerte (Wigner-Vermutung gegen Poisson-Statistik, Verhältnis-Statistik ⟨r⟩), die Marchenko-Pastur-Verteilung der Eigenwerte von Rausch-Kovarianzmatrizen mit dem BBP-Übergang, ab dem ein Signal aus dem Rauschen heraustritt, und den Übergang einer Spinkette mit zufälligen Feldern von Quantenchaos (GOE) zu Lokalisierung (Poisson).

Was bleibt von einer Matrix übrig, wenn ihre Einträge zufällig sind, und warum zeigen die Energieniveaus chaotischer Quantensysteme genau diese Statistik?

#### **Hintergrund**

Eine **Zufallsmatrix** ist eine Matrix, deren Einträge Zufallszahlen sind. Erstaunlicherweise haben ihre Eigenwerte, wenn die Matrix groß ist, ganz bestimmte Eigenschaften, die kaum davon abhängen, wie die Einträge verteilt sind: Ihre Dichte ist ein Halbkreis, und benachbarte Eigenwerte stoßen sich ab. Diese **Universalität** macht Zufallsmatrizen zu einem Modell für alles, was zu kompliziert ist, um es im Detail zu kennen.

Wigner führte sie in den 1950er Jahren ein, um die Energieniveaus schwerer Atomkerne zu beschreiben, deren Hamilton-Operator niemand kennt. Dyson ordnete 1962 die Ensembles nach ihren Symmetrien, Marchenko und Pastur fanden 1967 die Eigenwertverteilung von Kovarianzmatrizen, und Bohigas, Giannoni und Schmit vermuteten 1984, dass die Niveaus jedes quantenmechanischen Systems mit chaotischem klassischem Gegenstück der Zufallsmatrix-Statistik folgen. Heute tauchen Zufallsmatrizen in der Quantenphysik (Quantenchaos, Vielteilchenlokalisierung, Scrambling), in der Zahlentheorie (die Nullstellen der Riemannschen Zeta-Funktion haben die Abstände des GUE), in der Statistik (Hauptkomponenten hochdimensionaler Daten), in der Finanzmathematik, in der Nachrichtentechnik (Kapazität von Mehrantennensystemen) und im maschinellen Lernen auf (Spektren von Gewichtsmatrizen und Hesse-Matrizen großer Netze).

Landkarte:

- **Eigenwertdichte**
  - Wigner-Halbkreis (Exp. 1): Die Eigenwerte füllen $[-2\sqrt N, 2\sqrt N]$ mit Halbkreisdichte, gleich wie die Einträge verteilt sind; die Momente sind Catalan-Zahlen.
- **Abstände**
  - Niveauabstoßung (Exp. 2): Abstände benachbarter Eigenwerte folgen $P(s)\propto s^\beta$ für kleine $s$; unabhängige Zahlen folgen $e^{-s}$.
  - Verhältnis-Statistik (Exp. 3): $\langle r\rangle = 0.386$ (Poisson), $0.531$ (GOE), $0.600$ (GUE), ohne Entfalten.
- **Daten**
  - Marchenko–Pastur und BBP-Übergang (Exp. 4): Reines Rauschen streut die Eigenwerte einer geschätzten Kovarianz über einen ganzen Bereich; ein Signal tritt erst ab einer Schwelle heraus.
- **Physik**
  - Quantenchaos und Lokalisierung (Exp. 5): Eine Spinkette mit schwacher Unordnung hat GOE-Statistik, mit starker Unordnung Poisson-Statistik.

#### **Theorie**

##### *1. Gaußsche Ensembles*

Die wichtigsten Zufallsmatrizen sind hermitesch mit unabhängigen Gauß-verteilten Einträgen:
- **GOE** (Gaussian Orthogonal Ensemble): reell symmetrisch, Index $\beta = 1$; passend für Systeme mit Zeitumkehrsymmetrie.
- **GUE** (Gaussian Unitary Ensemble): komplex hermitesch, $\beta = 2$; passend ohne Zeitumkehrsymmetrie, etwa mit Magnetfeld.

Die Verteilung $P(H)\propto e^{-\mathrm{tr}\,H^2/4}$ (Varianz 1 außerhalb der Diagonale) ist invariant unter orthogonalen bzw. unitären Basiswechseln: Keine Richtung ist ausgezeichnet, alle Information steckt in den Eigenwerten. Schreibt man $H = O\Lambda O^T$, bringt der Variablenwechsel die Jacobi-Determinante $\prod_{i<j}\vert\lambda_i - \lambda_j\vert^\beta$ mit, und die gemeinsame Dichte der Eigenwerte ist
$$p(\lambda_1, \dots, \lambda_N)\propto\prod_{i<j}\vert\lambda_i - \lambda_j\vert^\beta\,e^{-\sum_i\lambda_i^2/4} .$$
Der Faktor $\vert\lambda_i - \lambda_j\vert^\beta$ macht gleiche Eigenwerte unwahrscheinlich: Um zwei Eigenwerte zusammenfallen zu lassen, müssen $\beta + 1$ unabhängige Bedingungen zugleich erfüllt sein, nicht nur eine.

##### *2. Der Wigner-Halbkreis*

Mit Varianz 1 außerhalb der Diagonale wachsen die Eigenwerte wie $\sqrt N$. Für $x = \lambda/\sqrt N$ ist die Dichte für großes $N$ der **Halbkreis**
$$\rho(x) = \frac1{2\pi}\sqrt{4 - x^2},\qquad\vert x\vert\leq2 .$$
Begründung über die **Momente** $\frac1N\mathbb E\,\mathrm{tr}(H/\sqrt N)^k$: $\mathrm{tr}\,H^{2k}$ ist eine Summe über geschlossene Pfade der Länge $2k$ durch die Indizes. Weil die Einträge den Mittelwert null haben, zählen nur Pfade, die jede Kante mindestens zweimal benutzen, und für großes $N$ dominieren die, die einen Baum bilden. Ihre Anzahl ist die **Catalan-Zahl** $C_k = \frac1{k+1}\binom{2k}k$ ($1, 2, 5, 14, \dots$), und genau das sind die Momente des Halbkreises; die ungeraden Momente verschwinden. Weil nur die Varianz der Einträge eingeht, gilt das für fast jede Verteilung mit endlicher Varianz (**Universalität**).

##### *3. Abstände und Niveauabstoßung*

- **Wigner-Vermutung:** Für eine $2\times2$-Matrix $\begin{pmatrix}a & c\\ c & b\end{pmatrix}$ ist der Abstand der Eigenwerte $s = \sqrt{(a - b)^2 + 4c^2}$, der Betrag eines zweidimensionalen Gauß-Vektors. Das Flächenelement in Polarkoordinaten bringt einen Faktor $s$; normiert auf mittleren Abstand 1:
$$P_\text{GOE}(s) = \frac\pi2s\,e^{-\pi s^2/4},\qquad P_\text{GUE}(s) = \frac{32}{\pi^2}s^2e^{-4s^2/\pi} .$$
Beim GUE ist $c$ komplex, der Vektor hat drei Komponenten, der Faktor wird $s^2$. Diese Formeln beschreiben auch große Matrizen bis auf wenige Prozent.
- **Poisson-Statistik:** Unabhängige, gleichmäßig verteilte Niveaus mit mittlerem Abstand 1 haben $P(s) = e^{-s}$, am häufigsten sind kleine Abstände.
- **Entfalten:** Weil die Dichte nicht konstant ist, ersetzt man jeden Eigenwert durch die erwartete Zahl der Eigenwerte unter ihm, $N\int_{-\infty}^\lambda\rho$; dann ist der mittlere Abstand überall 1.
- **Verhältnis-Statistik:** Ohne Kenntnis der Dichte nimmt man $r_i = \min(s_i, s_{i+1})/\max(s_i, s_{i+1})$, bei dem sich die lokale Dichte herauskürzt. $\langle r\rangle = 2\ln2 - 1\approx0.386$ (Poisson), $\approx0.531$ (GOE), $\approx0.600$ (GUE) (Atas et al. 2013).

##### *4. Kovarianzmatrizen: Marchenko–Pastur und der BBP-Übergang*

Schätzt man aus $n$ Messungen von $p$ Größen die Kovarianzmatrix $C = X^TX/n$, streuen ihre Eigenwerte auch dann, wenn die Größen in Wahrheit unkorreliert mit Varianz 1 sind. Für $n, p\to\infty$ mit $\gamma = p/n\leq1$ verteilen sie sich nach der **Marchenko–Pastur-Verteilung** über
$$\lambda_\pm = (1\pm\sqrt\gamma)^2,\qquad\rho(\lambda) = \frac{\sqrt{(\lambda_+ - \lambda)(\lambda - \lambda_-)}}{2\pi\gamma\lambda} .$$
Hat die wahre Kovarianz eine Richtung mit Varianz $1 + \ell$ (ein **Signal**, ein „Spike“), löst sich der größte Eigenwert nur ab, wenn $\ell>\sqrt\gamma$, und liegt dann bei $(1 + \ell)(1 + \gamma/\ell)$, über der wahren Varianz. Schwächere Signale verschwinden im Rauschen am Rand $\lambda_+$, gleich wie man rechnet (**BBP-Übergang**, Baik, Ben Arous und Péché 2005). Eigenwerte außerhalb des Rauschbereichs zeigen echte Korrelationen an; das nutzt man bei der **Hauptkomponentenanalyse**, in der Finanzmathematik und in der Genetik.

##### *5. Quantenchaos und Lokalisierung*

Die Energieniveaus eines Quantensystems, dessen klassisches Gegenstück chaotisch ist, folgen der Abstandsstatistik von GOE oder GUE (**Vermutung von Bohigas, Giannoni und Schmit** 1984); die Niveaus integrabler Systeme folgen meist der Poisson-Statistik (**Berry und Tabor**). Die Niveaustatistik ist deshalb ein Fingerabdruck von Chaos in Systemen ohne Trajektorien. Dabei darf man nur Niveaus mit denselben Quantenzahlen vergleichen: Niveaus aus verschiedenen **Symmetriesektoren** wissen nichts voneinander und täuschen Poisson-Statistik vor. Ein Beispiel ist die Heisenberg-Kette mit zufälligen Feldern, $H = \sum_i\vec S_i\cdot\vec S_{i+1} + \sum_ih_iS_i^z$ mit $h_i$ gleichverteilt in $[-W, W]$. Bei schwacher Unordnung ist sie chaotisch (GOE), bei starker **vielteilchenlokalisiert**: Die Spins bleiben in der Nähe ihres Anfangszustands, und die Niveaus verhalten sich wie unabhängige Zahlen. Für lange Ketten liegt der Übergang bei etwa $W\approx3.7$ (Luitz, Laflorencie und Alet 2015); ob er im Grenzfall unendlich langer Ketten bestehen bleibt, ist umstritten.

#### **Code**

Funktionen zum Ziehen der Ensembles, zum Entfalten mit dem Halbkreis und für die Verhältnis-Statistik.

```python
# Alles hier (numpy, scipy, matplotlib) ist in Colab vorinstalliert
import itertools
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)

def goe(N):
    A = rng.normal(size=(N, N))
    return (A + A.T) / np.sqrt(2)                     # Varianz 1 außerhalb, 2 auf der Diagonale

def gue(N):
    A = (rng.normal(size=(N, N)) + 1j * rng.normal(size=(N, N))) / np.sqrt(2)
    return (A + A.conj().T) / np.sqrt(2)              # E|H_ij|² = 1 außerhalb der Diagonale

semicircle = lambda x: np.sqrt(np.clip(4 - x ** 2, 0, None)) / (2 * np.pi)
semicircle_cdf = lambda x: 0.5 + (x * np.sqrt(np.clip(4 - x ** 2, 0, None)) / (4 * np.pi)) + np.arcsin(np.clip(x / 2, -1, 1)) / np.pi

def unfolded_spacings(ev, N, keep=0.5):
    """Entfalten mit dem Halbkreis, Abstände aus der mittleren Hälfte des Spektrums."""
    u = N * semicircle_cdf(np.sort(ev) / np.sqrt(N))
    s = np.diff(u)
    lo, hi = int(len(s) * (1 - keep) / 2), int(len(s) * (1 + keep) / 2)
    return s[lo:hi]

def ratios(ev):
    s = np.diff(np.sort(ev))
    return np.minimum(s[:-1], s[1:]) / np.maximum(s[:-1], s[1:])

wigner_goe = lambda s: np.pi / 2 * s * np.exp(-np.pi * s ** 2 / 4)
wigner_gue = lambda s: 32 / np.pi ** 2 * s ** 2 * np.exp(-4 * s ** 2 / np.pi)
poisson = lambda s: np.exp(-s)
print("Funktionen bereit")
```

#### **Experimente**

##### *1. Der Wigner-Halbkreis*

Folgen die Eigenwerte einer großen GOE- und GUE-Matrix dem Halbkreis, und hängt das von der Verteilung der Einträge ab? Vorher vorhersagen: Welchen Wert hat $\langle x^4\rangle$, das vierte Moment der skalierten Eigenwerte?

Je eine Matrix der Größe $N = 1500$ wird aus GOE und GUE gezogen; als dritter Fall dienen symmetrische Matrizen mit Einträgen $\pm1$ (gleich wahrscheinlich), also nicht Gauß-verteilt, aber mit derselben Varianz. Die Eigenwerte werden durch $\sqrt N$ geteilt und als Histogramm mit dem Halbkreis verglichen; dazu kommen die Momente 2, 4 und 6.

**Hintergrund & Fachbegriffe:**
- **Eigenwertdichte:** wie dicht die Eigenwerte an jeder Stelle liegen, als Histogramm über alle $N$ Eigenwerte einer Matrix.
- **Skalierung:** Mit Varianz 1 pro Eintrag liegen die Eigenwerte in $[-2\sqrt N, 2\sqrt N]$; geteilt durch $\sqrt N$ wird daraus $[-2, 2]$.
- **Momente und Catalan-Zahlen:** Das $k$-te Moment ist der Mittelwert von $x^k$; beim Halbkreis sind die geraden Momente die Catalan-Zahlen $1, 2, 5, 14, \dots$
- **Universalität:** Ein Ergebnis hängt nicht von den Details ab, hier nicht von der Verteilung der Einträge, solange ihre Varianz endlich ist.

```python
N = 1500
ev_goe = np.linalg.eigvalsh(goe(N))
ev_gue = np.linalg.eigvalsh(gue(N))
B = rng.choice([-1.0, 1.0], size=(N, N))
ev_pm = np.linalg.eigvalsh(np.triu(B) + np.triu(B, 1).T)

fig, ax = plt.subplots(figsize=(9, 4.5))
bins = np.linspace(-2.2, 2.2, 60)
for ev, name, c in [(ev_goe, "GOE", "tab:blue"), (ev_gue, "GUE", "tab:red"), (ev_pm, "Einträge ±1", "tab:green")]:
    ax.hist(ev / np.sqrt(N), bins=bins, density=True, histtype="step", lw=1.5, color=c, label=name)
xs = np.linspace(-2.2, 2.2, 400)
ax.plot(xs, semicircle(xs), c="black", lw=2, ls="--", label="Halbkreis √(4 − x²)/2π")
ax.set_xlabel("Eigenwert / √N")
ax.set_ylabel("Dichte")
ax.set_title(f"Eigenwertdichte, N = {N}")
ax.legend()
fig.tight_layout()
plt.show()
for ev, name in [(ev_goe, "GOE"), (ev_gue, "GUE"), (ev_pm, "±1")]:
    m = [np.mean((ev / np.sqrt(N)) ** k) for k in (2, 4, 6)]
    print(f"{name:4s}: Momente ⟨x²⟩, ⟨x⁴⟩, ⟨x⁶⟩ = {m[0]:.3f}, {m[1]:.3f}, {m[2]:.3f}   (Catalan-Zahlen: 1, 2, 5)")
```

* Das **Bild** zeigt die drei Histogramme der skalierten Eigenwerte über dem gestrichelten Halbkreis $\sqrt{4 - x^2}/2\pi$. GOE (blau), GUE (rot) und die Matrix mit Einträgen $\pm1$ (grün) liegen alle darauf, mit den kleinen Schwankungen eines Histogramms aus 1500 Werten. Die Ränder bei $\pm2$ sind scharf: Außerhalb liegt praktisch kein Eigenwert.
* Die **Ausgabe** gibt die Momente $\langle x^2\rangle, \langle x^4\rangle, \langle x^6\rangle$: $0.999, 2.001, 5.012$ (GOE), $1.000, 2.003, 5.013$ (GUE) und $1.000, 1.999, 4.995$ ($\pm1$), also die Catalan-Zahlen 1, 2, 5. Dass die Einträge nur $\pm1$ sind, spielt keine Rolle; es kommt nur auf ihre Varianz an.

##### *2. Abstoßung der Eigenwerte*

Wie sind die Abstände benachbarter Eigenwerte verteilt, und wie unterscheiden sie sich von unabhängigen Zufallszahlen? Vorher vorhersagen: Welcher Anteil der Abstände ist kleiner als 0.1, für unabhängige Zahlen und für das GOE?

Für GOE und GUE werden 40 Matrizen der Größe 400 gezogen und mit dem Halbkreis entfaltet (Abschnitt 3 der Theorie); verwendet wird die mittlere Hälfte des Spektrums. Zum Vergleich dienen unabhängige, gleichverteilte Zahlen mit mittlerem Abstand 1.

**Hintergrund & Fachbegriffe:**
- **Niveauabstand:** der Abstand $s$ zweier aufeinanderfolgender Eigenwerte, gemessen in Einheiten des mittleren Abstands.
- **Entfalten:** Umrechnen der Eigenwerte, sodass der mittlere Abstand überall 1 ist, obwohl die Dichte nicht konstant ist.
- **Wigner-Vermutung (Wigner surmise):** die Abstandsverteilung von $2\times2$-Matrizen, $\propto s\,e^{-\pi s^2/4}$ (GOE) und $\propto s^2e^{-4s^2/\pi}$ (GUE); sie beschreibt auch große Matrizen gut.
- **Niveauabstoßung:** $P(s)\propto s^\beta$ für kleine $s$; beim GOE linear, beim GUE quadratisch. Unabhängige Zahlen haben dagegen $P(s) = e^{-s}$, am häufigsten nahe 0 (**Poisson-Statistik**).

```python
N, REP = 400, 40
s_goe = np.concatenate([unfolded_spacings(np.linalg.eigvalsh(goe(N)), N) for _ in range(REP)])
s_gue = np.concatenate([unfolded_spacings(np.linalg.eigvalsh(gue(N)), N) for _ in range(REP)])
s_poi = np.diff(np.sort(rng.uniform(0, 20000, 20000)))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
bins = np.linspace(0, 4, 41)
ss = np.linspace(0, 4, 300)
for s, name, c in [(s_poi, "unabhängige Zahlen", "tab:green"), (s_goe, "GOE", "tab:blue"), (s_gue, "GUE", "tab:red")]:
    axes[0].hist(s / s.mean(), bins=bins, density=True, alpha=0.35, color=c, label=name)
axes[0].plot(ss, poisson(ss), c="tab:green", lw=2, label="Poisson e^(−s)")
axes[0].plot(ss, wigner_goe(ss), c="tab:blue", lw=2, label="Wigner GOE (π/2) s e^(−πs²/4)")
axes[0].plot(ss, wigner_gue(ss), c="tab:red", lw=2, label="Wigner GUE (32/π²) s² e^(−4s²/π)")
axes[0].set_xlabel("Abstand s (mittlerer Abstand 1)")
axes[0].set_ylabel("Dichte P(s)")
axes[0].set_title("Abstandsverteilung")
axes[0].legend(fontsize=8)
small = np.logspace(-2, 0, 12)
for s, name, c, beta in [(s_poi, "unabhängig", "tab:green", 0), (s_goe, "GOE", "tab:blue", 1), (s_gue, "GUE", "tab:red", 2)]:
    cum = [np.mean(s / s.mean() < x) for x in small]
    axes[1].loglog(small, cum, "o-", c=c, label=f"{name}: Anteil ~ s^{beta + 1}")
axes[1].set_xlabel("Abstand s")
axes[1].set_ylabel("Anteil der Abstände kleiner als s")
axes[1].set_title("Kleine Abstände: P(s) ~ s^β")
axes[1].legend(fontsize=8)
fig.tight_layout()
plt.show()
for s, name in [(s_poi, "unabhängig"), (s_goe, "GOE"), (s_gue, "GUE")]:
    print(f"{name:10s}: Anteil der Abstände < 0.1: {np.mean(s / s.mean() < 0.1):.4f}")
```

* Das **linke Bild** zeigt die Abstandsverteilungen. Die unabhängigen Zahlen (grün) folgen $e^{-s}$: Kleine Abstände sind am häufigsten. GOE (blau) und GUE (rot) gehen bei $s = 0$ auf null und haben ihr Maximum bei $s\approx0.8$ bis 0.9, wie die Wigner-Formeln; das GUE ist schmaler, die Abstoßung stärker.
* Das **rechte Bild** zeigt den Anteil der Abstände kleiner als $s$, doppelt logarithmisch. Die drei Kurven haben die Steigungen 1, 2 und 3: Aus $P(s)\propto s^\beta$ wird beim Aufsummieren $s^{\beta + 1}$. Ganz links wird die blaue Kurve flach, weil dort nur noch einzelne Abstände liegen und die Statistik nicht mehr reicht.
* Die **Ausgabe** gibt den Anteil der Abstände unter 0.1: $9.4\,\%$ für unabhängige Zahlen, $1.0\,\%$ für das GOE und $0.09\,\%$ für das GUE. Zwei fast gleiche Eigenwerte sind bei Zufallsmatrizen also selten, bei unabhängigen Zahlen häufig.

##### *3. Die Verhältnis-Statistik*

Welchen Mittelwert hat $r$ für Poisson, GOE und GUE, wenn man nicht entfaltet? Vorher vorhersagen: Liegt das GOE näher am Poisson- oder am GUE-Wert?

Je 150 Matrizen der Größe 400, aus der mittleren Hälfte des Spektrums. Den Fehler schätzt die Streuung der Mittelwerte zwischen den Matrizen, nicht die Streuung einzelner Verhältnisse: Benachbarte Verhältnisse teilen sich einen Abstand und sind deshalb korreliert.

**Hintergrund & Fachbegriffe:**
- **Verhältnis aufeinanderfolgender Abstände:** $r_i = \min(s_i, s_{i+1})/\max(s_i, s_{i+1})$ liegt zwischen 0 und 1. Die lokale Dichte kürzt sich heraus, deshalb braucht man kein Entfalten.
- **Korrelierte Daten und Standardfehler:** Sind Werte nicht unabhängig, unterschätzt die Formel Streuung$/\sqrt n$ die Unsicherheit; man nimmt dann unabhängige Blöcke, hier ganze Matrizen.

```python
N_MAT = 150
samples = {"Poisson": [ratios(rng.uniform(0, 1, 200)) for _ in range(N_MAT)],
           "GOE": [ratios(np.linalg.eigvalsh(goe(400))[100:300]) for _ in range(N_MAT)],
           "GUE": [ratios(np.linalg.eigvalsh(gue(400))[100:300]) for _ in range(N_MAT)]}
ref = {"Poisson": 2 * np.log(2) - 1, "GOE": 0.5307, "GUE": 0.5996}
for name, per_matrix in samples.items():
    means = np.array([r.mean() for r in per_matrix])
    # Fehler aus der Streuung zwischen den Matrizen: benachbarte Verhältnisse derselben Matrix sind korreliert
    print(f"{name:8s}: ⟨r⟩ = {means.mean():.4f} ± {means.std() / np.sqrt(N_MAT):.4f}   (Theorie {ref[name]:.4f})")
```

* Die **Ausgabe** gibt $\langle r\rangle = 0.3842\pm0.0021$ (Poisson), $0.5314\pm0.0017$ (GOE) und $0.5992\pm0.0017$ (GUE); die Theorie sagt $0.3863$, $0.5307$, $0.5996$. Alle drei liegen innerhalb ihrer Unsicherheit.
* **Deutung:** Ein einziger Mittelwert trennt die drei Fälle deutlich, ohne dass man die Dichte kennen oder entfalten muss. Deshalb ist $\langle r\rangle$ das Standardmaß, um in einem Spektrum unbekannter Herkunft Chaos (GOE) von Integrabilität oder Lokalisierung (Poisson) zu unterscheiden, wie in Experiment 5.

##### *4. Rauschen in Kovarianzmatrizen: Marchenko–Pastur und ein verstecktes Signal*

Wie verteilen sich die Eigenwerte einer Kovarianzmatrix aus reinem Rauschen, und ab welcher Stärke tritt ein echtes Signal aus diesem Rauschen heraus? Vorher vorhersagen: Wie groß ist der größte geschätzte Eigenwert, wenn alle wahren Varianzen 1 sind?

$n = 2000$ Messungen von $p = 500$ Größen, also $\gamma = 0.25$ und $\sqrt\gamma = 0.5$. Für das Signal bekommt eine zufällige Richtung die Varianz $1 + \ell$, mit $\ell$ zwischen 0.05 und 2; der größte Eigenwert wird über drei Ziehungen gemittelt.

**Hintergrund & Fachbegriffe:**
- **Stichproben-Kovarianzmatrix:** $C = X^TX/n$ aus $n$ Messungen (Zeilen) von $p$ Größen (Spalten). Ihre Eigenvektoren sind die **Hauptkomponenten**, ihre Eigenwerte die Varianzen entlang dieser Richtungen.
- **Marchenko–Pastur-Verteilung:** die Verteilung der Eigenwerte, wenn in Wahrheit alle Varianzen 1 und alle Größen unkorreliert sind; sie reicht von $(1 - \sqrt\gamma)^2$ bis $(1 + \sqrt\gamma)^2$ mit $\gamma = p/n$.
- **Spike-Modell und BBP-Übergang:** Ein einzelnes echtes Signal der Stärke $\ell$ ist im größten Eigenwert nur sichtbar, wenn $\ell>\sqrt\gamma$; darunter verschwindet es im Rauschen.

```python
n, p = 2000, 500
gam = p / n
lam_m, lam_p = (1 - np.sqrt(gam)) ** 2, (1 + np.sqrt(gam)) ** 2
mp = lambda x: np.sqrt(np.clip((lam_p - x) * (x - lam_m), 0, None)) / (2 * np.pi * gam * x)

X = rng.normal(size=(n, p))
ev_noise = np.linalg.eigvalsh(X.T @ X / n)

def top_eigenvalue(ell):
    v = rng.normal(size=p); v /= np.linalg.norm(v)
    Y = rng.normal(size=(n, p)) + np.sqrt(ell) * rng.normal(size=(n, 1)) * v      # Kovarianz I + ℓ v vᵀ
    return np.linalg.eigvalsh(Y.T @ Y / n)[-1]

ells = np.linspace(0.05, 2.0, 25)
tops = np.array([np.mean([top_eigenvalue(l) for _ in range(3)]) for l in ells])
theory = np.where(ells > np.sqrt(gam), (1 + ells) * (1 + gam / ells), lam_p)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
axes[0].hist(ev_noise, bins=60, density=True, alpha=0.5, color="tab:blue", label=f"Eigenwerte, reines Rauschen (p = {p}, n = {n})")
xs = np.linspace(lam_m, lam_p, 400)
axes[0].plot(xs, mp(xs), c="black", lw=2, label="Marchenko-Pastur, γ = 0.25")
axes[0].axvline(1, color="tab:red", ls=":", label="wahre Varianz: 1")
axes[0].set_xlabel("Eigenwert der geschätzten Kovarianz")
axes[0].set_ylabel("Dichte")
axes[0].set_title("Rauschen allein streut die Eigenwerte von 0.25 bis 2.25")
axes[0].legend(fontsize=8)
axes[1].plot(ells, tops, "o", label="größter Eigenwert (Mittel aus 3 Ziehungen)")
axes[1].plot(ells, theory, c="black", label="Theorie: Rand λ₊ oder (1 + ℓ)(1 + γ/ℓ)")
axes[1].plot(ells, 1 + ells, c="grey", ls=":", label="wahre Varianz 1 + ℓ")
axes[1].axvline(np.sqrt(gam), color="tab:red", ls="--", label="Schwelle ℓ = √γ")
axes[1].set_xlabel("Stärke des Signals ℓ")
axes[1].set_ylabel("größter Eigenwert")
axes[1].set_title("BBP-Übergang")
axes[1].legend(fontsize=8)
fig.tight_layout()
plt.show()
print(f"Rauschen: kleinster Eigenwert {ev_noise.min():.3f} (Theorie {lam_m:.3f}), größter {ev_noise.max():.3f} (Theorie {lam_p:.3f})")
```

* Das **linke Bild** zeigt die Eigenwerte der geschätzten Kovarianz aus reinem Rauschen. Obwohl alle wahren Varianzen 1 sind (rote Linie), streuen sie von 0.25 bis 2.25, in der schiefen Form der Marchenko–Pastur-Kurve (schwarz) mit dem Maximum bei etwa 0.45. Wer naiv die größten Eigenwerte als „wichtige Richtungen“ deutet, findet hier Strukturen, die es nicht gibt.
* Das **rechte Bild** zeigt den größten Eigenwert gegen die Signalstärke $\ell$. Unterhalb der Schwelle $\ell = 0.5$ (rot gestrichelt) bleibt er am Rauschrand 2.25 hängen: Das Signal ist unsichtbar. Darüber folgt er der Formel $(1 + \ell)(1 + \gamma/\ell)$ (schwarz) und liegt immer über der wahren Varianz $1 + \ell$ (grau gepunktet).
* Die **Ausgabe** gibt den kleinsten Eigenwert des Rauschens $0.255$ (Theorie $0.250$) und den größten $2.211$ (Theorie $2.250$). Mit viermal so vielen Messungen wie Größen überschätzt die Stichprobe die größte Varianz also um mehr als das Doppelte. Dass der größte Wert etwas innerhalb des Randes liegt, ist typisch für endliche Matrizen; die Schwankungen am Rand werden erst für $n, p\to\infty$ null.

##### *5. Quantenchaos und Lokalisierung in einer Spinkette*

Zeigt eine Kette wechselwirkender Spins die Niveaustatistik von GOE, und was passiert, wenn starke zufällige Felder die Spins festhalten? Vorher vorhersagen: Welchen Wert hat $\langle r\rangle$ bei sehr starker Unordnung?

Das Modell ist die Heisenberg-Kette mit zufälligen Feldern, $H = \sum_i\vec S_i\cdot\vec S_{i+1} + \sum_ih_iS^z_i$, mit $L = 12$ Spins, $h_i$ gleichverteilt in $[-W, W]$ und offenem Rand. Die Gesamtmagnetisierung ist erhalten; gerechnet wird im Sektor $\sum_iS^z_i = 0$ mit 924 Zuständen, denn Niveaus aus verschiedenen Symmetriesektoren sind unabhängig und würden eine Poisson-Statistik vortäuschen. Für neun Werte von $W$ zwischen 0.25 und 12 wird $\langle r\rangle$ aus der mittleren Hälfte des Spektrums bestimmt.

**Hintergrund & Fachbegriffe:**
- **Heisenberg-Kette:** Spins ½ auf einer Linie, die mit ihren Nachbarn über $\vec S_i\cdot\vec S_{i+1}$ wechselwirken; ein Grundmodell des Magnetismus.
- **Unordnung:** zufällige Felder $h_i$, die an jedem Platz anders sind, wie Verunreinigungen in einem Material; $W$ gibt ihre Stärke an.
- **Erhaltungsgröße und Symmetriesektor:** Weil $H$ die Gesamtmagnetisierung nicht ändert, zerfällt die Matrix in Blöcke, einen pro Wert von $\sum S^z$. Die Statistik darf man nur innerhalb eines Blocks berechnen.
- **Vielteilchenlokalisierung:** Bei starker Unordnung erreicht ein wechselwirkendes System kein thermisches Gleichgewicht; Information über den Anfangszustand bleibt lokal erhalten, die Niveaus sind unkorreliert.

```python
L = 12
basis = [s for s in itertools.product([0, 1], repeat=L) if sum(s) == L // 2]     # 1 = Spin hoch
index = {s: i for i, s in enumerate(basis)}
D = len(basis)

def heisenberg_random_field(W):
    h = rng.uniform(-W, W, L)
    H = np.zeros((D, D))
    for k, s in enumerate(basis):
        sz = np.array(s) - 0.5
        H[k, k] = np.sum(sz[:-1] * sz[1:]) + np.sum(h * sz)
        for i in range(L - 1):
            if s[i] != s[i + 1]:                                      # S⁺S⁻ + S⁻S⁺ tauscht antiparallele Nachbarn
                t_ = list(s); t_[i], t_[i + 1] = t_[i + 1], t_[i]
                H[index[tuple(t_)], k] = 0.5
    return H

Ws = [0.25, 0.5, 1, 2, 3, 4, 6, 8, 12]
r_mean = []
for W in Ws:
    rs_ = [ratios(np.linalg.eigvalsh(heisenberg_random_field(W))[D // 4: 3 * D // 4]) for _ in range(15)]
    r_mean.append(np.mean(np.concatenate(rs_)))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(Ws, r_mean, "o-", c="black", label=f"Heisenberg-Kette, L = {L}, Sektor Sᶻ = 0 ({D} Zustände)")
ax.axhline(0.5307, color="tab:blue", ls="--", label="GOE 0.531")
ax.axhline(2 * np.log(2) - 1, color="tab:green", ls="--", label="Poisson 0.386")
ax.set_xscale("log")
ax.set_xlabel("Stärke der Unordnung W")
ax.set_ylabel("⟨r⟩ (mittlere Hälfte des Spektrums)")
ax.set_title("Von Quantenchaos zu Lokalisierung")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
for W, r in zip(Ws, r_mean):
    print(f"W = {W:5.2f}:  ⟨r⟩ = {r:.3f}")
```

* Das **Bild** zeigt $\langle r\rangle$ gegen die Unordnung $W$ (logarithmische Achse). Bis $W = 1$ liegen die Punkte bei der GOE-Linie 0.531: Die Kette ist chaotisch, ihre Niveaus stoßen sich ab. Zwischen $W = 1$ und 3 fällt $\langle r\rangle$ ab, und ab $W\approx3$ liegt es nahe der Poisson-Linie 0.386: Die Niveaus sind unabhängig, die Kette ist lokalisiert.
* Die **Ausgabe** gibt die Werte: $0.533$, $0.529$, $0.532$ für $W\leq1$, dann $0.446$ bei $W = 2$ und $0.383$ bis $0.402$ für $W\geq3$. Für diese kurze offene Kette liegt der Übergang um $W\approx2$; Rechnungen mit längeren, geschlossenen Ketten finden ihn bei etwa $W\approx3.7$, weil kurze Ketten leichter lokalisieren.

#### **Ergebnis**

Beim Lauf vom 2026-10-12:

- **Halbkreis:** GOE, GUE und Matrizen mit Einträgen $\pm1$ haben dieselbe Eigenwertdichte; ihre Momente $\langle x^2\rangle, \langle x^4\rangle, \langle x^6\rangle$ sind $1, 2, 5$, die Catalan-Zahlen. Die Verteilung der Einträge spielt keine Rolle, nur ihre Varianz (Universalität).
- **Abstoßung:** Die Abstände folgen den Wigner-Formeln. Abstände unter 0.1 haben 9.4 % der unabhängigen Zahlen, aber nur 1.0 % beim GOE und 0.09 % beim GUE; der Anteil wächst wie $s^{\beta+1}$.
- **Verhältnis-Statistik:** $\langle r\rangle = 0.384 \pm 0.002$ (Poisson), $0.531 \pm 0.002$ (GOE), $0.599 \pm 0.002$ (GUE), wie vorhergesagt (0.386, 0.531, 0.600).
- **Rauschen in Kovarianzen:** Obwohl alle wahren Varianzen 1 sind, liegen die geschätzten Eigenwerte zwischen 0.255 und 2.211 (Theorie 0.25 bis 2.25). Mit viermal so vielen Messungen wie Größen überschätzt die Stichprobe die größte Varianz also um mehr als das Doppelte.
- **BBP-Übergang:** Ein Signal mit $\ell < \sqrt\gamma = 0.5$ ist im größten Eigenwert nicht zu sehen; darüber folgt dieser der Formel $(1 + \ell)(1 + \gamma/\ell)$ und liegt stets über der wahren Varianz $1 + \ell$.
- **Spinkette:** Bei schwacher Unordnung ($W \le 1$) ist $\langle r\rangle = 0.53$: Die Kette ist chaotisch, und ihre Niveaus verhalten sich wie die Eigenwerte einer GOE-Matrix. Ab $W \approx 3$ ist $\langle r\rangle \approx 0.39$: Die Niveaus sind unabhängig, die Kette ist lokalisiert. Für diese kurze offene Kette ($L = 12$) liegt der Übergang um $W \approx 2$; Rechnungen mit längeren, geschlossenen Ketten finden ihn bei etwa $W \approx 3.7$ (Luitz, Laflorencie & Alet 2015). Ob er für unendlich lange Ketten bestehen bleibt, ist umstritten.

#### **Wiederholung**

Erst selbst antworten, dann aufklappen.

<details><summary>1. Was unterscheidet GOE und GUE, und wann passt welches Ensemble?</summary>

GOE: reell symmetrisch ($\beta = 1$), für Systeme mit Zeitumkehrsymmetrie. GUE: komplex hermitesch ($\beta = 2$), ohne Zeitumkehrsymmetrie, etwa mit Magnetfeld.
</details>

<details><summary>2. Woher kommt die Abstoßung der Eigenwerte?</summary>

Aus der Jacobi-Determinante beim Diagonalisieren: Die gemeinsame Dichte enthält $\prod_{i<j}\vert\lambda_i - \lambda_j\vert^\beta$. Gleiche Eigenwerte verlangen $\beta + 1$ Bedingungen zugleich und sind deshalb unwahrscheinlich.
</details>

<details><summary>3. Welche Dichte haben die Eigenwerte einer großen Zufallsmatrix, und warum hängt sie nicht von der Verteilung der Einträge ab?</summary>

Den Halbkreis $\sqrt{4 - x^2}/2\pi$ für $x = \lambda/\sqrt N$. Seine Momente sind die Catalan-Zahlen, die aus dem Zählen baumartiger Pfade kommen; dabei geht nur die Varianz der Einträge ein.
</details>

<details><summary>4. Wie sind die Abstände unabhängiger Zahlen verteilt, wie die von Zufallsmatrizen?</summary>

Unabhängig: $P(s) = e^{-s}$, kleine Abstände am häufigsten. GOE: $\frac\pi2s\,e^{-\pi s^2/4}$, GUE: $\frac{32}{\pi^2}s^2e^{-4s^2/\pi}$; beide verschwinden bei $s = 0$ wie $s^\beta$.
</details>

<details><summary>5. Was heißt Entfalten, und warum braucht die Verhältnis-Statistik es nicht?</summary>

Entfalten rechnet die Eigenwerte so um, dass der mittlere Abstand überall 1 ist. Im Verhältnis zweier benachbarter Abstände kürzt sich die lokale Dichte heraus.
</details>

<details><summary>6. Welche Werte hat ⟨r⟩ für Poisson, GOE und GUE?</summary>

$2\ln2 - 1\approx0.386$, $0.531$ und $0.600$.
</details>

<details><summary>7. Warum streuen die Eigenwerte einer geschätzten Kovarianzmatrix, wenn alle wahren Varianzen 1 sind?</summary>

Weil jede Stichprobe zufällig ist und in $p$ Dimensionen mit nur $n$ Messungen viele Richtungen zufällig etwas mehr oder weniger Varianz zeigen. Die Eigenwerte füllen den Marchenko–Pastur-Bereich $[(1 - \sqrt\gamma)^2, (1 + \sqrt\gamma)^2]$, $\gamma = p/n$.
</details>

<details><summary>8. Ab welcher Stärke ist ein Signal in den Eigenwerten sichtbar, und wo liegt dann der größte Eigenwert?</summary>

Ab $\ell>\sqrt\gamma$ (BBP-Übergang); dann bei $(1 + \ell)(1 + \gamma/\ell)$, also über der wahren Varianz $1 + \ell$. Darunter bleibt er am Rauschrand.
</details>

<details><summary>9. Was besagt die Vermutung von Bohigas, Giannoni und Schmit?</summary>

Die Energieniveaus eines Quantensystems, dessen klassisches Gegenstück chaotisch ist, haben die Abstandsstatistik von GOE oder GUE; integrable Systeme haben meist Poisson-Statistik.
</details>

<details><summary>10. Warum muss man Symmetriesektoren trennen, bevor man die Niveaustatistik berechnet?</summary>

Niveaus aus verschiedenen Sektoren koppeln nicht und stoßen sich nicht ab. Mischt man sie, sieht ein chaotisches System wie ein Poisson-System aus.
</details>

<details><summary>11. Was zeigt ⟨r⟩ in der Heisenberg-Kette mit zufälligen Feldern?</summary>

Bei schwacher Unordnung 0.53 (GOE): Die Kette ist chaotisch. Bei starker Unordnung etwa 0.39 (Poisson): Sie ist vielteilchenlokalisiert. Für $L = 12$ liegt der Übergang um $W\approx2$, für lange Ketten bei etwa $3.7$.
</details>

#### **Nächste Schritte**

- **Spektraler Formfaktor:** $K(t) = |\sum_n e^{-iE_nt}|^2$ zeigt bei chaotischen Systemen die Folge aus Abfall, Rampe und Plateau; die Rampe ist die Signatur der Niveauabstoßung auf langen Zeitskalen.
- **Haar-Maß und typische Zustände:** Zufällige unitäre Matrizen (CUE) und die Verschränkung zufälliger Zustände (Page-Kurve).
- **Netzwerke:** Das Spektrum der Adjazenzmatrix eines Zufallsgraphen ist ebenfalls ein Halbkreis (Notebook `complexnetworks.ipynb`).
- **Kovarianzen in Daten:** Eigenwerte über dem Marchenko-Pastur-Rand als Kriterium, wie viele Hauptkomponenten man behält, etwa bei Aktienrenditen.
