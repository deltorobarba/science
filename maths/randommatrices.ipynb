# Zufallsmatrizen

Das Notebook zieht große Zufallsmatrizen aus den Gaußschen Ensembles GOE und GUE und prüft numerisch die zentralen Ergebnisse der Theorie: den Wigner-Halbkreis, die Abstoßung benachbarter Eigenwerte (Wigner-Vermutung gegen Poisson-Statistik, Verhältnis-Statistik ⟨r⟩), die Marchenko-Pastur-Verteilung der Eigenwerte von Rausch-Kovarianzmatrizen mit dem BBP-Übergang, ab dem ein Signal aus dem Rauschen heraustritt, und den Übergang einer Spinkette mit zufälligen Feldern von Quantenchaos (GOE) zu Lokalisierung (Poisson).

Was bleibt von einer Matrix übrig, wenn ihre Einträge zufällig sind, und warum zeigen die Energieniveaus chaotischer Quantensysteme genau diese Statistik?

#### **Theorie**

**Gaußsche Ensembles.** Eine Zufallsmatrix ist eine Matrix mit zufälligen Einträgen. Die wichtigsten sind hermitesch mit unabhängigen Gauß-verteilten Einträgen:
- **GOE** (Gaussian Orthogonal Ensemble): reell symmetrisch, $\beta = 1$; passend für Systeme mit Zeitumkehrsymmetrie.
- **GUE** (Gaussian Unitary Ensemble): komplex hermitesch, $\beta = 2$; passend ohne Zeitumkehrsymmetrie, etwa mit Magnetfeld.

Die Verteilung ist invariant unter orthogonalen bzw. unitären Basiswechseln: Keine Richtung ist ausgezeichnet, alle Information steckt in den Eigenwerten.

**Universelle Ergebnisse.** Für große Matrizen hängen viele Eigenschaften der Eigenwerte nicht von den Details der Verteilung ab:
- Die Dichte der Eigenwerte ist ein **Halbkreis** (Wigner).
- Benachbarte Eigenwerte **stoßen sich ab**: Sehr kleine Abstände sind selten, die Abstandsverteilung geht für $s \to 0$ wie $s^\beta$ gegen null. Unabhängige, zufällig verteilte Zahlen haben dagegen exponentiell verteilte Abstände (**Poisson-Statistik**), und kleine Abstände sind am häufigsten.

**Kovarianzmatrizen.** Schätzt man aus $n$ Messungen von $p$ Größen die Kovarianzmatrix, streuen deren Eigenwerte auch dann, wenn die Größen in Wahrheit unkorreliert sind. Die Verteilung dieser Rausch-Eigenwerte ist die **Marchenko-Pastur-Verteilung**. Eigenwerte außerhalb ihres Bereichs zeigen echte Korrelationen an; das nutzt man in der Finanzmathematik, in der Genetik und bei der Hauptkomponentenanalyse.

**Quantenchaos.** Die Energieniveaus eines Quantensystems, dessen klassisches Gegenstück chaotisch ist, folgen der Abstandsstatistik von GOE oder GUE (Vermutung von Bohigas, Giannoni und Schmit 1984). Die Niveaus integrabler Systeme folgen dagegen meist der Poisson-Statistik (Berry und Tabor). Die Statistik der Niveauabstände ist deshalb ein Fingerabdruck von Chaos in Systemen ohne Trajektorien.

#### **Herleitung**

**1. Woher die Abstoßung kommt.** Die Verteilung der GOE-Matrizen ist $P(H) \propto e^{-\operatorname{tr} H^2 / 4}$ (mit Varianz 1 außerhalb der Diagonale). Schreibt man $H = O\Lambda O^T$ mit Eigenwerten $\Lambda$ und orthogonaler Matrix $O$, enthält das Volumenelement $dH$ die Jacobi-Determinante dieses Variablenwechsels, $\prod_{i<j}|\lambda_i - \lambda_j|^\beta$. Die gemeinsame Dichte der Eigenwerte ist damit
$$p(\lambda_1, \dots, \lambda_N) \propto \prod_{i<j}|\lambda_i - \lambda_j|^\beta\; e^{-\sum_i \lambda_i^2/4} .$$
Der Faktor $|\lambda_i - \lambda_j|^\beta$ macht gleiche Eigenwerte unwahrscheinlich: Um zwei Eigenwerte zusammenfallen zu lassen, müssen $\beta + 1$ unabhängige Bedingungen gleichzeitig erfüllt sein, nicht nur eine.

**2. Die Wigner-Vermutung aus $2\times2$-Matrizen.** Für $H = \begin{pmatrix} a & c \\ c & b\end{pmatrix}$ mit $a, b \sim N(0, 2)$ und $c \sim N(0, 1)$ ist der Abstand der Eigenwerte
$$s = \sqrt{(a - b)^2 + 4c^2} .$$
Beide Größen unter der Wurzel, $u = a - b$ und $v = 2c$, sind unabhängig normalverteilt mit Varianz 4. Der Abstand ist der Betrag des Vektors $(u, v)$ in der Ebene; in Polarkoordinaten bringt das Flächenelement $s\,ds\,d\varphi$ einen Faktor $s$:
$$P(s) \propto s\, e^{-s^2/8} .$$
Normiert auf Fläche 1 und mittleren Abstand 1 wird daraus
$$P_\text{GOE}(s) = \frac{\pi}{2}\, s\, e^{-\pi s^2/4}, \qquad P_\text{GUE}(s) = \frac{32}{\pi^2}\, s^2 e^{-4s^2/\pi} .$$
Beim GUE ist $c$ komplex, der Vektor hat drei Komponenten, und der Faktor wird $s^2$. Diese Formeln für $2\times2$ beschreiben auch große Matrizen bis auf wenige Prozent.

**3. Poisson-Statistik.** Liegen Niveaus unabhängig und gleichmäßig mit mittlerem Abstand 1, ist die Wahrscheinlichkeit, auf einer Strecke $s$ kein weiteres Niveau zu finden, $e^{-s}$, und die Abstandsdichte ist $P(s) = e^{-s}$, maximal bei $s = 0$.

**4. Der Halbkreis aus Momenten.** Mit Varianz 1 außerhalb der Diagonale wachsen die Eigenwerte wie $\sqrt N$; für $x = \lambda/\sqrt N$ ist das $k$-te Moment $\frac{1}{N}\mathbb{E}\operatorname{tr}(H/\sqrt N)^k$. Ausgeschrieben ist $\operatorname{tr} H^{2k}$ eine Summe über geschlossene Pfade $i_1 \to i_2 \to \dots \to i_1$ der Länge $2k$. Weil die Einträge den Mittelwert null haben, zählen nur Pfade, die jede Kante mindestens zweimal benutzen. Für großes $N$ dominieren die Pfade, die jede Kante genau zweimal benutzen und einen Baum bilden; ihre Anzahl ist die Catalan-Zahl $C_k = \frac{1}{k+1}\binom{2k}{k}$, und die ungeraden Momente verschwinden. Die Catalan-Zahlen sind genau die Momente der Halbkreisdichte
$$\rho(x) = \frac{1}{2\pi}\sqrt{4 - x^2}, \qquad |x| \le 2 .$$

**5. Entfalten und die Verhältnis-Statistik.** Die Dichte der Eigenwerte ist nicht konstant, also hängt der mittlere Abstand vom Ort im Spektrum ab. Um Abstände zu vergleichen, **entfaltet** man das Spektrum: Man ersetzt jeden Eigenwert $\lambda$ durch die erwartete Zahl der Eigenwerte unter ihm, $N\!\int_{-\infty}^{\lambda}\rho$, dann ist der mittlere Abstand überall 1. Wenn man die Dichte nicht kennt, nimmt man stattdessen das Verhältnis aufeinanderfolgender Abstände,
$$r_i = \frac{\min(s_i, s_{i+1})}{\max(s_i, s_{i+1})} \in [0, 1],$$
bei dem sich die lokale Dichte herauskürzt. Sein Mittelwert ist $\langle r\rangle = 2\ln2 - 1 \approx 0.386$ für Poisson, $\approx 0.531$ für GOE und $\approx 0.600$ für GUE (Atas et al. 2013).

**6. Marchenko-Pastur und der BBP-Übergang.** Sei $X$ eine $n\times p$-Matrix mit unabhängigen Einträgen der Varianz 1 und $C = X^T X/n$ die geschätzte Kovarianzmatrix. Die wahre Kovarianz ist die Einheitsmatrix, aber für $n, p \to \infty$ mit $\gamma = p/n \le 1$ verteilen sich die Eigenwerte von $C$ über
$$\lambda_\pm = (1 \pm \sqrt\gamma)^2, \qquad \rho(\lambda) = \frac{\sqrt{(\lambda_+ - \lambda)(\lambda - \lambda_-)}}{2\pi\gamma\lambda} .$$
Hat die wahre Kovarianz eine Richtung mit Varianz $1 + \ell$ (ein Signal der Stärke $\ell$), dann löst sich der größte Eigenwert nur ab, wenn $\ell > \sqrt\gamma$, und liegt dann bei $(1 + \ell)(1 + \gamma/\ell)$. Schwächere Signale verschwinden im Rauschen am Rand $\lambda_+$, egal wie man rechnet (Baik, Ben Arous und Péché 2005).

#### **Code**

Funktionen zum Ziehen der Ensembles, zum Entfalten mit dem Halbkreis und für die Verhältnis-Statistik.

```python
# Alles hier (numpy, scipy, matplotlib) ist in Colab vorinstalliert
import itertools
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(1)

def goe(N):
    A = rng.normal(size=(N, N))
    return (A + A.T) / np.sqrt(2)                     # Varianz 1 außerhalb, 2 auf der Diagonale

def gue(N):
    A = (rng.normal(size=(N, N)) + 1j * rng.normal(size=(N, N))) / np.sqrt(2)
    return (A + A.conj().T) / np.sqrt(2)              # E|H_ij|² = 1 außerhalb der Diagonale

semicircle = lambda x: np.sqrt(np.clip(4 - x ** 2, 0, None)) / (2 * np.pi)
semicircle_cdf = lambda x: 0.5 + (x * np.sqrt(np.clip(4 - x ** 2, 0, None)) / (4 * np.pi)) + np.arcsin(np.clip(x / 2, -1, 1)) / np.pi

def unfolded_spacings(ev, N, keep=0.5):
    """Entfalten mit dem Halbkreis, Abstände aus der mittleren Hälfte des Spektrums."""
    u = N * semicircle_cdf(np.sort(ev) / np.sqrt(N))
    s = np.diff(u)
    lo, hi = int(len(s) * (1 - keep) / 2), int(len(s) * (1 + keep) / 2)
    return s[lo:hi]

def ratios(ev):
    s = np.diff(np.sort(ev))
    return np.minimum(s[:-1], s[1:]) / np.maximum(s[:-1], s[1:])

wigner_goe = lambda s: np.pi / 2 * s * np.exp(-np.pi * s ** 2 / 4)
wigner_gue = lambda s: 32 / np.pi ** 2 * s ** 2 * np.exp(-4 * s ** 2 / np.pi)
poisson = lambda s: np.exp(-s)
print("Funktionen bereit")
```

#### **Experimente**

##### *1. Der Wigner-Halbkreis*

Folgen die Eigenwerte einer großen GOE- und GUE-Matrix dem Halbkreis, und hängt das von der Verteilung der Einträge ab?

Als dritter Fall dienen symmetrische Matrizen mit Einträgen $\pm1$ (gleich wahrscheinlich): nicht Gauß-verteilt, aber mit derselben Varianz.

```python
N = 1500
ev_goe = np.linalg.eigvalsh(goe(N))
ev_gue = np.linalg.eigvalsh(gue(N))
B = rng.choice([-1.0, 1.0], size=(N, N))
ev_pm = np.linalg.eigvalsh(np.triu(B) + np.triu(B, 1).T)

fig, ax = plt.subplots(figsize=(9, 4.5))
bins = np.linspace(-2.2, 2.2, 60)
for ev, name, c in [(ev_goe, "GOE", "tab:blue"), (ev_gue, "GUE", "tab:red"), (ev_pm, "Einträge ±1", "tab:green")]:
    ax.hist(ev / np.sqrt(N), bins=bins, density=True, histtype="step", lw=1.5, color=c, label=name)
xs = np.linspace(-2.2, 2.2, 400)
ax.plot(xs, semicircle(xs), c="black", lw=2, ls="--", label="Halbkreis √(4 − x²)/2π")
ax.set_xlabel("Eigenwert / √N")
ax.set_ylabel("Dichte")
ax.set_title(f"Eigenwertdichte, N = {N}")
ax.legend()
fig.tight_layout()
plt.show()
for ev, name in [(ev_goe, "GOE"), (ev_gue, "GUE"), (ev_pm, "±1")]:
    m = [np.mean((ev / np.sqrt(N)) ** k) for k in (2, 4, 6)]
    print(f"{name:4s}: Momente ⟨x²⟩, ⟨x⁴⟩, ⟨x⁶⟩ = {m[0]:.3f}, {m[1]:.3f}, {m[2]:.3f}   (Catalan-Zahlen: 1, 2, 5)")
```

##### *2. Abstoßung der Eigenwerte*

Wie sind die Abstände benachbarter Eigenwerte verteilt, und wie unterscheiden sie sich von unabhängigen Zufallszahlen?

Für GOE und GUE werden 40 Matrizen der Größe 400 gezogen und mit dem Halbkreis entfaltet (Herleitung 5). Zum Vergleich: unabhängige, gleichverteilte Zahlen mit mittlerem Abstand 1.

```python
N, REP = 400, 40
s_goe = np.concatenate([unfolded_spacings(np.linalg.eigvalsh(goe(N)), N) for _ in range(REP)])
s_gue = np.concatenate([unfolded_spacings(np.linalg.eigvalsh(gue(N)), N) for _ in range(REP)])
s_poi = np.diff(np.sort(rng.uniform(0, 20000, 20000)))

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
bins = np.linspace(0, 4, 41)
ss = np.linspace(0, 4, 300)
for s, name, c in [(s_poi, "unabhängige Zahlen", "tab:green"), (s_goe, "GOE", "tab:blue"), (s_gue, "GUE", "tab:red")]:
    axes[0].hist(s / s.mean(), bins=bins, density=True, alpha=0.35, color=c, label=name)
axes[0].plot(ss, poisson(ss), c="tab:green", lw=2, label="Poisson e^(−s)")
axes[0].plot(ss, wigner_goe(ss), c="tab:blue", lw=2, label="Wigner GOE (π/2) s e^(−πs²/4)")
axes[0].plot(ss, wigner_gue(ss), c="tab:red", lw=2, label="Wigner GUE (32/π²) s² e^(−4s²/π)")
axes[0].set_xlabel("Abstand s (mittlerer Abstand 1)")
axes[0].set_ylabel("Dichte P(s)")
axes[0].set_title("Abstandsverteilung")
axes[0].legend(fontsize=8)
small = np.logspace(-2, 0, 12)
for s, name, c, beta in [(s_poi, "unabhängig", "tab:green", 0), (s_goe, "GOE", "tab:blue", 1), (s_gue, "GUE", "tab:red", 2)]:
    cum = [np.mean(s / s.mean() < x) for x in small]
    axes[1].loglog(small, cum, "o-", c=c, label=f"{name}: Anteil ~ s^{beta + 1}")
axes[1].set_xlabel("Abstand s")
axes[1].set_ylabel("Anteil der Abstände kleiner als s")
axes[1].set_title("Kleine Abstände: P(s) ~ s^β")
axes[1].legend(fontsize=8)
fig.tight_layout()
plt.show()
for s, name in [(s_poi, "unabhängig"), (s_goe, "GOE"), (s_gue, "GUE")]:
    print(f"{name:10s}: Anteil der Abstände < 0.1: {np.mean(s / s.mean() < 0.1):.4f}")
```

##### *3. Die Verhältnis-Statistik*

Welchen Mittelwert hat $r$ für Poisson, GOE und GUE, wenn man nicht entfaltet?

Je 150 Matrizen der Größe 400, aus der mittleren Hälfte des Spektrums. Den Fehler schätzt die Streuung der Mittelwerte zwischen den Matrizen, nicht die Streuung einzelner Verhältnisse: Benachbarte Verhältnisse teilen sich einen Abstand und sind deshalb korreliert.

```python
N_MAT = 150
samples = {"Poisson": [ratios(rng.uniform(0, 1, 200)) for _ in range(N_MAT)],
           "GOE": [ratios(np.linalg.eigvalsh(goe(400))[100:300]) for _ in range(N_MAT)],
           "GUE": [ratios(np.linalg.eigvalsh(gue(400))[100:300]) for _ in range(N_MAT)]}
ref = {"Poisson": 2 * np.log(2) - 1, "GOE": 0.5307, "GUE": 0.5996}
for name, per_matrix in samples.items():
    means = np.array([r.mean() for r in per_matrix])
    # Fehler aus der Streuung zwischen den Matrizen: benachbarte Verhältnisse derselben Matrix sind korreliert
    print(f"{name:8s}: ⟨r⟩ = {means.mean():.4f} ± {means.std() / np.sqrt(N_MAT):.4f}   (Theorie {ref[name]:.4f})")
```

##### *4. Rauschen in Kovarianzmatrizen: Marchenko-Pastur und ein verstecktes Signal*

Wie verteilen sich die Eigenwerte einer Kovarianzmatrix aus reinem Rauschen, und ab welcher Stärke tritt ein echtes Signal aus diesem Rauschen heraus?

$n = 2000$ Messungen von $p = 500$ Größen, also $\gamma = 0.25$ und $\sqrt\gamma = 0.5$. Für das Signal bekommt eine zufällige Richtung die Varianz $1 + \ell$.

```python
n, p = 2000, 500
gam = p / n
lam_m, lam_p = (1 - np.sqrt(gam)) ** 2, (1 + np.sqrt(gam)) ** 2
mp = lambda x: np.sqrt(np.clip((lam_p - x) * (x - lam_m), 0, None)) / (2 * np.pi * gam * x)

X = rng.normal(size=(n, p))
ev_noise = np.linalg.eigvalsh(X.T @ X / n)

def top_eigenvalue(ell):
    v = rng.normal(size=p); v /= np.linalg.norm(v)
    Y = rng.normal(size=(n, p)) + np.sqrt(ell) * rng.normal(size=(n, 1)) * v      # Kovarianz I + ℓ v vᵀ
    return np.linalg.eigvalsh(Y.T @ Y / n)[-1]

ells = np.linspace(0.05, 2.0, 25)
tops = np.array([np.mean([top_eigenvalue(l) for _ in range(3)]) for l in ells])
theory = np.where(ells > np.sqrt(gam), (1 + ells) * (1 + gam / ells), lam_p)

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
axes[0].hist(ev_noise, bins=60, density=True, alpha=0.5, color="tab:blue", label=f"Eigenwerte, reines Rauschen (p = {p}, n = {n})")
xs = np.linspace(lam_m, lam_p, 400)
axes[0].plot(xs, mp(xs), c="black", lw=2, label="Marchenko-Pastur, γ = 0.25")
axes[0].axvline(1, color="tab:red", ls=":", label="wahre Varianz: 1")
axes[0].set_xlabel("Eigenwert der geschätzten Kovarianz")
axes[0].set_ylabel("Dichte")
axes[0].set_title("Rauschen allein streut die Eigenwerte von 0.25 bis 2.25")
axes[0].legend(fontsize=8)
axes[1].plot(ells, tops, "o", label="größter Eigenwert (Mittel aus 3 Ziehungen)")
axes[1].plot(ells, theory, c="black", label="Theorie: Rand λ₊ oder (1 + ℓ)(1 + γ/ℓ)")
axes[1].plot(ells, 1 + ells, c="grey", ls=":", label="wahre Varianz 1 + ℓ")
axes[1].axvline(np.sqrt(gam), color="tab:red", ls="--", label="Schwelle ℓ = √γ")
axes[1].set_xlabel("Stärke des Signals ℓ")
axes[1].set_ylabel("größter Eigenwert")
axes[1].set_title("BBP-Übergang")
axes[1].legend(fontsize=8)
fig.tight_layout()
plt.show()
print(f"Rauschen: kleinster Eigenwert {ev_noise.min():.3f} (Theorie {lam_m:.3f}), größter {ev_noise.max():.3f} (Theorie {lam_p:.3f})")
```

##### *5. Quantenchaos und Lokalisierung in einer Spinkette*

Zeigt eine Kette wechselwirkender Spins die Niveaustatistik von GOE, und was passiert, wenn starke zufällige Felder die Spins festhalten?

Das Modell ist die Heisenberg-Kette mit zufälligen Feldern, $H = \sum_i \vec S_i\cdot\vec S_{i+1} + \sum_i h_i S^z_i$, mit $h_i$ gleichverteilt in $[-W, W]$ und offenem Rand. Die Gesamtmagnetisierung ist erhalten; gerechnet wird im Sektor $\sum_i S^z_i = 0$, denn Niveaus aus verschiedenen Symmetriesektoren sind unabhängig und würden eine Poisson-Statistik vortäuschen. Bei schwacher Unordnung ist die Kette chaotisch, bei starker Unordnung **vielteilchenlokalisiert**: Die Spins bleiben in der Nähe ihres Anfangszustands, und die Niveaus verhalten sich wie unabhängige Zahlen.

```python
L = 12
basis = [s for s in itertools.product([0, 1], repeat=L) if sum(s) == L // 2]     # 1 = Spin hoch
index = {s: i for i, s in enumerate(basis)}
D = len(basis)

def heisenberg_random_field(W):
    h = rng.uniform(-W, W, L)
    H = np.zeros((D, D))
    for k, s in enumerate(basis):
        sz = np.array(s) - 0.5
        H[k, k] = np.sum(sz[:-1] * sz[1:]) + np.sum(h * sz)
        for i in range(L - 1):
            if s[i] != s[i + 1]:                                      # S⁺S⁻ + S⁻S⁺ tauscht antiparallele Nachbarn
                t_ = list(s); t_[i], t_[i + 1] = t_[i + 1], t_[i]
                H[index[tuple(t_)], k] = 0.5
    return H

Ws = [0.25, 0.5, 1, 2, 3, 4, 6, 8, 12]
r_mean = []
for W in Ws:
    rs_ = [ratios(np.linalg.eigvalsh(heisenberg_random_field(W))[D // 4: 3 * D // 4]) for _ in range(15)]
    r_mean.append(np.mean(np.concatenate(rs_)))

fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(Ws, r_mean, "o-", c="black", label=f"Heisenberg-Kette, L = {L}, Sektor Sᶻ = 0 ({D} Zustände)")
ax.axhline(0.5307, color="tab:blue", ls="--", label="GOE 0.531")
ax.axhline(2 * np.log(2) - 1, color="tab:green", ls="--", label="Poisson 0.386")
ax.set_xscale("log")
ax.set_xlabel("Stärke der Unordnung W")
ax.set_ylabel("⟨r⟩ (mittlere Hälfte des Spektrums)")
ax.set_title("Von Quantenchaos zu Lokalisierung")
ax.legend(fontsize=8)
fig.tight_layout()
plt.show()
for W, r in zip(Ws, r_mean):
    print(f"W = {W:5.2f}:  ⟨r⟩ = {r:.3f}")
```

#### **Ergebnis**

Beim Lauf vom 2026-10-12:

- **Halbkreis:** GOE, GUE und Matrizen mit Einträgen $\pm1$ haben dieselbe Eigenwertdichte; ihre Momente $\langle x^2\rangle, \langle x^4\rangle, \langle x^6\rangle$ sind $1, 2, 5$, die Catalan-Zahlen. Die Verteilung der Einträge spielt keine Rolle, nur ihre Varianz (Universalität).
- **Abstoßung:** Die Abstände folgen den Wigner-Formeln. Abstände unter 0.1 haben 9.4 % der unabhängigen Zahlen, aber nur 1.0 % beim GOE und 0.09 % beim GUE; der Anteil wächst wie $s^{\beta+1}$.
- **Verhältnis-Statistik:** $\langle r\rangle = 0.384 \pm 0.002$ (Poisson), $0.531 \pm 0.002$ (GOE), $0.599 \pm 0.002$ (GUE), wie vorhergesagt (0.386, 0.531, 0.600).
- **Rauschen in Kovarianzen:** Obwohl alle wahren Varianzen 1 sind, liegen die geschätzten Eigenwerte zwischen 0.256 und 2.244 (Theorie 0.25 bis 2.25). Mit viermal so vielen Messungen wie Größen überschätzt die Stichprobe die größte Varianz also um mehr als das Doppelte.
- **BBP-Übergang:** Ein Signal mit $\ell < \sqrt\gamma = 0.5$ ist im größten Eigenwert nicht zu sehen; darüber folgt dieser der Formel $(1 + \ell)(1 + \gamma/\ell)$ und liegt stets über der wahren Varianz $1 + \ell$.
- **Spinkette:** Bei schwacher Unordnung ($W \le 1$) ist $\langle r\rangle = 0.53$: Die Kette ist chaotisch, und ihre Niveaus verhalten sich wie die Eigenwerte einer GOE-Matrix. Ab $W \approx 3$ ist $\langle r\rangle \approx 0.39$: Die Niveaus sind unabhängig, die Kette ist lokalisiert. Für diese kurze offene Kette ($L = 12$) liegt der Übergang um $W \approx 2$; Rechnungen mit längeren, geschlossenen Ketten finden ihn bei etwa $W \approx 3.7$ (Luitz, Laflorencie & Alet 2015). Ob er für unendlich lange Ketten bestehen bleibt, ist umstritten.

- **Spektraler Formfaktor:** $K(t) = |\sum_n e^{-iE_nt}|^2$ zeigt bei chaotischen Systemen die Folge aus Abfall, Rampe und Plateau; die Rampe ist die Signatur der Niveauabstoßung auf langen Zeitskalen.
- **Haar-Maß und typische Zustände:** Zufällige unitäre Matrizen (CUE) und die Verschränkung zufälliger Zustände (Page-Kurve).
