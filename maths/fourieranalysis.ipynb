#### **Hintergrund**

Die **harmonische Analysis** zerlegt Funktionen in Grundschwingungen, die „Harmonischen“. Ein Ton wird zur Liste seiner Frequenzen, ein Bild zur Liste seiner Streifenmuster, eine Temperaturverteilung zur Liste ihrer Wellen. Der Vorteil dieser Darstellung: Viele Operationen, die im Ort oder in der Zeit kompliziert sind, werden in den Frequenzen einfach. Ableiten wird zu Multiplizieren, Glätten (Falten) wird zu Multiplizieren, Verschieben wird zu einer Phase. Deshalb ist die Fourier-Transformation eines der meistbenutzten Werkzeuge der Mathematik, Physik und Technik.

Joseph Fourier behauptete 1807 in seiner Arbeit zur Wärmeleitung, dass sich jede Funktion als Reihe aus Sinus- und Kosinusfunktionen schreiben lasse; Lagrange hielt das für falsch. Den ersten Konvergenzbeweis lieferte Dirichlet 1829. Die Frage, für welche Funktionen und in welchem Sinn die Reihe konvergiert, trieb die Entwicklung des Integralbegriffs (Riemann, Lebesgue) und der Mengenlehre (Cantor) an; abgeschlossen wurde sie erst 1966 durch Carleson. Gibbs beschrieb 1899 das Überschwingen an Sprüngen, Fejér 1900 die gemittelten Partialsummen, Plancherel 1910 die Erhaltung der Energie. Pontryagin, Weyl und Peter verallgemeinerten die Theorie auf Gruppen. Die **schnelle Fourier-Transformation** (FFT) fand schon Gauß 1805, bekannt wurde sie 1965 durch Cooley und Tukey; Shannon formulierte 1949 das Abtasttheorem, und ab den 1980er Jahren kamen die Wavelets hinzu (Morlet, Grossmann, Meyer, Mallat, Daubechies).

#### **Theorie**

Hier steht das Allgemeine; die Theorie jedes einzelnen Themas steht in seinem Experiment, direkt nach den Fachbegriffen.

##### *1. Zerlegen in Schwingungen*

- **Schwingung:** $A\cos(2\pi\xi x + \varphi)$ mit **Amplitude** $A$, **Frequenz** $\xi$ (Schwingungen pro Längen- oder Zeiteinheit) und **Phase** $\varphi$. Die **Kreisfrequenz** ist $\omega = 2\pi\xi$, die **Periode** $1/\xi$.
- **Komplexe Exponentialfunktion:** Nach der Euler-Formel ist $e^{i\theta} = \cos\theta + i\sin\theta$. Eine Schwingung mit Amplitude und Phase ist deshalb eine einzige komplexe Zahl mal $e^{2\pi i\xi x}$: Der Betrag ist die Amplitude, das Argument die Phase. Mit diesen Funktionen rechnet es sich einfacher als mit Sinus und Kosinus getrennt.
- **Skalarprodukt** zweier $2\pi$-periodischer Funktionen: $\langle f, g\rangle = \frac1{2\pi}\int_{-\pi}^\pi f(x)\overline{g(x)}\,\mathrm dx$; die **Norm** ist $\Vert f\Vert = \sqrt{\langle f, f\rangle}$, ihr Quadrat die **Energie** oder **Leistung** der Funktion. Die Funktionen $e_k(x) = e^{ikx}$, $k\in\mathbb Z$, sind **orthonormal**: $\langle e_k, e_l\rangle = 1$ für $k = l$ und 0 sonst.
- **Fourier-Koeffizient** $c_k = \langle f, e_k\rangle = \frac1{2\pi}\int_{-\pi}^\pi f(x)e^{-ikx}\,\mathrm dx$: die Projektion von $f$ auf die Schwingung $e_k$, genau wie man die Koordinate eines Vektors als Skalarprodukt mit einem Basisvektor berechnet. Die **Fourier-Reihe** ist $f(x)\sim\sum_kc_ke^{ikx}$, die **Partialsumme** $S_Nf = \sum_{\vert k\vert\leq N}c_ke^{ikx}$. Reell geschrieben: $f\sim\tfrac{a_0}2 + \sum_{k\geq1}(a_k\cos kx + b_k\sin kx)$ mit $a_k = c_k + c_{-k}$ und $b_k = i(c_k - c_{-k})$.
- **Spektrum:** die Beträge $\vert c_k\vert$ (oder $\vert\hat f(\xi)\vert$) als Funktion der Frequenz; das **Leistungsspektrum** ist ihr Quadrat.

##### *2. Die vier Fourier-Transformationen*

Welche Transformation man benutzt, hängt davon ab, ob das Signal stetig oder diskret und ob es periodisch ist oder nicht:

| Signal lebt auf | Frequenzen leben auf | Name | Formel |
| --- | --- | --- | --- |
| Kreis (periodisch, stetig) | ganzen Zahlen $\mathbb Z$ | Fourier-Reihe | $c_k = \frac1{2\pi}\int_{-\pi}^\pi f(x)e^{-ikx}\,\mathrm dx$ |
| Gerade $\mathbb R$ | Gerade $\mathbb R$ | Fourier-Transformation | $\hat f(\xi) = \int_{-\infty}^\infty f(x)e^{-2\pi ix\xi}\,\mathrm dx$ |
| ganzen Zahlen $\mathbb Z$ (Abtastwerte) | Kreis | zeitdiskrete FT (DTFT) | $X(\omega) = \sum_nx_ne^{-i\omega n}$ |
| $\mathbb Z_N$ ($N$ Werte, periodisch fortgesetzt) | $\mathbb Z_N$ | diskrete FT (DFT) | $X_k = \sum_{n=0}^{N-1}x_ne^{-2\pi ikn/N}$ |

Faustregel: **Diskret auf der einen Seite heißt periodisch auf der anderen.** Ein Computer kennt nur endlich viele Werte und rechnet deshalb immer mit der DFT; die anderen drei nähert er damit an. Dieses Notebook benutzt für die Fourier-Transformation die Konvention mit $\xi$ in Schwingungen pro Einheit: Dann ist die Rücktransformation $f(x) = \int\hat f(\xi)e^{2\pi ix\xi}\,\mathrm d\xi$ ohne Vorfaktoren. In der Physik ist auch $\hat f(\omega) = \int f(x)e^{-i\omega x}\,\mathrm dx$ mit $\frac1{2\pi}$ bei der Rücktransformation üblich; die Formeln unterscheiden sich dann um Faktoren $2\pi$. numpy rechnet die DFT ohne Vorfaktor und teilt bei `ifft` durch $N$.

##### *3. Rechenregeln*

Für die Fourier-Transformation $f\mapsto\hat f$ gilt (für Reihen und DFT entsprechend):

| im Ort | in der Frequenz |
| --- | --- |
| Verschiebung $f(x - a)$ | Phase $e^{-2\pi ia\xi}\hat f(\xi)$ |
| Modulation $e^{2\pi ibx}f(x)$ | Verschiebung $\hat f(\xi - b)$ |
| Streckung $f(ax)$ | Stauchung $\frac1{\vert a\vert}\hat f(\xi/a)$ |
| Ableitung $f'(x)$ | Multiplikation $2\pi i\xi\,\hat f(\xi)$ |
| Faltung $(f\ast g)(x) = \int f(y)g(x - y)\,\mathrm dy$ | Produkt $\hat f(\xi)\,\hat g(\xi)$ |
| Produkt $f(x)g(x)$ | Faltung $(\hat f\ast\hat g)(\xi)$ |
| $f$ reell | $\hat f(-\xi) = \overline{\hat f(\xi)}$ |
| $f$ reell und gerade | $\hat f$ reell und gerade |
| Energie $\int\vert f\vert^2\,\mathrm dx$ | gleiche Energie $\int\vert\hat f\vert^2\,\mathrm d\xi$ (Plancherel) |

Zwei Grundmuster ziehen sich durch das ganze Notebook: **Glatt im Ort heißt schnell abfallend in der Frequenz** (Ableiten multipliziert mit $\xi$; eine $m$-mal ableitbare Funktion hat deshalb ein Spektrum, das schneller als $1/\xi^m$ fällt). **Schmal im Ort heißt breit in der Frequenz** (Streckungsregel; quantitativ das Unschärfeprinzip).

##### *4. Harmonische Analysis auf Gruppen*

Warum gerade Schwingungen? Die Antwort ist die Verschiebung. Auf $\mathbb R$, dem Kreis, $\mathbb Z_N$ oder $\mathbb Z_2^n$ kann man Funktionen verschieben, und diese Bereiche sind **Gruppen**: Man kann Elemente addieren.
- **Charakter:** eine Funktion $\chi$ mit $\vert\chi\vert = 1$ und $\chi(g + h) = \chi(g)\,\chi(h)$. Die Charaktere von $\mathbb R$ sind $e^{2\pi ix\xi}$, die des Kreises $e^{ikx}$, die von $\mathbb Z_N$ die Funktionen $e^{2\pi ikn/N}$, die von $\mathbb Z_2^n$ die Funktionen $(-1)^{S\cdot x}$. Die Charaktere bilden selbst wieder eine Gruppe, die **duale Gruppe** (Pontryagin-Dualität): $\mathbb R$ ist zu sich selbst dual, der Kreis zu $\mathbb Z$, $\mathbb Z_N$ zu sich selbst.
- **Fourier-Transformation** heißt immer: eine Funktion nach den Charakteren ihrer Gruppe zerlegen. Alle vier Varianten aus Abschnitt 2 und die Walsh-Hadamard-Transformation (Exp. 15) sind Spezialfälle.
- **Warum das hilft:** Ein verschobener Charakter ist ein Vielfaches seiner selbst, $\chi(x + a) = \chi(a)\,\chi(x)$. Charaktere sind also Eigenfunktionen *jeder* Verschiebung und damit jedes linearen Operators, der mit Verschiebungen vertauscht (**translationsinvariant**): Ableitungen, Differentialoperatoren mit konstanten Koeffizienten, Faltungen, lineare zeitinvariante Systeme (LTI-Systeme) in der Signalverarbeitung. In der Fourier-Basis werden all diese Operatoren **diagonal**: Jede Frequenz wird nur mit einer Zahl multipliziert. Das ist der eigentliche Grund für den Erfolg der Fourier-Analyse.
- **Nicht kommutative Gruppen:** Bei Drehungen im Raum ($SO(3)$) treten an die Stelle der Charaktere irreduzible Darstellungen (Satz von Peter-Weyl); auf der Kugel werden daraus die **Kugelflächenfunktionen**, mit denen man Drehimpulse in der Quantenmechanik, das Schwerefeld der Erde und die kosmische Hintergrundstrahlung beschreibt.

##### *5. Konvergenzbegriffe und Funktionenräume*

- **Punktweise Konvergenz:** $S_Nf(x)\to f(x)$ an jeder Stelle $x$ einzeln. **Gleichmäßige Konvergenz:** Der größte Fehler $\max_x\vert f(x) - S_Nf(x)\vert$ geht gegen 0. **Konvergenz im quadratischen Mittel** ($L^2$): $\Vert f - S_Nf\Vert^2 = \frac1{2\pi}\int\vert f - S_Nf\vert^2\to0$; einzelne Ausreißer an wenigen Stellen stören hier nicht.
- **$L^1$ und $L^2$:** die Funktionen mit $\int\vert f\vert < \infty$ bzw. $\int\vert f\vert^2 < \infty$. $L^2$ ist ein **Hilbertraum** (ein Vektorraum mit Skalarprodukt, in dem Cauchy-Folgen konvergieren); die $e^{ikx}$ sind eine **Orthonormalbasis** davon, und die Fourier-Transformation ist eine **unitäre** Abbildung von $L^2$ auf sich selbst: ein Basiswechsel, der Längen und Winkel erhält.
- **Lemma von Riemann-Lebesgue:** Für jede integrierbare Funktion gehen die Fourier-Koeffizienten gegen 0. Wie schnell, hängt von der Glattheit ab (Exp. 1).
- **Was konvergiert wann:** Für jede $L^2$-Funktion konvergiert die Fourier-Reihe im quadratischen Mittel (Riesz-Fischer, Exp. 3). Für stückweise glatte Funktionen konvergiert sie punktweise, an Sprungstellen gegen den Mittelwert aus linkem und rechtem Grenzwert (Dirichlet). Für stetige Funktionen kann sie an einzelnen Stellen divergieren (du Bois-Reymond 1873), ihre Fejér-Mittel konvergieren aber immer gleichmäßig (Exp. 2). Für $L^2$-Funktionen konvergiert sie fast überall (Carleson 1966), für $L^1$-Funktionen nicht unbedingt (Kolmogorov 1923).
- **Distributionen:** Auch der Dirac-Stoß $\delta$ hat eine Fourier-Transformierte, die Konstante 1; ein Dirac-Kamm (Stöße im Abstand 1) hat als Transformierte wieder einen Dirac-Kamm (Exp. 7).

#### **Code**

Vier Werkzeuge, die mehrere Experimente brauchen. `fourier_coefficients` berechnet die Fourier-Koeffizienten aus Abtastwerten über die FFT; auf einem gleichmäßigen Gitter ist das die Trapezregel für das Integral. `partial_sum` bildet Partialsummen, mit Gewichten auch die Fejér-Mittel. `continuous_ft` nähert die Fourier-Transformation einer Funktion auf der Geraden mit der FFT, einschließlich der Phase für den Anfang des Gitters. `spread` misst die Breite einer Verteilung $\vert f\vert^2$ für das Unschärfeprinzip. Die übrigen Hilfsfunktionen stehen in ihren Experimenten.

```python
# Alles hier (numpy, scipy, matplotlib) ist in Colab vorinstalliert
import time
from fractions import Fraction
import numpy as np
import matplotlib.pyplot as plt
from scipy import fft as sfft, integrate, signal, special

def fourier_coefficients(f, K, M=2 ** 14):
    """Fourier-Koeffizienten c_k (k = −K … K) einer 2π-periodischen Funktion aus M Abtastwerten auf [−π, π)."""
    x = -np.pi + 2 * np.pi * np.arange(M) / M
    k = np.arange(-K, K + 1)
    return k, np.fft.fft(f(x))[k % M] / M * (-1.0) ** k        # Faktor (−1)^k, weil das Gitter bei −π beginnt

def partial_sum(k, c, x, weights=None):
    """Fourier-Partialsumme Σ w_k c_k e^{ikx}; mit Gewichten w_k z. B. das Fejér-Mittel."""
    w = np.ones_like(c) if weights is None else weights
    return np.real(np.exp(1j * np.outer(x, k)) @ (w * c))

def continuous_ft(fx, x):
    """Fourier-Transformation f̂(ξ) = ∫ f(x) e^{−2πixξ} dx, genähert mit der FFT auf einem gleichmäßigen Gitter x."""
    dx = x[1] - x[0]
    xi = np.fft.fftshift(np.fft.fftfreq(len(x), dx))
    return xi, dx * np.exp(-2j * np.pi * xi * x[0]) * np.fft.fftshift(np.fft.fft(fx))

def spread(values, grid):
    """Breite (Standardabweichung) der Verteilung |f|² auf einem Gitter."""
    p = np.abs(values) ** 2
    p = p / p.sum()
    mean = np.sum(grid * p)
    return np.sqrt(np.sum((grid - mean) ** 2 * p))

print("Werkzeuge bereit: Fourier-Koeffizienten, Partialsummen, Fourier-Transformation per FFT, Breite einer Verteilung")
```

#### **Experimente**

##### *1. Glattheit und der Abfall der Fourier-Koeffizienten*

Wie hängt das Spektrum einer Funktion von ihrer Glattheit ab? Vorher vorhersagen: Wie schnell fallen die Koeffizienten des Dreiecks $\vert x\vert$ im Vergleich zum Rechteck, und wird der größte Fehler der Partialsumme beim Rechteck jemals klein?

Untersucht werden vier $2\pi$-periodische Funktionen mit wachsender Glattheit: ein Rechteck (Sprung), das Dreieck $\vert x\vert$ (stetig, aber mit Knick), $x(\pi - \vert x\vert)$ (einmal stetig differenzierbar, Sprung erst in der zweiten Ableitung) und der Poisson-Kern $0.75/(1.25 - \cos x)$ (analytisch, beliebig oft differenzierbar). Links stehen die Funktionen mit ihren Partialsummen bis $\vert k\vert = 5$, in der Mitte die Beträge der Koeffizienten doppelt logarithmisch, rechts der größte Fehler der Partialsumme $S_N$ gegen $N$. Die Ausgabe gibt die gemessenen Abfallraten und vergleicht einzelne Koeffizienten mit den Formeln.

**Hintergrund & Fachbegriffe:**
- **Partialsumme** $S_N$: die Fourier-Reihe abgebrochen nach den Frequenzen $\vert k\vert\leq N$.
- **Abfallrate:** wie schnell $\vert c_k\vert$ mit $k$ kleiner wird, etwa wie $1/k^p$ (Potenzgesetz) oder wie $r^k$ (exponentiell).
- **Doppelt logarithmischer Plot:** Ein Potenzgesetz $1/k^p$ wird zur Geraden mit Steigung $-p$; ein exponentieller Abfall krümmt sich nach unten.
- **Sprung, Knick:** Eine Funktion springt, wenn sie an einer Stelle unstetig ist; sie hat einen Knick, wenn sie stetig ist, aber ihre Ableitung springt.
- **Analytisch:** lokal durch eine konvergente Potenzreihe darstellbar; dann ist die Funktion beliebig oft differenzierbar und lässt sich ins Komplexe fortsetzen.
- **Gleichmäßige Konvergenz:** Der größte Fehler über alle $x$ geht gegen 0 (Theorie, Abschnitt 5).

**Theorie:**

- **Glattheit gegen Abfall:** Partielle Integration gibt $c_k(f') = ik\,c_k(f)$. Ist $f$ periodisch und $(m - 1)$-mal stetig differenzierbar und hat $f^{(m)}$ Sprünge, so fallen die Koeffizienten wie $1/k^{m+1}$. Ein Sprung in $f$ gibt $1/k$, ein Knick $1/k^2$, ein Sprung in $f''$ gibt $1/k^3$.
- **Die Formeln:** Rechteck $c_k = \frac2{i\pi k}$ für ungerade $k$; Dreieck $c_0 = \pi/2$, $c_k = -\frac2{\pi k^2}$ für ungerade $k$; $x(\pi - \vert x\vert)$: $\vert c_k\vert = \frac4{\pi k^3}$ für ungerade $k$; alle geraden $k\neq0$ sind 0.
- **Analytische Funktionen:** Lässt sich $f$ in einen Streifen $\vert\mathrm{Im}\,z\vert < a$ der komplexen Ebene fortsetzen, fallen die Koeffizienten wie $e^{-a\vert k\vert}$; umgekehrt folgt aus exponentiellem Abfall Analytizität (Paley-Wiener). Der Poisson-Kern $\frac{1 - r^2}{1 - 2r\cos x + r^2} = \sum_kr^{\vert k\vert}e^{ikx}$ hat Pole bei $\mathrm{Im}\,z = \pm\ln(1/r)$; für $r = 0.5$ ist $c_k = 0.5^{\vert k\vert}$.
- **Fehler der Partialsumme:** $\max_x\vert f - S_Nf\vert\leq\sum_{\vert k\vert > N}\vert c_k\vert$. Bei $1/k^2$ fällt die Schranke wie $1/N$, bei $1/k^3$ wie $1/N^2$, bei $r^k$ wie $r^N$. Bei $1/k$ divergiert die Summe, und tatsächlich bleibt der Fehler groß: Eine stetige Partialsumme kann keinen Sprung machen.

```python
funcs = {
    "Rechteck (Sprung)": lambda x: np.sign(x) * (np.abs(x) < np.pi),
    "Dreieck |x| (Knick)": lambda x: np.abs(x),
    "x(π − |x|) (Sprung erst in f'')": lambda x: x * (np.pi - np.abs(x)),
    "Poisson-Kern r = 0.5 (analytisch)": lambda x: 0.75 / (1.25 - np.cos(x)),
}
exact = [lambda k: 2 / (np.pi * k), lambda k: 2 / (np.pi * k ** 2), lambda k: 4 / (np.pi * k ** 3), lambda k: 0.5 ** k]
K = 400
x = np.linspace(-np.pi, np.pi, 4001)
colors = ["tab:blue", "tab:orange", "tab:green", "tab:red"]
coef = {name: fourier_coefficients(f, K) for name, f in funcs.items()}

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
for (name, f), c in zip(funcs.items(), colors):
    k, ck = coef[name]
    axes[0].plot(x, f(x), c=c, lw=1, ls="--")
    axes[0].plot(x, partial_sum(k[np.abs(k) <= 5], ck[np.abs(k) <= 5], x), c=c, lw=1.5, label=name)
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].set_title("Gestrichelt f, durchgezogen die Partialsumme bis |k| = 5", fontsize=10)
axes[0].legend(fontsize=7)
kk = np.arange(1, K + 1)
for (name, _), c, ex in zip(funcs.items(), colors, exact):
    k, ck = coef[name]
    mag = np.abs(ck[K + 1:])
    keep = mag > 1e-15
    axes[1].loglog(kk[keep], mag[keep], ".", c=c, ms=3, label=name)
for p, ls in zip([1, 2, 3], [":", "--", "-."]):
    axes[1].loglog(kk, 0.6 * kk ** -float(p), "k", ls=ls, lw=0.8, label=f"∝ 1/k^{p}")
axes[1].set_ylim(1e-12, 3)
axes[1].set_xlabel("Frequenz k")
axes[1].set_ylabel("|c_k|")
axes[1].set_title("Je glatter f, desto schneller fallen die Koeffizienten", fontsize=10)
axes[1].legend(fontsize=7, ncol=2)
Ns = np.unique(np.round(np.geomspace(1, 300, 30)).astype(int))
near = np.geomspace(1e-7, 0.1, 300)
xe = np.sort(np.r_[x, near, -near, np.pi - near, -np.pi + near])   # mit Punkten dicht an den Sprungstellen 0 und ±π
for (name, f), c in zip(funcs.items(), colors):
    k, ck = coef[name]
    err = [np.max(np.abs(f(xe) - partial_sum(k[np.abs(k) <= N], ck[np.abs(k) <= N], xe))) for N in Ns]
    axes[2].loglog(Ns, err, "o-", c=c, ms=3, label=name)
axes[2].set_xlabel("Zahl der Frequenzen N")
axes[2].set_ylabel("größter Fehler max |f − S_N|")
axes[2].set_title("Fehler der Partialsumme S_N", fontsize=10)
axes[2].legend(fontsize=7)
plt.tight_layout()
plt.show()

odd = np.arange(11, 202, 2)
for (name, _), ex in zip(funcs.items(), exact):
    k, ck = coef[name]
    if "Poisson" in name:
        ratio = np.abs(ck[K + 11]) / np.abs(ck[K + 10])
        print(f"{name:34s}: c_11/c_10 = {ratio:.4f} (Theorie r = 0.5), |c_20| = {abs(ck[K + 20]):.2e} (Theorie {0.5 ** 20:.2e})")
    else:
        slope = np.polyfit(np.log(odd), np.log(np.abs(ck[K + odd])), 1)[0]
        print(f"{name:34s}: Steigung von log|c_k| gegen log k: {slope:.3f};  |c_11| = {abs(ck[K + 11]):.5f} (Formel {ex(11):.5f})")
```

* Das **linke Bild** zeigt die vier Funktionen (gestrichelt) mit ihren Partialsummen aus nur fünf Frequenzen (durchgezogen). Poisson-Kern (rot) und $x(\pi - \vert x\vert)$ (grün) sind schon fast exakt, das Dreieck (orange) ist an der Spitze leicht abgerundet, das Rechteck (blau) wellt sich und schafft den Sprung nicht. Im **mittleren Bild** liegen die Koeffizienten von Rechteck, Dreieck und $x(\pi - \vert x\vert)$ auf Geraden der Steigungen $-1$, $-2$ und $-3$; laut **Ausgabe** gemessen $-1.000$, $-2.000$, $-3.000$, und $\vert c_{11}\vert = 0.05787$, $0.00526$, $0.00096$ wie die Formeln. * Im **rechten Bild** bleibt der größte Fehler beim Rechteck bei etwa 1, gleich wie viele Frequenzen man nimmt: Direkt neben dem Sprung ist $S_N\approx0$, während $f = \pm1$ ist. Beim Dreieck fällt der Fehler wie $1/N$, bei $x(\pi - \vert x\vert)$ wie $1/N^2$, beim Poisson-Kern exponentiell bis zur Rechengenauigkeit ($10^{-15}$ ab etwa $N = 50$). Das beantwortet die Vorhersage: Glattheit entscheidet, wie gut wenige Frequenzen reichen.
* **Wo es vorkommt:** **Bildkompression** (JPEG wirft hohe Frequenzen weg; glatte Bildbereiche brauchen wenige Koeffizienten, Kanten viele), **Spektralverfahren** für Differentialgleichungen (bei glatten Lösungen exponentiell genau, „spektrale Genauigkeit“), die Analyse von Turbulenz und Oberflächen (der Abfall des Spektrums misst die Rauheit), Elektronik (steile Flanken digitaler Signale enthalten viele Oberwellen und stören Nachbarschaltungen). Im maschinellen Lernen lernen neuronale Netze glatte Anteile einer Funktion zuerst (**spectral bias**).

##### *2. Das Gibbs-Phänomen und die Fejér-Mittel*

Wird das Überschwingen an einem Sprung kleiner, wenn man mehr Frequenzen nimmt? Vorher vorhersagen: Wie hoch ist das Maximum der Partialsumme des Rechtecks (Sprung von $-1$ auf $+1$) für $N = 10$, $50$ und $1000$?

Für das Rechteck werden die Partialsummen $S_N$ und die Fejér-Mittel $\sigma_N$ nahe am Sprung bei $x = 0$ berechnet. Die Koeffizienten stammen hier aus einem feineren Gitter ($2^{18}$ Punkte), damit auch die bis $k = 2000$ genau sind. Links stehen $S_{10}$, $S_{50}$, $S_{250}$ und $\sigma_{250}$, in der Mitte der Dirichlet-Kern und der Fejér-Kern für $N = 10$, rechts das Maximum beider Näherungen nahe am Sprung für $N$ von 5 bis 1000. Die Ausgabe gibt die Maxima und die Lebesgue-Konstanten.

**Hintergrund & Fachbegriffe:**
- **Gibbs-Phänomen:** das Überschwingen der Partialsummen an einer Sprungstelle.
- **Integralsinus:** $\mathrm{Si}(x) = \int_0^x\frac{\sin t}t\,\mathrm dt$; $\frac2\pi\mathrm{Si}(\pi) = 1.17898$.
- **Dirichlet-Kern** $D_N(x) = \sum_{\vert k\vert\leq N}e^{ikx} = \frac{\sin((N + \frac12)x)}{\sin(x/2)}$: die Funktion, mit der man $f$ falten muss, um $S_Nf$ zu erhalten.
- **Fejér-Mittel** $\sigma_N = \frac1{N + 1}(S_0 + S_1 + \dots + S_N)$ (Cesàro-Mittel): das arithmetische Mittel der Partialsummen. Gleichwertig: die Koeffizienten werden mit $1 - \vert k\vert/(N + 1)$ gewichtet statt abgeschnitten.
- **Lebesgue-Konstante** $L_N = \frac1{2\pi}\int\vert D_N\vert$: wie stark $S_N$ eine beschränkte Funktion höchstens vergrößern kann.

**Theorie:**

- **Faltung mit dem Dirichlet-Kern:** Abschneiden der Koeffizienten bei $\vert k\vert = N$ heißt, $f$ mit $D_N$ zu falten (Faltungssatz). $D_N$ hat einen hohen Hauptgipfel ($2N + 1$) und negative Nebenschwinger. Am Sprung des Rechtecks wird daraus $S_N(x)\approx\frac2\pi\mathrm{Si}((N + \frac12)x)$: Das Maximum liegt bei $x\approx\pi/(N + 1)$ und strebt gegen $\frac2\pi\mathrm{Si}(\pi) = 1.17898$. Das Überschwingen beträgt also $8.95\,\%$ der Sprunghöhe, unabhängig von $N$ und von der Funktion (Wilbraham 1848, Gibbs 1899). Es wandert nur immer näher an den Sprung.
- **Lebesgue-Konstanten:** $L_N\approx\frac4{\pi^2}\ln N + 1.27$ wächst ohne Grenze. Deshalb gibt es stetige Funktionen, deren Fourier-Reihe an einer Stelle divergiert (du Bois-Reymond 1873).
- **Fejér-Kern** $F_N(x) = \frac1{N + 1}\Big(\frac{\sin((N + 1)x/2)}{\sin(x/2)}\Big)^2$ ist überall $\geq0$ und hat Mittelwert 1. Ein Mittel mit positiven Gewichten kann nie über das Maximum von $f$ hinausgehen: kein Überschwingen. **Satz von Fejér** (1900): Für jede stetige periodische Funktion konvergiert $\sigma_N$ gleichmäßig. Der Preis: An glatten Stellen konvergiert $\sigma_N$ nur wie $1/N$, langsamer als $S_N$.
- Die Gewichte $1 - \vert k\vert/(N + 1)$ sind ein Dreiecksfenster in der Frequenz; dieselbe Idee, sanft statt hart abzuschneiden, steckt in den Fensterfunktionen von Exp. 10.

```python
square = funcs["Rechteck (Sprung)"]
k, ck = fourier_coefficients(square, 2000, M=2 ** 18)                # feines Gitter, damit auch c_2000 genau ist
xz = np.linspace(-0.3, 1.0, 3000)

def fejer_weights(k, N):
    return np.clip(1 - np.abs(k) / (N + 1), 0, None)

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
axes[0].plot(xz, square(xz), "k-", lw=1, label="Rechteck")
for N, c in zip([10, 50, 250], ["tab:blue", "tab:orange", "tab:green"]):
    m = np.abs(k) <= N
    axes[0].plot(xz, partial_sum(k[m], ck[m], xz), c=c, lw=1, label=f"S_{N}")
m = np.abs(k) <= 250
axes[0].plot(xz, partial_sum(k[m], ck[m], xz, fejer_weights(k[m], 250)), "r--", lw=1.5, label="Fejér-Mittel σ_250")
axes[0].axhline(2 / np.pi * special.sici(np.pi)[0], color="grey", ls=":", lw=1, label="Gibbs-Grenze 1.179")
axes[0].set_xlabel("x")
axes[0].set_ylabel("Wert")
axes[0].set_title("Überschwingen am Sprung wandert, bleibt aber gleich hoch", fontsize=10)
axes[0].legend(fontsize=7)
xs = np.linspace(-np.pi, np.pi, 4001)
N = 10
with np.errstate(invalid="ignore", divide="ignore"):
    D = np.where(np.abs(xs) < 1e-12, 2 * N + 1, np.sin((N + 0.5) * xs) / np.sin(xs / 2))
    F = np.where(np.abs(xs) < 1e-12, N + 1, (np.sin((N + 1) * xs / 2) / np.sin(xs / 2)) ** 2 / (N + 1))
axes[1].plot(xs, D, label=f"Dirichlet-Kern D_{N} (S_N = f ∗ D_N)")
axes[1].plot(xs, F, label=f"Fejér-Kern F_{N} (σ_N = f ∗ F_N), ≥ 0")
axes[1].axhline(0, color="grey", lw=0.5)
axes[1].set_xlabel("x")
axes[1].set_title(f"Die beiden Kerne für N = {N}", fontsize=10)
axes[1].legend(fontsize=8)
Ns = np.unique(np.round(np.geomspace(5, 1000, 25)).astype(int))
over_S, over_F, leb = [], [], []
for N in Ns:
    xf = np.linspace(1e-6, 6 * np.pi / (N + 1), 3000)
    m = np.abs(k) <= N
    over_S.append(partial_sum(k[m], ck[m], xf).max())
    over_F.append(partial_sum(k[m], ck[m], xf, fejer_weights(k[m], N)).max())
    xl = np.linspace(-np.pi, np.pi, 200_001)[1:]
    with np.errstate(invalid="ignore", divide="ignore"):
        leb.append(np.mean(np.abs(np.sin((N + 0.5) * xl) / np.sin(xl / 2))))
axes[2].semilogx(Ns, over_S, "o-", ms=3, label="max S_N (nahe am Sprung)")
axes[2].semilogx(Ns, over_F, "s-", ms=3, label="max σ_N (Fejér)")
axes[2].axhline(2 / np.pi * special.sici(np.pi)[0], color="k", ls="--", lw=1, label="(2/π)·Si(π) = 1.179")
axes[2].set_xlabel("N")
axes[2].set_ylabel("Maximum der Näherung")
axes[2].set_title("Das Überschwingen verschwindet nicht", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

gibbs = 2 / np.pi * special.sici(np.pi)[0]
print(f"Maximum von S_N bei N = {Ns[0]}, {Ns[12]}, {Ns[-1]}: {over_S[0]:.4f}, {over_S[12]:.4f}, {over_S[-1]:.4f}  (Grenzwert (2/π)Si(π) = {gibbs:.5f})")
print(f"Überschwingen in Prozent des Sprungs 2: {(over_S[-1] - 1) / 2:.2%}  (Theorie {(gibbs - 1) / 2:.2%});  Fejér-Maximum höchstens {max(over_F):.4f}")
print(f"Lebesgue-Konstanten (1/2π)∫|D_N| bei N = {Ns[0]}, {Ns[12]}, {Ns[-1]}: {leb[0]:.3f}, {leb[12]:.3f}, {leb[-1]:.3f}  "
      f"(Näherung (4/π²) ln N + 1.27: {4 / np.pi ** 2 * np.log(Ns[0]) + 1.27:.3f}, {4 / np.pi ** 2 * np.log(Ns[12]) + 1.27:.3f}, {4 / np.pi ** 2 * np.log(Ns[-1]) + 1.27:.3f})")
```

* Im **linken Bild** liegt das Überschwingen von $S_{10}$, $S_{50}$ und $S_{250}$ immer näher am Sprung, aber immer auf derselben Höhe von etwa 1.18 (gepunktete Linie). Das Fejér-Mittel $\sigma_{250}$ (rot gestrichelt) steigt ohne Überschwingen an, dafür flacher: Der Übergang ist breiter. Das **mittlere Bild** zeigt den Grund: Der Dirichlet-Kern hat negative Nebenschwinger, der Fejér-Kern ist überall positiv.
* Im **rechten Bild** bleibt das Maximum von $S_N$ bei 1.179, für alle $N$. Laut **Ausgabe**: 1.1884 ($N = 5$), 1.1790 ($N = 71$), 1.1790 ($N = 1000$), Grenzwert $1.17898$; das Überschwingen ist $8.95\,\%$ der Sprunghöhe 2, genau wie vorhergesagt. Mehr Frequenzen helfen also nicht. Das Fejér-Mittel bleibt in derselben Umgebung des Sprungs unter 1 (höchstens 0.966). Die Lebesgue-Konstanten wachsen logarithmisch: 1.961, 3.001, 4.070 gegen $\frac4{\pi^2}\ln N + 1.27 = 1.922$, $2.998$, $4.070$.
* **Wo es vorkommt:** **Ringing-Artefakte** an Kanten in komprimierten Bildern (JPEG) und als Vorecho in Audiodateien (MP3), **Abbruchartefakte** in der Magnetresonanztomographie an scharfen Gewebegrenzen, Überschwingen von Rechtecksignalen in bandbegrenzten Leitungen und von idealen Tiefpassfiltern, Spektralverfahren mit Stoßwellen. Die Abhilfe ist immer dieselbe wie bei Fejér: sanft gewichten statt hart abschneiden (Hann-Filter, Lanczos-Faktoren, Lanczos-Interpolation beim Skalieren von Bildern).

##### *3. Parseval und das Basel-Problem*

Wie viel Energie steckt in jeder Frequenz, und was kostet das Weglassen von Frequenzen? Vorher vorhersagen: Welcher Anteil der Energie eines Rechtecksignals steckt im Grundton $k = 1$?

Für Sägezahn $x$, Parabel $x^2$ und Rechteck wird der mittlere quadratische Fehler $\Vert f - S_N\Vert^2$ numerisch integriert (auf $2^{18}$ Punkten) und mit der Energie der weggelassenen Koeffizienten verglichen (links). In der Mitte stehen die Partialsummen von $\sum1/k^2$ und $\sum1/k^4$ im Vergleich mit $\pi^2/6$ und $\pi^4/90$, die aus der Parseval-Gleichung folgen. Rechts steht, welchen Anteil der Energie die einzelnen Harmonischen des Rechtecks tragen. Die Ausgabe vergleicht außerdem den Fehler der Partialsumme mit dem des Fejér-Mittels.

**Hintergrund & Fachbegriffe:**
- **Energie** (Leistung) einer periodischen Funktion: $\Vert f\Vert^2 = \frac1{2\pi}\int_{-\pi}^\pi\vert f\vert^2\,\mathrm dx$, der Mittelwert von $\vert f\vert^2$.
- **Parseval-Gleichung:** $\Vert f\Vert^2 = \sum_k\vert c_k\vert^2$.
- **Harmonische:** die Schwingung mit Frequenz $k$; $k = 1$ ist der **Grundton**, die höheren sind die **Obertöne**.
- **Basel-Problem:** der Wert von $\zeta(2) = \sum_{k\geq1}1/k^2$, gelöst von Euler 1734. Die **Riemannsche Zeta-Funktion** ist $\zeta(s) = \sum_{k\geq1}k^{-s}$.
- **Orthogonale Projektion:** die nächstgelegene Näherung in einem Unterraum; der Fehler steht senkrecht auf dem Unterraum.
- **Polygamma-Funktion** $\psi^{(m)}$: liefert Summen der Form $\sum_{k\geq0}(k + a)^{-(m + 1)}$ in geschlossener Form; der Code berechnet damit die Energie der weggelassenen Koeffizienten exakt.

**Theorie:**

- **Pythagoras im Funktionenraum:** Weil die $e^{ikx}$ orthonormal sind, ist $S_Nf$ die orthogonale Projektion von $f$ auf die trigonometrischen Polynome vom Grad $\leq N$. Daher gilt $\Vert f\Vert^2 = \Vert S_Nf\Vert^2 + \Vert f - S_Nf\Vert^2$ und $\Vert f - S_Nf\Vert^2 = \sum_{\vert k\vert > N}\vert c_k\vert^2$. Keine andere Kombination dieser Frequenzen, auch nicht das Fejér-Mittel, hat einen kleineren quadratischen Fehler.
- **Bessel und Parseval:** Für jede Orthonormalfamilie gilt $\sum\vert c_k\vert^2\leq\Vert f\Vert^2$ (Bessel). Weil die $e^{ikx}$ eine vollständige Basis von $L^2$ sind, gilt Gleichheit (Parseval), und $S_Nf\to f$ im quadratischen Mittel für jede $L^2$-Funktion, auch für das Rechteck, das nicht gleichmäßig konvergiert. Umgekehrt gehört zu jeder Folge mit $\sum\vert c_k\vert^2 < \infty$ eine $L^2$-Funktion (Riesz-Fischer).
- **Basel-Problem:** Der Sägezahn $f(x) = x$ hat $c_k = i(-1)^k/k$. Parseval gibt $\frac1{2\pi}\int x^2 = \frac{\pi^2}3 = \sum_{k\neq0}\frac1{k^2} = 2\zeta(2)$, also $\zeta(2) = \pi^2/6$. Die Parabel $x^2$ mit $c_0 = \pi^2/3$ und $c_k = 2(-1)^k/k^2$ gibt $\frac{\pi^4}5 = \frac{\pi^4}9 + 8\zeta(4)$, also $\zeta(4) = \pi^4/90$.
- **Rechteck:** Energie 1, $\vert c_k\vert^2 = 4/(\pi^2k^2)$ für ungerade $k$; die Harmonische $k$ (mit $\pm k$) trägt $8/(\pi^2k^2)$, der Grundton also $8/\pi^2 = 81\,\%$.

```python
xg = np.linspace(-np.pi, np.pi, 8193)
sawtooth = lambda x: x * (np.abs(x) < np.pi)
parabola = lambda x: x ** 2
M = 2 ** 18
xm = -np.pi + 2 * np.pi * np.arange(M) / M                        # feines periodisches Gitter für die Fehlerintegrale

def partial_sum_grid(k, c):
    """Partialsumme auf dem Gitter xm über eine inverse FFT (schnell auch für viele Punkte)."""
    C = np.zeros(M, dtype=complex)
    C[k % M] = c * (-1.0) ** k
    return np.real(np.fft.ifft(C) * M)

def tail(name, N):
    """Exakte Energie der weggelassenen Koeffizienten Σ_{|k|>N} |c_k|² (über die Polygamma-Funktion)."""
    if name.startswith("Säge"):
        return 2 * special.polygamma(1, N + 1)
    if name.startswith("Parabel"):
        return 8 * special.polygamma(3, N + 1) / 6
    m = N + 1 if N % 2 == 0 else N + 2                            # erste ungerade Zahl über N
    return 8 / np.pi ** 2 * special.polygamma(1, m / 2) / 4

tests = [("Sägezahn x", sawtooth), ("Parabel x²", parabola), ("Rechteck", square)]
Ns = np.unique(np.round(np.geomspace(1, 200, 20)).astype(int))

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
for (name, f), c in zip(tests, ["tab:blue", "tab:orange", "tab:green"]):
    k, ck = fourier_coefficients(f, 200, M=M)
    l2 = [np.mean((f(xm) - partial_sum_grid(k[np.abs(k) <= N], ck[np.abs(k) <= N])) ** 2) for N in Ns]
    axes[0].loglog(Ns, l2, "o", c=c, ms=4, label=f"{name}: gemessen")
    axes[0].loglog(Ns, [tail(name, N) for N in Ns], "-", c=c, lw=1, label=f"{name}: Σ_|k|>N |c_k|² exakt")
axes[0].set_xlabel("N")
axes[0].set_ylabel("mittlerer quadratischer Fehler")
axes[0].set_title("L²-Fehler = Energie der weggelassenen Koeffizienten", fontsize=10)
axes[0].legend(fontsize=7)
n = np.arange(1, 10_001)
axes[1].loglog(n, np.pi ** 2 / 6 - np.cumsum(1 / n ** 2), label="π²/6 − Σ_{k≤n} 1/k²")
axes[1].loglog(n, np.pi ** 4 / 90 - np.cumsum(1 / n ** 4.0), label="π⁴/90 − Σ_{k≤n} 1/k⁴")
axes[1].loglog(n, 1 / n, "k:", lw=1, label="1/n")
axes[1].loglog(n, 1 / (3 * n ** 3.0), "k--", lw=1, label="1/(3n³)")
axes[1].set_ylim(1e-13, 1)
axes[1].set_xlabel("Zahl der Summanden n")
axes[1].set_ylabel("Abstand zum Grenzwert")
axes[1].set_title("Aus Parseval: ζ(2) = π²/6 und ζ(4) = π⁴/90", fontsize=10)
axes[1].legend(fontsize=8)
k, ck = fourier_coefficients(square, 400)
odd_k = np.arange(1, 30, 2)
share = 2 * np.abs(ck[400 + odd_k]) ** 2
axes[2].bar(odd_k, share, width=1.2, alpha=0.7, label="Anteil der Harmonischen k an der Energie")
axes[2].plot(odd_k, np.cumsum(share), "ko-", ms=3, label="Summe bis k")
axes[2].set_xlabel("Harmonische k (nur ungerade)")
axes[2].set_ylabel("Anteil an (1/2π)∫f² = 1")
axes[2].set_title("Rechteck: Energie in den Harmonischen", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

k, ck = fourier_coefficients(sawtooth, 400)
print(f"Sägezahn: (1/2π)∫x² = {integrate.trapezoid(xg ** 2, xg) / (2 * np.pi):.6f} = π²/3 = {np.pi ** 2 / 3:.6f};  Σ|c_k|² über |k| ≤ 400 = {np.sum(np.abs(ck) ** 2):.6f}, "
      f"fehlender Rest Σ_|k|>400 1/k² = {2 * special.polygamma(1, 401):.6f}")
print(f"Daraus ζ(2) = π²/6 = {np.pi ** 2 / 6:.6f};  Σ 1/k² bis 10 000 = {np.sum(1 / n ** 2):.6f}")
print(f"Daraus ζ(4) = π⁴/90 = {np.pi ** 4 / 90:.8f};  Σ 1/k⁴ bis 10 000 = {np.sum(1 / n ** 4.0):.8f}")
print(f"Rechteck: Grundton trägt {share[0]:.4f} der Energie (Theorie 8/π² = {8 / np.pi ** 2:.4f}), bis k = 5: {np.cumsum(share)[2]:.4f}")
k, ck = fourier_coefficients(square, 400)
m = np.abs(k) <= 10
e_S = np.mean((square(xm) - partial_sum_grid(k[m], ck[m])) ** 2)
e_F = np.mean((square(xm) - partial_sum_grid(k[m], ck[m] * fejer_weights(k[m], 10))) ** 2)
print(f"Rechteck, N = 10: L²-Fehler der Partialsumme {e_S:.4f} (exakt {tail('Rechteck', 10):.4f}), des Fejér-Mittels {e_F:.4f} (die Partialsumme ist die beste L²-Näherung)")
```

##### *4. Fourier-Transformation: Paare und Eigenfunktionen*

Wie sehen die Fourier-Transformierten typischer Funktionen aus, und gibt es Funktionen, die ihre eigene Transformierte sind? Vorher vorhersagen: Welche Funktion bleibt unter der Fourier-Transformation unverändert?

Drei Funktionen auf der Geraden (Gauß $e^{-\pi x^2}$, Rechteck auf $[-\tfrac12, \tfrac12]$, Laplace $e^{-\vert x\vert}$) werden auf einem Gitter von $2^{14}$ Punkten in $[-20, 20)$ mit der FFT transformiert und mit den exakten Transformierten verglichen (links und Mitte). Rechts werden die Hermite-Funktionen $h_0$ bis $h_3$ transformiert; die Punkte zeigen $\hat h_n/(-i)^n$, das genau auf $h_n$ liegen soll. Die Ausgabe gibt die Fehler der Näherung, prüft Plancherel und berechnet die Eigenwerte $\langle h_n, \hat h_n\rangle$ für $n = 0$ bis 7.

**Hintergrund & Fachbegriffe:**
- **Fourier-Transformation** $\hat f(\xi) = \int f(x)e^{-2\pi ix\xi}\,\mathrm dx$ und **Rücktransformation** $f(x) = \int\hat f(\xi)e^{2\pi ix\xi}\,\mathrm d\xi$.
- **sinc-Funktion:** $\mathrm{sinc}(\xi) = \frac{\sin(\pi\xi)}{\pi\xi}$, mit Nullstellen bei allen ganzen Zahlen außer 0.
- **Lorentz-Kurve:** $\frac2{1 + 4\pi^2\xi^2}$; die typische Form einer Spektrallinie.
- **Hermite-Polynome** $H_n$: Polynome vom Grad $n$ ($H_0 = 1$, $H_1 = 2x$, $H_2 = 4x^2 - 2$, …); **Hermite-Funktionen** sind $h_n(x)\propto H_n(\sqrt{2\pi}\,x)\,e^{-\pi x^2}$.
- **Eigenfunktion:** eine Funktion, die eine Abbildung nur mit einer Zahl, dem **Eigenwert**, multipliziert.
- **Näherung per FFT:** Das Integral wird zur Riemann-Summe $\Delta x\sum_jf(x_j)e^{-2\pi ix_j\xi}$, und das ist an den Frequenzen $\xi_m = m/(N\Delta x)$ eine DFT.

**Theorie:**

- **Drei Paare:** $e^{-\pi x^2}\mapsto e^{-\pi\xi^2}$; mit der Streckungsregel $e^{-\pi(x/a)^2}\mapsto a\,e^{-\pi(a\xi)^2}$. Das Rechteck gibt $\int_{-1/2}^{1/2}e^{-2\pi ix\xi}\,\mathrm dx = \mathrm{sinc}(\xi)$; seine Sprünge machen den langsamen Abfall $1/\xi$. $e^{-\vert x\vert}\mapsto\frac2{1 + 4\pi^2\xi^2}$; der Knick bei 0 gibt den Abfall $1/\xi^2$. Wie bei den Reihen (Exp. 1) bestimmt die Glattheit den Abfall.
- **Vierfache Anwendung:** Zweimal transformieren spiegelt, $\hat{\hat f}(x) = f(-x)$; viermal ergibt wieder $f$. Deshalb kann die Fourier-Transformation nur die Eigenwerte $1, -i, -1, i$ haben (die vierten Einheitswurzeln).
- **Hermite-Funktionen:** $\hat h_n = (-i)^nh_n$. Sie bilden eine Orthonormalbasis von $L^2(\mathbb R)$, in der die Fourier-Transformation diagonal ist. In der Quantenmechanik sind sie die Eigenzustände des harmonischen Oszillators, die in Orts- und Impulsdarstellung gleich aussehen. Ersetzt man $(-i)^n = e^{-in\pi/2}$ durch $e^{-in\alpha}$, erhält man die **fraktionale Fourier-Transformation**, eine Drehung um den Winkel $\alpha$ in der Ebene aus Ort und Frequenz.
- **Fehler der FFT-Näherung:** Abschneiden (die Funktion muss am Rand des Gitters klein sein) und Aliasing (die Transformierte muss oberhalb der Nyquist-Frequenz $1/(2\Delta x)$ klein sein, Exp. 9).

```python
x = np.linspace(-20, 20, 2 ** 14, endpoint=False)
pairs = [
    ("Gauß e^(−πx²)", np.exp(-np.pi * x ** 2), lambda s: np.exp(-np.pi * s ** 2)),
    ("Rechteck auf [−½, ½]", (np.abs(x) < 0.5).astype(float), lambda s: np.sinc(s)),
    ("Laplace e^(−|x|)", np.exp(-np.abs(x)), lambda s: 2 / (1 + 4 * np.pi ** 2 * s ** 2)),
]

def hermite_function(n, x):
    """Hermite-Funktion H_n(√(2π) x) e^{−πx²}, normiert; Eigenfunktion der Fourier-Transformation."""
    h = special.eval_hermite(n, np.sqrt(2 * np.pi) * x) * np.exp(-np.pi * x ** 2)
    return h / np.sqrt(np.sum(h ** 2) * (x[1] - x[0]))

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
for (name, fx, fhat), c in zip(pairs, ["tab:blue", "tab:orange", "tab:green"]):
    axes[0].plot(x, fx, c=c, label=name)
    xi, F = continuous_ft(fx, x)
    sel = np.abs(xi) <= 3
    axes[1].plot(xi[sel][::6], F.real[sel][::6], "o", c=c, ms=3)
    axes[1].plot(xi[sel], fhat(xi[sel]), "-", c=c, lw=1, label=f"{name} → exakt")
axes[0].set_xlim(-3, 3)
axes[0].set_xlabel("x")
axes[0].set_ylabel("f(x)")
axes[0].set_title("Drei Funktionen", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].axhline(0, color="grey", lw=0.5)
axes[1].set_xlabel("Frequenz ξ")
axes[1].set_ylabel("f̂(ξ)")
axes[1].set_title("Transformierte: Punkte per FFT, Linien exakt", fontsize=10)
axes[1].legend(fontsize=8)
for n, c in zip(range(4), ["tab:blue", "tab:orange", "tab:green", "tab:red"]):
    h = hermite_function(n, x)
    xi, H = continuous_ft(h, x)
    axes[2].plot(x, h, c=c, lw=1.2, label=f"h_{n}, Eigenwert (−i)^{n} = {['1', '−i', '−1', 'i'][n]}")
    back = np.interp(x, xi, (H / (-1j) ** n).real)
    axes[2].plot(x[::160], back[::160], "o", c=c, ms=3)
axes[2].set_xlim(-2.5, 2.5)
axes[2].set_xlabel("x bzw. ξ")
axes[2].set_title("Hermite-Funktionen: ĥ_n = (−i)ⁿ h_n (Punkte: ĥ_n / (−i)ⁿ)", fontsize=10)
axes[2].legend(fontsize=7)
plt.tight_layout()
plt.show()

for name, fx, fhat in pairs:
    xi, F = continuous_ft(fx, x)
    sel = np.abs(xi) <= 3
    print(f"{name:22s}: größter Fehler der FFT-Näherung für |ξ| ≤ 3: {np.max(np.abs(F[sel] - fhat(xi[sel]))):.1e};  "
          f"Plancherel ∫|f|² = {np.sum(np.abs(fx) ** 2) * (x[1] - x[0]):.6f}, ∫|f̂|² = {np.sum(np.abs(F) ** 2) * (xi[1] - xi[0]):.6f}")
lam = []
for n in range(8):
    h = hermite_function(n, x)
    xi, H = continuous_ft(h, x)
    lam.append(np.sum(np.conj(h) * np.interp(x, xi, H.real) + 1j * np.conj(h) * np.interp(x, xi, H.imag)) * (x[1] - x[0]))
print("Eigenwerte ⟨h_n, ĥ_n⟩ für n = 0 … 7: " + ", ".join(f"{z.real:+.3f}{z.imag:+.3f}i" for z in lam))
```

* Das **linke Bild** zeigt die drei Funktionen, das **mittlere** ihre Transformierten: Punkte aus der FFT, Linien exakt. Gauß bleibt Gauß, das Rechteck wird zur sinc-Funktion mit negativen Nebenschwingern, der Laplace-Puls zur schmalen Lorentz-Kurve mit Höhe 2. Laut **Ausgabe** ist die FFT-Näherung beim Gauß auf $5\cdot10^{-15}$ genau, beim Laplace-Puls auf $10^{-6}$ (Knick), beim Rechteck nur auf $1.5\cdot10^{-3}$, weil die Kanten bei $\pm\frac12$ nicht genau auf Gitterpunkten liegen. Plancherel gilt in allen drei Fällen auf sechs Stellen ($0.707107$, $0.998535$, $1.000002$ auf beiden Seiten).
* Im **rechten Bild** liegen die Punkte $\hat h_n/(-i)^n$ genau auf den Kurven $h_n$: Die Hermite-Funktionen sind Eigenfunktionen. Die Ausgabe gibt die Eigenwerte $+1$, $-i$, $-1$, $+i$, $+1$, $-i$, … für $n = 0$ bis 7, mit Beträgen zwischen 0.998 und 1.000 (die kleine Abweichung kommt aus der Interpolation vom gröberen Frequenzgitter). Die Antwort auf die Vorhersage: die Gauß-Funktion, und allgemein jede Hermite-Funktion mit $n = 0, 4, 8, \dots$ sowie ihre Summen.
* **Wo es vorkommt:** In der **Quantenmechanik** ist die Wellenfunktion im Impulsraum die Fourier-Transformierte der Wellenfunktion im Ortsraum. In der **Spektroskopie** ist eine Lorentz-Linie die Transformierte eines exponentiell abklingenden Signals (natürliche Linienbreite; in der NMR wird der gemessene Zerfall, der FID, per FFT zum Spektrum). Die sinc-Funktion ist das Spektrum eines Rechteckpulses (Radar, Unterträger im OFDM-Mobilfunk). Gauß-Strahlen beschreiben Laser, und die fraktionale Fourier-Transformation beschreibt die Ausbreitung von Licht durch Linsensysteme.

##### *5. Unschärfeprinzipien*

Kann eine Funktion zugleich im Ort und in der Frequenz schmal sein? Vorher vorhersagen: Welche Funktion ist in beiden zugleich am schmalsten, und was passiert mit der Frequenzbreite, wenn man einer Gauß-Funktion eine wachsende Frequenz (Chirp) aufprägt?

Gemessen wird die Breite $\Delta x$ von $\vert f\vert^2$ und die Breite $\Delta\xi$ von $\vert\hat f\vert^2$ für vier Funktionen (links), für 300 zufällige Summen aus je drei Gaußschen Wellenpaketen und für Gauß-Funktionen verschiedener Breite (Mitte). Rechts geht es um die diskrete Version auf $\mathbb Z_{64}$: Gezählt wird, wie viele Einträge eines Vektors und seiner DFT ungleich 0 sind, für Dirac-Kämme, für Summen aus ein bis drei verschobenen und modulierten Kämmen und für zufällige dünne Vektoren.

**Hintergrund & Fachbegriffe:**
- **Breite** $\Delta x$: die Standardabweichung der Verteilung $\vert f(x)\vert^2/\int\vert f\vert^2$; ebenso $\Delta\xi$ für $\vert\hat f\vert^2$.
- **Chirp:** ein Signal, dessen Frequenz sich mit der Zeit ändert; hier $e^{-\pi x^2(1 + 2i)}$, eine Gauß-Hülle mit linear wachsender Frequenz.
- **Wellenpaket:** eine Schwingung unter einer Gauß-Hülle, $e^{-\pi((x - x_0)/a)^2}e^{2\pi i\nu x}$.
- **Träger** (support): die Menge der Stellen, an denen eine Funktion oder ein Vektor nicht 0 ist; $\vert\mathrm{supp}\,f\vert$ ist ihre Anzahl.
- **Dirac-Kamm:** ein Vektor mit Einsen in festem Abstand $d$ und Nullen dazwischen.

**Theorie:**

- **Heisenberg-Weyl-Ungleichung:** Für jede Funktion gilt $\Delta x\cdot\Delta\xi\geq\frac1{4\pi}$. Idee des Beweises (für $f$ um 0 zentriert, $\int\vert f\vert^2 = 1$): Partielle Integration gibt $1 = -2\,\mathrm{Re}\int xf\,\overline{f'}$, Cauchy-Schwarz $1\leq2\Vert xf\Vert\,\Vert f'\Vert$, und nach Plancherel ist $\Vert f'\Vert = 2\pi\Vert\xi\hat f\Vert$. Zusammen $1\leq4\pi\,\Delta x\,\Delta\xi$. Gleichheit gilt genau dann, wenn $f'$ proportional zu $xf$ ist: für Gauß-Funktionen (auch verschoben und moduliert).
- **Werte für die Beispiele:** Dreieck $\Delta x\,\Delta\xi = \sqrt{0.1}\cdot\frac{\sqrt3}{2\pi} = 1.095\cdot\frac1{4\pi}$, Laplace $\frac1{\sqrt2}\cdot\frac1{2\pi} = \sqrt2\cdot\frac1{4\pi}$. Ein Chirp $e^{-\pi x^2(1 + ic)}$ hat dasselbe $\Delta x$ wie die Gauß-Funktion, aber ein um $\sqrt{1 + c^2}$ breiteres Spektrum.
- **Quantenmechanik:** Mit dem Impuls $p = h\xi$ wird daraus $\Delta x\,\Delta p\geq\hbar/2$. Die Unschärferelation ist also eine Eigenschaft der Fourier-Transformation, keine zusätzliche Annahme.
- **Donoho-Stark** (1989): Für jeden Vektor $f\neq0$ in $\mathbb C^N$ gilt $\vert\mathrm{supp}\,f\vert\cdot\vert\mathrm{supp}\,\hat f\vert\geq N$. Gleichheit erreichen die Dirac-Kämme: Ein Kamm mit Abstand $d$ (bei $d$ Teiler von $N$) hat $N/d$ Einträge, seine DFT ist ein Kamm mit $d$ Einträgen. Ist $N$ eine Primzahl, gibt es keine solchen Kämme, und es gilt sogar $\vert\mathrm{supp}\,f\vert + \vert\mathrm{supp}\,\hat f\vert\geq N + 1$ (Tao 2005).
- **Hardy:** $f$ und $\hat f$ können nicht beide schneller als eine Gauß-Funktion abfallen. Unschärfeprinzipien sind auch die Grundlage des **Compressed Sensing**: Ein dünnbesetztes Signal lässt sich aus wenigen zufälligen Fourier-Messungen rekonstruieren (Candès, Romberg, Tao 2006).

```python
rng = np.random.default_rng(5)
x = np.linspace(-40, 40, 2 ** 15, endpoint=False)
bound = 1 / (4 * np.pi)
shapes = {
    "Gauß": np.exp(-np.pi * x ** 2),
    "Dreieck": np.clip(1 - np.abs(x), 0, None),
    "Laplace e^(−|x|)": np.exp(-np.abs(x)),
    "Gauß mit Chirp": np.exp(-np.pi * x ** 2 * (1 + 2j)),
}
theory = {"Gauß": 1.0, "Dreieck": np.sqrt(0.1) * np.sqrt(3) / (2 * np.pi) / bound, "Laplace e^(−|x|)": np.sqrt(0.5) / (2 * np.pi) / bound, "Gauß mit Chirp": np.sqrt(5)}
products = {}
for name, fx in shapes.items():
    xi, F = continuous_ft(fx, x)
    products[name] = spread(fx, x) * spread(F, xi)

def wave_packets(rng):
    """Summe von drei Gaußschen Wellenpaketen mit zufälliger Lage, Breite und Frequenz."""
    f = np.zeros_like(x, dtype=complex)
    for _ in range(3):
        a, x0, nu = rng.uniform(0.3, 3), rng.uniform(-4, 4), rng.uniform(-2, 2)
        f += rng.normal() * np.exp(-np.pi * ((x - x0) / a) ** 2 + 2j * np.pi * nu * x)
    return f

random_pts = []
for _ in range(300):
    fx = wave_packets(rng)
    xi, F = continuous_ft(fx, x)
    random_pts.append((spread(fx, x), spread(F, xi)))
random_pts = np.array(random_pts)
gauss_pts = []
for a in np.geomspace(0.2, 5, 9):
    fx = np.exp(-np.pi * (x / a) ** 2)
    xi, F = continuous_ft(fx, x)
    gauss_pts.append((spread(fx, x), spread(F, xi)))
gauss_pts = np.array(gauss_pts)

N = 64
ds_random = []                                                    # zufällige dünne Vektoren: fast immer volle DFT
for _ in range(400):
    s = rng.integers(1, N + 1)
    f = np.zeros(N, dtype=complex)
    f[rng.choice(N, s, replace=False)] = rng.normal(size=s) + 1j * rng.normal(size=s)
    ds_random.append((s, np.sum(np.abs(np.fft.fft(f)) > 1e-9)))
ds_random = np.array(ds_random)
ds_struct = []                                                    # Summen von 1–3 verschobenen, modulierten Kämmen
for _ in range(400):
    f = np.zeros(N, dtype=complex)
    for _ in range(rng.integers(1, 4)):
        d = rng.choice([2, 4, 8, 16, 32])
        comb = np.zeros(N)
        comb[rng.integers(d)::d] = 1
        f += rng.normal() * comb * np.exp(2j * np.pi * rng.integers(N) * np.arange(N) / N)
    ds_struct.append((np.sum(np.abs(f) > 1e-9), np.sum(np.abs(np.fft.fft(f)) > 1e-9)))
ds_struct = np.array(ds_struct)
combs = []
for d in [1, 2, 4, 8, 16, 32, 64]:
    f = np.zeros(N)
    f[::d] = 1
    combs.append((np.sum(f != 0), np.sum(np.abs(np.fft.fft(f)) > 1e-9)))
combs = np.array(combs)

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
names = list(products)
axes[0].bar(names, [products[n] / bound for n in names], alpha=0.7, label="gemessen")
axes[0].plot(names, [theory[n] for n in names], "ko", label="Theorie")
axes[0].axhline(1, color="tab:red", ls="--", label="Untergrenze 1/(4π)")
axes[0].set_ylabel("Δx · Δξ in Einheiten von 1/(4π)")
axes[0].set_title("Nur die Gauß-Funktion erreicht die Grenze", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].loglog(random_pts[:, 0], random_pts[:, 1], ".", ms=4, alpha=0.6, label="300 zufällige Wellenpakete")
axes[1].loglog(gauss_pts[:, 0], gauss_pts[:, 1], "o", c="tab:red", ms=5, label="Gauß-Funktionen")
dxs = np.geomspace(0.03, 10, 50)
axes[1].loglog(dxs, bound / dxs, "k--", lw=1, label="Δx · Δξ = 1/(4π)")
axes[1].set_xlabel("Breite Δx")
axes[1].set_ylabel("Breite Δξ")
axes[1].set_title("Kein Signal liegt unter der Hyperbel", fontsize=10)
axes[1].legend(fontsize=8)
axes[2].plot(ds_struct[:, 0], ds_struct[:, 1], ".", ms=5, alpha=0.5, label="Summen von 1–3 Kämmen (verschoben, moduliert)")
axes[2].plot(ds_random[:, 0], ds_random[:, 1], "x", c="grey", ms=3, alpha=0.5, label="zufällige dünne Vektoren")
axes[2].plot(combs[:, 0], combs[:, 1], "o", c="tab:red", ms=6, label="Dirac-Kämme (Gleichheit)")
ss = np.linspace(1, N, 200)
axes[2].plot(ss, N / ss, "k--", lw=1, label=f"|supp f| · |supp f̂| = N = {N}")
axes[2].set_xscale("log", base=2)
axes[2].set_yscale("log", base=2)
axes[2].set_xlabel("Zahl der Einträge ≠ 0 in f")
axes[2].set_ylabel("Zahl der Einträge ≠ 0 in der DFT")
axes[2].set_title("Donoho-Stark auf ℤ_64", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

for n in names:
    print(f"{n:17s}: Δx·Δξ = {products[n]:.4f} = {products[n] / bound:.3f} × 1/(4π)  (Theorie {theory[n]:.3f})")
print(f"Zufällige Wellenpakete: kleinstes Produkt {np.min(random_pts[:, 0] * random_pts[:, 1]) / bound:.3f} × 1/(4π)")
print(f"Donoho-Stark: kleinstes |supp f|·|supp f̂| bei den Kammsummen {np.min(ds_struct[:, 0] * ds_struct[:, 1])}, bei den reinen Kämmen {sorted(set(combs[:, 0] * combs[:, 1]))}; "
      f"zufällige dünne Vektoren mit voller DFT (64 Einträge ≠ 0): {np.mean(ds_random[:, 1] == N):.0%}")
```

* Im **linken Bild** erreicht nur die Gauß-Funktion die Grenze $1/(4\pi)$; laut **Ausgabe** 1.000, Dreieck 1.096 (Theorie 1.095), Laplace 1.415 (1.414), Gauß mit Chirp 2.236 ($\sqrt5$). Der Chirp ist im Ort genauso schmal wie die Gauß-Funktion, aber im Spektrum breiter, weil er viele Frequenzen nacheinander durchläuft. Das beantwortet den zweiten Teil der Vorhersage.
* Im **mittleren Bild** (doppelt logarithmisch) liegen alle 300 zufälligen Wellenpakete oberhalb der Hyperbel $\Delta x\,\Delta\xi = 1/(4\pi)$, das kleinste Produkt ist 1.019-mal die Grenze. Die Gauß-Funktionen liegen alle auf der Hyperbel: Je schmaler im Ort, desto breiter in der Frequenz. Im **rechten Bild** liegen die Summen aus Kämmen alle auf oder über der Hyperbel $\vert\mathrm{supp}\,f\vert\cdot\vert\mathrm{supp}\,\hat f\vert = 64$, die reinen Kämme genau darauf (Produkt 64). Zufällige dünne Vektoren (grau) haben dagegen fast nie eine dünne DFT: Alle hatten 64 Einträge ungleich 0. Wenige Einträge in beiden Bereichen verlangen eine Kammstruktur.
* **Wo es vorkommt:** die **Heisenbergsche Unschärferelation**; die **Gabor-Grenze** der Nachrichtentechnik (ein kurzer Puls braucht viel Bandbreite, deshalb begrenzt die Bandbreite eines Kanals die Datenrate); **Chirp-Radar** und Fledermausrufe, die mit langen Chirps zugleich Reichweite und Auflösung erreichen; die Auflösungsgrenze von Mikroskopen und Teleskopen (Exp. 14); in der MRT die Abwägung zwischen Messzeit (Umfang im Frequenzraum) und Bildschärfe; **Compressed Sensing** zum Beschleunigen von MRT-Aufnahmen.

##### *6. Der Faltungssatz*

Wie hängen Faltung und Fourier-Transformation zusammen, und wie viel spart die FFT beim Falten? Vorher vorhersagen: Um welchen Faktor ist die Faltung über die FFT bei zwei Folgen der Länge 32 768 schneller als die direkte Berechnung?

Ein verrauschtes Signal (Sprung plus Sinus plus normalverteiltes Rauschen, 2000 Werte) wird mit einem Gauß-Kern geglättet, einmal direkt und einmal über die FFT (links). In der Mitte stehen die Beträge der Spektren von Signal, Kern und Ergebnis. Rechts wird die Rechenzeit beider Wege für Längen von 64 bis 32 768 gemessen (bester von drei Läufen). Die Ausgabe prüft, dass beide Wege dasselbe ergeben, und zeigt, was ohne Auffüllen mit Nullen passiert.

**Hintergrund & Fachbegriffe:**
- **Faltung:** $(f\ast g)(x) = \int f(y)\,g(x - y)\,\mathrm dy$, diskret $(f\ast g)_n = \sum_mf_mg_{n - m}$: Jeder Wert des Ergebnisses ist ein mit $g$ gewichtetes Mittel von $f$ in seiner Umgebung.
- **Kern:** die Gewichtsfunktion $g$; ein Gauß-Kern glättet, er ist ein **Tiefpass**, der hohe Frequenzen dämpft.
- **Zyklische Faltung:** die Faltung auf $\mathbb Z_N$, bei der Indizes über das Ende hinaus wieder vorn beginnen. **Lineare Faltung:** ohne dieses Herumwickeln.
- **Auffüllen mit Nullen** (zero padding): Verlängern einer Folge mit Nullen, hier damit die zyklische Faltung der DFT zur linearen wird.
- **Rechenaufwand** $O(N^2)$ bzw. $O(N\log N)$: wie die Zahl der Rechenschritte mit $N$ wächst, bis auf einen konstanten Faktor.

**Theorie:**

- **Faltungssatz:** $\widehat{f\ast g} = \hat f\cdot\hat g$. Grund: Die Charaktere $e^{2\pi ix\xi}$ sind Eigenfunktionen jeder Verschiebung und damit jeder Faltung (Theorie, Abschnitt 4); die Faltung mit $g$ multipliziert die Frequenz $\xi$ mit $\hat g(\xi)$. Für die DFT gilt der Satz exakt für die zyklische Faltung.
- **Lineare Faltung mit der FFT:** Folgen der Längen $N$ und $M$ mit Nullen auf $L\geq N + M - 1$ verlängern, beide transformieren, multiplizieren, zurücktransformieren. Kosten drei FFTs, also $O(L\log L)$ statt $N\cdot M$.
- **Lineare zeitinvariante Systeme:** Jedes lineare System, das nicht von der absoluten Zeit abhängt (ein Filter, ein Raum für Schall, eine Optik), antwortet auf ein Signal mit dessen Faltung mit der **Impulsantwort** $h$; seine **Übertragungsfunktion** ist $\hat h$.
- **Entfalten** heißt durch $\hat g$ teilen. Wo $\hat g$ fast 0 ist (beim Gauß-Kern bei hohen Frequenzen), wird Rauschen ungeheuer verstärkt; das **Wiener-Filter** dämpft diese Frequenzen stattdessen.
- **Korrelation:** $\sum_mf_mg_{m + n}$ hat die Transformierte $\overline{\hat f}\,\hat g$. Faltungsschichten in neuronalen Netzen berechnen eigentlich Korrelationen.

```python
rng = np.random.default_rng(6)
n = 2000
t = np.arange(n) / n
clean = np.where(t > 0.5, 1.0, 0.0) + 0.5 * np.sin(2 * np.pi * 3 * t)
noisy = clean + 0.3 * rng.normal(size=n)
tk = np.arange(-200, 201) / n
kernel = np.exp(-tk ** 2 / (2 * 0.01 ** 2))
kernel /= kernel.sum()
direct = np.convolve(noisy, kernel, mode="same")
via_fft = signal.fftconvolve(noisy, kernel, mode="same")
L = n + len(kernel) - 1
circular = np.real(np.fft.ifft(np.fft.fft(noisy) * np.fft.fft(kernel, n)))           # ohne Auffüllen mit Nullen
linear = np.real(np.fft.ifft(np.fft.fft(noisy, L) * np.fft.fft(kernel, L)))            # mit Auffüllen

sizes = 2 ** np.arange(6, 16)
t_direct, t_fft = [], []
for m in sizes:
    a, b = rng.normal(size=m), rng.normal(size=m)
    for store, fun in [(t_direct, np.convolve), (t_fft, signal.fftconvolve)]:
        best = np.inf
        for _ in range(3):
            t0 = time.perf_counter(); fun(a, b); best = min(best, time.perf_counter() - t0)
        store.append(best)

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
axes[0].plot(t, noisy, lw=0.5, alpha=0.6, label="verrauschtes Signal")
axes[0].plot(t, direct, lw=1.5, label="gefaltet mit Gauß-Kern (σ = 0.01)")
axes[0].plot(t, clean, "k--", lw=1, label="ohne Rauschen")
axes[0].set_xlabel("t")
axes[0].set_title("Glätten = Falten mit einem Kern", fontsize=10)
axes[0].legend(fontsize=8)
f = np.fft.rfftfreq(n, 1 / n)
axes[1].semilogy(f, np.abs(np.fft.rfft(noisy)), lw=0.6, label="|Spektrum des Signals|")
axes[1].semilogy(f, np.abs(np.fft.rfft(kernel, n)) * 300, lw=1.5, label="|Spektrum des Kerns| × 300")
axes[1].semilogy(f, np.abs(np.fft.rfft(direct)), lw=0.8, label="|Spektrum des Ergebnisses|")
axes[1].set_xlim(0, 300)
axes[1].set_ylim(1e-3, 3e3)
axes[1].set_xlabel("Frequenz [Schwingungen pro Signallänge]")
axes[1].set_title("Im Spektrum: Faltung = Multiplikation", fontsize=10)
axes[1].legend(fontsize=8)
axes[2].loglog(sizes, t_direct, "o-", label="direkt, O(N²)")
axes[2].loglog(sizes, t_fft, "s-", label="über die FFT, O(N log N)")
axes[2].set_xlabel("Länge N beider Folgen")
axes[2].set_ylabel("Rechenzeit [s]")
axes[2].set_title("Rechenzeit der Faltung", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

print(f"Direkte Faltung gegen FFT-Faltung: größter Unterschied {np.max(np.abs(direct - via_fft)):.1e}")
lin_same = linear[len(kernel) // 2: len(kernel) // 2 + n]
print(f"FFT mit Auffüllen gegen direkt: {np.max(np.abs(lin_same - direct)):.1e};  ohne Auffüllen (zyklisch): {np.max(np.abs(np.roll(circular, -(len(kernel) // 2)) - direct)):.2f} (Fehler an den Rändern)")
print(f"Rechenzeit bei N = {sizes[-1]}: direkt {t_direct[-1] * 1e3:.1f} ms, FFT {t_fft[-1] * 1e3:.2f} ms, Verhältnis {t_direct[-1] / t_fft[-1]:.0f}")
```

* Im **linken Bild** entfernt die Faltung das Rauschen: Die geglättete Kurve (orange) folgt dem Signal ohne Rauschen (schwarz gestrichelt). Der Sprung bei $t = 0.5$ wird dabei auf die Breite des Kerns verschmiert, und am rechten Rand fällt die Kurve ab, weil die Faltung jenseits des Endes Nullen annimmt. Das **mittlere Bild** zeigt den Faltungssatz: Das Spektrum des Signals (blau) besteht aus wenigen tiefen Frequenzen und einem flachen Rauschboden; das Spektrum des Gauß-Kerns (orange, um 300 vergrößert) fällt ab etwa 40 steil ab; das Spektrum des Ergebnisses (grün) ist das Produkt beider: Die tiefen Frequenzen bleiben, der Rauschboden darüber verschwindet.
* Im **rechten Bild** wächst die direkte Rechenzeit wie $N^2$, die über die FFT wie $N\log N$. Laut **Ausgabe** braucht die direkte Faltung bei $N = 32\,768$ 301 ms, die FFT-Faltung 0.85 ms, ein Faktor von etwa 350; bei kurzen Folgen ist die direkte Faltung schneller, weil die FFT einen festen Mehraufwand hat. Beide Wege stimmen auf $10^{-15}$ überein. Ohne Auffüllen mit Nullen rechnet die FFT zyklisch: Das Ende des Signals wird an den Anfang gewickelt, und an den Rändern entsteht ein Fehler von 0.43.
* **Wo es vorkommt:** **Bildfilter** (Weichzeichnen, Schärfen, Kantenerkennung), **Hall** in Audiosoftware (Faltung mit der gemessenen Impulsantwort eines Raums), die **Punktspreizfunktion** von Teleskopen und Mikroskopen und das Entfalten von Bildern, Faltungsnetze (große Kerne werden per FFT gefaltet, Fourier Neural Operators lernen direkt im Frequenzraum), die Wahrscheinlichkeitsrechnung (die Dichte einer Summe unabhängiger Zufallsgrößen ist die Faltung der Dichten, ihre charakteristischen Funktionen multiplizieren sich), die Multiplikation großer Zahlen und Polynome (Schönhage-Strassen) und **Matched Filter** in GPS und Radar, die ein bekanntes Signal im Rauschen per Korrelation finden.

##### *7. Die Poissonsche Summenformel*

Wie kann man eine langsam konvergierende Summe über alle ganzen Zahlen schnell berechnen? Vorher vorhersagen: Wie viele Terme braucht $\sum_ne^{-\pi n^2t}$ bei $t = 0.01$ für 15 Stellen Genauigkeit, und wie viele die andere Seite der Poissonschen Summenformel?

Für die Gauß-Funktion $e^{-\pi x^2t}$ werden beide Seiten der Poissonschen Summenformel für $t$ von 0.01 bis 100 ausgewertet (Jacobis Theta-Identität), mit der Zahl der Terme, die jede Seite für 15 Stellen braucht (links und Mitte). Rechts wird $\sum_n1/(n^2 + 1)$ direkt summiert und über die Poisson-Seite berechnet und mit dem exakten Wert $\pi\coth\pi$ verglichen.

**Hintergrund & Fachbegriffe:**
- **Poissonsche Summenformel:** $\sum_{n\in\mathbb Z}f(n) = \sum_{k\in\mathbb Z}\hat f(k)$.
- **Periodisierung:** $F(x) = \sum_nf(x + n)$; aus einer Funktion auf der Geraden wird eine 1-periodische Funktion, indem man alle ganzzahligen Verschiebungen aufaddiert.
- **Theta-Funktion:** $\theta(t) = \sum_ne^{-\pi n^2t}$.
- **coth:** der hyperbolische Kotangens $\coth y = \frac{e^y + e^{-y}}{e^y - e^{-y}}$.

**Theorie:**

- **Aussage und Beweis:** $\sum_nf(x + n) = \sum_k\hat f(k)\,e^{2\pi ikx}$. Die Periodisierung $F$ ist 1-periodisch, und ihr $k$-ter Fourier-Koeffizient ist $\int_0^1F(x)e^{-2\pi ikx}\,\mathrm dx = \int_{\mathbb R}f(x)e^{-2\pi ikx}\,\mathrm dx = \hat f(k)$. Bei $x = 0$ ergibt sich die Summenformel. In der Sprache der Distributionen: Die Fourier-Transformierte des Dirac-Kamms $\sum_n\delta(x - n)$ ist der Dirac-Kamm $\sum_k\delta(\xi - k)$.
- **Theta-Identität:** Mit $\hat f(k) = t^{-1/2}e^{-\pi k^2/t}$ wird daraus $\theta(t) = t^{-1/2}\theta(1/t)$ (Jacobi). Riemann bewies damit 1859 die Funktionalgleichung der Zeta-Funktion; die Identität ist der Ausgangspunkt der Theorie der Modulformen.
- **Warum eine Seite schnell ist:** Eine breite, glatte Funktion hat eine schmale Transformierte (Exp. 5). Ist $f$ breit, konvergiert die linke Summe langsam, die rechte schnell, und umgekehrt.
- **Gitterbeispiel:** $f(x) = \frac1{x^2 + a^2}$ hat $\hat f(k) = \frac\pi ae^{-2\pi a\vert k\vert}$. Also $\sum_n\frac1{n^2 + a^2} = \frac\pi a\sum_ke^{-2\pi a\vert k\vert} = \frac\pi a\coth(\pi a)$. Links fallen die Terme nur wie $1/n^2$ (Fehler nach $N$ Termen etwa $2/N$), rechts exponentiell.
- Dieselbe Formel erklärt das Abtasttheorem (Exp. 9: Abtasten periodisiert das Spektrum) und den Wärmeleitungskern auf dem Ring (Exp. 13).

```python
def theta_terms(t, side, tol=1e-15):
    """Partialsummen beider Seiten von Σ e^{−πn²t} = t^{−1/2} Σ e^{−πk²/t}; gibt Wert und nötige Termzahl zurück."""
    s = 1.0 if side == "links" else 1 / np.sqrt(t)
    a = t if side == "links" else 1 / t
    pref = 1.0 if side == "links" else 1 / np.sqrt(t)
    for n in range(1, 100_000):
        term = 2 * pref * np.exp(-np.pi * n ** 2 * a)
        s += term
        if term < tol * s:
            return s, n
    return s, n

ts = np.geomspace(0.01, 100, 41)
left = np.array([theta_terms(t, "links") for t in ts])
right = np.array([theta_terms(t, "rechts") for t in ts])

a = 1.0
exact_sum = np.pi / a / np.tanh(np.pi * a)
Nn = np.arange(1, 2001)
left_partial = 1 / a ** 2 + 2 * np.cumsum(1 / (Nn ** 2 + a ** 2))
Kk = np.arange(0, 8)
right_partial = np.pi / a * (1 + 2 * np.cumsum(np.r_[0, np.exp(-2 * np.pi * a * Kk[1:])]))

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
axes[0].loglog(ts, left[:, 0], "o", ms=4, label="links: Σ e^(−πn²t)")
axes[0].loglog(ts, right[:, 0], "-", label="rechts: t^(−½) Σ e^(−πk²/t)")
axes[0].set_xlabel("t")
axes[0].set_ylabel("θ(t)")
axes[0].set_title("Jacobis Theta-Identität", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].loglog(ts, left[:, 1], "o-", ms=3, label="Terme links")
axes[1].loglog(ts, right[:, 1], "s-", ms=3, label="Terme rechts")
axes[1].set_xlabel("t")
axes[1].set_ylabel("nötige Terme für 15 Stellen")
axes[1].set_title("Wo eine Seite langsam ist, ist die andere schnell", fontsize=10)
axes[1].legend(fontsize=8)
axes[2].semilogy(Nn, np.abs(left_partial - exact_sum), label="Σ 1/(n² + 1) direkt, bis |n| ≤ N")
axes[2].semilogy(Kk, np.maximum(np.abs(right_partial - exact_sum), 1e-17), "s-", label="Poisson-Seite (π/a) Σ e^(−2πa|k|), bis |k| ≤ K")
axes[2].set_xscale("symlog", linthresh=10)
axes[2].set_xlabel("Zahl der Terme N bzw. K")
axes[2].set_ylabel("Fehler gegen (π/a)·coth(πa)")
axes[2].set_title("Langsame Summe, schnelle Poisson-Seite", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

for t in (0.01, 1.0, 100.0):
    i = np.argmin(np.abs(ts - t))
    print(f"t = {ts[i]:6.2f}: links {left[i, 0]:.15f} ({left[i, 1]:.0f} Terme), rechts {right[i, 0]:.15f} ({right[i, 1]:.0f} Terme)")
print(f"Σ_n 1/(n² + 1): exakt π·coth(π) = {exact_sum:.12f}; direkt mit |n| ≤ 2000: Fehler {abs(left_partial[-1] - exact_sum):.1e}; "
      f"Poisson-Seite mit |k| ≤ 7: Fehler {abs(right_partial[-1] - exact_sum):.1e}")
```

* Im **linken Bild** stimmen beide Seiten für alle $t$ überein; für kleine $t$ ist $\theta(t)\approx t^{-1/2}$, für große $t$ strebt es gegen 1. Das **mittlere Bild** zeigt die Zahl der nötigen Terme: Bei $t = 0.01$ braucht die linke Seite 33 Terme, die rechte einen einzigen; bei $t = 100$ ist es umgekehrt, bei $t = 1$ brauchen beide 4. Das beantwortet die Vorhersage. Laut **Ausgabe** ist $\theta(0.01) = 10.000000000000002$ (links) und $10.000000000000000$ (rechts), $\theta(1) = 1.086434811213308$ auf beiden Seiten.
* Im **rechten Bild** konvergiert die direkte Summe $\sum1/(n^2 + 1)$ langsam: Nach 2000 Termen auf jeder Seite ist der Fehler noch $10^{-3}$. Die Poisson-Seite erreicht die Rechengenauigkeit schon mit sechs bis sieben Termen, denn jeder Term ist um $e^{-2\pi}\approx0.0019$ kleiner als der vorige. Der exakte Wert ist $\pi\coth\pi = 3.153348094937$.
* **Wo es vorkommt:** **Ewald-Summation** in der Molekulardynamik und Festkörperphysik: Die langsam konvergierende Coulomb-Summe über ein Kristallgitter wird in einen kurzreichweitigen Teil im Ortsraum und einen glatten Teil im Fourier-Raum zerlegt (Particle-Mesh-Ewald in Simulationsprogrammen für Proteine). Außerdem das **reziproke Gitter** der Kristallographie, das Abtasttheorem (Exp. 9), Theta-Funktionen und Modulformen in der Zahlentheorie und die Wärmeleitung auf periodischen Gebieten (Exp. 13).

##### *8. Diskrete Fourier-Transformation und FFT*

Wie rechnet ein Computer die Fourier-Transformation, und warum ist die FFT so schnell? Vorher vorhersagen: Eine Matrix $F$ erfüllt $F^4 = I$. Welche Eigenwerte kann sie haben?

Links ist der Realteil der DFT-Matrix für $N = 32$ als Bild dargestellt, in der Mitte die Eigenwerte der unitären DFT-Matrix ($F/\sqrt N$) für $N = 16$. Eine eigene Cooley-Tukey-FFT in reinem Python wird mit numpy verglichen; rechts stehen die Rechenzeiten von Matrix mal Vektor (mit fertig berechneter Matrix, bis $N = 2048$), der eigenen FFT und `np.fft.fft` für $N$ von 16 bis 16 384.

**Hintergrund & Fachbegriffe:**
- **DFT:** $X_k = \sum_{n=0}^{N-1}x_n\omega^{kn}$ mit der **Einheitswurzel** $\omega = e^{-2\pi i/N}$, also $X = Fx$ mit der Matrix $F_{jk} = \omega^{jk}$.
- **Unitär:** Eine Matrix $U$ ist unitär, wenn $UU^\ast = I$; sie erhält Längen und Winkel, ihre Eigenwerte haben Betrag 1.
- **FFT** (Fast Fourier Transform): jedes Verfahren, das die DFT in $O(N\log N)$ Schritten berechnet; das bekannteste ist **Cooley-Tukey**.
- **Twiddle-Faktoren:** die Einheitswurzeln $\omega^k$, mit denen die Teilergebnisse zusammengesetzt werden.
- **Frequenzachse:** Der Index $k$ gehört zur Frequenz $k\,f_s/N$ bei Abtastrate $f_s$; die Indizes über $N/2$ sind die **negativen Frequenzen** (`np.fft.fftfreq`).

**Theorie:**

- **Unitarität:** Die Spalten von $F$ sind orthogonal, weil $\sum_n\omega^{(j - k)n} = 0$ für $j\neq k$ (geometrische Reihe der Einheitswurzeln); also ist $F/\sqrt N$ unitär und $F^{-1} = \overline F/N$.
- **Eigenwerte:** $F^2/N$ ist die Spiegelung $x_n\mapsto x_{-n}$, also ist $(F/\sqrt N)^4 = I$, und die Eigenwerte liegen in $\lbrace1, -i, -1, i\rbrace$, wie bei der kontinuierlichen Transformation (Exp. 4). Für $N = 4m$ sind die Vielfachheiten $m + 1$, $m$, $m$, $m - 1$ (für 1, $-1$, $-i$, $i$).
- **Cooley-Tukey:** Teilt man die Summe in gerade und ungerade Indizes, ist $X_k = E_k + \omega^kO_k$ und $X_{k + N/2} = E_k - \omega^kO_k$, wobei $E$ und $O$ die DFTs der Länge $N/2$ der geraden und ungeraden Werte sind. Zwei halbe Probleme plus $O(N)$ Arbeit: $T(N) = 2T(N/2) + O(N)$, also $T(N) = O(N\log_2N)$. Bei $N = 2^{14}$ sind das etwa $2.3\cdot10^5$ statt $2.7\cdot10^8$ Operationen.
- **Geschichte:** Gauß benutzte das Verfahren 1805 für die Bahnen der Asteroiden Pallas und Juno, unveröffentlicht; Cooley und Tukey machten es 1965 bekannt. Für beliebige $N$ gibt es Varianten (gemischte Basen, Bluestein für Primzahlen); numpy benutzt die Bibliothek pocketfft.
- **Reelle Signale** haben $X_{N - k} = \overline{X_k}$; deshalb speichert `rfft` nur die Hälfte.

```python
def fft_radix2(x):
    """Cooley-Tukey-FFT für Längen 2^m: zerlegt in gerade und ungerade Indizes und setzt zwei halbe DFTs zusammen."""
    n = len(x)
    if n == 1:
        return x.astype(complex)
    even, odd = fft_radix2(x[0::2]), fft_radix2(x[1::2])
    twiddle = np.exp(-2j * np.pi * np.arange(n // 2) / n) * odd
    return np.concatenate([even + twiddle, even - twiddle])

def dft_matrix(n):
    j = np.arange(n)
    return np.exp(-2j * np.pi * np.outer(j, j) / n)

rng = np.random.default_rng(8)
F16 = dft_matrix(16) / 4
ev = np.linalg.eigvals(F16)
labels = {1: "1", -1: "−1", -1j: "−i", 1j: "i"}
counts = {lab: int(np.sum(np.abs(ev - z) < 1e-8)) for z, lab in labels.items()}

sizes = 2 ** np.arange(4, 15)
times = {"Matrix · Vektor": [], "eigene FFT (Python)": [], "np.fft.fft": []}
for m in sizes:
    v = rng.normal(size=m) + 1j * rng.normal(size=m)
    cands = [("np.fft.fft", lambda: np.fft.fft(v)), ("eigene FFT (Python)", lambda: fft_radix2(v))]
    if m <= 2048:
        M = dft_matrix(m)
        cands.append(("Matrix · Vektor", lambda: M @ v))
    for name, fun in cands:
        best = np.inf
        for _ in range(3):
            t0 = time.perf_counter(); fun(); best = min(best, time.perf_counter() - t0)
        times[name].append(best)

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
im = axes[0].imshow(dft_matrix(32).real, cmap="RdBu", vmin=-1, vmax=1)
axes[0].set_title("Realteil der DFT-Matrix, N = 32: cos(2πjk/N)", fontsize=10)
axes[0].set_xlabel("Spalte k")
axes[0].set_ylabel("Zeile j")
plt.colorbar(im, ax=axes[0], fraction=0.046)
th = np.linspace(0, 2 * np.pi, 200)
axes[1].plot(np.cos(th), np.sin(th), color="grey", lw=0.5)
axes[1].plot(ev.real, ev.imag, "o", ms=8)
for z, lab in labels.items():
    axes[1].annotate(f"{lab}: {counts[lab]}×", (complex(z).real, complex(z).imag), textcoords="offset points", xytext=(8, 8), fontsize=9)
axes[1].set_aspect("equal")
axes[1].set_xlim(-1.5, 1.5); axes[1].set_ylim(-1.5, 1.5)
axes[1].set_xlabel("Realteil")
axes[1].set_ylabel("Imaginärteil")
axes[1].set_title("Eigenwerte der unitären DFT, N = 16", fontsize=10)
for name, mk in zip(times, ["o-", "s-", "^-"]):
    axes[2].loglog(sizes[:len(times[name])], times[name], mk, ms=4, label=name)
axes[2].loglog(sizes, 2e-9 * sizes ** 2.0, "k:", lw=1, label="∝ N²")
axes[2].loglog(sizes, 3e-9 * sizes * np.log2(sizes), "k--", lw=1, label="∝ N log N")
axes[2].set_xlabel("Länge N")
axes[2].set_ylabel("Rechenzeit [s]")
axes[2].set_title("Matrix-Vektor gegen FFT", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

v = rng.normal(size=1024)
print(f"Unitär: ‖F F* − I‖ = {np.linalg.norm(F16 @ F16.conj().T - np.eye(16)):.1e};  F⁴ = I: {np.allclose(np.linalg.matrix_power(F16, 4), np.eye(16))};  "
      f"F² kehrt die Indizes um (j → −j mod N): {np.allclose(np.linalg.matrix_power(F16, 2), np.eye(16)[(-np.arange(16)) % 16])}")
print(f"Vielfachheiten der Eigenwerte für N = 16: {counts}")
print(f"Eigene FFT gegen np.fft.fft (N = 1024): größter Unterschied {np.max(np.abs(fft_radix2(v) - np.fft.fft(v))):.1e}")
print(f"Operationen bei N = 2^14: N² = {16384 ** 2:.2e}, N·log₂N = {16384 * 14:.2e}; Zeit eigene FFT {times['eigene FFT (Python)'][-1] * 1e3:.1f} ms, np.fft {times['np.fft.fft'][-1] * 1e3:.3f} ms")
```

* Im **linken Bild** ist Zeile $j$ ein Kosinus mit $j$ Schwingungen über die Zeile; die Matrix ist symmetrisch, und Zeile $N - j$ gleicht Zeile $j$ (die negative Frequenz hat denselben Realteil). Im **mittleren Bild** liegen alle Eigenwerte auf den vier Punkten $1, -1, i, -i$ des Einheitskreises, mit den Vielfachheiten 5, 4, 3 und 4. Das beantwortet die Vorhersage. Laut **Ausgabe** ist $F/\sqrt N$ unitär bis auf $9\cdot10^{-15}$, $F^4 = I$, und $F^2$ kehrt die Indizes um.
* Im **rechten Bild** wächst die Zeit für Matrix mal Vektor wie $N^2$, die der FFTs wie $N\log N$. Die eigene Python-FFT stimmt mit numpy auf $5\cdot10^{-14}$ überein, ist aber wegen des Python-Mehraufwands langsam (45 ms bei $N = 16\,384$); `np.fft.fft` braucht dafür 0.08 ms. Bei kleinen $N$ dominiert der feste Aufwand des Funktionsaufrufs, deshalb ist dort die Matrix mal Vektor nicht langsamer.
* **Wo es vorkommt:** in praktisch aller digitalen Signalverarbeitung: Spektren und Equalizer in der Audiotechnik, **OFDM** im WLAN, in LTE und 5G (die Daten werden auf viele Unterträger verteilt, Sender und Empfänger rechnen eine IFFT und eine FFT), Radar, die Bildrekonstruktion in der MRT (inverse 2D-FFT der Messdaten), spektrale Löser für Strömungen und Wetter, schnelle Faltungen (Exp. 6). Die **Quanten-Fourier-Transformation** ist dieselbe DFT auf den $2^n$ Amplituden eines Registers aus $n$ Qubits und braucht nur $O(n^2)$ Gatter, exponentiell weniger als die FFT; dafür kann man das Ergebnis nicht auslesen, sondern nur einmal messen (Exp. 15).

##### *9. Abtasttheorem und Aliasing*

Wie oft muss man ein Signal abtasten, damit nichts verloren geht? Vorher vorhersagen: Ein Sinus mit 7 Hz wird zehnmal pro Sekunde abgetastet. Welche Frequenz zeigt das Spektrum der Abtastwerte?

Links werden ein 7-Hz-Sinus und ein 3-Hz-Sinus mit derselben Abtastrate $f_s = 10$ Hz verglichen. In der Mitte wird für wahre Frequenzen von 0.25 bis 29.75 Hz gemessen, bei welcher Frequenz der Gipfel des FFT-Spektrums der Abtastwerte liegt. Rechts wird ein Signal mit Frequenzen 1.3, 3.7 und 4.6 Hz aus Abtastwerten mit 20, 10 und 6 Hz über die sinc-Formel rekonstruiert (Abtastwerte über $\pm200$ s).

**Hintergrund & Fachbegriffe:**
- **Abtastrate** $f_s$: die Zahl der Messwerte pro Sekunde; **Nyquist-Frequenz** $f_s/2$.
- **Bandbegrenzt:** Das Spektrum ist oberhalb einer Frequenz $B$ null.
- **Aliasing:** Eine Frequenz oberhalb von $f_s/2$ erscheint nach dem Abtasten als falsche, tiefere Frequenz (ihr „Alias“).
- **sinc-Interpolation:** die Rekonstruktion $f(t) = \sum_nf(n/f_s)\,\mathrm{sinc}(f_st - n)$ aus den Abtastwerten.
- **Anti-Aliasing-Filter:** ein Tiefpass vor dem Abtasten, der alles über $f_s/2$ entfernt.

**Theorie:**

- **Abtasten periodisiert das Spektrum:** Abtasten heißt, $f$ mit einem Dirac-Kamm im Abstand $1/f_s$ zu multiplizieren. Nach dem Faltungssatz und der Poissonschen Summenformel (Exp. 7) wird das Spektrum dabei mit einem Kamm im Abstand $f_s$ gefaltet: $\hat f_{\text{abgetastet}}(\xi) = f_s\sum_m\hat f(\xi - mf_s)$, Kopien des Spektrums im Abstand $f_s$.
- **Abtasttheorem** (Whittaker 1915, Nyquist 1928, Kotelnikov 1933, Shannon 1949): Ist $f$ auf $[-B, B]$ bandbegrenzt und $f_s > 2B$, überlappen sich die Kopien nicht. Ein idealer Tiefpass bis $f_s/2$ schneidet die mittlere Kopie heraus, und das ist im Ort die sinc-Interpolation: $f$ ist aus den Abtastwerten exakt bestimmt.
- **Aliasing:** Bei $f_s < 2B$ überlappen die Kopien. Eine Frequenz $f$ erscheint bei $\vert f - f_s\cdot\mathrm{round}(f/f_s)\vert$; das Spektrum wird an $f_s/2$ gespiegelt („Faltung“ der Frequenzachse). Die Ungleichung muss streng sein: Ein Sinus mit genau $f_s/2$ kann bei jeder Abtastung auf seine Nullstellen fallen.
- **In der Praxis:** Man hat nur endlich viele Abtastwerte, und die sinc-Funktion fällt nur wie $1/n$; der Abbruch erzeugt einen kleinen Fehler. Überabtasten (deutlich über $2B$) erlaubt rascher abfallende Interpolationskerne.

```python
fs = 10.0
tc = np.linspace(0, 1, 2001)
tn = np.arange(0, 1.0001, 1 / fs)

def apparent_frequency(f, fs, T=20.0):
    """Frequenz des stärksten FFT-Gipfels eines mit fs abgetasteten Sinus der Frequenz f."""
    tt = np.arange(0, T, 1 / fs)
    spec = np.abs(np.fft.rfft(np.sin(2 * np.pi * f * tt + 0.3) * np.hanning(len(tt))))
    return np.fft.rfftfreq(len(tt), 1 / fs)[np.argmax(spec)]

freqs = np.linspace(0.25, 29.75, 119)
seen = [apparent_frequency(f, fs) for f in freqs]
folded = np.abs(freqs - fs * np.round(freqs / fs))

def band_limited(t):
    return np.sin(2 * np.pi * 1.3 * t) + 0.5 * np.cos(2 * np.pi * 3.7 * t + 1) + 0.3 * np.sin(2 * np.pi * 4.6 * t)

def sinc_reconstruct(fs, t_eval, T=200.0):
    tn = np.arange(-T, T, 1 / fs)
    return np.sinc(fs * t_eval[:, None] - fs * tn[None, :]) @ band_limited(tn)

te = np.linspace(-2, 2, 801)
recon = {r: sinc_reconstruct(r, te) for r in (20.0, 10.0, 6.0)}

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
axes[0].plot(tc, np.sin(2 * np.pi * 7 * tc), lw=1, label="Sinus mit 7 Hz")
axes[0].plot(tc, -np.sin(2 * np.pi * 3 * tc), "--", lw=1.5, label="Alias: Sinus mit 3 Hz (Vorzeichen gedreht)")
axes[0].plot(tn, np.sin(2 * np.pi * 7 * tn), "ko", ms=6, label=f"Abtastwerte, f_s = {fs:.0f} Hz")
axes[0].set_xlabel("Zeit t [s]")
axes[0].set_title("7 Hz und 3 Hz haben dieselben Abtastwerte", fontsize=10)
axes[0].legend(fontsize=7, loc="lower left")
axes[1].plot(freqs, seen, "o", ms=3, label="Gipfel im FFT-Spektrum")
axes[1].plot(freqs, folded, "k-", lw=1, label="|f − f_s · round(f / f_s)|")
for kf in range(1, 4):
    axes[1].axvline(kf * fs / 2, color="grey", ls=":", lw=0.8)
axes[1].set_xlabel("wahre Frequenz f [Hz]")
axes[1].set_ylabel("gesehene Frequenz [Hz]")
axes[1].set_title("Spiegelung an der Nyquist-Frequenz f_s/2 = 5 Hz", fontsize=10)
axes[1].legend(fontsize=8)
axes[2].plot(te, band_limited(te), "k-", lw=2, label="Signal (höchste Frequenz 4.6 Hz)")
for r, c in zip((10.0, 6.0), ["tab:green", "tab:red"]):
    axes[2].plot(te, recon[r], c=c, lw=1, label=f"sinc-Rekonstruktion aus f_s = {r:.0f} Hz")
axes[2].set_xlabel("Zeit t [s]")
axes[2].set_title("Whittaker-Shannon: über 2 · 4.6 Hz exakt, darunter falsch", fontsize=10)
axes[2].legend(fontsize=7)
plt.tight_layout()
plt.show()

print(f"Gesehene Frequenzen bei f_s = 10 Hz: 7 Hz → {apparent_frequency(7, fs):.2f} Hz, 12 Hz → {apparent_frequency(12, fs):.2f} Hz, 18.5 Hz → {apparent_frequency(18.5, fs):.2f} Hz")
for r in (20.0, 10.0, 6.0):
    print(f"f_s = {r:4.0f} Hz ({'über' if r > 9.2 else 'unter'} 2 · 4.6 = 9.2 Hz): größter Rekonstruktionsfehler auf [−2, 2] = {np.max(np.abs(recon[r] - band_limited(te))):.1e}")
```

* Im **linken Bild** laufen der 7-Hz-Sinus (blau) und ein 3-Hz-Sinus mit umgekehrtem Vorzeichen (orange) durch dieselben elf Abtastwerte: Nach dem Abtasten sind sie nicht mehr zu unterscheiden. Im **mittleren Bild** liegt die gesehene Frequenz auf einer Zickzacklinie zwischen 0 und 5 Hz: Bis 5 Hz stimmt sie, darüber wird sie an 5 Hz gespiegelt, bei 10 Hz ist sie 0, und so weiter. Laut **Ausgabe** erscheinen 7 Hz als 3.00 Hz (die Antwort auf die Vorhersage), 12 Hz als 2.00 Hz, 18.5 Hz als 1.50 Hz.
* Im **rechten Bild** liegt die Rekonstruktion aus 10 Hz (grün) genau auf dem Signal; laut Ausgabe ist der größte Fehler $2.5\cdot10^{-4}$, aus 20 Hz sogar $6\cdot10^{-6}$ (der Rest kommt vom Abbruch nach $\pm200$ s, und Überabtasten macht ihn kleiner). Aus 6 Hz (rot) ist die Rekonstruktion falsch, mit Fehler 1.6: Die Anteile bei 3.7 und 4.6 Hz liegen über der Nyquist-Frequenz 3 Hz und erscheinen als 2.3 und 1.4 Hz.
* **Wo es vorkommt:** Die **Audio-CD** tastet mit 44.1 kHz ab, etwas mehr als das Doppelte der Hörgrenze von 20 kHz, mit einem Anti-Aliasing-Filter davor. Rückwärts drehende Speichenräder im Film (24 Bilder pro Sekunde) und **Moiré-Muster** beim Fotografieren feiner Stoffe oder Bildschirme sind Aliasing in Zeit und Raum. Kameras, Oszilloskope und Sensoren werden nach dem Abtasttheorem ausgelegt. In Faltungsnetzen erzeugt das Verkleinern ohne Tiefpass (Pooling, Schrittweite > 1) Aliasing, und die Netze reagieren dadurch empfindlich auf Verschiebungen um ein Pixel; Abhilfe sind vorgeschaltete Tiefpässe (anti-aliased CNNs, StyleGAN3). Compressed Sensing kommt bei dünnbesetzten Signalen mit weniger Messungen aus.

##### *10. Leckeffekt und Fensterfunktionen*

Warum verschmiert ein endliches Messfenster einen reinen Ton, und wie findet man einen schwachen Ton neben einem starken? Vorher vorhersagen: Ein Ton bei 61 Hz ist 1000-mal (60 dB) schwächer als ein Ton bei 50.3 Hz. Ist er im Spektrum einer Messung von 1 s zu sehen?

Ein Signal aus den beiden Tönen wird 1 s lang mit 1000 Hz abgetastet und mit drei Fenstern (Rechteck, Hann, Blackman-Harris) gewichtet; die Spektren sind mit 16-facher Nullverlängerung fein abgetastet und so normiert, dass ein Ton der Amplitude 1 bei 0 dB liegt (links). In der Mitte stehen die Fenster, rechts ihre Spektren in Einheiten von Frequenzbins. Die Ausgabe misst die höchste Nebenkeule, die Breite der Hauptkeule, den Amplitudenverlust bei einem Ton zwischen zwei Bins und den Pegel bei 61 Hz mit und ohne schwachen Ton.

**Hintergrund & Fachbegriffe:**
- **Leckeffekt** (spectral leakage): Die Energie eines Tons verteilt sich auf viele Frequenzbins, statt in einem zu liegen.
- **Frequenzbin:** ein Raster der DFT im Abstand $1/T$ bei Messdauer $T$, hier 1 Hz.
- **Fensterfunktion** $w(t)$: Gewichte, mit denen man das Signal vor der FFT multipliziert; das **Rechteckfenster** ist das ungewichtete Abschneiden. **Hann:** $\sin^2(\pi t/T)$; **Blackman-Harris:** eine Summe aus vier Kosinustermen.
- **Hauptkeule** und **Nebenkeulen:** der zentrale Gipfel des Fensterspektrums und die kleineren Gipfel daneben.
- **Dezibel:** $20\log_{10}$ des Amplitudenverhältnisses; $-60$ dB ist ein Tausendstel, $-20$ dB ein Zehntel.
- **Nullverlängerung** (zero padding): Anhängen von Nullen vor der FFT; sie tastet das Spektrum feiner ab, verbessert aber nicht die Auflösung.

**Theorie:**

- **Messen heißt Multiplizieren mit einem Fenster:** Ein endliches Stück $x(t)\,w(t)$ hat das Spektrum $\hat x\ast\hat w$ (Faltungssatz). Jeder Ton wird zu einer Kopie von $\hat w$ an seiner Frequenz.
- **Rechteckfenster:** $\hat w$ ist der Dirichlet-Kern (Exp. 2), mit höchster Nebenkeule $-13.3$ dB und Nebenkeulen, die nur wie $1/k$ (6 dB pro Oktave) abfallen, weil das Fenster an den Rändern springt (Exp. 1). Liegt die Frequenz genau auf einem Bin, fallen alle anderen Bins auf die Nullstellen und man sieht nichts davon; liegt sie dazwischen (hier 50.3 Hz), leckt die Energie überallhin.
- **Glatte Fenster:** Das Hann-Fenster geht mit Wert und Ableitung stetig auf 0; seine Nebenkeulen beginnen bei $-31.5$ dB und fallen wie $1/k^3$. Blackman-Harris erreicht $-92$ dB. Der Preis ist eine breitere Hauptkeule: 2, 4 und 8 Bins von Nullstelle zu Nullstelle. Zwei Töne müssen ungefähr eine halbe Hauptkeulenbreite auseinander liegen, um getrennt zu werden.
- **Scalloping:** Liegt ein Ton genau zwischen zwei Bins, erscheint seine Amplitude zu klein, beim Rechteck um 3.92 dB. Für genaue Amplituden nimmt man Flat-Top-Fenster.
- **Auflösung:** Sie hängt nur von der Messdauer ab, etwa $1/T$; Nullverlängerung interpoliert nur (Variation 5).

```python
fs, n = 1000, 1000
t = np.arange(n) / fs
x = np.sin(2 * np.pi * 50.3 * t) + 1e-3 * np.sin(2 * np.pi * 61 * t)              # starker Ton neben einem um 60 dB schwächeren
windows = {"Rechteck": np.ones(n), "Hann": signal.windows.hann(n, sym=False), "Blackman-Harris": signal.windows.blackmanharris(n, sym=False)}
pad = 16 * n

def spectrum_db(x, w):
    S = np.abs(np.fft.rfft(x * w, pad)) / (w.sum() / 2)               # so normiert, dass ein Ton der Amplitude 1 bei 0 dB liegt
    return np.fft.rfftfreq(pad, 1 / fs), 20 * np.log10(np.maximum(S, 1e-12))

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
for (name, w), c in zip(windows.items(), ["tab:blue", "tab:orange", "tab:green"]):
    f, S = spectrum_db(x, w)
    sel = (f > 30) & (f < 80)
    axes[0].plot(f[sel], S[sel], c=c, lw=1, label=name)
axes[0].axvline(61, color="grey", ls=":", lw=1)
axes[0].annotate("schwacher Ton\n61 Hz, −60 dB", (61, -60), xytext=(66, -25), fontsize=8, arrowprops=dict(arrowstyle="->"))
axes[0].set_ylim(-140, 5)
axes[0].set_xlabel("Frequenz [Hz]")
axes[0].set_ylabel("Pegel [dB]")
axes[0].set_title("Starker Ton bei 50.3 Hz, schwacher bei 61 Hz", fontsize=10)
axes[0].legend(fontsize=8)
for (name, w), c in zip(windows.items(), ["tab:blue", "tab:orange", "tab:green"]):
    axes[1].plot(t, w, c=c, label=name)
axes[1].set_xlabel("Zeit [s]")
axes[1].set_ylabel("Gewicht w(t)")
axes[1].set_title("Die drei Fenster", fontsize=10)
axes[1].legend(fontsize=8)
stats_w = {}
for (name, w), c in zip(windows.items(), ["tab:blue", "tab:orange", "tab:green"]):
    W = np.abs(np.fft.fft(w, 64 * n))
    W_db = 20 * np.log10(np.maximum(W / W[0], 1e-12))
    bins = np.arange(64 * n) / 64
    sel = bins <= 20
    axes[2].plot(bins[sel], W_db[sel], c=c, lw=1, label=name)
    first_null = np.argmax(np.diff(W_db[: 64 * 20]) > 0)          # erstes Minimum = Ende der Hauptkeule
    stats_w[name] = (W_db[first_null:64 * 20].max(), 2 * bins[first_null], 20 * np.log10(W[32] / W[0]))
axes[2].set_ylim(-140, 5)
axes[2].set_xlabel("Abstand vom Ton [Frequenzbins = 1/T]")
axes[2].set_ylabel("Pegel [dB]")
axes[2].set_title("Spektrum der Fenster: Hauptkeule und Nebenkeulen", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

for name, w in windows.items():
    f, S = spectrum_db(x, w)
    f0, S0 = spectrum_db(np.sin(2 * np.pi * 50.3 * t), w)
    i61 = np.argmin(np.abs(f - 61))
    psl, width, scallop = stats_w[name]
    print(f"{name:15s}: höchste Nebenkeule {psl:6.1f} dB, Hauptkeule {width:.1f} Bins breit, Verlust bei halbem Bin {scallop:5.2f} dB; "
          f"bei 61 Hz: {S[i61]:6.1f} dB mit, {S0[i61]:6.1f} dB ohne schwachen Ton")
```

##### *11. Zeit-Frequenz-Analyse: das Spektrogramm*

Wie zeigt man, welche Frequenzen wann vorkommen? Vorher vorhersagen: Wie lang muss ein Hann-Fenster mindestens sein, um Töne bei 200 und 215 Hz zu trennen, und was passiert dann mit einem kurzen Klick?

Ein 2 s langes Signal (Abtastrate 4000 Hz) enthält zwei Töne bei 200 und 215 Hz, einen Chirp, der von 400 auf 900 Hz steigt, und einen Klick von 0.5 ms bei $t = 1$ s. Es wird mit Hann-Fenstern von 32 ms und 512 ms Länge in überlappende Stücke zerlegt (Schrittweite ein Achtel des Fensters) und als Spektrogramm dargestellt (links und Mitte, Pegel in dB). Rechts steht für beide Fenster der Schnitt bei $t = 0.5$ s im Bereich der beiden Töne. Die Ausgabe misst, wie lang der Klick im Spektrogramm erscheint und wie tief der Einbruch zwischen den Tönen ist.

**Hintergrund & Fachbegriffe:**
- **Kurzzeit-Fourier-Transformation** (STFT): die Fourier-Transformation eines gefensterten Stücks, für jede Fensterposition neu.
- **Spektrogramm:** $\vert\mathrm{STFT}\vert^2$ als Bild über Zeit und Frequenz.
- **Zeitauflösung** $\Delta t$ und **Frequenzauflösung** $\Delta f$: wie nah zwei Ereignisse in der Zeit bzw. zwei Töne in der Frequenz sein dürfen, um getrennt zu erscheinen.
- **Schwebung:** Zwei Töne mit Abstand $\Delta f$ ergeben eine Schwingung, deren Lautstärke mit $\Delta f$ pulsiert, hier 15-mal pro Sekunde.
- **Chirp:** ein Ton mit gleitender Frequenz.

**Theorie:**

- **Definition:** $X(\tau, \xi) = \int x(t)\,w(t - \tau)\,e^{-2\pi i\xi t}\,\mathrm dt$. Das Fenster der Länge $T_w$ legt beide Auflösungen fest: $\Delta t\approx T_w$ und $\Delta f\approx1/T_w$ (das Raster ist $f_s/N$, die Hauptkeule des Hann-Fensters 4 Bins breit, Exp. 10).
- **Gabor-Grenze:** Das Produkt $\Delta t\,\Delta f$ kann nicht beliebig klein werden; es ist das Unschärfeprinzip aus Exp. 5, $\Delta t\,\Delta f\geq1/(4\pi)$ für die Standardabweichungen. Am besten ist ein Gauß-Fenster (**Gabor-Transformation**, 1946).
- **Trennen zweier Töne:** Die Hauptkeule des Hann-Fensters ist $4/T_w$ breit; zwei Töne werden getrennt, wenn ihr Abstand größer als etwa $2/T_w$ ist. Für 15 Hz braucht man $T_w\gtrsim130$ ms. Ein Klick, der viel kürzer als $T_w$ ist, wird dann über die ganze Fensterlänge verschmiert.
- **Zwei richtige Antworten:** Die beiden Töne sind gleichwertig eine Schwingung bei 207.5 Hz, deren Lautstärke mit 15 Hz pulsiert. Das kurze Fenster zeigt diese Beschreibung, das lange die zwei Töne. Welche man sieht, entscheidet die Fensterlänge.
- Weil eine feste Fensterlänge nicht für alle Frequenzen passt, benutzt man Verfahren mit frequenzabhängiger Auflösung: die Constant-Q-Transformation in der Musik und die Wavelet-Transformation (Exp. 12).

```python
fs, T = 4000, 2.0
t = np.arange(int(fs * T)) / fs
x = (np.sin(2 * np.pi * 200 * t) + np.sin(2 * np.pi * 215 * t)                      # zwei nahe Töne
     + 0.7 * np.sin(2 * np.pi * (400 * t + 125 * t ** 2))                           # Chirp von 400 auf 900 Hz
     + 5 * np.exp(-((t - 1.0) / 0.0005) ** 2))                                      # kurzer Klick bei t = 1 s
specs = {}
for nper in (128, 2048):
    stft = signal.ShortTimeFFT(signal.windows.hann(nper), hop=nper // 8, fs=fs, scale_to="psd")
    specs[nper] = (stft.f, stft.t(len(x)), stft.spectrogram(x))

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
for ax, nper in zip(axes[:2], (128, 2048)):
    f, tt, S = specs[nper]
    ax.pcolormesh(tt, f, 10 * np.log10(S + 1e-12), shading="auto", vmin=-80, vmax=-10, cmap="magma")
    ax.set_ylim(0, 1100)
    ax.set_xlim(0.3, T - 0.3)                                     # an den Rändern ragt das lange Fenster über das Signal hinaus
    ax.set_xlabel("Zeit [s]")
    ax.set_ylabel("Frequenz [Hz]")
    ax.set_title(f"Fenster {nper / fs * 1000:.0f} ms: Δt ≈ {nper / fs * 1000:.0f} ms, Δf ≈ {fs / nper:.1f} Hz", fontsize=10)
for nper, c in zip((128, 2048), ["tab:blue", "tab:orange"]):
    f, tt, S = specs[nper]
    i = np.argmin(np.abs(tt - 0.5))
    axes[2].plot(f, 10 * np.log10(S[:, i] / S[:, i].max()), c=c, label=f"Fenster {nper / fs * 1000:.0f} ms")
axes[2].set_xlim(150, 260)
axes[2].set_ylim(-60, 3)
axes[2].set_xlabel("Frequenz [Hz]")
axes[2].set_ylabel("Pegel relativ zum Maximum [dB]")
axes[2].set_title("Schnitt bei t = 0.5 s: 200 und 215 Hz getrennt?", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

for nper in (128, 2048):
    f, tt, S = specs[nper]
    band = (f > 1000) & (f < 1400)
    e = S[band].sum(axis=0)
    above = tt[e > e.max() / 2]
    i = np.argmin(np.abs(tt - 0.5))
    cut = S[(f >= 195) & (f <= 220), i]
    dip = 10 * np.log10(S[np.argmin(np.abs(f - 207.5)), i] / cut.max())
    print(f"Fenster {nper:4d} Punkte ({nper / fs * 1000:3.0f} ms): Frequenzauflösung f_s/N = {fs / nper:5.1f} Hz; Klick oberhalb halber Höhe {(above.max() - above.min()) * 1000:5.1f} ms lang; "
          f"Einbruch zwischen 200 und 215 Hz: {dip:6.1f} dB")
```

* Im **linken Bild** (32 ms) ist der Klick eine scharfe senkrechte Linie bei $t = 1$ s; laut **Ausgabe** liegt er nur 8 ms oberhalb der halben Höhe. Der Chirp ist ein steigendes Band. Die beiden Töne verschmelzen dagegen zu einem breiten Band um 207 Hz, das im Takt der Schwebung 15-mal pro Sekunde pulsiert. Im **mittleren Bild** (512 ms) sind die Töne zwei feine Linien, der Chirp eine Treppe, weil jedes Fenster über etwa 250 Hz seines Anstiegs mittelt, und der Klick ist zu einem breiten Streifen verschmiert (sichtbar etwa von 0.75 bis 1.25 s, oberhalb der halben Höhe 128 ms).
* Im **rechten Bild** zeigt der Schnitt bei 0.5 s: Mit 512 ms zwei getrennte Gipfel und ein Einbruch von 42.3 dB dazwischen, mit 32 ms (Raster 31 Hz) ein flacher Verlauf ohne Einbruch (0.0 dB). Das beantwortet die Vorhersage: Das Fenster muss deutlich länger als 100 ms sein, und dann ist der Klick zeitlich nicht mehr scharf. An den Rändern bei 0 und 2 s ragt das lange Fenster über das Signal hinaus; deshalb ist dort nur der Bereich von 0.3 bis 1.7 s gezeigt.
* **Wo es vorkommt:** **Spracherkennung** (Mel-Spektrogramme als Eingabe für Modelle wie Whisper), Musik (kurze Fenster für Schlagzeug, lange für Tonhöhen; Notenerkennung), die Analyse von Vogelgesang und Fledermausrufen, das Chirp-Signal verschmelzender schwarzer Löcher in Gravitationswellen-Detektoren, Mikro-Doppler-Signaturen im Radar, EEG-Analyse und die Überwachung von Maschinen.

##### *12. Wavelets gegen Fourier: wer braucht weniger Koeffizienten?*

Gibt es Basen, die für Signale mit Sprüngen besser geeignet sind als Schwingungen? Vorher vorhersagen: Welche Basis braucht weniger Koeffizienten für ein Signal aus Stufen, welche für ein glattes Signal?

Zwei Signale mit 1024 Werten werden verglichen: ein Blocksignal mit acht Sprüngen und ein glattes Signal (zwei Gauß-Buckel und ein Kosinus). Beide werden in die diskrete Kosinustransformation (DCT, die Fourier-Basis von JPEG) und in die Haar-Wavelet-Basis zerlegt. Man behält nur die größten Koeffizienten und transformiert zurück. Links steht das Blocksignal mit beiden Rekonstruktionen aus 6 % der Koeffizienten, in der Mitte die sortierten Koeffizienten, rechts der relative Fehler gegen den behaltenen Anteil.

**Hintergrund & Fachbegriffe:**
- **Wavelet:** eine kurze Welle $\psi$, deren gestreckte und verschobene Kopien $\psi_{j,k}(x) = 2^{j/2}\psi(2^jx - k)$ eine Basis bilden; $j$ ist die Skala (Feinheit), $k$ die Lage.
- **Haar-Wavelet:** $\psi = +1$ auf der ersten und $-1$ auf der zweiten Hälfte eines Intervalls; das einfachste Wavelet (Haar 1909).
- **Skalierungsfunktion:** die Funktion, die die groben Mittelwerte trägt; beim Haar-System die Konstante auf einem Intervall.
- **DCT** (diskrete Kosinustransformation): eine Variante der DFT, die das Signal gespiegelt statt periodisch fortsetzt; die Basis von JPEG.
- **Nichtlineare Approximation:** Man behält die $m$ betragsgrößten Koeffizienten, gleich welche. In einer Orthonormalbasis ist der quadratische Fehler dann die Energie der weggelassenen (Parseval).
- **Relativer Fehler:** $\Vert\text{Näherung} - \text{Signal}\Vert/\Vert\text{Signal}\Vert$.

**Theorie:**

- **Schnelle Haar-Transformation:** Aus je zwei Nachbarwerten werden Mittelwert und Differenz (beide geteilt durch $\sqrt2$, damit die Transformation orthonormal ist). Die Differenzen sind die Wavelet-Koeffizienten der feinsten Stufe, mit den Mittelwerten wiederholt man das Ganze. Nach $\log_2N$ Stufen bleiben $N - 1$ Wavelet-Koeffizienten und ein Gesamtmittel, berechnet in $O(N)$ Schritten, schneller als die FFT. Diese Zerlegung in Stufen heißt **Multiskalenanalyse** (Mallat, Meyer 1989).
- **Lokal gegen global:** Ein Wavelet-Koeffizient sieht nur einen kurzen Abschnitt. Ein Sprung trifft auf jeder Stufe höchstens einen Koeffizienten, ein Stufensignal mit $J$ Sprüngen braucht also höchstens etwa $J\log_2N$ Koeffizienten (hier 55 für 8 Sprünge und 10 Stufen); alle anderen sind exakt 0. Eine Kosinusschwingung erstreckt sich über das ganze Signal: Ein Sprung macht alle Koeffizienten ungleich null (Abfall $1/k$, Exp. 1), und wenige Koeffizienten erzeugen das Gibbs-Phänomen.
- **Glatte Signale:** Hier fallen die DCT-Koeffizienten sehr schnell (Exp. 1); Haar nähert glatte Kurven durch Treppen, und der Fehler fällt nur langsam. Wavelets mit mehr **verschwindenden Momenten** (Daubechies 1988) behandeln auch glatte Abschnitte gut; JPEG 2000 benutzt die CDF-9/7-Wavelets.
- **Warum DCT statt DFT:** Die DFT setzt das Signal periodisch fort, und wenn Anfang und Ende verschieden sind, entsteht dabei ein Sprung. Die DCT spiegelt das Signal; die Fortsetzung ist stetig, und die Koeffizienten fallen schneller (Variation 6).

```python
def haar_dwt(x):
    """Orthonormale Haar-Wavelet-Transformation über alle Stufen (Länge 2^m). Gibt die Koeffizienten als ein Vektor zurück."""
    a, out = x.astype(float), []
    while len(a) > 1:
        out.append((a[0::2] - a[1::2]) / np.sqrt(2))                  # Details (Wavelet-Koeffizienten) dieser Stufe
        a = (a[0::2] + a[1::2]) / np.sqrt(2)                           # Mittelwerte (Skalierungskoeffizienten)
    return np.concatenate([a] + out[::-1])

def haar_idwt(c):
    a, pos = c[:1], 1
    while pos < len(c):
        d = c[pos:2 * pos]
        nxt = np.empty(2 * pos)
        nxt[0::2], nxt[1::2] = (a + d) / np.sqrt(2), (a - d) / np.sqrt(2)
        a, pos = nxt, 2 * pos
    return a

def keep_largest(c, frac):
    m = max(1, int(round(frac * len(c))))
    out = np.zeros_like(c)
    idx = np.argsort(np.abs(c))[-m:]
    out[idx] = c[idx]
    return out

n = 1024
u = np.arange(n) / n
jumps = [(0.1, 2.0), (0.13, -3.0), (0.32, 1.5), (0.45, 2.5), (0.6, -2.0), (0.65, 3.0), (0.78, -1.5), (0.82, 1.0)]
blocks = sum(h * (u >= p) for p, h in jumps)
smooth = np.exp(-((u - 0.3) / 0.08) ** 2) + 0.6 * np.exp(-((u - 0.7) / 0.12) ** 2) + 0.3 * np.cos(4 * np.pi * u)
bases = {"DCT (Fourier, wie JPEG)": (lambda s: sfft.dct(s, norm="ortho"), lambda c: sfft.idct(c, norm="ortho")),
         "Haar-Wavelets": (haar_dwt, haar_idwt)}
fracs = np.geomspace(0.005, 0.3, 20)

def rel_error(s, basis, frac):
    fwd, inv = bases[basis]
    return np.linalg.norm(inv(keep_largest(fwd(s), frac)) - s) / np.linalg.norm(s)

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
axes[0].plot(u, blocks, "k-", lw=2, label="Blocksignal")
for basis, c in zip(bases, ["tab:blue", "tab:red"]):
    fwd, inv = bases[basis]
    axes[0].plot(u, inv(keep_largest(fwd(blocks), 0.06)), c=c, lw=1, label=f"{basis}, 6 % der Koeffizienten")
axes[0].set_xlabel("x")
axes[0].set_title("Sprünge: Haar trifft, DCT schwingt", fontsize=10)
axes[0].legend(fontsize=7)
for (sname, s), ls in zip([("Blöcke", blocks), ("glatt", smooth)], ["-", "--"]):
    for basis, c in zip(bases, ["tab:blue", "tab:red"]):
        mag = np.sort(np.abs(bases[basis][0](s)))[::-1]
        axes[1].loglog(np.arange(1, n + 1), mag / mag[0], c=c, ls=ls, lw=1.2, label=f"{sname}, {basis}")
axes[1].set_ylim(1e-8, 2)
axes[1].set_xlabel("Rang (der größte zuerst)")
axes[1].set_ylabel("|Koeffizient| / größter")
axes[1].set_title("Sortierte Koeffizienten: wie schnell fallen sie?", fontsize=10)
axes[1].legend(fontsize=7)
for (sname, s), ls in zip([("Blöcke", blocks), ("glatt", smooth)], ["-", "--"]):
    for basis, c in zip(bases, ["tab:blue", "tab:red"]):
        axes[2].loglog(fracs, [rel_error(s, basis, fr) for fr in fracs], c=c, ls=ls, marker="o", ms=3, label=f"{sname}, {basis}")
axes[2].set_xlabel("Anteil der behaltenen Koeffizienten")
axes[2].set_ylabel("relativer Fehler")
axes[2].set_title("Kompression: Fehler gegen Anteil", fontsize=10)
axes[2].legend(fontsize=7)
plt.tight_layout()
plt.show()

print(f"Haar ist orthonormal: Rücktransformation exakt bis {np.max(np.abs(haar_idwt(haar_dwt(blocks)) - blocks)):.1e}, Energie erhalten {np.allclose(np.sum(haar_dwt(blocks) ** 2), np.sum(blocks ** 2))}")
for sname, s in [("Blöcke", blocks), ("glatt", smooth)]:
    print(f"{sname:6s}: relativer Fehler mit 6 % (3 %) der Koeffizienten: DCT {rel_error(s, 'DCT (Fourier, wie JPEG)', 0.06):.2e} ({rel_error(s, 'DCT (Fourier, wie JPEG)', 0.03):.2e}), "
          f"Haar {rel_error(s, 'Haar-Wavelets', 0.06):.2e} ({rel_error(s, 'Haar-Wavelets', 0.03):.2e}); "
          f"Haar-Koeffizienten ≠ 0: {np.sum(np.abs(haar_dwt(s)) > 1e-10)} von {n}")
```

* Im **linken Bild** trifft die Haar-Rekonstruktion (rot) aus 6 % der Koeffizienten (61 von 1024) das Blocksignal exakt; laut **Ausgabe** ist der relative Fehler $1.3\cdot10^{-15}$, weil das Blocksignal nur 55 Haar-Koeffizienten ungleich 0 hat. Die DCT-Rekonstruktion (blau) schwingt an jedem Sprung (Gibbs), relativer Fehler $8.6\cdot10^{-2}$. Im **mittleren Bild** fallen die sortierten Haar-Koeffizienten des Blocksignals nach Rang 55 senkrecht auf 0, seine DCT-Koeffizienten nur langsam; beim glatten Signal ist es umgekehrt: Die DCT-Koeffizienten (blau gestrichelt) stürzen ab, die Haar-Koeffizienten (rot gestrichelt) fallen gemächlich.
* Im **rechten Bild** fällt der Haar-Fehler für das Blocksignal bei etwa 5.4 % der Koeffizienten auf Rechengenauigkeit; darunter ist er so groß wie der der DCT (bei 3 %: Haar 0.085, DCT 0.122). Für das glatte Signal erreicht die DCT mit 6 % einen Fehler von $1.0\cdot10^{-5}$, Haar nur $3.3\cdot10^{-2}$. Das beantwortet die Vorhersage: Haar für Stufen, DCT für glatte Signale. Welche Basis gut ist, hängt davon ab, in welcher das Signal dünn besetzt ist.
* **Wo es vorkommt:** **JPEG** (DCT auf 8×8-Blöcken; bei starker Kompression sieht man Blockgrenzen und Schwingungen an Kanten), **JPEG 2000** und der Fingerabdruck-Standard des FBI (Wavelets), Entrauschen durch Abschneiden kleiner Wavelet-Koeffizienten (Donoho, Johnstone), Compressed Sensing in der MRT (Bilder sind in Wavelet-Basen dünn besetzt), Multiskalenanalyse in Geophysik und Turbulenz. Im maschinellen Lernen: Haar-artige Merkmale der Gesichtserkennung von Viola und Jones (2001) und die Wavelet-Streutransformation (Mallat) als festes, verständliches Gegenstück zu Faltungsnetzen.

##### *13. Die Wärmeleitungsgleichung: Fouriers ursprüngliches Problem*

Wie breitet sich Wärme auf einem Ring aus, und warum löst die Fourier-Reihe dieses Problem in einer Zeile? Vorher vorhersagen: Welche Fourier-Mode überlebt am längsten, und wie viel schneller zerfällt die Mode $k = 5$ als die Mode $k = 1$?

Auf einem Ring der Länge $2\pi$ (256 Gitterpunkte, Diffusionskonstante $D = 1$) ist anfangs ein Stück der Länge 2 heiß (Temperatur 1) und der Rest kalt. Die Lösung wird exakt über die Fourier-Reihe berechnet und mit der Faltung mit dem periodisierten Gauß-Kern verglichen. Links steht die Temperatur zu sechs Zeiten, in der Mitte die Beträge der Fourier-Koeffizienten zu denselben Zeiten, rechts der zeitliche Verlauf der Moden $k = 1, 2, 3, 5$. Die Ausgabe prüft außerdem ein einfaches Differenzenverfahren mit zwei Zeitschritten.

**Hintergrund & Fachbegriffe:**
- **Wärmeleitungsgleichung:** $\partial_tu = D\,\partial_x^2u$; die Temperatur steigt, wo sie gekrümmt ist wie ein Tal, und fällt, wo sie einen Gipfel hat.
- **Diffusionskonstante** $D$: wie schnell sich die Wärme ausbreitet.
- **Mode:** ein einzelner Summand $c_k(t)\,e^{ikx}$ der Fourier-Reihe.
- **Wärmeleitungskern:** die Lösung zu einem punktförmigen Anfangswert; auf der Geraden die Gauß-Funktion $G_t(x) = e^{-x^2/(4Dt)}/\sqrt{4\pi Dt}$.
- **Explizites Differenzenverfahren:** Ableitungen durch Differenzen benachbarter Gitterwerte ersetzen und in kleinen Zeitschritten vorwärts rechnen.
- **Stabilität:** Ein Verfahren ist stabil, wenn kleine Fehler (auch Rundungsfehler) nicht anwachsen.

**Theorie:**

- **Fouriers Lösung** (1807/1822): Setzt man $u = \sum_kc_k(t)\,e^{ikx}$ ein, wird jede Mode unabhängig: $c_k' = -Dk^2c_k$, also $c_k(t) = c_k(0)\,e^{-Dk^2t}$. Das klappt, weil die $e^{ikx}$ Eigenfunktionen der zweiten Ableitung sind (Theorie, Abschnitt 4): Die Fourier-Basis macht den Operator diagonal.
- **Folgen:** Hohe Frequenzen zerfallen zuerst, die Lösung wird glatter. Für jedes $t > 0$ fallen die Koeffizienten wie $e^{-Dk^2t}$, die Lösung ist also analytisch, auch wenn der Anfangswert springt (Exp. 1). Der Mittelwert $c_0$ bleibt erhalten (Energieerhaltung), und für lange Zeiten bleibt nur die langsamste Mode $k = 1$ mit Rate $D$ übrig.
- **Zwei Formeln, eine Lösung:** Auf der Geraden ist $u = u_0\ast G_t$. Auf dem Ring muss man die Gauß-Funktion periodisieren: $\sum_mG_t(x + 2\pi m) = \frac1{2\pi}\sum_ke^{-Dk^2t}e^{ikx}$, eine Theta-Funktion (Poissonsche Summenformel, Exp. 7). Die linke Seite (Spiegelbilder) konvergiert schnell für kleine $t$, die rechte (Fourier-Reihe) für große.
- **Rückwärts in der Zeit** wird aus $e^{-Dk^2t}$ ein $e^{+Dk^2t}$: Jeder Rundungsfehler in hohen Moden wird ungeheuer verstärkt (Variation 7). Die Vergangenheit eines Diffusionsprozesses lässt sich nicht stabil zurückrechnen.
- **Stabilität des Differenzenverfahrens** (von-Neumann-Analyse, auch eine Fourier-Analyse): Das Verfahren $u_j\leftarrow u_j + r\,(u_{j+1} - 2u_j + u_{j-1})$ mit $r = D\,\Delta t/\Delta x^2$ multipliziert die Mode $k$ in jedem Schritt mit $1 - 4r\sin^2(k\Delta x/2)$. Für $r > \tfrac12$ hat die höchste Mode einen Faktor vom Betrag größer 1 und explodiert.

```python
n, D = 256, 1.0
x = 2 * np.pi * np.arange(n) / n
u0 = (np.abs(x - np.pi) < 1).astype(float)                        # heißes Stück in der Mitte eines Rings
k = np.fft.fftfreq(n, 1 / n)
c0 = np.fft.fft(u0) / n

def heat_spectral(t):
    return np.real(np.fft.ifft(c0 * np.exp(-D * k ** 2 * t) * n))

def heat_kernel_images(t, images=20):
    """Lösung als Faltung mit der periodisierten Gauß-Funktion (Summe über Spiegelbilder, Poissonsche Summenformel)."""
    dxg = 2 * np.pi / n
    d = x[:, None] - x[None, :]
    G = sum(np.exp(-(d + 2 * np.pi * m) ** 2 / (4 * D * t)) for m in range(-images, images + 1)) / np.sqrt(4 * np.pi * D * t)
    return G @ u0 * dxg

times = [0, 0.001, 0.01, 0.05, 0.2, 1.0]
fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
for tt, c in zip(times, plt.cm.plasma(np.linspace(0, 0.85, len(times)))):
    axes[0].plot(x, heat_spectral(tt), c=c, label=f"t = {tt}")
    ck = np.abs(c0 * np.exp(-D * k ** 2 * tt))[: n // 2]
    sel = ck > 1e-16
    axes[1].semilogy(np.arange(n // 2)[sel], ck[sel], ".", c=c, ms=3, label=f"t = {tt}")
axes[0].set_xlabel("Ort x auf dem Ring")
axes[0].set_ylabel("Temperatur u(x, t)")
axes[0].set_title("Wärmeleitung auf einem Ring", fontsize=10)
axes[0].legend(fontsize=8)
axes[1].set_xlabel("Frequenz k")
axes[1].set_ylabel("|c_k(t)|")
axes[1].set_title("Jede Mode zerfällt wie e^(−Dk²t): hohe zuerst", fontsize=10)
axes[1].legend(fontsize=8)
ts = np.linspace(0, 2, 200)
for kk, c in zip([1, 2, 3, 5], ["tab:blue", "tab:orange", "tab:green", "tab:red"]):
    amp = [np.abs(np.fft.fft(heat_spectral(tt))[kk]) / n for tt in ts]
    axes[2].semilogy(ts, amp, c=c, lw=2, alpha=0.6, label=f"k = {kk} (gemessen)")
    axes[2].semilogy(ts, np.abs(c0[kk]) * np.exp(-D * kk ** 2 * ts), "k--", lw=0.8)
axes[2].set_ylim(1e-8, 1)
axes[2].set_xlabel("Zeit t")
axes[2].set_ylabel("|c_k(t)|")
axes[2].set_title("Zerfallsraten Dk² (gestrichelt: Theorie)", fontsize=10)
axes[2].legend(fontsize=8)
plt.tight_layout()
plt.show()

for tt in (0.01, 0.2):
    print(f"t = {tt}: Fourier-Lösung gegen Faltung mit dem periodisierten Gauß-Kern: größter Unterschied {np.max(np.abs(heat_spectral(tt) - heat_kernel_images(tt))):.1e}")
print(f"Mittelwert bleibt erhalten: {u0.mean():.6f} → {heat_spectral(1.0).mean():.6f};  Abweichung vom Mittelwert bei t = 2 und 3: "
      f"{np.max(np.abs(heat_spectral(2.0) - u0.mean())):.5f}, {np.max(np.abs(heat_spectral(3.0) - u0.mean())):.5f} (Verhältnis {np.max(np.abs(heat_spectral(3.0) - u0.mean())) / np.max(np.abs(heat_spectral(2.0) - u0.mean())):.4f}, Theorie e^−1 = {np.exp(-1):.4f})")
dx = 2 * np.pi / n
for factor in (0.45, 0.55):
    u, dt = u0.copy(), factor * dx ** 2 / D
    for _ in range(400):
        u = u + D * dt / dx ** 2 * (np.roll(u, 1) - 2 * u + np.roll(u, -1))
    print(f"Explizites Differenzenverfahren mit dt = {factor} · dx²/D, 400 Schritte: max |u| = {np.max(np.abs(u)):.3g} ({'stabil' if np.max(np.abs(u)) <= 1 else 'instabil'})")
```

* Im **linken Bild** rundet sich das heiße Stück sofort ab (schon bei $t = 0.001$ sind die Kanten weich), breitet sich aus und ist bei $t = 1$ fast eine einzelne Kosinuswelle um den Mittelwert 0.316. Im **mittleren Bild** fallen die Koeffizienten bei $t = 0$ nur wie $1/k$ (Sprünge); danach bilden sie auf der logarithmischen Achse nach unten gekrümmte Parabeln ($e^{-k^2t}$): Je höher $k$, desto schneller verschwindet die Mode, bei $t = 1$ sind nur noch wenige Moden über $10^{-16}$.
* Im **rechten Bild** liegen die gemessenen Amplituden der Moden $k = 1, 2, 3, 5$ auf den Geraden $e^{-k^2t}$ mit Raten 1, 4, 9 und 25: Die Mode $k = 5$ zerfällt 25-mal schneller als $k = 1$, und $k = 1$ überlebt am längsten (abgesehen vom Mittelwert $k = 0$). Das beantwortet die Vorhersage. Laut **Ausgabe** stimmen Fourier-Lösung und Faltung mit dem periodisierten Gauß-Kern auf $10^{-15}$ überein, der Mittelwert 0.316406 bleibt erhalten, und die Abweichung vom Mittelwert schrumpft von $t = 2$ bis $t = 3$ um den Faktor 0.3674 (Theorie $e^{-1} = 0.3679$). Das Differenzenverfahren bleibt mit $r = 0.45$ stabil (größter Wert 0.967) und explodiert mit $r = 0.55$ auf $5\cdot10^{29}$, genau an der Grenze $r = \tfrac12$.
* **Wo es vorkommt:** Wärmeleitung in Bauteilen und Gebäuden, Diffusion von Stoffen, die **Black-Scholes-Gleichung** der Optionsbewertung (sie lässt sich in die Wärmeleitungsgleichung umformen), der Gauß-Weichzeichner der Bildverarbeitung (Glätten ist Wärmeleitung, mehrere Glättungsstufen bilden einen Skalenraum), **Diffusionsmodelle** im maschinellen Lernen (beim Verrauschen gehen hohe Frequenzen zuerst verloren, und das Erzeugen eines Bildes baut sie von grob nach fein wieder auf) und spektrale Wettermodelle, die auf der Kugel mit Kugelflächenfunktionen statt mit $e^{ikx}$ rechnen.

##### *14. Beugung: Fourier-Optik in zwei Dimensionen*

Was hat Beugung mit der Fourier-Transformation zu tun? Vorher vorhersagen: Wo liegt der erste dunkle Ring hinter einer Kreisöffnung mit Durchmesser $D$, gemessen in Raumfrequenz $\xi = \sin\theta/\lambda$?

Eine Kreisöffnung mit Radius 1 und ein Doppelspalt (Spaltbreite 0.2, Abstand 1.0, Höhe 8) werden auf einem Gitter von $1024\times1024$ Punkten dargestellt und mit der zweidimensionalen FFT transformiert. Oben stehen Öffnung, Beugungsbild (logarithmisch) und das über Ringe gemittelte Intensitätsprofil des Kreises im Vergleich mit der Airy-Formel; unten Öffnung, Beugungsbild und ein Schnitt durch das Muster des Doppelspalts im Vergleich mit $\cos^2\cdot\mathrm{sinc}^2$. Die Pixelmitten liegen so, dass die Spaltränder genau zwischen zwei Pixel fallen.

**Hintergrund & Fachbegriffe:**
- **Fraunhofer-Beugung:** das Beugungsmuster weit hinter einer Öffnung (Fernfeld), oder in der Brennebene einer Linse.
- **Raumfrequenz** $\xi = \sin\theta/\lambda$: Richtung $\theta$ geteilt durch die Wellenlänge $\lambda$; sie zählt Streifen pro Längeneinheit in der Öffnung.
- **Intensität:** das Betragsquadrat der Amplitude, das, was ein Schirm oder eine Kamera misst.
- **Airy-Scheibchen:** das helle Zentrum des Beugungsbilds einer Kreisöffnung, umgeben von dunklen und hellen Ringen.
- **Bessel-Funktion** $J_1$: eine Schwingungsfunktion, die bei rotationssymmetrischen Problemen an die Stelle des Sinus tritt; ihre erste positive Nullstelle liegt bei 3.8317.
- **Hülle:** die langsame Kurve, die die schnellen Streifen umschließt.

**Theorie:**

- **Fernfeld = Fourier-Transformation:** Nach Huygens sendet jeder Punkt der Öffnung eine Kugelwelle aus. Weit entfernt (Abstand groß gegen $D^2/\lambda$) addieren sie sich in Richtung $\theta$ mit dem Gangunterschied $x\sin\theta$, also mit der Phase $e^{-2\pi ix\xi}$: Die Amplitude ist $\hat A(\xi_x, \xi_y)$, die zweidimensionale Fourier-Transformierte der Öffnungsfunktion $A(x, y)$ (1 im Loch, 0 sonst). Gemessen wird $\vert\hat A\vert^2$; die Phase geht verloren.
- **Kreis:** Für eine rotationssymmetrische Funktion wird die 2D-Transformation zur Hankel-Transformation mit der Bessel-Funktion. Für Radius $a$ ist $I(\rho)\propto\Big[\frac{2J_1(2\pi a\rho)}{2\pi a\rho}\Big]^2$. Der erste dunkle Ring liegt bei $2\pi a\rho = 3.8317$, also $\rho = 0.6098/a = 1.22/D$, in Winkeln $\theta\approx1.22\,\lambda/D$ (Airy 1835). Im zentralen Scheibchen stecken 83.8 % der Energie.
- **Doppelspalt:** Die Öffnung ist die Faltung eines Spalts der Breite $w$ mit zwei Punkten im Abstand $d$. Nach dem Faltungssatz ist ihre Transformierte das Produkt $2\cos(\pi d\xi)\cdot w\,\mathrm{sinc}(w\xi)$, die Intensität also $\cos^2(\pi d\xi)\,\mathrm{sinc}^2(w\xi)$: Streifen im Abstand $1/d$ unter der Hülle eines einzelnen Spalts mit Nullstellen bei Vielfachen von $1/w$.
- **Streckung:** Eine kleinere Öffnung gibt ein größeres Beugungsbild (Variation 8), das ist das Unschärfeprinzip in der Optik und begrenzt die Auflösung jedes Teleskops und Mikroskops.

```python
n, L = 1024, 51.2
xs = (np.arange(n) - n // 2 + 0.5) * L / n                       # Pixelmitten bei ±0.025, ±0.075, …: Spaltränder fallen zwischen Pixel
X, Y = np.meshgrid(xs, xs)
circle = (X ** 2 + Y ** 2 <= 1.0).astype(float)                   # kreisförmige Öffnung, Radius a = 1 (Durchmesser D = 2)
w, d = 0.2, 1.0
slits = ((np.abs(np.abs(X) - d / 2) < w / 2) & (np.abs(Y) < 4)).astype(float)   # Doppelspalt: Breite w, Abstand d
xi = np.fft.fftshift(np.fft.fftfreq(n, L / n))

def far_field(aperture):
    """Intensität im Fernfeld = |2D-Fourier-Transformierte der Öffnung|², auf das Maximum normiert."""
    I = np.abs(np.fft.fftshift(np.fft.fft2(np.fft.ifftshift(aperture)))) ** 2
    return I / I.max()

I_circ, I_slit = far_field(circle), far_field(slits)
R = np.hypot(*np.meshgrid(xi, xi))
rbins = np.arange(0, 2.0001, 2 * (xi[1] - xi[0]))
which = np.digitize(R.ravel(), rbins)
sums = np.bincount(which, weights=I_circ.ravel(), minlength=len(rbins) + 1)[1:len(rbins)]
counts = np.bincount(which, minlength=len(rbins) + 1)[1:len(rbins)]
rc = 0.5 * (rbins[1:] + rbins[:-1])[counts > 0]
radial = sums[counts > 0] / counts[counts > 0]                  # Mittel über Ringe der Breite zweier Frequenzschritte
with np.errstate(invalid="ignore", divide="ignore"):
    airy = np.where(rc > 0, (2 * special.j1(2 * np.pi * rc) / (2 * np.pi * rc)) ** 2, 1.0)
cut = I_slit[n // 2]
theory_slit = np.cos(np.pi * d * xi) ** 2 * np.sinc(w * xi) ** 2

fig, axes = plt.subplots(2, 3, figsize=(17, 8.6))
ext_x, ext_k = [xs[0], xs[-1], xs[0], xs[-1]], [xi[0], xi[-1], xi[0], xi[-1]]
for row, (ap, I, name) in enumerate([(circle, I_circ, "Kreisöffnung, Radius 1"), (slits, I_slit, f"Doppelspalt, Breite {w}, Abstand {d}")]):
    axes[row, 0].imshow(ap, extent=ext_x, cmap="gray", origin="lower")
    axes[row, 0].set_xlim(-3, 3)
    axes[row, 0].set_ylim(-3, 3) if row == 0 else axes[row, 0].set_ylim(-5, 5)
    axes[row, 0].set_title(f"Öffnung: {name}", fontsize=10)
    axes[row, 0].set_xlabel("x"); axes[row, 0].set_ylabel("y")
    axes[row, 1].imshow(np.log10(I + 1e-8), extent=ext_k, cmap="inferno", origin="lower", vmin=-5, vmax=0)
    axes[row, 1].set_xlim(-3, 3); axes[row, 1].set_ylim(-3, 3)
    axes[row, 1].set_title("Beugungsbild log₁₀ |FT|²", fontsize=10)
    axes[row, 1].set_xlabel("Raumfrequenz ξ_x"); axes[row, 1].set_ylabel("Raumfrequenz ξ_y")
axes[0, 2].semilogy(rc, radial, "o", ms=3, label="gemessen (Mittel über Ringe)")
axes[0, 2].semilogy(rc, airy, "k-", lw=1, label="Airy: [2J₁(2πρ)/(2πρ)]²")
axes[0, 2].axvline(0.6098, color="tab:red", ls="--", lw=1, label="erster dunkler Ring ρ = 1.22/D")
axes[0, 2].set_ylim(1e-6, 1.5)
axes[0, 2].set_xlabel("radiale Raumfrequenz ρ")
axes[0, 2].set_ylabel("Intensität")
axes[0, 2].set_title("Kreis: Airy-Scheibchen", fontsize=10)
axes[0, 2].legend(fontsize=8)
sel = np.abs(xi) <= 6
axes[1, 2].plot(xi[sel], cut[sel], lw=1, label="gemessen (Schnitt ξ_y = 0)")
axes[1, 2].plot(xi[sel], theory_slit[sel], "k--", lw=1, label="cos²(πdξ) · sinc²(wξ)")
axes[1, 2].plot(xi[sel], np.sinc(w * xi[sel]) ** 2, ":", c="grey", label="Hülle sinc²(wξ): ein Spalt")
axes[1, 2].set_xlabel("Raumfrequenz ξ_x")
axes[1, 2].set_ylabel("Intensität")
axes[1, 2].set_title("Doppelspalt: Streifen im Abstand 1/d", fontsize=10)
axes[1, 2].legend(fontsize=8)
plt.tight_layout()
plt.show()

inner = (rc > 0.3) & (rc < 0.8)
first_zero = rc[inner][np.argmin(radial[inner])]
print(f"Erster dunkler Ring: gemessen bei ρ = {first_zero:.3f} (Raster {rbins[1] - rbins[0]:.3f}), Theorie 3.8317/(2πa) = {3.8317 / (2 * np.pi):.4f} = 1.22/D")
print(f"Anteil der Energie im zentralen Scheibchen: {I_circ[R < 0.6098].sum() / I_circ.sum():.3f} (Theorie 0.838)")
mins = xi[sel][1:-1][(cut[sel][1:-1] < cut[sel][:-2]) & (cut[sel][1:-1] < cut[sel][2:])]
print(f"Doppelspalt: Minima bei ξ = {np.round(mins[np.abs(mins) < 3], 3)} (Theorie ±0.5, ±1.5, ±2.5: Abstand 1/d = {1 / d})")
```

* **Obere Reihe:** Die Kreisöffnung erzeugt das Airy-Muster aus konzentrischen Ringen; die leichte Eckigkeit der äußeren Ringe kommt vom treppenförmigen Rand der Öffnung auf dem Pixelgitter. Rechts liegt das über Ringe gemittelte Profil über vier Größenordnungen auf der Airy-Formel. Laut **Ausgabe** liegt der erste dunkle Ring bei $\rho = 0.605$ (Ringbreite 0.039) statt $0.6098 = 1.22/D$, und im zentralen Scheibchen stecken 0.838 der Energie, wie vorhergesagt.
* **Untere Reihe:** Der Doppelspalt erzeugt senkrechte Streifen bei ganzzahligen $\xi_x$; ihre Helligkeit folgt der Hülle eines einzelnen Spalts (gepunktet), die bei $1/w = 5$ ihre erste Nullstelle hat, deshalb sind die Streifen bei $\pm4$ schon schwach. Die waagerechte Streifung in $\xi_y$ kommt von der endlichen Spalthöhe 8. Der Schnitt bei $\xi_y = 0$ liegt auf $\cos^2(\pi d\xi)\,\mathrm{sinc}^2(w\xi)$; die Minima liegen laut Ausgabe bei $\pm0.508$, $\pm1.504$, $\pm2.5$ (Theorie $\pm0.5$, $\pm1.5$, $\pm2.5$; Frequenzraster 0.0195), im Abstand $1/d = 1$.
* **Wo es vorkommt:** Das **Auflösungsvermögen** von Teleskopen und Mikroskopen (Rayleigh-Kriterium $1.22\,\lambda/D$; die sechs Beugungsstrahlen auf Bildern des James-Webb-Teleskops kommen von den sechseckigen Spiegelsegmenten und den Streben). Die **Röntgenkristallographie** misst das Betragsquadrat der Fourier-Transformierten der Elektronendichte eines Kristalls und muss die fehlende Phase rekonstruieren (Phasenproblem; das Foto 51 der DNA-Doppelhelix ist ein solches Beugungsbild). Die **Radiointerferometrie** misst die 2D-Fourier-Transformierte der Himmelshelligkeit (Satz von van Cittert-Zernike; so entstand das Bild des Schwarzen Lochs in M87). Dazu kommen Holographie, die Lithographie von Computerchips und Gitterspektrometer.

##### *15. Harmonische Analysis auf endlichen Gruppen: Hyperwürfel und Periodenfindung*

Wie sieht Fourier-Analyse auf einer endlichen Gruppe aus, und was hat sie mit Quantencomputern zu tun? Vorher vorhersagen: Auf welcher Stufe $\vert S\vert$ liegt das Fourier-Gewicht der Mehrheitsfunktion von 9 Bits, und wo das der Parität? Und wo liegen die Gipfel, wenn man die Periode von $2^x\bmod21$ per Fourier-Abtastung sucht?

Links steht die Walsh-Hadamard-Matrix für $\mathbb Z_2^4$: Jede Zeile ist ein Charakter. In der Mitte werden vier boolesche Funktionen von 9 Bits mit der schnellen Walsh-Hadamard-Transformation zerlegt (Diktator, Mehrheit, Parität, eine zufällige Funktion), und ihr Fourier-Gewicht wird nach Stufen $\vert S\vert$ aufgeteilt. Rechts wird die Periodenfindung aus Shors Algorithmus klassisch nachgerechnet: Für $f(x) = 2^x\bmod21$ auf $\mathbb Z_{512}$ wird die Wahrscheinlichkeitsverteilung der Messwerte nach der Quanten-Fourier-Transformation berechnet. Die Ausgabe prüft Parseval, gibt Gewichte und Einflüsse, die Wahrscheinlichkeit an den Gipfeln und die Kettenbrüche.

**Hintergrund & Fachbegriffe:**
- **Hyperwürfel** $\mathbb Z_2^n$: die $2^n$ Bitstrings der Länge $n$, mit bitweiser Addition (XOR) als Gruppenoperation.
- **Charakter** $\chi_S(x) = (-1)^{S\cdot x}$: das Vorzeichen, das von der Parität der Bits von $x$ an den Stellen $S$ abhängt. $\vert S\vert$, die Zahl der beteiligten Bits, heißt **Stufe**.
- **Walsh-Hadamard-Transformation:** die Fourier-Transformation auf $\mathbb Z_2^n$; ihre Matrix ist das $n$-fache Tensorprodukt der $2\times2$-Matrix mit den Zeilen $(1, 1)$ und $(1, -1)$.
- **Boolesche Funktion:** $f:\lbrace-1, 1\rbrace^n\to\lbrace-1, 1\rbrace$. **Diktator** $f = x_1$; **Mehrheit** = Vorzeichen der Summe; **Parität** = Produkt aller Bits.
- **Einfluss** von Bit $i$: die Wahrscheinlichkeit, dass das Umdrehen von Bit $i$ den Funktionswert ändert. Der **Gesamteinfluss** ist die Summe über alle Bits.
- **Periodenfindung:** Gesucht ist die kleinste Zahl $r > 0$ mit $f(x + r) = f(x)$ für alle $x$.
- **Kettenbruch:** die Darstellung einer Zahl als $a_0 + 1/(a_1 + 1/(a_2 + \dots))$; ihre Abbrüche liefern die besten Brüche mit kleinem Nenner.

**Theorie:**

- **Endliche abelsche Gruppen:** $\hat f(\chi) = \frac1{\vert G\vert}\sum_gf(g)\overline{\chi(g)}$, $f = \sum_\chi\hat f(\chi)\,\chi$; Plancherel $\frac1{\vert G\vert}\sum\vert f\vert^2 = \sum\vert\hat f\vert^2$ und der Faltungssatz gelten wie in den anderen Fällen. Für $\mathbb Z_N$ ist das die DFT, für $\mathbb Z_2^n$ die Walsh-Hadamard-Transformation $\hat f(S) = 2^{-n}\sum_xf(x)(-1)^{S\cdot x}$, schnell berechenbar in $n\cdot2^n$ Schritten.
- **Boolesche Funktionen:** Für $f$ mit Werten $\pm1$ ist $\sum_S\hat f(S)^2 = 1$; die Gewichte sind also eine Wahrscheinlichkeitsverteilung über die Stufen. Der Diktator hat alles auf Stufe 1, die Parität alles auf Stufe $n$. Die Mehrheit hat auf Stufe 1 das Gewicht $n\big(\binom{n-1}{(n-1)/2}2^{1-n}\big)^2$, für $n = 9$ also 0.673 und für $n\to\infty$ gegen $2/\pi$; der Rest liegt auf ungeraden Stufen, weil $f(-x) = -f(x)$. Eine zufällige Funktion verteilt ihr Gewicht binomial um $n/2$. Der Einfluss von Bit $i$ ist $\sum_{S\ni i}\hat f(S)^2$, der Gesamteinfluss $\sum_S\vert S\vert\hat f(S)^2$.
- **Quantenrechner:** Ein Hadamard-Gatter auf jedem von $n$ Qubits ist die Walsh-Hadamard-Transformation der $2^n$ Amplituden; die Quanten-Fourier-Transformation ist die DFT auf $\mathbb Z_{2^n}$ mit $O(n^2)$ Gattern.
- **Periodenfindung** (Shor 1994): Man präpariert $\frac1{\sqrt N}\sum_x\vert x\rangle\vert f(x)\rangle$ und misst das zweite Register. Übrig bleibt eine gleichmäßige Überlagerung der $x$ mit gleichem Funktionswert, ein verschobener Kamm mit Abstand $r$. Seine Fourier-Transformierte ist ein Kamm mit Abstand $N/r$ (Exp. 5); die Messung liefert $y\approx jN/r$ für ein zufälliges $j$. Der Kettenbruch von $y/N$ ergibt $j/r$ in gekürzter Form; ist $j$ zu $r$ teilerfremd, ist der Nenner $r$, sonst ein Teiler davon, und man wiederholt. Aus $r$ folgt die Faktorisierung: $\mathrm{ggT}(a^{r/2}\pm1, M)$.
- **Verstecktes-Untergruppen-Problem:** Periodenfindung, diskreter Logarithmus und Simons Algorithmus sind Fälle davon; für abelsche Gruppen löst es die Fourier-Abtastung effizient, für nicht abelsche (Graphenisomorphie, Gitterprobleme) ist das offen.

```python
def walsh_hadamard(f):
    """Schnelle Walsh-Hadamard-Transformation: Σ_x f(x) (−1)^{S·x} für alle S, in n · 2^n Schritten."""
    a, h = np.array(f, dtype=float), 1
    while h < len(a):
        a = a.reshape(-1, 2 * h)
        left, right = a[:, :h].copy(), a[:, h:].copy()
        a[:, :h], a[:, h:] = left + right, left - right
        a, h = a.reshape(-1), 2 * h
    return a

rng = np.random.default_rng(15)
nbits = 9
xs_int = np.arange(2 ** nbits)
bits = (xs_int[:, None] >> np.arange(nbits)) & 1
pm = 1 - 2 * bits                                                 # Bits als ±1
boolean = {
    "Diktator x₁": pm[:, 0],
    "Mehrheit von 9": np.sign(pm.sum(axis=1)),
    "Parität x₁⋯x₉": np.prod(pm, axis=1),
    "zufällige Funktion": rng.choice([-1, 1], size=2 ** nbits),
}
level = bits.sum(axis=1)                                          # |S| für den Index S
weights = {}
for name, f in boolean.items():
    fhat = walsh_hadamard(f) / 2 ** nbits
    weights[name] = np.array([np.sum(fhat[level == j] ** 2) for j in range(nbits + 1)])

Nq, M, base = 512, 21, 2
fx = np.array([pow(base, int(v), M) for v in range(Nq)])
P = np.zeros(Nq)
for val in np.unique(fx):
    P += np.abs(np.fft.fft((fx == val).astype(float))) ** 2
P /= Nq ** 2
r = 6

fig, axes = plt.subplots(1, 3, figsize=(17, 4.3))
H = np.array([walsh_hadamard(row) for row in np.eye(16)])
axes[0].imshow(H, cmap="gray")
axes[0].set_title("Charaktere von ℤ₂⁴: Walsh-Hadamard-Matrix (16 × 16)", fontsize=10)
axes[0].set_xlabel("x")
axes[0].set_ylabel("S")
wdt = 0.2
for i, (name, wv) in enumerate(weights.items()):
    axes[1].bar(np.arange(nbits + 1) + (i - 1.5) * wdt, wv, width=wdt, label=name)
axes[1].set_xlabel("Stufe |S| (Zahl der Variablen im Charakter)")
axes[1].set_ylabel("Fourier-Gewicht Σ_{|S|=k} f̂(S)²")
axes[1].set_title("Boolesche Funktionen auf {−1, 1}⁹", fontsize=10)
axes[1].legend(fontsize=8)
axes[2].plot(np.arange(Nq), P, lw=0.8)
for j in range(r):
    axes[2].axvline(j * Nq / r, color="tab:red", ls=":", lw=0.8)
axes[2].set_xlabel("Messwert y")
axes[2].set_ylabel("Wahrscheinlichkeit")
axes[2].set_title(f"Fourier-Abtastung von 2^x mod {M} auf ℤ_{Nq}: Gipfel bei j·{Nq}/{r}", fontsize=10)
plt.tight_layout()
plt.show()

for name, wv in weights.items():
    infl = np.sum(np.arange(nbits + 1) * wv)
    print(f"{name:18s}: Σ f̂(S)² = {wv.sum():.3f};  Gewicht auf Stufe 1: {wv[1]:.3f};  Gesamteinfluss Σ|S| f̂(S)² = {infl:.3f}")
print(f"Mehrheit: Stufe-1-Gewicht exakt 9·(C(8,4)/2⁸)² = {9 * (special.comb(8, 4) / 2 ** 8) ** 2:.3f}, für n → ∞ gegen 2/π = {2 / np.pi:.3f}")
near = np.zeros(Nq, bool)
for j in range(r):
    near[[int(np.floor(j * Nq / r)) % Nq, int(np.ceil(j * Nq / r)) % Nq]] = True
print(f"Periode von 2^x mod {M}: {r}; Wahrscheinlichkeit, einen der beiden nächsten Werte zu j·{Nq}/{r} zu messen: {P[near].sum():.3f}")
from math import gcd
h = pow(base, r // 2, M)
print(f"Aus r = {r}: {base}^{r // 2} mod {M} = {h};  ggT({h} − 1, {M}) = {gcd(h - 1, M)},  ggT({h} + 1, {M}) = {gcd(h + 1, M)}  →  {M} = {gcd(h + 1, M)} · {gcd(h - 1, M)}")
for y in (85, 171, 427):
    fr = Fraction(y, Nq).limit_denominator(M)
    print(f"Messwert y = {y}: y/{Nq} ≈ {fr} → Nenner {fr.denominator}" + ("  (Teiler von r; ein zweiter Messwert gibt r = 6)" if fr.denominator != r else "  (= Periode)"))
```

<details><summary>1. Was ist ein Fourier-Koeffizient, anschaulich?</summary>

Die Projektion der Funktion auf eine Schwingung $e^{ikx}$, berechnet als Skalarprodukt $c_k = \frac1{2\pi}\int f(x)e^{-ikx}\,\mathrm dx$, genau wie die Koordinate eines Vektors in einer Orthonormalbasis. Sein Betrag ist die Amplitude, sein Argument die Phase dieser Schwingung.
</details>

<details><summary>2. Welche vier Fourier-Transformationen gibt es, und welche Regel verbindet sie?</summary>

Fourier-Reihe (periodisch, stetig → diskrete Frequenzen), Fourier-Transformation (Gerade → Gerade), DTFT (Abtastwerte → periodisches Spektrum), DFT ($N$ Werte → $N$ Frequenzen). Regel: Diskret auf der einen Seite heißt periodisch auf der anderen.
</details>

<details><summary>3. Warum zerlegt man ausgerechnet in Schwingungen?</summary>

Weil Schwingungen (die Charaktere der Gruppe) Eigenfunktionen jeder Verschiebung sind. Deshalb werden alle verschiebungsinvarianten linearen Operatoren in der Fourier-Basis diagonal: Ableitungen, Faltungen, Filter, Differentialgleichungen mit konstanten Koeffizienten.
</details>

<details><summary>4. Wie hängt die Glattheit einer Funktion mit dem Abfall ihrer Fourier-Koeffizienten zusammen?</summary>

Jede Ableitung multipliziert die Koeffizienten mit $ik$. Springt erst die $m$-te Ableitung, fallen die Koeffizienten wie $1/k^{m+1}$: Sprung $1/k$, Knick $1/k^2$. Analytische Funktionen haben exponentiell fallende Koeffizienten.
</details>

<details><summary>5. Was ist das Gibbs-Phänomen, und warum verschwindet es nicht mit mehr Frequenzen?</summary>

Das Überschwingen der Partialsummen an einem Sprung um $8.95\,\%$ der Sprunghöhe. Die Partialsumme ist eine Faltung mit dem Dirichlet-Kern, der negative Nebenschwinger hat; mit wachsendem $N$ wandert das Überschwingen näher an den Sprung, seine Höhe bleibt aber bei $8.95\,\%$ der Sprunghöhe.
</details>

<details><summary>6. Was leisten die Fejér-Mittel, und was kosten sie?</summary>

Sie mitteln die Partialsummen; das ist eine Faltung mit dem positiven Fejér-Kern. Dadurch gibt es kein Überschwingen, und für stetige Funktionen konvergieren sie gleichmäßig. Dafür konvergieren sie an glatten Stellen langsamer, und ihr quadratischer Fehler ist größer als der der Partialsumme.
</details>

<details><summary>7. Was sagt die Parseval-Gleichung, und wie folgt daraus ζ(2) = π²/6?</summary>

Die Energie einer Funktion ist die Summe der Betragsquadrate ihrer Koeffizienten. Beim Sägezahn $f(x) = x$ ist $\vert c_k\vert^2 = 1/k^2$ und $\frac1{2\pi}\int x^2 = \pi^2/3$, also $2\zeta(2) = \pi^2/3$.
</details>

<details><summary>8. Warum ist die Partialsumme die beste Näherung im quadratischen Mittel?</summary>

Weil sie die orthogonale Projektion auf die trigonometrischen Polynome vom Grad $\leq N$ ist. Ihr Fehler steht senkrecht auf diesem Raum, und nach Pythagoras hat jede andere Wahl einen größeren Fehler.
</details>

<details><summary>9. Welche Eigenwerte hat die Fourier-Transformation, und welche Funktionen sind Eigenfunktionen?</summary>

Viermal angewendet ergibt sie die Identität, also kommen nur $1, -i, -1, i$ vor. Eigenfunktionen sind die Hermite-Funktionen $h_n$ mit Eigenwert $(-i)^n$; die Gauß-Funktion ist $h_0$ mit Eigenwert 1.
</details>

<details><summary>10. Was besagt das Unschärfeprinzip, und welche Funktion erreicht die Grenze?</summary>

$\Delta x\cdot\Delta\xi\geq1/(4\pi)$: Eine Funktion und ihre Transformierte können nicht beide schmal sein. Gleichheit gilt nur für Gauß-Funktionen (auch verschoben und moduliert). In der Quantenmechanik wird daraus $\Delta x\,\Delta p\geq\hbar/2$.
</details>

<details><summary>11. Was sagt Donoho-Stark auf ℤ_N, und welche Vektoren erreichen die Grenze?</summary>

Die Zahl der Einträge ungleich 0 in $f$ mal die Zahl in $\hat f$ ist mindestens $N$. Gleichheit erreichen Dirac-Kämme mit Abstand $d$, wenn $d$ ein Teiler von $N$ ist; für Primzahlen $N$ gilt sogar die Summe $\geq N + 1$.
</details>

<details><summary>12. Was sagt der Faltungssatz, und warum braucht man beim Falten mit der FFT Nullen?</summary>

Die Transformierte einer Faltung ist das Produkt der Transformierten. Die DFT rechnet zyklisch, das Ende wickelt sich an den Anfang; mit Nullen auf mindestens $N + M - 1$ verlängert, wird die zyklische Faltung zur gewöhnlichen.
</details>

<details><summary>13. Was ist die Poissonsche Summenformel, und warum hilft sie beim Rechnen?</summary>

$\sum_nf(n) = \sum_k\hat f(k)$. Ist $f$ breit und glatt, ist $\hat f$ schmal, und die rechte Seite konvergiert schnell, wo die linke langsam ist; so berechnet man Theta-Funktionen, Gittersummen (Ewald) und $\sum1/(n^2 + a^2)$.
</details>

<details><summary>14. Warum ist die FFT schnell?</summary>

Sie teilt die DFT der Länge $N$ in zwei DFTs der Länge $N/2$ (gerade und ungerade Indizes) und setzt sie mit $N$ Multiplikationen zusammen. Wiederholt ergibt das $N\log_2N$ statt $N^2$ Schritte.
</details>

<details><summary>15. Was sagt das Abtasttheorem, und was ist Aliasing?</summary>

Ein auf $[-B, B]$ bandbegrenztes Signal ist durch Abtastwerte mit $f_s > 2B$ eindeutig bestimmt und per sinc-Interpolation rekonstruierbar. Tastet man langsamer ab, überlappen die Kopien des Spektrums, und hohe Frequenzen erscheinen als falsche tiefe (gespiegelt an $f_s/2$).
</details>

<details><summary>16. Warum benutzt man Fensterfunktionen, und welchen Preis haben sie?</summary>

Ein endliches Stück eines Signals ist mit einem Fenster multipliziert, sein Spektrum also mit dem Spektrum des Fensters gefaltet. Glatte Fenster haben kleine Nebenkeulen, sodass schwache Töne neben starken sichtbar werden; dafür ist die Hauptkeule breiter und die Frequenzauflösung schlechter.
</details>

<details><summary>17. Warum kann ein Spektrogramm Zeit und Frequenz nicht zugleich beliebig fein auflösen?</summary>

Ein Fenster der Länge $T_w$ löst die Zeit auf etwa $T_w$ und die Frequenz auf etwa $1/T_w$ auf; das Produkt ist durch das Unschärfeprinzip nach unten begrenzt. Kurze Fenster zeigen Klicks, lange Fenster trennen nahe Töne.
</details>

<details><summary>18. Wann sind Wavelets besser als die Fourier-Basis, und wann nicht?</summary>

Wavelets sind lokal: Ein Sprung trifft nur wenige Koeffizienten pro Skala, Signale mit Sprüngen sind in der Wavelet-Basis dünn besetzt. Für glatte Signale fallen die Fourier- oder DCT-Koeffizienten schneller, dort ist Fourier besser.
</details>

<details><summary>19. Wie löste Fourier die Wärmeleitungsgleichung, und warum wird die Lösung sofort glatt?</summary>

In der Fourier-Basis entkoppeln die Moden: $c_k(t) = c_k(0)e^{-Dk^2t}$. Hohe Frequenzen zerfallen am schnellsten; für jedes $t > 0$ fallen die Koeffizienten wie $e^{-Dk^2t}$, die Lösung ist also beliebig glatt.
</details>

<details><summary>20. Warum ist das Beugungsbild einer Öffnung ihre Fourier-Transformierte, und wo liegt der erste dunkle Ring hinter einer Kreisblende?</summary>

Im Fernfeld addieren sich die Wellen aller Punkte der Öffnung mit der Phase $e^{-2\pi ix\xi}$, $\xi = \sin\theta/\lambda$; das ist die Fourier-Transformation der Öffnung. Der erste dunkle Ring liegt bei $\xi = 1.22/D$, also $\theta\approx1.22\,\lambda/D$.
</details>

<details><summary>21. Wie findet Shors Algorithmus eine Periode, und welche Rolle spielt dabei die Fourier-Transformation?</summary>

Nach dem Messen des Funktionswerts bleibt eine Überlagerung der Eingaben mit Abstand $r$, ein Kamm. Die Quanten-Fourier-Transformation macht daraus einen Kamm mit Abstand $N/r$; die Messung liefert $y\approx jN/r$, und der Kettenbruch von $y/N$ ergibt $r$ (oder einen Teiler davon).
</details>
