# Variationsrechnung

Welche Kurve macht ein Integral über eine Funktion und ihre Ableitung stationär, und was folgt daraus für Erhaltungsgrößen, für die schnellste Rutschbahn und für kürzeste Wege auf einer Kugel?

## Theorie

**Funktional.** $J[y] = \int_a^b L(x, y, y')\,dx$ für Funktionen $y$ mit festen Randwerten $y(a) = y_a$, $y(b) = y_b$. Gesucht sind Funktionen, für die $J$ stationär ist: Eine kleine Änderung $y\to y + \epsilon\eta$ mit $\eta(a) = \eta(b) = 0$ ändert $J$ erst in zweiter Ordnung in $\epsilon$.

**Euler–Lagrange-Gleichung.**
$$\frac{\partial L}{\partial y} - \frac{d}{dx}\frac{\partial L}{\partial y'} = 0 .$$

**Beltrami-Identität.** Hängt $L$ nicht explizit von $x$ ab, so ist $L - y'\,\partial L/\partial y'$ entlang jeder Lösung konstant.

**Mechanik.** Mit Koordinaten $q_i(t)$ und $L = T - V$ wird die Euler–Lagrange-Gleichung zur Bewegungsgleichung (Prinzip der stationären Wirkung), eine Gleichung pro Koordinate.

**Noether-Theorem.** Ist $L$ invariant unter $q\to q + \epsilon K(q)$, so ist $Q = \sum_i\frac{\partial L}{\partial\dot q_i}K_i$ erhalten. Zeitverschiebung ergibt die Energie $H = \sum_i\dot q_i\,\partial L/\partial\dot q_i - L$, Verschiebung den Impuls, Drehung den Drehimpuls.

**Brachistochrone.** Ein Körper gleitet reibungsfrei unter der Schwerkraft von $(0, 0)$ nach $(x_1, y_1)$, mit $y$ nach unten. Die Laufzeit
$$T[y] = \int_0^{x_1}\frac{\sqrt{1 + y'^2}}{\sqrt{2gy}}\,dx$$
ist minimal für die Zykloide $x = a(\theta - \sin\theta)$, $y = a(1 - \cos\theta)$.

**Geodäten.** Auf einer Fläche mit der Metrik $ds^2 = g_{ij}\,dx^i dx^j$ sind Geodäten die stationären Kurven des Energiefunktionals $\tfrac12\int g_{ij}\dot x^i\dot x^j\,dt$. Sie erfüllen
$$\ddot x^k + \Gamma^k_{ij}\,\dot x^i\dot x^j = 0,\qquad \Gamma^k_{ij} = \tfrac12\,g^{kl}\big(\partial_ig_{jl} + \partial_jg_{il} - \partial_lg_{ij}\big).$$
Auf der Kugel sind die Geodäten Großkreise.

## Herleitung

**1. Euler–Lagrange-Gleichung.** Setze $y_\epsilon = y + \epsilon\eta$ mit $\eta(a) = \eta(b) = 0$. Dann
$$\frac{d}{d\epsilon}J[y_\epsilon]\Big\vert_{\epsilon=0} = \int_a^b\Big(\frac{\partial L}{\partial y}\,\eta + \frac{\partial L}{\partial y'}\,\eta'\Big)dx .$$
Partielle Integration des zweiten Terms ergibt $\int_a^b\frac{\partial L}{\partial y'}\eta'\,dx = \Big[\frac{\partial L}{\partial y'}\eta\Big]_a^b - \int_a^b\Big(\frac{d}{dx}\frac{\partial L}{\partial y'}\Big)\eta\,dx$; der Randterm verschwindet, weil $\eta$ an den Rändern null ist. Also
$$\delta J = \int_a^b\Big(\frac{\partial L}{\partial y} - \frac{d}{dx}\frac{\partial L}{\partial y'}\Big)\eta\,dx .$$
Stationär heißt $\delta J = 0$ für alle $\eta$. Daraus folgt die Euler–Lagrange-Gleichung nach dem Fundamentallemma: Ist eine stetige Funktion $h$ an einer Stelle ungleich null, etwa positiv, dann ist sie es auf einem kleinen Intervall; ein Buckel $\eta\geq0$, der nur dort ungleich null ist, gibt $\int h\eta\,dx>0$, ein Widerspruch.

**2. Beltrami-Identität.** Mit der Kettenregel ist
$$\frac{d}{dx}\Big(L - y'\frac{\partial L}{\partial y'}\Big) = \frac{\partial L}{\partial x} + \frac{\partial L}{\partial y}y' + \frac{\partial L}{\partial y'}y'' - y''\frac{\partial L}{\partial y'} - y'\frac{d}{dx}\frac{\partial L}{\partial y'} = \frac{\partial L}{\partial x} + y'\Big(\frac{\partial L}{\partial y} - \frac{d}{dx}\frac{\partial L}{\partial y'}\Big).$$
Ist $\partial L/\partial x = 0$ und gilt die Euler–Lagrange-Gleichung, so verschwindet die rechte Seite.

**3. Noether-Theorem.** Invarianz heißt: $L(q + \epsilon K, \dot q + \epsilon\dot K) = L(q, \dot q) + O(\epsilon^2)$ für jede Bahn, mit $\dot K_i = \sum_j\frac{\partial K_i}{\partial q_j}\dot q_j$. In erster Ordnung in $\epsilon$:
$$\sum_i\Big(\frac{\partial L}{\partial q_i}K_i + \frac{\partial L}{\partial\dot q_i}\dot K_i\Big) = 0 .$$
Entlang einer Lösung ist $\partial L/\partial q_i = \frac{d}{dt}\partial L/\partial\dot q_i$, also
$$0 = \sum_i\Big(\frac{d}{dt}\frac{\partial L}{\partial\dot q_i}\,K_i + \frac{\partial L}{\partial\dot q_i}\,\dot K_i\Big) = \frac{d}{dt}\sum_i\frac{\partial L}{\partial\dot q_i}K_i = \frac{dQ}{dt}.$$
*Beispiel Drehung.* $L = \tfrac12m(\dot x^2 + \dot y^2) - V(r)$ mit $r = \sqrt{x^2 + y^2}$ ist invariant unter Drehungen; eine Drehung um den kleinen Winkel $\epsilon$ ist $K = (-y, x)$. Dann $Q = m\dot x(-y) + m\dot y\,x = m(x\dot y - y\dot x) = L_z$, der Drehimpuls.

*Energie.* Hängt $L$ nicht explizit von $t$ ab, so ergibt dieselbe Rechnung wie bei Beltrami (mit $t$ statt $x$) $dH/dt = 0$ für $H = \sum_i\dot q_i\,\partial L/\partial\dot q_i - L$. Für $L = T - V$ mit $T = \tfrac12m\vert\dot q\vert^2$ ist $\sum_i\dot q_i\,\partial L/\partial\dot q_i = 2T$, also $H = T + V$.

**4. Brachistochrone.** Energieerhaltung beim Start aus der Ruhe: $\tfrac12mv^2 = mgy$, also $v = \sqrt{2gy}$. Für ein Bogenstück $ds = \sqrt{1 + y'^2}\,dx$ braucht der Körper $dt = ds/v$. Der Integrand $L = \sqrt{1 + y'^2}/\sqrt{2gy}$ hängt nicht von $x$ ab, also gilt Beltrami:
$$L - y'\frac{\partial L}{\partial y'} = \frac{\sqrt{1 + y'^2}}{\sqrt{2gy}} - \frac{y'^2}{\sqrt{2gy}\sqrt{1 + y'^2}} = \frac{1}{\sqrt{2gy}\,\sqrt{1 + y'^2}} = C .$$
Quadrieren gibt $y(1 + y'^2) = 1/(2gC^2) =: 2a$. Ansatz $y = a(1 - \cos\theta) = 2a\sin^2(\theta/2)$: Dann ist $1 + y'^2 = 2a/y = 1/\sin^2(\theta/2)$, also $y' = \cot(\theta/2)$. Mit $dy = a\sin\theta\,d\theta$ folgt
$$dx = \frac{dy}{y'} = a\sin\theta\,\tan(\theta/2)\,d\theta = 2a\sin^2(\theta/2)\,d\theta = a(1 - \cos\theta)\,d\theta,$$
also $x = a(\theta - \sin\theta)$ mit $x(0) = 0$: eine Zykloide, die Bahn eines Punktes auf einem rollenden Rad vom Radius $a$.

*Laufzeit.* $ds = a\sqrt{(1 - \cos\theta)^2 + \sin^2\theta}\,d\theta = a\sqrt{2 - 2\cos\theta}\,d\theta = 2a\sin(\theta/2)\,d\theta$ und $v = \sqrt{2ga(1 - \cos\theta)} = 2\sqrt{ga}\,\sin(\theta/2)$, also $dt = \sqrt{a/g}\,d\theta$ und $T = \sqrt{a/g}\;\theta_1$. Die Parameter folgen aus dem Endpunkt: $\frac{\theta_1 - \sin\theta_1}{1 - \cos\theta_1} = \frac{x_1}{y_1}$ und $a = \frac{y_1}{1 - \cos\theta_1}$. Bei $\theta = \pi$ liegt der tiefste Punkt, dort ist $x/y = \pi/2$. Ist $x_1/y_1>\pi/2$, so führt die schnellste Bahn unter das Ziel und steigt am Ende wieder an.

**5. Geodätengleichung.** Für $L = \tfrac12g_{ij}(x)\,\dot x^i\dot x^j$ ist $\frac{\partial L}{\partial\dot x^l} = g_{lj}\dot x^j$, $\frac{d}{dt}\frac{\partial L}{\partial\dot x^l} = \partial_ig_{lj}\,\dot x^i\dot x^j + g_{lj}\ddot x^j$ und $\frac{\partial L}{\partial x^l} = \tfrac12\partial_lg_{ij}\,\dot x^i\dot x^j$. Die Euler–Lagrange-Gleichung lautet
$$g_{lj}\ddot x^j + \partial_ig_{lj}\,\dot x^i\dot x^j - \tfrac12\partial_lg_{ij}\,\dot x^i\dot x^j = 0 .$$
Weil $\dot x^i\dot x^j$ symmetrisch ist, darf man $\partial_ig_{lj}\,\dot x^i\dot x^j$ durch $\tfrac12(\partial_ig_{lj} + \partial_jg_{li})\,\dot x^i\dot x^j$ ersetzen. Multiplikation mit der inversen Metrik $g^{kl}$ gibt $\ddot x^k + \Gamma^k_{ij}\dot x^i\dot x^j = 0$.

*Warum Energie statt Länge.* Die Länge $\int\sqrt{g_{ij}\dot x^i\dot x^j}\,dt$ hängt nicht von der Parametrisierung ab, ihre Euler–Lagrange-Gleichungen legen den Parameter deshalb nicht fest. Das Energiefunktional hat dieselben stationären Kurven und wählt die Parametrisierung mit konstanter Geschwindigkeit: Nach Cauchy–Schwarz ist $\big(\int\vert\dot x\vert\,dt\big)^2\leq(t_1 - t_0)\int\vert\dot x\vert^2\,dt$, mit Gleichheit genau bei konstanter Geschwindigkeit.

**6. Großkreise auf der Kugel.** Auf der Einheitskugel mit Polarwinkel $\theta$ und Azimut $\varphi$ ist $ds^2 = d\theta^2 + \sin^2\theta\,d\varphi^2$, also $L = \tfrac12(\dot\theta^2 + \sin^2\theta\,\dot\varphi^2)$.
- $\varphi$ kommt in $L$ nicht vor (Drehsymmetrie um die $z$-Achse). Nach Noether ist $c = \partial L/\partial\dot\varphi = \sin^2\theta\,\dot\varphi$ erhalten. Mit $\rho = \sin\theta$, dem Abstand zur Achse, ist $c = \rho\cdot\rho\dot\varphi = \rho\,\vert v\vert\sin\psi$, wobei $\psi$ der Winkel zwischen Bahn und Meridian ist: die Clairaut-Relation.
- Euler–Lagrange für $\theta$: $\ddot\theta = \sin\theta\cos\theta\,\dot\varphi^2$. Für $\varphi$: $\frac{d}{dt}(\sin^2\theta\,\dot\varphi) = 0$, also $\ddot\varphi = -2\cot\theta\,\dot\theta\,\dot\varphi$. Die Christoffel-Symbole sind $\Gamma^\theta_{\varphi\varphi} = -\sin\theta\cos\theta$ und $\Gamma^\varphi_{\theta\varphi} = \Gamma^\varphi_{\varphi\theta} = \cot\theta$.
- Der Äquator $\theta = \pi/2$, $\varphi = \omega t$ löst beide Gleichungen. Die Gleichungen sind unter Drehungen der Kugel invariant, und eine Lösung ist durch Anfangspunkt und Anfangsgeschwindigkeit eindeutig bestimmt. Jede Geodäte ist deshalb ein gedrehter Äquator, also ein Großkreis: der Schnitt der Kugel mit einer Ebene durch den Mittelpunkt.

## Code

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp
from scipy.optimize import brentq, minimize

g = 9.81                       # Erdbeschleunigung in m/s²
x1, y1 = 1.0, 0.5              # Endpunkt in m; y zeigt nach unten


def laufzeit(x, y):
    """Laufzeit entlang des Polygonzugs (x, y), Start in Ruhe bei y = 0.

    Auf jedem geraden Stück ist die Beschleunigung konstant, also ist die Durchschnittsgeschwindigkeit
    der Mittelwert der Geschwindigkeiten an den Enden: dt = 2 ds / (v_a + v_b), exakt für den Polygonzug.
    """
    ds = np.hypot(np.diff(x), np.diff(y))
    v = np.sqrt(2 * g * np.clip(y, 0, None))
    m = ds > 0                                          # Stücke der Länge null überspringen
    return np.sum(2 * ds[m] / (v[1:][m] + v[:-1][m]))


# Zykloide durch den Endpunkt: (theta - sin theta) / (1 - cos theta) = x1 / y1
theta1 = brentq(lambda t: (t - np.sin(t)) / (1 - np.cos(t)) - x1 / y1, 1e-6, 2 * np.pi - 1e-6)
a = y1 / (1 - np.cos(theta1))
T_zykloide = np.sqrt(a / g) * theta1
print(f"Zykloide: a = {a:.4f} m, theta1 = {theta1:.4f} (tiefster Punkt bei theta = pi), T = {T_zykloide:.4f} s")
```

### Experiment 1: Laufzeiten verschiedener Bahnen

Wie schnell ist die Zykloide im Vergleich zur Geraden und zu Potenzkurven $y = y_1(x/x_1)^p$? Vorher schätzen: Ist die Gerade, der kürzeste Weg, auch der schnellste?

```python
s = np.linspace(0, 1, 20001) ** 2                     # Stützstellen, dicht am Start
th = theta1 * np.linspace(0, 1, 20001)                  # die Zykloide ist in theta schon dicht am Start
bahnen = {
    "Zykloide": (a * (th - np.sin(th)), a * (1 - np.cos(th))),
    "Gerade": (x1 * s, y1 * s),
    "Potenzkurve p = 0.5": (x1 * s, y1 * s ** 0.5),
}

T_gerade_exakt = np.sqrt(2 * (x1**2 + y1**2) / (g * y1))
print(f"Kontrolle: Gerade numerisch {laufzeit(*bahnen['Gerade']):.5f} s, exakt {T_gerade_exakt:.5f} s")
print(f"Kontrolle: Zykloide numerisch {laufzeit(*bahnen['Zykloide']):.5f} s, exakt {T_zykloide:.5f} s")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
for name, (x, y) in bahnen.items():
    ax1.plot(x, y, label=f"{name}: {laufzeit(x, y):.4f} s")
ax1.plot([0, x1], [0, y1], "ko")
ax1.invert_yaxis()
ax1.set_aspect("equal")
ax1.set_xlabel("x [m]")
ax1.set_ylabel("y [m], nach unten")
ax1.set_title("Bahnen von (0, 0) nach (1, 0.5)")
ax1.legend()

exponenten = np.linspace(0.15, 1.0, 60)
zeiten = [laufzeit(x1 * s, y1 * s ** p) for p in exponenten]
ax2.plot(exponenten, zeiten, label="Potenzkurven")
ax2.axhline(T_zykloide, color="k", ls="--", label="Zykloide")
ax2.set_xlabel("Exponent p")
ax2.set_ylabel("Laufzeit [s]")
ax2.set_title("Laufzeit der Potenzkurven y = y1 (x/x1)^p")
ax2.legend()
plt.tight_layout()
plt.show()
```

### Experiment 2: Die direkte Methode

Findet eine numerische Minimierung der Laufzeit über alle Polygonzüge von selbst die Zykloide, ohne die Euler–Lagrange-Gleichung zu kennen? Die Unbekannten sind die $y$-Werte an festen $x$-Stellen; gestartet wird mit der Geraden.

```python
n = 80
x_gitter = x1 * np.linspace(0, 1, n + 2) ** 2           # dicht am Start, wo die Bahn steil ist


def zeit_von(y_innen):
    y = np.concatenate([[0.0], y_innen, [y1]])
    return laufzeit(x_gitter, y)


start = y1 * x_gitter[1:-1] / x1
ergebnis = minimize(zeit_von, start, method="L-BFGS-B", bounds=[(1e-9, None)] * n, options={"maxiter": 5000})
y_opt = np.concatenate([[0.0], ergebnis.x, [y1]])
print(f"Minimierung: T = {ergebnis.fun:.5f} s nach {ergebnis.nit} Schritten, Zykloide exakt {T_zykloide:.5f} s")

fig, ax = plt.subplots(figsize=(8, 4.5))
th = np.linspace(0, theta1, 400)
ax.plot(a * (th - np.sin(th)), a * (1 - np.cos(th)), "k-", lw=3, alpha=0.4, label="Zykloide")
ax.plot(x_gitter, y_opt, "o-", ms=3, label="numerisch minimierter Polygonzug")
ax.plot(x_gitter, y1 * x_gitter / x1, ":", label="Startwert (Gerade)")
ax.invert_yaxis()
ax.set_aspect("equal")
ax.set_xlabel("x [m]")
ax.set_ylabel("y [m], nach unten")
ax.set_title("Direkte Methode gegen Euler–Lagrange")
ax.legend()
plt.show()
```

### Experiment 3: Noether am Zentralkraftproblem

Ein Teilchen bewegt sich im Potential $V = -1/r$ und einmal zusätzlich in $\frac{\epsilon}{2}(x^2 - y^2)$, das die Drehsymmetrie bricht, aber nicht von der Zeit abhängt. Welche Größe bleibt jeweils erhalten, Energie oder Drehimpuls? Vorher aus dem Noether-Theorem vorhersagen.

```python
def bewegung(t, z, eps):
    x, y, vx, vy = z
    r3 = (x**2 + y**2) ** 1.5
    return [vx, vy, -x / r3 - eps * x, -y / r3 + eps * y]


def energie(z, eps):
    x, y, vx, vy = z
    return 0.5 * (vx**2 + vy**2) - 1 / np.hypot(x, y) + 0.5 * eps * (x**2 - y**2)


def drehimpuls(z):
    x, y, vx, vy = z
    return x * vy - y * vx


z0 = [1.0, 0.0, 0.0, 0.8]
t = np.linspace(0, 60, 3000)
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for eps, farbe in [(0.0, "C0"), (0.01, "C1")]:
    lsg = solve_ivp(bewegung, (0, 60), z0, t_eval=t, args=(eps,), method="DOP853", rtol=1e-11, atol=1e-12)
    axes[0].plot(lsg.y[0], lsg.y[1], color=farbe, lw=0.7, label=f"ε = {eps}")
    axes[1].plot(t, energie(lsg.y, eps) - energie(z0, eps), color=farbe, label=f"ε = {eps}")
    axes[2].plot(t, drehimpuls(lsg.y), color=farbe, label=f"ε = {eps}")
axes[0].set_aspect("equal")
axes[0].set_title("Bahn")
axes[0].set_xlabel("x")
axes[0].set_ylabel("y")
axes[1].set_title("Energie: Abweichung vom Startwert")
axes[1].set_xlabel("Zeit")
axes[2].set_title("Drehimpuls L_z")
axes[2].set_xlabel("Zeit")
for ax in axes:
    ax.legend()
plt.tight_layout()
plt.show()
```

### Experiment 4: Geodäten auf der Kugel

Liegt eine numerisch integrierte Geodäte in einer Ebene durch den Mittelpunkt, und bleibt die Clairaut-Größe $c = \sin^2\theta\,\dot\varphi$ konstant?

```python
def geodaete(t, z):
    th, ph, dth, dph = z
    return [dth, dph, np.sin(th) * np.cos(th) * dph**2, -2 / np.tan(th) * dth * dph]


z0 = [1.0, 0.0, 0.3, 1.0]                                  # theta, phi, ihre Ableitungen
tempo = np.sqrt(z0[2] ** 2 + np.sin(z0[0]) ** 2 * z0[3] ** 2)
t = np.linspace(0, 2 * np.pi / tempo, 600)                 # eine volle Runde
lsg = solve_ivp(geodaete, (0, t[-1]), z0, t_eval=t, method="DOP853", rtol=1e-11, atol=1e-12)
th, ph, dth, dph = lsg.y
punkte = np.array([np.sin(th) * np.cos(ph), np.sin(th) * np.sin(ph), np.cos(th)])

# Normale der Ebene aus Startpunkt und Startrichtung
r0 = punkte[:, 0]
v0 = np.array([np.cos(th[0]) * np.cos(ph[0]) * dth[0] - np.sin(th[0]) * np.sin(ph[0]) * dph[0],
               np.cos(th[0]) * np.sin(ph[0]) * dth[0] + np.sin(th[0]) * np.cos(ph[0]) * dph[0],
               -np.sin(th[0]) * dth[0]])
normale = np.cross(r0, v0) / np.linalg.norm(np.cross(r0, v0))
abstand = normale @ punkte
c = np.sin(th) ** 2 * dph
print(f"größter Abstand von der Ebene durch den Mittelpunkt: {np.abs(abstand).max():.1e}")
print(f"Clairaut-Größe c: von {c.min():.10f} bis {c.max():.10f}")

fig = plt.figure(figsize=(12, 5))
ax = fig.add_subplot(1, 2, 1, projection="3d")
u, w = np.meshgrid(np.linspace(0, 2 * np.pi, 40), np.linspace(0, np.pi, 20))
ax.plot_wireframe(np.cos(u) * np.sin(w), np.sin(u) * np.sin(w), np.cos(w), color="0.8", lw=0.5)
ax.plot(*punkte, color="C3", lw=2, label="Geodäte")
ax.set_box_aspect((1, 1, 1))
ax.set_axis_off()
ax.set_title("Geodäte auf der Einheitskugel")
ax.legend()
ax2 = fig.add_subplot(1, 2, 2)
ax2.plot(t, abstand, label="Abstand von der Ebene")
ax2.plot(t, c - c[0], label="c(t) − c(0)")
ax2.set_xlabel("Parameter t")
ax2.set_title("Ebene und Clairaut-Größe entlang der Bahn")
ax2.legend()
plt.tight_layout()
plt.show()
```
